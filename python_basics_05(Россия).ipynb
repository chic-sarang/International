## 1. Зачем использовать цикл for со списком?

В Уроке 4 мы узнали, что список может хранить много значений, и что одно значение можно получить по его **индексу** (номеру позиции). Первое значение находится под индексом 0, второе - под индексом 1, и так далее.

Поэтому, если нужно вывести все оценки из списка, можно написать по одной строке для каждого индекса:

```python
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
```

Это работает. Но есть две проблемы:

- Если в списке 100 значений, придётся написать 100 строк.
- Если добавить новое значение с помощью `append()`, придётся самому добавить ещё одну строку `print()`.

Эту проблему решает **цикл for**. Цикл for достаёт значения из списка **по одному, от первого до последнего**, и выполняет один и тот же код для каждого значения. Неважно, какой длины список. Код нужно написать только один раз.

````python
# Длинный способ: по одной строке на каждое значение
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
````

````python
# Короткий способ: цикл for
# Результат точно такой же, но всего 2 строки
scores = [85, 92, 78, 90, 66]
for score in scores:
    print(score)
````

````python
# Даже если список становится длиннее, цикл for остаётся тем же
# Не нужно добавлять ни одной строки print()
scores = [85, 92, 78, 90, 66]
scores.append(100)      # Добавляем ещё одно значение в конец (теперь значений 6)
scores.append(73)       # И ещё одно (теперь значений 7)

for score in scores:    # Те же 2 строки, что и выше
    print(score)
````

## 2. Базовая форма: `for значение in список`

Код выглядит так:

```python
for score in scores:
    код, который нужно повторять
```

Его можно прочитать как предложение: **"для каждого score в scores выполни код ниже"**.

Разберём каждую часть:

| Часть | Значение |
|---|---|
| `for` | "повтори следующее для каждого значения" |
| `score` | переменная. Каждый раз сюда сохраняется одно значение из списка |
| `in` | "в" |
| `scores` | список, по которому мы проходим |
| `:` | двоеточие. Обязательно должно стоять в конце строки `for` |
| `    код, который нужно повторять` | код, который выполняется для каждого значения. Должен быть с отступом (4 пробела) |

**Как это работает, шаг за шагом** (при `scores = [85, 92, 78]`):

| Повторение | Что сохраняется в `score` | Что происходит |
|---|---|---|
| 1-е | `85` (значение под индексом 0) | Код с отступом выполняется при `score` = 85 |
| 2-е | `92` (значение под индексом 1) | Код с отступом выполняется при `score` = 92 |
| 3-е | `78` (значение под индексом 2) | Код с отступом выполняется при `score` = 78 |
| - | значений больше нет | Цикл заканчивается, и Python переходит к коду после цикла |

Значит, цикл выполняется **столько раз, сколько значений в списке**. Список из 3 значений → 3 повторения. Список из 100 значений → 100 повторений.

````python
# Выводим каждый фрукт по одному
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:        # 1-й раз: fruit = "apple", 2-й раз: fruit = "banana", 3-й раз: fruit = "cherry"
    print(fruit)            # Выводим значение, которое сейчас лежит в fruit
````

````python
# Список чисел
numbers = [10, 20, 30, 40]

for number in numbers:      # number становится 10, потом 20, потом 30, потом 40
    print(number)
````

````python
# Список логических значений (True / False)
answers = [True, False, True]

for answer in answers:      # answer становится True, потом False, потом True
    print(answer)
````

````python
# С каждым значением можно считать
# Каждая цена умножается на 2 и выводится
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)        # 1000 * 2, потом 2500 * 2, потом 3000 * 2
````

````python
# Значения внутри списка НЕ меняются
# В ячейке выше мы выводили price * 2, но сам список остался прежним
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)

print(prices)               # По-прежнему [1000, 2500, 3000]
````

````python
# Соединяем строки с помощью +
# Помните: + может соединить две строки в одну ("Hello, " + "Bill" -> "Hello, Bill")
names = ["Bill", "Anna", "Tom"]

for name in names:
    print("Hello, " + name)
````

````python
# Если список пустой, цикл выполняется 0 раз
# Ничего не выводится, и ошибки тоже нет
empty_list = []

for value in empty_list:
    print("This never prints.")

print("The loop ran 0 times.")
````

## 3. Отступы: какие строки повторяются?

**Отступ** - это сдвиг строки вправо с помощью пробелов в её начале. В Python используют **4 пробела**.

```python
for fruit in fruits:
    print(fruit)          # <- у этой строки есть отступ (4 пробела в начале)
```

### Зачем нужны отступы?

Цикл `for` повторяет какой-то код. Но как Python узнаёт, **какие именно строки** нужно повторять и где повторяемая часть заканчивается?

Python смотрит на отступы:

- Строки **с отступом** сразу после строки `for` → **внутри** цикла → повторяются для каждого значения
- Первая строка, где отступа **снова нет** → **вне** цикла → выполняется только один раз, после окончания цикла

Строки с отступом связаны между собой, как одна группа. Такая группа строк называется **блоком (block)**. Блок цикла `for` - это и есть "код, который нужно повторять".

(Это то же правило, что и для `if` в Уроке 2: строки с отступом под `if` выполняются, только когда условие равно True.)

### Тот же код, другой отступ, другой результат

В двух ячейках ниже абсолютно одинаковые строки. **Единственное** различие - есть ли отступ у последнего `print()`. Посмотрите, как меняется результат.

````python
# Вариант A: у второго print() есть отступ -> он ВНУТРИ цикла
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
    print("Done!")          # Есть отступ -> внутри цикла -> выводится 3 раза
````

````python
# Вариант B: у второго print() НЕТ отступа -> он ВНЕ цикла
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
print("Done!")              # Нет отступа -> вне цикла -> выводится только один раз, в конце
````

### В блоке может быть много строк

Каждая строка с отступом выполняется, по порядку сверху вниз, для каждого значения. Затем приходит следующее значение, и весь блок снова выполняется сверху.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Now the value is:")      # 1-я строка блока
    print(fruit)                    # 2-я строка блока
    print("----------")             # 3-я строка блока

print("The loop is finished.")      # Нет отступа -> вне цикла -> выполняется один раз
````

### У всех строк блока должен быть одинаковый отступ

Если у одной строки 4 пробела, а у следующей 2 пробела, Python запутается и покажет ошибку. Всегда используйте ровно 4 пробела. (В Jupyter клавиша **Tab** добавляет 4 пробела за вас.)

```python
for fruit in fruits:
    print(fruit)       # 4 пробела
  print("Done!")       # 2 пробела -> IndentationError!
```

## 4. Имя переменной цикла

В `for score in scores:` слово `score` - это **новая переменная**, которую создаёт цикл for. При каждом повторении в неё сохраняется одно значение из списка.

**Этой переменной можно дать любое имя.** Python не важно, как она называется. Он просто сохраняет каждое значение в коробку с этим именем.

В трёх ячейках ниже используются разные имена (`x`, `score`, `banana`), но результат абсолютно одинаковый.

````python
scores = [85, 92, 78]

for x in scores:            # Переменная называется x
    print(x)
````

````python
scores = [85, 92, 78]

for score in scores:        # Переменная называется score
    print(score)
````

````python
scores = [85, 92, 78]

for banana in scores:       # Переменная называется banana (странно, но работает!)
    print(banana)
````

Все три варианта работают. Но какой понятнее всего?

- `x` не говорит, что это за значение.
- `banana` сбивает с толку, ведь это оценки, а не бананы.
- `score` точно говорит, что это: **одна оценка** из списка оценок.

**Хорошая привычка:** называйте список словом во множественном числе, а переменную цикла - тем же словом в **единственном** числе.

| Список (много значений) | Переменная цикла (одно значение) |
|---|---|
| `scores` | `score` |
| `fruits` | `fruit` |
| `names` | `name` |
| `prices` | `price` |

Тогда при чтении кода легко отличить "весь список" (`scores`) от "одного значения" (`score`).

**Одно правило:** имя переменной должно **отличаться** от имени списка. Что будет, если они совпадут, мы увидим в разделе 12 (Частые ошибки).

## 5. Вывод нескольких значений сразу через запятую

До сих пор мы в основном выводили по одному значению: `print(score)`.

Но иногда хочется вывести подпись вместе со значением, например `Total: 255`. Для этого внутрь `print()` можно поставить **несколько значений**, разделив их **запятыми** `,`.

```python
print(значение1, значение2, значение3)
```

Правила:
1. Значения выводятся **в одну строку, по порядку**.
2. Python **автоматически ставит один пробел** между значениями. Добавлять пробелы самому не нужно.
3. Текст внутри кавычек `" "` выводится **как есть**.
4. То, что вне кавычек, **сначала вычисляется**, а потом выводится результат. Если это переменная, выводится значение, которое в ней лежит.

````python
# Несколько строк
print("apple", "banana", "cherry")     # Между значениями автоматически ставится один пробел
````

````python
# Несколько чисел
print(10, 20, 30)
````

````python
# Строка и число вместе
print("Total:", 255)                   # "Total:" выводится как есть, потом пробел, потом 255
````

````python
# Строка и переменная вместе
score = 85
print("Score:", score)                 # score - это переменная, поэтому выводится значение внутри неё (85)
````

````python
# Строка и вычисление вместе
price = 1000
print("Double:", price * 2)            # Сначала вычисляется price * 2 (2000), потом выводится
````

````python
# Внутри кавычек и вне кавычек
price = 1000
print("price * 2")                     # Внутри кавычек -> выводится как есть: price * 2
print(price * 2)                       # Вне кавычек -> вычисляется: 2000
````

### Запятая `,` и плюс `+`

Мы также знаем, что `+` может соединять строки. В чём разница?

| | Запятая `,` | Плюс `+` |
|---|---|---|
| Пробел между значениями | Ставится автоматически | Не ставится. Нужно добавить самому |
| Строка + число | ✅ Работает | ❌ Ошибка |

`+` может соединять только **строку со строкой**. Если попытаться соединить строку и число через `+`, Python покажет ошибку. С запятыми смешивать строки и числа можно без проблем.

````python
total = 255

print("Total:", total)                 # ✅ Запятая: строка и число вместе -> работает

# print("Total: " + total)             # ❌ Плюс: строка + число -> TypeError!
````

## 6. Сложение значений

Одна из самых частых задач для цикла for - **сложить** все значения списка. Например: какова сумма всех оценок?

Хитрость в том, чтобы использовать переменную как **накопитель (накопленную сумму)**:

1. **Перед циклом:** создайте переменную и присвойте ей 0. → `total = 0`
2. **Внутри цикла:** прибавляйте к ней текущее значение. → `total = total + score`
3. **После цикла:** в переменной лежит сумма всех значений.

Посмотрите внимательно на шаг 2: `total = total + score`

- `=` не значит "равно". Оно значит **"сохрани значение справа в коробку слева"**.
- Поэтому Python сначала вычисляет правую часть, `total + score` (старая сумма плюс текущая оценка),
- а потом сохраняет этот результат обратно в `total`.
- Это та же идея, что и `age = age + 1`: берём старое значение, что-то прибавляем и кладём обратно в ту же коробку.

````python
# Складываем все оценки
scores = [85, 92, 78, 90, 66]
total = 0                       # Шаг 1: накопитель начинается с 0 (ПЕРЕД циклом)

for score in scores:
    total = total + score       # Шаг 2: старая сумма + текущая оценка -> сохраняем обратно в total

print("Total:", total)          # Шаг 3: ПОСЛЕ цикла (без отступа) выводим результат один раз
````

````python
# Посмотрим, как меняется total при каждом повторении
# Тот же код, что и выше, но с print() ВНУТРИ цикла (с отступом)
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total = total + score
    print("score:", score, "total:", total)      # Есть отступ -> выводится при каждом повторении

# 1-е: 0 + 85 = 85
# 2-е: 85 + 92 = 177
# 3-е: 177 + 78 = 255
# 4-е: 255 + 90 = 345
# 5-е: 345 + 66 = 411
````

### Более короткая запись: `+=`

`total = total + score` используется так часто, что в Python есть более короткая запись:

```python
total += score
```

`+=` означает **"прибавь правую часть к переменной слева и сохрани результат обратно в неё"**.

`total += score` и `total = total + score` означают **абсолютно одно и то же**. Можно использовать любой вариант. Многие программисты на Python пишут коротко, поэтому вы будете часто это видеть.

````python
# Длинная запись
scores = [85, 92, 78]
total = 0
for score in scores:
    total = total + score
print("Long form:", total)

# Короткая запись
total = 0
for score in scores:
    total += score              # то же самое, что: total = total + score
print("Short form:", total)     # Результат тот же
````

````python
# Общая стоимость корзины
cart = [3500, 1200, 8000, 2300]
total_price = 0                 # Перед циклом: начинаем с 0

for price in cart:
    total_price += price        # то же самое, что: total_price = total_price + price

print("Total price:", total_price)
````

### Среднее значение

**Среднее значение** - это сумма, делённая на количество значений.

Как узнать количество значений? Вспомните `len()`: она показывает, сколько значений в списке. `len([85, 92, 78])` равно `3`.

Значит: **среднее = сумма / len(список)**

Важно: вычисляйте среднее **после** цикла, когда сумма уже готова.

````python
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total += score              # то же самое, что: total = total + score

average = total / len(scores)   # После цикла: сумма (411) / количество оценок (5)
print("Total:", total)
print("Number of scores:", len(scores))
print("Average:", average)
````

## 7. if внутри цикла

`if` можно поставить **внутрь** цикла for. Тогда каждое значение проверяется по очереди, и можно что-то делать **только с нужными значениями**.

Вспомните, как работает `if`:
- `if условие:` → код с отступом под ним выполняется **только когда условие равно True**.
- `else:` → код с отступом под ним выполняется, когда условие равно False.

Когда `if` находится внутри `for`, получается **два уровня** отступа:

```python
for score in scores:          # без пробелов
    if score >= 60:           # 4 пробела -> внутри for (выполняется для каждой оценки)
        print(score)          # 8 пробелов -> внутри if (выполняется, только когда условие равно True)
```

- У строки `if` 4 пробела, потому что она относится к блоку `for`.
- У строки `print()` 8 пробелов, потому что она относится к блоку `if`, а он находится внутри блока `for`.

````python
# Выводим только сданные оценки (60 и выше)
scores = [85, 42, 78, 55, 90]

for score in scores:            # Достаём каждую оценку
    if score >= 60:             # 4 пробела: проверяем эту оценку
        print(score)            # 8 пробелов: выводим, только если она 60 или выше
````

````python
# Посмотрим, какие оценки проверялись, а какие выводились
scores = [85, 42, 78, 55, 90]

for score in scores:
    print("Checking", score)    # 4 пробела: выполняется для КАЖДОЙ оценки
    if score >= 60:
        print("pass")           # 8 пробелов: выполняется, только когда score >= 60
````

````python
# if - else: для каждой оценки выводим pass или fail
scores = [85, 42, 78, 55, 90]

for score in scores:
    if score >= 60:
        print(score, "pass")    # Выполняется, когда условие равно True
    else:
        print(score, "fail")    # Выполняется, когда условие равно False
````

````python
# Выводим только чётные числа
# Число чётное, если остаток от деления на 2 равен 0
# % даёт остаток: 8 % 2 равно 0 (чётное), 5 % 2 равно 1 (нечётное)
numbers = [3, 8, 5, 12, 7, 6]

for number in numbers:
    if number % 2 == 0:         # Остаток равен 0? -> чётное
        print(number)
````

````python
# if - elif - else внутри цикла: ставим каждой оценке букву
# Python проверяет условия сверху вниз и выполняет только ПЕРВОЕ, которое равно True
scores = [95, 82, 74, 58]

for score in scores:
    if score >= 90:
        print(score, "A")
    elif score >= 80:
        print(score, "B")
    elif score >= 70:
        print(score, "C")
    else:
        print(score, "F")
````

````python
# Складываем только некоторые значения: if + накопитель
# Складываем только чётные числа
numbers = [3, 8, 5, 12, 7, 6]
even_total = 0                  # Перед циклом: начинаем с 0

for number in numbers:
    if number % 2 == 0:         # Только если число чётное...
        even_total += number    # ...прибавляем его (то же самое, что: even_total = even_total + number)

print("Sum of even numbers:", even_total)    # 8 + 12 + 6 = 26
````

## 8. Подсчёт

Сколько учеников сдали? Сколько человек проголосовало за пиццу?

Для **подсчёта** используется та же идея, что и для накопителя. Единственное отличие: вместо значения мы каждый раз прибавляем **1**.

1. **Перед циклом:** `count = 0`
2. **Внутри цикла:** когда условие равно True, `count = count + 1` (или `count += 1`)
3. **После цикла:** в `count` лежит, сколько раз условие было True

````python
# Сколько учеников сдали (60 и выше)?
scores = [85, 42, 78, 55, 90]
pass_count = 0                  # Перед циклом: начинаем с 0

for score in scores:
    if score >= 60:
        pass_count += 1         # то же самое, что: pass_count = pass_count + 1

print("Passed:", pass_count)    # 85, 78, 90 -> 3
````

````python
# Считаем сразу две вещи
scores = [85, 42, 78, 55, 90]
pass_count = 0
fail_count = 0

for score in scores:
    if score >= 60:
        pass_count += 1
    else:
        fail_count += 1

print("Passed:", pass_count)
print("Failed:", fail_count)
````

## 9. Поиск наибольшего и наименьшего значения

Как найти самую высокую оценку в списке? Подумайте, как это сделал бы человек со списком чисел на бумаге:

1. Посмотреть на первое число и запомнить его как **"самое большое на данный момент"**.
2. Посмотреть на следующее число. Оно больше запомненного? Если да, запомнить его вместо старого.
3. Делать так до конца. Число, которое запомнено в конце, и есть самое большое.

В Python можно записать ровно эти шаги. Переменная `highest` - это "число, которое мы запомнили".

**Почему начинаем с первого значения (`scores[0]`)?** Потому что нам нужно с чем-то сравнивать. Индекс 0 - это первое значение списка.

````python
# Находим самую высокую оценку
scores = [85, 92, 78, 95, 66]
highest = scores[0]             # Шаг 1: запоминаем первую оценку (85) как "самую большую на данный момент"

for score in scores:
    if score > highest:         # Шаг 2: эта оценка больше запомненной?
        highest = score         #        если да, запоминаем её вместо старой

print("Highest:", highest)      # Шаг 3: то, что запомнено в конце
````

````python
# Посмотрим шаг за шагом
scores = [85, 92, 78, 95, 66]
highest = scores[0]

for score in scores:
    if score > highest:
        highest = score
    print("score:", score, "highest:", highest)     # 4 пробела: выполняется при каждом повторении
````

````python
# Наименьшее значение ищется так же, только с < вместо >
temperatures = [12, 8, 15, 3, 10]
lowest = temperatures[0]        # Начинаем с первой температуры (12)

for temperature in temperatures:
    if temperature < lowest:    # Эта меньше запомненной?
        lowest = temperature

print("Lowest:", lowest)
````

````python
# Самый дорогой товар в корзине
cart = [3500, 1200, 8000, 2300]
most_expensive = cart[0]

for price in cart:
    if price > most_expensive:
        most_expensive = price

print("Most expensive:", most_expensive)
````

## 11. in и break внутри цикла

### in

`значение in список` проверяет, есть ли значение в списке. Результат - `True` или `False`, поэтому его можно использовать как условие `if`.

```python
"milk" in ["bread", "milk"]     # True
"eggs" in ["bread", "milk"]     # False
```

**Будьте внимательны:** `in` в `for fruit in fruits:` и `in` в `if fruit in store:` выглядят одинаково, но делают разную работу.

| | Значение |
|---|---|
| `for fruit in fruits:` | **достаёт** каждое значение из fruits, по одному |
| `if fruit in store:` | **проверяет**, есть ли fruit в store (True / False) |

````python
# Какие товары из моего списка покупок есть в магазине?
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]

for item in shopping_list:              # "in" №1: достаём каждый товар
    if item in store:                   # "in" №2: есть ли этот товар в store?
        print(item, "yes")
    else:
        print(item, "no")
````

````python
# Составим список товаров, которые нужно купить где-то ещё
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]
not_found = []

for item in shopping_list:
    if item not in store:               # not in: True, когда товара НЕТ в store
        not_found.append(item)

print("Not found:", not_found)
````

### break

`break` **немедленно останавливает цикл**. Остальные значения вообще не проверяются, и Python переходит к коду после цикла.

Это удобно, когда вы нашли то, что искали, и дальше искать не нужно. `break` почти всегда используется внутри `if`.

````python
# Останавливаемся, как только найдём первую несданную оценку
scores = [85, 92, 55, 78, 40]

for score in scores:
    if score < 60:
        print("Found a failing score:", score)
        break                           # Останавливаем цикл прямо здесь
    print(score, "is OK")

print("The loop is over.")              # 78 и 40 так и не проверялись
````

````python
# Есть ли в списке кто-то по имени Tom?
names = ["Bill", "Anna", "Tom", "Kate", "Mike"]

for name in names:
    print("Checking", name)
    if name == "Tom":
        print("Found Tom!")
        break                           # Проверять Kate и Mike не нужно
````

## 12. Частые ошибки

Эти ошибки делает почти каждый, кто использует цикл for со списком. Для каждой мы посмотрим:

- ❌ **Неправильный код**
- 🔍 **Что пошло не так**
- ✅ **Правильный способ**

Некоторый неправильный код вызывает ошибку. Такой код записан в виде комментария (с `#` в начале), чтобы он не останавливал блокнот. Если хотите увидеть настоящее сообщение об ошибке, уберите `#`, запустите ячейку, а потом верните `#` обратно.

Другой неправильный код **не вызывает ошибку, но даёт неправильный результат**. Такие ошибки ещё опаснее, потому что Python вас не предупреждает.

### Ошибка 1: Забытое двоеточие

````python
# ❌ Неправильно
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits           # SyntaxError
#     print(fruit)
````

🔍 **Что пошло не так?**

- Строка `for` всегда должна заканчиваться двоеточием `:`.
- Двоеточие говорит Python: "код, который нужно повторять, начинается со следующей строки".
- Без него Python не понимает код и показывает `SyntaxError` ("в коде нарушена грамматика").

✅ **Правильный способ:** Поставьте `:` в конце строки `for`.

````python
# ✅ Правильно
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # Двоеточие в конце
    print(fruit)
````

### Ошибка 2: Забытый отступ

````python
# ❌ Неправильно
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits:
# print(fruit)                  # IndentationError
````

🔍 **Что пошло не так?**

- После строки `for` должна быть хотя бы одна строка **с отступом**: код, который нужно повторять.
- Здесь у `print(fruit)` нет пробелов в начале, поэтому Python считает, что циклу `for` нечего повторять.
- Ошибка называется `IndentationError` ("неправильный отступ").

✅ **Правильный способ:** Сделайте отступ в 4 пробела у кода, который нужно повторять (или нажмите Tab один раз).

````python
# ✅ Правильно
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # 4 пробела в начале
````

### Ошибка 3: Отступ у строки, которая должна выполниться один раз

````python
# ❌ Неправильно: мы хотим вывести сумму только один раз, в конце
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score
    print("Total:", total)      # Есть отступ -> внутри цикла -> выводится 3 раза!
````

🔍 **Что пошло не так?**

- У `print("Total:", total)` есть отступ, поэтому эта строка находится **внутри** цикла.
- Значит, она выполняется при каждом повторении и выводит сумму, пока та ещё складывается.
- Ошибки нет, но мы получаем 3 строки вместо 1, и первые две - это не настоящая сумма.

✅ **Правильный способ:** Уберите отступ у строки, которая должна выполниться один раз. Тогда она окажется **вне** цикла и выполнится только после его окончания.

````python
# ✅ Правильно
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score              # Внутри цикла: складываем
print("Total:", total)          # Нет отступа -> вне цикла -> выводится один раз
````

### Ошибка 4: Одинаковое имя у списка и переменной цикла

````python
# ❌ Неправильно
fruits = ["apple", "banana", "cherry"]

for fruits in fruits:           # У переменной цикла ТАКОЕ ЖЕ имя, как у списка
    print(fruits)               # Это выводится правильно...

print(fruits)                   # ...но теперь в fruits просто "cherry". Списка больше нет!
````

🔍 **Что пошло не так?**

- Цикл как будто работает, поэтому ошибку трудно заметить.
- Но при каждом повторении значение сохраняется в коробку с именем `fruits`. Помните: когда в переменную сохраняется новое значение, старое исчезает.
- Поэтому список, который лежал в `fruits`, заменяется на `"apple"`, потом на `"banana"`, потом на `"cherry"`.
- После цикла в `fruits` лежит только `"cherry"`. Список пропал.

✅ **Правильный способ:** Дайте переменной цикла **другое** имя. Лучше всего - единственное число от имени списка: `fruits` → `fruit`.

````python
# ✅ Правильно
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # fruit (одно значение) и fruits (список) - разные имена
    print(fruit)

print(fruits)                   # Список в целости
````

### Ошибка 5: Имя списка вместо переменной цикла

````python
# ❌ Неправильно: мы хотим вывести каждый фрукт
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruits)               # fruits (с s) -> весь список выводится 3 раза
````

🔍 **Что пошло не так?**

- Внутри цикла мы написали `fruits` (весь список) вместо `fruit` (одно значение).
- Имена почти одинаковые, отличаются только одной буквой `s`. Поэтому эту ошибку так легко сделать.
- Цикл выполняется 3 раза, и каждый раз выводит весь список.

✅ **Правильный способ:** Внутри цикла используйте **переменную цикла** (`fruit`), чтобы получить текущее значение.

````python
# ✅ Правильно
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # fruit (без s) -> по одному значению за раз
````

### Ошибка 6: Имя списка в кавычках

````python
# ❌ Неправильно
fruits = ["apple", "banana", "cherry"]

for fruit in "fruits":          # "fruits" в кавычках!
    print(fruit)                # f, r, u, i, t, s ... по одной букве
````

🔍 **Что пошло не так?**

- `fruits` (без кавычек) - это **переменная**, в которой лежит наш список.
- `"fruits"` (в кавычках) - это просто **строка**: 6 букв f, r, u, i, t, s.
- Цикл for может проходить и по строке. Он достаёт **по одной букве**. Поэтому мы получаем 6 букв вместо 3 фруктов.
- Ошибки нет, поэтому это может сбить с толку.

✅ **Правильный способ:** Пишите имя переменной без кавычек.

````python
# ✅ Правильно
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # Без кавычек -> переменная (список)
    print(fruit)
````

### Ошибка 7: Обнуление накопителя внутри цикла

````python
# ❌ Неправильно
scores = [85, 92, 78]

for score in scores:
    total = 0                   # Есть отступ -> внутри цикла -> выполняется при каждом повторении!
    total += score

print("Total:", total)          # 78, а не 255
````

🔍 **Что пошло не так?**

- У `total = 0` есть отступ, поэтому строка находится внутри цикла. Она выполняется **3 раза**.
- При каждом повторении сумма сбрасывается в 0, и прибавляется только текущая оценка.
- 1-е: 0 + 85 = 85 → 2-е: **снова 0**, 0 + 92 = 92 → 3-е: **снова 0**, 0 + 78 = 78
- Поэтому в конце в `total` лежит только последняя оценка.

✅ **Правильный способ:** Напишите `total = 0` **один раз, перед** циклом (без отступа).

````python
# ✅ Правильно
scores = [85, 92, 78]
total = 0                       # Перед циклом -> выполняется только один раз

for score in scores:
    total += score              # то же самое, что: total = total + score

print("Total:", total)          # 85 + 92 + 78 = 255
````

### Ошибка 8: Вычисление среднего внутри цикла

````python
# ❌ Неправильно
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score
    average = total / len(scores)       # Есть отступ -> вычисляется при каждом повторении
    print("Average:", average)
````

🔍 **Что пошло не так?**

- Чтобы вычислить среднее, нужна сумма **всех** значений.
- Но внутри цикла сумма ещё не готова.
- 1-е: 80 / 3 → 2-е: 170 / 3 → 3-е: 270 / 3. Только последнее - настоящее среднее.

✅ **Правильный способ:** Внутри цикла только складывайте. Среднее вычисляйте **после** цикла, когда сумма готова.

````python
# ✅ Правильно
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score                      # Внутри цикла: только складываем

average = total / len(scores)           # После цикла: 270 / 3
print("Average:", average)
````

### Ошибка 9: Попытка изменить список через переменную цикла

````python
# ❌ Неправильно: мы хотим добавить 5 бонусных баллов к каждой оценке в списке
scores = [85, 92, 78]

for score in scores:
    score = score + 5           # Кажется, что это меняет оценку...

print(scores)                   # ...но список по-прежнему [85, 92, 78]!
````

🔍 **Что пошло не так?**

- `score` - это **отдельная коробка**. При каждом повторении значение из списка **копируется** в эту коробку.
- `score = score + 5` меняет только то, что лежит в коробке `score`. Список `scores` вообще не затрагивается.
- Это как переписать число из тетради на стикер, а потом исправить число на стикере. В тетради всё остаётся как было.
- Ошибки нет, поэтому это трудно заметить.

✅ **Правильный способ:** Создайте **новый список** с помощью `append()` (раздел 10) и положите в него изменённые значения.

````python
# ✅ Правильно
scores = [85, 92, 78]
new_scores = []                         # Пустой список для изменённых значений

for score in scores:
    new_scores.append(score + 5)        # Добавляем изменённое значение в новый список

print(new_scores)                       # [90, 97, 83]
````

### Ошибка 10: Удаление значений из списка во время цикла

````python
# ❌ Неправильно: мы хотим удалить все несданные оценки (ниже 60)
scores = [50, 40, 80, 30]

for score in scores:
    if score < 60:
        scores.remove(score)

print(scores)                   # [40, 80] -> 40 всё ещё здесь!
````

🔍 **Что пошло не так?**

Цикл for проходит по списку **по позициям**: 1-е значение, потом 2-е, потом 3-е ...
Но `remove()` убирает значение, и все значения после него **сдвигаются вперёд** на одну позицию.

| Повторение | Позиция, на которую смотрит цикл | Список в этот момент | Что происходит |
|---|---|---|---|
| 1-е | 1-я позиция | `[50, 40, 80, 30]` | 50 < 60 → удаляем 50. Список становится `[40, 80, 30]` |
| 2-е | 2-я позиция | `[40, 80, 30]` | На 2-й позиции теперь **80**. 40 сдвинулось на 1-ю позицию, поэтому цикл его **пропустил**! |
| 3-е | 3-я позиция | `[40, 80, 30]` | 30 < 60 → удаляем 30. Список становится `[40, 80]` |

- 40 так и не проверили, поэтому оно осталось в списке.
- Ошибки нет, поэтому это очень трудно заметить.

✅ **Правильный способ:** Не меняйте список, пока проходите по нему циклом. Вместо этого создайте **новый список** только с теми значениями, которые хотите оставить (раздел 10).

````python
# ✅ Правильно
scores = [50, 40, 80, 30]
kept = []

for score in scores:
    if score >= 60:             # Оставляем только сданные оценки
        kept.append(score)

print(kept)                     # [80]
````

### Ошибка 11: Значение вместо индекса

### Ошибка 12: Начинать "самое большое на данный момент" с 0

````python
# ❌ Неправильно: находим самую высокую температуру зимой
temperatures = [-5, -2, -8, -3]
highest = 0                     # Начинаем с 0...

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # 0 -> но 0 даже нет в списке!
````

🔍 **Что пошло не так?**

- Все температуры ниже 0. Поэтому ни одна из них не больше 0.
- `temperature > highest` ни разу не равно True, и `highest` остаётся 0 до самого конца.
- Легко подумать "надо начинать с 0", ведь накопитель мы начинаем с 0. Но поиск наибольшего значения - это другая задача.

✅ **Правильный способ:** Начинайте с **первого значения списка** (`temperatures[0]`). Тогда начальное значение всегда берётся из самого списка.

````python
# ✅ Правильно
temperatures = [-5, -2, -8, -3]
highest = temperatures[0]       # Начинаем с первого значения (-5)

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # -2
````

### Ошибка 13: Соединение строки и числа через +

````python
# ❌ Неправильно
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

# print("Total: " + total)      # TypeError: can only concatenate str (not "int") to str
````

🔍 **Что пошло не так?**

- `+` может соединять **строку со строкой**. Но `total` - это **число** (int).
- Python не знает, как соединить строку и число через `+`, поэтому показывает ошибку.
- Сообщение об ошибке значит: "строку (str) можно соединять только со строкой, а не с числом (int)".

✅ **Правильный способ:** Используйте **запятую** в `print()` (раздел 5). С запятыми строки и числа можно смешивать.

````python
# ✅ Правильно
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

print("Total:", total)          # Запятая -> работает со строкой и числом
````
