## 1. Why Do We Need a Dictionary?

Let's start with something we already know: the **list**.

A list holds many values in order, inside square brackets `[ ]`. We get a value with its **index** (its position number), and the index starts from **0**.

Now imagine we want to store the test scores of 3 students: Anna, Tom and Bill. With lists, we would need **two** lists: one for the names, and one for the scores.

```python
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]
```

The two lists are connected by their position:

| Index | 0 | 1 | 2 |
|---|---|---|---|
| `names` | "Anna" | "Tom" | "Bill" |
| `scores` | 85 | 92 | 78 |

So `names[1]` is Tom, and `scores[1]` is Tom's score.

**But this has problems:**

- To find Tom's score, you must first **remember** that Tom is at index 1. Python can't find it from the name "Tom".
- If you add or remove a name in one list but forget to do the same in the other list, the names and scores get **mixed up**.

Wouldn't it be easier if we could just ask: **"What is Tom's score?"**

That's exactly what a dictionary does.

````python
# With two lists, the name and the score are in separate places
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

# To get Tom's score, we must know that Tom is at index 1
print(names[1])         # Tom
print(scores[1])        # 92 -> Tom's score
````

````python
# The problem: if one list changes and the other doesn't, they get mixed up
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

names.remove("Anna")    # We removed Anna from names, but forgot to remove 85 from scores

print(names[0])         # Tom
print(scores[0])        # 85 -> This is Anna's score, not Tom's! Now they don't match
````

## 2. What Is a Dictionary?

Think of a real dictionary, the book you use to learn words.

- You look up a **word** (for example, "apple").
- You find its **meaning** ("a round fruit").

You don't need to know what page the word is on. You just look for the word, and the meaning is right next to it.

A Python **dictionary** works the same way. It stores values in **pairs**:

| Word in Python | Meaning | In a real dictionary | In our example |
|---|---|---|---|
| **key** | the name you look for | the word | the student's name ("Tom") |
| **value** | what you get | the meaning | the score (92) |
| **pair** | one key and its value together | one word and its meaning | "Tom" and 92 |

You can also think of it like **lockers with name tags**:
- In a **list**, each locker has a **number** (0, 1, 2 ...).
- In a **dictionary**, each locker has a **name tag** ("Anna", "Tom", "Bill" ...).

To open a locker, you use the name tag instead of the number.

## 3. Making a Dictionary

Let's start with the smallest dictionary: just **one pair**.

```python
scores = {"Tom": 92}
```

Let's look at each piece:

| Piece | Meaning |
|---|---|
| `{` | the dictionary starts (a **curly bracket**, not a square bracket) |
| `"Tom"` | the **key** (here, a string, so it's inside quotes) |
| `:` | a **colon**. It connects the key to its value |
| `92` | the **value** |
| `}` | the dictionary ends |

Remember: `=` means **"store the value on the right into the box on the left"**. So this line stores the whole dictionary in a variable named `scores`.

````python
# A dictionary with one pair
scores = {"Tom": 92}

print(scores)           # {'Tom': 92}
````

### More than one pair

To put more pairs in a dictionary, separate them with **commas** `,`. Just like in a list, where values are separated by commas.

```python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
```

- `"Anna": 85` is one pair
- `"Tom": 92` is one pair
- `"Bill": 78` is one pair

So this dictionary has **3 pairs**.

````python
# A dictionary with two pairs
scores = {"Anna": 85, "Tom": 92}
print(scores)
````

````python
# A dictionary with three pairs
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)
````

### Writing one pair per line

When there are many pairs, one long line is hard to read. You can write **one pair per line**. Python reads it exactly the same way. Don't forget the comma at the end of each pair.

````python
# The same dictionary, written one pair per line (easier to read)
scores = {
    "Anna": 85,
    "Tom": 92,
    "Bill": 78,
}
print(scores)
````

### Compare: list and dictionary

| | List | Dictionary |
|---|---|---|
| Brackets | square brackets `[ ]` | curly brackets `{ }` |
| What's inside | values only | **pairs** of `key: value` |
| Example | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92, "Bill": 78}` |

````python
# The same scores, as a list and as a dictionary
score_list = [85, 92, 78]
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list)       # Only the scores. We don't know whose they are
print(score_dict)       # Each score has a name next to it
````

````python
# A dictionary is a new data type: dict (short for "dictionary")
# type() tells you what kind of value something is
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(type(scores))     # <class 'dict'>
````

### Other kinds of dictionaries

The values can be any data type: strings, integers, floats, booleans. The keys are usually **strings**.

````python
# Values are strings: a word -> its meaning, like a real dictionary
english = {"apple": "a round fruit", "dog": "an animal that barks"}
print(english)
````

````python
# Values are strings: a country -> its capital city
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}
print(capitals)
````

````python
# Values are floats: a person -> height
heights = {"Anna": 162.5, "Tom": 175.0}
print(heights)
````

````python
# Values are booleans: a student -> did they come to class today?
attendance = {"Anna": True, "Tom": False, "Bill": True}
print(attendance)
````

````python
# An empty dictionary: no pairs at all
# We can start with an empty dictionary and add pairs later (section 5)
empty_dict = {}
print(empty_dict)
````

## 4. Getting a Value with a Key

To get a value from a dictionary, write the dictionary name and the **key** in square brackets `[ ]`:

```python
scores["Tom"]      # "the value of the key 'Tom' in scores"
```

This looks very similar to a list. The only difference is what's inside the square brackets:

| | List | Dictionary |
|---|---|---|
| What you write in `[ ]` | an **index** (a position number) | a **key** (a name) |
| Example | `score_list[1]` | `score_dict["Tom"]` |
| Meaning | "the value at position 1" | "the value whose key is 'Tom'" |

**Important:** even though we use curly brackets `{ }` to **make** a dictionary, we use square brackets `[ ]` to **get** a value from it.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # The value of the key "Tom" -> 92
print(scores["Anna"])       # The value of the key "Anna" -> 85
print(scores["Bill"])       # The value of the key "Bill" -> 78
````

````python
# Compare: getting Tom's score from a list and from a dictionary
score_list = [85, 92, 78]                           # Anna, Tom, Bill
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list[1])        # List: we must remember that Tom is at index 1
print(score_dict["Tom"])    # Dictionary: we just use the name "Tom"
````

````python
# More examples
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

print(capitals["Korea"])    # Seoul
print(capitals["France"])   # Paris
````

````python
english = {"apple": "a round fruit", "dog": "an animal that barks"}

print(english["dog"])       # Look up the word "dog" -> its meaning
````

### The value can be used like a normal variable

The value you get from a dictionary is just a normal value. You can store it in a variable, calculate with it, or compare it.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

tom_score = scores["Tom"]                   # Store Tom's score in a new variable
print(tom_score)

print(scores["Anna"] + scores["Tom"])       # 85 + 92 = 177 (we can calculate)
print(scores["Tom"] > scores["Bill"])       # 92 > 78 is True (we can compare)
````

````python
# Using the value in an if (if runs the indented code only when the condition is True)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

if scores["Bill"] >= 80:
    print("Bill passed with a high score!")
else:
    print("Bill needs to study more.")
````

### The key can be a variable

Instead of writing the key directly, you can store the key in a variable first. Then the variable goes inside `[ ]`, **without quotes**.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Anna"               # The variable name holds the string "Anna"
print(scores[name])         # scores[name] is the same as scores["Anna"] -> 85

name = "Bill"               # Now the variable holds "Bill"
print(scores[name])         # scores["Bill"] -> 78
````

**About printing several values:** in the next cell, we write `print(name, scores[name])`. When you put several values in `print()` separated by **commas** `,`, they are printed on one line, and Python puts **one space** between them automatically.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Tom"
print(name, scores[name])   # Tom 92 -> the name and the score on one line
````

## 5. Changing a Value and Adding a New Pair

To change a value or add a new pair, we use the same shape:

```python
dictionary_name[key] = value
```

Python does one of two things, depending on whether the key is **already** in the dictionary:

| Is the key already in the dictionary? | What happens |
|---|---|
| **Yes** | The old value is **replaced** by the new value |
| **No** | A **new pair** is added |

This is different from a list. In a list, you can only change values at indexes that already exist. To add, you need `append()`. In a dictionary, the same `[key] = value` both changes and adds.

### Changing a value (the key already exists)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Bill"] = 88         # "Bill" is already a key -> the value 78 is replaced by 88
print(scores)               # Only Bill's score changed
````

````python
# Changing a value using the old value (same idea as age = age + 1)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Anna"] = scores["Anna"] + 5     # Right side first: 85 + 5 = 90, then store it back
print(scores["Anna"])                   # 90
````

````python
# The short form += also works (same as: scores["Tom"] = scores["Tom"] + 3)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Tom"] += 3
print(scores["Tom"])                    # 95
````

### Adding a new pair (the key doesn't exist yet)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Kate"] = 95         # "Kate" is NOT a key yet -> a new pair is added
print(scores)               # Now there are 4 pairs
````

````python
# Starting with an empty dictionary, and adding pairs one by one
capitals = {}                   # Start empty
print(capitals)

capitals["Korea"] = "Seoul"     # Add a pair
print(capitals)

capitals["Japan"] = "Tokyo"     # Add another pair
print(capitals)

capitals["France"] = "Paris"    # And another
print(capitals)
````

````python
# Compare with a list: to add, a list needs append(), but a dictionary just uses [key] = value
fruit_list = ["apple", "banana"]
fruit_list.append("cherry")         # List: append() adds at the end
print(fruit_list)

prices = {"apple": 1000, "banana": 500}
prices["cherry"] = 3000             # Dictionary: [new key] = value
print(prices)
````

## 6. len() and in

### len(): how many pairs?

Remember `len()`: it tells you how many values are in a list. With a dictionary, `len()` tells you how many **pairs** there are.

One pair (`"Tom": 92`) counts as **1**, not 2.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(len(scores))          # 3 pairs -> 3

one_pair = {"Tom": 92}
print(len(one_pair))        # 1 pair -> 1 (not 2!)

empty_dict = {}
print(len(empty_dict))      # 0 pairs -> 0
````

````python
# len() changes when we add a pair
scores = {"Anna": 85, "Tom": 92}
print(len(scores))          # 2

scores["Bill"] = 78         # Add a new pair
print(len(scores))          # 3
````

### in: is this key in the dictionary?

Remember `in`: `value in list` checks whether the value is in the list, and gives `True` or `False`.

With a dictionary, `in` checks the **keys**.

```python
"Tom" in scores      # Is there a key "Tom"?
```

**Be careful:** `in` checks **only the keys**, not the values. Even if 92 is in the dictionary as a value, `92 in scores` is `False`.

To check that a key is **not** in the dictionary, use `not in`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # Is "Tom" a key? -> True
print("Kate" in scores)     # Is "Kate" a key? -> False
print("Kate" not in scores) # Is "Kate" NOT a key? -> True
````

````python
# in checks only the KEYS, not the values
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" is a key -> True
print(92 in scores)         # 92 is a value, not a key -> False
````

````python
# Using in with if: check first, then get the value
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    print(name, scores[name])
else:
    print(name, "is not in the list.")
````

````python
# A simple word dictionary program
english = {"apple": "a round fruit", "dog": "an animal that barks"}
word = "cat"

if word in english:
    print(english[word])
else:
    print("Sorry, I don't know that word.")
````

## 7. Deleting a Pair

To delete a pair, use `del` and the key:

```python
del dictionary_name[key]
```

`del` is short for **delete**. It removes the key **and** its value together.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

del scores["Tom"]           # Remove the key "Tom" and its value 92
print(scores)
print(len(scores))          # Now 2 pairs
````

````python
# Deleting a key that doesn't exist causes an error (we'll see it in section 10)
# So check with in first
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
    print(name, "was deleted.")
else:
    print(name, "is not in the dictionary.")
````

## 8. Using a for Loop with a Dictionary

Let's remember how a for loop works with a list:

```python
for score in score_list:
    print(score)
```

- Each repetition, one value from the list is stored in the variable `score`.
- The `for` line ends with a colon `:`.
- The code to repeat is **indented** (4 spaces). Python uses the indentation to know which lines belong to the loop.
- The first line that is not indented is outside the loop, and runs only once after the loop.

With a dictionary, the for loop gives you the **keys**, one by one.

```python
for name in scores:
    print(name)          # Anna, Tom, Bill
```

And once you have the key, you can get the value with `scores[name]`, just like in section 4.

````python
# The for loop gives the KEYS of the dictionary
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:         # name becomes "Anna", then "Tom", then "Bill"
    print(name)
````

````python
# Use the key to get the value
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(scores[name])     # scores["Anna"], then scores["Tom"], then scores["Bill"]
````

````python
# The key and the value together on one line (print with commas)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])
````

### The loop variable name

The variable after `for` (here, `name`) is a new variable. You can give it any name. Choose a name that says what the **keys** are: names → `name`, countries → `country`, words → `word`.

````python
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

for country in capitals:            # The keys are countries, so we call it country
    print(country, capitals[country])
````

### Indentation decides what is repeated

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])       # Indented -> repeated for each key
print("That's everyone!")           # Not indented -> outside the loop -> runs once
````

### Adding up the values

Remember the **running total**:

1. **Before the loop:** `total = 0`
2. **Inside the loop:** add the current value → `total = total + scores[name]`
3. **After the loop:** `total` holds the sum

`total = total + scores[name]` means: calculate the right side first (the old total + this student's score), and then store the result back in `total`. The short form is `total += scores[name]`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0                           # Before the loop: start from 0

for name in scores:
    total = total + scores[name]    # old total + this student's score

print("Total:", total)              # After the loop (not indented): 85 + 92 + 78 = 255
````

````python
# The average = total / number of values
# len(scores) is the number of pairs (= the number of students)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:
    total += scores[name]           # same as: total = total + scores[name]

average = total / len(scores)       # After the loop: 255 / 3
print("Average:", average)
````

### if inside the loop

When you put an `if` inside a `for`, there are two levels of indentation: 4 spaces for the `if` (inside the `for`), and 8 spaces for the code inside the `if`.

````python
# Print only the students who passed (60 or higher)
scores = {"Anna": 85, "Tom": 42, "Bill": 78, "Kate": 55}

for name in scores:
    if scores[name] >= 60:          # 4 spaces: check each student's score
        print(name, "passed")       # 8 spaces: only when the condition is True
````

````python
# Who came to class today?
attendance = {"Anna": True, "Tom": False, "Bill": True}

for name in attendance:
    if attendance[name]:            # The value is already True or False, so it can be the condition directly
        print(name, "is here.")
    else:
        print(name, "is absent.")
````

## 9. List vs Dictionary

| | List | Dictionary |
|---|---|---|
| Make | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92}` |
| Brackets to make it | `[ ]` | `{ }` |
| Find a value by | **index** (position number) | **key** (name) |
| Get a value | `score_list[1]` | `score_dict["Tom"]` |
| Change a value | `score_list[1] = 95` | `score_dict["Tom"] = 95` |
| Add | `score_list.append(88)` | `score_dict["Kate"] = 88` |
| Delete | `score_list.remove(92)` | `del score_dict["Tom"]` |
| `len()` counts | values | pairs |
| `in` checks | values | **keys** |
| `for x in ...` gives | values | **keys** |

**When to use which?**
- If the **order** matters, or the values don't need names → **list** (e.g. the scores of today's test, in order)
- If you want to find a value **by its name** → **dictionary** (e.g. each student's score, a word and its meaning)

## 10. Common Mistakes

These are the mistakes almost everyone makes when learning dictionaries. For each one, we will look at:

- ❌ **The wrong code**
- 🔍 **What went wrong**
- ✅ **The correct way**

Code that causes an error is written as a comment (with `#` at the beginning) so it won't stop the notebook. If you want to see the real error message, remove the `#`, run the cell, and then put the `#` back.

### Mistake 1: Using a key that doesn't exist

````python
# ❌ Wrong
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores["Kate"])     # KeyError: 'Kate'
# print(scores["tom"])      # KeyError: 'tom'
````

🔍 **What went wrong?**

- `"Kate"` is not a key in the dictionary. Python can't find it, so it shows a `KeyError` ("there is no such key").
- `"tom"` (small t) and `"Tom"` (capital T) are **different** strings. The key must be written **exactly** the same, including upper and lower case letters.

✅ **The correct way:** Write the key exactly as it is in the dictionary. If you're not sure whether the key exists, check with `in` first.

````python
# ✅ Correct
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # Exactly the same as the key

name = "Kate"
if name in scores:          # Check first
    print(scores[name])
else:
    print(name, "is not in the dictionary.")
````

### Mistake 2: Using square brackets to make a dictionary

````python
# ❌ Wrong
# scores = ["Anna": 85, "Tom": 92]      # SyntaxError
````

🔍 **What went wrong?**

- Square brackets `[ ]` are for making a **list**. A list holds only values, so Python doesn't understand `"Anna": 85` inside `[ ]`.
- The error is a `SyntaxError` ("the grammar of the code is wrong").

✅ **The correct way:** Use **curly brackets** `{ }` to make a dictionary. (To **get** a value, you still use square brackets: `scores["Anna"]`.)

````python
# ✅ Correct
scores = {"Anna": 85, "Tom": 92}    # Make: curly brackets
print(scores["Anna"])               # Get: square brackets
````

### Mistake 3: Forgetting the quotes around the key

````python
# ❌ Wrong
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[Anna])       # NameError: name 'Anna' is not defined
````

🔍 **What went wrong?**

- The key is the **string** `"Anna"`, with quotes.
- Without quotes, Python thinks `Anna` is the name of a **variable**. But there is no variable named `Anna`.
- The error is a `NameError` ("there is no variable with this name").

✅ **The correct way:** Put quotes around a string key. (No quotes are needed only when the key is stored in a variable, like `scores[name]` in section 4.)

````python
# ✅ Correct
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # A string key with quotes

name = "Anna"
print(scores[name])         # A variable without quotes (the variable holds "Anna")
````

### Mistake 4: Using an index number like a list

````python
# ❌ Wrong: we want the first student's score
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[0])          # KeyError: 0
````

🔍 **What went wrong?**

- In a list, `[0]` means "the first value". But a dictionary doesn't use index numbers. It uses **keys**.
- So Python looks for a **key** named `0`. There is no such key, so it shows a `KeyError: 0`.

✅ **The correct way:** Use the key (the name).

````python
# ✅ Correct
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # Use the key, not a number
````

### Mistake 5: Writing the same key twice

````python
# ❌ Wrong: two different students named Tom?
scores = {"Anna": 85, "Tom": 92, "Tom": 60}

print(scores)               # {'Anna': 85, 'Tom': 60} -> the first Tom (92) is gone!
print(len(scores))          # 2, not 3
````

🔍 **What went wrong?**

- In a dictionary, each key can appear **only once**.
- If you write the same key again, it works just like section 5: the key already exists, so the old value is **replaced**. 92 is replaced by 60.
- No error appears, so it's easy to miss.

✅ **The correct way:** Make every key different. If two people have the same name, give them different keys.

````python
# ✅ Correct
scores = {"Anna": 85, "Tom K.": 92, "Tom L.": 60}

print(scores)
print(len(scores))          # 3
````

### Mistake 6: Expecting in to check the values

````python
# ❌ Wrong: we want to know whether anyone got 92 points
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(92 in scores)         # False -> but Tom got 92!
````

🔍 **What went wrong?**

- With a dictionary, `in` checks **only the keys** (the names), not the values (the scores).
- 92 is a value, not a key, so the result is `False`.
- No error appears, but the answer is wrong.

✅ **The correct way:** Go through the dictionary with a for loop, and compare each value.

````python
# ✅ Correct
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    if scores[name] == 92:      # Compare each VALUE with 92
        print(name, "got 92 points.")
````

### Mistake 7: Thinking the for loop gives the values

````python
# ❌ Wrong: we want to add up all the scores
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

# for score in scores:
#     total += score        # TypeError: unsupported operand type(s) for +=: 'int' and 'str'
````

🔍 **What went wrong?**

- With a dictionary, the for loop gives the **keys**, not the values. So `score` is `"Anna"`, `"Tom"`, `"Bill"`, not 85, 92, 78.
- `total += score` tries to add a string (`"Anna"`) to a number (`0`). Python can't add a number and a string, so it shows a `TypeError`.
- Naming the variable `score` made it even more confusing. It's actually a name.

✅ **The correct way:** Name the loop variable after the keys (`name`), and use `scores[name]` to get the value.

````python
# ✅ Correct
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:             # name is the key
    total += scores[name]       # scores[name] is the value

print("Total:", total)
````

### Mistake 8: Deleting a key that doesn't exist

````python
# ❌ Wrong
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# del scores["Kate"]        # KeyError: 'Kate'
````

🔍 **What went wrong?**

- There is no key `"Kate"` in the dictionary, so there is nothing to delete.
- Just like Mistake 1, Python shows a `KeyError`.

✅ **The correct way:** Check with `in` first, and delete only if the key exists.

````python
# ✅ Correct
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
else:
    print(name, "is not in the dictionary, so nothing was deleted.")
````
