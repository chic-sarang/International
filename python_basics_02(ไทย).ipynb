## 1. คำสั่ง if คืออะไร?

ที่ผ่านมา โค้ดของเราทำงานจากบนลงล่าง ทีละบรรทัด เหมือนเดิมทุกครั้ง แต่ในชีวิตจริง เราต้องตัดสินใจเลือกอยู่ตลอดเวลา

- **ถ้า** ฝนตก ให้พกร่มไปด้วย
- **ถ้า** คุณอายุ 20 ปีขึ้นไป คุณสามารถซื้อตั๋วผู้ใหญ่ได้

คำสั่ง `if` ช่วยให้โปรแกรมตัดสินใจเลือกแบบนี้ได้ มันจะตรวจสอบ **เงื่อนไข** และรันโค้ดที่อยู่ข้างใน **เฉพาะเมื่อเงื่อนไขเป็น True เท่านั้น**

จำตัวดำเนินการเปรียบเทียบในบทที่ 1 (`>`, `<`, `==`, `!=`) ได้ไหม? ผลลัพธ์ของมันจะเป็น `True` หรือ `False` เสมอ ซึ่งก็คือสิ่งที่ `if` ต้องการใช้เป็นเงื่อนไขนั่นเอง

รูปแบบพื้นฐานมีหน้าตาแบบนี้:

```python
if เงื่อนไข:
    โค้ดที่จะรันเฉพาะเมื่อเงื่อนไขเป็น True
```

กฎสำคัญมาก 2 ข้อ:
- ท้ายบรรทัด `if` ต้องมีเครื่องหมายโคลอน `:` เสมอ
- โค้ดที่อยู่ใน `if` ต้อง **ย่อหน้า** (เลื่อนไปทางขวา 4 ช่องว่าง) การย่อหน้าคือวิธีที่ Python ใช้รู้ว่าบรรทัดไหนอยู่ "ข้างใน" `if`

````python
# เงื่อนไขคือ age >= 20 ("อายุมากกว่าหรือเท่ากับ 20")
# ผลลัพธ์ของการเปรียบเทียบนี้คือ True หรือ False
age = 25

if age >= 20:                         # 25 >= 20 เป็น True ดังนั้นโค้ดที่ย่อหน้าอยู่ข้างล่างจะทำงาน
    print("You are an adult.")        # บรรทัดนี้อยู่ใน if (ย่อหน้า 4 ช่องว่าง)

print("This line always runs.")      # บรรทัดนี้ไม่ได้ย่อหน้า จึงอยู่นอก if มันจะทำงานเสมอไม่ว่าอะไรจะเกิดขึ้น
````

````python
# ทีนี้ลองกรณีที่เงื่อนไขเป็น False ดูบ้าง
age = 15

if age >= 20:                         # 15 >= 20 เป็น False ดังนั้นโค้ดที่ย่อหน้าอยู่ข้างล่างจะถูกข้ามไป
    print("You are an adult.")        # บรรทัดนี้ไม่ทำงานเลย

print("This line always runs.")      # บรรทัดนี้อยู่นอก if จึงยังคงทำงานอยู่
````

````python
# ใน if สามารถมีได้มากกว่าหนึ่งบรรทัด
# ทุกบรรทัดที่ย่อหน้าเท่ากันจะอยู่ใน if เดียวกัน
temperature = 32

if temperature > 30:                          # 32 > 30 เป็น True
    print("It's very hot today.")             # อยู่ใน if
    print("Drink plenty of water.")           # อยู่ใน if เช่นกัน (ย่อหน้าเท่ากัน)
    print("Try to stay in the shade.")        # อยู่ใน if เช่นกัน (ย่อหน้าเท่ากัน)

print("Have a nice day!")                    # อยู่นอก if ทำงานเสมอ
````

````python
# เราสามารถใช้ == ตรวจสอบว่าค่าสองค่าเหมือนกันหรือไม่ได้ด้วย
# จำไว้ว่า: = หมายถึง "เก็บค่า" แต่ == หมายถึง "ค่าเหล่านี้เท่ากันไหม?"
password = "python123"

if password == "python123":           # สตริงสองตัวนี้เหมือนกันทุกตัวอักษร จึงเป็น True
    print("Login successful!")
````

## 2. if - else

ถ้ามีแค่ `if` โปรแกรมจะทำบางอย่างเมื่อเงื่อนไขเป็น True และจะ **ไม่ทำอะไรเลย** เมื่อเงื่อนไขเป็น False

แต่บ่อยครั้งที่เราต้องการทำ **อย่างใดอย่างหนึ่ง** ตัวอย่างเช่น:
- ถ้าสอบผ่าน ให้แสดงคำว่า "ยินดีด้วย" **ไม่อย่างนั้น** ให้แสดงคำว่า "ลองใหม่อีกครั้ง"

นี่คือหน้าที่ของ `else` คำว่า `else` หมายถึง "ไม่อย่างนั้น" โค้ดของมันจะทำงาน **เฉพาะเมื่อเงื่อนไขของ if เป็น False เท่านั้น**

```python
if เงื่อนไข:
    โค้ดที่ทำงานเมื่อเงื่อนไขเป็น True
else:
    โค้ดที่ทำงานเมื่อเงื่อนไขเป็น False
```

- `else` ไม่มีเงื่อนไขของตัวเอง มันแค่รับทุกกรณีที่ `if` ไม่ได้รับไว้
- `else` ก็ต้องมีโคลอน `:` ต่อท้าย และโค้ดของมันก็ต้องย่อหน้าด้วย
- จะมีแค่ **หนึ่ง** ในสองบล็อกที่ทำงานเสมอ ไม่มีทางทำงานทั้งคู่ และไม่มีทางที่จะไม่ทำงานเลยสักบล็อก

````python
# ตรวจสอบว่าคะแนนสอบผ่านหรือไม่ผ่าน
# สมมติว่าคะแนนผ่านคือ 60
score = 75

if score >= 60:                      # 75 >= 60 เป็น True
    print("Pass! Congratulations.")  # บรรทัดนี้จึงทำงาน
else:                                # "ไม่อย่างนั้น" - ส่วนนี้ถูกข้ามไป เพราะ if เป็น True
    print("Fail. Try again.")
````

````python
# โค้ดเดิม แต่คะแนนต่ำลง
score = 40

if score >= 60:                      # 40 >= 60 เป็น False ส่วนนี้จึงถูกข้ามไป
    print("Pass! Congratulations.")
else:                                # if เป็น False ส่วน else จึงทำงานแทน
    print("Fail. Try again.")
````

````python
# ตรวจสอบว่าตัวเลขเป็นเลขคู่หรือเลขคี่
# จำ % จากบทที่ 1 ได้ไหม? มันให้ผลลัพธ์เป็นเศษจากการหาร
# เลขคู่เมื่อหารด้วย 2 จะเหลือเศษ 0 เสมอ
number = 7

if number % 2 == 0:                  # 7 % 2 ได้ 1 และ 1 == 0 เป็น False
    print(number, "is even.")        # print() แสดงหลายค่าพร้อมกันได้ ถ้าคั่นด้วยเครื่องหมายจุลภาค
else:
    print(number, "is odd.")         # if เป็น False บรรทัดนี้จึงทำงาน
````

## 3. if - elif - else

บางครั้งก็มีตัวเลือก **มากกว่าสองทาง** ตัวอย่างเช่น การเปลี่ยนคะแนนเป็นเกรด:

- 90 ขึ้นไป → A
- 80 ขึ้นไป → B
- 70 ขึ้นไป → C
- นอกเหนือจากนั้น → F

ในกรณีนี้เราใช้ `elif` คำว่า `elif` ย่อมาจาก "else if" หมายถึง "ถ้าเงื่อนไขข้างบนเป็น False ทั้งหมด ให้ตรวจสอบเงื่อนไขนี้"

```python
if เงื่อนไข1:
    ทำงานเมื่อ เงื่อนไข1 เป็น True
elif เงื่อนไข2:
    ทำงานเมื่อ เงื่อนไข1 เป็น False และ เงื่อนไข2 เป็น True
elif เงื่อนไข3:
    ทำงานเมื่อ เงื่อนไข1 และ เงื่อนไข2 เป็น False และ เงื่อนไข3 เป็น True
else:
    ทำงานเมื่อทุกเงื่อนไขข้างบนเป็น False
```

วิธีการทำงาน:
- Python ตรวจสอบเงื่อนไข **จากบนลงล่าง ตามลำดับ**
- ทันทีที่เจอเงื่อนไขที่เป็น True **ตัวแรก** มันจะรันบล็อกนั้น แล้ว **ข้ามส่วนที่เหลือทั้งหมด**
- เพิ่ม `elif` ได้มากเท่าที่ต้องการ ส่วน `else` ตอนท้ายจะมีหรือไม่มีก็ได้

````python
# เปลี่ยนคะแนนเป็นเกรด
score = 85

if score >= 90:                # 85 >= 90 เป็น False -> ไปตรวจเงื่อนไขถัดไป
    print("Grade: A")
elif score >= 80:              # 85 >= 80 เป็น True -> บล็อกนี้ทำงาน และทุกอย่างข้างล่างถูกข้ามไป
    print("Grade: B")
elif score >= 70:              # ไม่ถูกตรวจเลย เพราะเจอเงื่อนไขที่เป็น True ข้างบนแล้ว
    print("Grade: C")
else:                          # ไม่ถูกตรวจเช่นกัน
    print("Grade: F")
````

````python
# ลองใช้คะแนนหลาย ๆ ค่ากับโค้ดเดิม
# เปลี่ยนค่าของ score แล้วรันใหม่ เพื่อดูว่าผลลัพธ์เปลี่ยนไปอย่างไร
score = 72

if score >= 90:                # 72 >= 90 เป็น False
    print("Grade: A")
elif score >= 80:              # 72 >= 80 เป็น False
    print("Grade: B")
elif score >= 70:              # 72 >= 70 เป็น True -> บล็อกนี้ทำงาน
    print("Grade: C")
else:
    print("Grade: F")
````

````python
# ทำไมลำดับจึงสำคัญ?
# Python หยุดที่เงื่อนไข True ตัวแรก ดังนั้นถ้าเราเอาเงื่อนไขที่เล็กที่สุดไว้ก่อน ผลลัพธ์จะผิด
score = 95

if score >= 70:                # 95 >= 70 เป็น True -> บล็อกนี้ทำงานก่อน และทุกอย่างที่เหลือถูกข้ามไป!
    print("Grade: C")
elif score >= 80:              # ไม่เคยถูกตรวจ
    print("Grade: B")
elif score >= 90:              # ไม่เคยถูกตรวจ ทั้งที่ 95 ควรได้ A
    print("Grade: A")

# ดังนั้นเวลาตรวจช่วงค่าแบบนี้ ให้เริ่มจากเงื่อนไขที่ใหญ่ที่สุด (เข้มงวดที่สุด) ก่อน
````

````python
# เลือกเสื้อผ้าตามอุณหภูมิ
temperature = 12

if temperature >= 25:                  # 12 >= 25 เป็น False
    print("Wear a T-shirt.")
elif temperature >= 15:                # 12 >= 15 เป็น False
    print("Wear a light jacket.")
elif temperature >= 5:                 # 12 >= 5 เป็น True -> บรรทัดนี้ทำงาน
    print("Wear a coat.")
else:                                  # ทำงานเฉพาะเมื่ออุณหภูมิต่ำกว่า 5
    print("Wear a thick padded jacket.")
````

## 4. การใช้ตัวแปร bool เป็นเงื่อนไข

เงื่อนไขของ `if` ขอแค่เป็น `True` หรือ `False` ก็พอ แปลว่าเราสามารถใช้ **ตัวแปร bool** (จากบทที่ 1) เป็นเงื่อนไขได้โดยตรง โดยไม่ต้องเปรียบเทียบอะไรเลย

```python
if is_student:        # ความหมายเหมือนกับ: if is_student == True:
```

การเขียน `== True` ไม่ผิด แต่ไม่จำเป็น นักเขียนโปรแกรม Python จึงมักไม่เขียน

ถ้าต้องการตรวจสอบว่าตัวแปร bool เป็น `False` ให้ใช้ `not`:

```python
if not is_student:    # ความหมายเหมือนกับ: if is_student == False:
```

````python
# ตัวแปร bool เก็บค่า True หรือ False อยู่แล้ว จึงใช้เป็นเงื่อนไขได้ทันที
is_student = True

if is_student:                        # is_student เป็น True บรรทัดนี้จึงทำงาน
    print("You get a student discount.")
else:
    print("Regular price.")
````

````python
# ใช้ not เพื่อตรวจสอบค่า False
# not จะกลับค่า: not True คือ False และ not False คือ True
is_raining = False

if not is_raining:                    # is_raining เป็น False ดังนั้น not is_raining เป็น True -> บรรทัดนี้ทำงาน
    print("No umbrella needed.")
else:
    print("Take an umbrella.")
````

````python
# เราสามารถเก็บผลการเปรียบเทียบไว้ในตัวแปร bool ก่อน แล้วค่อยนำมาใช้
# วิธีนี้มีประโยชน์เมื่อเงื่อนไขยาว หรือเมื่ออยากตั้งชื่อให้เงื่อนไขเข้าใจง่าย
age = 17
is_adult = age >= 20                  # 17 >= 20 เป็น False ดังนั้น is_adult จะเก็บค่า False
print(is_adult)

if is_adult:                          # is_adult เป็น False ส่วนนี้จึงถูกข้ามไป
    print("You can watch this movie.")
else:
    print("You can't watch this movie.")
````

## 5. การรวมเงื่อนไข (and, or, not)

บางครั้งเงื่อนไขเดียวก็ไม่พอ ตัวอย่างเช่น:
- คุณเล่นรถไฟเหาะได้ ถ้าคุณ **อายุ 10 ปีขึ้นไป และสูง 140 ซม. ขึ้นไป**
- คุณเข้าชมฟรี ถ้าคุณ **อายุต่ำกว่า 7 ปี หรือ 65 ปีขึ้นไป**

Python มีคำ 3 คำสำหรับรวมเงื่อนไข

| ตัวดำเนินการ | ความหมาย | ผลลัพธ์ |
|---|---|---|
| `and` | และ | เป็น True เฉพาะเมื่อ **ทั้งสองฝั่ง** เป็น True |
| `or` | หรือ | เป็น True เมื่อ **อย่างน้อยหนึ่งฝั่ง** เป็น True |
| `not` | ตรงข้าม | เปลี่ยน True เป็น False และ False เป็น True |

````python
# ก่อนอื่น มาดูกันว่า and / or ให้ผลลัพธ์อะไรเมื่อใช้เดี่ยว ๆ
print(True and True)     # ทั้งคู่เป็น True -> True
print(True and False)    # มีฝั่งหนึ่งเป็น False -> False
print(True or False)     # อย่างน้อยหนึ่งฝั่งเป็น True -> True
print(False or False)    # ทั้งคู่เป็น False -> False
print(not True)          # ตรงข้ามของ True -> False
````

````python
# and: ทุกเงื่อนไขต้องเป็น True
# กฎ: อายุ 10 ปีขึ้นไป และ สูง 140 ซม. ขึ้นไป
age = 12
height = 135

if age >= 10 and height >= 140:       # 12 >= 10 เป็น True แต่ 135 >= 140 เป็น False -> True and False -> False
    print("You can ride the roller coaster.")
else:
    print("Sorry, you can't ride it yet.")  # เงื่อนไขเป็น False บรรทัดนี้จึงทำงาน
````

````python
# or: อย่างน้อยหนึ่งเงื่อนไขต้องเป็น True
# กฎ: อายุต่ำกว่า 7 ปี หรือ 65 ปีขึ้นไป เข้าชมฟรี
age = 70

if age < 7 or age >= 65:              # 70 < 7 เป็น False แต่ 70 >= 65 เป็น True -> False or True -> True
    print("Free entry!")              # เงื่อนไขเป็น True บรรทัดนี้จึงทำงาน
else:
    print("Please buy a ticket.")
````

````python
# ตรวจสอบว่าตัวเลขอยู่ในช่วงที่กำหนดหรือไม่
# "score อยู่ระหว่าง 0 ถึง 100" หมายถึง "score ตั้งแต่ 0 ขึ้นไป และ score ไม่เกิน 100"
score = 105

if score >= 0 and score <= 100:       # 105 >= 0 เป็น True แต่ 105 <= 100 เป็น False -> False
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")

# Python ยังให้เราเขียนแบบสั้นลงได้ เหมือนในวิชาคณิตศาสตร์
if 0 <= score <= 100:                 # ความหมายเหมือนกับบรรทัดข้างบนทุกประการ
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")
````

## 6. คำสั่ง if ซ้อนกัน

เราสามารถใส่ `if` **ไว้ข้างใน** `if` อีกตัวหนึ่งได้ เรียกว่า if **ซ้อนกัน**

วิธีนี้มีประโยชน์เมื่อเราต้องการตรวจเงื่อนไขที่สอง **หลังจากที่** เงื่อนไขแรกเป็น True แล้วเท่านั้น ตัวอย่างเช่น ตรวจก่อนว่าผู้ใช้เข้าสู่ระบบแล้วหรือยัง แล้วจึงค่อยตรวจว่าเป็นผู้ดูแลระบบหรือไม่

`if` ตัวใน จะย่อหน้ามากกว่า `if` ตัวนอกหนึ่งระดับ (เพิ่มอีก 4 ช่องว่าง)

````python
# ตรวจการเข้าสู่ระบบก่อน และถ้าเข้าสู่ระบบแล้วเท่านั้น จึงตรวจสิทธิ์ผู้ดูแลระบบ
is_logged_in = True
is_admin = False

if is_logged_in:                               # if ตัวนอก: เป็น True จึงเข้าไปข้างใน
    print("Welcome!")                          # อยู่ใน if ตัวนอก (4 ช่องว่าง)
    if is_admin:                               # if ตัวใน: is_admin เป็น False
        print("You can open the admin page.")  # อยู่ใน if ตัวใน (8 ช่องว่าง) - ถูกข้ามไป
    else:
        print("You are a regular user.")       # อยู่ใน else ตัวใน (8 ช่องว่าง) - บรรทัดนี้ทำงาน
else:                                          # เป็นของ if ตัวนอก
    print("Please log in first.")              # ถูกข้ามไป เพราะ if ตัวนอกเป็น True
````

````python
# การซื้อของในร้านค้า
# ขั้นที่ 1: มีสินค้าไหม?  ขั้นที่ 2: มีเงินพอไหม?
in_stock = True
money = 8000
price = 12000

if in_stock:                                   # มีสินค้า -> เข้าไปข้างใน
    if money >= price:                         # 8000 >= 12000 เป็น False
        print("Purchase complete!")
    else:
        print("You don't have enough money.")  # บรรทัดนี้ทำงาน
        print("You need", price - money, "more won.")  # 12000 - 8000 = 4000
else:
    print("Sorry, this item is sold out.")
````

**เคล็ดลับ:** if ซ้อนกันมีประโยชน์ แต่ถ้าซ้อนกันหลายชั้นเกินไป โค้ดจะอ่านยาก ถ้าทำได้ ลองรวมเงื่อนไขด้วย `and` แทน

```python
# แบบซ้อนกัน
if is_logged_in:
    if is_admin:
        print("You can open the admin page.")

# ความหมายเดียวกัน เขียนด้วย and
if is_logged_in and is_admin:
    print("You can open the admin page.")
```

## 7. ข้อผิดพลาดที่พบบ่อย

นี่คือข้อผิดพลาดที่เกือบทุกคนเคยทำตอนเรียน `if` ถ้าเจอ error ให้ตรวจสอบข้อเหล่านี้ก่อน

1. **ใช้ `=` แทน `==`**: `=` ใช้เก็บค่า ส่วน `==` ใช้เปรียบเทียบค่า `if age = 20:` จะเกิด error
2. **ลืมโคลอน `:`**: บรรทัด `if`, `elif` และ `else` ต้องลงท้ายด้วย `:` ทั้งหมด
3. **ย่อหน้าผิด**: โค้ดข้างในต้องย่อหน้า และบรรทัดในบล็อกเดียวกันต้องย่อหน้าเท่ากัน
4. **ใส่เงื่อนไขให้ `else`**: `else` ไม่มีเงื่อนไขเด็ดขาด `else age < 20:` จะเกิด error ถ้าต้องการเงื่อนไข ให้ใช้ `elif`
5. **เปรียบเทียบตัวเลขกับสตริง**: `"20"` (สตริง) กับ `20` (จำนวนเต็ม) ไม่เหมือนกัน `"20" == 20` เป็น `False`

เซลล์ข้างล่างแสดงให้เห็นว่าข้อผิดพลาดแต่ละแบบเป็นอย่างไร โค้ดที่ผิดถูกเขียนไว้เป็นคอมเมนต์ เพื่อไม่ให้ notebook พัง ลองลบ `#` ออกจากบรรทัดโค้ดเพื่อดูข้อความ error จริง แล้วค่อยใส่ `#` กลับเหมือนเดิม

````python
# ข้อผิดพลาด 1: ใช้ = แทน ==
age = 20

# if age = 20:          # SyntaxError! = หมายถึง "เก็บค่า" จึงใช้เป็นเงื่อนไขไม่ได้
#     print("20 years old")

if age == 20:           # ถูกต้อง: == หมายถึง "เท่ากันไหม?"
    print("20 years old")
````

````python
# ข้อผิดพลาด 2: ลืมโคลอน
age = 20

# if age >= 20          # SyntaxError! ท้ายบรรทัดไม่มีโคลอน
#     print("adult")

if age >= 20:           # ถูกต้อง: บรรทัดลงท้ายด้วยโคลอน
    print("adult")
````

````python
# ข้อผิดพลาด 3: ย่อหน้าผิด
age = 20

# if age >= 20:
# print("adult")        # IndentationError! บรรทัดข้างใน if ไม่ได้ย่อหน้า

if age >= 20:
    print("adult")      # ถูกต้อง: ย่อหน้า 4 ช่องว่าง
````

````python
# ข้อผิดพลาด 4: ใส่เงื่อนไขให้ else
age = 15

# if age >= 20:
#     print("adult")
# else age < 20:        # SyntaxError! else มีเงื่อนไขไม่ได้
#     print("minor")

if age >= 20:
    print("adult")
else:                   # ถูกต้อง: else ไม่มีเงื่อนไข มันรับทุกกรณีที่ if ไม่ได้รับไว้
    print("minor")
````

````python
# ข้อผิดพลาด 5: เปรียบเทียบตัวเลขกับสตริง
# ไม่มีข้อความ error แสดงขึ้นมา แต่ผลลัพธ์ไม่เป็นอย่างที่คาดไว้ ข้อผิดพลาดนี้จึงหาเจอยาก
age_text = "20"          # นี่คือสตริง เพราะอยู่ในเครื่องหมายคำพูด (นึกถึง zip_code ในบทที่ 1)
age_number = 20          # นี่คือจำนวนเต็ม

print(age_text == age_number)     # False! สตริงกับจำนวนเต็มไม่มีทางเท่ากัน

if age_text == age_number:
    print("Same")
else:
    print("Different")            # บรรทัดนี้ทำงาน ทั้งที่ดูเหมือนจะเหมือนกัน
````
