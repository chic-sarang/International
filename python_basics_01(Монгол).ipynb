## 1. Хувьсагч ба өгөгдлийн төрөл

Хувьсагч гэдэг нь утга хадгалдаг хайрцаг юм. Хайрцагт нэр өгвөл, дараа нь тэр нэрээр нь дуудаж дотор нь байгаа утгыг ашиглаж болно. Шинэ утга хийвэл хайрцгийн доторх агуулга ч мөн өөрчлөгдөнө.

Гэхдээ хайрцагт хийж болох утгад төрөл ялгаа байдаг. Энэ төрлийг өгөгдлийн төрөл гэж нэрлэдэг. Python хэлний гол өгөгдлийн төрлүүд дараах байдалтай.

- `str` : мөр (үсэг)
- `int` : бүхэл тоо
- `float` : бутархай тоо (аравтын бутархайтай тоо)
- `bool` : булийн (үнэн/худал)

Мөн `print()` гэдэг нь хаалт доторх утгыг дэлгэц дээр харуулах команд юм.

```python
# = тэмдэг нь 'тэнцүү' гэсэн утгагүй, харин 'баруун талын утгыг зүүн талын нэртэй хайрцагт хадгалах' гэсэн утгатай
# Хувьсагчийн нэрс (name, age, height, is_student) нь хайрцагт өгсөн нэр юм

# Мөр (str) төрлийг хашилтаар хүрээлэхэд л хангалттай. Давхар хашилт (" ") болон ганц хашилт (' ') хоёуланг нь ашиглаж болно
name = "Bill"                    # Давхар хашилтаар хүрээлсэн үг
city = 'Seoul'                    # Ганц хашилтаар хүрээлсэн үг
greeting = "Hello, nice to meet you"  # Өгүүлбэр ч мөн мөр төрөл байдаг
zip_code = "06236"               # Тоо шиг харагдаж байсан ч хашилтын дотор байвал энэ нь мөр төрөл юм

# Бүхэл тоо (int) нь аравтын бутархайгүй тоо юм. Эерэг, сөрөг тоо, 0 бүгд бүхэл тоо
age = 20              # Эерэг бүхэл тоо
temperature = -5      # Сөрөг бүхэл тоо ч бас байж болно
score = 0             # 0 ч мөн бүхэл тоо

# Бутархай тоо (float) нь аравтын бутархайтай тоо
height = 175.5        # Эерэг бутархай тоо
account_balance = -3.75  # Сөрөг бутархай тоо ч бас байж болно
pi_value = 3.14159    # Аравтын бутархайн ард олон орон байсан ч асуудалгүй

# Булийн (bool) төрөл нь зөвхөн True (үнэн) эсвэл False (худал) гэсэн хоёр л утгатай байдаг (эхний үсэг үргэлж том үсэг байна)
is_student = True     # оюутан байна
is_married = False    # гэрлээгүй
has_passed = True     # шалгалтанд тэнцсэн

# print() нь утгыг дэлгэц дээр харуулах команд юм
print(name)
print(city)
print(greeting)
print(zip_code)
print(age)
print(temperature)
print(score)
print(height)
print(account_balance)
print(pi_value)
print(is_student)
print(is_married)
print(has_passed)
```

```python
# Хувьсагчийн утгыг дахин өөрчилж болно (бүхэл тооны жишээ)
age = 20
print(age)

age = 21   # Ижил хувьсагчид шинэ утга хадгалбал өмнөх утга алга болж, шинэ утгаар солигдоно
print(age)

age = age + 1   # Хуучин утгыг ашиглан шинэ утга үүсгэж ч болно (21 + 1 = 22)
print(age)
```

```python
# Хувьсагчийн утгыг дахин өөрчилж болно (булийн жишээ)
is_raining = False
print(is_raining)

is_raining = True   # Ижил хувьсагчид шинэ утга хадгалбал өмнөх утга алга болж, шинэ утгаар солигдоно
print(is_raining)

is_raining = not is_raining   # not ашиглавал утга нь эсрэгээрээ болно: True нь False, False нь True болно
print(is_raining)
```

```python
# Хувьсагчийн утгыг дахин өөрчилж болно (мөр төрлийн жишээ)
city = "Seoul"
print(city)

city = "Busan"   # Ижил хувьсагчид шинэ утга хадгалбал өмнөх утга алга болж, шинэ утгаар солигдоно
print(city)

city = city + " Travel"   # Мөрүүдийг + тэмдгээр холбож болно ("Busan" + " Travel" = "Busan Travel")
print(city)
```

```python
# Хувьсагчийн утгыг дахин өөрчилж болно (бутархай тооны жишээ)
height = 175.5
print(height)

height = 178.2   # Ижил хувьсагчид шинэ утга хадгалбал өмнөх утга алга болж, шинэ утгаар солигдоно
print(height)

height = height + 0.5   # Хуучин утгыг ашиглан шинэ утга үүсгэж ч болно (178.2 + 0.5 = 178.7)
print(height)
```

## 2. Үндсэн үйлдлүүд

Тоо тооцоолох, эсвэл утгыг харьцуулахад ашигладаг тэмдэгтүүд.

```python
# Дөрвөн үндсэн тооцоолол
a = 10
b = 3

print(a + b)   # Нэмэх
print(a - b)   # Хасах
print(a * b)   # Үржих
print(a / b)   # Хуваах (үр дүн нь бутархай тоо байна)
print(a // b)  # Хуваах (зөвхөн ноогдворыг үлдээнэ)
print(a % b)   # Хуваалтын үлдэгдэл
```

```python
# Харьцуулах үйлдэл -> үр дүн үргэлж True эсвэл False байна
print(a > b)   # a нь b-ээс их
print(a < b)   # a нь b-ээс бага
print(a == b)  # a нь b-тэй тэнцүү
print(a != b)  # a нь b-ээс ялгаатай
```
