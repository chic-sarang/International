## 1. Why Use a for Loop with a List?

In Lesson 4, we learned that a list can hold many values, and that we can get one value using its **index** (its position number). The first value is at index 0, the second at index 1, and so on.

So if you want to print every score in a list, you could write one line for each index:

```python
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
```

This works. But there are two problems:

- If the list has 100 values, you need to write 100 lines.
- If you add a new value with `append()`, you also have to add another `print()` line yourself.

A **for loop** solves this. A for loop takes the values out of the list **one by one, from the first to the last**, and runs the same code for each value. It doesn't matter how long the list is. You only write the code once.

````python
# The long way: one line for each value
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
````

````python
# The short way: a for loop
# The result is exactly the same, but it's only 2 lines
scores = [85, 92, 78, 90, 66]
for score in scores:
    print(score)
````

````python
# Even if the list gets longer, the for loop stays the same
# We don't need to add any print() lines
scores = [85, 92, 78, 90, 66]
scores.append(100)      # Add one more value at the end (now there are 6 values)
scores.append(73)       # And one more (now there are 7 values)

for score in scores:    # The same 2 lines as above
    print(score)
````

## 2. The Basic Form: `for value in list`

The shape of the code looks like this:

```python
for score in scores:
    code to repeat
```

You can read it like a sentence: **"for each score in scores, run the code below"**.

Let's look at each piece:

| Piece | Meaning |
|---|---|
| `for` | "repeat the following for each value" |
| `score` | a variable. Each time, one value from the list is stored here |
| `in` | "inside" |
| `scores` | the list we go through |
| `:` | a colon. It must be at the end of the `for` line |
| `    code to repeat` | the code that runs for each value. It must be indented (4 spaces) |

**How it works, step by step** (with `scores = [85, 92, 78]`):

| Repetition | What is stored in `score` | What happens |
|---|---|---|
| 1st | `85` (the value at index 0) | The indented code runs with `score` = 85 |
| 2nd | `92` (the value at index 1) | The indented code runs with `score` = 92 |
| 3rd | `78` (the value at index 2) | The indented code runs with `score` = 78 |
| - | no more values | The loop ends, and Python moves on to the code after the loop |

So the loop runs **as many times as there are values in the list**. A list with 3 values → 3 repetitions. A list with 100 values → 100 repetitions.

````python
# Print each fruit one by one
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:        # 1st time: fruit = "apple", 2nd time: fruit = "banana", 3rd time: fruit = "cherry"
    print(fruit)            # Print the value that is in fruit right now
````

````python
# A list of numbers
numbers = [10, 20, 30, 40]

for number in numbers:      # number becomes 10, then 20, then 30, then 40
    print(number)
````

````python
# A list of booleans (True / False)
answers = [True, False, True]

for answer in answers:      # answer becomes True, then False, then True
    print(answer)
````

````python
# We can calculate with each value
# Each price is multiplied by 2 and printed
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)        # 1000 * 2, then 2500 * 2, then 3000 * 2
````

````python
# The values inside the list do NOT change
# In the cell above we printed price * 2, but the list itself is still the same
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)

print(prices)               # Still [1000, 2500, 3000]
````

````python
# Joining strings with +
# Remember: + can join two strings into one ("Hello, " + "Bill" -> "Hello, Bill")
names = ["Bill", "Anna", "Tom"]

for name in names:
    print("Hello, " + name)
````

````python
# If the list is empty, the loop runs 0 times
# Nothing is printed, and there is no error
empty_list = []

for value in empty_list:
    print("This never prints.")

print("The loop ran 0 times.")
````

## 3. Indentation: Which Lines Are Repeated?

**Indentation** means pushing a line to the right by adding spaces at the beginning. In Python, we use **4 spaces**.

```python
for fruit in fruits:
    print(fruit)          # <- this line is indented (4 spaces at the beginning)
```

### Why do we need indentation?

The `for` loop repeats some code. But how does Python know **which lines** to repeat, and where the repeated part ends?

Python looks at the indentation:

- Lines that are **indented** right after the `for` line → **inside** the loop → repeated for every value
- The first line that is **not indented** again → **outside** the loop → runs only once, after the loop is finished

The indented lines belong together, like a group. This group of lines is called a **block**. The block of a `for` loop is "the code to repeat".

(This is the same rule we used for `if` in Lesson 2: the indented lines under `if` run only when the condition is True.)

### Same code, different indentation, different result

The cells below have exactly the same lines. The **only** difference is whether the last `print()` is indented or not. Look how the result changes.

````python
# Version A: the second print() is indented -> it is INSIDE the loop
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
    print("Done!")          # Indented -> inside the loop -> printed 3 times
````

````python
# Version B: the second print() is NOT indented -> it is OUTSIDE the loop
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
print("Done!")              # Not indented -> outside the loop -> printed only once, at the end
````

### A block can have many lines

Every indented line runs, in order from top to bottom, for each value. Then the next value comes, and the whole block runs again from the top.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Now the value is:")      # 1st line of the block
    print(fruit)                    # 2nd line of the block
    print("----------")             # 3rd line of the block

print("The loop is finished.")      # Not indented -> outside the loop -> runs once
````

### All lines in a block must have the same indentation

If one line has 4 spaces and the next line has 2 spaces, Python gets confused and shows an error. Always use exactly 4 spaces. (In Jupyter, pressing the **Tab** key adds 4 spaces for you.)

```python
for fruit in fruits:
    print(fruit)       # 4 spaces
  print("Done!")       # 2 spaces -> IndentationError!
```

## 4. The Name of the Loop Variable

In `for score in scores:`, the word `score` is a **new variable** that the for loop creates. Each repetition, one value from the list is stored in it.

**You can give this variable any name you want.** Python doesn't care about the name. It just stores each value in a box with that name.

The three cells below use different names (`x`, `score`, `banana`), but the result is exactly the same.

````python
scores = [85, 92, 78]

for x in scores:            # The variable is named x
    print(x)
````

````python
scores = [85, 92, 78]

for score in scores:        # The variable is named score
    print(score)
````

````python
scores = [85, 92, 78]

for banana in scores:       # The variable is named banana (strange, but it works!)
    print(banana)
````

All three work. But which one is easiest to understand?

- `x` doesn't tell you what the value is.
- `banana` is confusing, because these are scores, not bananas.
- `score` tells you exactly what it is: **one score** from the list of scores.

**A good habit:** name the list with a plural word, and the loop variable with the **singular** form of the same word.

| List (many values) | Loop variable (one value) |
|---|---|
| `scores` | `score` |
| `fruits` | `fruit` |
| `names` | `name` |
| `prices` | `price` |

Then when you read the code, you can easily tell "the whole list" (`scores`) from "one value" (`score`).

**One rule:** the variable name must be **different** from the list name. We'll see what happens if they are the same in section 12 (Common Mistakes).

## 5. Printing Several Values at Once with Commas

So far, we have mostly printed one value at a time: `print(score)`.

But sometimes we want to print a label together with a value, like `Total: 255`. For this, we can put **several values** inside `print()`, separated by **commas** `,`.

```python
print(value1, value2, value3)
```

Rules:
1. The values are printed **on one line, in order**.
2. Python **automatically puts one space** between the values. You don't need to add spaces yourself.
3. Text inside quotes `" "` is printed **as it is**.
4. Things outside quotes are **calculated first**, and then the result is printed. If it's a variable, the value inside the variable is printed.

````python
# Several strings
print("apple", "banana", "cherry")     # One space is put between each value automatically
````

````python
# Several numbers
print(10, 20, 30)
````

````python
# A string and a number together
print("Total:", 255)                   # "Total:" is printed as it is, then a space, then 255
````

````python
# A string and a variable together
score = 85
print("Score:", score)                 # score is a variable, so the value inside it (85) is printed
````

````python
# A string and a calculation together
price = 1000
print("Double:", price * 2)            # price * 2 is calculated first (2000), then printed
````

````python
# Inside quotes vs outside quotes
price = 1000
print("price * 2")                     # Inside quotes -> printed as it is: price * 2
print(price * 2)                       # Outside quotes -> calculated: 2000
````

### Commas `,` vs plus `+`

We also learned that `+` can join strings. What's the difference?

| | Comma `,` | Plus `+` |
|---|---|---|
| Space between values | Added automatically | Not added. You must add it yourself |
| String + number | ✅ Works | ❌ Error |

`+` can only join **a string with a string**. If you try to join a string with a number using `+`, Python shows an error. With commas, mixing strings and numbers is no problem.

````python
total = 255

print("Total:", total)                 # ✅ Comma: string and number together -> works

# print("Total: " + total)             # ❌ Plus: string + number -> TypeError!
````

## 6. Adding Up the Values

One of the most common jobs for a for loop is **adding up** all the values in a list. For example: what is the total of all the scores?

The trick is to use a variable as a **running total**:

1. **Before the loop:** make a variable and set it to 0. → `total = 0`
2. **Inside the loop:** add the current value to it. → `total = total + score`
3. **After the loop:** the variable holds the sum of all values.

Look at step 2 carefully: `total = total + score`

- `=` doesn't mean "equal". It means **"store the value on the right into the box on the left"**.
- So Python first calculates the right side, `total + score` (the old total plus the current score),
- and then stores that result back into `total`.
- It's the same idea as `age = age + 1`: take the old value, add something, and put it back in the same box.

````python
# Add up all the scores
scores = [85, 92, 78, 90, 66]
total = 0                       # Step 1: the running total starts at 0 (BEFORE the loop)

for score in scores:
    total = total + score       # Step 2: old total + current score -> store it back in total

print("Total:", total)          # Step 3: AFTER the loop (not indented), print the result once
````

````python
# Let's watch how total changes in each repetition
# Same code as above, but with a print() INSIDE the loop (indented)
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total = total + score
    print("score:", score, "total:", total)      # Indented -> printed every repetition

# 1st: 0 + 85 = 85
# 2nd: 85 + 92 = 177
# 3rd: 177 + 78 = 255
# 4th: 255 + 90 = 345
# 5th: 345 + 66 = 411
````

### A shorter way to write it: `+=`

`total = total + score` is used so often that Python has a shorter way to write it:

```python
total += score
```

`+=` means **"add the right side to the variable on the left, and store the result back in it"**.

`total += score` and `total = total + score` mean **exactly the same thing**. You can use either one. Many Python programmers use the short form, so you'll see it often.

````python
# The long form
scores = [85, 92, 78]
total = 0
for score in scores:
    total = total + score
print("Long form:", total)

# The short form
total = 0
for score in scores:
    total += score              # same as: total = total + score
print("Short form:", total)     # The same result
````

````python
# The total price of a shopping cart
cart = [3500, 1200, 8000, 2300]
total_price = 0                 # Before the loop: start from 0

for price in cart:
    total_price += price        # same as: total_price = total_price + price

print("Total price:", total_price)
````

### The average

The **average** is the total divided by the number of values.

How do we get the number of values? Remember `len()`: it tells you how many values are in a list. `len([85, 92, 78])` is `3`.

So: **average = total / len(list)**

Important: calculate the average **after** the loop, when the total is complete.

````python
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total += score              # same as: total = total + score

average = total / len(scores)   # After the loop: total (411) / number of scores (5)
print("Total:", total)
print("Number of scores:", len(scores))
print("Average:", average)
````

## 7. Using if Inside the Loop

You can put an `if` **inside** a for loop. Then each value is checked one by one, and you can do something **only for the values you want**.

Remember how `if` works:
- `if condition:` → the indented code under it runs **only when the condition is True**.
- `else:` → the indented code under it runs when the condition is False.

When you put an `if` inside a `for`, there are **two levels** of indentation:

```python
for score in scores:          # no spaces
    if score >= 60:           # 4 spaces  -> inside the for (runs for every score)
        print(score)          # 8 spaces  -> inside the if (runs only when the condition is True)
```

- The `if` line has 4 spaces, because it belongs to the `for` block.
- The `print()` line has 8 spaces, because it belongs to the `if` block, which is inside the `for` block.

````python
# Print only the scores that passed (60 or higher)
scores = [85, 42, 78, 55, 90]

for score in scores:            # Take out each score
    if score >= 60:             # 4 spaces: check this score
        print(score)            # 8 spaces: print it only if it is 60 or higher
````

````python
# Let's see which scores were checked and which were printed
scores = [85, 42, 78, 55, 90]

for score in scores:
    print("Checking", score)    # 4 spaces: runs for EVERY score
    if score >= 60:
        print("pass")           # 8 spaces: runs only when score >= 60
````

````python
# if - else: print pass or fail for every score
scores = [85, 42, 78, 55, 90]

for score in scores:
    if score >= 60:
        print(score, "pass")    # Runs when the condition is True
    else:
        print(score, "fail")    # Runs when the condition is False
````

````python
# Print only the even numbers
# A number is even if the remainder after dividing by 2 is 0
# % gives the remainder: 8 % 2 is 0 (even), 5 % 2 is 1 (odd)
numbers = [3, 8, 5, 12, 7, 6]

for number in numbers:
    if number % 2 == 0:         # Is the remainder 0? -> even
        print(number)
````

````python
# if - elif - else inside a loop: give each score a grade
# Python checks the conditions from top to bottom, and runs only the FIRST one that is True
scores = [95, 82, 74, 58]

for score in scores:
    if score >= 90:
        print(score, "A")
    elif score >= 80:
        print(score, "B")
    elif score >= 70:
        print(score, "C")
    else:
        print(score, "F")
````

````python
# Adding up only some of the values: if + running total
# Add only the even numbers
numbers = [3, 8, 5, 12, 7, 6]
even_total = 0                  # Before the loop: start from 0

for number in numbers:
    if number % 2 == 0:         # Only when the number is even...
        even_total += number    # ...add it (same as: even_total = even_total + number)

print("Sum of even numbers:", even_total)    # 8 + 12 + 6 = 26
````

## 8. Counting

How many students passed? How many people voted for pizza?

To **count**, we use the same idea as the running total. The only difference: instead of adding the value, we add **1** each time.

1. **Before the loop:** `count = 0`
2. **Inside the loop:** when the condition is True, `count = count + 1` (or `count += 1`)
3. **After the loop:** `count` holds how many times the condition was True

````python
# How many students passed (60 or higher)?
scores = [85, 42, 78, 55, 90]
pass_count = 0                  # Before the loop: start from 0

for score in scores:
    if score >= 60:
        pass_count += 1         # same as: pass_count = pass_count + 1

print("Passed:", pass_count)    # 85, 78, 90 -> 3
````

````python
# How many votes did pizza get?
votes = ["pizza", "pasta", "pizza", "salad", "pizza", "pasta"]
pizza_count = 0

for vote in votes:
    if vote == "pizza":         # == checks whether two values are the same
        pizza_count += 1        # same as: pizza_count = pizza_count + 1

print("Pizza:", pizza_count)
````

````python
# Counting two things at once
scores = [85, 42, 78, 55, 90]
pass_count = 0
fail_count = 0

for score in scores:
    if score >= 60:
        pass_count += 1
    else:
        fail_count += 1

print("Passed:", pass_count)
print("Failed:", fail_count)
````

````python
# Counting below-zero temperatures
temperatures = [3, -2, 5, -7, 0, -1]
below_zero = 0

for temperature in temperatures:
    if temperature < 0:
        below_zero += 1

print("Days below zero:", below_zero)
````

## 9. Finding the Biggest and the Smallest Value

How do we find the highest score in a list? Think about how a person would do it with a list of numbers on paper:

1. Look at the first number, and remember it as **"the biggest so far"**.
2. Look at the next number. Is it bigger than the one you remember? If yes, remember this one instead.
3. Keep doing this until the end. The number you remember at the end is the biggest.

We can write exactly these steps in Python. The variable `highest` is "the number we remember".

**Why start with the first value (`scores[0]`)?** Because we need something to compare with. Index 0 is the first value of the list.

````python
# Find the highest score
scores = [85, 92, 78, 95, 66]
highest = scores[0]             # Step 1: remember the first score (85) as "the biggest so far"

for score in scores:
    if score > highest:         # Step 2: is this score bigger than the one we remember?
        highest = score         #         if yes, remember this one instead

print("Highest:", highest)      # Step 3: the one we remember at the end
````

````python
# Let's watch it step by step
scores = [85, 92, 78, 95, 66]
highest = scores[0]

for score in scores:
    if score > highest:
        highest = score
    print("score:", score, "highest:", highest)     # 4 spaces: runs every repetition
````

````python
# The smallest value works the same way, just with < instead of >
temperatures = [12, 8, 15, 3, 10]
lowest = temperatures[0]        # Start with the first temperature (12)

for temperature in temperatures:
    if temperature < lowest:    # Is this one smaller than the one we remember?
        lowest = temperature

print("Lowest:", lowest)
````

````python
# The most expensive item in the cart
cart = [3500, 1200, 8000, 2300]
most_expensive = cart[0]

for price in cart:
    if price > most_expensive:
        most_expensive = price

print("Most expensive:", most_expensive)
````

## 10. Making a New List with append()

Remember `append()`: `list.append(value)` adds the value to the **end** of the list.

If we use `append()` inside a for loop, we can make a **new list** from an old list:

1. **Before the loop:** make an empty list. → `new_list = []`
2. **Inside the loop:** `append()` the values you want to the new list.
3. **After the loop:** the new list is complete.

The original list is not changed.

````python
# A new list with only the passing scores
scores = [85, 42, 78, 55, 90]
passed = []                     # Before the loop: an empty list

for score in scores:
    if score >= 60:
        passed.append(score)    # Add this score to the end of passed

print("All scores:", scores)    # The original list is not changed
print("Passed:", passed)
````

````python
# Let's watch the new list grow
scores = [85, 42, 78, 55, 90]
passed = []

for score in scores:
    if score >= 60:
        passed.append(score)
    print(passed)               # 4 spaces: print the new list every repetition
````

````python
# A new list with every price doubled
prices = [1000, 2500, 3000]
doubled = []

for price in prices:
    doubled.append(price * 2)   # price * 2 is calculated first, and then added

print(doubled)
````

````python
# A new list of greetings
names = ["Bill", "Anna", "Tom"]
greetings = []

for name in names:
    greetings.append("Hi, " + name)     # "Hi, " + "Bill" -> "Hi, Bill"

print(greetings)
````

````python
# A new list with only the long words
# len() also works on strings: it counts the characters. len("cat") is 3
words = ["cat", "elephant", "dog", "giraffe"]
long_words = []

for word in words:
    if len(word) > 3:           # More than 3 characters?
        long_words.append(word)

print(long_words)
````

## 11. Using in and break Inside the Loop

### in

`value in list` checks whether the value is inside the list. The result is `True` or `False`, so it can be used as the condition of an `if`.

```python
"milk" in ["bread", "milk"]     # True
"eggs" in ["bread", "milk"]     # False
```

**Be careful:** the `in` in `for fruit in fruits:` and the `in` in `if fruit in store:` look the same, but they do different jobs.

| | Meaning |
|---|---|
| `for fruit in fruits:` | **take out** each value from fruits, one by one |
| `if fruit in store:` | **check** whether fruit is inside store (True / False) |

````python
# Which items on my shopping list does the store have?
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]

for item in shopping_list:              # "in" #1: take out each item
    if item in store:                   # "in" #2: is this item inside store?
        print(item, "yes")
    else:
        print(item, "no")
````

````python
# Make a list of the items we still need to buy somewhere else
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]
not_found = []

for item in shopping_list:
    if item not in store:               # not in: True when the item is NOT inside store
        not_found.append(item)

print("Not found:", not_found)
````

### break

`break` **stops the loop immediately**. The rest of the values are not checked at all, and Python jumps to the code after the loop.

It's useful when you've found what you were looking for, and there's no need to look further. `break` is almost always used inside an `if`.

````python
# Stop as soon as we find the first failing score
scores = [85, 92, 55, 78, 40]

for score in scores:
    if score < 60:
        print("Found a failing score:", score)
        break                           # Stop the loop right here
    print(score, "is OK")

print("The loop is over.")              # 78 and 40 were never checked
````

````python
# Is there anyone named Tom in the list?
names = ["Bill", "Anna", "Tom", "Kate", "Mike"]

for name in names:
    print("Checking", name)
    if name == "Tom":
        print("Found Tom!")
        break                           # No need to check Kate and Mike
````

## 12. Common Mistakes

These are the mistakes almost everyone makes when using a for loop with a list. For each one, we will look at:

- ❌ **The wrong code**
- 🔍 **What went wrong**
- ✅ **The correct way**

Some wrong code causes an error. That code is written as a comment (with `#` at the beginning) so it won't stop the notebook. If you want to see the real error message, remove the `#`, run the cell, and then put the `#` back.

Other wrong code gives **no error, but the wrong result**. These mistakes are even more dangerous, because Python doesn't warn you.

### Mistake 1: Forgetting the colon

````python
# ❌ Wrong
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits           # SyntaxError
#     print(fruit)
````

🔍 **What went wrong?**

- The `for` line must always end with a colon `:`.
- The colon tells Python: "the code to repeat starts on the next line".
- Without it, Python doesn't understand the code and shows a `SyntaxError` ("the grammar of the code is wrong").

✅ **The correct way:** Put `:` at the end of the `for` line.

````python
# ✅ Correct
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # Colon at the end
    print(fruit)
````

### Mistake 2: Forgetting to indent

````python
# ❌ Wrong
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits:
# print(fruit)                  # IndentationError
````

🔍 **What went wrong?**

- After a `for` line, there must be at least one **indented** line: the code to repeat.
- Here, `print(fruit)` has no spaces at the beginning, so Python thinks the `for` loop has nothing to repeat.
- The error is `IndentationError` ("the indentation is wrong").

✅ **The correct way:** Indent the code to repeat by 4 spaces (or press Tab once).

````python
# ✅ Correct
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # 4 spaces at the beginning
````

### Mistake 3: Indenting a line that should run only once

````python
# ❌ Wrong: we want to print the total only once, at the end
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score
    print("Total:", total)      # Indented -> inside the loop -> printed 3 times!
````

🔍 **What went wrong?**

- `print("Total:", total)` is indented, so it is **inside** the loop.
- That means it runs every repetition, and prints the total while it is still being added up.
- No error appears, but we get 3 lines instead of 1, and the first two are not the real total.

✅ **The correct way:** Remove the indentation from the line that should run only once. Then it is **outside** the loop, and runs only after the loop is finished.

````python
# ✅ Correct
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score              # Inside the loop: add up
print("Total:", total)          # Not indented -> outside the loop -> printed once
````

### Mistake 4: Using the same name for the list and the loop variable

````python
# ❌ Wrong
fruits = ["apple", "banana", "cherry"]

for fruits in fruits:           # The loop variable has the SAME name as the list
    print(fruits)               # This prints correctly...

print(fruits)                   # ...but now fruits is just "cherry". The list is gone!
````

🔍 **What went wrong?**

- The loop seems to work, so it's hard to notice.
- But each repetition stores a value into the box named `fruits`. Remember: when you store a new value in a variable, the old value disappears.
- So the list that was in `fruits` is replaced by `"apple"`, then `"banana"`, then `"cherry"`.
- After the loop, `fruits` holds only `"cherry"`. The list is gone.

✅ **The correct way:** Give the loop variable a **different** name. The singular form of the list name is the best choice: `fruits` → `fruit`.

````python
# ✅ Correct
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # fruit (one value) and fruits (the list) are different names
    print(fruit)

print(fruits)                   # The list is still safe
````

### Mistake 5: Using the list name instead of the loop variable

````python
# ❌ Wrong: we want to print each fruit
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruits)               # fruits (with s) -> the whole list is printed 3 times
````

🔍 **What went wrong?**

- Inside the loop, we wrote `fruits` (the whole list) instead of `fruit` (one value).
- The names are almost the same, only one letter `s` is different. That's why this mistake is easy to make.
- The loop runs 3 times, and each time it prints the whole list.

✅ **The correct way:** Inside the loop, use the **loop variable** (`fruit`) to get the current value.

````python
# ✅ Correct
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # fruit (without s) -> one value at a time
````

### Mistake 6: Putting the list name in quotes

````python
# ❌ Wrong
fruits = ["apple", "banana", "cherry"]

for fruit in "fruits":          # "fruits" with quotes!
    print(fruit)                # f, r, u, i, t, s ... one letter at a time
````

🔍 **What went wrong?**

- `fruits` (without quotes) is the **variable** that holds our list.
- `"fruits"` (with quotes) is just a **string**: the 6 letters f, r, u, i, t, s.
- A for loop can go through a string too. It takes out **one letter at a time**. So we get 6 letters instead of 3 fruits.
- No error appears, so it can be confusing.

✅ **The correct way:** Write the variable name without quotes.

````python
# ✅ Correct
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # No quotes -> the variable (the list)
    print(fruit)
````

### Mistake 7: Setting the running total to 0 inside the loop

````python
# ❌ Wrong
scores = [85, 92, 78]

for score in scores:
    total = 0                   # Indented -> inside the loop -> runs every repetition!
    total += score

print("Total:", total)          # 78, not 255
````

🔍 **What went wrong?**

- `total = 0` is indented, so it is inside the loop. It runs **3 times**.
- Every repetition, the total goes back to 0, and then only the current score is added.
- 1st: 0 + 85 = 85 → 2nd: **back to 0**, 0 + 92 = 92 → 3rd: **back to 0**, 0 + 78 = 78
- So at the end, `total` holds only the last score.

✅ **The correct way:** Write `total = 0` **once, before** the loop (not indented).

````python
# ✅ Correct
scores = [85, 92, 78]
total = 0                       # Before the loop -> runs only once

for score in scores:
    total += score              # same as: total = total + score

print("Total:", total)          # 85 + 92 + 78 = 255
````

### Mistake 8: Calculating the average inside the loop

````python
# ❌ Wrong
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score
    average = total / len(scores)       # Indented -> calculated every repetition
    print("Average:", average)
````

🔍 **What went wrong?**

- To calculate the average, we need the total of **all** values.
- But inside the loop, the total isn't finished yet.
- 1st: 80 / 3 → 2nd: 170 / 3 → 3rd: 270 / 3. Only the last one is the real average.

✅ **The correct way:** Inside the loop, only add up. Calculate the average **after** the loop, when the total is complete.

````python
# ✅ Correct
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score                      # Inside the loop: only add up

average = total / len(scores)           # After the loop: 270 / 3
print("Average:", average)
````

### Mistake 9: Trying to change the list by changing the loop variable

````python
# ❌ Wrong: we want to add 5 bonus points to every score in the list
scores = [85, 92, 78]

for score in scores:
    score = score + 5           # This looks like it changes the score...

print(scores)                   # ...but the list is still [85, 92, 78]!
````

🔍 **What went wrong?**

- `score` is a **separate box**. Each repetition, a value is **copied** from the list into this box.
- `score = score + 5` only changes what's inside the `score` box. The list `scores` is not touched at all.
- It's like copying a number from your notebook onto a sticky note, and then changing the number on the sticky note. The notebook stays the same.
- No error appears, so this mistake is hard to notice.

✅ **The correct way:** Make a **new list** with `append()` (section 10), and put the changed values in it.

````python
# ✅ Correct
scores = [85, 92, 78]
new_scores = []                         # An empty list for the changed values

for score in scores:
    new_scores.append(score + 5)        # Add the changed value to the new list

print(new_scores)                       # [90, 97, 83]
````

### Mistake 10: Removing values from the list during the loop

````python
# ❌ Wrong: we want to remove every failing score (below 60)
scores = [50, 40, 80, 30]

for score in scores:
    if score < 60:
        scores.remove(score)

print(scores)                   # [40, 80] -> 40 is still there!
````

🔍 **What went wrong?**

The for loop goes through the list **by position**: the 1st value, then the 2nd value, then the 3rd value ...
But `remove()` takes a value out, and every value after it **moves forward** by one position.

| Repetition | Position the loop looks at | The list at that moment | What happens |
|---|---|---|---|
| 1st | 1st position | `[50, 40, 80, 30]` | 50 < 60 → remove 50. The list becomes `[40, 80, 30]` |
| 2nd | 2nd position | `[40, 80, 30]` | The 2nd value is now **80**. 40 moved to the 1st position, so the loop **skipped** it! |
| 3rd | 3rd position | `[40, 80, 30]` | 30 < 60 → remove 30. The list becomes `[40, 80]` |

- 40 was never checked, so it stayed in the list.
- No error appears, so it's very hard to notice.

✅ **The correct way:** Don't change the list while you are looping over it. Instead, make a **new list** with only the values you want to keep (section 10).

````python
# ✅ Correct
scores = [50, 40, 80, 30]
kept = []

for score in scores:
    if score >= 60:             # Keep only the passing scores
        kept.append(score)

print(kept)                     # [80]
````

### Mistake 11: Using the value as an index

🔍 **What went wrong?**

- In Lesson 4, we got values with an index: `scores[0]`, `scores[1]` ... So it's easy to think we need `scores[...]` here too.

### Mistake 12: Starting "the biggest so far" at 0

````python
# ❌ Wrong: find the highest temperature in winter
temperatures = [-5, -2, -8, -3]
highest = 0                     # Start from 0...

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # 0 -> but 0 is not even in the list!
````

🔍 **What went wrong?**

- All the temperatures are below 0. So none of them is bigger than 0.
- `temperature > highest` is never True, and `highest` stays 0 until the end.
- It's easy to think "I should start from 0", because we start the running total at 0. But finding the biggest value is a different job.

✅ **The correct way:** Start with the **first value of the list** (`temperatures[0]`). Then the starting value always comes from the list itself.

````python
# ✅ Correct
temperatures = [-5, -2, -8, -3]
highest = temperatures[0]       # Start from the first value (-5)

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # -2
````

### Mistake 13: Joining a string and a number with +

````python
# ❌ Wrong
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

# print("Total: " + total)      # TypeError: can only concatenate str (not "int") to str
````

🔍 **What went wrong?**

- `+` can join **a string with a string**. But `total` is a **number** (int).
- Python doesn't know how to join a string and a number with `+`, so it shows an error.
- The error message means: "you can only join a string (str) to a string, not a number (int)".

✅ **The correct way:** Use a **comma** in `print()` (section 5). With commas, you can mix strings and numbers.

````python
# ✅ Correct
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

print("Total:", total)          # Comma -> works with a string and a number
````
