## 1. Dictionary яагаад хэрэгтэй вэ?

Мэддэг зүйлээсээ эхэлье: **жагсаалт (list)**.

Жагсаалт нь дөрвөлжин хаалт `[ ]` дотор олон утгыг дарааллаар нь хадгалдаг. Утгыг **индекс**-ээр нь (байрлалын дугаараар) авдаг бөгөөд индекс **0**-ээс эхэлдэг.

Одоо Anna, Tom, Bill гэсэн 3 сурагчийн шалгалтын оноог хадгалахыг хүсэж байна гэж бодоорой. Жагсаалтаар бол **хоёр** жагсаалт хэрэгтэй болно: нэг нь нэрэнд, нөгөө нь оноонд.

```python
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]
```

Хоёр жагсаалт байрлалаараа холбоотой:

| Индекс | 0 | 1 | 2 |
|---|---|---|---|
| `names` | "Anna" | "Tom" | "Bill" |
| `scores` | 85 | 92 | 78 |

Тэгэхээр `names[1]` нь Tom, `scores[1]` нь Tom-ын оноо.

**Гэхдээ үүнд асуудал бий:**

- Tom-ын оноог олохын тулд эхлээд Tom 1 индекст байгааг **санаж** байх хэрэгтэй. Python "Tom" гэсэн нэрээр нь олж чадахгүй.
- Нэг жагсаалтад нэр нэмэх эсвэл устгаад, нөгөөд нь ижил зүйлийг хийхээ мартвал нэр ба оноо **холилдоно**.

Зүгээр л **"Tom-ын оноо хэд вэ?"** гэж асуух боломжтой бол илүү хялбар биш гэж үү?

Dictionary яг үүнийг хийдэг.

````python
# Хоёр жагсаалттай бол нэр ба оноо тусдаа газар байна
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

# Tom-ын оноог авахын тулд Tom 1 индекст байгааг мэдэх ёстой
print(names[1])         # Tom
print(scores[1])        # 92 -> Tom-ын оноо
````

````python
# Асуудал: нэг жагсаалт өөрчлөгдөөд нөгөө нь өөрчлөгдөхгүй бол холилдоно
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

names.remove("Anna")    # names-ээс Anna-г устгасан ч scores-оос 85-ыг устгахаа мартсан

print(names[0])         # Tom
print(scores[0])        # 85 -> Энэ бол Tom-ын биш, Anna-гийн оноо! Одоо таарахаа больсон
````

## 2. Dictionary гэж юу вэ?

Жинхэнэ толь бичгийг, үг сурахад ашигладаг номыг бодоорой.

- Та **үг** хайна (жишээ нь, "apple").
- Түүний **утгыг** олно ("бөөрөнхий жимс").

Тэр үг аль хуудсанд байгааг мэдэх шаардлагагүй. Зүгээр л үгийг хайхад утга нь яг хажууд нь байдаг.

Python-ий **dictionary** ч ингэж ажилладаг. Утгуудыг **хос**-оор хадгалдаг:

| Python дахь үг | Утга | Жинхэнэ толь бичигт | Бидний жишээнд |
|---|---|---|---|
| **түлхүүр (key)** | таны хайдаг нэр | үг | сурагчийн нэр ("Tom") |
| **утга (value)** | таны авдаг зүйл | тайлбар | оноо (92) |
| **хос (pair)** | нэг түлхүүр ба түүний утга хамтдаа | нэг үг ба түүний тайлбар | "Tom" ба 92 |

Үүнийг **нэрийн шошготой шүүгээ** гэж ч төсөөлж болно:
- **Жагсаалтад** шүүгээ бүр **дугаартай** (0, 1, 2 ...).
- **Dictionary-д** шүүгээ бүр **нэрийн шошготой** ("Anna", "Tom", "Bill" ...).

Шүүгээг онгойлгохдоо дугаарын оронд нэрийн шошгыг ашиглана.

## 3. Dictionary үүсгэх

Хамгийн жижиг dictionary-ээс эхэлье: ердөө **нэг хос**.

```python
scores = {"Tom": 92}
```

Хэсэг бүрийг харъя:

| Хэсэг | Утга |
|---|---|
| `{` | dictionary эхэлнэ (дөрвөлжин биш, **гоё хаалт**) |
| `"Tom"` | **түлхүүр** (энд тэмдэгт мөр тул хашилтанд байна) |
| `:` | **хоёр цэг**. Түлхүүрийг утгатай нь холбоно |
| `92` | **утга** |
| `}` | dictionary дуусна |

Санаарай: `=` нь **"баруун талын утгыг зүүн талын хайрцагт хадгал"** гэсэн утгатай. Тиймээс энэ мөр бүхэл dictionary-г `scores` нэртэй хувьсагчид хадгална.

````python
# Нэг хостой dictionary
scores = {"Tom": 92}

print(scores)           # {'Tom': 92}
````

### Нэгээс олон хос

Dictionary-д олон хос хийхийн тулд тэдгээрийг **таслал** `,`-аар тусгаарлана. Жагсаалтад утгуудыг таслалаар тусгаарладагтай адил.

```python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
```

- `"Anna": 85` нэг хос
- `"Tom": 92` нэг хос
- `"Bill": 78` нэг хос

Тэгэхээр энэ dictionary **3 хос**-той.

````python
# Хоёр хостой dictionary
scores = {"Anna": 85, "Tom": 92}
print(scores)
````

````python
# Гурван хостой dictionary
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)
````

### Мөр бүрд нэг хос бичих

Хос олон байвал нэг урт мөрийг уншихад хэцүү. **Мөр бүрд нэг хос** бичиж болно. Python үүнийг яг адилхан уншина. Хос бүрийн төгсгөлд таслал тавихаа бүү мартаарай.

````python
# Ижил dictionary, мөр бүрд нэг хос (уншихад хялбар)
scores = {
    "Anna": 85,
    "Tom": 92,
    "Bill": 78,
}
print(scores)
````

### Харьцуулаарай: жагсаалт ба dictionary

| | Жагсаалт | Dictionary |
|---|---|---|
| Хаалт | дөрвөлжин хаалт `[ ]` | гоё хаалт `{ }` |
| Дотор нь юу байх вэ | зөвхөн утгууд | `түлхүүр: утга` **хосууд** |
| Жишээ | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92, "Bill": 78}` |

````python
# Ижил оноонуудыг жагсаалт болон dictionary хэлбэрээр
score_list = [85, 92, 78]
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list)       # Зөвхөн оноо. Хэнийх болохыг мэдэхгүй
print(score_dict)       # Оноо бүрийн хажууд нэр байна
````

````python
# Dictionary бол шинэ өгөгдлийн төрөл: dict ("dictionary"-ийн товчлол)
# type() нь утга ямар төрлийнх болохыг хэлж өгдөг
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(type(scores))     # <class 'dict'>
````

### Өөр төрлийн dictionary-ууд

Утга нь ямар ч өгөгдлийн төрөл байж болно: тэмдэгт мөр, бүхэл тоо, бутархай тоо, bool. Түлхүүр нь ихэвчлэн **тэмдэгт мөр** байдаг.

````python
# Утга нь тэмдэгт мөр: үг -> түүний утга, жинхэнэ толь бичиг шиг
english = {"apple": "a round fruit", "dog": "an animal that barks"}
print(english)
````

````python
# Утга нь тэмдэгт мөр: улс -> нийслэл нь
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}
print(capitals)
````

````python
# Утга нь бутархай тоо: хүн -> өндөр
heights = {"Anna": 162.5, "Tom": 175.0}
print(heights)
````

````python
# Утга нь bool: сурагч -> өнөөдөр хичээлдээ ирсэн үү?
attendance = {"Anna": True, "Tom": False, "Bill": True}
print(attendance)
````

````python
# Хоосон dictionary: нэг ч хосгүй
# Хоосон dictionary-ээс эхлээд дараа нь хос нэмж болно (5-р хэсэг)
empty_dict = {}
print(empty_dict)
````

## 4. Түлхүүрээр утга авах

Dictionary-ээс утга авахын тулд dictionary-ийн нэр болон дөрвөлжин хаалт `[ ]` дотор **түлхүүрийг** бичнэ:

```python
scores["Tom"]      # "scores дахь 'Tom' түлхүүрийн утга"
```

Энэ нь жагсаалттай маш төстэй харагдана. Цорын ганц ялгаа нь дөрвөлжин хаалтан доторх зүйл:

| | Жагсаалт | Dictionary |
|---|---|---|
| `[ ]` дотор юу бичих вэ | **индекс** (байрлалын дугаар) | **түлхүүр** (нэр) |
| Жишээ | `score_list[1]` | `score_dict["Tom"]` |
| Утга | "1-р байрлал дахь утга" | "түлхүүр нь 'Tom' байгаа утга" |

**Чухал:** dictionary-г **үүсгэхдээ** гоё хаалт `{ }` ашигладаг ч түүнээс утга **авахдаа** дөрвөлжин хаалт `[ ]` ашиглана.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # "Tom" түлхүүрийн утга -> 92
print(scores["Anna"])       # "Anna" түлхүүрийн утга -> 85
print(scores["Bill"])       # "Bill" түлхүүрийн утга -> 78
````

````python
# Харьцуулаарай: Tom-ын оноог жагсаалт болон dictionary-ээс авах
score_list = [85, 92, 78]                           # Anna, Tom, Bill
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list[1])        # Жагсаалт: Tom 1 индекст байгааг санаж байх ёстой
print(score_dict["Tom"])    # Dictionary: зүгээр л "Tom" гэсэн нэрийг ашиглана
````

````python
# Өөр жишээнүүд
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

print(capitals["Korea"])    # Seoul
print(capitals["France"])   # Paris
````

````python
english = {"apple": "a round fruit", "dog": "an animal that barks"}

print(english["dog"])       # "dog" гэсэн үгийг хайна -> түүний утга
````

### Утгыг энгийн хувьсагч шиг ашиглаж болно

Dictionary-ээс авсан утга бол зүгээр л энгийн утга. Үүнийг хувьсагчид хадгалах, тооцоолох, харьцуулах боломжтой.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

tom_score = scores["Tom"]                   # Tom-ын оноог шинэ хувьсагчид хадгална
print(tom_score)

print(scores["Anna"] + scores["Tom"])       # 85 + 92 = 177 (тооцоолж болно)
print(scores["Tom"] > scores["Bill"])       # 92 > 78 нь True (харьцуулж болно)
````

````python
# Утгыг if дотор ашиглах (if нь нөхцөл True үед л догол мөртэй кодыг ажиллуулна)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

if scores["Bill"] >= 80:
    print("Bill passed with a high score!")
else:
    print("Bill needs to study more.")
````

### Түлхүүр нь хувьсагч байж болно

Түлхүүрийг шууд бичихийн оронд эхлээд хувьсагчид хадгалж болно. Тэгвэл `[ ]` дотор хувьсагчийг **хашилтгүй** бичнэ.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Anna"               # name хувьсагчид "Anna" гэсэн тэмдэгт мөр хадгалагдсан
print(scores[name])         # scores[name] нь scores["Anna"]-тай ижил -> 85

name = "Bill"               # Одоо хувьсагчид "Bill" байна
print(scores[name])         # scores["Bill"] -> 78
````

**Хэд хэдэн утгыг хэвлэх тухай:** дараагийн нүдэнд бид `print(name, scores[name])` гэж бичнэ. `print()` дотор хэд хэдэн утгыг **таслал** `,`-аар тусгаарлаж бичвэл тэдгээр нь нэг мөрөнд хэвлэгдэж, Python хооронд нь **нэг зай** автоматаар тавина.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Tom"
print(name, scores[name])   # Tom 92 -> нэр ба оноо нэг мөрөнд
````

## 5. Утга өөрчлөх ба шинэ хос нэмэх

Утга өөрчлөх эсвэл шинэ хос нэмэхэд ижил хэлбэрийг ашиглана:

```python
dictionary_нэр[түлхүүр] = утга
```

Түлхүүр dictionary-д **аль хэдийн** байгаа эсэхээс хамаарч Python хоёр зүйлийн аль нэгийг хийнэ:

| Түлхүүр dictionary-д аль хэдийн байна уу? | Юу болох вэ |
|---|---|
| **Тийм** | Хуучин утга шинэ утгаар **солигдоно** |
| **Үгүй** | **Шинэ хос** нэмэгдэнэ |

Энэ нь жагсаалтаас ялгаатай. Жагсаалтад зөвхөн аль хэдийн байгаа индексийн утгыг өөрчилж болдог. Нэмэхийн тулд `append()` хэрэгтэй. Dictionary-д бол ижил `[түлхүүр] = утга` нь өөрчлөх ч, нэмэх ч болно.

### Утга өөрчлөх (түлхүүр аль хэдийн байгаа)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Bill"] = 88         # "Bill" аль хэдийн түлхүүр -> 78 гэсэн утга 88-аар солигдоно
print(scores)               # Зөвхөн Bill-ийн оноо өөрчлөгдсөн
````

````python
# Хуучин утгыг нь ашиглан утгыг өөрчлөх (age = age + 1-тэй ижил санаа)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Anna"] = scores["Anna"] + 5     # Эхлээд баруун тал: 85 + 5 = 90, дараа нь буцааж хадгална
print(scores["Anna"])                   # 90
````

````python
# Богино хэлбэр += ч ажиллана (дараахтай ижил: scores["Tom"] = scores["Tom"] + 3)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Tom"] += 3
print(scores["Tom"])                    # 95
````

### Шинэ хос нэмэх (түлхүүр хараахан байхгүй)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Kate"] = 95         # "Kate" хараахан түлхүүр БИШ -> шинэ хос нэмэгдэнэ
print(scores)               # Одоо 4 хос байна
````

````python
# Хоосон dictionary-ээс эхлээд хосуудыг нэг нэгээр нэмэх
capitals = {}                   # Хоосноос эхэлнэ
print(capitals)

capitals["Korea"] = "Seoul"     # Хос нэмнэ
print(capitals)

capitals["Japan"] = "Tokyo"     # Дахиад нэг хос
print(capitals)

capitals["France"] = "Paris"    # Бас нэг
print(capitals)
````

````python
# Жагсаалттай харьцуулаарай: нэмэхэд жагсаалтад append() хэрэгтэй, dictionary-д [түлхүүр] = утга л болно
fruit_list = ["apple", "banana"]
fruit_list.append("cherry")         # Жагсаалт: append() төгсгөлд нэмнэ
print(fruit_list)

prices = {"apple": 1000, "banana": 500}
prices["cherry"] = 3000             # Dictionary: [шинэ түлхүүр] = утга
print(prices)
````

## 6. len() ба in

### len(): хэдэн хос байна вэ?

`len()`-ийг санаарай: жагсаалтад хэдэн утга байгааг хэлдэг. Dictionary-тэй бол `len()` нь хэдэн **хос** байгааг хэлнэ.

Нэг хос (`"Tom": 92`) нь 2 биш **1** гэж тоологдоно.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(len(scores))          # 3 хос -> 3

one_pair = {"Tom": 92}
print(len(one_pair))        # 1 хос -> 1 (2 биш!)

empty_dict = {}
print(len(empty_dict))      # 0 хос -> 0
````

````python
# Хос нэмэхэд len() өөрчлөгдөнө
scores = {"Anna": 85, "Tom": 92}
print(len(scores))          # 2

scores["Bill"] = 78         # Шинэ хос нэмнэ
print(len(scores))          # 3
````

### in: энэ түлхүүр dictionary-д байна уу?

`in`-ийг санаарай: `утга in жагсаалт` нь утга жагсаалтад байгаа эсэхийг шалгаж, `True` эсвэл `False` өгдөг.

Dictionary-тэй бол `in` нь **түлхүүрүүдийг** шалгана.

```python
"Tom" in scores      # "Tom" гэсэн түлхүүр байна уу?
```

**Анхаараарай:** `in` нь утгуудыг биш, **зөвхөн түлхүүрүүдийг** шалгадаг. 92 нь dictionary-д утга болон байгаа ч `92 in scores` нь `False`.

Түлхүүр dictionary-д **байхгүй** эсэхийг шалгахын тулд `not in`-ийг ашиглана.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" түлхүүр мөн үү? -> True
print("Kate" in scores)     # "Kate" түлхүүр мөн үү? -> False
print("Kate" not in scores) # "Kate" түлхүүр БИШ үү? -> True
````

````python
# in нь утгуудыг биш, зөвхөн ТҮЛХҮҮРҮҮДИЙГ шалгана
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" бол түлхүүр -> True
print(92 in scores)         # 92 бол түлхүүр биш, утга -> False
````

````python
# in-ийг if-тэй хамт: эхлээд шалгаад, дараа нь утгыг авна
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    print(name, scores[name])
else:
    print(name, "is not in the list.")
````

````python
# Энгийн толь бичгийн програм
english = {"apple": "a round fruit", "dog": "an animal that barks"}
word = "cat"

if word in english:
    print(english[word])
else:
    print("Sorry, I don't know that word.")
````

## 7. Хос устгах

Хос устгахын тулд `del` болон түлхүүрийг ашиглана:

```python
del dictionary_нэр[түлхүүр]
```

`del` нь **delete** (устгах)-ийн товчлол. Энэ нь түлхүүр **болон** түүний утгыг хамт устгана.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

del scores["Tom"]           # "Tom" түлхүүр болон 92 гэсэн утгыг нь устгана
print(scores)
print(len(scores))          # Одоо 2 хос
````

````python
# Байхгүй түлхүүрийг устгавал алдаа гарна (10-р хэсэгт үзнэ)
# Тиймээс эхлээд in-ээр шалгана
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
    print(name, "was deleted.")
else:
    print(name, "is not in the dictionary.")
````

## 8. Dictionary-тэй for давталт ашиглах

for давталт жагсаалттай хэрхэн ажилладгийг санацгаая:

```python
for score in score_list:
    print(score)
```

- Давталт бүрд жагсаалтын нэг утга `score` хувьсагчид хадгалагдана.
- `for` мөр хоёр цэг `:`-ээр төгсөнө.
- Давтах код **догол мөртэй** (4 зай) байна. Python догол мөрөөр аль мөрүүд давталтад хамаарахыг мэддэг.
- Догол мөргүй эхний мөр давталтын гадна байж, давталтын дараа зөвхөн нэг удаа ажиллана.

Dictionary-тэй бол for давталт **түлхүүрүүдийг** нэг нэгээр өгдөг.

```python
for name in scores:
    print(name)          # Anna, Tom, Bill
```

Түлхүүр байгаа бол 4-р хэсгийн нэгэн адил `scores[name]`-ээр утгыг авч болно.

````python
# for давталт dictionary-ийн ТҮЛХҮҮРҮҮДИЙГ өгдөг
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:         # name нь "Anna", дараа нь "Tom", дараа нь "Bill" болно
    print(name)
````

````python
# Түлхүүрээр утгыг авах
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(scores[name])     # scores["Anna"], дараа нь scores["Tom"], дараа нь scores["Bill"]
````

````python
# Түлхүүр ба утгыг нэг мөрөнд (таслалтай print)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])
````

### Давталтын хувьсагчийн нэр

`for`-ийн дараах хувьсагч (энд `name`) нь шинэ хувьсагч. Түүнд ямар ч нэр өгч болно. **Түлхүүрүүд** юу болохыг хэлдэг нэр сонгоорой: нэрс → `name`, улсууд → `country`, үгс → `word`.

````python
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

for country in capitals:            # Түлхүүрүүд нь улсууд тул country гэж нэрлэнэ
    print(country, capitals[country])
````

### Юу давтагдахыг догол мөр шийддэг

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])       # Догол мөртэй -> түлхүүр бүрд давтагдана
print("That's everyone!")           # Догол мөргүй -> давталтын гадна -> нэг удаа ажиллана
````

### Утгуудыг нэмэх

**Хуримтлагдах нийлбэр**-ийг санаарай:

1. **Давталтын өмнө:** `total = 0`
2. **Давталтын дотор:** одоогийн утгыг нэмнэ → `total = total + scores[name]`
3. **Давталтын дараа:** `total`-д нийлбэр байна

`total = total + scores[name]` нь: эхлээд баруун талыг (хуучин нийлбэр + энэ сурагчийн оноо) бодоод, үр дүнг `total` руу буцааж хадгал гэсэн утгатай. Богино хэлбэр нь `total += scores[name]`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0                           # Давталтын өмнө: 0-ээс эхэлнэ

for name in scores:
    total = total + scores[name]    # хуучин нийлбэр + энэ сурагчийн оноо

print("Total:", total)              # Давталтын дараа (догол мөргүй): 85 + 92 + 78 = 255
````

````python
# Дундаж = нийлбэр / утгын тоо
# len(scores) нь хосын тоо (= сурагчдын тоо)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:
    total += scores[name]           # дараахтай ижил: total = total + scores[name]

average = total / len(scores)       # Давталтын дараа: 255 / 3
print("Average:", average)
````

### Давталтын доторх if

`for`-ийн дотор `if` хийхэд догол мөр хоёр шат болно: `if`-д 4 зай (`for`-ийн дотор), `if`-ийн доторх кодод 8 зай.

````python
# Зөвхөн тэнцсэн (60 ба түүнээс дээш) сурагчдыг хэвлэх
scores = {"Anna": 85, "Tom": 42, "Bill": 78, "Kate": 55}

for name in scores:
    if scores[name] >= 60:          # 4 зай: сурагч бүрийн оноог шалгана
        print(name, "passed")       # 8 зай: зөвхөн нөхцөл True үед
````

````python
# Өнөөдөр хэн хичээлдээ ирсэн бэ?
attendance = {"Anna": True, "Tom": False, "Bill": True}

for name in attendance:
    if attendance[name]:            # Утга нь аль хэдийн True эсвэл False тул шууд нөхцөл болгож болно
        print(name, "is here.")
    else:
        print(name, "is absent.")
````

## 9. Жагсаалт ба dictionary

| | Жагсаалт | Dictionary |
|---|---|---|
| Үүсгэх | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92}` |
| Үүсгэх хаалт | `[ ]` | `{ }` |
| Утгыг юугаар олох вэ | **индекс** (байрлалын дугаар) | **түлхүүр** (нэр) |
| Утга авах | `score_list[1]` | `score_dict["Tom"]` |
| Утга өөрчлөх | `score_list[1] = 95` | `score_dict["Tom"] = 95` |
| Нэмэх | `score_list.append(88)` | `score_dict["Kate"] = 88` |
| Устгах | `score_list.remove(92)` | `del score_dict["Tom"]` |
| `len()` юуг тоолох вэ | утгууд | хосууд |
| `in` юуг шалгах вэ | утгууд | **түлхүүрүүд** |
| `for x in ...` юу өгөх вэ | утгууд | **түлхүүрүүд** |

**Алийг нь хэзээ ашиглах вэ?**
- **Дараалал** чухал, эсвэл утгуудад нэр хэрэггүй бол → **жагсаалт** (жишээ нь, өнөөдрийн шалгалтын оноонууд, дарааллаар)
- Утгыг **нэрээр нь** олохыг хүсвэл → **dictionary** (жишээ нь, сурагч бүрийн оноо, үг ба түүний утга)

## 10. Түгээмэл алдаанууд

Эдгээр нь dictionary-г сурч байхдаа бараг бүх хүний гаргадаг алдаанууд. Алдаа бүрийн хувьд бид дараахыг харна:

- ❌ **Буруу код**
- 🔍 **Юу буруу болсон бэ**
- ✅ **Зөв арга**

Алдаа гаргадаг кодыг notebook зогсохгүйн тулд тайлбар хэлбэрээр (эхэнд нь `#`-тэй) бичсэн. Жинхэнэ алдааны мэдэгдлийг харахыг хүсвэл `#`-г арилгаж, нүдийг ажиллуулаад, дараа нь `#`-г буцааж тавиарай.

### Алдаа 1: Байхгүй түлхүүр ашиглах

````python
# ❌ Буруу
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores["Kate"])     # KeyError: 'Kate'
# print(scores["tom"])      # KeyError: 'tom'
````

🔍 **Юу буруу болсон бэ?**

- `"Kate"` нь dictionary-ийн түлхүүр биш. Python үүнийг олж чадахгүй тул `KeyError` ("ийм түлхүүр байхгүй") гаргана.
- `"tom"` (жижиг t) ба `"Tom"` (том T) нь **өөр** тэмдэгт мөрүүд. Түлхүүрийг том, жижиг үсгийг нь оруулаад **яг адилхан** бичих ёстой.

✅ **Зөв арга:** Түлхүүрийг dictionary-д байгаатай яг адилхан бичнэ. Түлхүүр байгаа эсэхэд итгэлгүй байвал эхлээд `in`-ээр шалгана.

````python
# ✅ Зөв
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # Түлхүүртэй яг адилхан

name = "Kate"
if name in scores:          # Эхлээд шалгана
    print(scores[name])
else:
    print(name, "is not in the dictionary.")
````

✅ **Зөв арга:** Dictionary үүсгэхдээ **гоё хаалт** `{ }` ашиглана. (Утга **авахдаа** дөрвөлжин хаалт ашигласаар байна: `scores["Anna"]`.)

### Алдаа 3: Түлхүүрийн хашилтыг мартах

### Алдаа 5: Ижил түлхүүрийг хоёр удаа бичих

````python
# ❌ Буруу: Tom гэдэг хоёр өөр сурагч уу?
scores = {"Anna": 85, "Tom": 92, "Tom": 60}

print(scores)               # {'Anna': 85, 'Tom': 60} -> эхний Tom (92) алга болсон!
print(len(scores))          # 3 биш, 2
````

- Ижил түлхүүрийг дахин бичвэл 5-р хэсгийнх шиг ажиллана: түлхүүр аль хэдийн байгаа тул хуучин утга **солигдоно**. 92 нь 60-аар солигдсон.
- Алдаа гарахгүй тул анзааралгүй өнгөрөхөд амархан.

✅ **Зөв арга:** Бүх түлхүүрийг өөр өөр болгоно. Хоёр хүн ижил нэртэй бол тэдэнд өөр түлхүүр өгнө.

````python
# ✅ Зөв
scores = {"Anna": 85, "Tom K.": 92, "Tom L.": 60}

print(scores)
print(len(scores))          # 3
````

### Алдаа 6: in утгуудыг шалгана гэж бодох

````python
# ❌ Буруу: хэн нэг нь 92 оноо авсан эсэхийг мэдэхийг хүсэж байна
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(92 in scores)         # False -> гэхдээ Tom 92 авсан!
````

🔍 **Юу буруу болсон бэ?**

- Dictionary-тэй бол `in` нь утгуудыг (оноонуудыг) биш, **зөвхөн түлхүүрүүдийг** (нэрсийг) шалгана.
- 92 нь түлхүүр биш, утга тул үр дүн нь `False`.
- Алдаа гарахгүй ч хариулт буруу.

✅ **Зөв арга:** Dictionary-г for давталтаар гүйж, утга бүрийг харьцуулна.

````python
# ✅ Зөв
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    if scores[name] == 92:      # УТГА бүрийг 92-той харьцуулна
        print(name, "got 92 points.")
````

### Алдаа 7: for давталт утгуудыг өгнө гэж бодох

````python
# ❌ Буруу: бүх оноог нэмэхийг хүсэж байна
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

# for score in scores:
#     total += score        # TypeError: unsupported operand type(s) for +=: 'int' and 'str'
````

🔍 **Юу буруу болсон бэ?**

- Dictionary-тэй бол for давталт утгуудыг биш, **түлхүүрүүдийг** өгдөг. Тиймээс `score` нь 85, 92, 78 биш, `"Anna"`, `"Tom"`, `"Bill"` байна.
- `total += score` нь тоон (`0`) дээр тэмдэгт мөр (`"Anna"`) нэмэх гэж оролддог. Python тоо ба тэмдэгт мөрийг нэмж чадахгүй тул `TypeError` гаргана.
- Хувьсагчийг `score` гэж нэрлэсэн нь бүр ч будлиулсан. Үнэндээ энэ бол нэр.

✅ **Зөв арга:** Давталтын хувьсагчийг түлхүүрээр нь (`name`) нэрлэж, утгыг `scores[name]`-ээр авна.

````python
# ✅ Зөв
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:             # name бол түлхүүр
    total += scores[name]       # scores[name] бол утга

print("Total:", total)
````

### Алдаа 8: Байхгүй түлхүүрийг устгах

````python
# ❌ Буруу
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# del scores["Kate"]        # KeyError: 'Kate'
````

🔍 **Юу буруу болсон бэ?**

- Dictionary-д `"Kate"` гэсэн түлхүүр байхгүй тул устгах зүйл алга.
- Алдаа 1-ийн нэгэн адил Python `KeyError` гаргана.

✅ **Зөв арга:** Эхлээд `in`-ээр шалгаж, түлхүүр байгаа үед л устгана.

````python
# ✅ Зөв
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
else:
    print(name, "is not in the dictionary, so nothing was deleted.")
````
