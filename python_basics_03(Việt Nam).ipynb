## 1. Vòng lặp for là gì?

Hãy tưởng tượng bạn phải in ra "Hello" 5 lần. Bạn có thể viết `print("Hello")` 5 lần. Nhưng nếu là 100 lần thì sao? Hay 1000 lần? Viết đi viết lại cùng một dòng vừa mệt, vừa dễ sai.

**Vòng lặp** giải quyết vấn đề này. Vòng lặp chạy cùng một đoạn mã lặp đi lặp lại, bao nhiêu lần tùy bạn muốn.

Vòng lặp phổ biến nhất trong Python là vòng lặp `for`. Dạng cơ bản như sau:

```python
for i in range(5):
    mã cần lặp lại
```

Bạn có thể đọc như sau: **"với mỗi số i trong range(5), hãy chạy đoạn mã bên dưới"**.

- `range(5)` cho ra các số 0, 1, 2, 3, 4 (tổng cộng 5 số). Chúng ta sẽ học kỹ về `range()` ở phần tiếp theo.
- `i` là một biến. Mỗi lần vòng lặp lặp lại, số tiếp theo sẽ được lưu vào `i`.
- Giống như `if`, dòng `for` phải kết thúc bằng dấu hai chấm `:`.
- Giống như `if`, mã cần lặp lại phải được **thụt lề** (4 dấu cách). Chỉ những dòng được thụt lề mới được lặp lại.

````python
# Lặp lại print("Hello") 5 lần
# range(5) cho ra 5 số (0, 1, 2, 3, 4), nên đoạn mã thụt lề chạy 5 lần
for i in range(5):
    print("Hello")      # Dòng này nằm trong vòng lặp for (được thụt lề), nên nó được lặp lại
````

````python
# Hãy xem mỗi lần i chứa giá trị gì
# Lần 1: i = 0, lần 2: i = 1, lần 3: i = 2, lần 4: i = 3, lần 5: i = 4
for i in range(5):
    print(i)            # In ra giá trị hiện tại của i
````

````python
# Mỗi lần vòng lặp chạy hết một lượt đoạn mã thụt lề được gọi là một "lần lặp"
# Mọi dòng thụt lề đều chạy một lần trong mỗi lần lặp, theo thứ tự từ trên xuống dưới
for i in range(3):
    print("Repetition number", i)     # print() có thể hiển thị nhiều giá trị cùng lúc nếu bạn ngăn cách chúng bằng dấu phẩy
    print("----------")               # Dòng này cũng nằm trong vòng lặp (cùng mức thụt lề)

print("The loop is finished.")        # KHÔNG thụt lề, nên nằm ngoài vòng lặp. Chỉ chạy một lần, sau khi vòng lặp kết thúc
````

````python
# Tên biến không nhất thiết phải là i
# i chỉ là cái tên mà phần lớn mọi người hay dùng. Bạn có thể chọn bất kỳ tên nào, miễn là đúng quy tắc đặt tên biến
for number in range(3):
    print(number)

for count in range(3):
    print("count is", count)
````

## 2. Tìm hiểu kỹ về range()

`range()` tạo ra một dãy số. Có ba cách để sử dụng nó.

| Cách viết | Ý nghĩa | Các số được tạo ra |
|---|---|---|
| `range(end)` | từ 0 đến trước end (không bao gồm end) | `range(5)` → 0, 1, 2, 3, 4 |
| `range(start, end)` | từ start đến trước end (không bao gồm end) | `range(1, 6)` → 1, 2, 3, 4, 5 |
| `range(start, end, step)` | từ start đến trước end, mỗi lần nhảy step | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |

Điều quan trọng nhất cần nhớ:
- **Số cuối (end) không bao giờ được bao gồm.** `range(5)` dừng ở 4, không phải 5.
- Nếu không cho số bắt đầu, nó bắt đầu từ **0**.
- Nếu không cho bước nhảy (step), nó tăng lên **1** mỗi lần.

````python
# range(end): bắt đầu từ 0 và dừng ngay trước end
# range(4) cho ra 0, 1, 2, 3 -> 4 KHÔNG được bao gồm
for i in range(4):
    print(i)
````

````python
# range(start, end): bắt đầu từ start và dừng ngay trước end
# Nếu muốn từ 1 đến 5, end phải là 6 (vì số cuối không được bao gồm)
for i in range(1, 6):
    print(i)
````

````python
# range(start, end, step): mỗi lần nhảy step
# Bắt đầu từ 0, nhảy 2, dừng trước 10 -> 0, 2, 4, 6, 8
for i in range(0, 10, 2):
    print(i)
````

````python
# Một ví dụ khác về step: đếm cách 5 từ 5 đến 25
# end là 26 để 25 được bao gồm (cũng có thể viết 30, vì bản thân 30 sẽ không được bao gồm)
for i in range(5, 26, 5):
    print(i)
````

````python
# Nếu step là số âm, các số sẽ GIẢM dần (đếm ngược)
# Bắt đầu từ 5, giảm 1, dừng trước 0 -> 5, 4, 3, 2, 1
for i in range(5, 0, -1):
    print(i)

print("Lift off!")      # Nằm ngoài vòng lặp, chạy một lần ở cuối
````

````python
# Dùng i trong phép tính
# In bảng cửu chương 3 (3 x 1 đến 3 x 9)
for i in range(1, 10):           # i đi từ 1 đến 9 (10 không được bao gồm)
    print(3, "x", i, "=", 3 * i)  # Mỗi lần, 3 * i được tính lại với giá trị i mới
````

## 3. Cộng dồn bằng vòng lặp

Một trong những công việc phổ biến nhất của vòng lặp là **cộng dồn** các số. Ví dụ, 1 + 2 + 3 + ... + 10 bằng bao nhiêu?

Mẹo là dùng một biến làm "tổng tích lũy".
1. Trước vòng lặp, tạo một biến và gán bằng 0. (`total = 0`)
2. Bên trong vòng lặp, mỗi lần cộng số hiện tại vào nó. (`total = total + i`)
3. Sau vòng lặp, biến đó chứa kết quả cuối cùng.

Bạn còn nhớ `age = age + 1` ở Bài 1 không? Đây chính là ý tưởng đó: lấy giá trị cũ, cộng thêm gì đó, rồi lưu lại vào cùng một chiếc hộp.

````python
# Cộng dồn 1 + 2 + 3 + ... + 10
total = 0                     # Bước 1: tổng tích lũy bắt đầu từ 0 (TRƯỚC vòng lặp)

for i in range(1, 11):        # i đi từ 1 đến 10
    total = total + i         # Bước 2: cộng i hiện tại vào tổng tích lũy

print("Sum:", total)          # Bước 3: SAU vòng lặp, in kết quả cuối cùng (nằm ngoài vòng lặp nên chỉ in một lần)
````

````python
# Hãy xem total thay đổi thế nào qua mỗi lần lặp
# Đây là cùng đoạn mã ở trên, nhưng có thêm print() bên trong vòng lặp
total = 0

for i in range(1, 6):         # i đi từ 1 đến 5
    total = total + i
    print("i =", i, "-> total =", total)   # Nằm trong vòng lặp nên in ra mỗi lần

# Lần 1: 0 + 1 = 1
# Lần 2: 1 + 2 = 3
# Lần 3: 3 + 3 = 6
# Lần 4: 6 + 4 = 10
# Lần 5: 10 + 5 = 15
````

````python
# Bạn có thể viết total = total + i ngắn gọn hơn: total += i
# Hai dòng này có ý nghĩa hoàn toàn giống nhau. Nhiều người lập trình Python dùng cách viết ngắn
total = 0

for i in range(1, 101):       # i đi từ 1 đến 100
    total += i                # Giống với: total = total + i

print("Sum of 1 to 100:", total)
````

````python
# Giá trị tích lũy không nhất thiết phải là tổng. Chúng ta cũng có thể nhân
# 1 x 2 x 3 x 4 x 5 (trong toán học gọi là "giai thừa")
result = 1                    # Với phép nhân, bắt đầu từ 1 chứ không phải 0 (bất cứ số nào nhân 0 cũng bằng 0!)

for i in range(1, 6):         # i đi từ 1 đến 5
    result = result * i       # Nhân i hiện tại vào kết quả

print("1 x 2 x 3 x 4 x 5 =", result)
````

````python
# Đếm số lần vòng lặp đã chạy cũng là một kiểu tích lũy
# Ở đây mỗi lần ta cộng 1, thay vì cộng i
count = 0

for i in range(10, 20):       # i đi từ 10 đến 19
    count += 1                # Mỗi lần lặp cộng thêm 1

print("The loop ran", count, "times.")   # range(10, 20) có 10 số, nên vòng lặp chạy 10 lần
````

## 4. Dùng for và if cùng nhau

Bạn có thể đặt `if` **bên trong** vòng lặp `for`. Cách này cho phép bạn kiểm tra từng số một, và chỉ làm gì đó với những số bạn muốn.

```python
for i in range(...):
    if điều_kiện:
        mã chỉ chạy khi điều kiện là True
```

- `if` nằm trong `for`, nên được thụt lề 4 dấu cách.
- Mã bên trong `if` được thụt lề thêm một bậc, nên được thụt lề 8 dấu cách.

````python
# Chỉ in ra các số chẵn từ 1 đến 10
# Nhớ lại Bài 2: một số là số chẵn nếu number % 2 == 0
for i in range(1, 11):        # Kiểm tra từng số từ 1 đến 10, lần lượt từng số một
    if i % 2 == 0:            # i hiện tại có phải số chẵn không? (4 dấu cách: bên trong for)
        print(i)              # Chỉ số chẵn mới được in ra (8 dấu cách: bên trong if)
````

````python
# Cũng có thể tạo ra kết quả tương tự chỉ bằng bước nhảy của range(), không cần if
# Bắt đầu từ 2, nhảy 2 -> 2, 4, 6, 8, 10
for i in range(2, 11, 2):
    print(i)

# Cả hai cách đều đúng. Nhưng với if, bạn có thể kiểm tra bất kỳ điều kiện nào, chứ không chỉ "cách 2 số một"
````

````python
# Dùng if - else bên trong vòng lặp
for i in range(1, 6):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
````

````python
# Kết hợp for + if + tổng tích lũy
# Chỉ cộng các bội số của 3 trong khoảng từ 1 đến 20 (3 + 6 + 9 + 12 + 15 + 18)
total = 0

for i in range(1, 21):        # i đi từ 1 đến 20
    if i % 3 == 0:            # i có phải bội số của 3 không? (chia cho 3 không có số dư)
        total += i            # Chỉ khi đó mới cộng vào tổng

print("Sum of multiples of 3:", total)
````

````python
# Một bài tập kinh điển: FizzBuzz
# Với các số từ 1 đến 15:
#   - nếu là bội số của cả 3 và 5, in ra "FizzBuzz"
#   - nếu chỉ là bội số của 3, in ra "Fizz"
#   - nếu chỉ là bội số của 5, in ra "Buzz"
#   - các trường hợp còn lại, in ra chính số đó
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:     # Kiểm tra điều kiện khắt khe nhất TRƯỚC (nhớ Bài 2: thứ tự rất quan trọng!)
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

## 5. Vòng lặp while

Vòng lặp `for` phù hợp khi bạn biết **cần lặp bao nhiêu lần**. Nhưng đôi khi bạn không biết trước con số đó. Bạn chỉ biết **khi nào cần dừng**. Ví dụ:

- Tiếp tục để dành tiền **trong khi** tổng số tiền còn nhỏ hơn 10000.
- Tiếp tục cắt đôi tờ giấy **trong khi** nó còn dài hơn 1 cm.

Trong trường hợp này, chúng ta dùng vòng lặp `while`. `while` có nghĩa là "trong khi". Nó tiếp tục lặp lại **chừng nào điều kiện còn là True**, và dừng lại khi điều kiện trở thành False.

```python
while điều_kiện:
    mã cần lặp lại
```

Cách hoạt động:
1. Kiểm tra điều kiện.
2. Nếu là True, chạy đoạn mã thụt lề, rồi quay lại bước 1.
3. Nếu là False, dừng vòng lặp và chuyển sang đoạn mã phía sau nó.

**Rất quan trọng:** Phải có thứ gì đó bên trong vòng lặp thay đổi, để đến một lúc nào đó điều kiện trở thành False. Nếu điều kiện cứ mãi là True, vòng lặp sẽ không bao giờ kết thúc. Điều này được gọi là **vòng lặp vô hạn**.

````python
# In từ 1 đến 5 bằng while
# Với while, chúng ta phải tự tạo, tự kiểm tra và tự thay đổi biến
i = 1                     # 1. Tạo biến TRƯỚC vòng lặp

while i <= 5:             # 2. Kiểm tra điều kiện: tiếp tục chừng nào i còn nhỏ hơn hoặc bằng 5
    print(i)
    i = i + 1             # 3. Thay đổi biến. Nếu không có dòng này, i mãi mãi là 1 (vòng lặp vô hạn!)

print("Done. Now i is", i)    # Khi i trở thành 6, 6 <= 5 là False, nên vòng lặp dừng lại
````

````python
# So sánh: làm điều tương tự bằng for
# for tự động làm việc "tạo, kiểm tra, thay đổi" nhờ range()
for i in range(1, 6):
    print(i)

# Vì vậy khi bạn biết cần lặp bao nhiêu lần, dùng for thường đơn giản hơn
````

````python
# Trường hợp while phù hợp hơn: chúng ta không biết trước cần lặp bao nhiêu lần
# Mỗi ngày bạn để dành 3000 won. Cần bao nhiêu ngày để có từ 20000 won trở lên?
money = 0
days = 0

while money < 20000:          # Tiếp tục để dành chừng nào số tiền còn nhỏ hơn 20000
    money += 3000             # Để dành 3000 won
    days += 1                 # Thêm một ngày nữa đã trôi qua

print("Days needed:", days)
print("Money saved:", money)
````

````python
# Một ví dụ khác: tiếp tục chia đôi một số cho đến khi nó nhỏ hơn hoặc bằng 1
# Chúng ta dùng // (chỉ lấy thương, ở Bài 1) để số luôn là số nguyên
number = 100
steps = 0

while number > 1:             # Tiếp tục chừng nào number còn lớn hơn 1
    number = number // 2      # Chia đôi (100 -> 50 -> 25 -> 12 -> 6 -> 3 -> 1)
    steps += 1
    print("Step", steps, "->", number)

print("It took", steps, "steps.")
````

````python
# Nếu điều kiện là False ngay từ đầu, vòng lặp while không chạy lần nào
i = 10

while i < 5:                  # 10 < 5 là False ngay lập tức
    print("This never prints.")
    i += 1

print("The loop was skipped.")
````

## 6. break và continue

Đôi khi bạn muốn thay đổi luồng chạy của vòng lặp giữa chừng.

- `break`: **dừng toàn bộ vòng lặp ngay lập tức**, và thoát ra ngoài.
- `continue`: **bỏ qua phần còn lại của lần lặp này**, và chuyển thẳng sang lần lặp tiếp theo.

Cả hai thường được dùng cùng với `if`, vì chúng ta chỉ muốn break hoặc continue trong một tình huống nhất định.

````python
# break: dừng vòng lặp ngay khi tìm thấy thứ chúng ta cần
# Tìm số đầu tiên trong khoảng 1 đến 100 là bội số của 7
for i in range(1, 101):
    if i % 7 == 0:                # i có phải bội số của 7 không?
        print("Found it:", i)
        break                     # Dừng vòng lặp ngay tại đây. Các số sau 7 hoàn toàn không được kiểm tra

print("The loop is over.")
````

````python
# Hãy xem chính xác break dừng vòng lặp ở đâu
for i in range(1, 11):
    if i == 4:                    # Khi i trở thành 4...
        break                     # ...dừng vòng lặp ngay lập tức
    print(i)                      # Nên chỉ 1, 2, 3 được in ra (4 không được in, vì break chạy trước)
````

````python
# continue: chỉ bỏ qua lần lặp này, và tiếp tục với lần lặp tiếp theo
for i in range(1, 6):
    if i == 3:                    # Khi i là 3...
        continue                  # ...bỏ qua print() bên dưới và chuyển thẳng sang i = 4
    print(i)                      # Nên 1, 2, 4, 5 được in ra (chỉ thiếu 3)
````

````python
# break với while: một mẫu thường gặp là "while True" + break
# while True nghĩa là "lặp mãi mãi", nên break là cách DUY NHẤT để thoát ra
count = 0

while True:                       # Điều kiện luôn là True -> nếu không có break thì đây là vòng lặp vô hạn
    count += 1
    print("count:", count)
    if count == 3:                # Khi count đạt 3...
        break                     # ...thoát khỏi vòng lặp

print("Out of the loop!")
````

````python
# So sánh break và continue cạnh nhau
print("With break:")
for i in range(1, 6):
    if i == 3:
        break                     # Dừng tất cả tại 3
    print(i)                      # 1, 2

print("With continue:")
for i in range(1, 6):
    if i == 3:
        continue                  # Chỉ bỏ qua 3
    print(i)                      # 1, 2, 4, 5
````

## 7. Những lỗi thường gặp

Đây là những lỗi mà hầu như ai cũng mắc phải khi học vòng lặp. Nếu có gì đó không đúng, hãy kiểm tra những điểm này trước.

1. **Quên rằng số cuối của range() không được bao gồm**: `range(1, 10)` dừng ở 9. Muốn bao gồm 10, hãy viết `range(1, 11)`.
2. **Quên dấu hai chấm `:`**: Các dòng `for` và `while` phải kết thúc bằng `:`.
3. **Thụt lề sai**: Chỉ những dòng được thụt lề mới được lặp lại. Một dòng lẽ ra phải nằm trong vòng lặp nhưng không được thụt lề thì chỉ chạy một lần.
4. **Vòng lặp vô hạn với while**: Nếu quên thay đổi biến bên trong vòng lặp `while`, điều kiện sẽ mãi là True.
5. **Đặt lại tổng tích lũy bên trong vòng lặp**: `total = 0` phải được viết **trước** vòng lặp, không phải bên trong.

Các ô bên dưới cho thấy mỗi lỗi trông như thế nào. Mã bị lỗi được viết dưới dạng chú thích để không làm hỏng notebook.

**Nếu bạn lỡ chạy một vòng lặp vô hạn**, hãy nhấn nút ■ (dừng / interrupt) ở phía trên Jupyter, hoặc nhấn `I` hai lần (`I`, `I`) khi đang chọn ô đó.

````python
# Lỗi 1: quên rằng số cuối không được bao gồm
# Chúng ta muốn từ 1 đến 5, nhưng...
for i in range(1, 5):     # Sai: vòng lặp này dừng ở 4!
    print(i)

print("---")

for i in range(1, 6):     # Đúng: end phải lớn hơn số cuối cùng ta muốn 1 đơn vị
    print(i)
````

````python
# Lỗi 2: quên dấu hai chấm

# for i in range(3)       # SyntaxError! Thiếu dấu hai chấm ở cuối
#     print(i)

for i in range(3):        # Đúng: dòng kết thúc bằng dấu hai chấm
    print(i)
````

````python
# Lỗi 3: thụt lề sai
# print() thứ hai KHÔNG được thụt lề, nên nằm ngoài vòng lặp và chỉ chạy một lần
for i in range(3):
    print("Inside the loop:", i)
print("Outside the loop:", i)     # Chạy một lần, sau vòng lặp. (i vẫn giữ giá trị cuối cùng là 2)

# for i in range(3):
# print(i)                # IndentationError! Không có dòng mã nào được thụt lề bên trong for
````

````python
# Lỗi 4: vòng lặp vô hạn với while
# ĐỪNG xóa dấu # ở đoạn này! Nó sẽ không bao giờ dừng.

# i = 1
# while i <= 5:
#     print(i)            # i không bao giờ thay đổi, nên i <= 5 mãi là True -> vòng lặp vô hạn!

i = 1
while i <= 5:
    print(i)
    i += 1                # Đúng: thay đổi i mỗi lần, để cuối cùng điều kiện trở thành False
````

````python
# Lỗi 5: đặt lại tổng tích lũy bên trong vòng lặp
# Sai: total quay về 0 ở mỗi lần lặp, nên chỉ còn lại số cuối cùng
for i in range(1, 6):
    total = 0             # Sai vị trí! Dòng này chạy 5 lần
    total += i
print("Wrong total:", total)      # Chỉ là 5

# Đúng: chỉ gán total bằng 0 một lần, TRƯỚC vòng lặp
total = 0
for i in range(1, 6):
    total += i
print("Correct total:", total)    # 1 + 2 + 3 + 4 + 5 = 15
````
