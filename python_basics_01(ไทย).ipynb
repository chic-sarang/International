## 1. ตัวแปรและชนิดข้อมูล

ตัวแปรคือกล่องสำหรับเก็บค่า เมื่อตั้งชื่อให้กล่องแล้ว ภายหลังเราสามารถเรียกชื่อนั้นเพื่อดึงค่าที่อยู่ข้างในออกมาใช้ได้ ถ้าใส่ค่าใหม่เข้าไป เนื้อหาภายในกล่องก็จะเปลี่ยนไปด้วย

แต่ค่าที่สามารถใส่ลงในกล่องได้นั้นมีหลายประเภท ประเภทของค่านี้เรียกว่าชนิดข้อมูล ชนิดข้อมูลหลัก ๆ ในภาษาไพทอนมีดังนี้

- `str` : สตริง (ตัวอักษร)
- `int` : จำนวนเต็ม
- `float` : ทศนิยม (ตัวเลขที่มีจุดทศนิยม)
- `bool` : บูลีน (จริง/เท็จ)

นอกจากนี้ `print()` คือคำสั่งที่แสดงค่าภายในวงเล็บขึ้นบนหน้าจอ

```python
# เครื่องหมาย = ไม่ได้แปลว่า 'เท่ากับ' แต่แปลว่า 'เก็บค่าทางขวาไว้ในกล่องที่ชื่อทางซ้าย'
# ชื่อตัวแปร (name, age, height, is_student) คือชื่อที่ตั้งให้กับกล่อง

# สตริง (str) แค่ใส่เครื่องหมายคำพูดครอบก็พอ สามารถใช้ได้ทั้งเครื่องหมายคำพูดคู่ (" ") และเดี่ยว (' ')
name = "Bill"                    # คำที่ครอบด้วยเครื่องหมายคำพูดคู่
city = 'Seoul'                    # คำที่ครอบด้วยเครื่องหมายคำพูดเดี่ยว
greeting = "Hello, nice to meet you"  # แม้แต่ประโยคก็ยังถือเป็นสตริง
zip_code = "06236"               # แม้จะดูเหมือนตัวเลข แต่ถ้าอยู่ในเครื่องหมายคำพูดก็ถือเป็นสตริง

# จำนวนเต็ม (int) คือตัวเลขที่ไม่มีจุดทศนิยม ทั้งจำนวนบวก จำนวนลบ และ 0 ล้วนเป็นจำนวนเต็ม
age = 20              # จำนวนเต็มบวก
temperature = -5      # จำนวนเต็มลบก็ได้เช่นกัน
score = 0             # 0 ก็เป็นจำนวนเต็มเช่นกัน

# ทศนิยม (float) คือตัวเลขที่มีจุดทศนิยม
height = 175.5        # ทศนิยมบวก
account_balance = -3.75  # ทศนิยมลบก็ได้เช่นกัน
pi_value = 3.14159    # ต่อให้มีตัวเลขหลังจุดทศนิยมมากก็ไม่มีปัญหา

# บูลีน (bool) มีได้แค่สองค่าเท่านั้นคือ True (จริง) หรือ False (เท็จ) (ตัวอักษรแรกต้องเป็นตัวพิมพ์ใหญ่เสมอ)
is_student = True     # เป็นนักเรียน/นักศึกษา
is_married = False    # ยังไม่ได้แต่งงาน
has_passed = True     # สอบผ่านแล้ว

# print() คือคำสั่งที่แสดงค่าขึ้นบนหน้าจอ
print(name)
print(city)
print(greeting)
print(zip_code)
print(age)
print(temperature)
print(score)
print(height)
print(account_balance)
print(pi_value)
print(is_student)
print(is_married)
print(has_passed)
```

```python
# ค่าของตัวแปรสามารถเปลี่ยนแปลงได้ (ตัวอย่างจำนวนเต็ม)
age = 20
print(age)

age = 21   # ถ้าเก็บค่าใหม่ในตัวแปรเดิม ค่าเก่าจะหายไปและถูกแทนที่ด้วยค่าใหม่
print(age)

age = age + 1   # สามารถสร้างค่าใหม่โดยใช้ค่าเดิมก็ได้ (21 + 1 = 22)
print(age)
```

```python
# ค่าของตัวแปรสามารถเปลี่ยนแปลงได้ (ตัวอย่างบูลีน)
is_raining = False
print(is_raining)

is_raining = True   # ถ้าเก็บค่าใหม่ในตัวแปรเดิม ค่าเก่าจะหายไปและถูกแทนที่ด้วยค่าใหม่
print(is_raining)

is_raining = not is_raining   # การใช้ not จะกลับค่า: True กลายเป็น False และ False กลายเป็น True
print(is_raining)
```

```python
# ค่าของตัวแปรสามารถเปลี่ยนแปลงได้ (ตัวอย่างสตริง)
city = "Seoul"
print(city)

city = "Busan"   # ถ้าเก็บค่าใหม่ในตัวแปรเดิม ค่าเก่าจะหายไปและถูกแทนที่ด้วยค่าใหม่
print(city)

city = city + " Travel"   # สตริงสามารถต่อกันได้ด้วยเครื่องหมาย + ("Busan" + " Travel" = "Busan Travel")
print(city)
```

```python
# ค่าของตัวแปรสามารถเปลี่ยนแปลงได้ (ตัวอย่างทศนิยม)
height = 175.5
print(height)

height = 178.2   # ถ้าเก็บค่าใหม่ในตัวแปรเดิม ค่าเก่าจะหายไปและถูกแทนที่ด้วยค่าใหม่
print(height)

height = height + 0.5   # สามารถสร้างค่าใหม่โดยใช้ค่าเดิมก็ได้ (178.2 + 0.5 = 178.7)
print(height)
```

## 2. ตัวดำเนินการพื้นฐาน

สัญลักษณ์ที่ใช้ในการคำนวณตัวเลขหรือเปรียบเทียบค่า

```python
# การคำนวณสี่หลักพื้นฐาน
a = 10
b = 3

print(a + b)   # บวก
print(a - b)   # ลบ
print(a * b)   # คูณ
print(a / b)   # หาร (ผลลัพธ์เป็นทศนิยม)
print(a // b)  # หารเอาเฉพาะผลหาร
print(a % b)   # หารเอาเศษ
```

```python
# ตัวดำเนินการเปรียบเทียบ -> ผลลัพธ์จะเป็น True หรือ False เสมอ
print(a > b)   # a มากกว่า b
print(a < b)   # a น้อยกว่า b
print(a == b)  # a เท่ากับ b
print(a != b)  # a ไม่เท่ากับ b
```
