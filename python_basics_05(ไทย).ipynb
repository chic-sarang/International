## 1. ทำไมต้องใช้ลูป for กับลิสต์?

ในบทที่ 4 เราได้เรียนว่าลิสต์เก็บค่าได้หลายค่า และเราดึงค่าออกมาทีละค่าได้ด้วย **อินเด็กซ์ (index)** (หมายเลขตำแหน่ง) ค่าแรกอยู่ที่อินเด็กซ์ 0 ค่าที่สองอยู่ที่อินเด็กซ์ 1 ไปเรื่อย ๆ

ดังนั้นถ้าอยากแสดงคะแนนทุกค่าในลิสต์ เราอาจเขียนหนึ่งบรรทัดต่อหนึ่งอินเด็กซ์:

```python
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
```

แบบนี้ก็ใช้ได้ แต่มีปัญหาอยู่ 2 ข้อ:

- ถ้าลิสต์มี 100 ค่า ก็ต้องเขียน 100 บรรทัด
- ถ้าเพิ่มค่าใหม่ด้วย `append()` ก็ต้องเพิ่มบรรทัด `print()` เองอีกหนึ่งบรรทัด

**ลูป for** ช่วยแก้ปัญหานี้ ลูป for จะดึงค่าออกจากลิสต์ **ทีละค่า ตั้งแต่ค่าแรกจนถึงค่าสุดท้าย** แล้วรันโค้ดเดิมกับทุกค่า ลิสต์จะยาวแค่ไหนก็ไม่สำคัญ เขียนโค้ดแค่ครั้งเดียวก็พอ

````python
# วิธียาว: เขียนหนึ่งบรรทัดต่อหนึ่งค่า
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
````

````python
# วิธีสั้น: ลูป for
# ผลลัพธ์เหมือนกันทุกประการ แต่ใช้แค่ 2 บรรทัด
scores = [85, 92, 78, 90, 66]
for score in scores:
    print(score)
````

````python
# ถึงลิสต์จะยาวขึ้น ลูป for ก็ยังเหมือนเดิม
# ไม่ต้องเพิ่มบรรทัด print() เลยสักบรรทัด
scores = [85, 92, 78, 90, 66]
scores.append(100)      # เพิ่มอีกหนึ่งค่าไว้ท้ายลิสต์ (ตอนนี้มี 6 ค่า)
scores.append(73)       # และอีกหนึ่งค่า (ตอนนี้มี 7 ค่า)

for score in scores:    # 2 บรรทัดเดิมเหมือนข้างบน
    print(score)
````

## 2. รูปแบบพื้นฐาน: `for ค่า in ลิสต์`

หน้าตาของโค้ดเป็นแบบนี้:

```python
for score in scores:
    โค้ดที่ต้องการทำซ้ำ
```

อ่านได้เหมือนประโยค: **"สำหรับ score แต่ละตัวใน scores ให้รันโค้ดข้างล่าง"**

มาดูทีละส่วน:

| ส่วน | ความหมาย |
|---|---|
| `for` | "ทำสิ่งต่อไปนี้ซ้ำกับทุกค่า" |
| `score` | ตัวแปร แต่ละรอบจะมีค่าหนึ่งค่าจากลิสต์ถูกเก็บไว้ที่นี่ |
| `in` | "ใน" |
| `scores` | ลิสต์ที่เราไล่ดู |
| `:` | โคลอน ต้องมีอยู่ท้ายบรรทัด `for` เสมอ |
| `    โค้ดที่ต้องการทำซ้ำ` | โค้ดที่ทำงานกับทุกค่า ต้องย่อหน้า (4 ช่องว่าง) |

**วิธีการทำงาน ทีละขั้น** (เมื่อ `scores = [85, 92, 78]`):

| รอบที่ | ค่าที่ถูกเก็บใน `score` | เกิดอะไรขึ้น |
|---|---|---|
| 1 | `85` (ค่าที่อินเด็กซ์ 0) | โค้ดที่ย่อหน้าทำงานโดย `score` = 85 |
| 2 | `92` (ค่าที่อินเด็กซ์ 1) | โค้ดที่ย่อหน้าทำงานโดย `score` = 92 |
| 3 | `78` (ค่าที่อินเด็กซ์ 2) | โค้ดที่ย่อหน้าทำงานโดย `score` = 78 |
| - | ไม่มีค่าเหลือแล้ว | ลูปจบ แล้ว Python ไปทำโค้ดที่อยู่ถัดจากลูป |

ดังนั้นลูปจะทำงาน **เท่ากับจำนวนค่าในลิสต์** ลิสต์ 3 ค่า → ทำซ้ำ 3 รอบ ลิสต์ 100 ค่า → ทำซ้ำ 100 รอบ

````python
# แสดงผลไม้ทีละชนิด
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:        # รอบที่ 1: fruit = "apple", รอบที่ 2: fruit = "banana", รอบที่ 3: fruit = "cherry"
    print(fruit)            # แสดงค่าที่อยู่ใน fruit ตอนนี้
````

````python
# ลิสต์ของตัวเลข
numbers = [10, 20, 30, 40]

for number in numbers:      # number เป็น 10 แล้วเป็น 20 แล้วเป็น 30 แล้วเป็น 40
    print(number)
````

````python
# ลิสต์ของค่าบูลีน (True / False)
answers = [True, False, True]

for answer in answers:      # answer เป็น True แล้วเป็น False แล้วเป็น True
    print(answer)
````

````python
# นำแต่ละค่ามาคำนวณได้
# ราคาแต่ละค่าจะถูกคูณ 2 แล้วแสดงออกมา
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)        # 1000 * 2 แล้ว 2500 * 2 แล้ว 3000 * 2
````

````python
# ค่าในลิสต์ "ไม่" เปลี่ยน
# ในเซลล์ข้างบนเราแสดง price * 2 แต่ตัวลิสต์เองยังเหมือนเดิม
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)

print(prices)               # ยังเป็น [1000, 2500, 3000]
````

````python
# ต่อสตริงด้วย +
# จำไว้ว่า: + ต่อสตริงสองตัวให้เป็นตัวเดียวได้ ("Hello, " + "Bill" -> "Hello, Bill")
names = ["Bill", "Anna", "Tom"]

for name in names:
    print("Hello, " + name)
````

````python
# ถ้าลิสต์ว่าง ลูปจะทำงาน 0 รอบ
# ไม่มีอะไรแสดงออกมา และไม่มี error
empty_list = []

for value in empty_list:
    print("This never prints.")

print("The loop ran 0 times.")
````

## 3. การย่อหน้า: บรรทัดไหนบ้างที่ถูกทำซ้ำ?

**การย่อหน้า** คือการเลื่อนบรรทัดไปทางขวาโดยเติมช่องว่างไว้ข้างหน้า ใน Python เราใช้ **4 ช่องว่าง**

```python
for fruit in fruits:
    print(fruit)          # <- บรรทัดนี้ย่อหน้าอยู่ (มี 4 ช่องว่างข้างหน้า)
```

### ทำไมต้องย่อหน้า?

ลูป `for` ทำโค้ดบางส่วนซ้ำ แต่ Python รู้ได้อย่างไรว่า **บรรทัดไหนบ้าง** ที่ต้องทำซ้ำ และส่วนที่ทำซ้ำจบตรงไหน?

Python ดูจากการย่อหน้า:

- บรรทัดที่ **ย่อหน้า** ต่อจากบรรทัด `for` → อยู่ **ใน** ลูป → ถูกทำซ้ำกับทุกค่า
- บรรทัดแรกที่ **กลับมาไม่ย่อหน้า** → อยู่ **นอก** ลูป → ทำงานแค่ครั้งเดียว หลังจากลูปจบ

บรรทัดที่ย่อหน้าจะอยู่รวมกันเหมือนเป็นกลุ่มเดียวกัน กลุ่มของบรรทัดนี้เรียกว่า **บล็อก (block)** บล็อกของลูป `for` ก็คือ "โค้ดที่ต้องการทำซ้ำ" นั่นเอง

(นี่คือกฎเดียวกับที่เราใช้กับ `if` ในบทที่ 2: บรรทัดที่ย่อหน้าอยู่ใต้ `if` จะทำงานเฉพาะเมื่อเงื่อนไขเป็น True)

### โค้ดเดียวกัน ย่อหน้าต่างกัน ผลลัพธ์ต่างกัน

สองเซลล์ข้างล่างมีบรรทัดเหมือนกันทุกตัวอักษร ความแตกต่าง **อย่างเดียว** คือ `print()` ตัวสุดท้ายย่อหน้าหรือไม่ ลองดูว่าผลลัพธ์เปลี่ยนไปอย่างไร

````python
# แบบ A: print() ตัวที่สองย่อหน้า -> อยู่ "ใน" ลูป
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
    print("Done!")          # ย่อหน้า -> อยู่ในลูป -> แสดง 3 ครั้ง
````

````python
# แบบ B: print() ตัวที่สองไม่ย่อหน้า -> อยู่ "นอก" ลูป
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
print("Done!")              # ไม่ย่อหน้า -> อยู่นอกลูป -> แสดงแค่ครั้งเดียวตอนท้าย
````

### บล็อกหนึ่งมีได้หลายบรรทัด

ทุกบรรทัดที่ย่อหน้าจะทำงานตามลำดับจากบนลงล่าง กับทุกค่า จากนั้นค่าถัดไปก็มา และทั้งบล็อกก็ทำงานใหม่ตั้งแต่บรรทัดบนสุด

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Now the value is:")      # บรรทัดที่ 1 ของบล็อก
    print(fruit)                    # บรรทัดที่ 2 ของบล็อก
    print("----------")             # บรรทัดที่ 3 ของบล็อก

print("The loop is finished.")      # ไม่ย่อหน้า -> อยู่นอกลูป -> ทำงานครั้งเดียว
````

### ทุกบรรทัดในบล็อกต้องย่อหน้าเท่ากัน

ถ้าบรรทัดหนึ่งมี 4 ช่องว่าง แต่บรรทัดถัดไปมี 2 ช่องว่าง Python จะสับสนแล้วแสดง error ให้ใช้ 4 ช่องว่างพอดีเสมอ (ใน Jupyter กดปุ่ม **Tab** ก็จะเติม 4 ช่องว่างให้อัตโนมัติ)

```python
for fruit in fruits:
    print(fruit)       # 4 ช่องว่าง
  print("Done!")       # 2 ช่องว่าง -> IndentationError!
```

## 4. ชื่อของตัวแปรในลูป

ใน `for score in scores:` คำว่า `score` คือ **ตัวแปรใหม่** ที่ลูป for สร้างขึ้น ในแต่ละรอบ ค่าหนึ่งค่าจากลิสต์จะถูกเก็บไว้ในตัวแปรนี้

**ตั้งชื่อตัวแปรนี้ว่าอะไรก็ได้** Python ไม่สนใจชื่อ มันแค่เก็บแต่ละค่าไว้ในกล่องที่มีชื่อนั้น

สามเซลล์ข้างล่างใช้ชื่อต่างกัน (`x`, `score`, `banana`) แต่ผลลัพธ์เหมือนกันทุกประการ

````python
scores = [85, 92, 78]

for x in scores:            # ตัวแปรชื่อ x
    print(x)
````

````python
scores = [85, 92, 78]

for score in scores:        # ตัวแปรชื่อ score
    print(score)
````

````python
scores = [85, 92, 78]

for banana in scores:       # ตัวแปรชื่อ banana (แปลก แต่ก็ใช้ได้!)
    print(banana)
````

ทั้งสามแบบใช้ได้ แต่แบบไหนเข้าใจง่ายที่สุด?

- `x` ไม่ได้บอกว่าค่านั้นคืออะไร
- `banana` ทำให้สับสน เพราะนี่คือคะแนน ไม่ใช่กล้วย
- `score` บอกชัดเจนว่าคืออะไร: **คะแนนหนึ่งค่า** จากลิสต์ของคะแนน

**นิสัยที่ดี:** ตั้งชื่อลิสต์เป็นคำพหูพจน์ (หลายสิ่ง) และตั้งชื่อตัวแปรในลูปเป็นคำ **เอกพจน์** (สิ่งเดียว) ของคำเดียวกัน

| ลิสต์ (หลายค่า) | ตัวแปรในลูป (ค่าเดียว) |
|---|---|
| `scores` | `score` |
| `fruits` | `fruit` |
| `names` | `name` |
| `prices` | `price` |

แบบนี้เวลาอ่านโค้ด จะแยก "ทั้งลิสต์" (`scores`) กับ "ค่าเดียว" (`score`) ออกได้ง่าย

**กฎหนึ่งข้อ:** ชื่อตัวแปรต้อง **ไม่ซ้ำ** กับชื่อลิสต์ เราจะดูว่าถ้าซ้ำกันจะเกิดอะไรขึ้นในหัวข้อที่ 12 (ข้อผิดพลาดที่พบบ่อย)

## 5. การแสดงหลายค่าพร้อมกันด้วยเครื่องหมายจุลภาค

ที่ผ่านมา ส่วนใหญ่เราแสดงทีละค่า: `print(score)`

แต่บางครั้งเราอยากแสดงข้อความกำกับพร้อมกับค่า เช่น `Total: 255` เราทำได้โดยใส่ **หลายค่า** ไว้ใน `print()` แล้วคั่นด้วย **เครื่องหมายจุลภาค** `,`

```python
print(ค่า1, ค่า2, ค่า3)
```

กฎ:
1. ค่าต่าง ๆ จะแสดง **ในบรรทัดเดียว ตามลำดับ**
2. Python **เติมช่องว่างหนึ่งช่องให้อัตโนมัติ** ระหว่างค่า ไม่ต้องเติมเอง
3. ข้อความในเครื่องหมายคำพูด `" "` จะแสดง **ตามที่เขียนไว้**
4. สิ่งที่อยู่นอกเครื่องหมายคำพูดจะถูก **คำนวณก่อน** แล้วจึงแสดงผลลัพธ์ ถ้าเป็นตัวแปร ก็จะแสดงค่าที่อยู่ในตัวแปรนั้น

````python
# สตริงหลายตัว
print("apple", "banana", "cherry")     # มีช่องว่างหนึ่งช่องเติมให้ระหว่างค่าโดยอัตโนมัติ
````

````python
# ตัวเลขหลายตัว
print(10, 20, 30)
````

````python
# สตริงกับตัวเลขด้วยกัน
print("Total:", 255)                   # "Total:" แสดงตามที่เขียน แล้วช่องว่าง แล้ว 255
````

````python
# สตริงกับตัวแปรด้วยกัน
score = 85
print("Score:", score)                 # score เป็นตัวแปร จึงแสดงค่าที่อยู่ข้างใน (85)
````

````python
# สตริงกับการคำนวณด้วยกัน
price = 1000
print("Double:", price * 2)            # price * 2 ถูกคำนวณก่อน (2000) แล้วจึงแสดง
````

````python
# ในเครื่องหมายคำพูด กับ นอกเครื่องหมายคำพูด
price = 1000
print("price * 2")                     # ในเครื่องหมายคำพูด -> แสดงตามที่เขียน: price * 2
print(price * 2)                       # นอกเครื่องหมายคำพูด -> ถูกคำนวณ: 2000
````

### จุลภาค `,` กับ บวก `+`

เราได้เรียนแล้วว่า `+` ต่อสตริงได้ แล้วต่างกันอย่างไร?

| | จุลภาค `,` | บวก `+` |
|---|---|---|
| ช่องว่างระหว่างค่า | เติมให้อัตโนมัติ | ไม่เติมให้ ต้องเติมเอง |
| สตริง + ตัวเลข | ✅ ใช้ได้ | ❌ Error |

`+` ต่อได้แค่ **สตริงกับสตริง** ถ้าพยายามต่อสตริงกับตัวเลขด้วย `+` Python จะแสดง error ส่วนจุลภาคใช้ผสมสตริงกับตัวเลขได้ไม่มีปัญหา

````python
total = 255

print("Total:", total)                 # ✅ จุลภาค: สตริงกับตัวเลขด้วยกัน -> ใช้ได้

# print("Total: " + total)             # ❌ บวก: สตริง + ตัวเลข -> TypeError!
````

## 6. การบวกค่าต่าง ๆ รวมกัน

งานที่ลูป for ถูกใช้บ่อยที่สุดอย่างหนึ่งคือ **บวก** ค่าทั้งหมดในลิสต์ เช่น คะแนนทั้งหมดรวมกันได้เท่าไร?

เคล็ดลับคือใช้ตัวแปรหนึ่งตัวเป็น **ผลรวมสะสม**:

1. **ก่อนลูป:** สร้างตัวแปรและกำหนดค่าเป็น 0 → `total = 0`
2. **ในลูป:** บวกค่าปัจจุบันเข้าไป → `total = total + score`
3. **หลังลูป:** ตัวแปรนั้นเก็บผลรวมของทุกค่า

ดูขั้นที่ 2 ให้ดี: `total = total + score`

- `=` ไม่ได้แปลว่า "เท่ากับ" แต่แปลว่า **"เก็บค่าทางขวาไว้ในกล่องทางซ้าย"**
- ดังนั้น Python จะคำนวณทางขวาก่อน คือ `total + score` (ผลรวมเดิมบวกคะแนนปัจจุบัน)
- แล้วจึงเก็บผลลัพธ์นั้นกลับเข้าไปใน `total`
- เป็นแนวคิดเดียวกับ `age = age + 1`: เอาค่าเดิมมา บวกอะไรบางอย่าง แล้วใส่กลับเข้ากล่องใบเดิม

````python
# บวกคะแนนทั้งหมด
scores = [85, 92, 78, 90, 66]
total = 0                       # ขั้นที่ 1: ผลรวมสะสมเริ่มที่ 0 (ก่อนลูป)

for score in scores:
    total = total + score       # ขั้นที่ 2: ผลรวมเดิม + คะแนนปัจจุบัน -> เก็บกลับเข้า total

print("Total:", total)          # ขั้นที่ 3: หลังลูป (ไม่ย่อหน้า) แสดงผลลัพธ์ครั้งเดียว
````

````python
# มาดูกันว่า total เปลี่ยนไปอย่างไรในแต่ละรอบ
# โค้ดเดียวกับข้างบน แต่มี print() อยู่ "ใน" ลูป (ย่อหน้า)
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total = total + score
    print("score:", score, "total:", total)      # ย่อหน้า -> แสดงทุกรอบ

# รอบที่ 1: 0 + 85 = 85
# รอบที่ 2: 85 + 92 = 177
# รอบที่ 3: 177 + 78 = 255
# รอบที่ 4: 255 + 90 = 345
# รอบที่ 5: 345 + 66 = 411
````

### วิธีเขียนที่สั้นกว่า: `+=`

`total = total + score` ถูกใช้บ่อยมาก จน Python มีวิธีเขียนที่สั้นกว่า:

```python
total += score
```

`+=` หมายถึง **"บวกค่าทางขวาเข้าไปในตัวแปรทางซ้าย แล้วเก็บผลลัพธ์กลับไว้ในตัวแปรนั้น"**

`total += score` กับ `total = total + score` มี **ความหมายเหมือนกันทุกประการ** จะใช้แบบไหนก็ได้ นักเขียนโปรแกรม Python หลายคนนิยมแบบสั้น จึงจะเห็นบ่อย

````python
# แบบยาว
scores = [85, 92, 78]
total = 0
for score in scores:
    total = total + score
print("Long form:", total)

# แบบสั้น
total = 0
for score in scores:
    total += score              # เหมือนกับ: total = total + score
print("Short form:", total)     # ผลลัพธ์เหมือนกัน
````

````python
# ราคารวมของตะกร้าสินค้า
cart = [3500, 1200, 8000, 2300]
total_price = 0                 # ก่อนลูป: เริ่มจาก 0

for price in cart:
    total_price += price        # เหมือนกับ: total_price = total_price + price

print("Total price:", total_price)
````

### ค่าเฉลี่ย

**ค่าเฉลี่ย** คือผลรวมหารด้วยจำนวนค่า

จะรู้จำนวนค่าได้อย่างไร? จำ `len()` ได้ไหม: มันบอกว่าลิสต์มีกี่ค่า `len([85, 92, 78])` คือ `3`

ดังนั้น: **ค่าเฉลี่ย = ผลรวม / len(ลิสต์)**

สำคัญ: คำนวณค่าเฉลี่ย **หลัง** ลูป เมื่อผลรวมครบแล้ว

````python
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total += score              # เหมือนกับ: total = total + score

average = total / len(scores)   # หลังลูป: ผลรวม (411) / จำนวนคะแนน (5)
print("Total:", total)
print("Number of scores:", len(scores))
print("Average:", average)
````

## 7. การใช้ if ในลูป

เราใส่ `if` ไว้ **ใน** ลูป for ได้ แล้วแต่ละค่าจะถูกตรวจทีละค่า และเราทำบางอย่าง **เฉพาะกับค่าที่ต้องการ** ได้

จำวิธีทำงานของ `if`:
- `if เงื่อนไข:` → โค้ดที่ย่อหน้าอยู่ข้างล่างจะทำงาน **เฉพาะเมื่อเงื่อนไขเป็น True**
- `else:` → โค้ดที่ย่อหน้าอยู่ข้างล่างจะทำงานเมื่อเงื่อนไขเป็น False

เมื่อใส่ `if` ไว้ใน `for` จะมีการย่อหน้า **สองระดับ**:

```python
for score in scores:          # ไม่มีช่องว่าง
    if score >= 60:           # 4 ช่องว่าง -> อยู่ใน for (ทำงานกับทุก score)
        print(score)          # 8 ช่องว่าง -> อยู่ใน if (ทำงานเฉพาะเมื่อเงื่อนไขเป็น True)
```

- บรรทัด `if` มี 4 ช่องว่าง เพราะเป็นส่วนหนึ่งของบล็อก `for`
- บรรทัด `print()` มี 8 ช่องว่าง เพราะเป็นส่วนหนึ่งของบล็อก `if` ซึ่งอยู่ในบล็อก `for` อีกที

````python
# แสดงเฉพาะคะแนนที่ผ่าน (60 ขึ้นไป)
scores = [85, 42, 78, 55, 90]

for score in scores:            # ดึงคะแนนออกมาทีละค่า
    if score >= 60:             # 4 ช่องว่าง: ตรวจคะแนนนี้
        print(score)            # 8 ช่องว่าง: แสดงเฉพาะเมื่อได้ 60 ขึ้นไป
````

````python
# มาดูว่าคะแนนไหนถูกตรวจ และคะแนนไหนถูกแสดง
scores = [85, 42, 78, 55, 90]

for score in scores:
    print("Checking", score)    # 4 ช่องว่าง: ทำงานกับ "ทุก" คะแนน
    if score >= 60:
        print("pass")           # 8 ช่องว่าง: ทำงานเฉพาะเมื่อ score >= 60
````

````python
# if - else: แสดง pass หรือ fail ของทุกคะแนน
scores = [85, 42, 78, 55, 90]

for score in scores:
    if score >= 60:
        print(score, "pass")    # ทำงานเมื่อเงื่อนไขเป็น True
    else:
        print(score, "fail")    # ทำงานเมื่อเงื่อนไขเป็น False
````

````python
# แสดงเฉพาะเลขคู่
# ตัวเลขเป็นเลขคู่ถ้าหารด้วย 2 แล้วเหลือเศษ 0
# % ให้ค่าเศษ: 8 % 2 คือ 0 (คู่), 5 % 2 คือ 1 (คี่)
numbers = [3, 8, 5, 12, 7, 6]

for number in numbers:
    if number % 2 == 0:         # เศษเป็น 0 ไหม? -> คู่
        print(number)
````

````python
# if - elif - else ในลูป: ให้เกรดแต่ละคะแนน
# Python ตรวจเงื่อนไขจากบนลงล่าง และทำงานแค่เงื่อนไข "แรก" ที่เป็น True
scores = [95, 82, 74, 58]

for score in scores:
    if score >= 90:
        print(score, "A")
    elif score >= 80:
        print(score, "B")
    elif score >= 70:
        print(score, "C")
    else:
        print(score, "F")
````

````python
# บวกแค่บางค่า: if + ผลรวมสะสม
# บวกเฉพาะเลขคู่
numbers = [3, 8, 5, 12, 7, 6]
even_total = 0                  # ก่อนลูป: เริ่มจาก 0

for number in numbers:
    if number % 2 == 0:         # เฉพาะเมื่อเป็นเลขคู่...
        even_total += number    # ...จึงบวกเข้าไป (เหมือนกับ: even_total = even_total + number)

print("Sum of even numbers:", even_total)    # 8 + 12 + 6 = 26
````

## 8. การนับ

มีนักเรียนสอบผ่านกี่คน? มีคนโหวตให้พิซซ่ากี่คน?

ในการ **นับ** เราใช้แนวคิดเดียวกับผลรวมสะสม ต่างกันแค่อย่างเดียว: แทนที่จะบวกค่า เราบวก **1** ทุกครั้ง

1. **ก่อนลูป:** `count = 0`
2. **ในลูป:** เมื่อเงื่อนไขเป็น True ให้ `count = count + 1` (หรือ `count += 1`)
3. **หลังลูป:** `count` เก็บจำนวนครั้งที่เงื่อนไขเป็น True

````python
# มีนักเรียนสอบผ่าน (60 ขึ้นไป) กี่คน?
scores = [85, 42, 78, 55, 90]
pass_count = 0                  # ก่อนลูป: เริ่มจาก 0

for score in scores:
    if score >= 60:
        pass_count += 1         # เหมือนกับ: pass_count = pass_count + 1

print("Passed:", pass_count)    # 85, 78, 90 -> 3
````

````python
# พิซซ่าได้กี่โหวต?
votes = ["pizza", "pasta", "pizza", "salad", "pizza", "pasta"]
pizza_count = 0

for vote in votes:
    if vote == "pizza":         # == ตรวจว่าค่าสองค่าเหมือนกันหรือไม่
        pizza_count += 1        # เหมือนกับ: pizza_count = pizza_count + 1

print("Pizza:", pizza_count)
````

````python
# นับสองอย่างพร้อมกัน
scores = [85, 42, 78, 55, 90]
pass_count = 0
fail_count = 0

for score in scores:
    if score >= 60:
        pass_count += 1
    else:
        fail_count += 1

print("Passed:", pass_count)
print("Failed:", fail_count)
````

````python
# นับวันที่อุณหภูมิต่ำกว่า 0
temperatures = [3, -2, 5, -7, 0, -1]
below_zero = 0

for temperature in temperatures:
    if temperature < 0:
        below_zero += 1

print("Days below zero:", below_zero)
````

## 9. การหาค่ามากที่สุดและน้อยที่สุด

จะหาคะแนนสูงสุดในลิสต์ได้อย่างไร? ลองคิดดูว่าคนจะทำอย่างไรกับตัวเลขที่เขียนอยู่บนกระดาษ:

1. ดูตัวเลขตัวแรก แล้วจำไว้ว่าเป็น **"ตัวที่มากที่สุดจนถึงตอนนี้"**
2. ดูตัวเลขตัวถัดไป มันมากกว่าตัวที่จำไว้ไหม? ถ้าใช่ ให้จำตัวนี้แทน
3. ทำแบบนี้ไปเรื่อย ๆ จนจบ ตัวเลขที่จำไว้ตอนท้ายคือตัวที่มากที่สุด

เราเขียนขั้นตอนเหล่านี้เป็น Python ได้ตรง ๆ ตัวแปร `highest` คือ "ตัวเลขที่เราจำไว้"

**ทำไมเริ่มจากค่าแรก (`scores[0]`)?** เพราะเราต้องมีอะไรสักอย่างไว้เปรียบเทียบ อินเด็กซ์ 0 คือค่าแรกของลิสต์

````python
# หาคะแนนสูงสุด
scores = [85, 92, 78, 95, 66]
highest = scores[0]             # ขั้นที่ 1: จำคะแนนแรก (85) ไว้ว่าเป็น "ตัวที่มากที่สุดจนถึงตอนนี้"

for score in scores:
    if score > highest:         # ขั้นที่ 2: คะแนนนี้มากกว่าตัวที่จำไว้ไหม?
        highest = score         #           ถ้าใช่ ให้จำตัวนี้แทน

print("Highest:", highest)      # ขั้นที่ 3: ตัวที่จำไว้ตอนท้าย
````

````python
# มาดูทีละขั้น
scores = [85, 92, 78, 95, 66]
highest = scores[0]

for score in scores:
    if score > highest:
        highest = score
    print("score:", score, "highest:", highest)     # 4 ช่องว่าง: ทำงานทุกรอบ
````

````python
# ค่าน้อยที่สุดก็ทำแบบเดียวกัน แค่ใช้ < แทน >
temperatures = [12, 8, 15, 3, 10]
lowest = temperatures[0]        # เริ่มจากอุณหภูมิแรก (12)

for temperature in temperatures:
    if temperature < lowest:    # ค่านี้น้อยกว่าตัวที่จำไว้ไหม?
        lowest = temperature

print("Lowest:", lowest)
````

````python
# สินค้าที่แพงที่สุดในตะกร้า
cart = [3500, 1200, 8000, 2300]
most_expensive = cart[0]

for price in cart:
    if price > most_expensive:
        most_expensive = price

print("Most expensive:", most_expensive)
````

## 10. การสร้างลิสต์ใหม่ด้วย append()

จำ `append()` ได้ไหม: `ลิสต์.append(ค่า)` จะเพิ่มค่าไว้ **ท้าย** ลิสต์

ถ้าใช้ `append()` ในลูป for เราจะสร้าง **ลิสต์ใหม่** จากลิสต์เดิมได้:

1. **ก่อนลูป:** สร้างลิสต์ว่าง → `new_list = []`
2. **ในลูป:** ใช้ `append()` เพิ่มค่าที่ต้องการเข้าไปในลิสต์ใหม่
3. **หลังลูป:** ลิสต์ใหม่ก็เสร็จสมบูรณ์

ลิสต์เดิมไม่เปลี่ยนแปลง

````python
# ลิสต์ใหม่ที่มีแค่คะแนนที่ผ่าน
scores = [85, 42, 78, 55, 90]
passed = []                     # ก่อนลูป: ลิสต์ว่าง

for score in scores:
    if score >= 60:
        passed.append(score)    # เพิ่มคะแนนนี้ไว้ท้าย passed

print("All scores:", scores)    # ลิสต์เดิมไม่เปลี่ยน
print("Passed:", passed)
````

````python
# มาดูลิสต์ใหม่ค่อย ๆ ยาวขึ้น
scores = [85, 42, 78, 55, 90]
passed = []

for score in scores:
    if score >= 60:
        passed.append(score)
    print(passed)               # 4 ช่องว่าง: แสดงลิสต์ใหม่ทุกรอบ
````

````python
# ลิสต์ใหม่ที่ราคาทุกตัวคูณ 2
prices = [1000, 2500, 3000]
doubled = []

for price in prices:
    doubled.append(price * 2)   # price * 2 ถูกคำนวณก่อน แล้วจึงถูกเพิ่มเข้าไป

print(doubled)
````

````python
# ลิสต์ใหม่ที่มีแค่คำยาว ๆ
# len() ใช้กับสตริงได้ด้วย: มันนับตัวอักษร len("cat") คือ 3
words = ["cat", "elephant", "dog", "giraffe"]
long_words = []

for word in words:
    if len(word) > 3:           # มากกว่า 3 ตัวอักษรไหม?
        long_words.append(word)

print(long_words)
````

## 11. การใช้ in และ break ในลูป

### in

`ค่า in ลิสต์` ตรวจว่าค่านั้นอยู่ในลิสต์หรือไม่ ผลลัพธ์เป็น `True` หรือ `False` จึงใช้เป็นเงื่อนไขของ `if` ได้

```python
"milk" in ["bread", "milk"]     # True
"eggs" in ["bread", "milk"]     # False
```

**ระวัง:** `in` ใน `for fruit in fruits:` กับ `in` ใน `if fruit in store:` หน้าตาเหมือนกัน แต่ทำหน้าที่ต่างกัน

| | ความหมาย |
|---|---|
| `for fruit in fruits:` | **ดึง** ค่าออกจาก fruits ทีละค่า |
| `if fruit in store:` | **ตรวจ** ว่า fruit อยู่ใน store หรือไม่ (True / False) |

````python
# ร้านมีของในรายการซื้อของของฉันชิ้นไหนบ้าง?
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]

for item in shopping_list:              # "in" ตัวที่ 1: ดึงของออกมาทีละชิ้น
    if item in store:                   # "in" ตัวที่ 2: ของชิ้นนี้อยู่ใน store ไหม?
        print(item, "yes")
    else:
        print(item, "no")
````

````python
# ทำลิสต์ของที่ยังต้องไปซื้อที่อื่น
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]
not_found = []

for item in shopping_list:
    if item not in store:               # not in: เป็น True เมื่อของชิ้นนั้น "ไม่" อยู่ใน store
        not_found.append(item)

print("Not found:", not_found)
````

### break

`break` **หยุดลูปทันที** ค่าที่เหลือจะไม่ถูกตรวจเลย และ Python จะข้ามไปทำโค้ดที่อยู่ถัดจากลูป

มีประโยชน์เมื่อเจอสิ่งที่ตามหาแล้ว และไม่ต้องหาต่อ `break` แทบจะใช้อยู่ใน `if` เสมอ

````python
# หยุดทันทีที่เจอคะแนนที่ไม่ผ่านตัวแรก
scores = [85, 92, 55, 78, 40]

for score in scores:
    if score < 60:
        print("Found a failing score:", score)
        break                           # หยุดลูปตรงนี้เลย
    print(score, "is OK")

print("The loop is over.")              # 78 กับ 40 ไม่เคยถูกตรวจ
````

````python
# ในลิสต์มีคนชื่อ Tom ไหม?
names = ["Bill", "Anna", "Tom", "Kate", "Mike"]

for name in names:
    print("Checking", name)
    if name == "Tom":
        print("Found Tom!")
        break                           # ไม่ต้องตรวจ Kate กับ Mike
````

## 12. ข้อผิดพลาดที่พบบ่อย

นี่คือข้อผิดพลาดที่เกือบทุกคนเคยทำตอนใช้ลูป for กับลิสต์ ในแต่ละข้อเราจะดู:

- ❌ **โค้ดที่ผิด**

โค้ดที่ผิดบางตัว **ไม่มี error แต่ได้ผลลัพธ์ผิด** ข้อผิดพลาดแบบนี้อันตรายกว่า เพราะ Python ไม่เตือนเราเลย

### ข้อผิดพลาด 1: ลืมโคลอน

````python
# ❌ ผิด
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits           # SyntaxError
#     print(fruit)
````

🔍 **ผิดตรงไหน?**

- บรรทัด `for` ต้องลงท้ายด้วยโคลอน `:` เสมอ
- โคลอนบอก Python ว่า: "โค้ดที่ต้องทำซ้ำเริ่มที่บรรทัดถัดไป"
- ถ้าไม่มี Python จะไม่เข้าใจโค้ด และแสดง `SyntaxError` ("ไวยากรณ์ของโค้ดผิด")

✅ **วิธีที่ถูกต้อง:** ใส่ `:` ไว้ท้ายบรรทัด `for`

````python
# ✅ ถูกต้อง
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # มีโคลอนท้ายบรรทัด
    print(fruit)
````

### ข้อผิดพลาด 2: ลืมย่อหน้า

````python
# ❌ ผิด
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits:
# print(fruit)                  # IndentationError
````

🔍 **ผิดตรงไหน?**

- หลังบรรทัด `for` ต้องมีบรรทัดที่ **ย่อหน้า** อย่างน้อยหนึ่งบรรทัด นั่นคือโค้ดที่ต้องทำซ้ำ
- ตรงนี้ `print(fruit)` ไม่มีช่องว่างข้างหน้า Python จึงคิดว่าลูป `for` ไม่มีอะไรให้ทำซ้ำ
- error คือ `IndentationError` ("ย่อหน้าผิด")

✅ **วิธีที่ถูกต้อง:** ย่อหน้าโค้ดที่ต้องทำซ้ำ 4 ช่องว่าง (หรือกด Tab หนึ่งครั้ง)

````python
# ✅ ถูกต้อง
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # มี 4 ช่องว่างข้างหน้า
````

### ข้อผิดพลาด 3: ย่อหน้าบรรทัดที่ควรทำงานแค่ครั้งเดียว

````python
# ❌ ผิด: เราอยากแสดงผลรวมแค่ครั้งเดียวตอนท้าย
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score
    print("Total:", total)      # ย่อหน้า -> อยู่ในลูป -> แสดง 3 ครั้ง!
````

🔍 **ผิดตรงไหน?**

- `print("Total:", total)` ย่อหน้าอยู่ จึงอยู่ **ใน** ลูป
- แปลว่ามันทำงานทุกรอบ และแสดงผลรวมในขณะที่ยังบวกไม่เสร็จ
- ไม่มี error แต่ได้ 3 บรรทัดแทนที่จะเป็น 1 บรรทัด และสองบรรทัดแรกก็ไม่ใช่ผลรวมจริง

✅ **วิธีที่ถูกต้อง:** เอาการย่อหน้าออกจากบรรทัดที่ควรทำงานครั้งเดียว แล้วมันจะอยู่ **นอก** ลูป และทำงานหลังลูปจบเท่านั้น

````python
# ✅ ถูกต้อง
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score              # ในลูป: บวกสะสม
print("Total:", total)          # ไม่ย่อหน้า -> อยู่นอกลูป -> แสดงครั้งเดียว
````

### ข้อผิดพลาด 4: ตั้งชื่อลิสต์กับตัวแปรในลูปซ้ำกัน

````python
# ❌ ผิด
fruits = ["apple", "banana", "cherry"]

for fruits in fruits:           # ตัวแปรในลูปมีชื่อ "ซ้ำ" กับลิสต์
    print(fruits)               # ตรงนี้แสดงถูกต้อง...

print(fruits)                   # ...แต่ตอนนี้ fruits เหลือแค่ "cherry" ลิสต์หายไปแล้ว!
````

🔍 **ผิดตรงไหน?**

- ลูปดูเหมือนทำงานปกติ จึงสังเกตได้ยาก
- แต่ทุกรอบจะเก็บค่าลงในกล่องชื่อ `fruits` จำไว้ว่า: เมื่อเก็บค่าใหม่ลงในตัวแปร ค่าเก่าจะหายไป
- ดังนั้นลิสต์ที่อยู่ใน `fruits` จะถูกแทนที่ด้วย `"apple"` แล้ว `"banana"` แล้ว `"cherry"`
- หลังลูป `fruits` เหลือแค่ `"cherry"` ลิสต์หายไปแล้ว

✅ **วิธีที่ถูกต้อง:** ตั้งชื่อตัวแปรในลูปให้ **ต่าง** จากชื่อลิสต์ รูปเอกพจน์ของชื่อลิสต์เป็นตัวเลือกที่ดีที่สุด: `fruits` → `fruit`

````python
# ✅ ถูกต้อง
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # fruit (ค่าเดียว) กับ fruits (ลิสต์) เป็นคนละชื่อ
    print(fruit)

print(fruits)                   # ลิสต์ยังอยู่ครบ
````

### ข้อผิดพลาด 7: กำหนดผลรวมสะสมเป็น 0 ในลูป

````python
# ❌ ผิด
scores = [85, 92, 78]

for score in scores:
    total = 0                   # ย่อหน้า -> อยู่ในลูป -> ทำงานทุกรอบ!
    total += score

print("Total:", total)          # 78 ไม่ใช่ 255
````

🔍 **ผิดตรงไหน?**

- `total = 0` ย่อหน้าอยู่ จึงอยู่ในลูป มันทำงาน **3 ครั้ง**
- ทุกรอบ ผลรวมจะกลับไปเป็น 0 แล้วบวกแค่คะแนนปัจจุบัน
- รอบที่ 1: 0 + 85 = 85 → รอบที่ 2: **กลับเป็น 0**, 0 + 92 = 92 → รอบที่ 3: **กลับเป็น 0**, 0 + 78 = 78
- ตอนท้าย `total` จึงเก็บแค่คะแนนสุดท้าย

✅ **วิธีที่ถูกต้อง:** เขียน `total = 0` **ครั้งเดียว ก่อน** ลูป (ไม่ย่อหน้า)

````python
# ✅ ถูกต้อง
scores = [85, 92, 78]
total = 0                       # ก่อนลูป -> ทำงานครั้งเดียว

for score in scores:
    total += score              # เหมือนกับ: total = total + score

print("Total:", total)          # 85 + 92 + 78 = 255
````

### ข้อผิดพลาด 8: คำนวณค่าเฉลี่ยในลูป

````python
# ❌ ผิด
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score
    average = total / len(scores)       # ย่อหน้า -> ถูกคำนวณทุกรอบ
    print("Average:", average)
````

🔍 **ผิดตรงไหน?**

- การคำนวณค่าเฉลี่ยต้องใช้ผลรวมของ **ทุก** ค่า
- แต่ในลูป ผลรวมยังไม่ครบ
- รอบที่ 1: 80 / 3 → รอบที่ 2: 170 / 3 → รอบที่ 3: 270 / 3 มีแค่ค่าสุดท้ายที่เป็นค่าเฉลี่ยจริง

✅ **วิธีที่ถูกต้อง:** ในลูป ให้บวกสะสมอย่างเดียว แล้วคำนวณค่าเฉลี่ย **หลัง** ลูป เมื่อผลรวมครบแล้ว

````python
# ✅ ถูกต้อง
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score                      # ในลูป: บวกสะสมอย่างเดียว

average = total / len(scores)           # หลังลูป: 270 / 3
print("Average:", average)
````

### ข้อผิดพลาด 9: พยายามเปลี่ยนลิสต์ด้วยการเปลี่ยนตัวแปรในลูป

````python
# ❌ ผิด: เราอยากบวกคะแนนพิเศษ 5 คะแนนให้ทุกคะแนนในลิสต์
scores = [85, 92, 78]

for score in scores:
    score = score + 5           # ดูเหมือนจะเปลี่ยนคะแนน...

print(scores)                   # ...แต่ลิสต์ยังเป็น [85, 92, 78]!
````

🔍 **ผิดตรงไหน?**

- `score` เป็น **กล่องแยกต่างหาก** ทุกรอบ ค่าหนึ่งค่าจะถูก **คัดลอก** จากลิสต์มาไว้ในกล่องนี้
- `score = score + 5` เปลี่ยนแค่สิ่งที่อยู่ในกล่อง `score` ลิสต์ `scores` ไม่ถูกแตะต้องเลย
- เหมือนลอกตัวเลขจากสมุดลงกระดาษโน้ต แล้วแก้ตัวเลขบนกระดาษโน้ต ตัวเลขในสมุดยังเหมือนเดิม
- ไม่มี error จึงสังเกตได้ยาก

✅ **วิธีที่ถูกต้อง:** สร้าง **ลิสต์ใหม่** ด้วย `append()` (หัวข้อที่ 10) แล้วใส่ค่าที่เปลี่ยนแล้วลงไป

````python
# ✅ ถูกต้อง
scores = [85, 92, 78]
new_scores = []                         # ลิสต์ว่างสำหรับค่าที่เปลี่ยนแล้ว

for score in scores:
    new_scores.append(score + 5)        # เพิ่มค่าที่เปลี่ยนแล้วเข้าไปในลิสต์ใหม่

print(new_scores)                       # [90, 97, 83]
````

### ข้อผิดพลาด 10: ลบค่าออกจากลิสต์ระหว่างลูป

````python
# ❌ ผิด: เราอยากลบคะแนนที่ไม่ผ่าน (ต่ำกว่า 60) ทั้งหมด
scores = [50, 40, 80, 30]

for score in scores:
    if score < 60:
        scores.remove(score)

print(scores)                   # [40, 80] -> 40 ยังอยู่!
````

🔍 **ผิดตรงไหน?**

ลูป for ไล่ดูลิสต์ **ตามตำแหน่ง**: ค่าที่ 1 แล้วค่าที่ 2 แล้วค่าที่ 3 ...
แต่ `remove()` เอาค่าหนึ่งออก แล้วค่าทุกตัวที่อยู่ถัดไปจะ **เลื่อนขึ้นมา** หนึ่งตำแหน่ง

| รอบที่ | ตำแหน่งที่ลูปดูอยู่ | ลิสต์ในตอนนั้น | เกิดอะไรขึ้น |
|---|---|---|---|
| 1 | ตำแหน่งที่ 1 | `[50, 40, 80, 30]` | 50 < 60 → ลบ 50 ลิสต์กลายเป็น `[40, 80, 30]` |
| 2 | ตำแหน่งที่ 2 | `[40, 80, 30]` | ค่าที่ 2 ตอนนี้คือ **80** 40 เลื่อนไปอยู่ตำแหน่งที่ 1 แล้ว ลูปจึง **ข้าม** มันไป! |
| 3 | ตำแหน่งที่ 3 | `[40, 80, 30]` | 30 < 60 → ลบ 30 ลิสต์กลายเป็น `[40, 80]` |

- 40 ไม่เคยถูกตรวจ จึงยังอยู่ในลิสต์
- ไม่มี error จึงสังเกตได้ยากมาก

✅ **วิธีที่ถูกต้อง:** อย่าเปลี่ยนลิสต์ในขณะที่กำลังวนลูปกับมัน ให้สร้าง **ลิสต์ใหม่** ที่มีแค่ค่าที่ต้องการเก็บไว้แทน (หัวข้อที่ 10)

````python
# ✅ ถูกต้อง
scores = [50, 40, 80, 30]
kept = []

for score in scores:
    if score >= 60:             # เก็บไว้เฉพาะคะแนนที่ผ่าน
        kept.append(score)

print(kept)                     # [80]
````

### ข้อผิดพลาด 12: เริ่ม "ตัวที่มากที่สุดจนถึงตอนนี้" ที่ 0

````python
# ❌ ผิด: หาอุณหภูมิสูงสุดในฤดูหนาว
temperatures = [-5, -2, -8, -3]
highest = 0                     # เริ่มที่ 0...

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # 0 -> แต่ 0 ไม่ได้อยู่ในลิสต์ด้วยซ้ำ!
````

🔍 **ผิดตรงไหน?**

- อุณหภูมิทุกค่าต่ำกว่า 0 จึงไม่มีค่าไหนมากกว่า 0
- `temperature > highest` ไม่เคยเป็น True และ `highest` ก็ยังเป็น 0 จนจบ
- ง่ายที่จะคิดว่า "ควรเริ่มที่ 0" เพราะผลรวมสะสมเริ่มที่ 0 แต่การหาค่ามากที่สุดเป็นคนละงานกัน

✅ **วิธีที่ถูกต้อง:** เริ่มจาก **ค่าแรกของลิสต์** (`temperatures[0]`) แบบนี้ค่าเริ่มต้นจะมาจากลิสต์เองเสมอ

````python
# ✅ ถูกต้อง
temperatures = [-5, -2, -8, -3]
highest = temperatures[0]       # เริ่มจากค่าแรก (-5)

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # -2
````
