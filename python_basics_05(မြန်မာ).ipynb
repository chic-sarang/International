## 1. list နဲ့ for loop ကို ဘာလို့ သုံးတာလဲ?

သင်ခန်းစာ 4 မှာ list က တန်ဖိုး အများကြီးကို သိမ်းထားနိုင်ပြီး တန်ဖိုး တစ်ခုကို သူ့ရဲ့ **index** (နေရာ နံပါတ်) နဲ့ ယူလို့ရတယ်ဆိုတာ လေ့လာခဲ့ပါတယ်။ ပထမ တန်ဖိုးက index 0 မှာ၊ ဒုတိယ တန်ဖိုးက index 1 မှာ စသဖြင့် ရှိပါတယ်။

ဒါကြောင့် list ထဲက အမှတ်တိုင်းကို ပြချင်ရင် index တစ်ခုစီအတွက် စာကြောင်း တစ်ကြောင်းစီ ရေးလို့ရပါတယ်:

```python
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
```

ဒါက အလုပ်လုပ်ပါတယ်။ ဒါပေမဲ့ ပြဿနာ ၂ ခု ရှိပါတယ်:

- list မှာ တန်ဖိုး 100 ခု ရှိရင် စာကြောင်း 100 ကြောင်း ရေးရပါမယ်။
- `append()` နဲ့ တန်ဖိုးအသစ် ထည့်ရင် `print()` စာကြောင်း နောက်တစ်ကြောင်းကိုလည်း ကိုယ်တိုင် ထပ်ထည့်ရပါမယ်။

**for loop** က ဒါကို ဖြေရှင်းပေးပါတယ်။ for loop က list ထဲက တန်ဖိုးတွေကို **ပထမကနေ နောက်ဆုံးအထိ တစ်ခုချင်းစီ** ထုတ်ယူပြီး တန်ဖိုး တစ်ခုစီအတွက် code တစ်ခုတည်းကို run ပေးပါတယ်။ list က ဘယ်လောက်ရှည်ရှည် ကိစ္စမရှိပါဘူး။ code ကို တစ်ခါပဲ ရေးဖို့ လိုပါတယ်။

````python
# ရှည်တဲ့နည်း: တန်ဖိုး တစ်ခုစီအတွက် စာကြောင်း တစ်ကြောင်းစီ
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
````

````python
# တိုတဲ့နည်း: for loop
# ရလဒ်က အတိအကျ တူပေမဲ့ စာကြောင်း 2 ကြောင်းပဲ ရှိပါတယ်
scores = [85, 92, 78, 90, 66]
for score in scores:
    print(score)
````

````python
# list ရှည်လာရင်တောင် for loop က အတူတူပါပဲ
# print() စာကြောင်း တစ်ကြောင်းမှ ထပ်ထည့်စရာ မလိုပါဘူး
scores = [85, 92, 78, 90, 66]
scores.append(100)      # နောက်ဆုံးမှာ တန်ဖိုး တစ်ခု ထပ်ထည့်ပါတယ် (အခု တန်ဖိုး 6 ခု)
scores.append(73)       # နောက်ထပ် တစ်ခု (အခု တန်ဖိုး 7 ခု)

for score in scores:    # အပေါ်က စာကြောင်း 2 ကြောင်းနဲ့ အတူတူ
    print(score)
````

## 2. အခြေခံပုံစံ: `for တန်ဖိုး in list`

code ရဲ့ ပုံစံက ဒီလိုပါ:

```python
for score in scores:
    ထပ်ခါထပ်ခါ run မယ့် code
```

ဝါကျ တစ်ကြောင်းလို ဖတ်လို့ရပါတယ်: **"scores ထဲက score တစ်ခုစီအတွက် အောက်က code ကို run ပါ"**။

အစိတ်အပိုင်း တစ်ခုချင်းစီကို ကြည့်ရအောင်:

| အစိတ်အပိုင်း | အဓိပ္ပာယ် |
|---|---|
| `for` | "တန်ဖိုး တစ်ခုစီအတွက် အောက်ပါအတိုင်း ထပ်လုပ်ပါ" |
| `score` | variable တစ်ခု။ အကြိမ်တိုင်း list ထဲက တန်ဖိုး တစ်ခုကို ဒီမှာ သိမ်းပါတယ် |
| `in` | "ထဲမှာ" |
| `scores` | ကျွန်တော်တို့ ဖြတ်သွားမယ့် list |
| `:` | colon။ `for` စာကြောင်းရဲ့ အဆုံးမှာ မဖြစ်မနေ ရှိရပါမယ် |
| `    ထပ်ခါထပ်ခါ run မယ့် code` | တန်ဖိုး တစ်ခုစီအတွက် run မယ့် code။ indent (space 4 ခု) လုပ်ရပါမယ် |

**အလုပ်လုပ်ပုံ၊ အဆင့်ဆင့်** (`scores = [85, 92, 78]` ဖြစ်တဲ့အခါ):

| အကြိမ် | `score` ထဲမှာ ဘာသိမ်းလဲ | ဘာဖြစ်လဲ |
|---|---|---|
| 1 ကြိမ်မြောက် | `85` (index 0 က တန်ဖိုး) | `score` = 85 နဲ့ indent လုပ်ထားတဲ့ code run ပါတယ် |
| 2 ကြိမ်မြောက် | `92` (index 1 က တန်ဖိုး) | `score` = 92 နဲ့ indent လုပ်ထားတဲ့ code run ပါတယ် |
| 3 ကြိမ်မြောက် | `78` (index 2 က တန်ဖိုး) | `score` = 78 နဲ့ indent လုပ်ထားတဲ့ code run ပါတယ် |
| - | တန်ဖိုး မကျန်တော့ပါ | loop ဆုံးသွားပြီး Python က loop နောက်က code ကို ဆက်သွားပါတယ် |

ဒါကြောင့် loop က **list ထဲမှာ တန်ဖိုး ရှိသလောက်** run ပါတယ်။ တန်ဖိုး 3 ခုရှိတဲ့ list → 3 ကြိမ်။ တန်ဖိုး 100 ခုရှိတဲ့ list → 100 ကြိမ်။

````python
# သစ်သီး တစ်မျိုးချင်းစီကို ပြခြင်း
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:        # 1 ကြိမ်မြောက်: fruit = "apple"၊ 2 ကြိမ်မြောက်: fruit = "banana"၊ 3 ကြိမ်မြောက်: fruit = "cherry"
    print(fruit)            # အခု fruit ထဲမှာ ရှိတဲ့ တန်ဖိုးကို ပြပါတယ်
````

````python
# ဂဏန်းတွေရဲ့ list
numbers = [10, 20, 30, 40]

for number in numbers:      # number က 10၊ ပြီးတော့ 20၊ ပြီးတော့ 30၊ ပြီးတော့ 40 ဖြစ်ပါတယ်
    print(number)
````

````python
# bool တန်ဖိုးတွေရဲ့ list (True / False)
answers = [True, False, True]

for answer in answers:      # answer က True၊ ပြီးတော့ False၊ ပြီးတော့ True ဖြစ်ပါတယ်
    print(answer)
````

````python
# တန်ဖိုး တစ်ခုစီနဲ့ တွက်ချက်လို့ ရပါတယ်
# စျေးနှုန်း တစ်ခုစီကို 2 နဲ့ မြှောက်ပြီး ပြပါတယ်
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)        # 1000 * 2၊ ပြီးတော့ 2500 * 2၊ ပြီးတော့ 3000 * 2
````

````python
# list ထဲက တန်ဖိုးတွေ "မပြောင်းပါဘူး"
# အပေါ်က cell မှာ price * 2 ကို ပြခဲ့ပေမဲ့ list ကိုယ်တိုင်က အတူတူပဲ ရှိပါတယ်
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)

print(prices)               # [1000, 2500, 3000] အတိုင်းပဲ
````

````python
# + နဲ့ string တွေကို ဆက်ခြင်း
# မှတ်ထားပါ: + က string နှစ်ခုကို တစ်ခုတည်း ဆက်ပေးနိုင်ပါတယ် ("Hello, " + "Bill" -> "Hello, Bill")
names = ["Bill", "Anna", "Tom"]

for name in names:
    print("Hello, " + name)
````

````python
# list က ဗလာ ဖြစ်ရင် loop က 0 ကြိမ် run ပါတယ်
# ဘာမှ မပြပါဘူး၊ error လည်း မရှိပါဘူး
empty_list = []

for value in empty_list:
    print("This never prints.")

print("The loop ran 0 times.")
````

## 3. indent: ဘယ်စာကြောင်းတွေ ထပ်ခါထပ်ခါ run လဲ?

**indent** ဆိုတာ စာကြောင်းရဲ့ အစမှာ space တွေ ထည့်ပြီး ညာဘက်ကို တွန်းတာ ဖြစ်ပါတယ်။ Python မှာ **space 4 ခု** သုံးပါတယ်။

```python
for fruit in fruits:
    print(fruit)          # <- ဒီစာကြောင်းကို indent လုပ်ထားပါတယ် (အစမှာ space 4 ခု)
```

### indent ဘာလို့ လိုတာလဲ?

`for` loop က code တချို့ကို ထပ်ခါထပ်ခါ run ပါတယ်။ ဒါပေမဲ့ **ဘယ်စာကြောင်းတွေကို** ထပ်ခါထပ်ခါ run ရမလဲ၊ ထပ်ခါထပ်ခါ run မယ့် အပိုင်း ဘယ်မှာ ဆုံးလဲဆိုတာ Python က ဘယ်လို သိလဲ?

Python က indent ကို ကြည့်ပါတယ်:

- `for` စာကြောင်း နောက်မှာ တန်းရှိတဲ့ **indent လုပ်ထားတဲ့** စာကြောင်းတွေ → loop **ထဲမှာ** → တန်ဖိုး တစ်ခုစီအတွက် ထပ်ခါထပ်ခါ run ပါတယ်
- **indent မလုပ်တော့တဲ့** ပထမဆုံး စာကြောင်း → loop **အပြင်မှာ** → loop ပြီးမှ တစ်ကြိမ်ပဲ run ပါတယ်

indent လုပ်ထားတဲ့ စာကြောင်းတွေက အုပ်စု တစ်ခုလို အတူတကွ ရှိပါတယ်။ ဒီလို စာကြောင်းအုပ်စုကို **block** လို့ ခေါ်ပါတယ်။ `for` loop ရဲ့ block က "ထပ်ခါထပ်ခါ run မယ့် code" ပါပဲ။

(ဒါက သင်ခန်းစာ 2 မှာ `if` အတွက် သုံးခဲ့တဲ့ စည်းမျဉ်းနဲ့ အတူတူပါပဲ: `if` အောက်က indent လုပ်ထားတဲ့ စာကြောင်းတွေက condition True ဖြစ်မှသာ run ပါတယ်။)

### code အတူတူ၊ indent မတူ၊ ရလဒ် မတူ

အောက်က cell နှစ်ခုမှာ စာကြောင်းတွေ အတိအကျ တူပါတယ်။ **တစ်ခုတည်းသော** ကွာခြားချက်က နောက်ဆုံး `print()` ကို indent လုပ်ထားလား မလုပ်ထားလား ဖြစ်ပါတယ်။ ရလဒ် ဘယ်လို ပြောင်းလဲလဲ ကြည့်ပါ။

````python
# ပုံစံ A: ဒုတိယ print() ကို indent လုပ်ထားပါတယ် -> loop "ထဲမှာ"
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
    print("Done!")          # indent လုပ်ထား -> loop ထဲမှာ -> 3 ကြိမ် ပြပါတယ်
````

````python
# ပုံစံ B: ဒုတိယ print() ကို indent မလုပ်ထားပါ -> loop "အပြင်မှာ"
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
print("Done!")              # indent မလုပ်ထား -> loop အပြင်မှာ -> နောက်ဆုံးမှာ တစ်ကြိမ်ပဲ ပြပါတယ်
````

### block တစ်ခုမှာ စာကြောင်း အများကြီး ရှိလို့ရပါတယ်

indent လုပ်ထားတဲ့ စာကြောင်းတိုင်းက တန်ဖိုး တစ်ခုစီအတွက် အပေါ်ကနေ အောက်ကို အစဉ်လိုက် run ပါတယ်။ ပြီးရင် နောက်တန်ဖိုး ရောက်လာပြီး block တစ်ခုလုံးက အပေါ်ကနေ ပြန် run ပါတယ်။

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Now the value is:")      # block ရဲ့ 1 ကြောင်းမြောက်
    print(fruit)                    # block ရဲ့ 2 ကြောင်းမြောက်
    print("----------")             # block ရဲ့ 3 ကြောင်းမြောက်

print("The loop is finished.")      # indent မလုပ်ထား -> loop အပြင်မှာ -> တစ်ကြိမ် run ပါတယ်
````

### block တစ်ခုထဲက စာကြောင်းအားလုံး indent တူရပါမယ်

စာကြောင်း တစ်ကြောင်းမှာ space 4 ခု ရှိပြီး နောက်တစ်ကြောင်းမှာ space 2 ခု ရှိရင် Python က ရှုပ်ထွေးသွားပြီး error ပြပါတယ်။ အမြဲ space 4 ခု အတိအကျ သုံးပါ။ (Jupyter မှာ **Tab** ခလုတ်ကို နှိပ်ရင် space 4 ခုကို အလိုအလျောက် ထည့်ပေးပါတယ်။)

```python
for fruit in fruits:
    print(fruit)       # space 4 ခု
  print("Done!")       # space 2 ခု -> IndentationError!
```

## 4. loop variable ရဲ့ နာမည်

`for score in scores:` မှာ `score` ဆိုတဲ့ စကားလုံးက for loop ဖန်တီးတဲ့ **variable အသစ်** ဖြစ်ပါတယ်။ အကြိမ်တိုင်း list ထဲက တန်ဖိုး တစ်ခုကို အဲဒီထဲ သိမ်းပါတယ်။

**ဒီ variable ကို ကြိုက်တဲ့ နာမည် ပေးလို့ရပါတယ်။** Python က နာမည်ကို ဂရုမစိုက်ပါဘူး။ တန်ဖိုး တစ်ခုစီကို အဲဒီနာမည်ရှိတဲ့ သေတ္တာထဲ သိမ်းရုံပါပဲ။

အောက်က cell သုံးခုမှာ နာမည် မတူတာတွေ (`x`, `score`, `banana`) သုံးထားပေမဲ့ ရလဒ်က အတိအကျ တူပါတယ်။

````python
scores = [85, 92, 78]

for x in scores:            # variable နာမည်က x
    print(x)
````

````python
scores = [85, 92, 78]

for score in scores:        # variable နာမည်က score
    print(score)
````

````python
scores = [85, 92, 78]

for banana in scores:       # variable နာမည်က banana (ထူးဆန်းပေမဲ့ အလုပ်လုပ်ပါတယ်!)
    print(banana)
````

သုံးခုလုံး အလုပ်လုပ်ပါတယ်။ ဒါပေမဲ့ ဘယ်ဟာက နားလည်ရ အလွယ်ဆုံးလဲ?

- `x` က တန်ဖိုးဟာ ဘာလဲ မပြောပါဘူး။
- `banana` က ရှုပ်ထွေးစေပါတယ်၊ ဒါတွေက ငှက်ပျောသီး မဟုတ်ဘဲ အမှတ်တွေ ဖြစ်လို့ပါ။
- `score` က ဘာလဲဆိုတာ တိတိကျကျ ပြောပါတယ်: အမှတ်တွေရဲ့ list ထဲက **အမှတ် တစ်ခု**။

**အလေ့အထ ကောင်း:** list ကို အများကိန်း စကားလုံးနဲ့ နာမည်ပေးပြီး loop variable ကို အဲဒီ စကားလုံးရဲ့ **ဧကဝုစ်** (တစ်ခုတည်း) ပုံစံနဲ့ နာမည်ပေးပါ။

| list (တန်ဖိုး အများကြီး) | loop variable (တန်ဖိုး တစ်ခု) |
|---|---|
| `scores` | `score` |
| `fruits` | `fruit` |
| `names` | `name` |
| `prices` | `price` |

ဒါဆိုရင် code ဖတ်တဲ့အခါ "list တစ်ခုလုံး" (`scores`) နဲ့ "တန်ဖိုး တစ်ခု" (`score`) ကို လွယ်လွယ် ခွဲခြားနိုင်ပါတယ်။

**စည်းမျဉ်း တစ်ခု:** variable နာမည်က list နာမည်နဲ့ **မတူရပါဘူး**။ တူသွားရင် ဘာဖြစ်လဲဆိုတာ အပိုင်း 12 (မကြာခဏ ဖြစ်တတ်သော အမှားများ) မှာ ကြည့်ပါမယ်။

## 5. comma နဲ့ တန်ဖိုး အများကြီးကို တစ်ခါတည်း ပြခြင်း

အခုအထိ များသောအားဖြင့် တစ်ခါမှာ တန်ဖိုး တစ်ခုပဲ ပြခဲ့ပါတယ်: `print(score)`။

ဒါပေမဲ့ တစ်ခါတလေ `Total: 255` လို တန်ဖိုးနဲ့အတူ ခေါင်းစဉ်စာသား တစ်ခုကိုပါ ပြချင်ပါတယ်။ ဒါအတွက် `print()` ထဲမှာ **တန်ဖိုး အများကြီး** ကို **comma** `,` နဲ့ ခွဲပြီး ထည့်လို့ရပါတယ်။

## 6. တန်ဖိုးတွေကို ပေါင်းခြင်း

for loop ရဲ့ အသုံးအများဆုံး အလုပ်တစ်ခုက list ထဲက တန်ဖိုးအားလုံးကို **ပေါင်းခြင်း** ဖြစ်ပါတယ်။ ဥပမာ: အမှတ်အားလုံးရဲ့ စုစုပေါင်း ဘယ်လောက်လဲ?

နည်းလမ်းက variable တစ်ခုကို **ပေါင်းစုတန်ဖိုး** အဖြစ် သုံးတာပါ:

1. **loop မစခင်:** variable တစ်ခု ဖန်တီးပြီး 0 ထည့်ပါ။ → `total = 0`
2. **loop ထဲမှာ:** လက်ရှိ တန်ဖိုးကို အဲဒီထဲ ပေါင်းထည့်ပါ။ → `total = total + score`
3. **loop ပြီးရင်:** အဲဒီ variable ထဲမှာ တန်ဖိုးအားလုံးရဲ့ ပေါင်းလဒ် ရှိပါတယ်။

အဆင့် 2 ကို သေချာ ကြည့်ပါ: `total = total + score`

- `=` က "ညီတယ်" လို့ အဓိပ္ပာယ် မရပါဘူး။ **"ညာဘက်က တန်ဖိုးကို ဘယ်ဘက်က သေတ္တာထဲ သိမ်းပါ"** လို့ အဓိပ္ပာယ်ရပါတယ်။
- ဒါကြောင့် Python က ညာဘက် `total + score` (စုစုပေါင်းဟောင်း နဲ့ လက်ရှိ အမှတ်) ကို အရင်တွက်ပြီး၊
- အဲဒီရလဒ်ကို `total` ထဲ ပြန်သိမ်းပါတယ်။
- ဒါက `age = age + 1` နဲ့ စိတ်ကူး အတူတူပါပဲ: တန်ဖိုးဟောင်းကို ယူ၊ တစ်ခုခု ပေါင်းပြီး သေတ္တာ အဟောင်းထဲ ပြန်ထည့်ပါတယ်။

````python
# အမှတ်အားလုံးကို ပေါင်းခြင်း
scores = [85, 92, 78, 90, 66]
total = 0                       # အဆင့် 1: ပေါင်းစုတန်ဖိုးက 0 ကနေ စပါတယ် (loop မစခင်)

for score in scores:
    total = total + score       # အဆင့် 2: စုစုပေါင်းဟောင်း + လက်ရှိ အမှတ် -> total ထဲ ပြန်သိမ်းပါတယ်

print("Total:", total)          # အဆင့် 3: loop ပြီးမှ (indent မလုပ်ထား) ရလဒ်ကို တစ်ကြိမ် ပြပါတယ်
````

````python
# အကြိမ်တိုင်း total ဘယ်လို ပြောင်းလဲလဲ ကြည့်ရအောင်
# အပေါ်က code နဲ့ အတူတူ၊ ဒါပေမဲ့ loop "ထဲမှာ" (indent လုပ်ပြီး) print() ထည့်ထားပါတယ်
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total = total + score
    print("score:", score, "total:", total)      # indent လုပ်ထား -> အကြိမ်တိုင်း ပြပါတယ်

# 1 ကြိမ်မြောက်: 0 + 85 = 85
# 2 ကြိမ်မြောက်: 85 + 92 = 177
# 3 ကြိမ်မြောက်: 177 + 78 = 255
# 4 ကြိမ်မြောက်: 255 + 90 = 345
# 5 ကြိမ်မြောက်: 345 + 66 = 411
````

### ပိုတိုတဲ့ ရေးနည်း: `+=`

`total = total + score` ကို အရမ်းအသုံးများလို့ Python မှာ ပိုတိုတဲ့ ရေးနည်း ရှိပါတယ်:

```python
total += score
```

`+=` က **"ညာဘက်ကို ဘယ်ဘက်က variable ထဲ ပေါင်းပြီး ရလဒ်ကို အဲဒီ variable ထဲ ပြန်သိမ်းပါ"** လို့ အဓိပ္ပာယ်ရပါတယ်။

`total += score` နဲ့ `total = total + score` က **အဓိပ္ပာယ် လုံးဝ တူပါတယ်**။ ကြိုက်တာကို သုံးလို့ရပါတယ်။ Python programmer အများစုက အတိုပုံစံကို သုံးကြလို့ မကြာခဏ တွေ့ရပါမယ်။

````python
# အရှည်ပုံစံ
scores = [85, 92, 78]
total = 0
for score in scores:
    total = total + score
print("Long form:", total)

# အတိုပုံစံ
total = 0
for score in scores:
    total += score              # အဓိပ္ပာယ်တူ: total = total + score
print("Short form:", total)     # ရလဒ် အတူတူပဲ
````

````python
# ဈေးခြင်းတောင်းရဲ့ စုစုပေါင်း စျေးနှုန်း
cart = [3500, 1200, 8000, 2300]
total_price = 0                 # loop မစခင်: 0 ကနေ စပါတယ်

for price in cart:
    total_price += price        # အဓိပ္ပာယ်တူ: total_price = total_price + price

print("Total price:", total_price)
````

### ပျမ်းမျှ

**ပျမ်းမျှ** ဆိုတာ စုစုပေါင်းကို တန်ဖိုး အရေအတွက်နဲ့ စားထားတာ ဖြစ်ပါတယ်။

တန်ဖိုး အရေအတွက်ကို ဘယ်လို ရမလဲ? `len()` ကို မှတ်မိပါ: list ထဲမှာ တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ ပြောပြပါတယ်။ `len([85, 92, 78])` က `3` ပါ။

ဒါကြောင့်: **ပျမ်းမျှ = စုစုပေါင်း / len(list)**

အရေးကြီးပါတယ်: စုစုပေါင်း ပြည့်စုံသွားတဲ့အခါ၊ loop **ပြီးမှ** ပျမ်းမျှကို တွက်ပါ။

````python
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total += score              # အဓိပ္ပာယ်တူ: total = total + score

average = total / len(scores)   # loop ပြီးမှ: စုစုပေါင်း (411) / အမှတ် အရေအတွက် (5)
print("Total:", total)
print("Number of scores:", len(scores))
print("Average:", average)
````

## 7. loop ထဲမှာ if ကို သုံးခြင်း

for loop ရဲ့ **ထဲမှာ** `if` ကို ထည့်လို့ရပါတယ်။ ဒါဆိုရင် တန်ဖိုး တစ်ခုချင်းစီကို စစ်ပြီး **လိုချင်တဲ့ တန်ဖိုးတွေအတွက်ပဲ** တစ်ခုခု လုပ်လို့ရပါတယ်။

`if` ဘယ်လို အလုပ်လုပ်လဲ မှတ်မိပါ:
- `if condition:` → အောက်က indent လုပ်ထားတဲ့ code က **condition True ဖြစ်မှသာ** run ပါတယ်။
- `else:` → အောက်က indent လုပ်ထားတဲ့ code က condition False ဖြစ်တဲ့အခါ run ပါတယ်။

`for` ထဲမှာ `if` ကို ထည့်ရင် indent **နှစ်ဆင့်** ဖြစ်ပါတယ်:

```python
for score in scores:          # space မရှိ
    if score >= 60:           # space 4 ခု -> for ထဲမှာ (အမှတ်တိုင်းအတွက် run ပါတယ်)
        print(score)          # space 8 ခု -> if ထဲမှာ (condition True ဖြစ်မှသာ run ပါတယ်)
```

- `if` စာကြောင်းမှာ space 4 ခု ရှိပါတယ်၊ `for` block ထဲမှာ ပါဝင်လို့ပါ။
- `print()` စာကြောင်းမှာ space 8 ခု ရှိပါတယ်၊ `if` block ထဲမှာ ပါဝင်ပြီး အဲဒီ block က `for` block ထဲမှာ ရှိလို့ပါ။

````python
# အောင်တဲ့ အမှတ် (60 နဲ့အထက်) တွေကိုပဲ ပြခြင်း
scores = [85, 42, 78, 55, 90]

for score in scores:            # အမှတ် တစ်ခုစီကို ထုတ်ယူပါတယ်
    if score >= 60:             # space 4 ခု: ဒီအမှတ်ကို စစ်ပါတယ်
        print(score)            # space 8 ခု: 60 နဲ့အထက် ဖြစ်မှသာ ပြပါတယ်
````

````python
# ဘယ်အမှတ်တွေကို စစ်ပြီး ဘယ်အမှတ်တွေကို ပြလဲ ကြည့်ရအောင်
scores = [85, 42, 78, 55, 90]

for score in scores:
    print("Checking", score)    # space 4 ခု: အမှတ် "တိုင်း" အတွက် run ပါတယ်
    if score >= 60:
        print("pass")           # space 8 ခု: score >= 60 ဖြစ်မှသာ run ပါတယ်
````

````python
# if - else: အမှတ်တိုင်းအတွက် pass ဒါမှမဟုတ် fail ကို ပြခြင်း
scores = [85, 42, 78, 55, 90]

for score in scores:
    if score >= 60:
        print(score, "pass")    # condition True ဖြစ်တဲ့အခါ run ပါတယ်
    else:
        print(score, "fail")    # condition False ဖြစ်တဲ့အခါ run ပါတယ်
````

````python
# စုံကိန်းတွေကိုပဲ ပြခြင်း
# 2 နဲ့ စားလို့ အကြွင်း 0 ဖြစ်ရင် စုံကိန်းပါ
# % က အကြွင်းကို ပေးပါတယ်: 8 % 2 က 0 (စုံ)၊ 5 % 2 က 1 (မ)
numbers = [3, 8, 5, 12, 7, 6]

for number in numbers:
    if number % 2 == 0:         # အကြွင်း 0 လား? -> စုံ
        print(number)
````

````python
# loop ထဲမှာ if - elif - else: အမှတ်တိုင်းကို grade ပေးခြင်း
# Python က condition တွေကို အပေါ်ကနေ အောက်ကို စစ်ပြီး True ဖြစ်တဲ့ "ပထမဆုံး" တစ်ခုကိုပဲ run ပါတယ်
scores = [95, 82, 74, 58]

for score in scores:
    if score >= 90:
        print(score, "A")
    elif score >= 80:
        print(score, "B")
    elif score >= 70:
        print(score, "C")
    else:
        print(score, "F")
````

````python
# တန်ဖိုး တချို့ကိုပဲ ပေါင်းခြင်း: if + ပေါင်းစုတန်ဖိုး
# စုံကိန်းတွေကိုပဲ ပေါင်းပါမယ်
numbers = [3, 8, 5, 12, 7, 6]
even_total = 0                  # loop မစခင်: 0 ကနေ စပါတယ်

for number in numbers:
    if number % 2 == 0:         # စုံကိန်း ဖြစ်မှသာ...
        even_total += number    # ...ပေါင်းပါတယ် (အဓိပ္ပာယ်တူ: even_total = even_total + number)

print("Sum of even numbers:", even_total)    # 8 + 12 + 6 = 26
````

## 8. ရေတွက်ခြင်း

ကျောင်းသား ဘယ်နှစ်ယောက် အောင်လဲ? pizza ကို ဘယ်နှစ်ယောက် မဲပေးလဲ?

**ရေတွက်** ဖို့ ပေါင်းစုတန်ဖိုးနဲ့ စိတ်ကူး အတူတူကို သုံးပါတယ်။ တစ်ခုတည်းသော ကွာခြားချက်: တန်ဖိုးကို ပေါင်းမယ့်အစား အကြိမ်တိုင်း **1** ကို ပေါင်းပါတယ်။

1. **loop မစခင်:** `count = 0`
2. **loop ထဲမှာ:** condition True ဖြစ်တဲ့အခါ `count = count + 1` (ဒါမှမဟုတ် `count += 1`)
3. **loop ပြီးရင်:** `count` ထဲမှာ condition ဘယ်နှစ်ကြိမ် True ဖြစ်ခဲ့လဲ ရှိပါတယ်

## 9. အကြီးဆုံး နဲ့ အငယ်ဆုံး တန်ဖိုးကို ရှာခြင်း

list ထဲက အမြင့်ဆုံး အမှတ်ကို ဘယ်လို ရှာမလဲ? စက္ကူပေါ်က ဂဏန်းတွေကို လူတစ်ယောက်က ဘယ်လို ရှာမလဲ စဉ်းစားကြည့်ပါ:

1. ပထမ ဂဏန်းကို ကြည့်ပြီး **"အခုထိ အကြီးဆုံး"** လို့ မှတ်ထားပါ။
2. နောက်ဂဏန်းကို ကြည့်ပါ။ မှတ်ထားတာထက် ကြီးလား? ကြီးရင် အဲဒါကို အစားထိုး မှတ်ထားပါ။
3. အဆုံးအထိ ဒီလိုပဲ ဆက်လုပ်ပါ။ နောက်ဆုံး မှတ်ထားတဲ့ ဂဏန်းက အကြီးဆုံး ဖြစ်ပါတယ်။

ဒီအဆင့်တွေကို Python နဲ့ အတိအကျ ရေးလို့ရပါတယ်။ `highest` variable က "ကျွန်တော်တို့ မှတ်ထားတဲ့ ဂဏန်း" ပါ။

**ပထမ တန်ဖိုး (`scores[0]`) ကနေ ဘာလို့ စတာလဲ?** နှိုင်းယှဉ်ဖို့ တစ်ခုခု လိုလို့ပါ။ index 0 က list ရဲ့ ပထမ တန်ဖိုး ဖြစ်ပါတယ်။

````python
# အမြင့်ဆုံး အမှတ်ကို ရှာခြင်း
scores = [85, 92, 78, 95, 66]
highest = scores[0]             # အဆင့် 1: ပထမ အမှတ် (85) ကို "အခုထိ အကြီးဆုံး" လို့ မှတ်ထားပါတယ်

for score in scores:
    if score > highest:         # အဆင့် 2: ဒီအမှတ်က မှတ်ထားတာထက် ကြီးလား?
        highest = score         #         ကြီးရင် ဒါကို အစားထိုး မှတ်ထားပါတယ်

print("Highest:", highest)      # အဆင့် 3: နောက်ဆုံး မှတ်ထားတဲ့ အမှတ်
````

````python
# အဆင့်ဆင့် ကြည့်ရအောင်
scores = [85, 92, 78, 95, 66]
highest = scores[0]

for score in scores:
    if score > highest:
        highest = score
    print("score:", score, "highest:", highest)     # space 4 ခု: အကြိမ်တိုင်း run ပါတယ်
````

````python
# အငယ်ဆုံး တန်ဖိုးကိုလည်း နည်းတူ ရှာပါတယ်၊ > အစား < ကို သုံးရုံပါပဲ
temperatures = [12, 8, 15, 3, 10]
lowest = temperatures[0]        # ပထမ အပူချိန် (12) ကနေ စပါတယ်

for temperature in temperatures:
    if temperature < lowest:    # ဒါက မှတ်ထားတာထက် ငယ်လား?
        lowest = temperature

print("Lowest:", lowest)
````

````python
# ဈေးခြင်းတောင်းထဲက အဈေးအကြီးဆုံး ပစ္စည်း
cart = [3500, 1200, 8000, 2300]
most_expensive = cart[0]

for price in cart:
    if price > most_expensive:
        most_expensive = price

print("Most expensive:", most_expensive)
````

for loop ထဲမှာ `append()` ကို သုံးရင် list ဟောင်းကနေ **list အသစ်** ကို ဖန်တီးလို့ရပါတယ်:

2. **loop ထဲမှာ:** လိုချင်တဲ့ တန်ဖိုးတွေကို list အသစ်ထဲ `append()` လုပ်ပါ။

## 11. loop ထဲမှာ in နဲ့ break ကို သုံးခြင်း

### in

`တန်ဖိုး in list` က တန်ဖိုးဟာ list ထဲမှာ ရှိမရှိ စစ်ပါတယ်။ ရလဒ်က `True` ဒါမှမဟုတ် `False` ဖြစ်လို့ `if` ရဲ့ condition အဖြစ် သုံးလို့ရပါတယ်။

```python
"milk" in ["bread", "milk"]     # True
"eggs" in ["bread", "milk"]     # False
```

**သတိထားပါ:** `for fruit in fruits:` ထဲက `in` နဲ့ `if fruit in store:` ထဲက `in` က ကြည့်ရတာ တူပေမဲ့ အလုပ် မတူပါဘူး။

| | အဓိပ္ပာယ် |
|---|---|
| `for fruit in fruits:` | fruits ထဲက တန်ဖိုး တစ်ခုစီကို တစ်ခုချင်း **ထုတ်ယူပါတယ်** |
| `if fruit in store:` | fruit က store ထဲမှာ ရှိမရှိ **စစ်ပါတယ်** (True / False) |

````python
# ကျွန်တော့် ဝယ်စရာစာရင်းထဲက ဘယ်ပစ္စည်းတွေ ဆိုင်မှာ ရှိလဲ?
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]

for item in shopping_list:              # "in" နံပါတ် 1: ပစ္စည်း တစ်ခုစီကို ထုတ်ယူပါတယ်
    if item in store:                   # "in" နံပါတ် 2: ဒီပစ္စည်း store ထဲမှာ ရှိလား?
        print(item, "yes")
    else:
        print(item, "no")
````

````python
# တခြားနေရာမှာ ဝယ်ရမယ့် ပစ္စည်းတွေရဲ့ list ဖန်တီးခြင်း
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]
not_found = []

for item in shopping_list:
    if item not in store:               # not in: ပစ္စည်းက store ထဲမှာ "မရှိ" ရင် True
        not_found.append(item)

print("Not found:", not_found)
````

### break

`break` က **loop ကို ချက်ချင်း ရပ်ပါတယ်**။ ကျန်တဲ့ တန်ဖိုးတွေကို လုံးဝ မစစ်တော့ဘဲ Python က loop နောက်က code ကို ခုန်သွားပါတယ်။

ရှာနေတာကို တွေ့ပြီး ဆက်ရှာစရာ မလိုတော့တဲ့အခါ အသုံးဝင်ပါတယ်။ `break` ကို အမြဲလိုလို `if` ထဲမှာ သုံးပါတယ်။

````python
# ပထမဆုံး ကျတဲ့ အမှတ်ကို တွေ့တာနဲ့ ရပ်ခြင်း
scores = [85, 92, 55, 78, 40]

for score in scores:
    if score < 60:
        print("Found a failing score:", score)
        break                           # loop ကို ဒီမှာပဲ ရပ်ပါတယ်
    print(score, "is OK")

print("The loop is over.")              # 78 နဲ့ 40 ကို ဘယ်တော့မှ မစစ်ခဲ့ပါဘူး
````

````python
# list ထဲမှာ Tom ဆိုတဲ့ နာမည်ရှိတဲ့သူ ရှိလား?
names = ["Bill", "Anna", "Tom", "Kate", "Mike"]

for name in names:
    print("Checking", name)
    if name == "Tom":
        print("Found Tom!")
        break                           # Kate နဲ့ Mike ကို စစ်စရာ မလိုပါဘူး
````

## 12. မကြာခဏ ဖြစ်တတ်သော အမှားများ

ဒါတွေက list နဲ့ for loop ကို သုံးတဲ့အခါ လူတိုင်းနီးပါး လုပ်မိကြတဲ့ အမှားတွေ ဖြစ်ပါတယ်။ တစ်ခုချင်းစီအတွက် ဒါတွေကို ကြည့်ပါမယ်:

- ❌ **မှားတဲ့ code**

တခြား မှားတဲ့ code တွေက **error မပြပေမဲ့ ရလဒ် မှားပါတယ်**။ Python က သတိမပေးလို့ ဒီအမှားတွေက ပိုအန္တရာယ်များပါတယ်။

### အမှား 1: colon မေ့ခြင်း

````python
# ❌ မှားပါတယ်
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits           # SyntaxError
#     print(fruit)
````

🔍 **ဘာမှားသွားလဲ?**

- `for` စာကြောင်းက colon `:` နဲ့ အမြဲ ဆုံးရပါမယ်။
- colon က Python ကို "ထပ်ခါထပ်ခါ run မယ့် code က နောက်စာကြောင်းကနေ စပါတယ်" လို့ ပြောပါတယ်။
- မပါရင် Python က code ကို နားမလည်ဘဲ `SyntaxError` ("code ရဲ့ သဒ္ဒါ မှားနေပါတယ်") ကို ပြပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** `for` စာကြောင်းရဲ့ အဆုံးမှာ `:` ထည့်ပါ။

````python
# ✅ မှန်ပါတယ်
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # အဆုံးမှာ colon ပါပါတယ်
    print(fruit)
````

### အမှား 2: indent မေ့ခြင်း

````python
# ❌ မှားပါတယ်
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits:
# print(fruit)                  # IndentationError
````

🔍 **ဘာမှားသွားလဲ?**

- `for` စာကြောင်း နောက်မှာ **indent လုပ်ထားတဲ့** စာကြောင်း အနည်းဆုံး တစ်ကြောင်း ရှိရပါမယ်: အဲဒါက ထပ်ခါထပ်ခါ run မယ့် code ပါ။
- ဒီမှာ `print(fruit)` ရဲ့ အစမှာ space မရှိလို့ Python က `for` loop မှာ ထပ်လုပ်စရာ ဘာမှ မရှိဘူးလို့ ထင်ပါတယ်။
- error က `IndentationError` ("indent မှားနေပါတယ်") ပါ။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** ထပ်ခါထပ်ခါ run မယ့် code ကို space 4 ခု indent လုပ်ပါ (ဒါမှမဟုတ် Tab ကို တစ်ခါ နှိပ်ပါ)။

````python
# ✅ မှန်ပါတယ်
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # အစမှာ space 4 ခု
````

### အမှား 3: တစ်ကြိမ်ပဲ run သင့်တဲ့ စာကြောင်းကို indent လုပ်ခြင်း

````python
# ❌ မှားပါတယ်: စုစုပေါင်းကို နောက်ဆုံးမှာ တစ်ကြိမ်ပဲ ပြချင်တာပါ
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score
    print("Total:", total)      # indent လုပ်ထား -> loop ထဲမှာ -> 3 ကြိမ် ပြပါတယ်!
````

🔍 **ဘာမှားသွားလဲ?**

- `print("Total:", total)` ကို indent လုပ်ထားလို့ loop **ထဲမှာ** ရှိပါတယ်။
- ဆိုလိုတာက အကြိမ်တိုင်း run ပြီး စုစုပေါင်းကို ပေါင်းနေဆဲမှာ ပြပါတယ်။
- error မရှိပေမဲ့ 1 ကြောင်းအစား 3 ကြောင်း ရပြီး ပထမ နှစ်ကြောင်းက တကယ့် စုစုပေါင်း မဟုတ်ပါဘူး။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** တစ်ကြိမ်ပဲ run သင့်တဲ့ စာကြောင်းရဲ့ indent ကို ဖယ်ပါ။ ဒါဆိုရင် loop **အပြင်မှာ** ရှိပြီး loop ပြီးမှ run ပါမယ်။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score              # loop ထဲမှာ: ပေါင်းပါတယ်
print("Total:", total)          # indent မလုပ်ထား -> loop အပြင်မှာ -> တစ်ကြိမ် ပြပါတယ်
````

### အမှား 4: list နဲ့ loop variable ကို နာမည်တူ ပေးခြင်း

````python
# ❌ မှားပါတယ်
fruits = ["apple", "banana", "cherry"]

for fruits in fruits:           # loop variable က list နဲ့ နာမည် "တူ" နေပါတယ်
    print(fruits)               # ဒါက မှန်မှန် ပြပါတယ်...

print(fruits)                   # ...ဒါပေမဲ့ အခု fruits က "cherry" ပဲ ဖြစ်နေပါပြီ။ list ပျောက်သွားပါပြီ!
````

🔍 **ဘာမှားသွားလဲ?**

- loop က အလုပ်လုပ်နေသလို ထင်ရလို့ သတိထားမိဖို့ ခက်ပါတယ်။
- ဒါပေမဲ့ အကြိမ်တိုင်း `fruits` ဆိုတဲ့ နာမည်ရှိတဲ့ သေတ္တာထဲကို တန်ဖိုး သိမ်းပါတယ်။ မှတ်ထားပါ: variable ထဲကို တန်ဖိုးအသစ် သိမ်းရင် တန်ဖိုးဟောင်း ပျောက်သွားပါတယ်။
- ဒါကြောင့် `fruits` ထဲမှာ ရှိခဲ့တဲ့ list ကို `"apple"`၊ ပြီးတော့ `"banana"`၊ ပြီးတော့ `"cherry"` နဲ့ အစားထိုးသွားပါတယ်။
- loop ပြီးရင် `fruits` ထဲမှာ `"cherry"` ပဲ ကျန်ပါတယ်။ list ပျောက်သွားပါပြီ။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** loop variable ကို **မတူတဲ့** နာမည် ပေးပါ။ list နာမည်ရဲ့ ဧကဝုစ် ပုံစံက အကောင်းဆုံးပါ: `fruits` → `fruit`။

````python
# ✅ မှန်ပါတယ်
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # fruit (တန်ဖိုး တစ်ခု) နဲ့ fruits (list) က နာမည် မတူပါဘူး
    print(fruit)

print(fruits)                   # list က ဘေးကင်းပါတယ်
````

### အမှား 5: loop variable အစား list နာမည်ကို သုံးခြင်း

````python
# ❌ မှားပါတယ်: သစ်သီး တစ်မျိုးချင်းစီကို ပြချင်တာပါ
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruits)               # fruits (s ပါ) -> list တစ်ခုလုံးကို 3 ကြိမ် ပြပါတယ်
````

🔍 **ဘာမှားသွားလဲ?**

- loop ထဲမှာ `fruit` (တန်ဖိုး တစ်ခု) အစား `fruits` (list တစ်ခုလုံး) လို့ ရေးမိပါတယ်။
- နာမည်တွေက အရမ်းနီးစပ်ပြီး `s` တစ်လုံးပဲ ကွာပါတယ်။ ဒါကြောင့် ဒီအမှားကို လုပ်မိဖို့ လွယ်ပါတယ်။
- loop က 3 ကြိမ် run ပြီး အကြိမ်တိုင်း list တစ်ခုလုံးကို ပြပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** loop ထဲမှာ လက်ရှိ တန်ဖိုးကို ယူဖို့ **loop variable** (`fruit`) ကို သုံးပါ။

````python
# ✅ မှန်ပါတယ်
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # fruit (s မပါ) -> တစ်ခါမှာ တန်ဖိုး တစ်ခု
````

### အမှား 6: list နာမည်ကို quote ထဲ ထည့်ခြင်း

````python
# ❌ မှားပါတယ်
fruits = ["apple", "banana", "cherry"]

for fruit in "fruits":          # quote ပါတဲ့ "fruits"!
    print(fruit)                # f, r, u, i, t, s ... တစ်ခါမှာ စာလုံး တစ်လုံး
````

🔍 **ဘာမှားသွားလဲ?**

- `fruits` (quote မပါ) က ကျွန်တော်တို့ list ကို သိမ်းထားတဲ့ **variable** ပါ။
- `"fruits"` (quote ပါ) က **string** တစ်ခုပဲ ဖြစ်ပါတယ်: f, r, u, i, t, s ဆိုတဲ့ စာလုံး 6 လုံး။
- for loop က string ကိုလည်း ဖြတ်သွားနိုင်ပါတယ်။ **တစ်ခါမှာ စာလုံး တစ်လုံး** ထုတ်ယူပါတယ်။ ဒါကြောင့် သစ်သီး 3 မျိုးအစား စာလုံး 6 လုံး ရပါတယ်။
- error မရှိလို့ ရှုပ်ထွေးစေနိုင်ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** variable နာမည်ကို quote မပါဘဲ ရေးပါ။

````python
# ✅ မှန်ပါတယ်
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # quote မပါ -> variable (list)
    print(fruit)
````

### အမှား 7: ပေါင်းစုတန်ဖိုးကို loop ထဲမှာ 0 ထားခြင်း

````python
# ❌ မှားပါတယ်
scores = [85, 92, 78]

for score in scores:
    total = 0                   # indent လုပ်ထား -> loop ထဲမှာ -> အကြိမ်တိုင်း run ပါတယ်!
    total += score

print("Total:", total)          # 255 မဟုတ်ဘဲ 78
````

🔍 **ဘာမှားသွားလဲ?**

- `total = 0` ကို indent လုပ်ထားလို့ loop ထဲမှာ ရှိပါတယ်။ ဒါက **3 ကြိမ်** run ပါတယ်။
- အကြိမ်တိုင်း စုစုပေါင်းက 0 ကို ပြန်ရောက်ပြီး လက်ရှိ အမှတ်ကိုပဲ ပေါင်းပါတယ်။
- 1 ကြိမ်မြောက်: 0 + 85 = 85 → 2 ကြိမ်မြောက်: **0 ပြန်ဖြစ်**, 0 + 92 = 92 → 3 ကြိမ်မြောက်: **0 ပြန်ဖြစ်**, 0 + 78 = 78
- ဒါကြောင့် နောက်ဆုံးမှာ `total` ထဲမှာ နောက်ဆုံး အမှတ်ပဲ ရှိပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** `total = 0` ကို loop **မစခင် တစ်ကြိမ်** (indent မလုပ်ဘဲ) ရေးပါ။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78]
total = 0                       # loop မစခင် -> တစ်ကြိမ်ပဲ run ပါတယ်

for score in scores:
    total += score              # အဓိပ္ပာယ်တူ: total = total + score

print("Total:", total)          # 85 + 92 + 78 = 255
````

### အမှား 8: ပျမ်းမျှကို loop ထဲမှာ တွက်ခြင်း

````python
# ❌ မှားပါတယ်
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score
    average = total / len(scores)       # indent လုပ်ထား -> အကြိမ်တိုင်း တွက်ပါတယ်
    print("Average:", average)
````

🔍 **ဘာမှားသွားလဲ?**

- ပျမ်းမျှ တွက်ဖို့ တန်ဖိုး **အားလုံး** ရဲ့ စုစုပေါင်း လိုပါတယ်။
- ဒါပေမဲ့ loop ထဲမှာ စုစုပေါင်းက မပြည့်စုံသေးပါဘူး။
- 1 ကြိမ်မြောက်: 80 / 3 → 2 ကြိမ်မြောက်: 170 / 3 → 3 ကြိမ်မြောက်: 270 / 3။ နောက်ဆုံး တစ်ခုပဲ တကယ့် ပျမ်းမျှ ဖြစ်ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** loop ထဲမှာ ပေါင်းရုံပဲ ပေါင်းပါ။ စုစုပေါင်း ပြည့်စုံသွားတဲ့အခါ loop **ပြီးမှ** ပျမ်းမျှကို တွက်ပါ။

````python
# ✅ မှန်ပါတယ်
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score                      # loop ထဲမှာ: ပေါင်းရုံပဲ ပေါင်းပါတယ်

average = total / len(scores)           # loop ပြီးမှ: 270 / 3
print("Average:", average)
````

### အမှား 9: loop variable ကို ပြောင်းပြီး list ကို ပြောင်းဖို့ ကြိုးစားခြင်း

````python
# ❌ မှားပါတယ်: list ထဲက အမှတ်တိုင်းကို ဆုအမှတ် 5 မှတ် ပေါင်းချင်တာပါ
scores = [85, 92, 78]

for score in scores:
    score = score + 5           # အမှတ်ကို ပြောင်းနေသလို ထင်ရပါတယ်...

print(scores)                   # ...ဒါပေမဲ့ list က [85, 92, 78] အတိုင်းပဲ!
````

🔍 **ဘာမှားသွားလဲ?**

- `score` က **သီးခြား သေတ္တာ** တစ်ခုပါ။ အကြိမ်တိုင်း list ထဲက တန်ဖိုး တစ်ခုကို ဒီသေတ္တာထဲ **ကူးယူ** ပါတယ်။
- `score = score + 5` က `score` သေတ္တာထဲက အရာကိုပဲ ပြောင်းပါတယ်။ `scores` list ကို လုံးဝ မထိပါဘူး။
- ဒါက မှတ်စုစာအုပ်ထဲက ဂဏန်းကို sticky note ပေါ် ကူးရေးပြီး sticky note ပေါ်က ဂဏန်းကို ပြင်တာနဲ့ တူပါတယ်။ မှတ်စုစာအုပ်က အတူတူပဲ ရှိနေပါတယ်။
- error မရှိလို့ သတိထားမိဖို့ ခက်ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** `append()` နဲ့ **list အသစ်** (အပိုင်း 10) ကို ဖန်တီးပြီး ပြောင်းထားတဲ့ တန်ဖိုးတွေကို အဲဒီထဲ ထည့်ပါ။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78]
new_scores = []                         # ပြောင်းထားတဲ့ တန်ဖိုးတွေအတွက် ဗလာ list

for score in scores:
    new_scores.append(score + 5)        # ပြောင်းထားတဲ့ တန်ဖိုးကို list အသစ်ထဲ ထည့်ပါတယ်

print(new_scores)                       # [90, 97, 83]
````

### အမှား 10: loop လုပ်နေစဉ် list ထဲက တန်ဖိုးတွေကို ဖျက်ခြင်း

````python
# ❌ မှားပါတယ်: ကျတဲ့ အမှတ် (60 အောက်) အားလုံးကို ဖျက်ချင်တာပါ
scores = [50, 40, 80, 30]

for score in scores:
    if score < 60:
        scores.remove(score)

print(scores)                   # [40, 80] -> 40 က ရှိနေဆဲပါ!
````

🔍 **ဘာမှားသွားလဲ?**

for loop က list ကို **နေရာအလိုက်** ဖြတ်သွားပါတယ်: 1 ခုမြောက် တန်ဖိုး၊ ပြီးတော့ 2 ခုမြောက်၊ ပြီးတော့ 3 ခုမြောက် ...
ဒါပေမဲ့ `remove()` က တန်ဖိုး တစ်ခုကို ထုတ်လိုက်ရင် သူ့နောက်က တန်ဖိုးအားလုံးက တစ်နေရာစီ **ရှေ့ကို ရွှေ့** သွားပါတယ်။

| အကြိမ် | loop ကြည့်နေတဲ့ နေရာ | အဲဒီအချိန်က list | ဘာဖြစ်လဲ |
|---|---|---|---|
| 1 ကြိမ်မြောက် | 1 ခုမြောက် နေရာ | `[50, 40, 80, 30]` | 50 < 60 → 50 ကို ဖျက်ပါတယ်။ list က `[40, 80, 30]` ဖြစ်သွားပါတယ် |
| 2 ကြိမ်မြောက် | 2 ခုမြောက် နေရာ | `[40, 80, 30]` | 2 ခုမြောက် တန်ဖိုးက အခု **80** ပါ။ 40 က 1 ခုမြောက် နေရာကို ရွှေ့သွားလို့ loop က အဲဒါကို **ကျော်သွားပါတယ်**! |
| 3 ကြိမ်မြောက် | 3 ခုမြောက် နေရာ | `[40, 80, 30]` | 30 < 60 → 30 ကို ဖျက်ပါတယ်။ list က `[40, 80]` ဖြစ်သွားပါတယ် |

- 40 ကို ဘယ်တော့မှ မစစ်ခဲ့လို့ list ထဲမှာ ကျန်နေပါတယ်။
- error မရှိလို့ သတိထားမိဖို့ အရမ်းခက်ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** loop ဖြတ်သွားနေတဲ့ list ကို မပြောင်းပါနဲ့။ အစား ထားချင်တဲ့ တန်ဖိုးတွေပဲ ပါတဲ့ **list အသစ်** ကို ဖန်တီးပါ (အပိုင်း 10)။

````python
# ✅ မှန်ပါတယ်
scores = [50, 40, 80, 30]
kept = []

for score in scores:
    if score >= 60:             # အောင်တဲ့ အမှတ်တွေကိုပဲ ထားပါတယ်
        kept.append(score)

print(kept)                     # [80]
````

### အမှား 11: တန်ဖိုးကို index အဖြစ် သုံးခြင်း

````python
# ❌ မှားပါတယ်
scores = [85, 92, 78]

# for score in scores:
#     print(scores[score])      # IndexError: list index out of range
````

- သင်ခန်းစာ 4 မှာ တန်ဖိုးတွေကို index နဲ့ ယူခဲ့ပါတယ်: `scores[0]`, `scores[1]` ... ဒါကြောင့် ဒီမှာလည်း `scores[...]` လိုတယ်လို့ ထင်ဖို့ လွယ်ပါတယ်။
- ဒါပေမဲ့ for loop မှာ `score` က index မဟုတ်ဘဲ **တန်ဖိုး ဖြစ်ပြီးသား** (85, 92, 78) ပါ။
- ဒါကြောင့် ပထမ အကြိမ်မှာ `scores[score]` က `scores[85]` ဖြစ်သွားပါတယ်။ list မှာ တန်ဖိုး 3 ခု (index 0, 1, 2) ပဲ ရှိပါတယ်။ index 85 မရှိပါဘူး။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** `score` ကိုယ်တိုင်ကို သုံးလိုက်ပါ။ တန်ဖိုး ရှိပြီးသားပါ။

### အမှား 12: "အခုထိ အကြီးဆုံး" ကို 0 ကနေ စခြင်း

````python
# ❌ မှားပါတယ်: ဆောင်းရာသီရဲ့ အမြင့်ဆုံး အပူချိန်ကို ရှာခြင်း
temperatures = [-5, -2, -8, -3]
highest = 0                     # 0 ကနေ စပါတယ်...

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # 0 -> ဒါပေမဲ့ 0 က list ထဲမှာတောင် မရှိပါဘူး!
````

🔍 **ဘာမှားသွားလဲ?**

- အပူချိန်အားလုံးက 0 အောက်မှာ ရှိပါတယ်။ ဒါကြောင့် ဘယ်တစ်ခုမှ 0 ထက် မကြီးပါဘူး။
- `temperature > highest` က ဘယ်တော့မှ True မဖြစ်ဘဲ `highest` က အဆုံးအထိ 0 အတိုင်း ရှိနေပါတယ်။
- ပေါင်းစုတန်ဖိုးကို 0 ကနေ စလို့ "0 ကနေ စရမယ်" လို့ ထင်ဖို့ လွယ်ပါတယ်။ ဒါပေမဲ့ အကြီးဆုံး တန်ဖိုး ရှာတာက တခြား အလုပ်ပါ။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** **list ရဲ့ ပထမ တန်ဖိုး** (`temperatures[0]`) ကနေ စပါ။ ဒါဆိုရင် စမယ့် တန်ဖိုးက list ကိုယ်တိုင်ကနေ အမြဲ လာပါတယ်။

````python
# ✅ မှန်ပါတယ်
temperatures = [-5, -2, -8, -3]
highest = temperatures[0]       # ပထမ တန်ဖိုး (-5) ကနေ စပါတယ်

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # -2
````

### အမှား 13: string နဲ့ ဂဏန်းကို + နဲ့ ဆက်ခြင်း

````python
# ❌ မှားပါတယ်
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

# print("Total: " + total)      # TypeError: can only concatenate str (not "int") to str
````

🔍 **ဘာမှားသွားလဲ?**

- `+` က **string နဲ့ string** ကို ဆက်နိုင်ပါတယ်။ ဒါပေမဲ့ `total` က **ဂဏန်း** (int) ပါ။
- Python က string နဲ့ ဂဏန်းကို `+` နဲ့ ဘယ်လို ဆက်ရမလဲ မသိလို့ error ပြပါတယ်။
- error message ရဲ့ အဓိပ္ပာယ်က: "string (str) ကို string နဲ့ပဲ ဆက်လို့ရပါတယ်၊ ဂဏန်း (int) နဲ့ မရပါ"။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** `print()` ထဲမှာ **comma** ကို သုံးပါ (အပိုင်း 5)။ comma နဲ့ဆိုရင် string နဲ့ ဂဏန်းကို ရောလို့ ရပါတယ်။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

print("Total:", total)          # comma -> string နဲ့ ဂဏန်းကို အတူ သုံးလို့ ရပါတယ်
````
