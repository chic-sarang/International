## 1. ฟังก์ชันคืออะไร?

### ปัญหา

สมมติว่าเราอยากพิมพ์ข้อความทักทายสั้น ๆ:

```python
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
```

ทีนี้ ลองนึกว่าเราต้องพิมพ์ข้อความทักทายนี้ **3 ครั้ง** ในที่ต่าง ๆ ของโปรแกรม เราก็ต้องเขียน 3 บรรทัดเดิมซ้ำแล้วซ้ำอีก

````python
# เขียน 3 บรรทัดเดิมซ้ำ 3 ครั้ง
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
````

แบบนี้ก็ทำงานได้ แต่มีปัญหา:

- ยาวและน่าเบื่อที่จะเขียน
- ถ้าอยากเปลี่ยน "Hello!" เป็น "Hi!" ต้องเปลี่ยน **ทุก** ที่ ถ้าลืมไปที่หนึ่ง ข้อความทักทายก็จะไม่เหมือนกัน

### แนวคิดของฟังก์ชัน

จะดีไหม ถ้าเราเขียน 3 บรรทัดนั้น **แค่ครั้งเดียว** ตั้ง **ชื่อ** ให้มัน แล้วเมื่อต้องการใช้ก็แค่เรียกชื่อนั้น?

นั่นแหละคือ **ฟังก์ชัน**: **กลุ่มของโค้ดที่มีชื่อ**

### ลองนึกถึงสูตรอาหาร

ฟังก์ชันก็เหมือน **สูตรอาหาร**

1. ก่อนอื่น คุณ **เขียนสูตร** ลงบนกระดาษ: "วิธีทำไข่เจียว: ขั้นที่ 1, ขั้นที่ 2, ขั้นที่ 3" การเขียนสูตรไม่ได้ทำให้เกิดไข่เจียวขึ้นมา มันเป็นแค่คำแนะนำที่เก็บไว้ใช้ทีหลัง
2. จากนั้น ทุกครั้งที่อยากกินไข่เจียว คุณก็ **ทำตามสูตร** คุณทำตามสูตรเดิมได้กี่ครั้งก็ได้

ฟังก์ชันทำงานแบบเดียวกัน:

| สูตรอาหาร | ฟังก์ชัน |
|---|---|
| เขียนสูตร | **สร้าง** ฟังก์ชัน (นิยาม) |
| ทำอาหารตามสูตร | **ใช้** ฟังก์ชัน (เรียก) |

### คุณเคยใช้ฟังก์ชันมาแล้ว!

คุณใช้ฟังก์ชันมาตั้งแต่บทที่ 1 แล้ว `print()`, `len()` และ `type()` ล้วนเป็นฟังก์ชัน มีคนเขียนไว้ให้คุณแล้ว คุณแค่เขียนชื่อและวงเล็บ `( )` เพื่อใช้มัน

ในบทนี้ คุณจะได้เรียนวิธีสร้างฟังก์ชัน **ของคุณเอง**

````python
# ฟังก์ชันที่คุณเคยใช้มาแล้ว
print("Hello")              # print() เป็นฟังก์ชัน: แสดงค่าบนหน้าจอ
print(len("apple"))         # len() เป็นฟังก์ชัน: นับจำนวน
print(type(10))             # type() เป็นฟังก์ชัน: บอกชนิดข้อมูล
````

## 2. การสร้างและการใช้ฟังก์ชัน

การใช้ฟังก์ชันมี **2 ขั้นตอนแยกกัน**:

1. **สร้าง** ฟังก์ชัน (เขียนสูตร)
2. **ใช้** ฟังก์ชัน (ทำตามสูตร)

### ขั้นที่ 1: สร้างฟังก์ชัน

```python
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")
```

มาดูทีละส่วน:

| ส่วน | ความหมาย |
|---|---|
| `def` | ย่อมาจาก **define** (นิยาม) หมายถึง "ฉันกำลังจะสร้างฟังก์ชัน" |
| `say_hello` | **ชื่อ** ของฟังก์ชัน คุณตั้งเองได้ (กฎเหมือนการตั้งชื่อตัวแปร) |
| `()` | วงเล็บ ต้องมีหลังชื่อเสมอ (จะเรียนว่าใส่อะไรข้างในได้ในหัวข้อที่ 3) |
| `:` | เครื่องหมายโคลอนท้ายบรรทัด เหมือน `if` และ `for` |
| บรรทัดที่ย่อหน้า | **ตัว (body)** ของฟังก์ชัน: โค้ดที่เป็นของฟังก์ชันนี้ |

**เรื่องย่อหน้า:** เหมือน `if` และ `for` คือ Python ใช้ย่อหน้า (เว้นวรรค 4 ช่องที่ต้นบรรทัด) เพื่อรู้ว่าบรรทัดไหนเป็นของฟังก์ชัน บรรทัดที่ย่อหน้าใต้ `def` คือตัวของฟังก์ชัน บรรทัดแรกที่ไม่ย่อหน้าจะอยู่ **นอก** ฟังก์ชัน

### การสร้างฟังก์ชันไม่ได้ทำให้มันทำงาน

นี่คือเรื่องที่สำคัญที่สุดในหัวข้อนี้

เมื่อ Python อ่าน `def` มันแค่ **จำ** ฟังก์ชันไว้ มันยังไม่ได้รันโค้ดข้างใน เหมือนการเขียนสูตรไม่ได้ทำให้เกิดไข่เจียว

ดูเซลล์ด้านล่าง เราสร้างฟังก์ชัน แต่ **ไม่มีอะไรถูกพิมพ์ออกมาเลย**

````python
# ขั้นที่ 1: สร้างฟังก์ชัน
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")

# ไม่มีอะไรถูกพิมพ์! เราแค่สร้าง (เขียน) ฟังก์ชัน ยังไม่ได้ใช้มัน
````

### ขั้นที่ 2: ใช้ฟังก์ชัน

ในการใช้ฟังก์ชัน ให้เขียน **ชื่อ** ของมันตามด้วย **วงเล็บ** `()`:

```python
say_hello()
```

การใช้ฟังก์ชันเรียกว่าการ **เรียก (call)** ฟังก์ชัน เมื่อเรียกฟังก์ชัน Python จะไปที่ฟังก์ชันนั้น รันโค้ดข้างในทั้งหมดจากบนลงล่าง แล้วกลับมา

````python
# ขั้นที่ 2: เรียก (ใช้) ฟังก์ชัน
say_hello()
````

### สูตรเดียว ใช้ได้หลายครั้ง

เมื่อสร้างฟังก์ชันแล้ว คุณเรียกมันได้ **กี่ครั้งก็ได้** ทุกครั้งที่เรียก ตัวฟังก์ชันทั้งหมดจะทำงานอีกครั้ง

````python
# ข้อความทักทายเดิม 3 ครั้ง แต่ตอนนี้ใช้แค่ 3 บรรทัดสั้น ๆ
say_hello()
say_hello()
say_hello()
````

ลองเทียบกับตอนต้นบท เมื่อก่อนเราต้องใช้ 9 บรรทัด ตอนนี้แค่เรียก 3 ครั้ง

ถ้าอยากเปลี่ยน "Hello!" เป็น "Hi!" ก็เปลี่ยนแค่ **ที่เดียว**: ข้างในฟังก์ชัน ทุกการเรียกจะใช้ข้อความใหม่

````python
# เปลี่ยนฟังก์ชันที่เดียว...
def say_hello():
    print("Hi!")                        # เปลี่ยนแค่ตรงนี้
    print("Welcome to Python class.")
    print("Have a nice day!")

# ...ทุกการเรียกจะใช้แบบใหม่
say_hello()
say_hello()
````

### สร้างและใช้ในเซลล์เดียวกัน

ส่วนใหญ่เราจะสร้างฟังก์ชันก่อน แล้วเรียกมันข้างล่าง มาตามดูทีละขั้นว่า Python ทำอะไร

````python
def say_goodbye():          # 1. Python จำฟังก์ชัน "say_goodbye" ไว้ ยังไม่มีอะไรทำงาน
    print("Goodbye!")
    print("See you next time.")

print("Class is over.")     # 2. บรรทัดนี้ไม่ย่อหน้า -> อยู่นอกฟังก์ชัน -> ทำงานตอนนี้
say_goodbye()               # 3. เรียกฟังก์ชัน -> 2 บรรทัดข้างในทำงานตอนนี้
print("The end.")           # 4. เมื่อฟังก์ชันจบ Python กลับมาที่นี่
````

สังเกตลำดับของผลลัพธ์:

1. `Class is over.` - บรรทัดแรกที่อยู่ **นอก** ฟังก์ชัน
2. `Goodbye!` และ `See you next time.` - ตัวฟังก์ชันทำงานเฉพาะตอนที่เรา **เรียก** มัน
3. `The end.` - หลังฟังก์ชันจบ Python ทำต่อที่บรรทัดถัดจากการเรียก

### สร้างก่อน แล้วค่อยใช้

คุณต้องเขียนสูตร **ก่อน** จะทำตามสูตรได้ เช่นเดียวกัน ฟังก์ชันต้อง **ถูกสร้างก่อนที่จะถูกเรียก** Python อ่านโค้ดจากบนลงล่าง ถ้าเรียกฟังก์ชันที่ยังไม่ได้สร้าง Python จะไม่รู้จักมันและเกิดข้อผิดพลาด (เราจะดูข้อผิดพลาดนี้ในหัวข้อที่ 5)

### ตัวอย่างเพิ่มเติม

````python
# ฟังก์ชันที่วาดเส้น
def draw_line():
    print("--------------------")

draw_line()
print("Python Basics")
draw_line()
````

````python
# ฟังก์ชันที่พิมพ์กติกาของเกม
def show_rules():
    print("Rule 1: Be kind.")
    print("Rule 2: Take turns.")
    print("Rule 3: Have fun!")

print("Before we start, here are the rules:")
show_rules()
````

## 3. การส่งค่าให้ฟังก์ชัน

### ปัญหา

ฟังก์ชัน `say_hello()` ของเราพูดเหมือนเดิมทุกครั้ง แต่ถ้าเราอยากทักทาย **คนต่าง ๆ** ด้วยชื่อของเขาล่ะ?

- "Hello, Anna!"
- "Hello, Tom!"
- "Hello, Bill!"

เราไม่อยากสร้างฟังก์ชันใหม่สำหรับทุกคน เราอยากมีฟังก์ชัน **เดียว** แล้ว **บอก** ชื่อมันทุกครั้ง

### ลองนึกถึงการสั่งที่ร้านกาแฟ

เวลาสั่งที่ร้านกาแฟ คุณ **บอก** พนักงานว่าต้องการอะไร: "กาแฟหนึ่งแก้วครับ" หรือ "ชาหนึ่งแก้วค่ะ" พนักงานทำงานแบบเดิม (ทำเครื่องดื่ม) แต่ผลลัพธ์ขึ้นอยู่กับ **สิ่งที่คุณบอก**

ฟังก์ชันก็ทำงานแบบนี้ได้ เวลาเรียก เราสามารถ **ส่ง** ค่าให้มัน แล้วฟังก์ชันจะทำงาน **ด้วยค่านั้น**

### สร้างฟังก์ชันที่รับค่า

```python
def greet(name):
    print("Hello,", name)
```

ดูในวงเล็บ: มีคำว่า `name` อยู่ นี่คือ **ตัวแปร** ที่รับค่าที่เราส่งมาตอนเรียกฟังก์ชัน ตัวแปรนี้เรียกว่า **พารามิเตอร์ (parameter)**

### เรียกพร้อมกับค่า

```python
greet("Anna")
```

ตอนนี้เวลาเรียกฟังก์ชัน เราใส่ค่าไว้ในวงเล็บ ค่าที่เราส่งไปเรียกว่า **อาร์กิวเมนต์ (argument)**

| คำ | อยู่ที่ไหน | ตัวอย่าง |
|---|---|---|
| **พารามิเตอร์ (parameter)** | ใน `( )` ตอน **สร้าง** ฟังก์ชัน | `name` ใน `def greet(name):` |
| **อาร์กิวเมนต์ (argument)** | ใน `( )` ตอน **เรียก** ฟังก์ชัน | `"Anna"` ใน `greet("Anna")` |

ไม่ต้องกังวลกับสองคำนี้มากนัก แนวคิดสำคัญคือ: **ค่าที่เราส่งไปจะถูกเก็บไว้ในตัวแปรข้างในฟังก์ชัน**

````python
# สร้างฟังก์ชันที่รับค่าหนึ่งค่า
def greet(name):
    print("Hello,", name)

# เรียกด้วยค่าที่ต่างกัน
greet("Anna")
greet("Tom")
greet("Bill")
````

**เรื่อง `print("Hello,", name)`:** เมื่อเขียนหลายค่าใน `print()` โดยคั่นด้วย **เครื่องหมายจุลภาค** `,` ค่าเหล่านั้นจะถูกพิมพ์ในบรรทัดเดียว และ Python จะใส่ **ช่องว่างหนึ่งช่อง** ระหว่างค่าให้อัตโนมัติ ข้อความในเครื่องหมายคำพูดจะถูกพิมพ์ตามที่เขียน ส่วนตัวแปรจะพิมพ์ค่าที่อยู่ข้างในแทน

### เกิดอะไรขึ้น ทีละขั้น

เมื่อเรียก `greet("Anna")`:

| ขั้น | Python ทำอะไร |
|---|---|
| 1 | หาฟังก์ชัน `greet` |
| 2 | เก็บค่าที่ส่งมาในพารามิเตอร์: `name = "Anna"` |
| 3 | รันตัวฟังก์ชัน: `print("Hello,", name)` → `name` คือ `"Anna"` → พิมพ์ `Hello, Anna` |
| 4 | ฟังก์ชันจบแล้ว Python กลับไปที่บรรทัดถัดจากการเรียก |

จากนั้นเมื่อเรียก `greet("Tom")` ก็เกิดสิ่งเดิมอีกครั้ง แต่ครั้งนี้ `name = "Tom"`

ดังนั้น **โค้ดในฟังก์ชันเหมือนเดิมทุกครั้ง** แต่ **ผลลัพธ์เปลี่ยน** ตามค่าที่เราส่งไป

````python
# พารามิเตอร์ก็คือตัวแปรธรรมดา เราดูได้ว่าข้างในมีอะไร
def show_value(x):
    print("The function received:", x)

show_value(10)
show_value("apple")
show_value(3.14)
````

````python
# ฟังก์ชันที่ใช้ค่าในการคำนวณ
def show_double(number):
    print(number * 2)       # number ถูกแทนด้วยค่าที่เราส่งไป

show_double(5)              # number = 5  -> 5 * 2
show_double(100)            # number = 100 -> 100 * 2
````

````python
# อาร์กิวเมนต์เป็นตัวแปรก็ได้
def greet(name):
    print("Hello,", name)

student = "Kate"
greet(student)              # ค่าใน student ("Kate") ถูกส่งให้ฟังก์ชัน
````

### ส่งสองค่า

ฟังก์ชันรับได้ **มากกว่าหนึ่ง** ค่า ให้คั่นด้วยเครื่องหมายจุลภาค `,` ทั้งตอนสร้างและตอนเรียกฟังก์ชัน

```python
def add(a, b):
    print(a + b)

add(2, 3)
```

ค่าจะจับคู่ **ตามลำดับ**:
- ค่าที่ 1 (`2`) เข้าไปในพารามิเตอร์ที่ 1 (`a`)
- ค่าที่ 2 (`3`) เข้าไปในพารามิเตอร์ที่ 2 (`b`)

````python
def add(a, b):
    print(a + b)

add(2, 3)                   # a = 2, b = 3  -> 2 + 3
add(10, 20)                 # a = 10, b = 20 -> 10 + 20
````

````python
# ฟังก์ชันที่มีสองพารามิเตอร์: ชื่อและอายุ
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce("Anna", 20)       # name = "Anna", age = 20
introduce("Tom", 25)        # name = "Tom", age = 25
````

### ลำดับมีความสำคัญ

เพราะค่าจับคู่ตามลำดับ ถ้าสลับลำดับ ผลลัพธ์อาจเปลี่ยน

````python
def subtract(a, b):
    print(a - b)

subtract(10, 3)             # a = 10, b = 3  -> 10 - 3 = 7
subtract(3, 10)             # a = 3, b = 10  -> 3 - 10 = -7 (ผลลัพธ์ต่างกัน!)
````

````python
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce(20, "Anna")       # ลำดับผิด: name = 20, age = "Anna"
````

## 4. การรับผลลัพธ์กลับมา: return

### ปัญหา

ฟังก์ชัน `add(a, b)` ของเรา **พิมพ์** ผลลัพธ์บนหน้าจอ แต่ถ้าเราอยาก **ใช้** ผลลัพธ์ทีหลังล่ะ? เช่น เก็บไว้ในตัวแปร หรือนำไปคำนวณต่อ?

การพิมพ์แค่ **แสดง** ผลลัพธ์ มันไม่ได้ **ส่งผลลัพธ์กลับ** มาให้เรา เมื่อพิมพ์แล้ว เราก็เอามันมาใช้อีกไม่ได้

### ลองนึกถึงการฝากซื้อของ

สมมติคุณขอเพื่อนว่า "ช่วยไปร้านซื้อขนมปังให้หน่อย"

- เพื่อนไป ซื้อขนมปัง แล้วกลับมา
- จากนั้นเพื่อน **ยื่นขนมปังใส่มือคุณ**
- ตอนนี้ **คุณมีขนมปัง** แล้ว และจะทำอะไรกับมันก็ได้: กิน แบ่ง หรือทำแซนด์วิช

`return` ก็เหมือน **การยื่นผลลัพธ์กลับมา** ฟังก์ชันทำงานเสร็จ แล้วส่งผลลัพธ์กลับไปยังที่ที่เรียกมัน

### สร้างฟังก์ชันที่มี return

```python
def add(a, b):
    return a + b
```

`return` หมายถึง: **"ส่งค่านี้กลับไปยังที่ที่เรียกฟังก์ชัน"**

สังเกต: ฟังก์ชันนี้ไม่ได้พิมพ์อะไรเลย มันแค่ส่งผลลัพธ์กลับมา

````python
def add(a, b):
    return a + b            # ส่งผลลัพธ์กลับ

result = add(2, 3)          # ผลลัพธ์ (5) กลับมาและถูกเก็บใน result
print(result)
````

### การเรียกฟังก์ชันจะถูกแทนที่ด้วยค่าที่ return

นี่คือแนวคิดที่สำคัญที่สุดของ `return`

เมื่อฟังก์ชัน return ค่า **ตำแหน่งที่คุณเรียกฟังก์ชันจะถูกแทนที่ด้วยค่านั้น**

มาดูบรรทัดนี้ช้า ๆ:

```python
result = add(2, 3)
```

| ขั้น | Python เห็นอะไร |
|---|---|
| 1 | `result = add(2, 3)` |
| 2 | Python เรียก `add(2, 3)`: `a = 2`, `b = 3` ฟังก์ชัน return `2 + 3` ซึ่งก็คือ `5` |
| 3 | `add(2, 3)` ถูก **แทนที่** ด้วย `5` → `result = 5` |
| 4 | `5` ถูกเก็บใน `result` |

ดังนั้นหลังจากฟังก์ชัน return ก็เหมือนคุณเขียน `result = 5`

คุณอาจนึกภาพ `add(2, 3)` เป็นกล่องที่ **กลายเป็น 5** เมื่อฟังก์ชันทำงานเสร็จ

````python
def add(a, b):
    return a + b

x = add(2, 3)               # add(2, 3) -> 5  ดังนั้นจะกลายเป็น: x = 5
y = add(10, 20)             # add(10, 20) -> 30  ดังนั้นจะกลายเป็น: y = 30

print(x)
print(y)
````

### ใช้ค่าที่ return ในการคำนวณ

เพราะ `add(2, 3)` กลายเป็น `5` คุณจึงใช้มันได้ทุกที่ที่ใช้เลข `5` ได้

```python
add(2, 3) + add(3, 4)
```

| ขั้น | Python เห็นอะไร |
|---|---|
| 1 | `add(2, 3) + add(3, 4)` |
| 2 | `add(2, 3)` ทำงานและ return 5 → **`5`** `+ add(3, 4)` |
| 3 | `add(3, 4)` ทำงานและ return 7 → `5 +` **`7`** |
| 4 | `12` |

````python
def add(a, b):
    return a + b

total = add(2, 3) + add(3, 4)   # กลายเป็น 5 + 7
print(total)                    # 12
````

````python
def add(a, b):
    return a + b

print(add(2, 3) * 10)           # add(2, 3) กลายเป็น 5 -> print(5 * 10) -> 50
print(add(2, 3) > 4)            # add(2, 3) กลายเป็น 5 -> print(5 > 4) -> True
````

### เขียนการเรียกไว้ใน print() โดยตรง

`print(add(2, 3))` ก็ทำงานแบบเดียวกัน:

| ขั้น | Python เห็นอะไร |
|---|---|
| 1 | `print(add(2, 3))` |
| 2 | `add(2, 3)` return 5 → `print(5)` |
| 3 | พิมพ์ `5` |

````python
def add(a, b):
    return a + b

print(add(2, 3))                # กลายเป็น print(5)
````

### ส่งค่าที่ return ให้ฟังก์ชันอื่น

คุณส่งค่าที่ return ให้ฟังก์ชันอีกครั้งก็ได้ Python ทำงาน **จากข้างในออกมาข้างนอก** เสมอ: การเรียกข้างในจะถูกแทนที่ก่อน

```python
add(add(1, 2), 3)
```

| ขั้น | Python เห็นอะไร |
|---|---|
| 1 | `add(add(1, 2), 3)` |
| 2 | `add(1, 2)` ข้างใน return 3 → `add(` **`3`** `, 3)` |
| 3 | `add(3, 3)` return 6 → **`6`** |

````python
def add(a, b):
    return a + b

result = add(add(1, 2), 3)      # add(1, 2) -> 3 จากนั้น add(3, 3) -> 6
print(result)
````

### ตัวอย่างเพิ่มเติมของ return

````python
def double(number):
    return number * 2

a = double(4)                   # a = 8
b = double(a)                   # a คือ 8 ดังนั้น double(8) -> 16
print(a)
print(b)
````

````python
# ฟังก์ชันที่ return สตริง
def make_greeting(name):
    return "Hello, " + name     # + ต่อสองสตริงเข้าด้วยกัน

message = make_greeting("Anna")  # message = "Hello, Anna"
print(message)
````

````python
# ราคาหลังลด
def discount_price(price):
    return price - 1000

coffee = discount_price(4500)   # coffee = 3500
cake = discount_price(6000)     # cake = 5000

print("Coffee:", coffee)
print("Cake:", cake)
print("Total:", coffee + cake)  # คำนวณด้วยค่าที่ return มาได้
````

### print กับ return: ต่างกันอย่างไร?

มาสร้างสองฟังก์ชันที่ดูเกือบเหมือนกัน อันหนึ่งใช้ `print` อีกอันใช้ `return`

```python
def show_add(a, b):        # ใช้ print
    print(a + b)

def add(a, b):             # ใช้ return
    return a + b
```

| | `print` ในฟังก์ชัน | `return` ในฟังก์ชัน |
|---|---|---|
| ทำอะไร | **แสดง** ผลลัพธ์บนหน้าจอ | **ส่งผลลัพธ์กลับ** ไปยังที่ที่เรียก |
| เก็บผลลัพธ์ในตัวแปรได้ไหม? | ❌ | ✅ |
| นำผลลัพธ์ไปคำนวณได้ไหม? | ❌ | ✅ |
| เหมือนกับ... | เพื่อนที่โชว์ขนมปังให้ดูแล้วเก็บไว้เอง | เพื่อนที่ยื่นขนมปังใส่มือคุณ |

### ถ้าไม่มี return จะได้อะไรกลับมา?

ถึงฟังก์ชันจะไม่มี `return` มันก็ยังส่งบางอย่างกลับมา: ค่าพิเศษที่ชื่อ **`None`** `None` หมายถึง **"ไม่มีอะไร"**

ดังนั้น `show_add(2, 3)` จะถูกแทนที่ด้วย `None` ไม่ใช่ `5`

````python
def show_add(a, b):
    print(a + b)                # แค่แสดงผลลัพธ์ ไม่มี return

result = show_add(2, 3)         # 5 ถูกพิมพ์บนหน้าจอ... แต่อะไรถูกเก็บใน result?
print("result is:", result)     # None! show_add(2, 3) ถูกแทนที่ด้วย None ไม่ใช่ 5
````

ดูผลลัพธ์ให้ดี:

1. `5` - พิมพ์โดย **ตัวฟังก์ชันเอง** (`print` ข้างใน)
2. `result is: None` - ฟังก์ชันไม่ได้ return อะไร `result` จึงได้ `None`

เลข 5 ปรากฏบนหน้าจอ แต่ **เราไม่ได้รับมันกลับมา** มันหายไปแล้ว

### ดังนั้น add(2, 3) + add(3, 4) จึงใช้กับ print ไม่ได้

ถ้าลอง `show_add(2, 3) + show_add(3, 4)` จะเกิดอะไรขึ้น?

| ขั้น | Python เห็นอะไร |
|---|---|
| 1 | `show_add(2, 3) + show_add(3, 4)` |
| 2 | `show_add(2, 3)` พิมพ์ 5 และถูกแทนที่ด้วย **`None`** |
| 3 | `show_add(3, 4)` พิมพ์ 7 และถูกแทนที่ด้วย **`None`** |
| 4 | `None + None` → **ข้อผิดพลาด!** Python เอา "ไม่มีอะไร" บวกกับ "ไม่มีอะไร" ไม่ได้ |

````python
def show_add(a, b):
    print(a + b)

# total = show_add(2, 3) + show_add(3, 4)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'NoneType'
````

(บรรทัดข้างบนเป็นคอมเมนต์ไว้เพื่อไม่ให้ notebook หยุดทำงาน ถ้าลบ `#` ทั้งสองตัวของบรรทัด `total = ...` คุณจะเห็น `5` และ `7` ถูกพิมพ์ก่อน แล้วจึงเกิดข้อผิดพลาด)

ข้อความแสดงข้อผิดพลาดหมายถึง: "ใช้ `+` กับ `None` และ `None` ไม่ได้" (`NoneType` คือชนิดข้อมูลของ `None`)

### ควรใช้อันไหนเมื่อไร?

- ถ้าแค่อยาก **แสดง** บางอย่างบนหน้าจอ → ใช้ `print` ได้
- ถ้าอยาก **ใช้** ผลลัพธ์ทีหลัง (เก็บ คำนวณ ส่งให้ฟังก์ชันอื่น) → ใช้ `return`

ส่วนใหญ่ ฟังก์ชันที่คำนวณอะไรบางอย่างควรใช้ `return` แล้วเมื่ออยากเห็นผลลัพธ์ ก็พิมพ์ด้วย `print(add(2, 3))` ได้เสมอ

````python
def add(a, b):
    return a + b

# ถ้ามี return ทำได้ทั้งสองอย่าง: ใช้ผลลัพธ์ และ พิมพ์มัน
total = add(2, 3) + add(3, 4)
print(total)
````

## 5. ข้อผิดพลาดที่พบบ่อย

นี่คือข้อผิดพลาดที่แทบทุกคนทำเมื่อเรียนเรื่องฟังก์ชัน สำหรับแต่ละข้อ เราจะดู:

- ❌ **โค้ดที่ผิด**
- 🔍 **ผิดตรงไหน**
- ✅ **วิธีที่ถูกต้อง**

โค้ดที่ทำให้เกิดข้อผิดพลาดถูกเขียนเป็นคอมเมนต์ (มี `#` ข้างหน้า) เพื่อไม่ให้ notebook หยุด ถ้าอยากเห็นข้อความแสดงข้อผิดพลาดจริง ให้ลบ `#` แล้วรันเซลล์ จากนั้นใส่ `#` กลับคืน

### ข้อผิดพลาด 1: สร้างฟังก์ชันแล้วคิดว่ามันจะทำงานเอง

````python
# ❌ ผิด: อยากพิมพ์ "Good morning!"
def morning():
    print("Good morning!")

# ไม่มีอะไรถูกพิมพ์...
````

🔍 **ผิดตรงไหน?**

- `def` แค่ **สร้าง** ฟังก์ชัน ไม่ได้รันมัน (การเขียนสูตรไม่ได้ทำให้เกิดอาหาร)
- ไม่มีข้อผิดพลาดแสดงขึ้นมา จึงอาจสับสนได้: "ทำไมไม่มีอะไรเกิดขึ้น?"

✅ **วิธีที่ถูกต้อง:** หลังสร้างฟังก์ชันแล้ว ให้ **เรียก** มัน

````python
# ✅ ถูกต้อง
def morning():
    print("Good morning!")

morning()               # เรียกมัน
````

### ข้อผิดพลาด 2: ลืมวงเล็บตอนเรียก

````python
# ❌ ผิด
def morning():
    print("Good morning!")

morning                 # ไม่มีวงเล็บ -> ฟังก์ชันไม่ทำงาน
print("Did you see 'Good morning!' above? No!")
````

🔍 **ผิดตรงไหน?**

- `morning` ที่ไม่มี `()` เป็นแค่ **ชื่อ** ของฟังก์ชัน ไม่ใช่การเรียก
- การเรียกฟังก์ชัน **ต้อง** มีวงเล็บ: `morning()` วงเล็บหมายถึง "ทำงานตอนนี้"
- ไม่มีข้อผิดพลาดแสดงขึ้นมา แต่ฟังก์ชันไม่ทำงาน

✅ **วิธีที่ถูกต้อง:** ใส่ `()` หลังชื่อทุกครั้งที่เรียก

````python
# ✅ ถูกต้อง
def morning():
    print("Good morning!")

morning()               # มีวงเล็บ
````

### ข้อผิดพลาด 3: เรียกฟังก์ชันก่อนสร้าง

✅ **วิธีที่ถูกต้อง:** สร้างฟังก์ชันก่อน แล้วเรียกข้างล่าง

### ข้อผิดพลาด 4: ลืมโคลอนหรือย่อหน้า

````python
# ❌ ผิด
# def night()                 # SyntaxError: ขาดโคลอน
#     print("Good night!")

# def night():
# print("Good night!")        # IndentationError: ตัวฟังก์ชันไม่ได้ย่อหน้า
````

- ตัวฟังก์ชันต้องย่อหน้า (เว้นวรรค 4 ช่อง) ถ้าไม่ย่อหน้า Python จะไม่รู้ว่าบรรทัดไหนเป็นของฟังก์ชัน

### ข้อผิดพลาด 5: ส่งจำนวนค่าผิด

````python
# ❌ ผิด
def add(a, b):
    return a + b

# add(2)          # TypeError: add() missing 1 required positional argument: 'b'
# add(2, 3, 4)    # TypeError: add() takes 2 positional arguments but 3 were given
````

🔍 **ผิดตรงไหน?**

- `add` ถูกสร้างด้วยพารามิเตอร์ **2** ตัว (`a` และ `b`) จึงต้องการค่า **2** ค่าพอดี
- `add(2)`: ส่งแค่ 1 ค่า `a` ได้ 2 แต่ `b` ไม่ได้อะไรเลย ข้อผิดพลาดบอกว่า "ขาดไป 1 ค่า: `b`"
- `add(2, 3, 4)`: ส่ง 3 ค่า แต่มีที่แค่ 2 ที่ ข้อผิดพลาดบอกว่า "รับ 2 แต่ได้ 3"

✅ **วิธีที่ถูกต้อง:** ส่งค่าให้มีจำนวนเท่ากับพารามิเตอร์พอดี

````python
# ✅ ถูกต้อง
def add(a, b):
    return a + b

print(add(2, 3))        # 2 พารามิเตอร์ 2 ค่า
````

### ข้อผิดพลาด 6: ใช้ print ในที่ที่ต้องใช้ return

````python
# ❌ ผิด: อยากเก็บผลลัพธ์ไว้ใช้ทีหลัง
def multiply(a, b):
    print(a * b)                # แค่แสดงผลลัพธ์

answer = multiply(3, 4)         # 12 ถูกพิมพ์... แต่ answer เป็น None
print("answer:", answer)
````

🔍 **ผิดตรงไหน?**

- `multiply` ใช้ `print` ไม่ใช่ `return` มัน **แสดง** 12 บนหน้าจอ แต่ **ไม่ได้ส่งกลับมา**
- ฟังก์ชันที่ไม่มี `return` จะส่ง `None` กลับมา ดังนั้น `multiply(3, 4)` ถูกแทนที่ด้วย `None` และ `answer` เป็น `None`
- ถ้าลองนำไปคำนวณ (เช่น `answer + 1`) จะเกิดข้อผิดพลาด

✅ **วิธีที่ถูกต้อง:** ใช้ `return` เพื่อส่งผลลัพธ์กลับ ถ้าอยากเห็นก็พิมพ์ข้างนอกฟังก์ชัน

````python
# ✅ ถูกต้อง
def multiply(a, b):
    return a * b                # ส่งผลลัพธ์กลับ

answer = multiply(3, 4)         # answer = 12
print("answer:", answer)
print("answer + 1:", answer + 1)
````

### ข้อผิดพลาด 7: return แล้วแต่ไม่ได้ใช้ผลลัพธ์

````python
# ❌ ผิด: อยากเห็นผลลัพธ์
def multiply(a, b):
    return a * b

multiply(3, 4)                  # ผลลัพธ์ (12) กลับมาแล้ว... แต่ไม่มีใครรับไว้!
print("Where is the result?")
````

🔍 **ผิดตรงไหน?**

- `multiply(3, 4)` ถูกแทนที่ด้วย `12` แต่ในบรรทัดนั้นมีแค่ `12` อยู่เฉย ๆ เราไม่ได้เก็บหรือพิมพ์มัน
- เหมือนเพื่อนยื่นขนมปังให้ แต่คุณทำมันตกพื้น
- ไม่มีข้อผิดพลาด แต่ผลลัพธ์หายไป

✅ **วิธีที่ถูกต้อง:** รับค่าที่ return มา: เก็บในตัวแปร หรือเขียนการเรียกไว้ใน `print()`

````python
# ✅ ถูกต้อง
def multiply(a, b):
    return a * b

result = multiply(3, 4)         # เก็บไว้
print(result)

print(multiply(3, 4))           # หรือพิมพ์โดยตรง
````
