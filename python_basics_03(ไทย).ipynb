## 1. ลูป for คืออะไร?

ลองนึกภาพว่าคุณต้องแสดงคำว่า "Hello" 5 ครั้ง คุณอาจเขียน `print("Hello")` 5 ครั้งก็ได้ แต่ถ้าเป็น 100 ครั้งล่ะ? หรือ 1000 ครั้ง? การเขียนบรรทัดเดิมซ้ำไปซ้ำมาทั้งเหนื่อยและผิดพลาดได้ง่าย

**ลูป (loop)** ช่วยแก้ปัญหานี้ ลูปจะรันโค้ดเดิมซ้ำแล้วซ้ำอีก กี่ครั้งก็ได้ตามที่คุณต้องการ

ลูปที่ใช้บ่อยที่สุดใน Python คือลูป `for` รูปแบบพื้นฐานมีหน้าตาแบบนี้:

```python
for i in range(5):
    โค้ดที่ต้องการทำซ้ำ
```

อ่านได้แบบนี้: **"สำหรับตัวเลข i แต่ละตัวใน range(5) ให้รันโค้ดข้างล่าง"**

- `range(5)` ให้ตัวเลข 0, 1, 2, 3, 4 (รวม 5 ตัว) เราจะเรียน `range()` อย่างละเอียดในหัวข้อถัดไป
- `i` คือตัวแปร ทุกครั้งที่ลูปทำซ้ำ ตัวเลขถัดไปจะถูกเก็บไว้ใน `i`
- เหมือนกับ `if` บรรทัด `for` ต้องลงท้ายด้วยโคลอน `:`
- เหมือนกับ `if` โค้ดที่ต้องการทำซ้ำต้อง **ย่อหน้า** (4 ช่องว่าง) เฉพาะบรรทัดที่ย่อหน้าเท่านั้นที่จะถูกทำซ้ำ

````python
# ทำซ้ำ print("Hello") 5 ครั้ง
# range(5) ให้ตัวเลข 5 ตัว (0, 1, 2, 3, 4) โค้ดที่ย่อหน้าจึงทำงาน 5 ครั้ง
for i in range(5):
    print("Hello")      # บรรทัดนี้อยู่ในลูป for (ย่อหน้าอยู่) จึงถูกทำซ้ำ
````

````python
# มาดูกันว่าแต่ละครั้ง i เก็บค่าอะไรไว้
# ครั้งที่ 1: i = 0, ครั้งที่ 2: i = 1, ครั้งที่ 3: i = 2, ครั้งที่ 4: i = 3, ครั้งที่ 5: i = 4
for i in range(5):
    print(i)            # แสดงค่าปัจจุบันของ i
````

````python
# การรันโค้ดที่ย่อหน้าครบหนึ่งรอบ เรียกว่า "การทำซ้ำ" หนึ่งรอบ
# ทุกบรรทัดที่ย่อหน้าจะทำงานรอบละหนึ่งครั้ง ตามลำดับจากบนลงล่าง
for i in range(3):
    print("Repetition number", i)     # print() แสดงหลายค่าพร้อมกันได้ ถ้าคั่นด้วยเครื่องหมายจุลภาค
    print("----------")               # บรรทัดนี้ก็อยู่ในลูปเช่นกัน (ย่อหน้าเท่ากัน)

print("The loop is finished.")        # ไม่ได้ย่อหน้า จึงอยู่นอกลูป ทำงานแค่ครั้งเดียวหลังจากลูปจบ
````

````python
# ชื่อตัวแปรไม่จำเป็นต้องเป็น i
# i เป็นแค่ชื่อที่คนส่วนใหญ่นิยมใช้ จะตั้งชื่ออะไรก็ได้ ขอแค่ถูกตามกฎการตั้งชื่อตัวแปร
for number in range(3):
    print(number)

for count in range(3):
    print("count is", count)
````

## 2. เจาะลึก range()

`range()` สร้างชุดตัวเลขที่เรียงต่อกัน มีวิธีใช้ 3 แบบ

| วิธีเขียน | ความหมาย | ตัวเลขที่ได้ |
|---|---|---|
| `range(end)` | จาก 0 ถึงก่อน end (ไม่รวม end) | `range(5)` → 0, 1, 2, 3, 4 |
| `range(start, end)` | จาก start ถึงก่อน end (ไม่รวม end) | `range(1, 6)` → 1, 2, 3, 4, 5 |
| `range(start, end, step)` | จาก start ถึงก่อน end โดยกระโดดทีละ step | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |

สิ่งที่สำคัญที่สุดที่ต้องจำ:
- **ตัวเลขสุดท้าย (end) ไม่เคยถูกรวมเข้าไป** `range(5)` หยุดที่ 4 ไม่ใช่ 5
- ถ้าไม่ระบุตัวเลขเริ่มต้น จะเริ่มจาก **0**
- ถ้าไม่ระบุ step จะเพิ่มขึ้นทีละ **1**

````python
# range(end): เริ่มจาก 0 และหยุดก่อนถึง end
# range(4) ให้ 0, 1, 2, 3 -> ไม่รวม 4
for i in range(4):
    print(i)
````

````python
# range(start, end): เริ่มจาก start และหยุดก่อนถึง end
# ถ้าต้องการ 1 ถึง 5 ค่า end ต้องเป็น 6 (เพราะตัวเลขสุดท้ายไม่ถูกรวม)
for i in range(1, 6):
    print(i)
````

````python
# range(start, end, step): กระโดดทีละ step
# เริ่มจาก 0 กระโดดทีละ 2 หยุดก่อนถึง 10 -> 0, 2, 4, 6, 8
for i in range(0, 10, 2):
    print(i)
````

````python
# อีกตัวอย่างของ step: นับทีละ 5 จาก 5 ถึง 25
# ใช้ end เป็น 26 เพื่อให้รวม 25 ด้วย (จะเขียน 30 ก็ได้ เพราะ 30 เองไม่ถูกรวมอยู่แล้ว)
for i in range(5, 26, 5):
    print(i)
````

````python
# ถ้า step เป็นเลขติดลบ ตัวเลขจะ "ลดลง" (นับถอยหลัง)
# เริ่มจาก 5 ลดลงทีละ 1 หยุดก่อนถึง 0 -> 5, 4, 3, 2, 1
for i in range(5, 0, -1):
    print(i)

print("Lift off!")      # อยู่นอกลูป ทำงานครั้งเดียวตอนท้าย
````

````python
# ใช้ i ในการคำนวณ
# แสดงสูตรคูณแม่ 3 (3 x 1 ถึง 3 x 9)
for i in range(1, 10):           # i มีค่าตั้งแต่ 1 ถึง 9 (ไม่รวม 10)
    print(3, "x", i, "=", 3 * i)  # 3 * i ถูกคำนวณใหม่ด้วยค่า i ใหม่ทุกครั้ง
````

## 3. การบวกสะสมด้วยลูป

งานที่ลูปถูกใช้บ่อยที่สุดอย่างหนึ่งคือการ **บวกสะสม** ตัวเลข เช่น 1 + 2 + 3 + ... + 10 ได้เท่าไร?

เคล็ดลับคือใช้ตัวแปรหนึ่งตัวเป็น "ผลรวมสะสม"
1. ก่อนลูป สร้างตัวแปรและกำหนดค่าเป็น 0 (`total = 0`)
2. ในลูป บวกตัวเลขปัจจุบันเข้าไปทุกครั้ง (`total = total + i`)
3. หลังลูป ตัวแปรนั้นจะเก็บผลลัพธ์สุดท้ายไว้

จำ `age = age + 1` ในบทที่ 1 ได้ไหม? เป็นแนวคิดเดียวกันเลย: เอาค่าเดิมมา บวกอะไรบางอย่างเพิ่ม แล้วเก็บกลับเข้ากล่องใบเดิม

````python
# บวก 1 + 2 + 3 + ... + 10
total = 0                     # ขั้นที่ 1: ผลรวมสะสมเริ่มที่ 0 (ก่อนลูป)

for i in range(1, 11):        # i มีค่าตั้งแต่ 1 ถึง 10
    total = total + i         # ขั้นที่ 2: บวก i ปัจจุบันเข้าไปในผลรวมสะสม

print("Sum:", total)          # ขั้นที่ 3: หลังลูป แสดงผลลัพธ์สุดท้าย (อยู่นอกลูป จึงแสดงแค่ครั้งเดียว)
````

````python
# มาดูกันว่า total เปลี่ยนไปอย่างไรในแต่ละรอบ
# เป็นโค้ดเดียวกับข้างบน แต่เพิ่ม print() ไว้ในลูป
total = 0

for i in range(1, 6):         # i มีค่าตั้งแต่ 1 ถึง 5
    total = total + i
    print("i =", i, "-> total =", total)   # อยู่ในลูป จึงแสดงทุกครั้ง

# ครั้งที่ 1: 0 + 1 = 1
# ครั้งที่ 2: 1 + 2 = 3
# ครั้งที่ 3: 3 + 3 = 6
# ครั้งที่ 4: 6 + 4 = 10
# ครั้งที่ 5: 10 + 5 = 15
````

````python
# total = total + i เขียนให้สั้นลงได้: total += i
# สองบรรทัดนี้มีความหมายเหมือนกันทุกประการ นักเขียนโปรแกรม Python หลายคนนิยมใช้แบบสั้น
total = 0

for i in range(1, 101):       # i มีค่าตั้งแต่ 1 ถึง 100
    total += i                # เหมือนกับ: total = total + i

print("Sum of 1 to 100:", total)
````

````python
# ค่าสะสมไม่จำเป็นต้องเป็นผลบวกเสมอไป เราคูณก็ได้
# 1 x 2 x 3 x 4 x 5 (ในวิชาคณิตศาสตร์เรียกว่า "แฟกทอเรียล")
result = 1                    # สำหรับการคูณ ให้เริ่มที่ 1 ไม่ใช่ 0 (อะไรคูณ 0 ก็ได้ 0!)

for i in range(1, 6):         # i มีค่าตั้งแต่ 1 ถึง 5
    result = result * i       # คูณ i ปัจจุบันเข้าไปในผลลัพธ์

print("1 x 2 x 3 x 4 x 5 =", result)
````

````python
# การนับว่าลูปทำงานไปกี่ครั้งก็เป็นการสะสมแบบหนึ่ง
# คราวนี้เราบวก 1 ทุกครั้ง แทนที่จะบวก i
count = 0

for i in range(10, 20):       # i มีค่าตั้งแต่ 10 ถึง 19
    count += 1                # บวก 1 ทุกรอบ

print("The loop ran", count, "times.")   # range(10, 20) มีตัวเลข 10 ตัว ลูปจึงทำงาน 10 ครั้ง
````

## 4. การใช้ for และ if ร่วมกัน

เราสามารถใส่ `if` ไว้ **ข้างใน** ลูป `for` ได้ ทำให้ตรวจสอบตัวเลขได้ทีละตัว และทำบางอย่างเฉพาะกับตัวเลขที่เราต้องการ

```python
for i in range(...):
    if เงื่อนไข:
        โค้ดที่ทำงานเฉพาะเมื่อเงื่อนไขเป็น True
```

- `if` อยู่ใน `for` จึงย่อหน้า 4 ช่องว่าง
- โค้ดใน `if` ย่อหน้าเพิ่มอีกหนึ่งระดับ จึงย่อหน้า 8 ช่องว่าง

````python
# แสดงเฉพาะเลขคู่ตั้งแต่ 1 ถึง 10
# จำจากบทที่ 2: ตัวเลขเป็นเลขคู่ถ้า number % 2 == 0
for i in range(1, 11):        # ตรวจตัวเลขตั้งแต่ 1 ถึง 10 ทีละตัว
    if i % 2 == 0:            # i ตอนนี้เป็นเลขคู่ไหม? (4 ช่องว่าง: อยู่ใน for)
        print(i)              # แสดงเฉพาะเลขคู่ (8 ช่องว่าง: อยู่ใน if)
````

````python
# ผลลัพธ์เดียวกันนี้ทำได้ด้วย step ของ range() อย่างเดียว ไม่ต้องใช้ if
# เริ่มที่ 2 กระโดดทีละ 2 -> 2, 4, 6, 8, 10
for i in range(2, 11, 2):
    print(i)

# ทั้งสองวิธีถูกต้อง แต่ถ้าใช้ if เราตรวจเงื่อนไขอะไรก็ได้ ไม่ใช่แค่ "ทุก ๆ 2 ตัว"
````

````python
# ใช้ if - else ในลูป
for i in range(1, 6):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
````

````python
# ใช้ for + if + ผลรวมสะสม ร่วมกัน
# บวกเฉพาะพหุคูณของ 3 ตั้งแต่ 1 ถึง 20 (3 + 6 + 9 + 12 + 15 + 18)
total = 0

for i in range(1, 21):        # i มีค่าตั้งแต่ 1 ถึง 20
    if i % 3 == 0:            # i เป็นพหุคูณของ 3 ไหม? (หารด้วย 3 ลงตัว)
        total += i            # ถ้าใช่ จึงบวกเข้าไปในผลรวม

print("Sum of multiples of 3:", total)
````

````python
# แบบฝึกหัดคลาสสิก: FizzBuzz
# สำหรับตัวเลข 1 ถึง 15:
#   - ถ้าเป็นพหุคูณของทั้ง 3 และ 5 ให้แสดง "FizzBuzz"
#   - ถ้าเป็นพหุคูณของ 3 อย่างเดียว ให้แสดง "Fizz"
#   - ถ้าเป็นพหุคูณของ 5 อย่างเดียว ให้แสดง "Buzz"
#   - นอกเหนือจากนั้น ให้แสดงตัวเลขนั้นเลย
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:     # ตรวจเงื่อนไขที่เข้มงวดที่สุดก่อน (จำบทที่ 2: ลำดับสำคัญ!)
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

## 5. ลูป while

ลูป `for` เหมาะเมื่อเรารู้ว่าต้องทำซ้ำ **กี่ครั้ง** แต่บางครั้งเราไม่รู้จำนวนนั้นล่วงหน้า เรารู้แค่ว่า **จะหยุดเมื่อไร** ตัวอย่างเช่น:

- เก็บเงินต่อไป **ตราบใดที่** เงินรวมยังน้อยกว่า 10000
- ตัดกระดาษครึ่งหนึ่งต่อไป **ตราบใดที่** มันยังยาวกว่า 1 ซม.

กรณีแบบนี้เราใช้ลูป `while` คำว่า `while` หมายถึง "ตราบใดที่" มันจะทำซ้ำต่อไป **ตราบใดที่เงื่อนไขยังเป็น True** และหยุดเมื่อเงื่อนไขกลายเป็น False

```python
while เงื่อนไข:
    โค้ดที่ต้องการทำซ้ำ
```

วิธีการทำงาน:
1. ตรวจเงื่อนไข
2. ถ้าเป็น True ให้รันโค้ดที่ย่อหน้า แล้วกลับไปที่ขั้นที่ 1
3. ถ้าเป็น False ให้หยุดลูป แล้วไปทำโค้ดที่อยู่ถัดจากลูป

**สำคัญมาก:** ต้องมีบางอย่างในลูปที่เปลี่ยนไป เพื่อให้เงื่อนไขกลายเป็น False ในสักวัน ถ้าเงื่อนไขเป็น True ตลอดไป ลูปจะไม่มีวันจบ เรียกว่า **ลูปไม่รู้จบ (infinite loop)**

````python
# แสดง 1 ถึง 5 ด้วย while
# ใน while เราต้องสร้าง ตรวจสอบ และเปลี่ยนค่าตัวแปรเอง
i = 1                     # 1. สร้างตัวแปรก่อนลูป

while i <= 5:             # 2. ตรวจเงื่อนไข: ทำต่อไปตราบใดที่ i ยังน้อยกว่าหรือเท่ากับ 5
    print(i)
    i = i + 1             # 3. เปลี่ยนค่าตัวแปร ถ้าไม่มีบรรทัดนี้ i จะเป็น 1 ตลอดไป (ลูปไม่รู้จบ!)

print("Done. Now i is", i)    # เมื่อ i กลายเป็น 6 เงื่อนไข 6 <= 5 เป็น False ลูปจึงหยุด
````

````python
# เปรียบเทียบ: ทำสิ่งเดียวกันด้วย for
# for ทำงาน "สร้าง ตรวจสอบ เปลี่ยนค่า" ให้อัตโนมัติด้วย range()
for i in range(1, 6):
    print(i)

# ดังนั้นถ้ารู้ว่าต้องทำซ้ำกี่ครั้ง ใช้ for มักจะง่ายกว่า
````

````python
# กรณีที่ while เหมาะกว่า: เราไม่รู้ล่วงหน้าว่าต้องทำซ้ำกี่ครั้ง
# คุณเก็บเงินวันละ 3000 วอน ต้องใช้กี่วันจึงจะมีเงิน 20000 วอนขึ้นไป?
money = 0
days = 0

while money < 20000:          # เก็บเงินต่อไปตราบใดที่เงินยังน้อยกว่า 20000
    money += 3000             # เก็บเงิน 3000 วอน
    days += 1                 # ผ่านไปอีกหนึ่งวัน

print("Days needed:", days)
print("Money saved:", money)
````

````python
# อีกตัวอย่าง: แบ่งตัวเลขครึ่งหนึ่งไปเรื่อย ๆ จนกว่าจะเหลือ 1 หรือน้อยกว่า
# ใช้ // (เอาแค่ผลหาร จากบทที่ 1) เพื่อให้ตัวเลขยังเป็นจำนวนเต็ม
number = 100
steps = 0

while number > 1:             # ทำต่อไปตราบใดที่ number ยังมากกว่า 1
    number = number // 2      # แบ่งครึ่ง (100 -> 50 -> 25 -> 12 -> 6 -> 3 -> 1)
    steps += 1
    print("Step", steps, "->", number)

print("It took", steps, "steps.")
````

````python
# ถ้าเงื่อนไขเป็น False ตั้งแต่แรก ลูป while จะไม่ทำงานเลยสักครั้ง
i = 10

while i < 5:                  # 10 < 5 เป็น False ทันที
    print("This never prints.")
    i += 1

print("The loop was skipped.")
````

## 6. break และ continue

บางครั้งเราอยากเปลี่ยนการทำงานของลูประหว่างทาง

- `break`: **หยุดลูปทั้งหมดทันที** แล้วออกจากลูป
- `continue`: **ข้ามส่วนที่เหลือของรอบนี้** แล้วไปรอบถัดไปเลย

ทั้งสองมักใช้คู่กับ `if` เพราะเราต้องการหยุดหรือข้ามเฉพาะในสถานการณ์ที่กำหนด

````python
# break: หยุดลูปทันทีที่เจอสิ่งที่ตามหา
# หาตัวเลขตัวแรกตั้งแต่ 1 ถึง 100 ที่เป็นพหุคูณของ 7
for i in range(1, 101):
    if i % 7 == 0:                # i เป็นพหุคูณของ 7 ไหม?
        print("Found it:", i)
        break                     # หยุดลูปตรงนี้เลย ตัวเลขหลัง 7 ไม่ถูกตรวจเลย

print("The loop is over.")
````

````python
# มาดูกันว่า break หยุดลูปตรงไหนกันแน่
for i in range(1, 11):
    if i == 4:                    # เมื่อ i กลายเป็น 4...
        break                     # ...หยุดลูปทันที
    print(i)                      # จึงแสดงแค่ 1, 2, 3 (ไม่แสดง 4 เพราะ break ทำงานก่อน)
````

````python
# continue: ข้ามแค่รอบนี้ แล้วทำรอบถัดไปต่อ
for i in range(1, 6):
    if i == 3:                    # เมื่อ i เป็น 3...
        continue                  # ...ข้าม print() ข้างล่าง แล้วไปที่ i = 4 เลย
    print(i)                      # จึงแสดง 1, 2, 4, 5 (ขาดแค่ 3)
````

````python
# break กับ while: รูปแบบที่ใช้บ่อยคือ "while True" + break
# while True หมายถึง "ทำซ้ำตลอดไป" break จึงเป็นทางออกทางเดียว
count = 0

while True:                       # เงื่อนไขเป็น True เสมอ -> ถ้าไม่มี break จะเป็นลูปไม่รู้จบ
    count += 1
    print("count:", count)
    if count == 3:                # เมื่อ count ถึง 3...
        break                     # ...ออกจากลูป

print("Out of the loop!")
````

````python
# เปรียบเทียบ break กับ continue แบบเคียงข้างกัน
print("With break:")
for i in range(1, 6):
    if i == 3:
        break                     # หยุดทุกอย่างที่ 3
    print(i)                      # 1, 2

print("With continue:")
for i in range(1, 6):
    if i == 3:
        continue                  # ข้ามแค่ 3
    print(i)                      # 1, 2, 4, 5
````

## 7. ข้อผิดพลาดที่พบบ่อย

นี่คือข้อผิดพลาดที่เกือบทุกคนเคยทำตอนเรียนเรื่องลูป ถ้ามีอะไรผิดปกติ ให้ตรวจสอบข้อเหล่านี้ก่อน

1. **ลืมว่าตัวเลขสุดท้ายของ range() ไม่ถูกรวม**: `range(1, 10)` หยุดที่ 9 ถ้าต้องการรวม 10 ให้เขียน `range(1, 11)`
2. **ลืมโคลอน `:`**: บรรทัด `for` และ `while` ต้องลงท้ายด้วย `:`
3. **ย่อหน้าผิด**: เฉพาะบรรทัดที่ย่อหน้าเท่านั้นที่ถูกทำซ้ำ บรรทัดที่ควรอยู่ในลูปแต่ไม่ได้ย่อหน้าจะทำงานแค่ครั้งเดียว
4. **ลูปไม่รู้จบกับ while**: ถ้าลืมเปลี่ยนค่าตัวแปรในลูป `while` เงื่อนไขจะเป็น True ตลอดไป
5. **รีเซ็ตผลรวมสะสมในลูป**: `total = 0` ต้องเขียน **ก่อน** ลูป ไม่ใช่ในลูป

เซลล์ข้างล่างแสดงให้เห็นว่าข้อผิดพลาดแต่ละแบบเป็นอย่างไร โค้ดที่ผิดถูกเขียนไว้เป็นคอมเมนต์ เพื่อไม่ให้ notebook พัง

**ถ้าคุณเผลอรันลูปไม่รู้จบ** ให้กดปุ่ม ■ (หยุด / interrupt) ด้านบนของ Jupyter หรือกด `I` สองครั้ง (`I`, `I`) ขณะที่เลือกเซลล์นั้นอยู่

````python
# ข้อผิดพลาด 1: ลืมว่าตัวเลขสุดท้ายไม่ถูกรวม
# เราต้องการ 1 ถึง 5 แต่...
for i in range(1, 5):     # ผิด: ลูปนี้หยุดที่ 4!
    print(i)

print("---")

for i in range(1, 6):     # ถูกต้อง: end ต้องมากกว่าตัวเลขสุดท้ายที่ต้องการอยู่ 1
    print(i)
````

````python
# ข้อผิดพลาด 2: ลืมโคลอน

# for i in range(3)       # SyntaxError! ท้ายบรรทัดไม่มีโคลอน
#     print(i)

for i in range(3):        # ถูกต้อง: บรรทัดลงท้ายด้วยโคลอน
    print(i)
````

````python
# ข้อผิดพลาด 3: ย่อหน้าผิด
# print() ตัวที่สองไม่ได้ย่อหน้า จึงอยู่นอกลูปและทำงานแค่ครั้งเดียว
for i in range(3):
    print("Inside the loop:", i)
print("Outside the loop:", i)     # ทำงานครั้งเดียวหลังลูป (i ยังเก็บค่าสุดท้ายคือ 2 อยู่)

# for i in range(3):
# print(i)                # IndentationError! ไม่มีโค้ดที่ย่อหน้าอยู่ใน for เลย
````

````python
# ข้อผิดพลาด 4: ลูปไม่รู้จบกับ while
# อย่าลบ # ในตัวอย่างนี้! มันจะไม่มีวันหยุด

# i = 1
# while i <= 5:
#     print(i)            # i ไม่เคยเปลี่ยนค่า i <= 5 จึงเป็น True ตลอดไป -> ลูปไม่รู้จบ!

i = 1
while i <= 5:
    print(i)
    i += 1                # ถูกต้อง: เปลี่ยนค่า i ทุกครั้ง เพื่อให้เงื่อนไขกลายเป็น False ในที่สุด
````

````python
# ข้อผิดพลาด 5: รีเซ็ตผลรวมสะสมในลูป
# ผิด: total กลับไปเป็น 0 ทุกรอบ จึงเหลือแค่ตัวเลขตัวสุดท้าย
for i in range(1, 6):
    total = 0             # ผิดที่! บรรทัดนี้ทำงาน 5 ครั้ง
    total += i
print("Wrong total:", total)      # ได้แค่ 5

# ถูกต้อง: กำหนด total เป็น 0 แค่ครั้งเดียว ก่อนลูป
total = 0
for i in range(1, 6):
    total += i
print("Correct total:", total)    # 1 + 2 + 3 + 4 + 5 = 15
````
