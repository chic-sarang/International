## 1. for loop ဆိုတာဘာလဲ?

"Hello" ကို 5 ကြိမ် ပြရမယ်လို့ စဉ်းစားကြည့်ပါ။ `print("Hello")` ကို 5 ကြိမ် ရေးလို့ရပါတယ်။ ဒါပေမဲ့ 100 ကြိမ်ဆိုရင်ရော? 1000 ကြိမ်ဆိုရင်ရော? စာကြောင်းတစ်ကြောင်းတည်းကို ထပ်ခါထပ်ခါ ရေးရတာ ပင်ပန်းပြီး မှားဖို့လည်း လွယ်ပါတယ်။

**loop (ထပ်ခါထပ်ခါ လုပ်ခြင်း)** က ဒီပြဿနာကို ဖြေရှင်းပေးပါတယ်။ loop က code တစ်ခုတည်းကို လိုချင်သလောက် အကြိမ်ကြိမ် ထပ်ပြီး run ပေးပါတယ်။

Python မှာ အသုံးအများဆုံး loop က `for` loop ဖြစ်ပါတယ်။ အခြေခံပုံစံက ဒီလိုပါ:

```python
for i in range(5):
    ထပ်ခါထပ်ခါ run မယ့် code
```

ဒီလို ဖတ်လို့ရပါတယ်: **"range(5) ထဲက ဂဏန်း i တစ်ခုချင်းစီအတွက် အောက်က code ကို run ပါ"**။

- `range(5)` က 0, 1, 2, 3, 4 ဂဏန်းတွေကို ပေးပါတယ် (စုစုပေါင်း 5 လုံး)။ `range()` ကို နောက်အပိုင်းမှာ အသေးစိတ် လေ့လာပါမယ်။
- `i` က variable တစ်ခုပါ။ loop တစ်ကြိမ် ပြန်လုပ်တိုင်း နောက်ဂဏန်းကို `i` ထဲ သိမ်းပါတယ်။
- `if` လိုပဲ `for` စာကြောင်းက colon `:` နဲ့ ဆုံးရပါမယ်။
- `if` လိုပဲ ထပ်ခါထပ်ခါ run မယ့် code ကို **indent** (space 4 ခု) လုပ်ရပါမယ်။ indent လုပ်ထားတဲ့ စာကြောင်းတွေကိုပဲ ထပ်ခါထပ်ခါ run ပါတယ်။

````python
# print("Hello") ကို 5 ကြိမ် ထပ်လုပ်ခြင်း
# range(5) က ဂဏန်း 5 လုံး (0, 1, 2, 3, 4) ပေးလို့ indent လုပ်ထားတဲ့ code က 5 ကြိမ် run ပါတယ်
for i in range(5):
    print("Hello")      # ဒီစာကြောင်းက for loop ထဲမှာ ရှိလို့ (indent လုပ်ထားလို့) ထပ်ခါထပ်ခါ run ပါတယ်
````

````python
# အကြိမ်တိုင်း i ထဲမှာ ဘာသိမ်းထားလဲ ကြည့်ရအောင်
# 1 ကြိမ်မြောက်: i = 0, 2 ကြိမ်မြောက်: i = 1, 3 ကြိမ်မြောက်: i = 2, 4 ကြိမ်မြောက်: i = 3, 5 ကြိမ်မြောက်: i = 4
for i in range(5):
    print(i)            # i ရဲ့ လက်ရှိ တန်ဖိုးကို ပြပါတယ်
````

````python
# indent လုပ်ထားတဲ့ code ကို တစ်ပတ်လုံး run တာကို "တစ်ကြိမ် ထပ်လုပ်ခြင်း" လို့ ခေါ်ပါတယ်
# indent လုပ်ထားတဲ့ စာကြောင်းတိုင်းက တစ်ကြိမ်ထပ်လုပ်တိုင်း အပေါ်ကနေ အောက်ကို အစဉ်လိုက် တစ်ခါစီ run ပါတယ်
for i in range(3):
    print("Repetition number", i)     # comma နဲ့ ခွဲပေးရင် print() က တန်ဖိုး အများကြီးကို တစ်ခါတည်း ပြနိုင်ပါတယ်
    print("----------")               # ဒီစာကြောင်းလည်း loop ထဲမှာ ရှိပါတယ် (indent တူတူ)

print("The loop is finished.")        # indent မလုပ်ထားလို့ loop ရဲ့ အပြင်မှာ ရှိပါတယ်။ loop ပြီးမှ တစ်ကြိမ်ပဲ run ပါတယ်
````

````python
# variable နာမည်က i ဖြစ်စရာ မလိုပါဘူး
# i က လူအများစု သုံးလေ့ရှိတဲ့ နာမည်တစ်ခုပဲ ဖြစ်ပါတယ်။ variable နာမည်ပေးစည်းမျဉ်းနဲ့ ကိုက်ရင် ကြိုက်တဲ့ နာမည် ပေးလို့ရပါတယ်
for number in range(3):
    print(number)

for count in range(3):
    print("count is", count)
````

## 2. range() ကို အသေးစိတ် လေ့လာခြင်း

`range()` က ဆက်တိုက် ဂဏန်းတန်းတစ်ခုကို ဖန်တီးပေးပါတယ်။ သုံးနည်း ၃ မျိုး ရှိပါတယ်။

| ရေးပုံ | အဓိပ္ပာယ် | ရလာမယ့် ဂဏန်းများ |
|---|---|---|
| `range(end)` | 0 ကနေ end မတိုင်ခင်အထိ (end မပါ) | `range(5)` → 0, 1, 2, 3, 4 |
| `range(start, end)` | start ကနေ end မတိုင်ခင်အထိ (end မပါ) | `range(1, 6)` → 1, 2, 3, 4, 5 |
| `range(start, end, step)` | start ကနေ end မတိုင်ခင်အထိ step အလိုက် ခုန်ပြီး | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |

မှတ်ထားရမယ့် အရေးအကြီးဆုံး အချက်:
- **နောက်ဆုံးဂဏန်း (end) က ဘယ်တော့မှ မပါဝင်ပါဘူး။** `range(5)` က 5 မှာ မဟုတ်ဘဲ 4 မှာ ရပ်ပါတယ်။
- စမယ့် ဂဏန်း မပေးရင် **0** ကနေ စပါတယ်။
- step မပေးရင် **1** စီ တိုးပါတယ်။

````python
# range(end): 0 ကနေ စပြီး end မတိုင်ခင်မှာ ရပ်ပါတယ်
# range(4) က 0, 1, 2, 3 ကို ပေးပါတယ် -> 4 မပါဝင်ပါဘူး
for i in range(4):
    print(i)
````

````python
# range(start, end): start ကနေ စပြီး end မတိုင်ခင်မှာ ရပ်ပါတယ်
# 1 ကနေ 5 အထိ လိုချင်ရင် end က 6 ဖြစ်ရပါမယ် (နောက်ဆုံးဂဏန်း မပါဝင်လို့ပါ)
for i in range(1, 6):
    print(i)
````

````python
# range(start, end, step): တစ်ခါစီ step အလိုက် ခုန်ပါတယ်
# 0 ကနေ စပြီး 2 စီ ခုန်ကာ 10 မတိုင်ခင် ရပ်ပါတယ် -> 0, 2, 4, 6, 8
for i in range(0, 10, 2):
    print(i)
````

````python
# step ရဲ့ နောက်ထပ် ဥပမာ: 5 ကနေ 25 အထိ 5 စီ ရေတွက်ခြင်း
# 25 ပါဝင်အောင် end ကို 26 ထားပါတယ် (30 လို့ ရေးလည်း ရပါတယ်၊ 30 ကိုယ်တိုင်က မပါဝင်လို့ပါ)
for i in range(5, 26, 5):
    print(i)
````

````python
# step က အနုတ်ကိန်း ဖြစ်ရင် ဂဏန်းတွေက "လျော့" သွားပါတယ် (နောက်ပြန် ရေတွက်ခြင်း)
# 5 ကနေ စပြီး 1 စီ လျော့ကာ 0 မတိုင်ခင် ရပ်ပါတယ် -> 5, 4, 3, 2, 1
for i in range(5, 0, -1):
    print(i)

print("Lift off!")      # loop ရဲ့ အပြင်မှာ ရှိပြီး နောက်ဆုံးမှာ တစ်ကြိမ် run ပါတယ်
````

````python
# i ကို တွက်ချက်မှုမှာ သုံးခြင်း
# 3 ဆခွက် (3 x 1 ကနေ 3 x 9 အထိ) ကို ပြပါမယ်
for i in range(1, 10):           # i က 1 ကနေ 9 အထိ (10 မပါဝင်ပါ)
    print(3, "x", i, "=", 3 * i)  # တစ်ခါစီ i အသစ်နဲ့ 3 * i ကို ပြန်တွက်ပါတယ်
````

## 3. loop နဲ့ ပေါင်းစုခြင်း

loop ရဲ့ အသုံးအများဆုံး အလုပ်တစ်ခုက ဂဏန်းတွေကို **ပေါင်းစုခြင်း** ဖြစ်ပါတယ်။ ဥပမာ 1 + 2 + 3 + ... + 10 က ဘယ်လောက်လဲ?

နည်းလမ်းက variable တစ်ခုကို "စုစုပေါင်း ပေါင်းစုထားတဲ့ တန်ဖိုး" အဖြစ် သုံးတာပါ။
1. loop မစခင် variable တစ်ခု ဖန်တီးပြီး 0 ထည့်ပါ။ (`total = 0`)
2. loop ထဲမှာ အကြိမ်တိုင်း လက်ရှိ ဂဏန်းကို အဲဒီထဲ ပေါင်းထည့်ပါ။ (`total = total + i`)
3. loop ပြီးတဲ့အခါ အဲဒီ variable ထဲမှာ နောက်ဆုံး ရလဒ် ရှိနေပါမယ်။

သင်ခန်းစာ 1 က `age = age + 1` ကို မှတ်မိသေးလား? အတူတူပါပဲ: တန်ဖိုးဟောင်းကို ယူ၊ တစ်ခုခု ပေါင်းထည့်ပြီး သေတ္တာ အဟောင်းထဲကို ပြန်သိမ်းပါတယ်။

````python
# 1 + 2 + 3 + ... + 10 ကို ပေါင်းခြင်း
total = 0                     # အဆင့် 1: ပေါင်းစုတန်ဖိုးက 0 ကနေ စပါတယ် (loop မစခင်)

for i in range(1, 11):        # i က 1 ကနေ 10 အထိ
    total = total + i         # အဆင့် 2: လက်ရှိ i ကို ပေါင်းစုတန်ဖိုးထဲ ပေါင်းထည့်ပါတယ်

print("Sum:", total)          # အဆင့် 3: loop ပြီးမှ နောက်ဆုံး ရလဒ်ကို ပြပါတယ် (loop အပြင်မှာ ရှိလို့ တစ်ကြိမ်ပဲ ပြပါတယ်)
````

````python
# အကြိမ်တိုင်းမှာ total ဘယ်လို ပြောင်းလဲလဲ ကြည့်ရအောင်
# အပေါ်က code နဲ့ အတူတူပါပဲ၊ ဒါပေမဲ့ loop ထဲမှာ print() ထည့်ထားပါတယ်
total = 0

for i in range(1, 6):         # i က 1 ကနေ 5 အထိ
    total = total + i
    print("i =", i, "-> total =", total)   # loop ထဲမှာ ရှိလို့ အကြိမ်တိုင်း ပြပါတယ်

# 1 ကြိမ်မြောက်: 0 + 1 = 1
# 2 ကြိမ်မြောက်: 1 + 2 = 3
# 3 ကြိမ်မြောက်: 3 + 3 = 6
# 4 ကြိမ်မြောက်: 6 + 4 = 10
# 5 ကြိမ်မြောက်: 10 + 5 = 15
````

````python
# total = total + i ကို ပိုတိုတိုနဲ့ ရေးလို့ရပါတယ်: total += i
# စာကြောင်း နှစ်ကြောင်းက အဓိပ္ပာယ် လုံးဝ တူပါတယ်။ Python programmer အများစုက အတိုပုံစံကို သုံးကြပါတယ်
total = 0

for i in range(1, 101):       # i က 1 ကနေ 100 အထိ
    total += i                # အဓိပ္ပာယ်တူ: total = total + i

print("Sum of 1 to 100:", total)
````

````python
# ပေါင်းစုတန်ဖိုးက ပေါင်းလဒ် ဖြစ်စရာ မလိုပါဘူး။ မြှောက်လို့လည်း ရပါတယ်
# 1 x 2 x 3 x 4 x 5 (သင်္ချာမှာ "factorial" လို့ ခေါ်ပါတယ်)
result = 1                    # မြှောက်တဲ့အခါ 0 မဟုတ်ဘဲ 1 ကနေ စပါ (ဘာကိုပဲ 0 နဲ့ မြှောက်မြှောက် 0 ပဲ ရပါတယ်!)

for i in range(1, 6):         # i က 1 ကနေ 5 အထိ
    result = result * i       # လက်ရှိ i ကို ရလဒ်ထဲ မြှောက်ထည့်ပါတယ်

print("1 x 2 x 3 x 4 x 5 =", result)
````

````python
# loop ဘယ်နှစ်ကြိမ် run ခဲ့လဲ ရေတွက်တာလည်း ပေါင်းစုခြင်း တစ်မျိုးပါပဲ
# ဒီမှာ i ကို ပေါင်းမယ့်အစား အကြိမ်တိုင်း 1 ကို ပေါင်းပါတယ်
count = 0

for i in range(10, 20):       # i က 10 ကနေ 19 အထိ
    count += 1                # အကြိမ်တိုင်း 1 ပေါင်းပါတယ်

print("The loop ran", count, "times.")   # range(10, 20) မှာ ဂဏန်း 10 လုံး ရှိလို့ 10 ကြိမ် run ခဲ့ပါတယ်
````

## 4. for နဲ့ if ကို တွဲသုံးခြင်း

`for` loop ရဲ့ **အထဲမှာ** `if` ကို ထည့်လို့ရပါတယ်။ ဒါဆိုရင် ဂဏန်းတစ်ခုချင်းစီကို စစ်ပြီး လိုချင်တဲ့ ဂဏန်းတွေအတွက်ပဲ တစ်ခုခု လုပ်လို့ရပါတယ်။

```python
for i in range(...):
    if condition:
        condition က True ဖြစ်မှသာ run မယ့် code
```

- `if` က `for` ထဲမှာ ရှိလို့ space 4 ခု indent လုပ်ထားပါတယ်။
- `if` ထဲက code က တစ်ဆင့် ပိုပြီး indent လုပ်ထားလို့ space 8 ခု indent လုပ်ထားပါတယ်။

````python
# 1 ကနေ 10 အထိ စုံကိန်းတွေကိုပဲ ပြခြင်း
# သင်ခန်းစာ 2 ကို သတိရပါ: number % 2 == 0 ဖြစ်ရင် စုံကိန်းပါ
for i in range(1, 11):        # 1 ကနေ 10 အထိ ဂဏန်းတစ်ခုချင်းစီကို စစ်ပါတယ်
    if i % 2 == 0:            # လက်ရှိ i က စုံကိန်းလား? (space 4 ခု: for ထဲမှာ)
        print(i)              # စုံကိန်းတွေပဲ ပြပါတယ် (space 8 ခု: if ထဲမှာ)
````

````python
# if မသုံးဘဲ range() ရဲ့ step တစ်ခုတည်းနဲ့လည်း ရလဒ် အတူတူ ရနိုင်ပါတယ်
# 2 ကနေ စပြီး 2 စီ ခုန်ပါတယ် -> 2, 4, 6, 8, 10
for i in range(2, 11, 2):
    print(i)

# နည်းလမ်း နှစ်ခုလုံး မှန်ပါတယ်။ ဒါပေမဲ့ if သုံးရင် "2 လုံးခြား တစ်လုံး" သာမက ကြိုက်တဲ့ condition ကို စစ်လို့ရပါတယ်
````

````python
# loop ထဲမှာ if - else ကို သုံးခြင်း
for i in range(1, 6):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
````

````python
# for + if + ပေါင်းစုတန်ဖိုး ကို တွဲသုံးခြင်း
# 1 ကနေ 20 အထိထဲက 3 ရဲ့ ဆတိုးကိန်းတွေကိုပဲ ပေါင်းပါမယ် (3 + 6 + 9 + 12 + 15 + 18)
total = 0

for i in range(1, 21):        # i က 1 ကနေ 20 အထိ
    if i % 3 == 0:            # i က 3 ရဲ့ ဆတိုးကိန်းလား? (3 နဲ့ စားရင် အကြွင်းမရှိ)
        total += i            # ဟုတ်မှသာ total ထဲ ပေါင်းထည့်ပါတယ်

print("Sum of multiples of 3:", total)
````

````python
# ဂန္ထဝင် လေ့ကျင့်ခန်း: FizzBuzz
# 1 ကနေ 15 အထိ ဂဏန်းတွေအတွက်:
#   - 3 ရော 5 ရဲ့ ဆတိုးကိန်း ဖြစ်ရင် "FizzBuzz" လို့ ပြပါ
#   - 3 ရဲ့ ဆတိုးကိန်း တစ်ခုတည်း ဖြစ်ရင် "Fizz" လို့ ပြပါ
#   - 5 ရဲ့ ဆတိုးကိန်း တစ်ခုတည်း ဖြစ်ရင် "Buzz" လို့ ပြပါ
#   - မဟုတ်ရင် ဂဏန်းကိုယ်တိုင်ကို ပြပါ
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:     # အတင်းကျပ်ဆုံး condition ကို အရင်စစ်ပါ (သင်ခန်းစာ 2 ကို သတိရပါ: အစီအစဉ်က အရေးကြီးပါတယ်!)
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

## 5. while loop

`for` loop က **ဘယ်နှစ်ကြိမ်** ထပ်လုပ်ရမလဲ သိတဲ့အခါ သင့်တော်ပါတယ်။ ဒါပေမဲ့ တစ်ခါတလေ အဲဒီအရေအတွက်ကို ကြိုမသိပါဘူး။ **ဘယ်အချိန် ရပ်ရမလဲ** ဆိုတာပဲ သိပါတယ်။ ဥပမာ:

- စုစုပေါင်း ငွေက 10000 ထက် နည်းနေ **သရွေ့** ငွေဆက်စုပါ။
- စက္ကူက 1 cm ထက် ရှည်နေ **သရွေ့** တစ်ဝက်စီ ဆက်ဖြတ်ပါ။

ဒီလိုအခါ `while` loop ကို သုံးပါတယ်။ `while` ဆိုတာ "... နေသရွေ့" လို့ အဓိပ္ပာယ်ရပါတယ်။ **condition က True ဖြစ်နေသရွေ့** ထပ်ခါထပ်ခါ လုပ်ပြီး condition က False ဖြစ်သွားတဲ့အခါ ရပ်ပါတယ်။

```python
while condition:
    ထပ်ခါထပ်ခါ run မယ့် code
```

အလုပ်လုပ်ပုံ:
1. condition ကို စစ်ပါတယ်။
2. True ဖြစ်ရင် indent လုပ်ထားတဲ့ code ကို run ပြီး အဆင့် 1 ကို ပြန်သွားပါတယ်။
3. False ဖြစ်ရင် loop ကို ရပ်ပြီး loop နောက်က code ကို ဆက်သွားပါတယ်။

**အရမ်းအရေးကြီးပါတယ်:** loop ထဲမှာ တစ်ခုခု ပြောင်းလဲနေရပါမယ်၊ ဒါမှ တစ်ချိန်ချိန်မှာ condition က False ဖြစ်လာမှာပါ။ condition က အမြဲတမ်း True ဖြစ်နေရင် loop က ဘယ်တော့မှ မဆုံးပါဘူး။ ဒါကို **infinite loop (အဆုံးမရှိ loop)** လို့ ခေါ်ပါတယ်။

````python
# while ကို သုံးပြီး 1 ကနေ 5 အထိ ပြခြင်း
# while မှာ variable ကို ကိုယ်တိုင် ဖန်တီး၊ စစ်ဆေး၊ ပြောင်းလဲရပါတယ်
i = 1                     # 1. loop မစခင် variable ကို ဖန်တီးပါ

while i <= 5:             # 2. condition ကို စစ်ပါ: i က 5 နဲ့ ညီ သို့မဟုတ် ငယ်နေသရွေ့ ဆက်လုပ်ပါ
    print(i)
    i = i + 1             # 3. variable ကို ပြောင်းပါ။ ဒီစာကြောင်း မရှိရင် i က အမြဲ 1 ဖြစ်နေပါမယ် (infinite loop!)

print("Done. Now i is", i)    # i က 6 ဖြစ်သွားတဲ့အခါ 6 <= 5 က False ဖြစ်လို့ loop ရပ်သွားပါတယ်
````

````python
# နှိုင်းယှဉ်ကြည့်ပါ: အတူတူကို for နဲ့ ရေးခြင်း
# for က range() နဲ့ "ဖန်တီး၊ စစ်ဆေး၊ ပြောင်းလဲ" အလုပ်ကို အလိုအလျောက် လုပ်ပေးပါတယ်
for i in range(1, 6):
    print(i)

# ဒါကြောင့် ဘယ်နှစ်ကြိမ် ထပ်လုပ်ရမလဲ သိရင် for က များသောအားဖြင့် ပိုရိုးရှင်းပါတယ်
````

````python
# while က ပိုသင့်တော်တဲ့ အခြေအနေ: ဘယ်နှစ်ကြိမ် ထပ်လုပ်ရမလဲ ကြိုမသိပါဘူး
# နေ့တိုင်း 3000 won စုပါတယ်။ 20000 won နဲ့အထက် ရဖို့ ဘယ်နှစ်ရက် လိုလဲ?
money = 0
days = 0

while money < 20000:          # ငွေက 20000 ထက် နည်းနေသရွေ့ ဆက်စုပါ
    money += 3000             # 3000 won စုပါတယ်
    days += 1                 # နောက်တစ်ရက် ကုန်သွားပါပြီ

print("Days needed:", days)
print("Money saved:", money)
````

````python
# နောက်ထပ် ဥပမာ: ဂဏန်းကို 1 သို့မဟုတ် အောက် ရောက်တဲ့အထိ တစ်ဝက်စီ ဆက်ခွဲခြင်း
# ဂဏန်းက integer အဖြစ် ဆက်ရှိနေအောင် // (စားလဒ်ပဲယူ၊ သင်ခန်းစာ 1 က) ကို သုံးပါတယ်
number = 100
steps = 0

while number > 1:             # number က 1 ထက် ကြီးနေသရွေ့ ဆက်လုပ်ပါ
    number = number // 2      # တစ်ဝက် ခွဲပါတယ် (100 -> 50 -> 25 -> 12 -> 6 -> 3 -> 1)
    steps += 1
    print("Step", steps, "->", number)

print("It took", steps, "steps.")
````

````python
# condition က အစကတည်းက False ဖြစ်နေရင် while loop က တစ်ကြိမ်မှ မ run ပါဘူး
i = 10

while i < 5:                  # 10 < 5 က ချက်ချင်း False ပါ
    print("This never prints.")
    i += 1

print("The loop was skipped.")
````

## 6. break နဲ့ continue

တစ်ခါတလေ loop ရဲ့ လုပ်ဆောင်ပုံကို လမ်းခုလတ်မှာ ပြောင်းချင်ပါတယ်။

- `break`: **loop တစ်ခုလုံးကို ချက်ချင်း ရပ်ပြီး** အပြင်ကို ထွက်ပါတယ်။
- `continue`: **ဒီတစ်ကြိမ်ရဲ့ ကျန်တဲ့အပိုင်းကို ကျော်ပြီး** နောက်တစ်ကြိမ်ကို တန်းသွားပါတယ်။

သတ်မှတ်ထားတဲ့ အခြေအနေမှာပဲ ရပ်ချင် ကျော်ချင်လို့ နှစ်ခုလုံးကို များသောအားဖြင့် `if` နဲ့ တွဲသုံးပါတယ်။

````python
# break: ရှာနေတာကို တွေ့တာနဲ့ loop ကို ရပ်ပါတယ်
# 1 ကနေ 100 အထိထဲက 7 ရဲ့ ဆတိုးကိန်း ဖြစ်တဲ့ ပထမဆုံး ဂဏန်းကို ရှာပါမယ်
for i in range(1, 101):
    if i % 7 == 0:                # i က 7 ရဲ့ ဆတိုးကိန်းလား?
        print("Found it:", i)
        break                     # loop ကို ဒီမှာပဲ ရပ်ပါတယ်။ 7 နောက်က ဂဏန်းတွေကို လုံးဝ မစစ်တော့ပါဘူး

print("The loop is over.")
````

````python
# break က loop ကို ဘယ်နေရာမှာ အတိအကျ ရပ်လဲ ကြည့်ရအောင်
for i in range(1, 11):
    if i == 4:                    # i က 4 ဖြစ်လာတဲ့အခါ...
        break                     # ...loop ကို ချက်ချင်း ရပ်ပါတယ်
    print(i)                      # ဒါကြောင့် 1, 2, 3 ပဲ ပြပါတယ် (break က အရင် အလုပ်လုပ်လို့ 4 ကို မပြပါဘူး)
````

````python
# continue: ဒီတစ်ကြိမ်ကိုပဲ ကျော်ပြီး နောက်တစ်ကြိမ်ကို ဆက်လုပ်ပါတယ်
for i in range(1, 6):
    if i == 3:                    # i က 3 ဖြစ်တဲ့အခါ...
        continue                  # ...အောက်က print() ကို ကျော်ပြီး i = 4 ကို တန်းသွားပါတယ်
    print(i)                      # ဒါကြောင့် 1, 2, 4, 5 ကို ပြပါတယ် (3 တစ်ခုပဲ မပါပါဘူး)
````

````python
# while နဲ့ break: အသုံးများတဲ့ ပုံစံက "while True" + break ပါ
# while True ဆိုတာ "အမြဲတမ်း ထပ်လုပ်ပါ" လို့ အဓိပ္ပာယ်ရလို့ break က ထွက်ဖို့ တစ်ခုတည်းသော နည်းလမ်းပါ
count = 0

while True:                       # condition က အမြဲ True -> break မရှိရင် infinite loop ဖြစ်ပါမယ်
    count += 1
    print("count:", count)
    if count == 3:                # count က 3 ရောက်တဲ့အခါ...
        break                     # ...loop ထဲက ထွက်ပါတယ်

print("Out of the loop!")
````

````python
# break နဲ့ continue ကို ဘေးချင်းယှဉ်ပြီး နှိုင်းယှဉ်ခြင်း
print("With break:")
for i in range(1, 6):
    if i == 3:
        break                     # 3 မှာ အားလုံးကို ရပ်ပါတယ်
    print(i)                      # 1, 2

print("With continue:")
for i in range(1, 6):
    if i == 3:
        continue                  # 3 ကိုပဲ ကျော်ပါတယ်
    print(i)                      # 1, 2, 4, 5
````

## 7. မကြာခဏ ဖြစ်တတ်သော အမှားများ

loop ကို လေ့လာတဲ့အခါ လူတိုင်းနီးပါး ဒီအမှားတွေ လုပ်မိကြပါတယ်။ တစ်ခုခု မှားနေရင် ဒါတွေကို အရင်စစ်ကြည့်ပါ။

1. **range() ရဲ့ နောက်ဆုံးဂဏန်း မပါဝင်တာကို မေ့ခြင်း**: `range(1, 10)` က 9 မှာ ရပ်ပါတယ်။ 10 ပါဝင်စေချင်ရင် `range(1, 11)` လို့ ရေးပါ။
2. **colon `:` မေ့ခြင်း**: `for` နဲ့ `while` စာကြောင်းတွေက `:` နဲ့ ဆုံးရပါမယ်။
3. **indent မှားခြင်း**: indent လုပ်ထားတဲ့ စာကြောင်းတွေကိုပဲ ထပ်ခါထပ်ခါ run ပါတယ်။ loop ထဲမှာ ရှိသင့်ပေမဲ့ indent မလုပ်ထားတဲ့ စာကြောင်းက တစ်ကြိမ်ပဲ run ပါတယ်။
4. **while နဲ့ infinite loop ဖြစ်ခြင်း**: `while` loop ထဲမှာ variable ကို ပြောင်းဖို့ မေ့ရင် condition က အမြဲတမ်း True ဖြစ်နေပါမယ်။
5. **ပေါင်းစုတန်ဖိုးကို loop ထဲမှာ ပြန်စခြင်း**: `total = 0` ကို loop ထဲမှာ မဟုတ်ဘဲ loop **မစခင်** ရေးရပါမယ်။

အောက်က cell တွေမှာ အမှားတစ်ခုချင်းစီ ဘယ်လိုပုံစံလဲ ပြထားပါတယ်။ notebook မပျက်အောင် အမှား code ကို comment အဖြစ် ရေးထားပါတယ်။

**မတော်တဆ infinite loop ကို run မိသွားရင်** Jupyter အပေါ်ဘက်က ■ (ရပ် / interrupt) ခလုတ်ကို နှိပ်ပါ၊ ဒါမှမဟုတ် cell ကို ရွေးထားချိန်မှာ `I` ကို နှစ်ခါ (`I`, `I`) နှိပ်ပါ။

````python
# အမှား 1: နောက်ဆုံးဂဏန်း မပါဝင်တာကို မေ့ခြင်း
# 1 ကနေ 5 အထိ လိုချင်တာပါ၊ ဒါပေမဲ့...
for i in range(1, 5):     # မှားပါတယ်: ဒါက 4 မှာ ရပ်ပါတယ်!
    print(i)

print("---")

for i in range(1, 6):     # မှန်ပါတယ်: end က လိုချင်တဲ့ နောက်ဆုံးဂဏန်းထက် 1 ပိုရပါမယ်
    print(i)
````

````python
# အမှား 2: colon မေ့ခြင်း

# for i in range(3)       # SyntaxError! အဆုံးမှာ colon ကျန်နေပါတယ်
#     print(i)

for i in range(3):        # မှန်ပါတယ်: စာကြောင်းက colon နဲ့ ဆုံးပါတယ်
    print(i)
````

````python
# အမှား 3: indent မှားခြင်း
# ဒုတိယ print() ကို indent မလုပ်ထားလို့ loop ရဲ့ အပြင်မှာ ရှိပြီး တစ်ကြိမ်ပဲ run ပါတယ်
for i in range(3):
    print("Inside the loop:", i)
print("Outside the loop:", i)     # loop ပြီးမှ တစ်ကြိမ် run ပါတယ်။ (i က နောက်ဆုံးတန်ဖိုး 2 ကို သိမ်းထားဆဲပါ)

# for i in range(3):
# print(i)                # IndentationError! for ထဲမှာ indent လုပ်ထားတဲ့ code လုံးဝ မရှိပါဘူး
````

````python
# အမှား 4: while နဲ့ infinite loop ဖြစ်ခြင်း
# ဒီဥပမာမှာ # ကို မဖျက်ပါနဲ့! ဘယ်တော့မှ မရပ်ပါဘူး။

# i = 1
# while i <= 5:
#     print(i)            # i က ဘယ်တော့မှ မပြောင်းလို့ i <= 5 က အမြဲ True -> infinite loop!

i = 1
while i <= 5:
    print(i)
    i += 1                # မှန်ပါတယ်: အကြိမ်တိုင်း i ကို ပြောင်းလို့ နောက်ဆုံးမှာ condition က False ဖြစ်လာပါတယ်
````

````python
# အမှား 5: ပေါင်းစုတန်ဖိုးကို loop ထဲမှာ ပြန်စခြင်း
# မှားပါတယ်: အကြိမ်တိုင်း total က 0 ပြန်ဖြစ်သွားလို့ နောက်ဆုံးဂဏန်းပဲ ကျန်ပါတယ်
for i in range(1, 6):
    total = 0             # နေရာမှားပါတယ်! ဒီစာကြောင်းက 5 ကြိမ် run ပါတယ်
    total += i
print("Wrong total:", total)      # 5 ပဲ ရပါတယ်

# မှန်ပါတယ်: total ကို loop မစခင် တစ်ကြိမ်တည်း 0 ထားပါ
total = 0
for i in range(1, 6):
    total += i
print("Correct total:", total)    # 1 + 2 + 3 + 4 + 5 = 15
````
