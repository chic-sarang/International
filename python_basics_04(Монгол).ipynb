## 1. Жагсаалт (list) гэж юу вэ?

Хичээл 1-д бид хувьсагч нь **нэг** утга хадгалдаг хайрцагтай адил гэж сурсан.

Гэтэл 5 сурагчийн шалгалтын оноог хадгалахыг хүсвэл яах вэ? Одоогийн мэдлэгээрээ бол 5 хувьсагч хэрэгтэй болно:

```python
score1 = 85
score2 = 92
score3 = 78
score4 = 90
score5 = 66
```

Энэ нь аль хэдийн их байна. Харин 30 сурагч байвал? Эсвэл 300 бол?

**Жагсаалт (list)** энэ асуудлыг шийддэг. Жагсаалт бол **олон утгыг дарааллаар нь** хадгалж чаддаг ганц хувьсагч юм.

Жагсаалтыг эгнээ **шүүгээ (locker)** гэж төсөөлж болно:
- Бүх эгнээ шүүгээ нэг нэртэй (жагсаалтын нэр).
- Шүүгээ бүрд нэг утга хадгалагдана.
- Шүүгээ бүр дугаартай тул олоход хялбар.

Жагсаалт үүсгэх арга:
- Утгуудыг дөрвөлжин хаалт `[ ]` дотор бичнэ.
- Утгуудыг таслал `,`-аар тусгаарлана.

```python
scores = [85, 92, 78, 90, 66]
```

````python
# Бүхэл тоонуудын жагсаалт
# Одоо нэг хувьсагч (scores) 5 утга хадгалж байна
scores = [85, 92, 78, 90, 66]
print(scores)           # print() нь дөрвөлжин хаалттай нь хамт бүх жагсаалтыг харуулна
````

````python
# Жагсаалт нь Хичээл 1-д сурсан ямар ч өгөгдлийн төрлийг хадгалж чадна
fruits = ["apple", "banana", "cherry"]      # Тэмдэгт мөрүүдийн (str) жагсаалт
heights = [175.5, 162.3, 180.1]             # Бутархай тоонуудын (float) жагсаалт
answers = [True, False, True, True]         # bool утгуудын жагсаалт

print(fruits)
print(heights)
print(answers)
````

````python
# Нэг жагсаалтад өөр өөр төрлийн өгөгдлийг хольж болно
# Энэ нь зөвшөөрөгдөнө, гэхдээ ихэвчлэн жагсаалт нэг л төрлийн утгатай байвал ойлгоход хялбар
student = ["Bill", 20, 175.5, True]         # нэр, нас, өндөр, сурагч эсэх
print(student)
````

````python
# Жагсаалт хоосон (ямар ч утгагүй) байж болно
# Хоосноос эхлээд дараа нь утга нэмэхийг хүсвэл хэрэгтэй (хэрхэн нэмэхийг 5-р хэсэгт сурна)
empty_list = []
print(empty_list)
````

````python
# Жагсаалт бол шинэ өгөгдлийн төрөл: list
# type() нь утгын өгөгдлийн төрлийг хэлж өгдөг
print(type(scores))     # <class 'list'> нь "энэ бол жагсаалт" гэсэн утгатай
print(type(85))         # <class 'int'> нь "энэ бол бүхэл тоо" гэсэн утгатай (жагсаалт доторх утгуудын нэг)
````

## 2. Индекс: байрлалаар нь утгыг олох

Жагсаалтын утга бүр байрлалын дугаартай. Энэ дугаарыг **индекс (index)** гэдэг. Шүүгээ бүр дээр бичсэн дугаар гэж төсөөлөөрэй.

Хамгийн чухал санах зүйл: **Python 1-ээс биш 0-ээс тоолж эхэлдэг.**

Тиймээс эхний утга 0 индекст, хоёр дахь утга 1 индекст, гурав дахь утга 2 индекст байх гэх мэт. (Хичээл 3-ын `range(5)`-ыг санаж байна уу? Тэр ч бас 0-ээс эхэлсэн: 0, 1, 2, 3, 4.)

```python
fruits = ["apple", "banana", "cherry", "grape", "melon"]
```

| Утга | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |
| Байрлал (хүмүүсийн тоолдогоор) | 1-р | 2-р | 3-р | 4-р | 5-р |

Санахад хялбар арга: **индекс = байрлал - 1**. 1-р утгын индекс 0, 5-р утгын индекс 4.

Утгыг авахын тулд жагсаалтын нэрийг бичээд, араас нь дөрвөлжин хаалтанд индексийг бичнэ:

```python
fruits[0]     # "fruits-ийн 0 индекс дээрх утга"
```

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Индексээр утга авах
print(fruits[0])        # 1-р утга 0 индекст -> apple
print(fruits[1])        # 2-р утга 1 индекст -> banana
print(fruits[2])        # 3-р утга 2 индекст -> cherry
print(fruits[3])        # 4-р утга 3 индекст -> grape
print(fruits[4])        # 5-р (сүүлийн) утга 4 индекст -> melon
````

````python
# Тоонуудын жагсаалт дээр ижил зүйл
# Индекс (байрлал) ба утга (доторх зүйл)-ыг андуурахгүй байхыг анхаараарай
numbers = [10, 20, 30, 40]
#  индекс:  0   1   2   3

print(numbers[0])       # индекс 0 -> утга нь 10
print(numbers[2])       # индекс 2 -> утга нь 30 (2 биш!)
print(numbers[3])       # индекс 3 -> утга нь 40
````

````python
# Утгууд ямар нэг тусгай дарааллаар байх албагүй
# Индекс нь зөвхөн утга ХААНА байгааг хэлдэг, хэр том болохыг биш
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[1])        # 2-р сурагчийн оноо -> 92
print(scores[4])        # 5-р сурагчийн оноо -> 66
````

````python
# Нэг индексийг хүссэн удаагаараа авч болно
# Утга авах нь зөвхөн УНШИХ үйлдэл. Жагсаалт өөрчлөгдөхгүй
colors = ["red", "green", "blue"]

print(colors[1])        # green
print(colors[1])        # green хэвээр
print(colors)           # Жагсаалт өмнөхтэйгээ адилхан
````

````python
# Жагсаалтаас авсан утгыг энгийн хувьсагч шиг ашиглаж болно
scores = [85, 92, 78, 90, 66]

first = scores[0]                  # Эхний оноог (85) шинэ хувьсагчид хадгална
print(first)

print(scores[0] + scores[1])       # 85 + 92 = 177 (тэдгээрээр тооцоолж болно)
print(scores[1] > scores[2])       # 92 > 78 нь True (Хичээл 2-ын нэгэн адил харьцуулж болно)
````

````python
# Жагсаалтаас авсан тэмдэгт мөрүүдийг Хичээл 1-ийн нэгэн адил +-ээр залгаж болно
words = ["Hello", "nice", "to", "meet", "you"]

print(words[0] + ", " + words[1] + " " + words[2] + " " + words[3] + " " + words[4])
````

````python
# Индекс нь хувьсагч ч байж болно
# Энд i-д 2 хадгалагдсан тул scores[i] нь scores[2]-той адил
scores = [85, 92, 78, 90, 66]

i = 2
print(scores[i])        # scores[2] -> 78

i = 0
print(scores[i])        # Одоо i-д 0 хадгалагдсан тул scores[0] -> 85
````

````python
# Индекс нь тооцоолол ч байж болно
# Python эхлээд [ ] доторх тоог бодоод, дараа нь тэр индексийг хайна
scores = [85, 92, 78, 90, 66]

print(scores[1 + 2])    # 1 + 2 = 3 -> scores[3] -> 90

i = 1
print(scores[i + 1])    # i + 1 = 2 -> scores[2] -> 78
````

### Сөрөг индекс: төгсгөлөөс тоолох

Энгийн (эерэг) индексийг мэдсэн тул одоо утга олох өөр нэг аргыг үзье.

Python нь сөрөг тоо ашиглан жагсаалтын **төгсгөлөөс буцаж** тоолох боломжийг олгодог.

- `-1` нь **сүүлийн** утга
- `-2` нь сүүлээсээ хоёр дахь утга
- `-3` нь сүүлээсээ гурав дахь утга гэх мэт

| Утга | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |
| Сөрөг индекс | -5 | -4 | -3 | -2 | -1 |

Сөрөг индекс -0-ээс биш **-1**-ээс эхэлдгийг анхаараарай. (-0 гэж байхгүй, учир нь -0 бол зүгээр л 0, өөрөөр хэлбэл эхний утга юм.)

**Сөрөг индексийг яагаад ашигладаг вэ?**

Гол шалтгаан нь **сүүлийн утгыг хурдан авах**.

Эерэг индексээр сүүлийн утгыг авахын тулд эхлээд жагсаалтад хэдэн утга байгааг мэдэх хэрэгтэй. Жагсаалт 5 утгатай бол сүүлийн индекс 4. 100 утгатай бол сүүлийн индекс 99. Мөн утга нэмэх эсвэл устгах үед сүүлийн индекс ч өөрчлөгдөнө.

Сөрөг индексээр бол юу ч тоолох шаардлагагүй. Жагсаалт хэчнээн урт байсан ч **`-1` үргэлж сүүлийн утга** байна.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Сөрөг индекс: төгсгөлөөс тоолох
print(fruits[-1])       # -1 үргэлж сүүлийн утга -> melon
print(fruits[-2])       # -2 нь сүүлээсээ хоёр дахь утга -> grape
print(fruits[-3])       # -3 нь сүүлээсээ гурав дахь утга -> cherry
````

````python
# Эерэг индекс ба сөрөг индекс нэг утгыг заасан байж болно
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[4])        # Эерэг индексээр сүүлийн утга -> melon
print(fruits[-1])       # Сөрөг индексээр сүүлийн утга -> melon (адилхан!)

print(fruits[0])        # Эерэг индексээр эхний утга -> apple
print(fruits[-5])       # Сөрөг индексээр эхний утга -> apple (адилхан!)
````

````python
# -1 яагаад хэрэгтэй вэ: урт нь ямар ч байсан БҮХ жагсаалтад ажилладаг
# Эерэг индексээр бол жагсаалт бүрийн сүүлийн индекс өөр (2, 6, 0 ...)
# -1 бол үргэлж адилхан
short_list = ["a", "b", "c"]
long_list = [10, 20, 30, 40, 50, 60, 70]
one_item = ["only"]

print(short_list[-1])   # c       (эерэг индекс бол 2)
print(long_list[-1])    # 70      (эерэг индекс бол 6)
print(one_item[-1])     # only    (эерэг индекс бол 0)
````

## 3. Утгыг өөрчлөх ба len()

### Утгыг өөрчлөх

Жагсаалт тогтмол биш. Түүний доторх утгыг хүссэн үедээ **өөрчилж** болно. Нэг шүүгээнд шинэ юм хийхтэй адил:

```python
жагсаалтын_нэр[индекс] = шинэ_утга
```

Энэ нь хувьсагчид утга хадгалахтай яг адилхан ажилладаг (`=` нь "хадгалах" гэсэн утгатай), гэхдээ зөвхөн тэр нэг байрлал дахь утга солигдоно. Бусад утгууд хэвээрээ үлдэнэ.

````python
# Тодорхой индекс дээрх утгыг өөрчлөх
fruits = ["apple", "banana", "cherry"]
print(fruits)

fruits[1] = "orange"        # 1 индекс дээрх утгыг ("banana") "orange"-аар солино
print(fruits)               # Зөвхөн 1 индекс өөрчлөгдсөн. "apple" ба "cherry" хэвээрээ
````

````python
# Сөрөг индексээр ч утгыг өөрчилж болно
fruits = ["apple", "banana", "cherry"]

fruits[-1] = "grape"        # Сүүлийн утгыг ("cherry") "grape"-ээр солино
print(fruits)
````

````python
# Хуучин утгыг нь ашиглан утгыг өөрчлөх
# Хичээл 1-ийн age = age + 1-тэй ижил санаа
scores = [85, 92, 78]

scores[2] = scores[2] + 5   # Гурав дахь сурагч 5 урамшууллын оноо авна (78 + 5 = 83)
print(scores)
````

### len(): жагсаалтад хэдэн утга байна вэ?

`len()` нь жагсаалтад хэдэн утга байгааг хэлж өгдөг. `len` нь **length** (урт) гэдгийн товчлол.

```python
len(жагсаалтын_нэр)
```

Жагсаалтыг дугуй хаалтан дотор бичихэд `len()` нэг тоо буцаана: тэр жагсаалт доторх утгын тоо.

````python
# len(): жагсаалтад хэдэн утга байгааг тоолох
fruits = ["apple", "banana", "cherry", "grape", "melon"]
print(len(fruits))          # 5 утга байгаа тул 5
````

````python
# Өөр өөр урттай жагсаалтууд
three = ["a", "b", "c"]
five = [1, 2, 3, 4, 5]
one = ["only"]
empty_list = []

print(len(three))           # 3 утга -> 3
print(len(five))            # 5 утга -> 5
print(len(one))             # 1 утга -> 1
print(len(empty_list))      # Хоосон жагсаалтад 0 утга -> 0
````

````python
# len() нь ямар ч төрлийн жагсаалт дээр адилхан ажилладаг
names = ["Bill", "Anna", "Tom", "Kate"]          # Тэмдэгт мөрүүдийн жагсаалт
numbers = [3, 7, 1, 9, 4, 6]                     # Бүхэл тоонуудын жагсаалт
answers = [True, False, True]                    # bool утгуудын жагсаалт

print(len(names))           # 4
print(len(numbers))         # 6
print(len(answers))         # 3
````

````python
# len() нь утга ХЭДЭН ширхэг байгааг тоолдог. Утгуудыг нэмдэггүй
# Утгууд том ч зөвхөн 3 ширхэг байна
prices = [1000, 2500, 3000]
print(len(prices))          # 3 (6500 биш!)
````

````python
# len()-ийн үр дүн нь зүгээр л тоо (int) тул хувьсагчид хадгалж болно
students = ["Bill", "Anna", "Tom", "Kate", "Mike"]

count = len(students)       # count хувьсагчид 5-ыг хадгална
print(count)
print("There are", count, "students.")
````

````python
# len() нь тэмдэгт мөр дээр ч ажилладаг: тэмдэгтийн тоог тоолно
# (Зай ч бас нэг тэмдэгтэд тооцогдоно)
print(len("apple"))         # a, p, p, l, e -> 5
print(len("hi there"))      # h, i, (зай), t, h, e, r, e -> 8
````

### Сүүлийн индекс нь len() - 1

Одоо `len()`-ийг индекстэй холбож болно.

Жагсаалт 5 утгатай бол `len()` нь 5-ыг өгөх боловч сүүлийн индекс нь **4**. Индекс 0-ээс эхэлдэг тул **сүүлийн индекс үргэлж `len() - 1`** байна.

Тиймээс сүүлийн утгыг авах хоёр арга бий:
- `fruits[len(fruits) - 1]`: утгуудыг тоолж, 1-ийг хасаад, тэр индексийг ашиглана
- `fruits[-1]`: зүгээр л -1-ийг ашиглана

Хоёулаа ижил утга өгөх боловч `-1` нь хамаагүй богино бөгөөд хялбар. Сөрөг индекс яг үүнд л хэрэгтэй.

````python
# Сүүлийн индекс нь len() - 1
fruits = ["apple", "banana", "cherry", "grape", "melon"]

last_index = len(fruits) - 1       # 5 - 1 = 4
print(last_index)
print(fruits[last_index])          # fruits[4] -> melon
````

````python
# Сүүлийн утгыг авах хоёр аргыг харьцуулах
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[len(fruits) - 1])     # Урт арга: len(fruits) нь 5, 5 - 1 = 4, fruits[4] -> melon
print(fruits[-1])                  # Богино арга: -1 нь сүүлийн утга -> melon (үр дүн адилхан!)
````

## 4. Утга жагсаалтад байгаа эсэхийг шалгах (in)

Заримдаа **"Энэ утга жагсаалтад байна уу?"** гэдгийг мэдэхийг хүсдэг.

Үүнд `in`-ийг ашиглана.

```python
утга in жагсаалтын_нэр
```

- Утга жагсаалтад байвал үр дүн нь `True`.
- Байхгүй бол үр дүн нь `False`.

Үр дүн нь Хичээл 2-ын харьцуулах операторуудын нэгэн адил `True` эсвэл `False` байна. Энэ нь үүнийг `if`-ийн нөхцөл болгон шууд ашиглаж болно гэсэн үг!

Утга жагсаалтад **байхгүй** эсэхийг шалгахын тулд `not in`-ийг ашиглана.

**Анхаар:** Та `in` гэдэг үгийг Хичээл 3-т ч харсан (`for i in range(5)`). Тэр `in` нь `for` давталтын нэг хэсэг. Энд сурч буй `in` нь өөр: утга жагсаалтад байгаа эсэхийг **шалгаж**, True эсвэл False өгдөг.

````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)       # "banana" жагсаалтад байна -> True
print("grape" in fruits)        # "grape" жагсаалтад байхгүй -> False
````

````python
# not in: in-ийн эсрэг
print("grape" not in fruits)    # "grape" жагсаалтад БАЙХГҮЙ -> True
print("apple" not in fruits)    # "apple" жагсаалтад БАЙГАА тул "not in" нь False
````

````python
# in-ийг if-ийн нөхцөл болгон ашиглах (Хичээл 2)
fruits = ["apple", "banana", "cherry"]
wanted = "banana"

if wanted in fruits:                    # "banana" in fruits нь True
    print("Yes, we have it!")
else:
    print("Sorry, it's sold out.")
````

````python
# Бодит амьдралын жишээ: зочдын жагсаалтыг шалгах
guests = ["Bill", "Anna", "Tom"]
name = "Kate"

if name in guests:                      # "Kate" in guests нь False
    print("Welcome!")
else:
    print("Sorry, you are not on the list.")    # Тиймээс энэ ажиллана
````

## 5. Утга нэмэх ба устгах

Жагсаалт томорч, жижгэрч болно. Шинэ утга нэмж, хэрэггүй утгаа устгаж болно.

| Бичих арга | Юу хийдэг вэ |
|---|---|
| `жагсаалтын_нэр.append(утга)` | Утгыг жагсаалтын **төгсгөлд** нэмнэ |
| `жагсаалтын_нэр.insert(индекс, утга)` | Утгыг **сонгосон байрлалд** нэмнэ |
| `жагсаалтын_нэр.remove(утга)` | Заасан утгыг устгана (хамгийн эхэнд олдсоныг) |
| `жагсаалтын_нэр.pop()` | **Сүүлийн** утгыг гаргаж авна |
| `жагсаалтын_нэр.pop(индекс)` | **Сонгосон байрлал** дахь утгыг гаргаж авна |

Цэг `.` нь "энэ жагсаалт дээр үүнийг хий" гэсэн утгатай. Жишээ нь, `fruits.append("grape")` нь "fruits жагсаалтад "grape"-ийг нэм" гэсэн утгатай. Ийм цэгийн араас бичдэг командуудыг **метод (method)** гэдэг.

Эдгээр метод нь жагсаалтыг өөрийг нь өөрчилдөг. `fruits = ...` гэж бичих шаардлагагүй.

````python
# append(): утгыг төгсгөлд нэмэх
fruits = ["apple", "banana"]
print(fruits)

fruits.append("cherry")         # "cherry" төгсгөлд нэмэгдэнэ
print(fruits)

fruits.append("grape")          # "grape" дахин төгсгөлд нэмэгдэнэ
print(fruits)
print(len(fruits))              # Жагсаалт 2 утгаас 4 утга болж томорсон
````

````python
# Хоосон жагсаалтаас эхлээд утгуудыг нэг нэгээр нэмэх
# Энэ бол маш түгээмэл хэрэглэгддэг арга
shopping = []                   # Хоосон жагсаалтаас эхэлнэ
shopping.append("milk")
shopping.append("eggs")
shopping.append("bread")
print(shopping)
````

````python
# insert(): тодорхой байрлалд утга нэмэх
# insert(индекс, утга) -> шинэ утга тэр индекст орж, араас нь байсан утгууд нэг байрлалаар хойшилно
fruits = ["apple", "banana", "cherry"]

fruits.insert(1, "orange")      # "orange"-ийг 1 индекст оруулна
print(fruits)                   # "banana" ба "cherry" тус бүр нэг байрлалаар хойшилсон

fruits.insert(0, "melon")       # "melon"-ийг 0 индекст (хамгийн урд) оруулна
print(fruits)
````

````python
# remove(): утгыг өөрийг нь нэрлэж устгах
fruits = ["apple", "banana", "cherry"]

fruits.remove("banana")         # "banana"-г олж устгана
print(fruits)                   # Араас нь байсан утгууд урагшилж хоосон зайг нөхнө
````

````python
# Нэг утга олон удаа байвал remove() нь зөвхөн ЭХНИЙХИЙГ нь устгана
numbers = [1, 2, 3, 2, 1]

numbers.remove(2)               # Зөвхөн эхний 2 (1 индекс дээрх) устгагдана
print(numbers)                  # Хоёр дахь 2 хэвээрээ байна
````

````python
# pop(): байрлалаар нь утгыг гаргаж авах
# Индексгүй бол pop() нь СҮҮЛИЙН утгыг гаргаж авна
fruits = ["apple", "banana", "cherry"]

fruits.pop()                    # Сүүлийн утгыг ("cherry") гаргаж авна
print(fruits)
````

````python
# pop(индекс): тэр индекс дээрх утгыг гаргаж авах
fruits = ["apple", "banana", "cherry"]

fruits.pop(0)                   # 0 индекс дээрх утгыг ("apple") гаргаж авна
print(fruits)
````

````python
# pop() нь гаргаж авсан утгаа БУЦААЖ өгдөг тул хувьсагчид хадгалж болно
# remove() зөвхөн устгадаг. pop() нь устгаад ӨӨРТ ЧИНЬ утгыг өгдөг
tasks = ["wash dishes", "do homework", "clean room"]

done = tasks.pop(0)             # Эхний ажлыг гаргаж аваад done-д хадгална
print("Finished:", done)
print("Still to do:", tasks)
````

````python
# remove() ба pop()-ийг нэг дор харах
# remove(утга): "ЭНЭ УТГЫГ устга"                    -> та утгыг мэдэж байна
# pop(индекс):  "ЭНЭ БАЙРЛАЛД байгааг устга"          -> та байрлалыг мэдэж байна
colors = ["red", "green", "blue"]

colors.remove("green")          # Утгаар нь устгах
print(colors)

colors.pop(0)                   # Байрлалаар нь устгах (одоо 0 индекс дээр "red" байна)
print(colors)
````

## 6. Түгээмэл алдаанууд

Жагсаалтыг сурч байхдаа бараг бүх хүн эдгээр алдааг гаргадаг. Алдаа гарвал эхлээд эдгээрийг шалгаарай.

1. **Индекс 0-ээс эхэлдгийг мартах**: `fruits[1]` нь эхний биш **хоёр дахь** утга.
2. **Байхгүй индекс ашиглах**: Жагсаалт 3 утгатай бол индексүүд нь 0, 1, 2. `fruits[3]` нь `IndexError: list index out of range` ("индекс жагсаалтын хүрээнээс гарсан") алдаа гаргана.
3. **Шинэ индексээр утга нэмэхийг оролдох**: 3 утгатай жагсаалтад `fruits[3] = "grape"` гэвэл утга **нэмэгдэхгүй**. `IndexError: list assignment index out of range` ("жагсаалтын гадна байгаа индекст утга хадгалж болохгүй") алдаа гарна. Нэмэхийн тулд `append()`-ийг ашиглаарай.
4. **Жагсаалтад байхгүй утгыг устгах**: `fruits.remove("grape")` нь `ValueError: list.remove(x): x not in list` ("тэр утга жагсаалтад байхгүй") алдаа гаргана. Эхлээд `in`-ээр шалгаарай.
5. **`( )` ба `[ ]`-ийг андуурах**: Жагсаалт үүсгэх болон индексээр утга авахад (`fruits[0]`) `[ ]`-ийг ашиглана. `len(fruits)`, `fruits.append("grape")` зэрэг командуудад `( )`-ийг ашиглана.

Доорх нүднүүд алдаа бүр ямар харагддагийг үзүүлнэ. Notebook эвдрэхгүйн тулд алдаатай кодыг тайлбар хэлбэрээр бичсэн. Жинхэнэ алдааны мэдэгдлийг харахын тулд кодын мөрөөс `#`-г арилгаж үзээд, дараа нь буцааж тавиарай.

````python
# Алдаа 1: индекс 0-ээс эхэлдгийг мартах
fruits = ["apple", "banana", "cherry"]

print(fruits[1])        # Олон хүн "apple" гэж бодох боловч энэ бол ХОЁР ДАХЬ утга -> banana
print(fruits[0])        # Зөв: эхний утга 0 индекст -> apple
````

````python
# Алдаа 2: байхгүй индекс ашиглах
fruits = ["apple", "banana", "cherry"]      # 3 утга -> индексүүд нь 0, 1, 2

# print(fruits[3])      # IndexError: list index out of range -> 3 индекс байхгүй!

print(fruits[2])                    # Зөв: сүүлийн индекс нь 2
print(fruits[len(fruits) - 1])      # Мөн зөв: len(fruits) - 1 = 3 - 1 = 2
print(fruits[-1])                   # Мөн зөв: -1 үргэлж сүүлийн утга
````

````python
# Алдаа 3: шинэ индексээр утга нэмэхийг оролдох
fruits = ["apple", "banana", "cherry"]

# fruits[3] = "grape"   # IndexError: list assignment index out of range -> зөвхөн аль хэдийн байгаа индексийн утгыг ӨӨРЧИЛЖ болно

fruits.append("grape")  # Зөв: шинэ утгыг төгсгөлд нэмэхийн тулд append() ашиглана
print(fruits)
````

````python
# Алдаа 4: жагсаалтад байхгүй утгыг устгах
fruits = ["apple", "banana", "cherry"]

# fruits.remove("grape")    # ValueError: list.remove(x): x not in list -> "grape" жагсаалтад байхгүй!

# Зөв: эхлээд in-ээр шалгаад (4-р хэсэг), байгаа үед л устгана
if "grape" in fruits:
    fruits.remove("grape")
else:
    print("grape is not in the list, so nothing was removed.")
````

````python
# Алдаа 5: ( ) ба [ ]-ийг андуурах
fruits = ["apple", "banana", "cherry"]

# print(fruits(0))          # TypeError: 'list' object is not callable -> ( ) нь индексэд биш, командад зориулагдсан
# fruits.append["grape"]    # TypeError -> append бол команд тул [ ] биш ( ) хэрэгтэй

print(fruits[0])            # Зөв: индексээр утга авахад [ ]
fruits.append("grape")      # Зөв: командад ( )
print(fruits)
````
