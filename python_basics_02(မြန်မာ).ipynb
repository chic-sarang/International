## 1. if ကြေညာချက် ဆိုတာဘာလဲ?

အခုအချိန်ထိ ကျွန်တော်တို့ရဲ့ code က အပေါ်ကနေ အောက်ကို တစ်ကြောင်းပြီးတစ်ကြောင်း အမြဲတမ်း တစ်ပုံစံတည်း run ခဲ့ပါတယ်။ ဒါပေမဲ့ လက်တွေ့ဘဝမှာ ကျွန်တော်တို့ အမြဲတမ်း ရွေးချယ်မှုတွေ လုပ်နေရပါတယ်။

- မိုးရွာနေ **ရင်** ထီးယူသွားပါ။
- အသက် 20 နှစ် နဲ့အထက် ဖြစ် **ရင်** လူကြီးလက်မှတ် ဝယ်လို့ရပါတယ်။

`if` ကြေညာချက်က ပရိုဂရမ်ကို ဒီလို ရွေးချယ်မှုတွေ လုပ်နိုင်စေပါတယ်။ သူက **condition (အခြေအနေ)** ကို စစ်ဆေးပြီး **condition က True ဖြစ်မှသာ** အထဲက code ကို run ပါတယ်။

သင်ခန်းစာ 1 က နှိုင်းယှဉ် operator တွေ (`>`, `<`, `==`, `!=`) ကို မှတ်မိသေးလား? သူတို့ရဲ့ ရလဒ်က အမြဲတမ်း `True` ဒါမှမဟုတ် `False` ဖြစ်ပါတယ်။ `if` က condition အဖြစ် လိုအပ်တာ အဲဒါပါပဲ။

အခြေခံပုံစံက ဒီလိုပါ:

```python
if condition:
    condition က True ဖြစ်မှသာ run မယ့် code
```

အရမ်းအရေးကြီးတဲ့ စည်းမျဉ်း ၂ ခု:
- `if` စာကြောင်းရဲ့ အဆုံးမှာ colon `:` မဖြစ်မနေ ရှိရပါမယ်။
- `if` ထဲမှာပါတဲ့ code ကို **indent (စာကြောင်းရှေ့ နေရာချန်ခြင်း)** လုပ်ရပါမယ် (ညာဘက်ကို space 4 ခု တွန်းပေးရပါမယ်)။ ဘယ်စာကြောင်းတွေက `if` ရဲ့ "အထဲမှာ" ရှိလဲဆိုတာကို Python က indent နဲ့ သိပါတယ်။

````python
# condition က age >= 20 ("အသက်က 20 ထက်ကြီးသည် သို့မဟုတ် ညီသည်") ဖြစ်ပါတယ်
# ဒီနှိုင်းယှဉ်ချက်ရဲ့ ရလဒ်က True ဒါမှမဟုတ် False ဖြစ်ပါတယ်
age = 25

if age >= 20:                         # 25 >= 20 က True ဖြစ်လို့ အောက်က indent လုပ်ထားတဲ့ code က run ပါမယ်
    print("You are an adult.")        # ဒီစာကြောင်းက if ထဲမှာ ရှိပါတယ် (space 4 ခု indent လုပ်ထားပါတယ်)

print("This line always runs.")      # ဒီစာကြောင်းကို indent မလုပ်ထားလို့ if ရဲ့ အပြင်မှာ ရှိပါတယ်။ ဘာပဲဖြစ်ဖြစ် run ပါတယ်
````

````python
# အခု condition က False ဖြစ်တဲ့ အခြေအနေကို စမ်းကြည့်ရအောင်
age = 15

if age >= 20:                         # 15 >= 20 က False ဖြစ်လို့ အောက်က indent လုပ်ထားတဲ့ code ကို ကျော်သွားပါမယ်
    print("You are an adult.")        # ဒီစာကြောင်း လုံးဝ မ run ပါဘူး

print("This line always runs.")      # ဒီစာကြောင်းက if ရဲ့ အပြင်မှာ ရှိလို့ ဆက်ပြီး run ပါတယ်
````

````python
# if ထဲမှာ စာကြောင်း တစ်ကြောင်းထက် ပိုပြီး ထည့်လို့ရပါတယ်
# indent တူညီတဲ့ စာကြောင်းအားလုံးက if တစ်ခုတည်းမှာ ပါဝင်ပါတယ်
temperature = 32

if temperature > 30:                          # 32 > 30 က True
    print("It's very hot today.")             # if ထဲမှာ
    print("Drink plenty of water.")           # ဒါလည်း if ထဲမှာ (indent တူတူ)
    print("Try to stay in the shade.")        # ဒါလည်း if ထဲမှာ (indent တူတူ)

print("Have a nice day!")                    # if ရဲ့ အပြင်မှာ၊ အမြဲတမ်း run ပါတယ်
````

````python
# == ကို သုံးပြီး တန်ဖိုး နှစ်ခု တူမတူ စစ်လို့လည်း ရပါတယ်
# မှတ်ထားပါ: = က "တန်ဖိုးကို သိမ်းပါ" ဆိုတဲ့ အဓိပ္ပာယ်ဖြစ်ပြီး == က "ဒီတန်ဖိုးတွေ ညီသလား?" ဆိုတဲ့ အဓိပ္ပာယ်ပါ
password = "python123"

if password == "python123":           # string နှစ်ခုက အတိအကျ တူညီလို့ True ဖြစ်ပါတယ်
    print("Login successful!")
````

## 2. if - else

`if` တစ်ခုတည်းဆိုရင် condition က True ဖြစ်တဲ့အခါ တစ်ခုခု လုပ်ပြီး False ဖြစ်တဲ့အခါ **ဘာမှ မလုပ်ပါဘူး**။

ဒါပေမဲ့ များသောအားဖြင့် **ဒါ ဒါမှမဟုတ် ဟိုဟာ** လုပ်ချင်တာ များပါတယ်။ ဥပမာ:
- စာမေးပွဲ အောင်ရင် "ဂုဏ်ယူပါတယ်" လို့ ပြပါ။ **မဟုတ်ရင်** "ထပ်ကြိုးစားပါ" လို့ ပြပါ။

ဒါအတွက် `else` ကို သုံးပါတယ်။ `else` ဆိုတာ "မဟုတ်ရင်" လို့ အဓိပ္ပာယ်ရပါတယ်။ သူ့ code က **if ရဲ့ condition က False ဖြစ်မှသာ** run ပါတယ်။

```python
if condition:
    condition က True ဖြစ်ရင် run မယ့် code
else:
    condition က False ဖြစ်ရင် run မယ့် code
```

- `else` မှာ ကိုယ်ပိုင် condition မရှိပါဘူး။ `if` က မဖမ်းထားတဲ့ အခြေအနေ အားလုံးကို သူက ဖမ်းယူပါတယ်။
- `else` ရဲ့ အဆုံးမှာလည်း colon `:` လိုပါတယ်၊ သူ့ code ကိုလည်း indent လုပ်ရပါမယ်။
- block နှစ်ခုထဲက **တစ်ခု** က အမြဲတမ်း အတိအကျ run ပါတယ်။ နှစ်ခုလုံး run တာလည်း မရှိ၊ တစ်ခုမှ မ run တာလည်း မရှိပါဘူး။

````python
# စာမေးပွဲ အမှတ်က အောင်လား ကျလား စစ်ဆေးခြင်း
# အောင်မှတ်က 60 လို့ ဆိုကြပါစို့
score = 75

if score >= 60:                      # 75 >= 60 က True
    print("Pass! Congratulations.")  # ဒါကြောင့် ဒီစာကြောင်း run ပါတယ်
else:                                # "မဟုတ်ရင်" - if က True ဖြစ်ခဲ့လို့ ဒီအပိုင်းကို ကျော်သွားပါတယ်
    print("Fail. Try again.")
````

````python
# code အတူတူပါပဲ၊ ဒါပေမဲ့ အမှတ် နည်းသွားပါတယ်
score = 40

if score >= 60:                      # 40 >= 60 က False ဖြစ်လို့ ဒီအပိုင်းကို ကျော်သွားပါတယ်
    print("Pass! Congratulations.")
else:                                # if က False ဖြစ်ခဲ့လို့ အစား else အပိုင်းက run ပါတယ်
    print("Fail. Try again.")
````

````python
# ဂဏန်းတစ်ခုက စုံကိန်းလား မကိန်းလား စစ်ဆေးခြင်း
# သင်ခန်းစာ 1 က % ကို မှတ်မိသေးလား? သူက စားလို့ ကျန်တဲ့ အကြွင်းကို ပေးပါတယ်
# စုံကိန်းကို 2 နဲ့ စားရင် အကြွင်း အမြဲ 0 ဖြစ်ပါတယ်
number = 7

if number % 2 == 0:                  # 7 % 2 က 1 ဖြစ်ပြီး 1 == 0 က False
    print(number, "is even.")        # comma နဲ့ ခွဲပေးရင် print() က တန်ဖိုး အများကြီးကို တစ်ခါတည်း ပြနိုင်ပါတယ်
else:
    print(number, "is odd.")         # if က False ဖြစ်ခဲ့လို့ ဒီစာကြောင်း run ပါတယ်
````

## 3. if - elif - else

တစ်ခါတလေ ရွေးချယ်စရာက **နှစ်ခုထက် ပိုပါတယ်**။ ဥပမာ အမှတ်ကို အဆင့် (grade) အဖြစ် ပြောင်းခြင်း:

- 90 နဲ့အထက် → A
- 80 နဲ့အထက် → B
- 70 နဲ့အထက် → C
- ကျန်တာအားလုံး → F

ဒီလိုအခါ `elif` ကို သုံးပါတယ်။ `elif` က "else if" ရဲ့ အတိုကောက်ဖြစ်ပြီး "အပေါ်က condition တွေ False ဖြစ်ခဲ့ရင် ဒီ condition ကို စစ်ပါ" လို့ အဓိပ္ပာယ်ရပါတယ်။

```python
if condition1:
    condition1 က True ဖြစ်ရင် run ပါတယ်
elif condition2:
    condition1 က False ဖြစ်ပြီး condition2 က True ဖြစ်ရင် run ပါတယ်
elif condition3:
    condition1 နဲ့ condition2 က False ဖြစ်ပြီး condition3 က True ဖြစ်ရင် run ပါတယ်
else:
    အပေါ်က condition အားလုံး False ဖြစ်ရင် run ပါတယ်
```

အလုပ်လုပ်ပုံ:
- Python က condition တွေကို **အပေါ်ကနေ အောက်ကို အစဉ်လိုက်** စစ်ပါတယ်။
- True ဖြစ်တဲ့ **ပထမဆုံး** condition ကို တွေ့တာနဲ့ အဲဒီ block ကို run ပြီး **ကျန်တာအားလုံးကို ကျော်သွားပါတယ်**။
- `elif` ကို လိုသလောက် ထည့်လို့ရပါတယ်။ နောက်ဆုံးက `else` က ထည့်ချင်မှ ထည့်ပါ။

````python
# အမှတ်ကို grade အဖြစ် ပြောင်းခြင်း
score = 85

if score >= 90:                # 85 >= 90 က False -> နောက် condition ကို ဆက်သွားပါတယ်
    print("Grade: A")
elif score >= 80:              # 85 >= 80 က True -> ဒီ block run ပြီး အောက်က အားလုံးကို ကျော်သွားပါတယ်
    print("Grade: B")
elif score >= 70:              # အပေါ်မှာ True condition ကို တွေ့ပြီးသားမို့ လုံးဝ မစစ်တော့ပါဘူး
    print("Grade: C")
else:                          # ဒါလည်း မစစ်ပါဘူး
    print("Grade: F")
````

````python
# code တူတူနဲ့ အမှတ် အမျိုးမျိုး စမ်းကြည့်ပါ
# score ရဲ့ တန်ဖိုးကို ပြောင်းပြီး ပြန် run ကြည့်ရင် ရလဒ် ဘယ်လိုပြောင်းလဲလဲ မြင်ရပါမယ်
score = 72

if score >= 90:                # 72 >= 90 က False
    print("Grade: A")
elif score >= 80:              # 72 >= 80 က False
    print("Grade: B")
elif score >= 70:              # 72 >= 70 က True -> ဒီ block run ပါတယ်
    print("Grade: C")
else:
    print("Grade: F")
````

````python
# အစီအစဉ်က ဘာလို့ အရေးကြီးတာလဲ?
# Python က ပထမဆုံး True condition မှာ ရပ်လို့ အသေးဆုံး စစ်ချက်ကို အရင်ထားရင် မှားသွားပါမယ်
score = 95

if score >= 70:                # 95 >= 70 က True -> ဒါ အရင် run ပြီး ကျန်တာအားလုံးကို ကျော်သွားပါတယ်!
    print("Grade: C")
elif score >= 80:              # ဘယ်တော့မှ မစစ်ပါဘူး
    print("Grade: B")
elif score >= 90:              # 95 က A ရသင့်ပေမဲ့ ဘယ်တော့မှ မစစ်ပါဘူး
    print("Grade: A")

# ဒါကြောင့် ဒီလို အပိုင်းအခြားတွေ စစ်တဲ့အခါ အကြီးဆုံး (အတင်းကျပ်ဆုံး) condition ကနေ စပါ
````

````python
# အပူချိန်ပေါ်မူတည်ပြီး ဘာဝတ်ရမလဲ ဆုံးဖြတ်ခြင်း
temperature = 12

if temperature >= 25:                  # 12 >= 25 က False
    print("Wear a T-shirt.")
elif temperature >= 15:                # 12 >= 15 က False
    print("Wear a light jacket.")
elif temperature >= 5:                 # 12 >= 5 က True -> ဒီစာကြောင်း run ပါတယ်
    print("Wear a coat.")
else:                                  # အပူချိန် 5 အောက်ဖြစ်မှသာ run ပါတယ်
    print("Wear a thick padded jacket.")
````

## 4. bool variable ကို condition အဖြစ် အသုံးပြုခြင်း

`if` ရဲ့ condition က `True` ဒါမှမဟုတ် `False` ဖြစ်ရုံနဲ့ လုံလောက်ပါတယ်။ ဆိုလိုတာက **bool variable** (သင်ခန်းစာ 1 က) ကို ဘာနှိုင်းယှဉ်ချက်မှ မလိုဘဲ condition အဖြစ် တိုက်ရိုက် သုံးလို့ရပါတယ်။

```python
if is_student:        # အဓိပ္ပာယ်တူ: if is_student == True:
```

`== True` လို့ ရေးတာ မမှားပေမဲ့ မလိုအပ်လို့ Python programmer တွေက များသောအားဖြင့် မရေးကြပါဘူး။

bool variable က `False` ဖြစ်လား စစ်ချင်ရင် `not` ကို သုံးပါ:

```python
if not is_student:    # အဓိပ္ပာယ်တူ: if is_student == False:
```

## 5. condition များကို ပေါင်းစပ်ခြင်း (and, or, not)

တစ်ခါတလေ condition တစ်ခုတည်းနဲ့ မလုံလောက်ပါဘူး။ ဥပမာ:
- **အသက် 10 နှစ်နဲ့အထက် ဖြစ်ပြီး အရပ် 140 cm နဲ့အထက်** ဖြစ်ရင် roller coaster စီးလို့ရပါတယ်။
- **အသက် 7 နှစ်အောက် ဒါမှမဟုတ် 65 နှစ်နဲ့အထက်** ဖြစ်ရင် အခမဲ့ ဝင်ခွင့်ရပါတယ်။

Python မှာ condition တွေကို ပေါင်းစပ်ဖို့ စကားလုံး ၃ လုံး ရှိပါတယ်။

| Operator | အဓိပ္ပာယ် | ရလဒ် |
|---|---|---|
| `and` | နှင့် | ဘက် **နှစ်ဘက်လုံး** True ဖြစ်မှသာ True |
| `or` | သို့မဟုတ် | **အနည်းဆုံး တစ်ဘက်** True ဖြစ်ရင် True |
| `not` | ဆန့်ကျင်ဘက် | True ကို False၊ False ကို True အဖြစ် ပြောင်းပေးပါတယ် |

````python
# ပထမဆုံး and / or တစ်ခုတည်း သုံးရင် ဘာရလဒ်ရလဲ ကြည့်ရအောင်
print(True and True)     # နှစ်ခုလုံး True -> True
print(True and False)    # တစ်ခုက False -> False
print(True or False)     # အနည်းဆုံး တစ်ခုက True -> True
print(False or False)    # နှစ်ခုလုံး False -> False
print(not True)          # True ရဲ့ ဆန့်ကျင်ဘက် -> False
````

````python
# and: condition အားလုံး True ဖြစ်ရပါမယ်
# စည်းမျဉ်း: အသက် 10 နှစ်နဲ့အထက် ဖြစ်ပြီး (AND) အရပ် 140 cm နဲ့အထက်
age = 12
height = 135

if age >= 10 and height >= 140:       # 12 >= 10 က True ဒါပေမဲ့ 135 >= 140 က False -> True and False -> False
    print("You can ride the roller coaster.")
else:
    print("Sorry, you can't ride it yet.")  # condition က False ဖြစ်ခဲ့လို့ ဒါ run ပါတယ်
````

````python
# or: အနည်းဆုံး condition တစ်ခု True ဖြစ်ရပါမယ်
# စည်းမျဉ်း: အသက် 7 နှစ်အောက် ဒါမှမဟုတ် (OR) 65 နှစ်နဲ့အထက် ဆိုရင် အခမဲ့ ဝင်ခွင့်
age = 70

if age < 7 or age >= 65:              # 70 < 7 က False ဒါပေမဲ့ 70 >= 65 က True -> False or True -> True
    print("Free entry!")              # condition က True ဖြစ်ခဲ့လို့ ဒါ run ပါတယ်
else:
    print("Please buy a ticket.")
````

````python
# ဂဏန်းတစ်ခုက သတ်မှတ်ထားတဲ့ အပိုင်းအခြားထဲမှာ ရှိမရှိ စစ်ဆေးခြင်း
# "score က 0 နဲ့ 100 ကြားမှာ ရှိတယ်" ဆိုတာ "score က 0 နဲ့အထက် ဖြစ်ပြီး (AND) score က 100 နဲ့အောက်" လို့ အဓိပ္ပာယ်ရပါတယ်
score = 105

if score >= 0 and score <= 100:       # 105 >= 0 က True ဒါပေမဲ့ 105 <= 100 က False -> False
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")

# Python မှာ သင်္ချာလိုပဲ ပိုတိုတိုနဲ့ ရေးလို့လည်း ရပါတယ်
if 0 <= score <= 100:                 # အပေါ်က စာကြောင်းနဲ့ အဓိပ္ပာယ် အတူတူပါပဲ
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")
````

## 6. if အထပ်ထပ် (nested if)

`if` တစ်ခုကို **နောက်ထပ်** `if` တစ်ခုရဲ့ **အထဲမှာ** ထည့်လို့ရပါတယ်။ ဒါကို **nested (အထပ်ထပ်)** if လို့ ခေါ်ပါတယ်။

ပထမ condition က True ဖြစ်ပြီး **နောက်မှသာ** ဒုတိယ condition ကို စစ်ချင်တဲ့အခါ အသုံးဝင်ပါတယ်။ ဥပမာ user က login ဝင်ထားလား အရင်စစ်ပြီး၊ ဝင်ထားမှသာ admin ဟုတ်မဟုတ် စစ်တာမျိုးပါ။

အတွင်း `if` က အပြင် `if` ထက် တစ်ဆင့် (space 4 ခု) ပိုပြီး indent လုပ်ထားပါတယ်။

````python
# login ကို အရင်စစ်ပြီး login ဝင်ထားမှသာ admin ဟုတ်မဟုတ် စစ်ပါတယ်
is_logged_in = True
is_admin = False

if is_logged_in:                               # အပြင် if: True ဖြစ်လို့ အထဲကို ဝင်ပါတယ်
    print("Welcome!")                          # အပြင် if မှာ ပါဝင်ပါတယ် (space 4 ခု)
    if is_admin:                               # အတွင်း if: is_admin က False
        print("You can open the admin page.")  # အတွင်း if မှာ ပါဝင်ပါတယ် (space 8 ခု) - ကျော်သွားပါတယ်
    else:
        print("You are a regular user.")       # အတွင်း else မှာ ပါဝင်ပါတယ် (space 8 ခု) - ဒါ run ပါတယ်
else:                                          # အပြင် if မှာ ပါဝင်ပါတယ်
    print("Please log in first.")              # အပြင် if က True ဖြစ်ခဲ့လို့ ကျော်သွားပါတယ်
````

````python
# ဆိုင်မှာ ပစ္စည်းဝယ်ခြင်း
# အဆင့် 1: ပစ္စည်း ရှိသေးလား?  အဆင့် 2: ပိုက်ဆံ လုံလောက်လား?
in_stock = True
money = 8000
price = 12000

if in_stock:                                   # ပစ္စည်း ရှိပါတယ် -> အထဲကို ဝင်ပါတယ်
    if money >= price:                         # 8000 >= 12000 က False
        print("Purchase complete!")
    else:
        print("You don't have enough money.")  # ဒါ run ပါတယ်
        print("You need", price - money, "more won.")  # 12000 - 8000 = 4000
else:
    print("Sorry, this item is sold out.")
````

**အကြံပြုချက်:** nested if က အသုံးဝင်ပေမဲ့ အထပ်တွေ များလွန်းရင် code ဖတ်ရခက်သွားပါတယ်။ ဖြစ်နိုင်ရင် condition တွေကို `and` နဲ့ ပေါင်းပြီး ရေးကြည့်ပါ။

```python
# အထပ်ထပ် ပုံစံ
if is_logged_in:
    if is_admin:
        print("You can open the admin page.")

# and နဲ့ ရေးထားတဲ့ အဓိပ္ပာယ်တူ ပုံစံ
if is_logged_in and is_admin:
    print("You can open the admin page.")
```

## 7. မကြာခဏ ဖြစ်တတ်သော အမှားများ

`if` ကို လေ့လာတဲ့အခါ လူတိုင်းနီးပါး ဒီအမှားတွေ လုပ်မိကြပါတယ်။ error တက်ရင် ဒါတွေကို အရင်စစ်ကြည့်ပါ။

1. **`==` အစား `=` ကို သုံးခြင်း**: `=` က တန်ဖိုးကို သိမ်းပြီး `==` က တန်ဖိုးတွေကို နှိုင်းယှဉ်ပါတယ်။ `if age = 20:` က error ဖြစ်ပါတယ်။
2. **colon `:` မေ့ခြင်း**: `if`, `elif`, `else` စာကြောင်းတွေ အားလုံး `:` နဲ့ ဆုံးရပါမယ်။
3. **indent မှားခြင်း**: အထဲက code ကို indent လုပ်ရမယ်၊ block တစ်ခုတည်းက စာကြောင်းတွေ indent တူရပါမယ်။
4. **`else` ကို condition ပေးခြင်း**: `else` မှာ condition ဘယ်တော့မှ မရှိပါဘူး။ `else age < 20:` က error ဖြစ်ပါတယ်။ condition လိုရင် `elif` ကို သုံးပါ။
5. **ဂဏန်းကို string နဲ့ နှိုင်းယှဉ်ခြင်း**: `"20"` (string) နဲ့ `20` (integer) က မတူပါဘူး။ `"20" == 20` က `False` ဖြစ်ပါတယ်။

အောက်က cell တွေမှာ အမှားတစ်ခုချင်းစီ ဘယ်လိုပုံစံလဲ ပြထားပါတယ်။ notebook မပျက်အောင် အမှား code ကို comment အဖြစ် ရေးထားပါတယ်။ တကယ့် error message ကို မြင်ချင်ရင် code စာကြောင်းက `#` ကို ဖျက်ကြည့်ပြီး ပြီးရင် `#` ကို ပြန်ထည့်ပါ။

````python
# အမှား 1: == အစား = ကို သုံးခြင်း
age = 20

# if age = 20:          # SyntaxError! = က "သိမ်းပါ" ဆိုတဲ့ အဓိပ္ပာယ်မို့ condition အဖြစ် သုံးလို့မရပါဘူး
#     print("20 years old")

if age == 20:           # မှန်ပါတယ်: == က "ညီသလား?" ဆိုတဲ့ အဓိပ္ပာယ်ပါ
    print("20 years old")
````

````python
# အမှား 2: colon မေ့ခြင်း
age = 20

# if age >= 20          # SyntaxError! အဆုံးမှာ colon ကျန်နေပါတယ်
#     print("adult")

if age >= 20:           # မှန်ပါတယ်: စာကြောင်းက colon နဲ့ ဆုံးပါတယ်
    print("adult")
````

````python
# အမှား 3: indent မှားခြင်း
age = 20

# if age >= 20:
# print("adult")        # IndentationError! if ထဲက စာကြောင်းကို indent မလုပ်ထားပါဘူး

if age >= 20:
    print("adult")      # မှန်ပါတယ်: space 4 ခု indent လုပ်ထားပါတယ်
````
