## 1. Tại sao dùng vòng lặp for với list?

Ở Bài 4, chúng ta đã học rằng list có thể chứa nhiều giá trị, và có thể lấy một giá trị bằng **chỉ số (index)** của nó (số chỉ vị trí). Giá trị đầu tiên ở chỉ số 0, giá trị thứ hai ở chỉ số 1, và cứ thế tiếp tục.

Vì vậy nếu muốn in mọi điểm số trong list, bạn có thể viết một dòng cho mỗi chỉ số:

```python
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
```

Cách này chạy được. Nhưng có hai vấn đề:

- Nếu list có 100 giá trị, bạn phải viết 100 dòng.
- Nếu thêm một giá trị mới bằng `append()`, bạn cũng phải tự thêm một dòng `print()` nữa.

**Vòng lặp for** giải quyết vấn đề này. Vòng lặp for lấy các giá trị ra khỏi list **lần lượt từng cái một, từ đầu đến cuối**, và chạy cùng một đoạn mã cho mỗi giá trị. List dài bao nhiêu cũng không sao. Bạn chỉ cần viết mã một lần.

````python
# Cách dài: mỗi giá trị một dòng
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
````

````python
# Cách ngắn: vòng lặp for
# Kết quả hoàn toàn giống nhau, nhưng chỉ có 2 dòng
scores = [85, 92, 78, 90, 66]
for score in scores:
    print(score)
````

````python
# Dù list dài thêm, vòng lặp for vẫn giữ nguyên
# Chúng ta không cần thêm dòng print() nào
scores = [85, 92, 78, 90, 66]
scores.append(100)      # Thêm một giá trị nữa vào cuối (bây giờ có 6 giá trị)
scores.append(73)       # Thêm một giá trị nữa (bây giờ có 7 giá trị)

for score in scores:    # Vẫn là 2 dòng giống như trên
    print(score)
````

## 2. Dạng cơ bản: `for giá_trị in list`

Hình dạng của mã như sau:

```python
for score in scores:
    mã cần lặp lại
```

Bạn có thể đọc nó như một câu: **"với mỗi score trong scores, hãy chạy đoạn mã bên dưới"**.

Hãy xem từng phần:

| Phần | Ý nghĩa |
|---|---|
| `for` | "lặp lại phần sau cho mỗi giá trị" |
| `score` | một biến. Mỗi lần, một giá trị từ list được lưu vào đây |
| `in` | "trong" |
| `scores` | list mà chúng ta duyệt qua |
| `:` | dấu hai chấm. Bắt buộc phải có ở cuối dòng `for` |
| `    mã cần lặp lại` | đoạn mã chạy cho mỗi giá trị. Phải được thụt lề (4 dấu cách) |

**Cách hoạt động, từng bước một** (với `scores = [85, 92, 78]`):

| Lần lặp | Giá trị được lưu vào `score` | Điều gì xảy ra |
|---|---|---|
| Lần 1 | `85` (giá trị ở chỉ số 0) | Đoạn mã thụt lề chạy với `score` = 85 |
| Lần 2 | `92` (giá trị ở chỉ số 1) | Đoạn mã thụt lề chạy với `score` = 92 |
| Lần 3 | `78` (giá trị ở chỉ số 2) | Đoạn mã thụt lề chạy với `score` = 78 |
| - | hết giá trị | Vòng lặp kết thúc, và Python chuyển sang đoạn mã phía sau vòng lặp |

Vì vậy vòng lặp chạy **số lần đúng bằng số giá trị trong list**. List có 3 giá trị → lặp 3 lần. List có 100 giá trị → lặp 100 lần.

````python
# In từng loại trái cây một
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:        # Lần 1: fruit = "apple", lần 2: fruit = "banana", lần 3: fruit = "cherry"
    print(fruit)            # In giá trị đang nằm trong fruit lúc này
````

````python
# Một list các số
numbers = [10, 20, 30, 40]

for number in numbers:      # number lần lượt là 10, rồi 20, rồi 30, rồi 40
    print(number)
````

````python
# Một list các giá trị bool (True / False)
answers = [True, False, True]

for answer in answers:      # answer lần lượt là True, rồi False, rồi True
    print(answer)
````

````python
# Chúng ta có thể tính toán với mỗi giá trị
# Mỗi giá được nhân 2 rồi in ra
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)        # 1000 * 2, rồi 2500 * 2, rồi 3000 * 2
````

````python
# Các giá trị bên trong list KHÔNG thay đổi
# Ở ô trên chúng ta in price * 2, nhưng bản thân list vẫn như cũ
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)

print(prices)               # Vẫn là [1000, 2500, 3000]
````

````python
# Nối chuỗi bằng +
# Hãy nhớ: + có thể nối hai chuỗi thành một ("Hello, " + "Bill" -> "Hello, Bill")
names = ["Bill", "Anna", "Tom"]

for name in names:
    print("Hello, " + name)
````

````python
# Nếu list rỗng, vòng lặp chạy 0 lần
# Không có gì được in ra, và cũng không có lỗi
empty_list = []

for value in empty_list:
    print("This never prints.")

print("The loop ran 0 times.")
````

## 3. Thụt lề: những dòng nào được lặp lại?

**Thụt lề** nghĩa là đẩy một dòng sang phải bằng cách thêm dấu cách ở đầu dòng. Trong Python, chúng ta dùng **4 dấu cách**.

```python
for fruit in fruits:
    print(fruit)          # <- dòng này được thụt lề (có 4 dấu cách ở đầu)
```

### Tại sao cần thụt lề?

Vòng lặp `for` lặp lại một đoạn mã. Nhưng làm sao Python biết **những dòng nào** cần lặp lại, và phần lặp lại kết thúc ở đâu?

Python nhìn vào thụt lề:

- Các dòng được **thụt lề** ngay sau dòng `for` → **bên trong** vòng lặp → được lặp lại cho mỗi giá trị
- Dòng đầu tiên **không thụt lề** trở lại → **bên ngoài** vòng lặp → chỉ chạy một lần, sau khi vòng lặp kết thúc

Các dòng được thụt lề thuộc về nhau, giống như một nhóm. Nhóm dòng này được gọi là **khối (block)**. Khối của vòng lặp `for` chính là "đoạn mã cần lặp lại".

(Đây cũng là quy tắc chúng ta đã dùng cho `if` ở Bài 2: các dòng thụt lề dưới `if` chỉ chạy khi điều kiện là True.)

### Cùng mã, thụt lề khác, kết quả khác

Hai ô bên dưới có các dòng hoàn toàn giống nhau. Điểm khác biệt **duy nhất** là dòng `print()` cuối cùng có được thụt lề hay không. Hãy xem kết quả thay đổi thế nào.

````python
# Phiên bản A: print() thứ hai được thụt lề -> nằm BÊN TRONG vòng lặp
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
    print("Done!")          # Được thụt lề -> bên trong vòng lặp -> in ra 3 lần
````

````python
# Phiên bản B: print() thứ hai KHÔNG thụt lề -> nằm BÊN NGOÀI vòng lặp
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
print("Done!")              # Không thụt lề -> bên ngoài vòng lặp -> chỉ in một lần, ở cuối
````

### Một khối có thể có nhiều dòng

Mọi dòng được thụt lề đều chạy, theo thứ tự từ trên xuống dưới, cho mỗi giá trị. Sau đó giá trị tiếp theo đến, và toàn bộ khối lại chạy từ đầu.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Now the value is:")      # Dòng thứ 1 của khối
    print(fruit)                    # Dòng thứ 2 của khối
    print("----------")             # Dòng thứ 3 của khối

print("The loop is finished.")      # Không thụt lề -> bên ngoài vòng lặp -> chạy một lần
````

### Mọi dòng trong một khối phải thụt lề bằng nhau

Nếu một dòng có 4 dấu cách còn dòng tiếp theo có 2 dấu cách, Python sẽ bị rối và báo lỗi. Hãy luôn dùng đúng 4 dấu cách. (Trong Jupyter, nhấn phím **Tab** sẽ tự thêm 4 dấu cách cho bạn.)

```python
for fruit in fruits:
    print(fruit)       # 4 dấu cách
  print("Done!")       # 2 dấu cách -> IndentationError!
```

## 4. Tên của biến lặp

Trong `for score in scores:`, từ `score` là một **biến mới** do vòng lặp for tạo ra. Mỗi lần lặp, một giá trị từ list được lưu vào đó.

**Bạn có thể đặt cho biến này bất kỳ tên nào bạn muốn.** Python không quan tâm đến tên. Nó chỉ lưu mỗi giá trị vào một chiếc hộp mang tên đó.

Ba ô bên dưới dùng các tên khác nhau (`x`, `score`, `banana`), nhưng kết quả hoàn toàn giống nhau.

````python
scores = [85, 92, 78]

for x in scores:            # Biến được đặt tên là x
    print(x)
````

````python
scores = [85, 92, 78]

for score in scores:        # Biến được đặt tên là score
    print(score)
````

````python
scores = [85, 92, 78]

for banana in scores:       # Biến được đặt tên là banana (lạ, nhưng vẫn chạy được!)
    print(banana)
````

Cả ba đều chạy được. Nhưng cái nào dễ hiểu nhất?

- `x` không cho biết giá trị đó là gì.
- `banana` gây nhầm lẫn, vì đây là điểm số chứ không phải chuối.
- `score` cho biết chính xác nó là gì: **một điểm số** trong list các điểm số.

**Một thói quen tốt:** đặt tên list bằng một từ số nhiều, và đặt tên biến lặp bằng dạng **số ít** của cùng từ đó.

| List (nhiều giá trị) | Biến lặp (một giá trị) |
|---|---|
| `scores` | `score` |
| `fruits` | `fruit` |
| `names` | `name` |
| `prices` | `price` |

Như vậy khi đọc mã, bạn có thể dễ dàng phân biệt "cả list" (`scores`) với "một giá trị" (`score`).

**Một quy tắc:** tên biến phải **khác** với tên list. Chúng ta sẽ xem điều gì xảy ra nếu chúng giống nhau ở phần 12 (Những lỗi thường gặp).

## 5. In nhiều giá trị cùng lúc bằng dấu phẩy

3. Chữ bên trong dấu ngoặc kép `" "` được in **nguyên như vậy**.
4. Những thứ bên ngoài dấu ngoặc kép được **tính trước**, rồi kết quả mới được in ra. Nếu đó là một biến, giá trị bên trong biến sẽ được in ra.

````python
# Nhiều chuỗi
print("apple", "banana", "cherry")     # Một dấu cách được tự động thêm vào giữa các giá trị
````

````python
# Nhiều số
print(10, 20, 30)
````

## 6. Cộng dồn các giá trị

Một trong những công việc phổ biến nhất của vòng lặp for là **cộng dồn** tất cả các giá trị trong list. Ví dụ: tổng của tất cả các điểm số là bao nhiêu?

Mẹo là dùng một biến làm **tổng tích lũy**:

1. **Trước vòng lặp:** tạo một biến và gán bằng 0. → `total = 0`
2. **Bên trong vòng lặp:** cộng giá trị hiện tại vào nó. → `total = total + score`
3. **Sau vòng lặp:** biến đó chứa tổng của tất cả các giá trị.

Hãy xem kỹ bước 2: `total = total + score`

- `=` không có nghĩa là "bằng". Nó có nghĩa là **"lưu giá trị bên phải vào chiếc hộp bên trái"**.
- Vì vậy Python tính vế phải trước, `total + score` (tổng cũ cộng điểm hiện tại),
- rồi lưu kết quả đó trở lại vào `total`.
- Đây là cùng ý tưởng với `age = age + 1`: lấy giá trị cũ, cộng thêm gì đó, rồi đặt lại vào cùng chiếc hộp.

````python
# Cộng tất cả các điểm số
scores = [85, 92, 78, 90, 66]
total = 0                       # Bước 1: tổng tích lũy bắt đầu từ 0 (TRƯỚC vòng lặp)

for score in scores:
    total = total + score       # Bước 2: tổng cũ + điểm hiện tại -> lưu lại vào total

print("Total:", total)          # Bước 3: SAU vòng lặp (không thụt lề), in kết quả một lần
````

````python
# Hãy xem total thay đổi thế nào qua mỗi lần lặp
# Cùng mã như trên, nhưng có print() BÊN TRONG vòng lặp (được thụt lề)
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total = total + score
    print("score:", score, "total:", total)      # Được thụt lề -> in ra mỗi lần lặp

# Lần 1: 0 + 85 = 85
# Lần 2: 85 + 92 = 177
# Lần 3: 177 + 78 = 255
# Lần 4: 255 + 90 = 345
# Lần 5: 345 + 66 = 411
````

### Cách viết ngắn hơn: `+=`

`total = total + score` được dùng thường xuyên đến mức Python có một cách viết ngắn hơn:

```python
total += score
```

`+=` có nghĩa là **"cộng vế phải vào biến bên trái, rồi lưu kết quả trở lại vào biến đó"**.

`total += score` và `total = total + score` có **ý nghĩa hoàn toàn giống nhau**. Bạn có thể dùng cách nào cũng được. Nhiều người lập trình Python dùng cách viết ngắn, nên bạn sẽ thấy nó thường xuyên.

````python
# Cách viết dài
scores = [85, 92, 78]
total = 0
for score in scores:
    total = total + score
print("Long form:", total)

# Cách viết ngắn
total = 0
for score in scores:
    total += score              # giống với: total = total + score
print("Short form:", total)     # Kết quả giống nhau
````

````python
# Tổng tiền của giỏ hàng
cart = [3500, 1200, 8000, 2300]
total_price = 0                 # Trước vòng lặp: bắt đầu từ 0

for price in cart:
    total_price += price        # giống với: total_price = total_price + price

print("Total price:", total_price)
````

### Trung bình

**Trung bình** là tổng chia cho số lượng giá trị.

Làm sao để có số lượng giá trị? Hãy nhớ `len()`: nó cho biết list có bao nhiêu giá trị. `len([85, 92, 78])` là `3`.

Vậy: **trung bình = tổng / len(list)**

Quan trọng: hãy tính trung bình **sau** vòng lặp, khi tổng đã hoàn chỉnh.

````python
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total += score              # giống với: total = total + score

average = total / len(scores)   # Sau vòng lặp: tổng (411) / số lượng điểm (5)
print("Total:", total)
print("Number of scores:", len(scores))
print("Average:", average)
````

## 7. Dùng if bên trong vòng lặp

Bạn có thể đặt `if` **bên trong** vòng lặp for. Khi đó từng giá trị sẽ được kiểm tra lần lượt, và bạn có thể làm gì đó **chỉ với những giá trị bạn muốn**.

Hãy nhớ cách `if` hoạt động:
- `if điều_kiện:` → đoạn mã thụt lề bên dưới chỉ chạy **khi điều kiện là True**.
- `else:` → đoạn mã thụt lề bên dưới chạy khi điều kiện là False.

Khi đặt `if` bên trong `for`, sẽ có **hai mức** thụt lề:

```python
for score in scores:          # không có dấu cách
    if score >= 60:           # 4 dấu cách -> bên trong for (chạy với mọi score)
        print(score)          # 8 dấu cách -> bên trong if (chỉ chạy khi điều kiện là True)
```

- Dòng `if` có 4 dấu cách, vì nó thuộc về khối của `for`.
- Dòng `print()` có 8 dấu cách, vì nó thuộc về khối của `if`, mà khối này lại nằm trong khối của `for`.

````python
# Chỉ in những điểm đậu (từ 60 trở lên)
scores = [85, 42, 78, 55, 90]

for score in scores:            # Lấy ra từng điểm
    if score >= 60:             # 4 dấu cách: kiểm tra điểm này
        print(score)            # 8 dấu cách: chỉ in nếu điểm từ 60 trở lên
````

````python
# Hãy xem những điểm nào được kiểm tra và những điểm nào được in ra
scores = [85, 42, 78, 55, 90]

for score in scores:
    print("Checking", score)    # 4 dấu cách: chạy với MỌI điểm
    if score >= 60:
        print("pass")           # 8 dấu cách: chỉ chạy khi score >= 60
````

````python
# if - else: in đậu hay rớt cho mọi điểm
scores = [85, 42, 78, 55, 90]

for score in scores:
    if score >= 60:
        print(score, "pass")    # Chạy khi điều kiện là True
    else:
        print(score, "fail")    # Chạy khi điều kiện là False
````

````python
# Chỉ in các số chẵn
# Một số là số chẵn nếu số dư khi chia cho 2 bằng 0
# % cho ra số dư: 8 % 2 là 0 (chẵn), 5 % 2 là 1 (lẻ)
numbers = [3, 8, 5, 12, 7, 6]

for number in numbers:
    if number % 2 == 0:         # Số dư có bằng 0 không? -> chẵn
        print(number)
````

````python
# if - elif - else bên trong vòng lặp: xếp loại cho mỗi điểm
# Python kiểm tra các điều kiện từ trên xuống dưới, và chỉ chạy điều kiện True ĐẦU TIÊN
scores = [95, 82, 74, 58]

for score in scores:
    if score >= 90:
        print(score, "A")
    elif score >= 80:
        print(score, "B")
    elif score >= 70:
        print(score, "C")
    else:
        print(score, "F")
````

````python
# Chỉ cộng một số giá trị: if + tổng tích lũy
# Chỉ cộng các số chẵn
numbers = [3, 8, 5, 12, 7, 6]
even_total = 0                  # Trước vòng lặp: bắt đầu từ 0

for number in numbers:
    if number % 2 == 0:         # Chỉ khi số đó là số chẵn...
        even_total += number    # ...mới cộng vào (giống với: even_total = even_total + number)

print("Sum of even numbers:", even_total)    # 8 + 12 + 6 = 26
````

## 8. Đếm

1. **Trước vòng lặp:** `count = 0`

## 9. Tìm giá trị lớn nhất và nhỏ nhất

Làm sao tìm điểm cao nhất trong một list? Hãy nghĩ xem một người sẽ làm thế nào với một dãy số viết trên giấy:

1. Nhìn số đầu tiên, và ghi nhớ nó là **"số lớn nhất cho đến giờ"**.
2. Nhìn số tiếp theo. Nó có lớn hơn số đang nhớ không? Nếu có, hãy nhớ số này thay thế.
3. Cứ làm như vậy cho đến cuối. Số bạn nhớ cuối cùng chính là số lớn nhất.

Chúng ta có thể viết đúng các bước này trong Python. Biến `highest` là "số chúng ta đang nhớ".

**Tại sao bắt đầu bằng giá trị đầu tiên (`scores[0]`)?** Vì chúng ta cần một thứ gì đó để so sánh. Chỉ số 0 là giá trị đầu tiên của list.

````python
# Tìm điểm cao nhất
scores = [85, 92, 78, 95, 66]
highest = scores[0]             # Bước 1: nhớ điểm đầu tiên (85) là "lớn nhất cho đến giờ"

for score in scores:
    if score > highest:         # Bước 2: điểm này có lớn hơn điểm đang nhớ không?
        highest = score         #         nếu có, hãy nhớ điểm này thay thế

print("Highest:", highest)      # Bước 3: điểm chúng ta nhớ cuối cùng
````

````python
# Hãy xem từng bước một
scores = [85, 92, 78, 95, 66]
highest = scores[0]

for score in scores:
    if score > highest:
        highest = score
    print("score:", score, "highest:", highest)     # 4 dấu cách: chạy mỗi lần lặp
````

````python
# Giá trị nhỏ nhất cũng làm tương tự, chỉ dùng < thay cho >
temperatures = [12, 8, 15, 3, 10]
lowest = temperatures[0]        # Bắt đầu với nhiệt độ đầu tiên (12)

for temperature in temperatures:
    if temperature < lowest:    # Giá trị này có nhỏ hơn giá trị đang nhớ không?
        lowest = temperature

print("Lowest:", lowest)
````

````python
# Món đắt nhất trong giỏ hàng
cart = [3500, 1200, 8000, 2300]
most_expensive = cart[0]

for price in cart:
    if price > most_expensive:
        most_expensive = price

print("Most expensive:", most_expensive)
````

## 11. Dùng in và break bên trong vòng lặp

### in

`giá_trị in list` kiểm tra xem giá trị có nằm trong list không. Kết quả là `True` hoặc `False`, nên có thể dùng làm điều kiện của `if`.

```python
"milk" in ["bread", "milk"]     # True
"eggs" in ["bread", "milk"]     # False
```

**Hãy cẩn thận:** `in` trong `for fruit in fruits:` và `in` trong `if fruit in store:` trông giống nhau, nhưng làm hai việc khác nhau.

| | Ý nghĩa |
|---|---|
| `for fruit in fruits:` | **lấy ra** từng giá trị trong fruits, lần lượt từng cái |
| `if fruit in store:` | **kiểm tra** xem fruit có nằm trong store không (True / False) |

````python
# Cửa hàng có những món nào trong danh sách mua sắm của tôi?
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]

for item in shopping_list:              # "in" số 1: lấy ra từng món
    if item in store:                   # "in" số 2: món này có nằm trong store không?
        print(item, "yes")
    else:
        print(item, "no")
````

````python
# Tạo list các món vẫn cần mua ở chỗ khác
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]
not_found = []

for item in shopping_list:
    if item not in store:               # not in: True khi món đó KHÔNG nằm trong store
        not_found.append(item)

print("Not found:", not_found)
````

### break

`break` **dừng vòng lặp ngay lập tức**. Các giá trị còn lại hoàn toàn không được kiểm tra, và Python nhảy đến đoạn mã phía sau vòng lặp.

Nó hữu ích khi bạn đã tìm thấy thứ cần tìm, và không cần tìm tiếp nữa. `break` hầu như luôn được dùng bên trong `if`.

````python
# Dừng ngay khi tìm thấy điểm rớt đầu tiên
scores = [85, 92, 55, 78, 40]

for score in scores:
    if score < 60:
        print("Found a failing score:", score)
        break                           # Dừng vòng lặp ngay tại đây
    print(score, "is OK")

print("The loop is over.")              # 78 và 40 không bao giờ được kiểm tra
````

````python
# Trong list có ai tên Tom không?
names = ["Bill", "Anna", "Tom", "Kate", "Mike"]

for name in names:
    print("Checking", name)
    if name == "Tom":
        print("Found Tom!")
        break                           # Không cần kiểm tra Kate và Mike
````

## 12. Những lỗi thường gặp

Đây là những lỗi mà hầu như ai cũng mắc phải khi dùng vòng lặp for với list. Với mỗi lỗi, chúng ta sẽ xem:

- ❌ **Mã sai**
- 🔍 **Sai ở đâu**
- ✅ **Cách làm đúng**

Một số mã sai gây ra lỗi. Mã đó được viết dưới dạng chú thích (có `#` ở đầu) để không làm notebook dừng lại. Nếu muốn xem thông báo lỗi thật, hãy xóa `#`, chạy ô đó, rồi đặt `#` lại.

Một số mã sai khác **không báo lỗi, nhưng cho ra kết quả sai**. Những lỗi này còn nguy hiểm hơn, vì Python không cảnh báo bạn.

### Lỗi 1: Quên dấu hai chấm

````python
# ❌ Sai
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits           # SyntaxError
#     print(fruit)
````

🔍 **Sai ở đâu?**

- Dòng `for` luôn phải kết thúc bằng dấu hai chấm `:`.
- Dấu hai chấm báo cho Python biết: "đoạn mã cần lặp lại bắt đầu từ dòng tiếp theo".
- Thiếu nó, Python không hiểu mã và báo lỗi `SyntaxError` ("ngữ pháp của mã bị sai").

✅ **Cách làm đúng:** Đặt `:` ở cuối dòng `for`.

````python
# ✅ Đúng
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # Có dấu hai chấm ở cuối
    print(fruit)
````

### Lỗi 2: Quên thụt lề

````python
# ❌ Sai
fruits = ["apple", "banana", "cherry"]

# for fruit in fruits:
# print(fruit)                  # IndentationError
````

🔍 **Sai ở đâu?**

- Sau dòng `for`, phải có ít nhất một dòng được **thụt lề**: đó là đoạn mã cần lặp lại.
- Ở đây, `print(fruit)` không có dấu cách nào ở đầu, nên Python nghĩ vòng lặp `for` không có gì để lặp.
- Lỗi là `IndentationError` ("thụt lề bị sai").

✅ **Cách làm đúng:** Thụt lề đoạn mã cần lặp lại 4 dấu cách (hoặc nhấn Tab một lần).

````python
# ✅ Đúng
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # 4 dấu cách ở đầu
````

### Lỗi 3: Thụt lề một dòng lẽ ra chỉ chạy một lần

````python
# ❌ Sai: chúng ta muốn in tổng chỉ một lần, ở cuối
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score
    print("Total:", total)      # Được thụt lề -> bên trong vòng lặp -> in ra 3 lần!
````

🔍 **Sai ở đâu?**

- `print("Total:", total)` được thụt lề, nên nó nằm **bên trong** vòng lặp.
- Nghĩa là nó chạy ở mọi lần lặp, và in ra tổng khi tổng vẫn đang được cộng dồn.
- Không có lỗi nào, nhưng chúng ta nhận được 3 dòng thay vì 1, và hai dòng đầu không phải tổng thật.

✅ **Cách làm đúng:** Bỏ thụt lề ở dòng chỉ nên chạy một lần. Khi đó nó nằm **bên ngoài** vòng lặp, và chỉ chạy sau khi vòng lặp kết thúc.

````python
# ✅ Đúng
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score              # Bên trong vòng lặp: cộng dồn
print("Total:", total)          # Không thụt lề -> bên ngoài vòng lặp -> in một lần
````

### Lỗi 4: Dùng cùng một tên cho list và biến lặp

````python
# ❌ Sai
fruits = ["apple", "banana", "cherry"]

for fruits in fruits:           # Biến lặp có CÙNG tên với list
    print(fruits)               # Dòng này in ra đúng...

print(fruits)                   # ...nhưng bây giờ fruits chỉ còn là "cherry". List đã mất!
````

🔍 **Sai ở đâu?**

- Vòng lặp có vẻ chạy đúng, nên khó nhận ra.
- Nhưng mỗi lần lặp lại lưu một giá trị vào chiếc hộp tên `fruits`. Hãy nhớ: khi bạn lưu một giá trị mới vào biến, giá trị cũ sẽ biến mất.
- Vì vậy list đang nằm trong `fruits` bị thay bằng `"apple"`, rồi `"banana"`, rồi `"cherry"`.
- Sau vòng lặp, `fruits` chỉ còn chứa `"cherry"`. List đã mất.

✅ **Cách làm đúng:** Đặt cho biến lặp một tên **khác**. Dạng số ít của tên list là lựa chọn tốt nhất: `fruits` → `fruit`.

````python
# ✅ Đúng
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # fruit (một giá trị) và fruits (cả list) là hai tên khác nhau
    print(fruit)

print(fruits)                   # List vẫn an toàn
````

🔍 **Sai ở đâu?**

- Ở mỗi lần lặp, tổng quay về 0, rồi chỉ có điểm hiện tại được cộng vào.
- Lần 1: 0 + 85 = 85 → Lần 2: **quay về 0**, 0 + 92 = 92 → Lần 3: **quay về 0**, 0 + 78 = 78
- Vì vậy cuối cùng, `total` chỉ chứa điểm cuối cùng.

✅ **Cách làm đúng:** Viết `total = 0` **một lần, trước** vòng lặp (không thụt lề).

````python
# ✅ Đúng
scores = [85, 92, 78]
total = 0                       # Trước vòng lặp -> chỉ chạy một lần

for score in scores:
    total += score              # giống với: total = total + score

print("Total:", total)          # 85 + 92 + 78 = 255
````

### Lỗi 8: Tính trung bình bên trong vòng lặp

````python
# ❌ Sai
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score
    average = total / len(scores)       # Được thụt lề -> được tính ở mọi lần lặp
    print("Average:", average)
````

🔍 **Sai ở đâu?**

- Để tính trung bình, chúng ta cần tổng của **tất cả** các giá trị.
- Nhưng bên trong vòng lặp, tổng vẫn chưa hoàn chỉnh.
- Lần 1: 80 / 3 → Lần 2: 170 / 3 → Lần 3: 270 / 3. Chỉ có kết quả cuối là trung bình thật.

✅ **Cách làm đúng:** Bên trong vòng lặp, chỉ cộng dồn. Tính trung bình **sau** vòng lặp, khi tổng đã hoàn chỉnh.

````python
# ✅ Đúng
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score                      # Bên trong vòng lặp: chỉ cộng dồn

average = total / len(scores)           # Sau vòng lặp: 270 / 3
print("Average:", average)
````

### Lỗi 9: Cố thay đổi list bằng cách thay đổi biến lặp

````python
# ❌ Sai: chúng ta muốn cộng 5 điểm thưởng cho mọi điểm trong list
scores = [85, 92, 78]

for score in scores:
    score = score + 5           # Trông như đang thay đổi điểm...

print(scores)                   # ...nhưng list vẫn là [85, 92, 78]!
````

🔍 **Sai ở đâu?**

- `score` là một **chiếc hộp riêng**. Ở mỗi lần lặp, một giá trị được **sao chép** từ list vào chiếc hộp này.
- `score = score + 5` chỉ thay đổi thứ bên trong hộp `score`. List `scores` hoàn toàn không bị động đến.
- Giống như chép một con số từ vở sang tờ giấy nhớ, rồi sửa con số trên tờ giấy nhớ. Con số trong vở vẫn giữ nguyên.
- Không có lỗi nào, nên lỗi này khó nhận ra.

✅ **Cách làm đúng:** Tạo một **list mới** bằng `append()` (phần 10), và đặt các giá trị đã thay đổi vào đó.

````python
# ✅ Đúng
scores = [85, 92, 78]
new_scores = []                         # Một list rỗng cho các giá trị đã thay đổi

for score in scores:
    new_scores.append(score + 5)        # Thêm giá trị đã thay đổi vào list mới

print(new_scores)                       # [90, 97, 83]
````

### Lỗi 10: Xóa giá trị khỏi list trong lúc lặp

````python
# ❌ Sai: chúng ta muốn xóa mọi điểm rớt (dưới 60)
scores = [50, 40, 80, 30]

for score in scores:
    if score < 60:
        scores.remove(score)

print(scores)                   # [40, 80] -> 40 vẫn còn!
````

🔍 **Sai ở đâu?**

Vòng lặp for duyệt list **theo vị trí**: giá trị thứ 1, rồi giá trị thứ 2, rồi giá trị thứ 3 ...
Nhưng `remove()` lấy một giá trị ra, và mọi giá trị phía sau nó **tiến lên** một vị trí.

| Lần lặp | Vị trí vòng lặp nhìn vào | List lúc đó | Điều gì xảy ra |
|---|---|---|---|
| Lần 1 | vị trí thứ 1 | `[50, 40, 80, 30]` | 50 < 60 → xóa 50. List trở thành `[40, 80, 30]` |
| Lần 2 | vị trí thứ 2 | `[40, 80, 30]` | Giá trị thứ 2 bây giờ là **80**. 40 đã tiến lên vị trí thứ 1, nên vòng lặp **bỏ qua** nó! |
| Lần 3 | vị trí thứ 3 | `[40, 80, 30]` | 30 < 60 → xóa 30. List trở thành `[40, 80]` |

- 40 không bao giờ được kiểm tra, nên nó vẫn ở lại trong list.
- Không có lỗi nào, nên rất khó nhận ra.

✅ **Cách làm đúng:** Đừng thay đổi list trong khi đang lặp qua nó. Thay vào đó, hãy tạo một **list mới** chỉ gồm những giá trị bạn muốn giữ lại (phần 10).

````python
# ✅ Đúng
scores = [50, 40, 80, 30]
kept = []

for score in scores:
    if score >= 60:             # Chỉ giữ lại các điểm đậu
        kept.append(score)

print(kept)                     # [80]
````

### Lỗi 12: Bắt đầu "lớn nhất cho đến giờ" từ 0

````python
# ❌ Sai: tìm nhiệt độ cao nhất vào mùa đông
temperatures = [-5, -2, -8, -3]
highest = 0                     # Bắt đầu từ 0...

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # 0 -> nhưng 0 thậm chí không có trong list!
````

🔍 **Sai ở đâu?**

- Tất cả nhiệt độ đều dưới 0. Vì vậy không có giá trị nào lớn hơn 0.
- `temperature > highest` không bao giờ là True, và `highest` vẫn là 0 cho đến cuối.
- Rất dễ nghĩ "mình nên bắt đầu từ 0", vì tổng tích lũy bắt đầu từ 0. Nhưng tìm giá trị lớn nhất là một việc khác.

✅ **Cách làm đúng:** Bắt đầu bằng **giá trị đầu tiên của list** (`temperatures[0]`). Khi đó giá trị bắt đầu luôn lấy từ chính list.

````python
# ✅ Đúng
temperatures = [-5, -2, -8, -3]
highest = temperatures[0]       # Bắt đầu từ giá trị đầu tiên (-5)

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # -2
````

### Lỗi 13: Nối chuỗi và số bằng +

````python
# ❌ Sai
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

# print("Total: " + total)      # TypeError: can only concatenate str (not "int") to str
````

🔍 **Sai ở đâu?**

✅ **Cách làm đúng:** Dùng **dấu phẩy** trong `print()` (phần 5). Với dấu phẩy, bạn có thể trộn chuỗi và số.

````python
# ✅ Đúng
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

print("Total:", total)          # Dấu phẩy -> dùng được với chuỗi và số
````
