## 1. slicing ဆိုတာဘာလဲ?

သိပြီးသား အရာကနေ စကြရအောင်။

**list** က square bracket `[ ]` ထဲမှာ တန်ဖိုး အများကြီးကို အစီအစဉ်လိုက် သိမ်းထားပါတယ်:

```python
scores = [85, 92, 78, 90, 66]
```

တန်ဖိုးတိုင်းမှာ **index** လို့ ခေါ်တဲ့ နေရာ နံပါတ် ရှိပါတယ်။ အရေးအကြီးဆုံး စည်းမျဉ်း: **Python က 1 ကနေ မဟုတ်ဘဲ 0 ကနေ စရေတွက်ပါတယ်။**

| တန်ဖိုး | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| index | 0 | 1 | 2 | 3 | 4 |

တန်ဖိုး **တစ်ခု** ကို ယူဖို့ list နာမည်နဲ့ square bracket ထဲမှာ index ကို ရေးပါတယ်: `scores[0]` က `85`၊ `scores[2]` က `78` ပါ။

ဒါပေမဲ့ **တန်ဖိုး အများကြီးကို တစ်ခါတည်း** ယူချင်ရင်ရော? ဥပမာ ပထမ ကျောင်းသား 3 ယောက်ရဲ့ အမှတ်တွေကိုပဲ?

သိထားတာနဲ့ဆိုရင် တစ်ခုချင်း ယူရပါမယ်: `scores[0]`, `scores[1]`, `scores[2]`။

**slicing** က list ရဲ့ **အစိတ်အပိုင်း တစ်ခု** ကို အဆင့် တစ်ဆင့်တည်းနဲ့ ယူခွင့်ပေးပါတယ်။ "slice" ဆိုတဲ့ စကားလုံးက "တစ်ခုခုကနေ လှီးယူထားတဲ့ အပိုင်း" လို့ အဓိပ္ပာယ်ရပါတယ်၊ ပေါင်မုန့် တစ်ချပ် ဒါမှမဟုတ် pizza တစ်ချပ်လိုပါပဲ။ list ထဲကနေ အပိုင်း တစ်ခုကို လှီးယူပါတယ်။

slicing ရဲ့ ရလဒ်က အဲဒီ အပိုင်းပဲ ပါတဲ့ **list အသစ်** တစ်ခု ဖြစ်ပါတယ်။

````python
scores = [85, 92, 78, 90, 66]

# index နဲ့ တန်ဖိုး "တစ်ခု" ယူခြင်း (သိပြီးသား အရာ)
print(scores[0])        # index 0 -> 1 ခုမြောက် တန်ဖိုး -> 85
print(scores[1])        # index 1 -> 2 ခုမြောက် တန်ဖိုး -> 92
print(scores[2])        # index 2 -> 3 ခုမြောက် တန်ဖိုး -> 78
````

````python
scores = [85, 92, 78, 90, 66]

# slicing နဲ့ တန်ဖိုး "အများကြီး" ကို တစ်ခါတည်း ယူခြင်း (ဒီနေ့ လေ့လာမယ့် အရာ)
# scores[0:3] ဆိုတာ "index 0 ကနေ index 3 မတိုင်ခင်အထိ (3 မပါ)" လို့ အဓိပ္ပာယ်ရပါတယ်
print(scores[0:3])      # [85, 92, 78] -> ပထမ တန်ဖိုး 3 ခု ပါတဲ့ list အသစ်
````

အပေါ်က ရလဒ် နှစ်ခုကို သေချာ ကြည့်ပါ:

- `scores[0]` က **တန်ဖိုး တစ်ခု** ကို ပေးပါတယ်: `85` (square bracket မပါ)
- `scores[0:3]` က **list** ကို ပေးပါတယ်: `[85, 92, 78]` (square bracket ပါ)

ဒီကွာခြားချက်က အရမ်းအရေးကြီးပါတယ်။ index က တန်ဖိုးကို ပေးပါတယ်။ slice က list ကို ပေးပါတယ်။

## 2. အခြေခံပုံစံ: `list[start:end]`

```python
list_နာမည်[start:end]
```

- `start`: slice **စတဲ့** index (ဒီတန်ဖိုး **ပါပါတယ်**)
- `end`: slice **ရပ်တဲ့** index (ဒီတန်ဖိုး **မပါပါဘူး**)
- `:` ဂဏန်း နှစ်ခုကြားက colon

`scores[1:4]` ကို ဒီလို ဖတ်လို့ရပါတယ်: **"index 1 ကနေ index 4 မတိုင်ခင်အထိ (4 မပါ)"**။

### end မပါပါဘူး

ဒါက စတင်လေ့လာသူတွေကို အရှုပ်ထွေးဆုံး ဖြစ်စေတဲ့ အပိုင်းပါ။ ဇယားနဲ့ ဖြည်းဖြည်း ကြည့်ရအောင်။

```python
scores = [85, 92, 78, 90, 66]
```

| index | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| တန်ဖိုး | 85 | 92 | 78 | 90 | 66 |
| `scores[1:4]` ထဲမှာ ပါလား? | ❌ | ✅ | ✅ | ✅ | ❌ |

`scores[1:4]` → index 1, 2, 3 → `[92, 78, 90]`။ index 4 က slice **ရပ်တဲ့** နေရာ ဖြစ်လို့ 66 မပါပါဘူး။

**ဒီစည်းမျဉ်းကို အရင်က မြင်ဖူးလား?** မြင်ဖူးပါတယ်! သင်ခန်းစာ 3 မှာ `range(1, 4)` က 1, 2, 3 ဂဏန်းတွေကို ပေးပြီး 4 က **မပါခဲ့ပါဘူး**။ slicing က အတိအကျ တူညီတဲ့ စည်းမျဉ်းကို သုံးပါတယ်: **start ပါပြီး end မပါပါဘူး**။

### တန်ဖိုး ဘယ်နှစ်ခု ရလဲ?

end မပါလို့ တန်ဖိုး အရေအတွက်က ရိုးရိုးရှင်းရှင်း **end - start** ဖြစ်ပါတယ်။

- `scores[1:4]` → 4 - 1 = တန်ဖိုး **3** ခု
- `scores[0:2]` → 2 - 0 = တန်ဖိုး **2** ခု
- `scores[2:5]` → 5 - 2 = တန်ဖိုး **3** ခု

````python
scores = [85, 92, 78, 90, 66]
#  index:   0   1   2   3   4

print(scores[1:4])      # index 1, 2, 3 (4 မပါ) -> [92, 78, 90]
````

````python
scores = [85, 92, 78, 90, 66]
#  index:   0   1   2   3   4

print(scores[0:2])      # index 0, 1 -> [85, 92]
print(scores[2:5])      # index 2, 3, 4 -> [78, 90, 66]
print(scores[3:4])      # index 3 တစ်ခုတည်း -> [90] (တန်ဖိုး တစ်ခုတည်း ပါတဲ့ list)
````

````python
# သင်ခန်းစာ 3 က range() နဲ့ စည်းမျဉ်း အတူတူ: end မပါ
# range(1, 4) က 1, 2, 3 ကို ပေးပါတယ်
for i in range(1, 4):
    print(i)
````

````python
# string တွေနဲ့လည်း အတိအကျ အတူတူ အလုပ်လုပ်ပါတယ်
fruits = ["apple", "banana", "cherry", "grape", "melon"]
#  index:      0        1         2        3        4

print(fruits[0:2])      # index 0, 1 -> ['apple', 'banana']
print(fruits[1:3])      # index 1, 2 -> ['banana', 'cherry']
print(fruits[2:5])      # index 2, 3, 4 -> ['cherry', 'grape', 'melon']
````

### len() နဲ့ တန်ဖိုး အရေအတွက်ကို ရေတွက်ခြင်း

`len()` ကို မှတ်မိသေးလား? list ထဲမှာ **တန်ဖိုး ဘယ်နှစ်ခု** ရှိလဲ ပြောပြပါတယ်။ `len([85, 92, 78])` က `3` ပါ။

slice က list ဖြစ်လို့ သူ့ကိုလည်း `len()` သုံးလို့ရပါတယ်။ "end - start" စည်းမျဉ်းကို စစ်ကြည့်ရအောင်။

````python
scores = [85, 92, 78, 90, 66]

print(len(scores))          # list တစ်ခုလုံးမှာ တန်ဖိုး 5 ခု
print(len(scores[1:4]))     # 4 - 1 = တန်ဖိုး 3 ခု
print(len(scores[0:2]))     # 2 - 0 = တန်ဖိုး 2 ခု
````

### slice က list အသစ်ပါ။ မူလ list က မပြောင်းပါဘူး။

slicing က list ရဲ့ အစိတ်အပိုင်း တစ်ခုကို **list အသစ်** ထဲကို **ကူးယူ** ပါတယ်။ မူလ list က အတိအကျ အတူတူ ရှိနေပါတယ်။

slice ကို နောက်မှ သုံးချင်ရင် တခြား တန်ဖိုးတွေလိုပဲ variable ထဲ သိမ်းပါ။ မှတ်ထားပါ: `=` က **"ညာဘက်က တန်ဖိုးကို ဘယ်ဘက်က သေတ္တာထဲ သိမ်းပါ"** လို့ အဓိပ္ပာယ်ရပါတယ်။

````python
scores = [85, 92, 78, 90, 66]

first_three = scores[0:3]   # list အသစ်ကို first_three ဆိုတဲ့ variable ထဲ သိမ်းပါတယ်

print(first_three)          # [85, 92, 78] -> list အသစ်
print(scores)               # [85, 92, 78, 90, 66] -> မူလ list က ပြည့်စုံဆဲ
````

````python
# list အသစ်က တကယ့် list ဖြစ်လို့ type က list ပါ
# type() က တန်ဖိုးတစ်ခုဟာ ဘယ်လို အမျိုးအစားလဲ ပြောပြပါတယ်
scores = [85, 92, 78, 90, 66]
first_three = scores[0:3]

print(type(first_three))    # <class 'list'>
print(type(scores[0]))      # <class 'int'> -> index က list မဟုတ်ဘဲ တန်ဖိုး တစ်ခု (integer) ကို ပေးပါတယ်
````

````python
# list အသစ်မှာလည်း index ကို သုံးလို့ရပါတယ်
# မှတ်ထားပါ: list အသစ်က index 0 ကနေ ပြန်စပါတယ်!
scores = [85, 92, 78, 90, 66]
last_two = scores[3:5]      # [90, 66]

print(last_two[0])          # list "အသစ်" ရဲ့ 1 ခုမြောက် တန်ဖိုး -> 90 (မူလ list မှာ index 3 ဖြစ်ခဲ့ပါတယ်)
print(last_two[1])          # list "အသစ်" ရဲ့ 2 ခုမြောက် တန်ဖိုး -> 66
````

## 3. start ဒါမှမဟုတ် end ကို ချန်ထားခြင်း

ဂဏန်း နှစ်ခုလုံးကို အမြဲ ရေးစရာ မလိုပါဘူး။ ကျန်နေတဲ့ ဂဏန်းကို Python က ဖြည့်ပေးပါတယ်။

| ရေးပုံ | အဓိပ္ပာယ် | Python နားလည်ပုံ |
|---|---|---|
| `scores[:3]` | **အစ** ကနေ index 3 မတိုင်ခင်အထိ | `scores[0:3]` |
| `scores[2:]` | index 2 ကနေ **အဆုံး** အထိ | `scores[2:5]` (တန်ဖိုး 5 ခု ရှိတဲ့အခါ) |
| `scores[:]` | အစကနေ အဆုံးအထိ (list တစ်ခုလုံး) | `scores[0:5]` |

- start မရှိ → **အစ** (index 0) ကနေ စပါတယ်
- end မရှိ → **အဆုံး** အထိ သွားပါတယ် (နောက်ဆုံး တန်ဖိုး ပါပါတယ်)

**ဒါက ဘာလို့ အသုံးဝင်တာလဲ?**
- `scores[:3]` က `scores[0:3]` ထက် တိုပြီး ဖတ်ရလွယ်ပါတယ်: "ပထမ 3 ခု"။
- `scores[2:]` က list ဘယ်လောက်ရှည်လဲ မသိဘဲ အလုပ်လုပ်ပါတယ်။ တန်ဖိုးတွေကို ရေတွက်စရာ မလိုပါဘူး။

````python
scores = [85, 92, 78, 90, 66]
#  index:   0   1   2   3   4

# start မရှိ -> အစကနေ
print(scores[:3])       # scores[0:3] နဲ့ အတူတူ -> [85, 92, 78]
print(scores[0:3])      # ရလဒ် အတူတူပဲ
````

````python
scores = [85, 92, 78, 90, 66]
#  index:   0   1   2   3   4

# end မရှိ -> အဆုံးအထိ (နောက်ဆုံး တန်ဖိုး "ပါပါတယ်")
print(scores[2:])       # index 2, 3, 4 -> [78, 90, 66]
print(scores[1:])       # ပထမ တစ်ခုကလွဲပြီး အားလုံး -> [92, 78, 90, 66]
````

````python
# [:n] = "ပထမ တန်ဖိုး n ခု"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[:1])       # ပထမ 1 ခု -> ['apple']
print(fruits[:2])       # ပထမ 2 ခု -> ['apple', 'banana']
print(fruits[:4])       # ပထမ 4 ခု -> ['apple', 'banana', 'cherry', 'grape']
````

````python
# [n:] = "ပထမ တန်ဖိုး n ခုကို ကျော်ပြီး ကျန်တာကို ယူပါ"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[1:])       # 1 ခု ကျော် -> ['banana', 'cherry', 'grape', 'melon']
print(fruits[3:])       # 3 ခု ကျော် -> ['grape', 'melon']
````

````python
# list ရှည်လာရင်တောင် [n:] က အလုပ်လုပ်ပါတယ်
# append() က list ရဲ့ နောက်ဆုံးမှာ တန်ဖိုး ထည့်ပါတယ်
scores = [85, 92, 78]
scores.append(90)           # အခု: [85, 92, 78, 90]
scores.append(66)           # အခု: [85, 92, 78, 90, 66]

print(scores[1:])           # အခု တန်ဖိုး 5 ခု ရှိတယ်ဆိုတာ သိစရာ မလိုခဲ့ပါဘူး
````

````python
# [:] = list တစ်ခုလုံး (သူ့ရဲ့ မိတ္တူ)
scores = [85, 92, 78, 90, 66]
print(scores[:])
````

````python
# list ကို အပိုင်း နှစ်ပိုင်း ခွဲခြင်း
# [:2] နဲ့ [2:] က ဂဏန်း အတူတူ သုံးလို့ ပေါင်းလိုက်ရင် list တစ်ခုလုံး ဖြစ်ပါတယ်
# (2 က [:2] ထဲမှာ "မပါ" ပေမဲ့ [2:] ထဲမှာ "ပါ" လို့ပါ)
scores = [85, 92, 78, 90, 66]

front = scores[:2]
back = scores[2:]

print(front)            # [85, 92]
print(back)             # [78, 90, 66]
````

## 4. အနုတ် index နဲ့ slicing

**အနုတ် index** ကို ပြန်သတိရကြရအောင်။ Python က list ရဲ့ **နောက်ဆုံးကနေ နောက်ပြန်** လည်း ရေတွက်နိုင်ပါတယ်:

- `-1` က **နောက်ဆုံး** တန်ဖိုး
- `-2` က နောက်ဆုံးကနေ ဒုတိယ တန်ဖိုး
- `-3` က နောက်ဆုံးကနေ တတိယ တန်ဖိုး၊ စသဖြင့်

| တန်ဖိုး | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| index | 0 | 1 | 2 | 3 | 4 |
| အနုတ် index | -5 | -4 | -3 | -2 | -1 |

အနုတ် index ကို သုံးတဲ့ အဓိက အကြောင်းရင်း: **တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ မရေတွက်ဘဲ list ရဲ့ အဆုံးကို ရောက်နိုင်ပါတယ်**။ list မှာ တန်ဖိုး 5 ခုပဲ ရှိရှိ 500 ခုပဲ ရှိရှိ `-1` က အမြဲ နောက်ဆုံး တန်ဖိုး ဖြစ်ပါတယ်။

အနုတ် index က slicing မှာလည်း အလုပ်လုပ်ပါတယ်။ စည်းမျဉ်းက အတူတူပါပဲ: **start ပါပြီး end မပါပါဘူး**။

| ရေးပုံ | အဓိပ္ပာယ် |
|---|---|
| `scores[-3:]` | **နောက်ဆုံး တန်ဖိုး 3 ခု** |
| `scores[-2:]` | **နောက်ဆုံး တန်ဖိုး 2 ခု** |
| `scores[:-1]` | **နောက်ဆုံး တန်ဖိုး ကလွဲပြီး** အားလုံး |
| `scores[:-2]` | **နောက်ဆုံး တန်ဖိုး 2 ခု ကလွဲပြီး** အားလုံး |

````python
scores = [85, 92, 78, 90, 66]
#  အနုတ် index: -5  -4  -3  -2  -1

# [-n:] = "နောက်ဆုံး တန်ဖိုး n ခု"
print(scores[-3:])      # -3 ကနေ အဆုံးအထိ -> [78, 90, 66]
print(scores[-2:])      # -2 ကနေ အဆုံးအထိ -> [90, 66]
print(scores[-1:])      # -1 ကနေ အဆုံးအထိ -> [66] (တန်ဖိုး တစ်ခုတည်း ပါတဲ့ list)
````

````python
# နှိုင်းယှဉ်ကြည့်ပါ: index က တန်ဖိုး တစ်ခု၊ slice က list ကို ပေးပါတယ်
scores = [85, 92, 78, 90, 66]

print(scores[-1])       # 66   -> တန်ဖိုး တစ်ခု (နောက်ဆုံး တစ်ခု)
print(scores[-1:])      # [66] -> နောက်ဆုံး တန်ဖိုး ပါတဲ့ list
````

````python
scores = [85, 92, 78, 90, 66]
#  အနုတ် index: -5  -4  -3  -2  -1

# [:-n] = "နောက်ဆုံး တန်ဖိုး n ခု ကလွဲပြီး အားလုံး"
print(scores[:-1])      # -1 အထိ၊ ဒါပေမဲ့ -1 "မပါ" -> [85, 92, 78, 90]
print(scores[:-2])      # -2 အထိ၊ ဒါပေမဲ့ -2 "မပါ" -> [85, 92, 78]
````

````python
# အရှည် ဘယ်လောက်ပဲ ရှိရှိ နောက်ဆုံး တန်ဖိုး 3 ခု
# တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ သိစရာ မလိုပါဘူး
short_list = [1, 2, 3, 4]
long_list = [10, 20, 30, 40, 50, 60, 70, 80]

print(short_list[-3:])  # [2, 3, 4]
print(long_list[-3:])   # [60, 70, 80]
````

## 5. slice ကို for loop နဲ့ သုံးခြင်း

slice က list ဖြစ်ပါတယ်။ ဒါကြောင့် list နဲ့ လုပ်လို့ရတာ အားလုံးကို slice နဲ့လည်း လုပ်လို့ရပါတယ်။ **for loop** လည်း ပါပါတယ်။

for loop ဘယ်လို အလုပ်လုပ်လဲ ပြန်သတိရကြရအောင်:

```python
for score in scores:
    print(score)
```

- `for score in scores:` က **"scores ထဲက score တစ်ခုစီအတွက် အောက်က code ကို run ပါ"** လို့ အဓိပ္ပာယ်ရပါတယ်။
- အကြိမ်တိုင်း list ထဲက တန်ဖိုး တစ်ခုကို ပထမကနေ နောက်ဆုံးအထိ `score` variable ထဲ သိမ်းပါတယ်။
- စာကြောင်းက colon `:` နဲ့ ဆုံးရပါမယ်။
- ထပ်ခါထပ်ခါ run မယ့် code ကို **indent** (အစမှာ space 4 ခု) လုပ်ရပါမယ်။ ဘယ်စာကြောင်းတွေက loop ထဲမှာ ပါလဲဆိုတာ Python က indent နဲ့ သိပါတယ်။ indent လုပ်ထားတဲ့ ဒီစာကြောင်း အုပ်စုကို **block** လို့ ခေါ်ပါတယ်။
- indent **မလုပ်ထားတဲ့** ပထမ စာကြောင်းက loop အပြင်မှာ ရှိပြီး loop ပြီးမှ တစ်ကြိမ်ပဲ run ပါတယ်။

အခု `in` နောက်မှာ **slice** ကို ထည့်လိုက်ရင် loop က list ရဲ့ **အဲဒီ အပိုင်းကိုပဲ** ဖြတ်သွားပါတယ်။

```python
for score in scores[:3]:      # ပထမ တန်ဖိုး 3 ခုပဲ
    print(score)
```

````python
# list တစ်ခုလုံး
scores = [85, 92, 78, 90, 66]

for score in scores:            # တန်ဖိုး 5 ခု -> loop က 5 ကြိမ် run ပါတယ်
    print(score)
````

````python
# ပထမ တန်ဖိုး 3 ခုပဲ
scores = [85, 92, 78, 90, 66]

for score in scores[:3]:        # scores[:3] က [85, 92, 78] -> loop က 3 ကြိမ် run ပါတယ်
    print(score)                # indent (space 4 ခု) လုပ်ထား -> loop ထဲမှာ -> ထပ်ခါထပ်ခါ run ပါတယ်
````

### loop variable ကို ကြိုက်တဲ့ နာမည် ပေးလို့ရပါတယ်

မှတ်ထားပါ: `for` နောက်က variable (ဒီမှာ `score`) က loop ဖန်တီးတဲ့ **variable အသစ်** ပါ။ ကြိုက်တဲ့ နာမည် ပေးလို့ရပါတယ်။ ဒါပေမဲ့ list နာမည်ရဲ့ **ဧကဝုစ်** ပုံစံကို သုံးတာက အလေ့အထ ကောင်းပါ: `scores` → `score`, `fruits` → `fruit`။ ဒါဆိုရင် "list တစ်ခုလုံး" နဲ့ "တန်ဖိုး တစ်ခု" ကို ခွဲခြားရ လွယ်ပါတယ်။

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[1:4]:       # fruits[1:4] က ['banana', 'cherry', 'grape']
    print(fruit)                # fruit (တန်ဖိုး တစ်ခု), fruits (list တစ်ခုလုံး)
````

### ဘာကို ထပ်ခါထပ်ခါ run မလဲ ဆိုတာ indent က ဆုံးဖြတ်ပါတယ်

သင်ခန်းစာ 5 လိုပဲ indent လုပ်ထားတဲ့ စာကြောင်းတွေက ထပ်ခါထပ်ခါ run ပြီး indent မလုပ်ထားတဲ့ ပထမ စာကြောင်းက loop ပြီးမှ တစ်ကြိမ်ပဲ run ပါတယ်။

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[:2]:        # 'apple' နဲ့ 'banana' ပဲ
    print("I like", fruit)      # indent လုပ်ထား -> 2 ကြိမ် run ပါတယ်
print("That's all!")            # indent မလုပ်ထား -> loop အပြင်မှာ -> တစ်ကြိမ် run ပါတယ်
````

**`print("I like", fruit)` အကြောင်း:** `print()` ထဲမှာ တန်ဖိုး အများကြီးကို **comma** `,` နဲ့ ခွဲထည့်ရင် စာကြောင်း တစ်ကြောင်းတည်းမှာ ပြပြီး Python က သူတို့ကြားမှာ **space တစ်ခု** ကို အလိုအလျောက် ထည့်ပေးပါတယ်။ quote ထဲက စာသားကို ရေးထားတဲ့အတိုင်း ပြပြီး variable နေရာမှာ အထဲက တန်ဖိုးကို ပြပါတယ်။

## 6. slice နဲ့ တွက်ချက်ခြင်း

### list ရဲ့ အစိတ်အပိုင်း တစ်ခုကို ပေါင်းခြင်း

သင်ခန်းစာ 5 မှာ **ပေါင်းစုတန်ဖိုး** ကို သုံးပြီး list ထဲက တန်ဖိုးအားလုံးကို ပေါင်းခဲ့ပါတယ်:

1. **loop မစခင်:** variable တစ်ခု ဖန်တီးပြီး 0 ထည့်ပါ → `total = 0`
2. **loop ထဲမှာ:** လက်ရှိ တန်ဖိုးကို ပေါင်းထည့်ပါ → `total = total + score`
3. **loop ပြီးရင်:** variable ထဲမှာ ပေါင်းလဒ် ရှိပါတယ်

`total = total + score` ရဲ့ အဓိပ္ပာယ်ကို သတိရပါ: `=` က "သိမ်းပါ" လို့ အဓိပ္ပာယ်ရပါတယ်။ Python က ညာဘက် (စုစုပေါင်းဟောင်း + လက်ရှိ အမှတ်) ကို အရင်တွက်ပြီး ရလဒ်ကို `total` ထဲ ပြန်သိမ်းပါတယ်။

slice နဲ့ဆိုရင် list ရဲ့ **အစိတ်အပိုင်း တစ်ခုကိုပဲ** ပေါင်းလို့ရပါတယ်။ `in` နောက်မှာ slice ကို ထည့်ရုံပါပဲ။

````python
# ပထမ အမှတ် 3 ခုကိုပဲ ပေါင်းခြင်း
scores = [85, 92, 78, 90, 66]
total = 0                           # loop မစခင်: 0 ကနေ စပါတယ်

for score in scores[:3]:            # 85, 92, 78 ပဲ
    total = total + score           # စုစုပေါင်းဟောင်း + လက်ရှိ အမှတ် -> total ထဲ ပြန်သိမ်းပါတယ်

print("Total of the first 3:", total)   # loop ပြီးမှ (indent မလုပ်ထား): 85 + 92 + 78 = 255
````

````python
# total ဘယ်လို ပြောင်းလဲလဲ ကြည့်ရအောင်
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total = total + score
    print("score:", score, "total:", total)     # indent လုပ်ထား -> အကြိမ်တိုင်း ပြပါတယ်
````

**အတိုပုံစံ `+=` ကို သတိရပါ:** `total += score` က `total = total + score` နဲ့ အဓိပ္ပာယ် အတိအကျ တူပါတယ်။ "ညာဘက်ကို variable ထဲ ပေါင်းပြီး ရလဒ်ကို အဲဒီ variable ထဲ ပြန်သိမ်းပါ"။

````python
# အတူတူကို += နဲ့ ရေးခြင်း
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total += score                  # အဓိပ္ပာယ်တူ: total = total + score

print("Total of the first 3:", total)
````

````python
# နောက်ဆုံး အမှတ် 2 ခုကိုပဲ ပေါင်းခြင်း
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[-2:]:           # 90, 66 ပဲ
    total += score                  # အဓိပ္ပာယ်တူ: total = total + score

print("Total of the last 2:", total)    # 90 + 66 = 156
````

### အစိတ်အပိုင်း တစ်ခုရဲ့ ပျမ်းမျှ

**ပျမ်းမျှ** ဆိုတာ စုစုပေါင်းကို တန်ဖိုး အရေအတွက်နဲ့ စားထားတာပါ: **ပျမ်းမျှ = စုစုပေါင်း / တန်ဖိုး အရေအတွက်**

တန်ဖိုး အရေအတွက်ကို `len()` နဲ့ ရပါတယ်။ ဒါပေမဲ့ သတိထားပါ: list တစ်ခုလုံးရဲ့ မဟုတ်ဘဲ **slice ထဲက** တန်ဖိုး အရေအတွက် လိုပါတယ်။ ဒါကြောင့် slice ကို `len()` သုံးပါ။

ပြီးတော့ မှတ်ထားပါ: စုစုပေါင်း ပြည့်စုံသွားတဲ့အခါ loop **ပြီးမှ** (indent မလုပ်ဘဲ) ပျမ်းမျှကို တွက်ပါ။

````python
# ပထမ အမှတ် 3 ခုရဲ့ ပျမ်းမျှ
scores = [85, 92, 78, 90, 66]
first_three = scores[:3]            # slice ကို variable ထဲ သိမ်းပါတယ်: [85, 92, 78]
total = 0

for score in first_three:
    total += score                  # အဓိပ္ပာယ်တူ: total = total + score

average = total / len(first_three)  # 255 / 3 (list တစ်ခုလုံးရဲ့ မဟုတ်ဘဲ "slice" ရဲ့ အရှည်)
print("Total:", total)
print("Count:", len(first_three))
print("Average:", average)
````

````python
# ပထမ တစ်ဝက် နဲ့ ဒုတိယ တစ်ဝက်ရဲ့ ပျမ်းမျှကို နှိုင်းယှဉ်ခြင်း
scores = [60, 65, 70, 85, 90, 95]
first_half = scores[:3]             # [60, 65, 70]
second_half = scores[3:]            # [85, 90, 95]

total = 0
for score in first_half:
    total += score
print("First half average:", total / len(first_half))

total = 0                           # ဒုတိယ တစ်ဝက်အတွက် 0 ကနေ ပြန်စပါ!
for score in second_half:
    total += score
print("Second half average:", total / len(second_half))
````

### အစိတ်အပိုင်း တစ်ခုထဲက အကြီးဆုံး တန်ဖိုး

သင်ခန်းစာ 5 မှာ လူတစ်ယောက်က စက္ကူပေါ်မှာ ရှာသလို အကြီးဆုံး တန်ဖိုးကို ရှာခဲ့ပါတယ်:

1. **ပထမ တန်ဖိုး** ကို "အခုထိ အကြီးဆုံး" လို့ မှတ်ထားပါ။
2. တန်ဖိုး တစ်ခုစီကို ကြည့်ပါ။ မှတ်ထားတာထက် ကြီးရင် အဲဒါကို အစားထိုး မှတ်ထားပါ။
3. နောက်ဆုံး မှတ်ထားတဲ့ တန်ဖိုးက အကြီးဆုံးပါ။

slice နဲ့ဆိုရင် list ရဲ့ **အစိတ်အပိုင်း တစ်ခုထဲမှာပဲ** အကြီးဆုံး တန်ဖိုးကို ရှာလို့ရပါတယ်။

````python
# နောက်ဆုံး ကျောင်းသား 3 ယောက်ထဲက အမြင့်ဆုံး အမှတ်
scores = [95, 82, 74, 88, 79]
last_three = scores[-3:]            # [74, 88, 79]

highest = last_three[0]             # အဆင့် 1: "slice" ရဲ့ ပထမ တန်ဖိုး (74) ကနေ စပါတယ်

for score in last_three:
    if score > highest:             # အဆင့် 2: မှတ်ထားတာထက် ကြီးလား?
        highest = score             #         ကြီးရင် ဒါကို အစားထိုး မှတ်ထားပါတယ်

print("Highest of the last 3:", highest)    # အဆင့် 3: 88
````

## 7. slice ကို if, in, append() နဲ့ သုံးခြင်း

### loop ထဲက if

`for` loop ထဲမှာ `if` ကို ထည့်ရင် indent **နှစ်ဆင့်** ဖြစ်ပါတယ်:

```python
for score in scores[:3]:      # space မရှိ
    if score >= 80:           # space 4 ခု -> for ထဲမှာ (အမှတ်တိုင်းအတွက် run ပါတယ်)
        print(score)          # space 8 ခု -> if ထဲမှာ (condition True ဖြစ်မှသာ run ပါတယ်)
```

`if` ဘယ်လို အလုပ်လုပ်လဲ သတိရပါ: သူ့အောက်က indent လုပ်ထားတဲ့ code က **condition True ဖြစ်မှသာ** run ပါတယ်။

````python
# ပထမ အမှတ် 4 ခုထဲက 80 နဲ့အထက် ဖြစ်တာတွေကိုပဲ ပြခြင်း
scores = [85, 72, 91, 78, 95, 60]

for score in scores[:4]:            # 85, 72, 91, 78 ပဲ
    if score >= 80:                 # space 4 ခု: အမှတ် တစ်ခုစီကို စစ်ပါတယ်
        print(score)                # space 8 ခု: 80 နဲ့အထက် ဖြစ်မှသာ ပြပါတယ်
````

````python
# ရေတွက်ခြင်း: နောက်ဆုံး အမှတ် 3 ခုထဲက ဘယ်နှစ်ခု အောင်လဲ (60 နဲ့အထက်)?
# ရေတွက်ခြင်းက ပေါင်းစုတန်ဖိုးလိုပဲ အလုပ်လုပ်ပေမဲ့ တန်ဖိုးအစား 1 ကို ပေါင်းပါတယ်
scores = [85, 42, 78, 55, 90]
pass_count = 0                      # loop မစခင်: 0 ကနေ စပါတယ်

for score in scores[-3:]:           # 78, 55, 90 ပဲ
    if score >= 60:
        pass_count += 1             # အဓိပ္ပာယ်တူ: pass_count = pass_count + 1

print("Passed among the last 3:", pass_count)
````

### in

`တန်ဖိုး in list` က တန်ဖိုးဟာ list ထဲမှာ ရှိမရှိ စစ်ပြီး `True` ဒါမှမဟုတ် `False` ကို ပေးပါတယ်။ slice က list ဖြစ်လို့ `in` ကို slice နဲ့လည်း သုံးလို့ရပါတယ်။ ဒါဆိုရင် list ရဲ့ **အစိတ်အပိုင်း တစ်ခုကိုပဲ** စစ်ပါတယ်။

````python
# ထိပ်ဆုံး ကျောင်းသား 3 ယောက် (list က အဆင့်လိုက် စီပြီးသား: ပထမ၊ ဒုတိယ၊ ...)
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]

print("Tom" in ranking[:3])         # Tom က ထိပ်ဆုံး 3 ယောက်ထဲမှာ ပါလား? -> True
print("Kate" in ranking[:3])        # Kate က ထိပ်ဆုံး 3 ယောက်ထဲမှာ ပါလား? -> False
print("Kate" in ranking)            # Kate က list တစ်ခုလုံးထဲမှာ ပါလား? -> True
````

````python
# in ကို if ရဲ့ condition အဖြစ် သုံးခြင်း
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]
name = "Bill"

if name in ranking[:3]:             # Bill က ထိပ်ဆုံး 3 ယောက်ထဲမှာ ပါလား?
    print(name, "gets a medal!")
else:
    print(name, "doesn't get a medal.")
````

### append(): list အသစ် ဖန်တီးခြင်း

`append()` ကို သတိရပါ: `list.append(တန်ဖိုး)` က တန်ဖိုးကို list ရဲ့ **နောက်ဆုံး** မှာ ထည့်ပါတယ်။

ဗလာ list `[]` နဲ့ စပြီး loop ထဲမှာ slice ထဲက လိုချင်တဲ့ တန်ဖိုးတွေကိုပဲ `append()` လုပ်လို့ရပါတယ်။

````python
# ပထမ အမှတ် 4 ခုထဲက အောင်တဲ့ အမှတ်တွေပဲ ပါတဲ့ list အသစ် ဖန်တီးခြင်း
scores = [85, 42, 78, 55, 90, 30]
passed = []                         # loop မစခင်: ဗလာ list

for score in scores[:4]:            # 85, 42, 78, 55 ပဲ
    if score >= 60:
        passed.append(score)        # passed ရဲ့ နောက်ဆုံးမှာ ထည့်ပါတယ်

print(passed)                       # [85, 78]
````

````python
# နောက်ဆုံး စျေးနှုန်း 3 ခုကို 2 ဆ လုပ်ထားတဲ့ list အသစ် ဖန်တီးခြင်း
prices = [1000, 2500, 3000, 1500, 4000]
doubled = []

for price in prices[-3:]:           # 3000, 1500, 4000 ပဲ
    doubled.append(price * 2)       # price * 2 ကို အရင်တွက်ပြီးမှ ထည့်ပါတယ်

print(doubled)
````

## 8. မကြာခဏ ဖြစ်တတ်သော အမှားများ

ဒါတွေက slicing ကို လေ့လာတဲ့အခါ လူတိုင်းနီးပါး လုပ်မိကြတဲ့ အမှားတွေ ဖြစ်ပါတယ်။ တစ်ခုချင်းစီအတွက် ဒါတွေကို ကြည့်ပါမယ်:

### အမှား 1: end ပါတယ်လို့ ထင်ခြင်း

````python
# ❌ မှားပါတယ်: index 1, 2, 3 က တန်ဖိုးတွေကို လိုချင်တာပါ
scores = [85, 92, 78, 90, 66]
#  index:   0   1   2   3   4

print(scores[1:3])      # [92, 78] -> တန်ဖိုး 2 ခုပဲ! 90 (index 3) ပျောက်နေပါတယ်
````

🔍 **ဘာမှားသွားလဲ?**

- `scores[1:3]` က "index 1 ကနေ index 3 မတိုင်ခင်အထိ (3 **မပါ**)" လို့ အဓိပ္ပာယ်ရပါတယ်။
- ဒါကြောင့် index 1 နဲ့ 2 ပဲ ရပါတယ်။ index 3 က slice **ရပ်တဲ့** နေရာပါ။
- error မရှိလို့ လွတ်သွားဖို့ လွယ်ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** end က လိုချင်တဲ့ နောက်ဆုံး index ထက် **1 ပို** ရပါမယ်။ index 3 ပါဖို့ `4` လို့ ရေးပါ။ (သင်ခန်းစာ 3 က `range()` နဲ့ အတူတူ: 3 ပါဖို့ `range(1, 4)` လို့ ရေးပါတယ်။)

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # index 1, 2, 3 -> [92, 78, 90]
````

### အမှား 2: colon အစား comma ကို သုံးခြင်း

````python
# ❌ မှားပါတယ်
scores = [85, 92, 78, 90, 66]

# print(scores[1, 4])     # TypeError: list indices must be integers or slices, not tuple
````

🔍 **ဘာမှားသွားလဲ?**

- slicing က ဂဏန်း နှစ်ခုကြားမှာ comma `,` မဟုတ်ဘဲ **colon** `:` လိုပါတယ်။
- comma နဲ့ဆိုရင် Python က slice လို့ နားမလည်ဘဲ error ပြပါတယ်။
- error message ရဲ့ အဓိပ္ပာယ်က: "`[ ]` ထဲမှာ ဂဏန်း တစ်ခု (index) ဒါမှမဟုတ် `:` ပါတဲ့ slice ကို ရေးရပါမယ်"။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** colon `:` ကို သုံးပါ။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # ဂဏန်းတွေကြားမှာ colon
````

### အမှား 3: square bracket အစား ကွင်း ကို သုံးခြင်း

````python
# ❌ မှားပါတယ်
scores = [85, 92, 78, 90, 66]

# print(scores(1:4))      # SyntaxError
````

🔍 **ဘာမှားသွားလဲ?**

- list ထဲက တန်ဖိုး ယူတဲ့အခါ (index နဲ့ပဲ ဖြစ်ဖြစ် slice နဲ့ပဲ ဖြစ်ဖြစ်) **square bracket** `[ ]` ကို အမြဲ သုံးပါတယ်။
- ကွင်း `( )` က `print()`, `len()`, `append()` လို command တွေအတွက်ပါ။
- Python က `scores(1:4)` ကို နားမလည်လို့ `SyntaxError` ("code ရဲ့ သဒ္ဒါ မှားနေပါတယ်") ကို ပြပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** square bracket `[ ]` ကို သုံးပါ။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # square bracket
````

### အမှား 4: slice နဲ့ index ကို ရောထွေးခြင်း

````python
# ❌ မှားပါတယ်: ပထမ အမှတ်ကို 5 ပေါင်းချင်တာပါ
scores = [85, 92, 78]

first = scores[0:1]     # ဒါက "list" ပါ: [85]၊ ဂဏန်း 85 မဟုတ်ပါဘူး
print(first)

# print(first + 5)      # TypeError: can only concatenate list (not "int") to list
````

🔍 **ဘာမှားသွားလဲ?**

- `scores[0:1]` က **slice** ဖြစ်လို့ ရလဒ်က တန်ဖိုး တစ်ခုတည်း ပါတဲ့ **list**: `[85]` ပါ။
- list နဲ့ ဂဏန်းကို `+` နဲ့ ပေါင်းလို့ မရလို့ Python က error ပြပါတယ်။
- error message ရဲ့ အဓိပ္ပာယ်က: "list ကို list နဲ့ပဲ ဆက်လို့ရပါတယ်၊ ဂဏန်း (int) နဲ့ မရပါ"။
- output ထဲက square bracket ကို ကြည့်ပါ: `[85]`။ bracket ကို မြင်ရင် list ပါ။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** **တန်ဖိုး တစ်ခု** ယူဖို့ **index** (colon မပါတဲ့ ဂဏန်း တစ်ခု) ကို သုံးပါ: `scores[0]`။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78]

first = scores[0]       # ဒါက ဂဏန်း 85 ပါ (bracket မပါ)
print(first)
print(first + 5)        # 90
````

### အမှား 5: slice က မူလ list ကို ပြောင်းတယ်လို့ ထင်ခြင်း

````python
# ❌ မှားပါတယ်: ပထမ အမှတ် 3 ခုကိုပဲ ထားချင်တာပါ
scores = [85, 92, 78, 90, 66]

scores[:3]              # ဒါက list အသစ် ဖန်တီးပါတယ်... ဒါပေမဲ့ ဘယ်မှာမှ မသိမ်းထားပါဘူး!
print(scores)           # [85, 92, 78, 90, 66] -> ဘာမှ မပြောင်းပါဘူး
````

🔍 **ဘာမှားသွားလဲ?**

- slicing က **list အသစ်** ကို ဖန်တီးပါတယ်။ မူလ list ကို **မပြောင်းပါဘူး**။
- `scores[:3]` စာကြောင်းမှာ Python က `[85, 92, 78]` ဆိုတဲ့ list အသစ်ကို ဖန်တီးခဲ့ပေမဲ့ ဘယ် variable ထဲမှ မသိမ်းခဲ့လို့ ပျောက်သွားပါတယ်။
- error မရှိလို့ သတိထားမိဖို့ ခက်ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** slice ကို `=` နဲ့ variable ထဲ သိမ်းပါ။ မူလ list ကို အစားထိုးချင်ရင် variable အတူတူထဲကို ပြန်သိမ်းပါ (`age = age + 1` လိုပဲ: တန်ဖိုးအသစ်က တန်ဖိုးဟောင်းကို အစားထိုးပါတယ်)။

````python
# ✅ မှန်ပါတယ် (1): variable အသစ်ထဲ သိမ်းခြင်း
scores = [85, 92, 78, 90, 66]

first_three = scores[:3]
print(first_three)      # [85, 92, 78]
print(scores)           # မူလ list က ပြည့်စုံဆဲ
````

````python
# ✅ မှန်ပါတယ် (2): variable အတူတူထဲကို ပြန်သိမ်းခြင်း
scores = [85, 92, 78, 90, 66]

scores = scores[:3]     # ညာဘက်ကို အရင်တွက် ([85, 92, 78])၊ ပြီးမှ scores ထဲ သိမ်းပါတယ်
print(scores)           # [85, 92, 78] -> list ဟောင်းကို အစားထိုးလိုက်ပါပြီ
````

### အမှား 6: start ကို end နောက်မှာ ထားခြင်း

````python
# ❌ မှားပါတယ်: index 1 ကနေ index 3 အထိ တန်ဖိုးတွေကို လိုချင်တာပါ
scores = [85, 92, 78, 90, 66]

print(scores[4:1])      # [] -> ဗလာ list!
````

🔍 **ဘာမှားသွားလဲ?**

- slice က အမြဲ **ဘယ်ကနေ ညာ** ကို သွားပါတယ်: start ကနေ end ကို။
- `scores[4:1]` က index 4 ကနေ စပြီး index 1 ကို သွားဖို့ ကြိုးစားပါတယ်။ ဒါပေမဲ့ 1 က 4 ရဲ့ **ဘယ်ဘက်** မှာ ရှိလို့ ကြားမှာ ဘာမှ မရှိပါဘူး။
- ရလဒ်က ဗလာ list `[]` ပါ။ error မရှိပါဘူး။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** **ငယ်တဲ့** index ကို အရင် (start)၊ **ကြီးတဲ့** index ကို နောက်မှ (end) ရေးပါ။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # index 1 ကနေ index 3 အထိ -> [92, 78, 90]
````

### အမှား 7: နောက်ဆုံး တန်ဖိုး 3 ခုကို ယူဖို့ [-3:-1] ကို သုံးခြင်း

````python
# ❌ မှားပါတယ်: နောက်ဆုံး တန်ဖိုး 3 ခုကို လိုချင်တာပါ
scores = [85, 92, 78, 90, 66]
#  အနုတ် index: -5  -4  -3  -2  -1

print(scores[-3:-1])    # [78, 90] -> တန်ဖိုး 2 ခုပဲ! နောက်ဆုံး တန်ဖိုး (66) ပျောက်နေပါတယ်
````

🔍 **ဘာမှားသွားလဲ?**

- end က **မပါပါဘူး**။ အနုတ် index နဲ့လည်း ဒီစည်းမျဉ်း မပြောင်းပါဘူး။
- `scores[-3:-1]` က "-3 ကနေ -1 မတိုင်ခင်အထိ (-1 **မပါ**)" လို့ အဓိပ္ပာယ်ရပါတယ်။ ဒါကြောင့် နောက်ဆုံး တန်ဖိုး (-1) ကျန်ခဲ့ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** နောက်ဆုံး တန်ဖိုး ပါဖို့ **end ကို ချန်ထားပါ**: `scores[-3:]`။ end မရှိရင် "အဆုံးအထိ" လို့ အဓိပ္ပာယ်ရပါတယ်။

````python
# ✅ မှန်ပါတယ်
scores = [85, 92, 78, 90, 66]

print(scores[-3:])      # -3 ကနေ အဆုံးအထိ -> [78, 90, 66]
````

### အမှား 8: slice ရဲ့ ပျမ်းမျှအတွက် list တစ်ခုလုံးရဲ့ အရှည်ကို သုံးခြင်း

````python
# ❌ မှားပါတယ်: ပထမ အမှတ် 3 ခုရဲ့ ပျမ်းမျှကို လိုချင်တာပါ
scores = [80, 90, 100, 50, 60]
total = 0

for score in scores[:3]:
    total += score              # 80 + 90 + 100 = 270

average = total / len(scores)   # 270 / 5 -> မှားပါတယ်! အမှတ် 3 ခုပဲ ပေါင်းခဲ့တာပါ
print("Average:", average)      # 54.0
````

🔍 **ဘာမှားသွားလဲ?**

- ပထမ အမှတ် **3** ခု (270) ကိုပဲ ပေါင်းခဲ့ပါတယ်။
- ဒါပေမဲ့ `len(scores)` က list တစ်ခုလုံးထဲက တန်ဖိုး **5 ခုလုံး** ကို ရေတွက်ပါတယ်။
- ဒါကြောင့် 3 အစား 5 နဲ့ စားခဲ့ပါတယ်။ ပျမ်းမျှက အရမ်း နည်းသွားပါတယ်။
- error မရှိပါဘူး။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** **slice ထဲက** တန်ဖိုး အရေအတွက်နဲ့ စားပါ။ အလွယ်ဆုံးနည်း: slice ကို variable ထဲ အရင်သိမ်းပြီး အဲဒီ variable ကို `len()` သုံးပါ။

````python
# ✅ မှန်ပါတယ်
scores = [80, 90, 100, 50, 60]
first_three = scores[:3]        # [80, 90, 100]
total = 0

for score in first_three:
    total += score

average = total / len(first_three)  # 270 / 3
print("Average:", average)          # 90.0
````

### အမှား 9: slice က list အဆုံးကို ကျော်ရင် error ဖြစ်မယ်လို့ ထင်ခြင်း

````python
# ❓ အံ့သြစရာ: ဒါက error "မဖြစ်ပါဘူး"
scores = [85, 92, 78]           # တန်ဖိုး 3 ခုပဲ (index 0, 1, 2)

print(scores[0:10])             # [85, 92, 78] -> error မရှိပါဘူး!

# ဒါပေမဲ့ အရမ်းကြီးတဲ့ index က error "ဖြစ်ပါတယ်"
# print(scores[10])             # IndexError: list index out of range
````

🔍 **ဘာဖြစ်နေတာလဲ?**

- **index** နဲ့ဆိုရင် Python က အဲဒီ နေရာ အတိအကျမှာ တန်ဖိုး **တစ်ခု** ကို ရှာရပါတယ်။ အဲဒီ နေရာ မရှိရင် `IndexError` ("index က list အပြင်ကို ရောက်နေပါတယ်") ကို ပြပါတယ်။
- **slice** နဲ့ဆိုရင် Python က အဲဒီ အပိုင်းအခြားထဲမှာ **ရှာတွေ့သလောက် တန်ဖိုးတွေကိုပဲ** ပေးပါတယ်။ အပိုင်းအခြားက အဆုံးကို ကျော်သွားရင် အဆုံးမှာ ရပ်ပါတယ်။ error မရှိပါဘူး။
- ဒါက အမှား အတိအကျ မဟုတ်ပေမဲ့ ရှုပ်ထွေးစေနိုင်ပါတယ်: မှားနေသလို ထင်ရတဲ့ slice က "အလုပ်လုပ်" နေနိုင်ပြီး ဂဏန်းတွေ မှားရေးထားတာကို သတိမထားမိနိုင်ပါဘူး။

✅ **အကြံပြုချက်:** အဆုံးအထိ အားလုံးကို လိုချင်ရင် ဂဏန်းကြီးကြီး ခန့်မှန်းမနေပါနဲ့။ end ကို ချန်ထားလိုက်ရုံပါပဲ: `scores[0:]` ဒါမှမဟုတ် `scores[:]`။

````python
# ✅ ပိုကောင်းပါတယ်
scores = [85, 92, 78]

print(scores[0:])               # index 0 ကနေ အဆုံးအထိ -> [85, 92, 78]
````
