## 1. Slicing là gì?

Hãy bắt đầu từ những gì chúng ta đã biết.

**List** chứa nhiều giá trị theo thứ tự, bên trong dấu ngoặc vuông `[ ]`:

```python
scores = [85, 92, 78, 90, 66]
```

Mỗi giá trị có một số chỉ vị trí gọi là **chỉ số (index)**. Quy tắc quan trọng nhất: **Python bắt đầu đếm từ 0, không phải từ 1.**

| Giá trị | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Chỉ số | 0 | 1 | 2 | 3 | 4 |

Để lấy **một** giá trị, chúng ta viết tên list và chỉ số trong dấu ngoặc vuông: `scores[0]` là `85`, `scores[2]` là `78`.

Nhưng nếu muốn lấy **nhiều giá trị cùng lúc** thì sao? Ví dụ, chỉ lấy điểm của 3 học sinh đầu tiên?

Với những gì đã biết, chúng ta phải lấy từng cái một: `scores[0]`, `scores[1]`, `scores[2]`.

**Slicing** cho phép lấy **một phần** của list chỉ trong một bước. Từ "slice" có nghĩa là "một miếng được cắt ra", giống như một lát bánh mì hay một miếng pizza. Chúng ta cắt một miếng ra khỏi list.

Kết quả của slicing là một **list mới** chỉ chứa phần đó.

````python
scores = [85, 92, 78, 90, 66]

# Lấy MỘT giá trị bằng chỉ số (điều chúng ta đã biết)
print(scores[0])        # chỉ số 0 -> giá trị thứ 1 -> 85
print(scores[1])        # chỉ số 1 -> giá trị thứ 2 -> 92
print(scores[2])        # chỉ số 2 -> giá trị thứ 3 -> 78
````

````python
scores = [85, 92, 78, 90, 66]

# Lấy NHIỀU giá trị cùng lúc bằng slicing (điều chúng ta học hôm nay)
# scores[0:3] nghĩa là "từ chỉ số 0 đến trước chỉ số 3 (không bao gồm 3)"
print(scores[0:3])      # [85, 92, 78] -> một list mới gồm 3 giá trị đầu tiên
````

Hãy nhìn kỹ hai kết quả ở trên:

- `scores[0]` cho ra **một giá trị**: `85` (không có dấu ngoặc vuông)
- `scores[0:3]` cho ra **một list**: `[85, 92, 78]` (có dấu ngoặc vuông)

Sự khác biệt này rất quan trọng. Chỉ số cho ra một giá trị. Slice cho ra một list.

## 2. Dạng cơ bản: `list[start:end]`

```python
tên_list[start:end]
```

- `start`: chỉ số nơi slice **bắt đầu** (giá trị này **được** bao gồm)
- `end`: chỉ số nơi slice **dừng lại** (giá trị này **KHÔNG** được bao gồm)
- `:` dấu hai chấm nằm giữa hai số

Bạn có thể đọc `scores[1:4]` là: **"từ chỉ số 1, đến trước chỉ số 4 (không bao gồm 4)"**.

### End không được bao gồm

Đây là phần làm người mới học bối rối nhất. Hãy xem chậm rãi bằng một bảng.

```python
scores = [85, 92, 78, 90, 66]
```

| Chỉ số | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Giá trị | 85 | 92 | 78 | 90 | 66 |
| Có trong `scores[1:4]` không? | ❌ | ✅ | ✅ | ✅ | ❌ |

`scores[1:4]` → chỉ số 1, 2, 3 → `[92, 78, 90]`. Chỉ số 4 là nơi slice **dừng lại**, nên 66 không được bao gồm.

**Bạn đã thấy quy tắc này trước đây chưa?** Rồi! Ở Bài 3, `range(1, 4)` cho ra các số 1, 2, 3, và 4 **không** được bao gồm. Slicing dùng đúng quy tắc đó: **start được bao gồm, end thì không**.

### Lấy được bao nhiêu giá trị?

Vì end không được bao gồm, số lượng giá trị chỉ đơn giản là **end - start**.

- `scores[1:4]` → 4 - 1 = **3** giá trị
- `scores[0:2]` → 2 - 0 = **2** giá trị
- `scores[2:5]` → 5 - 2 = **3** giá trị

````python
scores = [85, 92, 78, 90, 66]
#  chỉ số:  0   1   2   3   4

print(scores[1:4])      # chỉ số 1, 2, 3 (4 không được bao gồm) -> [92, 78, 90]
````

````python
scores = [85, 92, 78, 90, 66]
#  chỉ số:  0   1   2   3   4

print(scores[0:2])      # chỉ số 0, 1 -> [85, 92]
print(scores[2:5])      # chỉ số 2, 3, 4 -> [78, 90, 66]
print(scores[3:4])      # chỉ có chỉ số 3 -> [90] (một list chỉ có một giá trị)
````

````python
# Cùng quy tắc với range() ở Bài 3: end không được bao gồm
# range(1, 4) cho ra 1, 2, 3
for i in range(1, 4):
    print(i)
````

````python
# Với chuỗi, cách làm hoàn toàn giống vậy
fruits = ["apple", "banana", "cherry", "grape", "melon"]
#  chỉ số:     0        1         2        3        4

print(fruits[0:2])      # chỉ số 0, 1 -> ['apple', 'banana']
print(fruits[1:3])      # chỉ số 1, 2 -> ['banana', 'cherry']
print(fruits[2:5])      # chỉ số 2, 3, 4 -> ['cherry', 'grape', 'melon']
````

### Đếm số lượng giá trị bằng len()

Bạn còn nhớ `len()` không? Nó cho biết list có **bao nhiêu giá trị**. `len([85, 92, 78])` là `3`.

Slice là một list, nên chúng ta cũng dùng `len()` với nó được. Hãy kiểm tra quy tắc "end - start".

````python
scores = [85, 92, 78, 90, 66]

print(len(scores))          # Cả list có 5 giá trị
print(len(scores[1:4]))     # 4 - 1 = 3 giá trị
print(len(scores[0:2]))     # 2 - 0 = 2 giá trị
````

### Slice là một list mới. List ban đầu không thay đổi.

Slicing **sao chép** một phần của list vào một **list mới**. List ban đầu vẫn giữ nguyên.

Nếu muốn dùng slice sau này, hãy lưu nó vào một biến, giống như mọi giá trị khác. Hãy nhớ: `=` có nghĩa là **"lưu giá trị bên phải vào chiếc hộp bên trái"**.

````python
scores = [85, 92, 78, 90, 66]

first_three = scores[0:3]   # Lưu list mới vào một biến tên first_three

print(first_three)          # [85, 92, 78] -> list mới
print(scores)               # [85, 92, 78, 90, 66] -> list ban đầu vẫn đầy đủ
````

````python
# List mới là một list thật, nên kiểu của nó là list
# type() cho biết một giá trị thuộc loại gì
scores = [85, 92, 78, 90, 66]
first_three = scores[0:3]

print(type(first_three))    # <class 'list'>
print(type(scores[0]))      # <class 'int'> -> chỉ số cho ra một giá trị (số nguyên), không phải list
````

````python
# Bạn cũng có thể dùng chỉ số với list mới
# Hãy nhớ: list mới lại bắt đầu từ chỉ số 0!
scores = [85, 92, 78, 90, 66]
last_two = scores[3:5]      # [90, 66]

print(last_two[0])          # Giá trị thứ 1 của list MỚI -> 90 (ở list ban đầu nó là chỉ số 3)
print(last_two[1])          # Giá trị thứ 2 của list MỚI -> 66
````

## 3. Bỏ trống start hoặc end

Không phải lúc nào cũng cần viết cả hai số. Python sẽ tự điền số còn thiếu cho bạn.

| Cách viết | Ý nghĩa | Python hiểu là |
|---|---|---|
| `scores[:3]` | từ **đầu** đến trước chỉ số 3 | `scores[0:3]` |
| `scores[2:]` | từ chỉ số 2 đến **cuối** | `scores[2:5]` (khi có 5 giá trị) |
| `scores[:]` | từ đầu đến cuối (cả list) | `scores[0:5]` |

- Không có start → bắt đầu từ **đầu** (chỉ số 0)
- Không có end → đi đến tận **cuối** (bao gồm cả giá trị cuối cùng)

**Tại sao điều này hữu ích?**
- `scores[:3]` ngắn hơn `scores[0:3]`, và dễ đọc: "3 cái đầu tiên".
- `scores[2:]` dùng được mà không cần biết list dài bao nhiêu. Bạn không cần đếm các giá trị.

````python
scores = [85, 92, 78, 90, 66]
#  chỉ số:  0   1   2   3   4

# Không có start -> từ đầu
print(scores[:3])       # Giống scores[0:3] -> [85, 92, 78]
print(scores[0:3])      # Kết quả giống nhau
````

````python
scores = [85, 92, 78, 90, 66]
#  chỉ số:  0   1   2   3   4

# Không có end -> đến cuối (giá trị cuối cùng CÓ được bao gồm)
print(scores[2:])       # chỉ số 2, 3, 4 -> [78, 90, 66]
print(scores[1:])       # Tất cả trừ cái đầu tiên -> [92, 78, 90, 66]
````

````python
# [:n] = "n giá trị đầu tiên"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[:1])       # 1 cái đầu -> ['apple']
print(fruits[:2])       # 2 cái đầu -> ['apple', 'banana']
print(fruits[:4])       # 4 cái đầu -> ['apple', 'banana', 'cherry', 'grape']
````

````python
# [n:] = "bỏ qua n giá trị đầu tiên, lấy phần còn lại"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[1:])       # Bỏ qua 1 -> ['banana', 'cherry', 'grape', 'melon']
print(fruits[3:])       # Bỏ qua 3 -> ['grape', 'melon']
````

````python
# [n:] vẫn dùng được kể cả khi list dài thêm
# append() thêm một giá trị vào cuối list
scores = [85, 92, 78]
scores.append(90)           # Bây giờ: [85, 92, 78, 90]
scores.append(66)           # Bây giờ: [85, 92, 78, 90, 66]

print(scores[1:])           # Chúng ta không cần biết bây giờ có 5 giá trị
````

````python
# [:] = cả list (một bản sao của nó)
scores = [85, 92, 78, 90, 66]
print(scores[:])
````

````python
# Cắt list thành hai phần
# [:2] và [2:] dùng cùng một số, nên ghép lại sẽ thành cả list
# (vì 2 KHÔNG nằm trong [:2], nhưng CÓ nằm trong [2:])
scores = [85, 92, 78, 90, 66]

front = scores[:2]
back = scores[2:]

print(front)            # [85, 92]
print(back)             # [78, 90, 66]
````

## 4. Slicing với chỉ số âm

Hãy nhớ lại **chỉ số âm**. Python còn có thể **đếm ngược từ cuối** list:

- `-1` là giá trị **cuối cùng**
- `-2` là giá trị kế cuối
- `-3` là giá trị thứ ba tính từ cuối, và cứ thế tiếp tục

| Giá trị | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Chỉ số | 0 | 1 | 2 | 3 | 4 |
| Chỉ số âm | -5 | -4 | -3 | -2 | -1 |

Lý do chính để dùng chỉ số âm: **chúng ta có thể đến cuối list mà không cần đếm xem có bao nhiêu giá trị**. `-1` luôn là giá trị cuối cùng, dù list có 5 giá trị hay 500 giá trị.

Chỉ số âm cũng dùng được trong slicing. Quy tắc vẫn vậy: **start được bao gồm, end thì không**.

| Cách viết | Ý nghĩa |
|---|---|
| `scores[-3:]` | **3 giá trị cuối** |
| `scores[-2:]` | **2 giá trị cuối** |
| `scores[:-1]` | tất cả **trừ giá trị cuối** |
| `scores[:-2]` | tất cả **trừ 2 giá trị cuối** |

````python
scores = [85, 92, 78, 90, 66]
#  chỉ số âm: -5  -4  -3  -2  -1

# [-n:] = "n giá trị cuối"
print(scores[-3:])      # Từ -3 đến cuối -> [78, 90, 66]
print(scores[-2:])      # Từ -2 đến cuối -> [90, 66]
print(scores[-1:])      # Từ -1 đến cuối -> [66] (một list có một giá trị)
````

````python
# So sánh: chỉ số cho ra một giá trị, slice cho ra một list
scores = [85, 92, 78, 90, 66]

print(scores[-1])       # 66   -> một giá trị (giá trị cuối cùng)
print(scores[-1:])      # [66] -> một list chứa giá trị cuối cùng
````

````python
scores = [85, 92, 78, 90, 66]
#  chỉ số âm: -5  -4  -3  -2  -1

# [:-n] = "tất cả trừ n giá trị cuối"
print(scores[:-1])      # Đến -1, nhưng -1 KHÔNG được bao gồm -> [85, 92, 78, 90]
print(scores[:-2])      # Đến -2, nhưng -2 KHÔNG được bao gồm -> [85, 92, 78]
````

````python
# 3 giá trị cuối, với list dài bao nhiêu cũng được
# Chúng ta không cần biết có bao nhiêu giá trị
short_list = [1, 2, 3, 4]
long_list = [10, 20, 30, 40, 50, 60, 70, 80]

print(short_list[-3:])  # [2, 3, 4]
print(long_list[-3:])   # [60, 70, 80]
````

````python
# Một ví dụ thực tế: 3 tin nhắn gần đây nhất
# Tin nhắn mới được thêm vào cuối bằng append(), nên tin mới nhất nằm ở cuối
messages = ["Hi", "How are you?", "Fine, thanks", "See you", "Bye"]

print(messages[-3:])    # 3 tin nhắn gần đây nhất
````

## 5. Dùng slice với vòng lặp for

Slice là một list. Vì vậy mọi thứ làm được với list thì cũng làm được với slice. Bao gồm cả **vòng lặp for**.

Hãy nhớ lại cách vòng lặp for hoạt động:

```python
for score in scores:
    print(score)
```

- `for score in scores:` có nghĩa là **"với mỗi score trong scores, hãy chạy đoạn mã bên dưới"**.
- Ở mỗi lần lặp, một giá trị từ list được lưu vào biến `score`, từ giá trị đầu tiên đến giá trị cuối cùng.
- Dòng này phải kết thúc bằng dấu hai chấm `:`.
- Đoạn mã cần lặp lại phải được **thụt lề** (4 dấu cách ở đầu). Python dùng thụt lề để biết những dòng nào thuộc về vòng lặp. Nhóm dòng được thụt lề này được gọi là **khối (block)**.
- Dòng đầu tiên **không** thụt lề nằm ngoài vòng lặp, và chỉ chạy một lần, sau khi vòng lặp kết thúc.

Bây giờ, nếu đặt một **slice** sau `in`, vòng lặp sẽ chỉ duyệt qua **phần đó** của list.

```python
for score in scores[:3]:      # chỉ 3 giá trị đầu tiên
    print(score)
```

````python
# Cả list
scores = [85, 92, 78, 90, 66]

for score in scores:            # 5 giá trị -> vòng lặp chạy 5 lần
    print(score)
````

````python
# Chỉ 3 giá trị đầu tiên
scores = [85, 92, 78, 90, 66]

for score in scores[:3]:        # scores[:3] là [85, 92, 78] -> vòng lặp chạy 3 lần
    print(score)                # Được thụt lề (4 dấu cách) -> bên trong vòng lặp -> được lặp lại
````

````python
# Chỉ 2 giá trị cuối
scores = [85, 92, 78, 90, 66]

for score in scores[-2:]:       # scores[-2:] là [90, 66] -> vòng lặp chạy 2 lần
    print(score)
````

````python
# Tất cả trừ cái đầu tiên
# Hữu ích khi giá trị đầu tiên là tiêu đề, không phải dữ liệu thật
row = ["Scores", 85, 92, 78]

for value in row[1:]:           # row[1:] là [85, 92, 78] -> bỏ qua "Scores"
    print(value)
````

### Biến lặp có thể mang bất kỳ tên nào

Hãy nhớ: biến đứng sau `for` (ở đây là `score`) là một **biến mới** do vòng lặp tạo ra. Bạn có thể đặt cho nó bất kỳ tên nào. Nhưng một thói quen tốt là dùng dạng **số ít** của tên list: `scores` → `score`, `fruits` → `fruit`. Như vậy rất dễ phân biệt "cả list" với "một giá trị".

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[1:4]:       # fruits[1:4] là ['banana', 'cherry', 'grape']
    print(fruit)                # fruit (một giá trị), fruits (cả list)
````

### Thụt lề quyết định cái gì được lặp lại

Giống như ở Bài 5, các dòng được thụt lề sẽ được lặp lại, và dòng đầu tiên không thụt lề chỉ chạy một lần sau vòng lặp.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[:2]:        # Chỉ 'apple' và 'banana'
    print("I like", fruit)      # Được thụt lề -> chạy 2 lần
print("That's all!")            # Không thụt lề -> bên ngoài vòng lặp -> chạy một lần
````

**Về `print("I like", fruit)`:** khi đặt nhiều giá trị vào `print()` và ngăn cách bằng **dấu phẩy** `,`, chúng được in trên một dòng, và Python tự động đặt **một dấu cách** giữa chúng. Chữ bên trong dấu ngoặc kép được in nguyên như vậy, còn biến thì được thay bằng giá trị bên trong nó.

## 6. Tính toán với slice

### Cộng dồn một phần của list

Ở Bài 5, chúng ta đã cộng tất cả các giá trị trong list bằng **tổng tích lũy**:

1. **Trước vòng lặp:** tạo một biến và gán bằng 0 → `total = 0`
2. **Bên trong vòng lặp:** cộng giá trị hiện tại vào → `total = total + score`
3. **Sau vòng lặp:** biến đó chứa tổng

Hãy nhớ `total = total + score` có nghĩa là gì: `=` có nghĩa là "lưu". Python tính vế phải trước (tổng cũ + điểm hiện tại), rồi lưu kết quả trở lại vào `total`.

Với slice, chúng ta có thể chỉ cộng **một phần** của list. Chỉ cần đặt slice sau `in`.

````python
# Chỉ cộng 3 điểm đầu tiên
scores = [85, 92, 78, 90, 66]
total = 0                           # Trước vòng lặp: bắt đầu từ 0

for score in scores[:3]:            # Chỉ 85, 92, 78
    total = total + score           # tổng cũ + điểm hiện tại -> lưu lại vào total

print("Total of the first 3:", total)   # Sau vòng lặp (không thụt lề): 85 + 92 + 78 = 255
````

````python
# Hãy xem total thay đổi thế nào
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total = total + score
    print("score:", score, "total:", total)     # Được thụt lề -> in ra mỗi lần lặp
````

**Hãy nhớ cách viết ngắn `+=`:** `total += score` có nghĩa hoàn toàn giống `total = total + score`. "Cộng vế phải vào biến, rồi lưu kết quả trở lại vào biến đó."

````python
# Cùng việc đó, viết bằng +=
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total += score                  # giống với: total = total + score

print("Total of the first 3:", total)
````

````python
# Chỉ cộng 2 điểm cuối
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[-2:]:           # Chỉ 90, 66
    total += score                  # giống với: total = total + score

print("Total of the last 2:", total)    # 90 + 66 = 156
````

### Trung bình của một phần

**Trung bình** là tổng chia cho số lượng giá trị: **trung bình = tổng / số lượng giá trị**

Chúng ta lấy số lượng giá trị bằng `len()`. Nhưng hãy cẩn thận: chúng ta cần số lượng giá trị **trong slice**, không phải trong cả list. Vì vậy hãy dùng `len()` với slice.

Cũng hãy nhớ: tính trung bình **sau** vòng lặp (không thụt lề), khi tổng đã hoàn chỉnh.

````python
# Trung bình của 3 điểm đầu tiên
scores = [85, 92, 78, 90, 66]
first_three = scores[:3]            # Lưu slice vào một biến: [85, 92, 78]
total = 0

for score in first_three:
    total += score                  # giống với: total = total + score

average = total / len(first_three)  # 255 / 3 (độ dài của SLICE, không phải cả list)
print("Total:", total)
print("Count:", len(first_three))
print("Average:", average)
````

````python
# So sánh trung bình của nửa đầu và nửa sau
scores = [60, 65, 70, 85, 90, 95]
first_half = scores[:3]             # [60, 65, 70]
second_half = scores[3:]            # [85, 90, 95]

total = 0
for score in first_half:
    total += score
print("First half average:", total / len(first_half))

total = 0                           # Bắt đầu lại từ 0 cho nửa sau!
for score in second_half:
    total += score
print("Second half average:", total / len(second_half))
````

### Giá trị lớn nhất trong một phần

Ở Bài 5, chúng ta tìm giá trị lớn nhất giống như cách một người làm trên giấy:

1. Ghi nhớ **giá trị đầu tiên** là "lớn nhất cho đến giờ".
2. Nhìn từng giá trị. Nếu nó lớn hơn giá trị đang nhớ, hãy nhớ giá trị này thay thế.
3. Giá trị bạn nhớ cuối cùng là giá trị lớn nhất.

Với slice, chúng ta có thể tìm giá trị lớn nhất **chỉ trong một phần** của list.

````python
# Điểm cao nhất trong 3 học sinh cuối
scores = [95, 82, 74, 88, 79]
last_three = scores[-3:]            # [74, 88, 79]

highest = last_three[0]             # Bước 1: bắt đầu bằng giá trị đầu tiên của SLICE (74)

for score in last_three:
    if score > highest:             # Bước 2: có lớn hơn giá trị đang nhớ không?
        highest = score             #         nếu có, hãy nhớ giá trị này thay thế

print("Highest of the last 3:", highest)    # Bước 3: 88
````

## 7. Dùng slice với if, in và append()

### if bên trong vòng lặp

Khi đặt `if` bên trong vòng lặp `for`, sẽ có **hai mức** thụt lề:

```python
for score in scores[:3]:      # không có dấu cách
    if score >= 80:           # 4 dấu cách -> bên trong for (chạy với mọi score)
        print(score)          # 8 dấu cách -> bên trong if (chỉ chạy khi điều kiện là True)
```

Hãy nhớ cách `if` hoạt động: đoạn mã thụt lề bên dưới nó chỉ chạy **khi điều kiện là True**.

````python
# Trong 4 điểm đầu tiên, chỉ in những điểm từ 80 trở lên
scores = [85, 72, 91, 78, 95, 60]

for score in scores[:4]:            # Chỉ 85, 72, 91, 78
    if score >= 80:                 # 4 dấu cách: kiểm tra từng điểm
        print(score)                # 8 dấu cách: chỉ in nếu từ 80 trở lên
````

````python
# Đếm: trong 3 điểm cuối có bao nhiêu điểm đậu (từ 60 trở lên)?
# Đếm hoạt động giống tổng tích lũy, nhưng chúng ta cộng 1 thay vì cộng giá trị
scores = [85, 42, 78, 55, 90]
pass_count = 0                      # Trước vòng lặp: bắt đầu từ 0

for score in scores[-3:]:           # Chỉ 78, 55, 90
    if score >= 60:
        pass_count += 1             # giống với: pass_count = pass_count + 1

print("Passed among the last 3:", pass_count)
````

### in

`giá_trị in list` kiểm tra xem giá trị có nằm trong list không, và cho ra `True` hoặc `False`. Slice là một list, nên chúng ta cũng dùng `in` với slice được. Khi đó chúng ta chỉ kiểm tra **một phần** của list.

````python
# Top 3 học sinh (list đã được sắp xếp theo thứ tự: hạng 1, hạng 2, ...)
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]

print("Tom" in ranking[:3])         # Tom có trong top 3 không? -> True
print("Kate" in ranking[:3])        # Kate có trong top 3 không? -> False
print("Kate" in ranking)            # Kate có trong cả list không? -> True
````

````python
# Dùng in làm điều kiện của if
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]
name = "Bill"

if name in ranking[:3]:             # Bill có trong top 3 không?
    print(name, "gets a medal!")
else:
    print(name, "doesn't get a medal.")
````

### append(): tạo list mới

Hãy nhớ `append()`: `list.append(giá_trị)` thêm giá trị vào **cuối** list.

Chúng ta có thể bắt đầu với một list rỗng `[]`, rồi bên trong vòng lặp, dùng `append()` để thêm chỉ những giá trị mình muốn từ một slice.

````python
# Từ 4 điểm đầu tiên, tạo một list mới chỉ gồm các điểm đậu
scores = [85, 42, 78, 55, 90, 30]
passed = []                         # Trước vòng lặp: một list rỗng

for score in scores[:4]:            # Chỉ 85, 42, 78, 55
    if score >= 60:
        passed.append(score)        # Thêm vào cuối passed

print(passed)                       # [85, 78]
````

````python
# Tạo một list mới với 3 giá cuối đều nhân đôi
prices = [1000, 2500, 3000, 1500, 4000]
doubled = []

for price in prices[-3:]:           # Chỉ 3000, 1500, 4000
    doubled.append(price * 2)       # price * 2 được tính trước, rồi mới được thêm vào

print(doubled)
````

## 8. Những lỗi thường gặp

Đây là những lỗi mà hầu như ai cũng mắc phải khi học slicing. Với mỗi lỗi, chúng ta sẽ xem:

- ❌ **Mã sai**
- 🔍 **Sai ở đâu**
- ✅ **Cách làm đúng**

Một số mã sai gây ra lỗi. Mã đó được viết dưới dạng chú thích (có `#` ở đầu) để không làm notebook dừng lại. Nếu muốn xem thông báo lỗi thật, hãy xóa `#`, chạy ô đó, rồi đặt `#` lại.

Một số mã sai khác **không báo lỗi, nhưng cho ra kết quả sai**. Những lỗi này nguy hiểm hơn, vì Python không cảnh báo bạn.

### Lỗi 1: Nghĩ rằng end được bao gồm

````python
# ❌ Sai: chúng ta muốn các giá trị ở chỉ số 1, 2, 3
scores = [85, 92, 78, 90, 66]
#  chỉ số:  0   1   2   3   4

print(scores[1:3])      # [92, 78] -> chỉ có 2 giá trị! Thiếu 90 (chỉ số 3)
````

🔍 **Sai ở đâu?**

- `scores[1:3]` có nghĩa là "từ chỉ số 1, đến trước chỉ số 3 (**không bao gồm** 3)".
- Vì vậy chúng ta chỉ nhận được chỉ số 1 và 2. Chỉ số 3 là nơi slice **dừng lại**.
- Không có lỗi nào, nên rất dễ bỏ sót.

✅ **Cách làm đúng:** End phải **lớn hơn 1** so với chỉ số cuối cùng bạn muốn. Để bao gồm chỉ số 3, hãy viết `4`. (Giống `range()` ở Bài 3: để bao gồm 3, bạn viết `range(1, 4)`.)

````python
# ✅ Đúng
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # chỉ số 1, 2, 3 -> [92, 78, 90]
````

### Lỗi 2: Dùng dấu phẩy thay cho dấu hai chấm

````python
# ❌ Sai
scores = [85, 92, 78, 90, 66]

# print(scores[1, 4])     # TypeError: list indices must be integers or slices, not tuple
````

🔍 **Sai ở đâu?**

- Slicing cần **dấu hai chấm** `:` giữa hai số, không phải dấu phẩy `,`.
- Với dấu phẩy, Python không hiểu đó là slice, và báo lỗi.
- Thông báo lỗi có nghĩa là: "bên trong `[ ]`, bạn phải viết một số (một chỉ số) hoặc một slice có `:`".

✅ **Cách làm đúng:** Dùng dấu hai chấm `:`.

````python
# ✅ Đúng
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # Dấu hai chấm giữa hai số
````

### Lỗi 3: Dùng dấu ngoặc tròn thay cho dấu ngoặc vuông

````python
# ❌ Sai
scores = [85, 92, 78, 90, 66]

# print(scores(1:4))      # SyntaxError
````

🔍 **Sai ở đâu?**

- Lấy giá trị từ list (bằng chỉ số hay slice) luôn dùng **dấu ngoặc vuông** `[ ]`.
- Dấu ngoặc tròn `( )` dùng cho các lệnh như `print()`, `len()` và `append()`.
- Python không hiểu `scores(1:4)`, nên báo lỗi `SyntaxError` ("ngữ pháp của mã bị sai").

✅ **Cách làm đúng:** Dùng dấu ngoặc vuông `[ ]`.

````python
# ✅ Đúng
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # Dấu ngoặc vuông
````

### Lỗi 4: Nhầm lẫn giữa slice và chỉ số

````python
# ❌ Sai: chúng ta muốn cộng 5 vào điểm đầu tiên
scores = [85, 92, 78]

first = scores[0:1]     # Đây là một LIST: [85], không phải số 85
print(first)

# print(first + 5)      # TypeError: can only concatenate list (not "int") to list
````

🔍 **Sai ở đâu?**

- `scores[0:1]` là một **slice**, nên kết quả là một **list** có một giá trị: `[85]`.
- Không thể cộng một list với một số bằng `+`, nên Python báo lỗi.
- Thông báo lỗi có nghĩa là: "chỉ có thể nối list với list, không thể nối với số (int)".
- Hãy nhìn dấu ngoặc vuông trong kết quả: `[85]`. Khi thấy dấu ngoặc vuông, đó là một list.

✅ **Cách làm đúng:** Để lấy **một giá trị**, hãy dùng **chỉ số** (một số, không có dấu hai chấm): `scores[0]`.

````python
# ✅ Đúng
scores = [85, 92, 78]

first = scores[0]       # Đây là số 85 (không có dấu ngoặc vuông)
print(first)
print(first + 5)        # 90
````

### Lỗi 5: Nghĩ rằng slice làm thay đổi list ban đầu

````python
# ❌ Sai: chúng ta muốn chỉ giữ lại 3 điểm đầu tiên
scores = [85, 92, 78, 90, 66]

scores[:3]              # Dòng này tạo ra một list mới... nhưng không được lưu ở đâu cả!
print(scores)           # [85, 92, 78, 90, 66] -> không có gì thay đổi
````

🔍 **Sai ở đâu?**

- Slicing tạo ra một **list mới**. Nó **không** thay đổi list ban đầu.
- Ở dòng `scores[:3]`, Python đã tạo ra list mới `[85, 92, 78]`, nhưng chúng ta không lưu nó vào biến nào. Nên nó biến mất luôn.
- Không có lỗi nào, nên khó nhận ra.

✅ **Cách làm đúng:** Lưu slice vào một biến bằng `=`. Nếu muốn thay thế list ban đầu, hãy lưu nó trở lại vào cùng biến đó (giống như `age = age + 1`: giá trị mới thay thế giá trị cũ).

````python
# ✅ Đúng (1): lưu vào một biến mới
scores = [85, 92, 78, 90, 66]

first_three = scores[:3]
print(first_three)      # [85, 92, 78]
print(scores)           # List ban đầu vẫn đầy đủ
````

````python
# ✅ Đúng (2): lưu trở lại vào cùng biến
scores = [85, 92, 78, 90, 66]

scores = scores[:3]     # Vế phải được tính trước ([85, 92, 78]), rồi được lưu vào scores
print(scores)           # [85, 92, 78] -> list cũ đã bị thay thế
````

### Lỗi 6: Đặt start sau end

````python
# ❌ Sai: chúng ta muốn các giá trị từ chỉ số 1 đến chỉ số 3
scores = [85, 92, 78, 90, 66]

print(scores[4:1])      # [] -> một list rỗng!
````

🔍 **Sai ở đâu?**

- Slice luôn đi từ **trái sang phải**: từ start đến end.
- `scores[4:1]` bắt đầu ở chỉ số 4 và cố đi đến chỉ số 1. Nhưng 1 nằm bên **trái** của 4, nên ở giữa không có gì cả.
- Kết quả là một list rỗng `[]`. Không có lỗi nào.

✅ **Cách làm đúng:** Viết chỉ số **nhỏ hơn** trước (start), và chỉ số **lớn hơn** sau (end).

````python
# ✅ Đúng
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # Từ chỉ số 1 đến chỉ số 3 -> [92, 78, 90]
````

### Lỗi 7: Dùng [-3:-1] để lấy 3 giá trị cuối

````python
# ❌ Sai: chúng ta muốn 3 giá trị cuối
scores = [85, 92, 78, 90, 66]
#  chỉ số âm: -5  -4  -3  -2  -1

print(scores[-3:-1])    # [78, 90] -> chỉ có 2 giá trị! Thiếu giá trị cuối (66)
````

🔍 **Sai ở đâu?**

- End **không được bao gồm**. Quy tắc đó không thay đổi với chỉ số âm.
- `scores[-3:-1]` có nghĩa là "từ -3, đến trước -1 (**không bao gồm** -1)". Vì vậy giá trị cuối (-1) bị bỏ ra.

✅ **Cách làm đúng:** Để bao gồm giá trị cuối, hãy **bỏ trống end**: `scores[-3:]`. Không có end nghĩa là "đi đến tận cuối".

````python
# ✅ Đúng
scores = [85, 92, 78, 90, 66]

print(scores[-3:])      # Từ -3 đến cuối -> [78, 90, 66]
````

### Lỗi 8: Dùng độ dài của cả list để tính trung bình của slice

````python
# ❌ Sai: chúng ta muốn trung bình của 3 điểm đầu tiên
scores = [80, 90, 100, 50, 60]
total = 0

for score in scores[:3]:
    total += score              # 80 + 90 + 100 = 270

average = total / len(scores)   # 270 / 5 -> sai! Chúng ta chỉ cộng 3 điểm
print("Average:", average)      # 54.0
````

🔍 **Sai ở đâu?**

- Chúng ta chỉ cộng **3** điểm đầu tiên (270).
- Nhưng `len(scores)` đếm **cả 5** giá trị trong cả list.
- Vì vậy chúng ta chia cho 5 thay vì 3. Trung bình thấp hơn nhiều so với đúng.
- Không có lỗi nào.

✅ **Cách làm đúng:** Chia cho số lượng giá trị **trong slice**. Cách dễ nhất: lưu slice vào một biến trước, rồi dùng `len()` với biến đó.

````python
# ✅ Đúng
scores = [80, 90, 100, 50, 60]
first_three = scores[:3]        # [80, 90, 100]
total = 0

for score in first_three:
    total += score

average = total / len(first_three)  # 270 / 3
print("Average:", average)          # 90.0
````

### Lỗi 9: Tưởng sẽ có lỗi khi slice vượt quá cuối list

````python
# ❓ Bất ngờ: dòng này KHÔNG gây ra lỗi
scores = [85, 92, 78]           # Chỉ có 3 giá trị (chỉ số 0, 1, 2)

print(scores[0:10])             # [85, 92, 78] -> không có lỗi!

# Nhưng một chỉ số quá lớn thì CÓ gây ra lỗi
# print(scores[10])             # IndexError: list index out of range
````

🔍 **Chuyện gì đang xảy ra?**

- Với **chỉ số**, Python phải tìm **một** giá trị ở đúng vị trí đó. Nếu không có vị trí như vậy, nó báo lỗi `IndexError` ("chỉ số nằm ngoài list").
- Với **slice**, Python chỉ đưa ra **những giá trị nào nó tìm được** trong khoảng đó. Nếu khoảng đó vượt quá cuối list, nó dừng ở cuối. Không có lỗi.
- Đây không hẳn là một lỗi, nhưng có thể gây bối rối: một slice trông có vẻ sai vẫn có thể "chạy được", và bạn sẽ không nhận ra các con số đã bị viết sai.

✅ **Mẹo:** Nếu bạn muốn lấy tất cả cho đến cuối, đừng đoán một số lớn. Chỉ cần bỏ trống end: `scores[0:]` hoặc `scores[:]`.

````python
# ✅ Tốt hơn
scores = [85, 92, 78]

print(scores[0:])               # Từ chỉ số 0 đến cuối -> [85, 92, 78]
````
