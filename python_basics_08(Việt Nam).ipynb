## 1. Hàm là gì?

### Một vấn đề

Giả sử chúng ta muốn in ra một lời chào ngắn:

```python
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
```

Bây giờ hãy tưởng tượng chúng ta cần in cùng lời chào này **3 lần** ở những chỗ khác nhau trong chương trình. Chúng ta sẽ phải viết đi viết lại 3 dòng đó.

````python
# Cùng 3 dòng, được viết 3 lần
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
````

Cách này chạy được, nhưng có vấn đề:

- Nó dài và viết rất mệt.
- Nếu muốn đổi "Hello!" thành "Hi!", chúng ta phải đổi ở **mọi** chỗ. Nếu quên một chỗ, các lời chào sẽ khác nhau.

### Ý tưởng của hàm

Sẽ thế nào nếu chúng ta chỉ viết 3 dòng đó **một lần**, đặt cho chúng một **cái tên**, rồi mỗi khi cần thì chỉ việc gọi tên đó?

Đó chính là **hàm**: **một nhóm mã có tên**.

### Hãy nghĩ đến một công thức nấu ăn

Hàm giống như một **công thức nấu ăn**.

1. Trước tiên, bạn **viết công thức** ra giấy: "Cách làm bánh kếp: bước 1, bước 2, bước 3." Viết công thức không làm ra cái bánh nào cả. Nó chỉ là hướng dẫn được lưu lại để dùng sau.
2. Sau đó, mỗi khi muốn ăn bánh kếp, bạn **làm theo công thức**. Bạn có thể làm theo cùng một công thức bao nhiêu lần tùy thích.

Hàm hoạt động giống như vậy:

| Công thức | Hàm |
|---|---|
| Viết công thức | **Tạo** (định nghĩa) hàm |
| Làm theo công thức để nấu | **Sử dụng** (gọi) hàm |

### Bạn đã từng dùng hàm rồi!

Bạn đã dùng hàm từ Bài 1. `print()`, `len()` và `type()` đều là hàm. Có người đã viết sẵn chúng cho bạn, và bạn chỉ cần dùng bằng cách viết tên của chúng và dấu ngoặc tròn `( )`.

Trong bài này, bạn sẽ học cách **tự tạo** hàm của riêng mình.

````python
# Những hàm bạn đã từng dùng
print("Hello")              # print() là một hàm: nó hiển thị giá trị lên màn hình
print(len("apple"))         # len() là một hàm: nó đếm
print(type(10))             # type() là một hàm: nó cho biết kiểu dữ liệu
````

## 2. Tạo hàm và sử dụng hàm

Hàm được dùng qua **hai bước riêng biệt**:

1. **Tạo** hàm (viết công thức)
2. **Sử dụng** hàm (làm theo công thức)

### Bước 1: Tạo một hàm

```python
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")
```

Hãy xem từng phần:

| Phần | Ý nghĩa |
|---|---|
| `def` | viết tắt của **define** (định nghĩa). Nó có nghĩa là "tôi sắp tạo một hàm" |
| `say_hello` | **tên** của hàm. Bạn tự chọn (quy tắc giống như đặt tên biến) |
| `()` | dấu ngoặc tròn. Luôn cần có sau tên (chúng ta sẽ xem bên trong đặt gì ở phần 3) |
| `:` | dấu hai chấm ở cuối dòng, giống như `if` và `for` |
| các dòng thụt lề | **thân** của hàm: đoạn mã thuộc về hàm |

**Về thụt lề:** giống như `if` và `for`, Python dùng thụt lề (4 dấu cách ở đầu dòng) để biết những dòng nào thuộc về hàm. Các dòng thụt lề bên dưới `def` là thân hàm. Dòng đầu tiên không thụt lề nằm **bên ngoài** hàm.

### Tạo hàm KHÔNG làm nó chạy

Đây là điều quan trọng nhất trong phần này.

Khi Python đọc `def`, nó chỉ **ghi nhớ** hàm đó. Nó không chạy đoạn mã bên trong. Giống như viết công thức không làm ra bánh kếp.

Hãy nhìn ô bên dưới. Chúng ta tạo một hàm, nhưng **không có gì được in ra**.

````python
# Bước 1: tạo hàm
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")

# Không có gì được in ra! Chúng ta chỉ mới tạo (viết) hàm. Chúng ta chưa dùng nó.
````

### Bước 2: Sử dụng một hàm

Để dùng một hàm, hãy viết **tên** của nó và theo sau là **dấu ngoặc tròn** `()`:

```python
say_hello()
```

Việc sử dụng hàm được gọi là **gọi (call)** hàm. Khi bạn gọi một hàm, Python đi đến hàm đó, chạy toàn bộ mã bên trong từ trên xuống dưới, rồi quay trở lại.

````python
# Bước 2: gọi (sử dụng) hàm
say_hello()
````

### Một công thức, nhiều lần

Một khi hàm đã được tạo, bạn có thể gọi nó **bao nhiêu lần tùy thích**. Mỗi lần gọi sẽ chạy lại toàn bộ thân hàm.

````python
# Cùng lời chào 3 lần, nhưng giờ chỉ cần 3 dòng ngắn
say_hello()
say_hello()
say_hello()
````

Hãy so sánh với phần đầu bài. Trước đây chúng ta cần 9 dòng. Bây giờ chỉ cần 3 lần gọi.

Và nếu muốn đổi "Hello!" thành "Hi!", chúng ta chỉ cần đổi **ở một chỗ**: bên trong hàm. Mọi lần gọi sẽ dùng lời chào mới.

````python
# Thay đổi hàm ở một chỗ...
def say_hello():
    print("Hi!")                        # Chỉ đổi ở đây
    print("Welcome to Python class.")
    print("Have a nice day!")

# ...và mọi lần gọi đều dùng phiên bản mới
say_hello()
say_hello()
````

### Tạo và sử dụng trong cùng một ô

Thông thường, chúng ta tạo hàm trước rồi gọi nó ở bên dưới. Hãy theo dõi những gì Python làm, từng bước một.

````python
def say_goodbye():          # 1. Python ghi nhớ hàm "say_goodbye". Chưa có gì chạy cả
    print("Goodbye!")
    print("See you next time.")

print("Class is over.")     # 2. Dòng này không thụt lề -> nằm ngoài hàm -> nó chạy ngay bây giờ
say_goodbye()               # 3. Gọi hàm -> 2 dòng bên trong chạy bây giờ
print("The end.")           # 4. Sau khi hàm kết thúc, Python quay lại đây
````

Hãy chú ý thứ tự của kết quả:

1. `Class is over.` - dòng đầu tiên **bên ngoài** hàm
2. `Goodbye!` và `See you next time.` - thân hàm chỉ chạy khi chúng ta **gọi** nó
3. `The end.` - sau khi hàm kết thúc, Python tiếp tục từ dòng sau lời gọi

### Tạo trước, dùng sau

Công thức phải được viết **trước** khi bạn có thể làm theo. Tương tự, hàm phải được **tạo trước khi được gọi**. Python đọc mã từ trên xuống dưới, nên nếu bạn gọi một hàm chưa được tạo, Python không biết nó, và báo lỗi. (Chúng ta sẽ thấy lỗi này ở phần 5.)

### Thêm ví dụ

## 3. Đưa một giá trị cho hàm

### Một vấn đề

Hàm `say_hello()` của chúng ta luôn nói đúng một câu giống nhau. Nhưng nếu chúng ta muốn chào **những người khác nhau** bằng tên của họ thì sao?

- "Hello, Anna!"
- "Hello, Tom!"
- "Hello, Bill!"

Chúng ta không muốn tạo một hàm khác cho mỗi người. Chúng ta muốn **một** hàm, và mỗi lần thì **nói cho** nó biết tên.

### Hãy nghĩ đến việc gọi món ở quán cà phê

Khi gọi món ở quán cà phê, bạn **nói** với nhân viên bạn muốn gì: "Cho tôi một ly cà phê." hoặc "Cho tôi một ly trà." Nhân viên làm cùng một công việc (pha đồ uống), nhưng kết quả phụ thuộc vào **điều bạn đã nói**.

Hàm cũng có thể hoạt động như vậy. Khi gọi hàm, chúng ta có thể **đưa** cho nó một giá trị, và hàm làm công việc của nó **với giá trị đó**.

### Tạo một hàm nhận một giá trị

```python
def greet(name):
    print("Hello,", name)
```

Hãy nhìn vào dấu ngoặc tròn: có một từ bên trong, `name`. Đây là một **biến** sẽ nhận giá trị mà chúng ta đưa vào khi gọi hàm. Biến này được gọi là **tham số (parameter)**.

### Gọi hàm với một giá trị

```python
greet("Anna")
```

Bây giờ chúng ta đặt một giá trị vào trong dấu ngoặc khi gọi hàm. Giá trị chúng ta đưa vào được gọi là **đối số (argument)**.

| Từ | Nằm ở đâu | Ví dụ |
|---|---|---|
| **tham số (parameter)** | bên trong `( )` khi **tạo** hàm | `name` trong `def greet(name):` |
| **đối số (argument)** | bên trong `( )` khi **gọi** hàm | `"Anna"` trong `greet("Anna")` |

Đừng lo quá về hai từ này. Ý quan trọng là: **giá trị chúng ta đưa vào được lưu vào biến bên trong hàm.**

````python
# Tạo một hàm nhận một giá trị
def greet(name):
    print("Hello,", name)

# Gọi hàm với các giá trị khác nhau
greet("Anna")
greet("Tom")
greet("Bill")
````

**Về `print("Hello,", name)`:** khi đặt nhiều giá trị vào `print()` và ngăn cách bằng **dấu phẩy** `,`, chúng được in trên một dòng, và Python tự động đặt **một dấu cách** giữa chúng. Chữ trong dấu ngoặc kép được in nguyên như vậy, còn biến thì được thay bằng giá trị bên trong nó.

### Điều gì xảy ra, từng bước một

Khi chúng ta gọi `greet("Anna")`:

| Bước | Python làm gì |
|---|---|
| 1 | Tìm hàm `greet` |
| 2 | Lưu giá trị được đưa vào tham số: `name = "Anna"` |
| 3 | Chạy thân hàm: `print("Hello,", name)` → `name` là `"Anna"` → in ra `Hello, Anna` |
| 4 | Hàm đã kết thúc, nên Python quay lại dòng sau lời gọi |

Rồi khi gọi `greet("Tom")`, điều tương tự lại xảy ra, nhưng lần này `name = "Tom"`.

Vì vậy **mã bên trong hàm lần nào cũng giống nhau**, nhưng **kết quả thay đổi** tùy vào giá trị chúng ta đưa vào.

````python
# Tham số chỉ là một biến. Chúng ta có thể xem bên trong nó có gì
def show_value(x):
    print("The function received:", x)

show_value(10)
show_value("apple")
show_value(3.14)
````

````python
# Một hàm dùng giá trị trong phép tính
def show_double(number):
    print(number * 2)       # number được thay bằng giá trị chúng ta đưa vào

show_double(5)              # number = 5  -> 5 * 2
show_double(100)            # number = 100 -> 100 * 2
````

````python
# Đối số cũng có thể là một biến
def greet(name):
    print("Hello,", name)

student = "Kate"
greet(student)              # Giá trị bên trong student ("Kate") được đưa cho hàm
````

### Đưa hai giá trị

Một hàm có thể nhận **nhiều hơn một** giá trị. Hãy ngăn cách chúng bằng dấu phẩy `,`, cả khi tạo lẫn khi gọi hàm.

```python
def add(a, b):
    print(a + b)

add(2, 3)
```

Các giá trị được ghép **theo thứ tự**:
- giá trị thứ 1 (`2`) đi vào tham số thứ 1 (`a`)
- giá trị thứ 2 (`3`) đi vào tham số thứ 2 (`b`)

````python
def add(a, b):
    print(a + b)

add(2, 3)                   # a = 2, b = 3  -> 2 + 3
add(10, 20)                 # a = 10, b = 20 -> 10 + 20
````

````python
# Một hàm có hai tham số: tên và tuổi
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce("Anna", 20)       # name = "Anna", age = 20
introduce("Tom", 25)        # name = "Tom", age = 25
````

### Thứ tự rất quan trọng

Vì các giá trị được ghép theo thứ tự, đổi thứ tự có thể làm thay đổi kết quả.

````python
def subtract(a, b):
    print(a - b)

subtract(10, 3)             # a = 10, b = 3  -> 10 - 3 = 7
subtract(3, 10)             # a = 3, b = 10  -> 3 - 10 = -7 (kết quả khác!)
````

````python
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce(20, "Anna")       # Sai thứ tự: name = 20, age = "Anna"
````

## 4. Nhận lại kết quả: return

### Một vấn đề

Hàm `add(a, b)` của chúng ta **in** kết quả lên màn hình. Nhưng nếu muốn **sử dụng** kết quả sau này thì sao? Ví dụ, lưu nó vào một biến, hoặc tính toán với nó?

In ra chỉ **hiển thị** kết quả. Nó không **đưa** kết quả lại cho chúng ta. Một khi đã in ra, chúng ta không dùng nó được nữa.

### Hãy nghĩ đến việc nhờ ai đó đi mua đồ

Hãy tưởng tượng bạn nhờ một người bạn: "Đi ra cửa hàng mua giúp mình ít bánh mì nhé."

- Bạn ấy đi, mua bánh mì, rồi quay về.
- Rồi bạn ấy **đưa bánh mì vào tay bạn**.
- Bây giờ **bạn có** bánh mì, và bạn có thể làm gì với nó cũng được: ăn, chia cho người khác, làm bánh sandwich.

`return` giống như **đưa kết quả lại**. Hàm làm công việc của nó, rồi đưa kết quả về đúng chỗ đã gọi nó.

### Tạo một hàm có return

```python
def add(a, b):
    return a + b
```

`return` có nghĩa là: **"đưa giá trị này trở lại chỗ đã gọi hàm."**

Chú ý: hàm này không in gì cả. Nó chỉ đưa kết quả trở lại.

````python
def add(a, b):
    return a + b            # Đưa kết quả trở lại

result = add(2, 3)          # Kết quả (5) quay trở lại và được lưu vào result
print(result)
````

### Lời gọi hàm được thay bằng giá trị trả về

Đây là ý tưởng quan trọng nhất của `return`.

Khi một hàm trả về một giá trị, **chỗ bạn gọi hàm sẽ được thay bằng giá trị đó.**

Hãy xem chậm rãi dòng này:

```python
result = add(2, 3)
```

| Bước | Python nhìn thấy gì |
|---|---|
| 1 | `result = add(2, 3)` |
| 2 | Python gọi `add(2, 3)`: `a = 2`, `b = 3`, và hàm trả về `2 + 3`, tức là `5` |
| 3 | `add(2, 3)` được **thay** bằng `5` → `result = 5` |
| 4 | `5` được lưu vào `result` |

Vì vậy sau khi hàm trả về, mọi thứ giống như bạn đã viết `result = 5`.

Bạn có thể hình dung `add(2, 3)` như một chiếc hộp **biến thành 5** khi hàm chạy xong.

````python
def add(a, b):
    return a + b

x = add(2, 3)               # add(2, 3) -> 5,  nên dòng này trở thành: x = 5
y = add(10, 20)             # add(10, 20) -> 30, nên dòng này trở thành: y = 30

print(x)
print(y)
````

### Dùng giá trị trả về trong phép tính

Vì `add(2, 3)` biến thành `5`, chúng ta có thể dùng nó ở bất cứ đâu mà ta có thể dùng số `5`.

```python
add(2, 3) + add(3, 4)
```

| Bước | Python nhìn thấy gì |
|---|---|
| 1 | `add(2, 3) + add(3, 4)` |
| 2 | `add(2, 3)` chạy và trả về 5 → **`5`** `+ add(3, 4)` |
| 3 | `add(3, 4)` chạy và trả về 7 → `5 +` **`7`** |
| 4 | `12` |

````python
def add(a, b):
    return a + b

total = add(2, 3) + add(3, 4)   # trở thành 5 + 7
print(total)                    # 12
````

````python
def add(a, b):
    return a + b

print(add(2, 3) * 10)           # add(2, 3) trở thành 5 -> print(5 * 10) -> 50
print(add(2, 3) > 4)            # add(2, 3) trở thành 5 -> print(5 > 4) -> True
````

### Đặt lời gọi trực tiếp vào trong print()

`print(add(2, 3))` cũng hoạt động theo cách tương tự:

| Bước | Python nhìn thấy gì |
|---|---|
| 1 | `print(add(2, 3))` |
| 2 | `add(2, 3)` trả về 5 → `print(5)` |
| 3 | `5` được in ra |

````python
def add(a, b):
    return a + b

print(add(2, 3))                # trở thành print(5)
````

### Đưa giá trị trả về cho một hàm khác

Một giá trị trả về thậm chí có thể được đưa vào hàm một lần nữa. Python luôn làm **từ trong ra ngoài**: lời gọi bên trong được thay thế trước.

```python
add(add(1, 2), 3)
```

| Bước | Python nhìn thấy gì |
|---|---|
| 1 | `add(add(1, 2), 3)` |
| 2 | `add(1, 2)` bên trong trả về 3 → `add(` **`3`** `, 3)` |
| 3 | `add(3, 3)` trả về 6 → **`6`** |

````python
def add(a, b):
    return a + b

result = add(add(1, 2), 3)      # add(1, 2) -> 3, rồi add(3, 3) -> 6
print(result)
````

### Thêm ví dụ với return

### print và return: khác nhau ở đâu?

Bây giờ hãy tạo hai hàm trông gần như giống nhau. Một hàm dùng `print`, hàm kia dùng `return`.

```python
def show_add(a, b):        # dùng print
    print(a + b)

def add(a, b):             # dùng return
    return a + b
```

| | `print` bên trong hàm | `return` bên trong hàm |
|---|---|---|
| Nó làm gì | **hiển thị** kết quả lên màn hình | **đưa** kết quả trở lại chỗ gọi |
| Có lưu kết quả vào biến được không? | ❌ | ✅ |
| Có tính toán với kết quả được không? | ❌ | ✅ |
| Giống như... | người bạn cho bạn xem bánh mì rồi giữ lại | người bạn đưa bánh mì vào tay bạn |

### Khi không có return thì cái gì quay trở lại?

Nếu hàm không có `return`, nó vẫn đưa lại một thứ: một giá trị đặc biệt gọi là **`None`**. `None` có nghĩa là **"không có gì"**.

Vì vậy `show_add(2, 3)` được thay bằng `None`, không phải bằng `5`.

````python
def show_add(a, b):
    print(a + b)                # Chỉ hiển thị kết quả. Không có return

result = show_add(2, 3)         # 5 được in lên màn hình... nhưng result chứa gì?
print("result is:", result)     # None! show_add(2, 3) được thay bằng None, không phải 5
````

Hãy nhìn kỹ kết quả:

1. `5` - được in ra **bởi hàm** (bởi lệnh `print` bên trong nó)
2. `result is: None` - giá trị lưu trong `result` là `None`, vì hàm không trả về gì

Số 5 đã xuất hiện trên màn hình, nhưng **chúng ta không nhận lại được nó**. Nó đã mất.

### Vì vậy add(2, 3) + add(3, 4) không chạy được với print

Điều gì xảy ra nếu chúng ta thử `show_add(2, 3) + show_add(3, 4)`?

| Bước | Python nhìn thấy gì |
|---|---|
| 1 | `show_add(2, 3) + show_add(3, 4)` |
| 2 | `show_add(2, 3)` in ra 5, và được thay bằng **`None`** |
| 3 | `show_add(3, 4)` in ra 7, và được thay bằng **`None`** |
| 4 | `None + None` → **lỗi!** Python không thể cộng "không có gì" với "không có gì" |

````python
def show_add(a, b):
    print(a + b)

# total = show_add(2, 3) + show_add(3, 4)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'NoneType'
````

(Dòng ở trên là chú thích để không làm notebook dừng lại. Nếu bạn xóa hai dấu `#` ở dòng `total = ...`, bạn sẽ thấy `5` và `7` được in ra trước, rồi đến lỗi.)

Thông báo lỗi có nghĩa là: "không thể dùng `+` với `None` và `None`". (`NoneType` là kiểu dữ liệu của `None`.)

### Khi nào dùng cái nào?

- Nếu bạn chỉ muốn **hiển thị** gì đó lên màn hình → dùng `print` là được
- Nếu bạn muốn **sử dụng** kết quả sau này (lưu lại, tính toán, đưa cho hàm khác) → dùng `return`

Phần lớn các hàm tính toán nên dùng `return`. Khi đó bạn vẫn có thể in kết quả bất cứ lúc nào, bằng `print(add(2, 3))`.

````python
def add(a, b):
    return a + b

# Với return, chúng ta làm được cả hai: dùng kết quả VÀ in nó ra
total = add(2, 3) + add(3, 4)
print(total)
````

## 5. Những lỗi thường gặp

Đây là những lỗi mà hầu như ai cũng mắc phải khi học về hàm. Với mỗi lỗi, chúng ta sẽ xem:

- ❌ **Mã sai**
- 🔍 **Sai ở đâu**
- ✅ **Cách làm đúng**

Mã gây ra lỗi được viết dưới dạng chú thích (có `#` ở đầu) để không làm notebook dừng lại. Nếu muốn xem thông báo lỗi thật, hãy xóa `#`, chạy ô đó, rồi đặt `#` lại.

### Lỗi 1: Tạo hàm rồi tưởng nó sẽ tự chạy

````python
# ❌ Sai: chúng ta muốn in "Good morning!"
def morning():
    print("Good morning!")

# Không có gì được in ra...
````

🔍 **Sai ở đâu?**

- `def` chỉ **tạo** hàm. Nó không chạy hàm. (Viết công thức không làm ra món ăn.)
- Không có lỗi nào, nên có thể gây bối rối: "Sao không có gì xảy ra vậy?"

✅ **Cách làm đúng:** **Gọi** hàm sau khi tạo nó.

````python
# ✅ Đúng
def morning():
    print("Good morning!")

morning()               # Gọi hàm
````

### Lỗi 2: Quên dấu ngoặc tròn khi gọi

````python
# ❌ Sai
def morning():
    print("Good morning!")

morning                 # Không có dấu ngoặc tròn -> hàm KHÔNG chạy
print("Did you see 'Good morning!' above? No!")
````

🔍 **Sai ở đâu?**

- `morning` không có `()` chỉ là **tên** của hàm. Nó không gọi hàm.
- Để gọi hàm, bạn **phải** viết dấu ngoặc tròn: `morning()`. Dấu ngoặc tròn có nghĩa là "chạy nó ngay bây giờ".
- Không có lỗi nào, nhưng hàm không chạy.

✅ **Cách làm đúng:** Luôn viết `()` sau tên khi gọi hàm.

````python
# ✅ Đúng
def morning():
    print("Good morning!")

morning()               # Có dấu ngoặc tròn
````

### Lỗi 3: Gọi hàm trước khi tạo nó

````python
# ❌ Sai
# evening()             # NameError: name 'evening' is not defined
#
# def evening():
#     print("Good evening!")
````

🔍 **Sai ở đâu?**

✅ **Cách làm đúng:** Tạo hàm trước, rồi gọi nó ở bên dưới.

### Lỗi 5: Đưa sai số lượng giá trị

````python
# ❌ Sai
def add(a, b):
    return a + b

# add(2)          # TypeError: add() missing 1 required positional argument: 'b'
# add(2, 3, 4)    # TypeError: add() takes 2 positional arguments but 3 were given
````

🔍 **Sai ở đâu?**

- `add` được tạo với **2** tham số (`a` và `b`), nên nó cần đúng **2** giá trị.
- `add(2)`: chỉ đưa 1 giá trị. `a` nhận 2, nhưng `b` không nhận được gì. Lỗi nói "thiếu 1 giá trị: `b`".
- `add(2, 3, 4)`: đưa 3 giá trị, nhưng chỉ có 2 chỗ để chứa. Lỗi nói "nó nhận 2, nhưng được đưa 3".

✅ **Cách làm đúng:** Đưa đúng số lượng giá trị bằng số tham số.

````python
# ✅ Đúng
def add(a, b):
    return a + b

print(add(2, 3))        # 2 tham số, 2 giá trị
````

### Lỗi 6: Dùng print khi cần return

````python
# ❌ Sai: chúng ta muốn lưu kết quả và dùng nó sau này
def multiply(a, b):
    print(a * b)                # Chỉ hiển thị kết quả

answer = multiply(3, 4)         # 12 được in ra... nhưng answer là None
print("answer:", answer)
````

🔍 **Sai ở đâu?**

- `multiply` dùng `print`, không dùng `return`. Nó **hiển thị** 12 lên màn hình, nhưng không **đưa nó trở lại**.
- Hàm không có `return` sẽ đưa lại `None`. Vì vậy `multiply(3, 4)` được thay bằng `None`, và `answer` là `None`.
- Nếu chúng ta thử tính toán với nó (như `answer + 1`), sẽ bị lỗi.

✅ **Cách làm đúng:** Dùng `return` để đưa kết quả trở lại. Sau đó in nó ra ở bên ngoài hàm nếu muốn xem.

````python
# ✅ Đúng
def multiply(a, b):
    return a * b                # Đưa kết quả trở lại

answer = multiply(3, 4)         # trở thành answer = 12
print("answer:", answer)
print("answer + 1:", answer + 1)
````

### Lỗi 7: Có return nhưng không dùng kết quả

````python
# ❌ Sai: chúng ta muốn xem kết quả
def multiply(a, b):
    return a * b

multiply(3, 4)                  # Kết quả (12) quay trở lại... nhưng không ai đón lấy nó!
print("Where is the result?")
````

🔍 **Sai ở đâu?**

- `multiply(3, 4)` được thay bằng `12`. Nhưng dòng này chỉ có `12` đứng một mình. Chúng ta không lưu nó, cũng không in nó ra.
- Giống như bạn của bạn đưa bánh mì cho bạn, và bạn đánh rơi nó xuống đất.
- Không có lỗi nào, nhưng kết quả bị mất.

✅ **Cách làm đúng:** Đón lấy giá trị trả về: lưu nó vào biến, hoặc đặt lời gọi vào trong `print()`.

````python
# ✅ Đúng
def multiply(a, b):
    return a * b

result = multiply(3, 4)         # Lưu lại
print(result)

print(multiply(3, 4))           # Hoặc in ra trực tiếp
````
