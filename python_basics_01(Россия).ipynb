## 1. Переменные и типы данных

Переменная — это коробка для хранения значения. Если дать коробке имя, позже можно просто вызвать это имя, чтобы получить значение внутри. Если положить новое значение, содержимое коробки тоже изменится.

Однако значения, которые можно положить в коробку, бывают разных видов. Этот вид значения называется типом данных. Основные типы данных в Python следующие.

- `str` : строка (текст)
- `int` : целое число
- `float` : число с плавающей точкой (число с десятичной дробью)
- `bool` : логический тип (истина/ложь)

Кстати, `print()` — это команда, которая выводит значение в скобках на экран.

```python
# = означает не 'равно', а 'сохранить значение справа в коробку с именем слева'
# Имена переменных (name, age, height, is_student) — это имена, данные коробкам

# Строку (str) достаточно заключить в кавычки. Можно использовать и двойные (" "), и одинарные (' ') кавычки
name = "Bill"                    # Слово в двойных кавычках
city = 'Seoul'                    # Слово в одинарных кавычках
greeting = "Hello, nice to meet you"  # Предложение тоже остаётся строкой
zip_code = "06236"               # Даже если выглядит как число, если оно в кавычках — это строка

# Целое число (int) — это число без десятичной дроби. Положительные, отрицательные числа и 0 — всё целые числа
age = 20              # Положительное целое число
temperature = -5      # Отрицательное целое число тоже возможно
score = 0             # 0 тоже целое число

# Число с плавающей точкой (float) — это число с десятичной дробью
height = 175.5        # Положительное число с плавающей точкой
account_balance = -3.75  # Отрицательное тоже возможно
pi_value = 3.14159    # Не важно, сколько знаков после запятой

# Логический тип (bool) может быть только True (истина) или False (ложь) (первая буква всегда заглавная)
is_student = True     # является студентом
is_married = False    # не состоит в браке
has_passed = True     # сдал экзамен

# print() — команда, которая выводит значение на экран
print(name)
print(city)
print(greeting)
print(zip_code)
print(age)
print(temperature)
print(score)
print(height)
print(account_balance)
print(pi_value)
print(is_student)
print(is_married)
print(has_passed)
```

```python
# Значение переменной можно изменить (пример с целым числом)
age = 20
print(age)

age = 21   # Если сохранить новое значение в ту же переменную, старое значение исчезает и заменяется новым
print(age)

age = age + 1   # Можно также создать новое значение на основе старого (21 + 1 = 22)
print(age)
```

```python
# Значение переменной можно изменить (пример с логическим типом)
is_raining = False
print(is_raining)

is_raining = True   # Если сохранить новое значение в ту же переменную, старое значение исчезает и заменяется новым
print(is_raining)

is_raining = not is_raining   # С помощью not значение переворачивается: True становится False, а False — True
print(is_raining)
```

```python
# Значение переменной можно изменить (пример со строкой)
city = "Seoul"
print(city)

city = "Busan"   # Если сохранить новое значение в ту же переменную, старое значение исчезает и заменяется новым
print(city)

city = city + " Travel"   # Строки можно соединять с помощью + ("Busan" + " Travel" = "Busan Travel")
print(city)
```

```python
# Значение переменной можно изменить (пример с числом с плавающей точкой)
height = 175.5
print(height)

height = 178.2   # Если сохранить новое значение в ту же переменную, старое значение исчезает и заменяется новым
print(height)

height = height + 0.5   # Можно также создать новое значение на основе старого (178.2 + 0.5 = 178.7)
print(height)
```

## 2. Основные операторы

Символы, используемые для вычислений с числами и сравнения значений.

```python
# Четыре арифметических действия
a = 10
b = 3

print(a + b)   # Сложение
print(a - b)   # Вычитание
print(a * b)   # Умножение
print(a / b)   # Деление (результат — число с плавающей точкой)
print(a // b)  # Деление (остаётся только частное)
print(a % b)   # Остаток от деления
```

```python
# Операторы сравнения -> результат всегда True или False
print(a > b)   # a больше b
print(a < b)   # a меньше b
print(a == b)  # a равно b
print(a != b)  # a не равно b
```
