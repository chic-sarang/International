## ၁။ ပြောင်းလဲကိန်း (Variable) နှင့် ဒေတာအမျိုးအစား

Variable ဆိုသည်မှာ တန်ဖိုးတစ်ခုကို သိမ်းဆည်းထားသည့် သေတ္တာတစ်ခုဖြစ်သည်။ သေတ္တာကို နာမည်တပ်ထားလျှင် နောက်ပိုင်း ထိုနာမည်ကို ခေါ်ရုံဖြင့် အထဲရှိတန်ဖိုးကို ပြန်ယူသုံးနိုင်သည်။ တန်ဖိုးအသစ်ထည့်လိုက်လျှင် သေတ္တာအတွင်းရှိအရာသည်လည်း ပြောင်းလဲသွားသည်။

သို့သော် သေတ္တာထဲသို့ထည့်နိုင်သည့်တန်ဖိုးများတွင် အမျိုးအစားများရှိသည်။ ဤအမျိုးအစားကို ဒေတာအမျိုးအစား (data type) ဟုခေါ်သည်။ Python ၏ အဓိကဒေတာအမျိုးအစားများမှာ အောက်ပါအတိုင်းဖြစ်သည်။

- `str` : string (စာလုံး)
- `int` : ကိန်းပြည့်
- `float` : ဒဿမကိန်း (ဒဿမနေရာပါသောကိန်း)
- `bool` : boolean (မှန်/မှား)

ထို့ပြင် `print()` သည် ကွင်းအတွင်းရှိတန်ဖိုးကို မျက်နှာပြင်ပေါ်တွင် ပြသပေးသည့် command ဖြစ်သည်။

```python
# = သည် 'ညီမျှသည်' ဟူသောအဓိပ္ပာယ်မဟုတ်ဘဲ 'ညာဘက်ရှိတန်ဖိုးကို ဘယ်ဘက်နာမည်ရှိသေတ္တာထဲသို့ သိမ်းဆည်းသည်' ဟူသောအဓိပ္ပာယ်ဖြစ်သည်
# Variable အမည်များ (name, age, height, is_student) သည် သေတ္တာများကို ပေးထားသောအမည်များဖြစ်သည်

# String (str) ကို quotation mark ဖြင့်ခြုံရုံဖြင့်ရသည်။ double quote (" ") နှင့် single quote (' ') နှစ်မျိုးလုံးသုံးနိုင်သည်
name = "Bill"                    # double quote ဖြင့်ခြုံထားသောစကားလုံး
city = 'Seoul'                    # single quote ဖြင့်ခြုံထားသောစကားလုံး
greeting = "Hello, nice to meet you"  # စာကြောင်းတစ်ကြောင်းလည်း string ဖြစ်သည်
zip_code = "06236"               # ကိန်းဂဏန်းလိုမြင်ရသော်လည်း quotation mark ထဲတွင်ရှိလျှင် string ဖြစ်သည်

# ကိန်းပြည့် (int) သည် ဒဿမနေရာမပါသောကိန်းဖြစ်သည်။ အပေါင်းကိန်း၊ အနှုတ်ကိန်း၊ 0 အားလုံးသည် ကိန်းပြည့်ဖြစ်သည်
age = 20              # အပေါင်းကိန်းပြည့်
temperature = -5      # အနှုတ်ကိန်းပြည့်လည်းရသည်
score = 0             # 0 သည်လည်း ကိန်းပြည့်ဖြစ်သည်

# ဒဿမကိန်း (float) သည် ဒဿမနေရာပါသောကိန်းဖြစ်သည်
height = 175.5        # အပေါင်းဒဿမကိန်း
account_balance = -3.75  # အနှုတ်ဒဿမကိန်းလည်းရသည်
pi_value = 3.14159    # ဒဿမနေရာများများပါလည်း ပြဿနာမရှိပါ

# Boolean (bool) တွင် True (မှန်) သို့မဟုတ် False (မှား) ဟူ၍ တန်ဖိုးနှစ်မျိုးသာရှိသည် (ပထမစာလုံးသည် အမြဲစာလုံးကြီးဖြစ်ရမည်)
is_student = True     # ကျောင်းသား/သူဖြစ်သည်
is_married = False    # အိမ်ထောင်မရှိသေးပါ
has_passed = True     # စာမေးပွဲအောင်ပြီးပါပြီ

# print() သည် တန်ဖိုးကို မျက်နှာပြင်ပေါ်တွင်ပြသပေးသည့် command ဖြစ်သည်
print(name)
print(city)
print(greeting)
print(zip_code)
print(age)
print(temperature)
print(score)
print(height)
print(account_balance)
print(pi_value)
print(is_student)
print(is_married)
print(has_passed)
```

```python
# Variable ၏တန်ဖိုးကို ပြန်လည်ပြောင်းလဲနိုင်သည် (ကိန်းပြည့်ဥပမာ)
age = 20
print(age)

age = 21   # Variable တစ်ခုတည်းတွင် တန်ဖိုးအသစ်ကို သိမ်းဆည်းလိုက်လျှင် ယခင်တန်ဖိုးပျောက်သွားပြီး တန်ဖိုးအသစ်ဖြင့်အစားထိုးသွားသည်
print(age)

age = age + 1   # ယခင်တန်ဖိုးကိုအသုံးပြု၍ တန်ဖိုးအသစ်ကို ဖန်တီးနိုင်သည် (21 + 1 = 22)
print(age)
```

```python
# Variable ၏တန်ဖိုးကို ပြန်လည်ပြောင်းလဲနိုင်သည် (boolean ဥပမာ)
is_raining = False
print(is_raining)

is_raining = True   # Variable တစ်ခုတည်းတွင် တန်ဖိုးအသစ်ကို သိမ်းဆည်းလိုက်လျှင် ယခင်တန်ဖိုးပျောက်သွားပြီး တန်ဖိုးအသစ်ဖြင့်အစားထိုးသွားသည်
print(is_raining)

is_raining = not is_raining   # not ကိုသုံးလျှင် တန်ဖိုးပြောင်းပြန်ဖြစ်သွားသည်: True သည် False ဖြစ်သွားပြီး False သည် True ဖြစ်သွားသည်
print(is_raining)
```

```python
# Variable ၏တန်ဖိုးကို ပြန်လည်ပြောင်းလဲနိုင်သည် (string ဥပမာ)
city = "Seoul"
print(city)

city = "Busan"   # Variable တစ်ခုတည်းတွင် တန်ဖိုးအသစ်ကို သိမ်းဆည်းလိုက်လျှင် ယခင်တန်ဖိုးပျောက်သွားပြီး တန်ဖိုးအသစ်ဖြင့်အစားထိုးသွားသည်
print(city)

city = city + " Travel"   # String များကို + ဖြင့်ဆက်စပ်နိုင်သည် ("Busan" + " Travel" = "Busan Travel")
print(city)
```

```python
# Variable ၏တန်ဖိုးကို ပြန်လည်ပြောင်းလဲနိုင်သည် (ဒဿမကိန်းဥပမာ)
height = 175.5
print(height)

height = 178.2   # Variable တစ်ခုတည်းတွင် တန်ဖိုးအသစ်ကို သိမ်းဆည်းလိုက်လျှင် ယခင်တန်ဖိုးပျောက်သွားပြီး တန်ဖိုးအသစ်ဖြင့်အစားထိုးသွားသည်
print(height)

height = height + 0.5   # ယခင်တန်ဖိုးကိုအသုံးပြု၍ တန်ဖိုးအသစ်ကို ဖန်တီးနိုင်သည် (178.2 + 0.5 = 178.7)
print(height)
```

## ၂။ အခြေခံ Operator များ

ကိန်းဂဏန်းတွက်ချက်ရန် သို့မဟုတ် တန်ဖိုးများနှိုင်းယှဉ်ရန်သုံးသော သင်္ကေတများ။

```python
# အခြေခံသင်္ချာလုပ်ဆောင်ချက်လေးမျိုး
a = 10
b = 3

print(a + b)   # ပေါင်းခြင်း
print(a - b)   # နှုတ်ခြင်း
print(a * b)   # မြှောက်ခြင်း
print(a / b)   # စားခြင်း (ရလဒ်သည် ဒဿမကိန်းဖြစ်သည်)
print(a // b)  # စားခြင်း (အကြွင်းမဲ့ကိန်းသာကျန်သည်)
print(a % b)   # စားခြင်း၏အကြွင်း
```

```python
# နှိုင်းယှဉ် operator -> ရလဒ်သည် အမြဲတမ်း True သို့မဟုတ် False ဖြစ်သည်
print(a > b)   # a သည် b ထက်ကြီးသည်
print(a < b)   # a သည် b ထက်ငယ်သည်
print(a == b)  # a သည် b နှင့်ညီသည်
print(a != b)  # a သည် b နှင့်မတူညီပါ
```
