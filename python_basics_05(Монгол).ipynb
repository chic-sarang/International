## 1. Жагсаалттай for давталтыг яагаад ашигладаг вэ?

Хичээл 4-т бид жагсаалт олон утга хадгалж чаддаг, мөн нэг утгыг **индекс**-ээр нь (байрлалын дугаараар) авч болдгийг сурсан. Эхний утга 0 индекст, хоёр дахь утга 1 индекст байх гэх мэт.

Тиймээс жагсаалт дахь бүх оноог хэвлэхийг хүсвэл индекс бүрд нэг мөр бичиж болно:

```python
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
```

Энэ нь ажиллана. Гэхдээ хоёр асуудал бий:

- Жагсаалт 100 утгатай бол 100 мөр бичих хэрэгтэй.
- `append()`-ээр шинэ утга нэмбэл өөрөө дахин нэг `print()` мөр нэмэх хэрэгтэй.

**for давталт** үүнийг шийддэг. for давталт нь жагсаалтаас утгуудыг **эхнээс нь сүүл хүртэл нэг нэгээр** гаргаж, утга бүрд ижил кодыг ажиллуулдаг. Жагсаалт хэр урт байх нь хамаагүй. Кодоо ганцхан удаа бичихэд л болно.

````python
# Урт арга: утга бүрд нэг мөр
scores = [85, 92, 78, 90, 66]
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
````

````python
# Богино арга: for давталт
# Үр дүн яг адилхан ч ердөө 2 мөр
scores = [85, 92, 78, 90, 66]
for score in scores:
    print(score)
````

````python
# Жагсаалт уртсан ч for давталт хэвээрээ
# Нэг ч print() мөр нэмэх шаардлагагүй
scores = [85, 92, 78, 90, 66]
scores.append(100)      # Төгсгөлд нэг утга нэмнэ (одоо 6 утга)
scores.append(73)       # Дахиад нэгийг (одоо 7 утга)

for score in scores:    # Дээрхтэй ижил 2 мөр
    print(score)
````

## 2. Үндсэн хэлбэр: `for утга in жагсаалт`

Кодын хэлбэр нь ийм байна:

```python
for score in scores:
    давтах код
```

Үүнийг өгүүлбэр шиг уншиж болно: **"scores доторх score бүрийн хувьд доорх кодыг ажиллуул"**.

Хэсэг бүрийг харъя:

| Хэсэг | Утга |
|---|---|
| `for` | "утга бүрийн хувьд дараахыг давт" |
| `score` | хувьсагч. Удаа бүр жагсаалтын нэг утга энд хадгалагдана |
| `in` | "дотор" |
| `scores` | бидний гүйж үзэх жагсаалт |
| `:` | хоёр цэг. `for` мөрийн төгсгөлд заавал байх ёстой |
| `    давтах код` | утга бүрд ажиллах код. Догол мөртэй (4 зай) байх ёстой |

**Хэрхэн ажилладаг вэ, алхам алхмаар** (`scores = [85, 92, 78]` үед):

| Давталт | `score`-д юу хадгалагдах вэ | Юу болох вэ |
|---|---|---|
| 1 дэх | `85` (0 индекс дээрх утга) | `score` = 85 байхад догол мөртэй код ажиллана |
| 2 дахь | `92` (1 индекс дээрх утга) | `score` = 92 байхад догол мөртэй код ажиллана |
| 3 дахь | `78` (2 индекс дээрх утга) | `score` = 78 байхад догол мөртэй код ажиллана |
| - | утга дууссан | Давталт дуусаж, Python давталтын дараах код руу шилжинэ |

Тиймээс давталт **жагсаалтад хэдэн утга байна, төдий удаа** ажиллана. 3 утгатай жагсаалт → 3 удаа давтана. 100 утгатай жагсаалт → 100 удаа давтана.

````python
# Жимс бүрийг нэг нэгээр хэвлэх
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:        # 1 дэх удаа: fruit = "apple", 2 дахь удаа: fruit = "banana", 3 дахь удаа: fruit = "cherry"
    print(fruit)            # Яг одоо fruit-д байгаа утгыг хэвлэнэ
````

````python
# Тоонуудын жагсаалт
numbers = [10, 20, 30, 40]

for number in numbers:      # number нь 10, дараа нь 20, дараа нь 30, дараа нь 40 болно
    print(number)
````

````python
# bool утгуудын жагсаалт (True / False)
answers = [True, False, True]

for answer in answers:      # answer нь True, дараа нь False, дараа нь True болно
    print(answer)
````

````python
# Утга бүрээр тооцоолж болно
# Үнэ бүрийг 2-оор үржүүлээд хэвлэнэ
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)        # 1000 * 2, дараа нь 2500 * 2, дараа нь 3000 * 2
````

````python
# Жагсаалт доторх утгууд ӨӨРЧЛӨГДӨХГҮЙ
# Дээрх нүдэнд price * 2-ыг хэвлэсэн ч жагсаалт өөрөө хэвээрээ
prices = [1000, 2500, 3000]

for price in prices:
    print(price * 2)

print(prices)               # [1000, 2500, 3000] хэвээр
````

````python
# Тэмдэгт мөрүүдийг +-ээр залгах
# Санаарай: + нь хоёр тэмдэгт мөрийг нэг болгон залгаж чадна ("Hello, " + "Bill" -> "Hello, Bill")
names = ["Bill", "Anna", "Tom"]

for name in names:
    print("Hello, " + name)
````

````python
# Жагсаалт хоосон бол давталт 0 удаа ажиллана
# Юу ч хэвлэгдэхгүй, алдаа ч гарахгүй
empty_list = []

for value in empty_list:
    print("This never prints.")

print("The loop ran 0 times.")
````

## 3. Догол мөр: аль мөрүүд давтагддаг вэ?

**Догол мөр** гэдэг нь мөрийн эхэнд зай нэмж баруун тийш шилжүүлэхийг хэлнэ. Python-д **4 зай** ашигладаг.

```python
for fruit in fruits:
    print(fruit)          # <- энэ мөр догол мөртэй (эхэндээ 4 зайтай)
```

### Догол мөр яагаад хэрэгтэй вэ?

`for` давталт ямар нэг кодыг давтдаг. Гэхдээ Python **аль мөрүүдийг** давтах, давтах хэсэг хаана дуусахыг яаж мэдэх вэ?

Python догол мөрийг хардаг:

- `for` мөрийн шууд дараах **догол мөртэй** мөрүүд → давталтын **дотор** → утга бүрд давтагдана
- Дахин **догол мөргүй** болсон эхний мөр → давталтын **гадна** → давталт дууссаны дараа зөвхөн нэг удаа ажиллана

Догол мөртэй мөрүүд нэг бүлэг шиг хамт байдаг. Ийм мөрүүдийн бүлгийг **блок (block)** гэдэг. `for` давталтын блок нь яг "давтах код" юм.

(Энэ нь Хичээл 2-т `if`-д ашигласан дүрэмтэй адил: `if`-ийн доорх догол мөртэй мөрүүд зөвхөн нөхцөл True үед ажилладаг.)

### Ижил код, өөр догол мөр, өөр үр дүн

Доорх хоёр нүдний мөрүүд яг адилхан. **Цорын ганц** ялгаа нь сүүлийн `print()` догол мөртэй эсэх. Үр дүн хэрхэн өөрчлөгдөхийг хараарай.

````python
# А хувилбар: хоёр дахь print() догол мөртэй -> давталтын ДОТОР
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
    print("Done!")          # Догол мөртэй -> давталтын дотор -> 3 удаа хэвлэгдэнэ
````

````python
# Б хувилбар: хоёр дахь print() догол мөргүй -> давталтын ГАДНА
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
print("Done!")              # Догол мөргүй -> давталтын гадна -> төгсгөлд зөвхөн нэг удаа хэвлэгдэнэ
````

### Блок олон мөртэй байж болно

Догол мөртэй мөр бүр утга бүрд дээрээс доош дарааллаар ажиллана. Дараа нь дараагийн утга ирж, бүх блок дахин дээрээсээ ажиллана.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Now the value is:")      # Блокийн 1-р мөр
    print(fruit)                    # Блокийн 2-р мөр
    print("----------")             # Блокийн 3-р мөр

print("The loop is finished.")      # Догол мөргүй -> давталтын гадна -> нэг удаа ажиллана
````

### Нэг блокийн бүх мөр ижил догол мөртэй байх ёстой

Нэг мөр 4 зайтай, дараагийн мөр 2 зайтай бол Python андуурч алдаа гаргана. Үргэлж яг 4 зай ашиглаарай. (Jupyter-т **Tab** товч дарахад 4 зайг автоматаар нэмдэг.)

```python
for fruit in fruits:
    print(fruit)       # 4 зай
  print("Done!")       # 2 зай -> IndentationError!
```

## 4. Давталтын хувьсагчийн нэр

`for score in scores:` дотор `score` гэдэг үг нь for давталтын үүсгэдэг **шинэ хувьсагч** юм. Давталт бүрд жагсаалтын нэг утга түүнд хадгалагдана.

**Энэ хувьсагчид хүссэн нэрээ өгч болно.** Python нэрэнд ач холбогдол өгдөггүй. Зүгээр л утга бүрийг тэр нэртэй хайрцагт хадгалдаг.

Доорх гурван нүд өөр өөр нэр (`x`, `score`, `banana`) ашигласан ч үр дүн яг адилхан.

````python
scores = [85, 92, 78]

for x in scores:            # Хувьсагчийн нэр x
    print(x)
````

````python
scores = [85, 92, 78]

for score in scores:        # Хувьсагчийн нэр score
    print(score)
````

````python
scores = [85, 92, 78]

for banana in scores:       # Хувьсагчийн нэр banana (хачин ч ажиллана!)
    print(banana)
````

Гурвуулаа ажиллана. Гэхдээ аль нь хамгийн ойлгомжтой вэ?

- `x` нь ямар утга болохыг хэлдэггүй.
- `banana` нь будлиулдаг, учир нь эдгээр нь гадил биш оноо.
- `score` нь юу болохыг яг хэлдэг: онооны жагсаалтаас **нэг оноо**.

**Сайн дадал:** жагсаалтыг олон тооны үгээр, давталтын хувьсагчийг тэр үгийн **ганц тооны** хэлбэрээр нэрлээрэй.

| Жагсаалт (олон утга) | Давталтын хувьсагч (нэг утга) |
|---|---|
| `scores` | `score` |
| `fruits` | `fruit` |
| `names` | `name` |
| `prices` | `price` |

Ингэвэл кодыг уншихдаа "бүх жагсаалт" (`scores`) ба "нэг утга" (`score`)-ыг амархан ялгана.

**Нэг дүрэм:** хувьсагчийн нэр жагсаалтын нэрээс **өөр** байх ёстой. Ижил байвал юу болохыг 12-р хэсэгт (Түгээмэл алдаанууд) үзнэ.

## 5. Таслалаар хэд хэдэн утгыг нэг дор хэвлэх

Одоог хүртэл бид ихэвчлэн нэг удаад нэг утга хэвлэсэн: `print(score)`.

Гэхдээ заримдаа `Total: 255` шиг тайлбар үгийг утгатай хамт хэвлэхийг хүсдэг. Үүний тулд `print()` дотор **хэд хэдэн утгыг** **таслал** `,`-аар тусгаарлан бичиж болно.

```python
print(утга1, утга2, утга3)
```

Дүрэм:
1. Утгууд **нэг мөрөнд, дарааллаараа** хэвлэгдэнэ.
2. Python утгуудын хооронд **автоматаар нэг зай** тавьдаг. Өөрөө зай нэмэх шаардлагагүй.
3. Хашилт `" "` доторх бичвэр **байгаагаараа** хэвлэгдэнэ.
4. Хашилтын гаднах зүйл **эхлээд бодогдож**, дараа нь үр дүн хэвлэгдэнэ. Хэрэв хувьсагч бол түүн доторх утга хэвлэгдэнэ.

````python
# Хэд хэдэн тэмдэгт мөр
print("apple", "banana", "cherry")     # Утга бүрийн хооронд автоматаар нэг зай тавигдана
````

````python
# Хэд хэдэн тоо
print(10, 20, 30)
````

````python
# Тэмдэгт мөр ба тоо хамт
print("Total:", 255)                   # "Total:" байгаагаараа, дараа нь зай, дараа нь 255
````

````python
# Тэмдэгт мөр ба хувьсагч хамт
score = 85
print("Score:", score)                 # score бол хувьсагч тул түүн доторх утга (85) хэвлэгдэнэ
````

````python
# Тэмдэгт мөр ба тооцоолол хамт
price = 1000
print("Double:", price * 2)            # Эхлээд price * 2 бодогдоод (2000), дараа нь хэвлэгдэнэ
````

````python
# Хашилтын дотор ба гадна
price = 1000
print("price * 2")                     # Хашилтын дотор -> байгаагаараа хэвлэгдэнэ: price * 2
print(price * 2)                       # Хашилтын гадна -> бодогдоно: 2000
````

### Таслал `,` ба нэмэх `+`

Бид `+` нь тэмдэгт мөрүүдийг залгаж чаддагийг бас сурсан. Ялгаа нь юу вэ?

| | Таслал `,` | Нэмэх `+` |
|---|---|---|
| Утгуудын хоорондох зай | Автоматаар нэмэгдэнэ | Нэмэгдэхгүй. Өөрөө нэмэх хэрэгтэй |
| Тэмдэгт мөр + тоо | ✅ Ажиллана | ❌ Алдаа |

`+` нь зөвхөн **тэмдэгт мөрийг тэмдэгт мөртэй** залгаж чадна. Тэмдэгт мөрийг тоотой `+`-ээр залгах гэвэл Python алдаа гаргана. Таслалаар бол тэмдэгт мөр, тоог хольж болно.

````python
total = 255

print("Total:", total)                 # ✅ Таслал: тэмдэгт мөр ба тоо хамт -> ажиллана

# print("Total: " + total)             # ❌ Нэмэх: тэмдэгт мөр + тоо -> TypeError!
````

## 6. Утгуудыг нэмэх

for давталтын хамгийн түгээмэл ажлын нэг бол жагсаалтын бүх утгыг **нэмэх** юм. Жишээ нь: бүх онооны нийлбэр хэд вэ?

Арга нь нэг хувьсагчийг **хуримтлагдах нийлбэр** болгон ашиглах:

1. **Давталтын өмнө:** хувьсагч үүсгээд 0 гэж оноо. → `total = 0`
2. **Давталтын дотор:** одоогийн утгыг түүн дээр нэм. → `total = total + score`
3. **Давталтын дараа:** тэр хувьсагчид бүх утгын нийлбэр байна.

2-р алхмыг анхааралтай хараарай: `total = total + score`

- `=` нь "тэнцүү" гэсэн утгагүй. Энэ нь **"баруун талын утгыг зүүн талын хайрцагт хадгал"** гэсэн утгатай.
- Тиймээс Python эхлээд баруун талыг, `total + score`-ыг (хуучин нийлбэр дээр одоогийн оноог нэмсэн) бодоод,
- дараа нь тэр үр дүнг `total` руу буцааж хадгална.
- Энэ нь `age = age + 1`-тэй ижил санаа: хуучин утгыг аваад, ямар нэг зүйл нэмээд, нөгөө хайрцагтаа буцааж хийнэ.

````python
# Бүх оноог нэмэх
scores = [85, 92, 78, 90, 66]
total = 0                       # 1-р алхам: хуримтлагдах нийлбэр 0-ээс эхэлнэ (давталтын ӨМНӨ)

for score in scores:
    total = total + score       # 2-р алхам: хуучин нийлбэр + одоогийн оноо -> total руу буцааж хадгална

print("Total:", total)          # 3-р алхам: давталтын ДАРАА (догол мөргүй) үр дүнг нэг удаа хэвлэнэ
````

````python
# Давталт бүрт total хэрхэн өөрчлөгдөхийг харъя
# Дээрхтэй ижил код, гэхдээ давталтын ДОТОР (догол мөртэй) print() бий
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total = total + score
    print("score:", score, "total:", total)      # Догол мөртэй -> давталт бүрт хэвлэгдэнэ

# 1 дэх: 0 + 85 = 85
# 2 дахь: 85 + 92 = 177
# 3 дахь: 177 + 78 = 255
# 4 дэх: 255 + 90 = 345
# 5 дахь: 345 + 66 = 411
````

### Илүү богино бичих арга: `+=`

`total = total + score` маш олон хэрэглэгддэг тул Python-д илүү богино бичих арга бий:

```python
total += score
```

`+=` нь **"баруун талыг зүүн талын хувьсагч дээр нэмээд, үр дүнг түүнд буцааж хадгал"** гэсэн утгатай.

`total += score` ба `total = total + score` нь **яг ижил утгатай**. Аль нэгийг нь ашиглаж болно. Python програмистуудын олонх нь богино хэлбэрийг ашигладаг тул та үүнийг олон удаа харах болно.

````python
# Урт хэлбэр
scores = [85, 92, 78]
total = 0
for score in scores:
    total = total + score
print("Long form:", total)

# Богино хэлбэр
total = 0
for score in scores:
    total += score              # дараахтай ижил: total = total + score
print("Short form:", total)     # Үр дүн адилхан
````

````python
# Сагсны нийт үнэ
cart = [3500, 1200, 8000, 2300]
total_price = 0                 # Давталтын өмнө: 0-ээс эхэлнэ

for price in cart:
    total_price += price        # дараахтай ижил: total_price = total_price + price

print("Total price:", total_price)
````

### Дундаж

**Дундаж** гэдэг нь нийлбэрийг утгын тоонд хуваасан утга.

Утгын тоог яаж олох вэ? `len()`-ийг санаарай: жагсаалтад хэдэн утга байгааг хэлж өгдөг. `len([85, 92, 78])` нь `3`.

Тэгэхээр: **дундаж = нийлбэр / len(жагсаалт)**

Чухал: дунджийг нийлбэр бүрэн болсон үед, давталтын **дараа** бодоорой.

````python
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores:
    total += score              # дараахтай ижил: total = total + score

average = total / len(scores)   # Давталтын дараа: нийлбэр (411) / онооны тоо (5)
print("Total:", total)
print("Number of scores:", len(scores))
print("Average:", average)
````

## 7. Давталтын дотор if ашиглах

for давталтын **дотор** `if`-г хийж болно. Тэгвэл утга бүр нэг нэгээр шалгагдаж, **зөвхөн хүссэн утгууд дээрээ** ямар нэг зүйл хийх боломжтой.

`if` хэрхэн ажилладгийг санаарай:
- `if нөхцөл:` → доорх догол мөртэй код **зөвхөн нөхцөл True үед** ажиллана.
- `else:` → доорх догол мөртэй код нөхцөл False үед ажиллана.

`for`-ийн дотор `if` хийхэд догол мөр **хоёр шат** болно:

```python
for score in scores:          # зайгүй
    if score >= 60:           # 4 зай -> for-ийн дотор (оноо бүрд ажиллана)
        print(score)          # 8 зай -> if-ийн дотор (зөвхөн нөхцөл True үед ажиллана)
```

- `if` мөр 4 зайтай, учир нь энэ нь `for` блокт хамаарна.
- `print()` мөр 8 зайтай, учир нь энэ нь `if` блокт хамаарах бөгөөд тэр блок `for` блокийн дотор байна.

````python
# Зөвхөн тэнцсэн оноог (60 ба түүнээс дээш) хэвлэх
scores = [85, 42, 78, 55, 90]

for score in scores:            # Оноо бүрийг гаргана
    if score >= 60:             # 4 зай: энэ оноог шалгана
        print(score)            # 8 зай: 60 ба түүнээс дээш бол л хэвлэнэ
````

````python
# Аль оноо шалгагдаж, аль нь хэвлэгдсэнийг харъя
scores = [85, 42, 78, 55, 90]

for score in scores:
    print("Checking", score)    # 4 зай: БҮХ оноонд ажиллана
    if score >= 60:
        print("pass")           # 8 зай: зөвхөн score >= 60 үед ажиллана
````

````python
# if - else: оноо бүрд pass эсвэл fail гэж хэвлэх
scores = [85, 42, 78, 55, 90]

for score in scores:
    if score >= 60:
        print(score, "pass")    # Нөхцөл True үед ажиллана
    else:
        print(score, "fail")    # Нөхцөл False үед ажиллана
````

````python
# Зөвхөн тэгш тоонуудыг хэвлэх
# 2-т хуваахад үлдэгдэл 0 бол тэгш тоо
# % нь үлдэгдлийг өгдөг: 8 % 2 нь 0 (тэгш), 5 % 2 нь 1 (сондгой)
numbers = [3, 8, 5, 12, 7, 6]

for number in numbers:
    if number % 2 == 0:         # Үлдэгдэл 0 юу? -> тэгш
        print(number)
````

````python
# Давталтын дотор if - elif - else: оноо бүрд үнэлгээ өгөх
# Python нөхцөлүүдийг дээрээс доош шалгаж, True болсон ЭХНИЙХИЙГ л ажиллуулна
scores = [95, 82, 74, 58]

for score in scores:
    if score >= 90:
        print(score, "A")
    elif score >= 80:
        print(score, "B")
    elif score >= 70:
        print(score, "C")
    else:
        print(score, "F")
````

````python
# Зарим утгыг л нэмэх: if + хуримтлагдах нийлбэр
# Зөвхөн тэгш тоонуудыг нэмнэ
numbers = [3, 8, 5, 12, 7, 6]
even_total = 0                  # Давталтын өмнө: 0-ээс эхэлнэ

for number in numbers:
    if number % 2 == 0:         # Тоо тэгш үед л...
        even_total += number    # ...нэмнэ (дараахтай ижил: even_total = even_total + number)

print("Sum of even numbers:", even_total)    # 8 + 12 + 6 = 26
````

## 8. Тоолох

Хэдэн сурагч тэнцсэн бэ? Хэдэн хүн пиццаны төлөө санал өгсөн бэ?

**Тоолохдоо** хуримтлагдах нийлбэртэй ижил санааг ашиглана. Цорын ганц ялгаа: утгыг нэмэхийн оронд удаа бүр **1**-ийг нэмнэ.

1. **Давталтын өмнө:** `count = 0`
2. **Давталтын дотор:** нөхцөл True үед `count = count + 1` (эсвэл `count += 1`)
3. **Давталтын дараа:** `count`-д нөхцөл хэдэн удаа True болсон нь хадгалагдана

````python
# Хэдэн сурагч тэнцсэн бэ (60 ба түүнээс дээш)?
scores = [85, 42, 78, 55, 90]
pass_count = 0                  # Давталтын өмнө: 0-ээс эхэлнэ

for score in scores:
    if score >= 60:
        pass_count += 1         # дараахтай ижил: pass_count = pass_count + 1

print("Passed:", pass_count)    # 85, 78, 90 -> 3
````

````python
# Пицца хэдэн санал авсан бэ?
votes = ["pizza", "pasta", "pizza", "salad", "pizza", "pasta"]
pizza_count = 0

for vote in votes:
    if vote == "pizza":         # == нь хоёр утга ижил эсэхийг шалгана
        pizza_count += 1        # дараахтай ижил: pizza_count = pizza_count + 1

print("Pizza:", pizza_count)
````

````python
# Хоёр зүйлийг зэрэг тоолох
scores = [85, 42, 78, 55, 90]
pass_count = 0
fail_count = 0

for score in scores:
    if score >= 60:
        pass_count += 1
    else:
        fail_count += 1

print("Passed:", pass_count)
print("Failed:", fail_count)
````

````python
# 0-ээс доош температуртай өдрийг тоолох
temperatures = [3, -2, 5, -7, 0, -1]
below_zero = 0

for temperature in temperatures:
    if temperature < 0:
        below_zero += 1

print("Days below zero:", below_zero)
````

## 9. Хамгийн их ба хамгийн бага утгыг олох

Жагсаалтаас хамгийн өндөр оноог яаж олох вэ? Цаасан дээрх тоонуудаас хүн яаж олохыг бодоод үзээрэй:

1. Эхний тоог хараад **"одоогоор хамгийн том"** гэж санана.
2. Дараагийн тоог харна. Санасан тооноос том уу? Том бол оронд нь энийг санана.
3. Төгсгөл хүртэл ингэсээр байна. Эцэст нь санаж буй тоо бол хамгийн том нь.

Яг эдгээр алхмыг Python-оор бичиж болно. `highest` хувьсагч бол "бидний санаж буй тоо".

**Яагаад эхний утгаас (`scores[0]`) эхэлдэг вэ?** Учир нь харьцуулах ямар нэг зүйл хэрэгтэй. 0 индекс бол жагсаалтын эхний утга.

````python
# Хамгийн өндөр оноог олох
scores = [85, 92, 78, 95, 66]
highest = scores[0]             # 1-р алхам: эхний оноог (85) "одоогоор хамгийн том" гэж санана

for score in scores:
    if score > highest:         # 2-р алхам: энэ оноо санасан онооноос том уу?
        highest = score         #            том бол оронд нь энийг санана

print("Highest:", highest)      # 3-р алхам: эцэст нь санаж буй оноо
````

````python
# Алхам алхмаар харъя
scores = [85, 92, 78, 95, 66]
highest = scores[0]

for score in scores:
    if score > highest:
        highest = score
    print("score:", score, "highest:", highest)     # 4 зай: давталт бүрт ажиллана
````

````python
# Хамгийн бага утгыг ч ижил аргаар олно, зөвхөн >-ийн оронд < ашиглана
temperatures = [12, 8, 15, 3, 10]
lowest = temperatures[0]        # Эхний температураас (12) эхэлнэ

for temperature in temperatures:
    if temperature < lowest:    # Энэ нь санаж буйгаас бага уу?
        lowest = temperature

print("Lowest:", lowest)
````

````python
# Сагсан дахь хамгийн үнэтэй бараа
cart = [3500, 1200, 8000, 2300]
most_expensive = cart[0]

for price in cart:
    if price > most_expensive:
        most_expensive = price

print("Most expensive:", most_expensive)
````

## 10. append()-ээр шинэ жагсаалт үүсгэх

for давталтын дотор `append()` ашиглавал хуучин жагсаалтаас **шинэ жагсаалт** үүсгэж болно:

## 11. Давталтын дотор in ба break ашиглах

### in

`утга in жагсаалт` нь утга жагсаалтад байгаа эсэхийг шалгана. Үр дүн нь `True` эсвэл `False` тул `if`-ийн нөхцөл болгон ашиглаж болно.

```python
"milk" in ["bread", "milk"]     # True
"eggs" in ["bread", "milk"]     # False
```

**Анхаар:** `for fruit in fruits:` доторх `in` ба `if fruit in store:` доторх `in` адилхан харагдах боловч өөр ажил хийдэг.

| | Утга |
|---|---|
| `for fruit in fruits:` | fruits-аас утга бүрийг нэг нэгээр **гаргана** |
| `if fruit in store:` | fruit нь store-д байгаа эсэхийг **шалгана** (True / False) |

````python
# Миний худалдан авах жагсаалтын аль бараа дэлгүүрт байна вэ?
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]

for item in shopping_list:              # 1-р "in": бараа бүрийг гаргана
    if item in store:                   # 2-р "in": энэ бараа store-д байна уу?
        print(item, "yes")
    else:
        print(item, "no")
````

````python
# Өөр газраас авах шаардлагатай бараануудын жагсаалт үүсгэх
shopping_list = ["milk", "eggs", "bread", "cheese"]
store = ["bread", "milk", "apple", "rice"]
not_found = []

for item in shopping_list:
    if item not in store:               # not in: бараа store-д БАЙХГҮЙ үед True
        not_found.append(item)

print("Not found:", not_found)
````

### break

`break` нь **давталтыг шууд зогсооно**. Үлдсэн утгууд огт шалгагдахгүй бөгөөд Python давталтын дараах код руу шилжинэ.

Хайж байсан зүйлээ олсон, цааш хайх шаардлагагүй үед хэрэгтэй. `break` бараг үргэлж `if`-ийн дотор хэрэглэгддэг.

````python
# Эхний унасан оноог олмогц зогсоох
scores = [85, 92, 55, 78, 40]

for score in scores:
    if score < 60:
        print("Found a failing score:", score)
        break                           # Давталтыг яг энд зогсооно
    print(score, "is OK")

print("The loop is over.")              # 78 ба 40 огт шалгагдаагүй
````

````python
# Жагсаалтад Tom гэдэг хүн байна уу?
names = ["Bill", "Anna", "Tom", "Kate", "Mike"]

for name in names:
    print("Checking", name)
    if name == "Tom":
        print("Found Tom!")
        break                           # Kate ба Mike-ийг шалгах шаардлагагүй
````

Бусад буруу код **алдаа гаргахгүй ч буруу үр дүн өгдөг**. Python анхааруулдаггүй тул эдгээр алдаа бүр ч аюултай.

### Алдаа 3: Нэг л удаа ажиллах ёстой мөрийг догол мөртэй болгох

````python
# ❌ Буруу: нийлбэрийг төгсгөлд зөвхөн нэг удаа хэвлэхийг хүсэж байна
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score
    print("Total:", total)      # Догол мөртэй -> давталтын дотор -> 3 удаа хэвлэгдэнэ!
````

✅ **Зөв арга:** Нэг л удаа ажиллах ёстой мөрийн догол мөрийг арилгана. Тэгвэл тэр давталтын **гадна** байж, давталт дууссаны дараа л ажиллана.

````python
# ✅ Зөв
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score              # Давталтын дотор: нэмнэ
print("Total:", total)          # Догол мөргүй -> давталтын гадна -> нэг удаа хэвлэгдэнэ
````

### Алдаа 4: Жагсаалт ба давталтын хувьсагчид ижил нэр өгөх

🔍 **Юу буруу болсон бэ?**

- Гэхдээ давталт бүр `fruits` нэртэй хайрцагт утга хадгалдаг. Санаарай: хувьсагчид шинэ утга хадгалахад хуучин утга алга болдог.
- Тиймээс `fruits`-д байсан жагсаалт `"apple"`, дараа нь `"banana"`, дараа нь `"cherry"`-гээр солигдоно.
- Давталтын дараа `fruits`-д зөвхөн `"cherry"` үлдэнэ. Жагсаалт алга болсон.

✅ **Зөв арга:** Давталтын хувьсагчид **өөр** нэр өгнө. Жагсаалтын нэрийн ганц тооны хэлбэр хамгийн тохиромжтой: `fruits` → `fruit`.

````python
# ✅ Зөв
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:            # fruit (нэг утга) ба fruits (жагсаалт) нь өөр нэр
    print(fruit)

print(fruits)                   # Жагсаалт эсэн мэнд
````

### Алдаа 5: Давталтын хувьсагчийн оронд жагсаалтын нэрийг ашиглах

````python
# ❌ Буруу: жимс бүрийг хэвлэхийг хүсэж байна
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruits)               # fruits (s-тэй) -> бүх жагсаалт 3 удаа хэвлэгдэнэ
````

🔍 **Юу буруу болсон бэ?**

- Нэрнүүд бараг адилхан, зөвхөн нэг `s` үсгээр ялгаатай. Тийм учраас энэ алдааг гаргахад амархан.
- Давталт 3 удаа ажиллаж, удаа бүр бүх жагсаалтыг хэвлэнэ.

✅ **Зөв арга:** Давталтын дотор одоогийн утгыг авахын тулд **давталтын хувьсагч**-ийг (`fruit`) ашиглана.

````python
# ✅ Зөв
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)                # fruit (s-гүй) -> нэг удаад нэг утга
````

- for давталт тэмдэгт мөрөөр ч гүйж чадна. **Нэг удаад нэг үсэг** гаргадаг. Тиймээс 3 жимсний оронд 6 үсэг гарна.
- Алдаа гарахгүй тул будлиулж болно.

### Алдаа 7: Хуримтлагдах нийлбэрийг давталтын дотор 0 болгох

````python
# ❌ Буруу
scores = [85, 92, 78]

for score in scores:
    total = 0                   # Догол мөртэй -> давталтын дотор -> давталт бүрт ажиллана!
    total += score

print("Total:", total)          # 255 биш 78
````

🔍 **Юу буруу болсон бэ?**

- `total = 0` догол мөртэй тул давталтын дотор байна. Энэ нь **3 удаа** ажиллана.
- Давталт бүрт нийлбэр 0 болж, зөвхөн одоогийн оноо нэмэгдэнэ.
- 1 дэх: 0 + 85 = 85 → 2 дахь: **дахин 0**, 0 + 92 = 92 → 3 дахь: **дахин 0**, 0 + 78 = 78
- Тиймээс эцэст нь `total`-д зөвхөн сүүлийн оноо үлдэнэ.

✅ **Зөв арга:** `total = 0`-ийг давталтын **өмнө нэг удаа** (догол мөргүй) бичнэ.

````python
# ✅ Зөв
scores = [85, 92, 78]
total = 0                       # Давталтын өмнө -> зөвхөн нэг удаа ажиллана

for score in scores:
    total += score              # дараахтай ижил: total = total + score

print("Total:", total)          # 85 + 92 + 78 = 255
````

### Алдаа 8: Дунджийг давталтын дотор бодох

````python
# ❌ Буруу
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score
    average = total / len(scores)       # Догол мөртэй -> давталт бүрт бодогдоно
    print("Average:", average)
````

🔍 **Юу буруу болсон бэ?**

- Дунджийг бодохын тулд **бүх** утгын нийлбэр хэрэгтэй.
- Гэхдээ давталтын дотор нийлбэр хараахан бүрэн болоогүй.
- 1 дэх: 80 / 3 → 2 дахь: 170 / 3 → 3 дахь: 270 / 3. Зөвхөн сүүлийнх нь жинхэнэ дундаж.

✅ **Зөв арга:** Давталтын дотор зөвхөн нэмнэ. Нийлбэр бүрэн болсон үед, давталтын **дараа** дунджийг бодно.

````python
# ✅ Зөв
scores = [80, 90, 100]
total = 0

for score in scores:
    total += score                      # Давталтын дотор: зөвхөн нэмнэ

average = total / len(scores)           # Давталтын дараа: 270 / 3
print("Average:", average)
````

### Алдаа 9: Давталтын хувьсагчийг өөрчилж жагсаалтыг өөрчлөх гэж оролдох

````python
# ❌ Буруу: жагсаалтын оноо бүр дээр 5 урамшууллын оноо нэмэхийг хүсэж байна
scores = [85, 92, 78]

for score in scores:
    score = score + 5           # Оноог өөрчилж байгаа мэт харагдана...

print(scores)                   # ...гэхдээ жагсаалт [85, 92, 78] хэвээр!
````

🔍 **Юу буруу болсон бэ?**

- `score` бол **тусдаа хайрцаг**. Давталт бүрт жагсаалтаас нэг утга энэ хайрцаг руу **хуулагдана**.
- `score = score + 5` нь зөвхөн `score` хайрцаг доторхыг өөрчилнө. `scores` жагсаалтад огт хүрэхгүй.
- Энэ нь дэвтрээсээ тоог наалттай цаасан дээр хуулж бичээд, наалттай цаасан дээрх тоог засахтай адил. Дэвтэр хэвээрээ үлдэнэ.
- Алдаа гарахгүй тул анзаарахад хэцүү.

✅ **Зөв арга:** `append()`-ээр **шинэ жагсаалт** (10-р хэсэг) үүсгэж, өөрчилсөн утгуудаа түүнд хийнэ.

````python
# ✅ Зөв
scores = [85, 92, 78]
new_scores = []                         # Өөрчилсөн утгуудад зориулсан хоосон жагсаалт

for score in scores:
    new_scores.append(score + 5)        # Өөрчилсөн утгыг шинэ жагсаалтад нэмнэ

print(new_scores)                       # [90, 97, 83]
````

### Алдаа 10: Давталтын явцад жагсаалтаас утга устгах

````python
# ❌ Буруу: унасан оноо (60-аас доош) бүрийг устгахыг хүсэж байна
scores = [50, 40, 80, 30]

for score in scores:
    if score < 60:
        scores.remove(score)

print(scores)                   # [40, 80] -> 40 байсаар байна!
````

🔍 **Юу буруу болсон бэ?**

for давталт жагсаалтыг **байрлалаар нь** гүйдэг: 1-р утга, дараа нь 2-р утга, дараа нь 3-р утга ...
Гэтэл `remove()` нэг утгыг гаргахад түүний араас байсан бүх утга нэг байрлалаар **урагшилна**.

| Давталт | Давталтын харж буй байрлал | Тэр үеийн жагсаалт | Юу болох вэ |
|---|---|---|---|
| 1 дэх | 1-р байрлал | `[50, 40, 80, 30]` | 50 < 60 → 50-ыг устгана. Жагсаалт `[40, 80, 30]` болно |
| 2 дахь | 2-р байрлал | `[40, 80, 30]` | 2-р утга одоо **80**. 40 нь 1-р байрлал руу шилжсэн тул давталт түүнийг **алгассан**! |
| 3 дахь | 3-р байрлал | `[40, 80, 30]` | 30 < 60 → 30-ыг устгана. Жагсаалт `[40, 80]` болно |

- 40 хэзээ ч шалгагдаагүй тул жагсаалтад үлдсэн.
- Алдаа гарахгүй тул анзаарахад маш хэцүү.

✅ **Зөв арга:** Давталтаар гүйж байх үедээ тэр жагсаалтыг өөрчилж болохгүй. Оронд нь үлдээхийг хүссэн утгуудаар **шинэ жагсаалт** үүсгэнэ (10-р хэсэг).

````python
# ✅ Зөв
scores = [50, 40, 80, 30]
kept = []

for score in scores:
    if score >= 60:             # Зөвхөн тэнцсэн оноог үлдээнэ
        kept.append(score)

print(kept)                     # [80]
````

### Алдаа 12: "Одоогоор хамгийн том"-ыг 0-ээс эхлүүлэх

````python
# ❌ Буруу: өвлийн хамгийн өндөр температурыг олох
temperatures = [-5, -2, -8, -3]
highest = 0                     # 0-ээс эхэлнэ...

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # 0 -> гэхдээ 0 жагсаалтад ч байхгүй!
````

🔍 **Юу буруу болсон бэ?**

- Бүх температур 0-ээс доош. Тиймээс аль нь ч 0-ээс их биш.
- `temperature > highest` хэзээ ч True болохгүй бөгөөд `highest` төгсгөл хүртэл 0 хэвээр үлдэнэ.
- Хуримтлагдах нийлбэрийг 0-ээс эхлүүлдэг тул "0-ээс эхлэх ёстой" гэж бодоход амархан. Гэхдээ хамгийн их утгыг олох нь өөр ажил.

✅ **Зөв арга:** **Жагсаалтын эхний утгаас** (`temperatures[0]`) эхэлнэ. Тэгвэл эхлэх утга үргэлж жагсаалтаас өөрөөс нь ирнэ.

````python
# ✅ Зөв
temperatures = [-5, -2, -8, -3]
highest = temperatures[0]       # Эхний утгаас (-5) эхэлнэ

for temperature in temperatures:
    if temperature > highest:
        highest = temperature

print("Highest:", highest)      # -2
````

### Алдаа 13: Тэмдэгт мөр ба тоог +-ээр залгах

````python
# ❌ Буруу
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

# print("Total: " + total)      # TypeError: can only concatenate str (not "int") to str
````

🔍 **Юу буруу болсон бэ?**

- `+` нь **тэмдэгт мөрийг тэмдэгт мөртэй** залгаж чадна. Гэхдээ `total` бол **тоо** (int).
- Python тэмдэгт мөр ба тоог `+`-ээр яаж залгахаа мэдэхгүй тул алдаа гаргана.
- Алдааны мэдэгдэл нь: "тэмдэгт мөрийг (str) зөвхөн тэмдэгт мөртэй залгаж болно, тоотой (int) биш" гэсэн утгатай.

✅ **Зөв арга:** `print()` дотор **таслал** ашиглана (5-р хэсэг). Таслалаар тэмдэгт мөр ба тоог хольж болно.

````python
# ✅ Зөв
scores = [85, 92, 78]
total = 0

for score in scores:
    total += score

print("Total:", total)          # Таслал -> тэмдэгт мөр ба тоотой ажиллана
````
