## 1. ลิสต์ (list) คืออะไร?

ในบทที่ 1 เราได้เรียนว่าตัวแปรเหมือนกล่องที่เก็บค่าได้ **หนึ่ง** ค่า

แต่ถ้าเราต้องการเก็บคะแนนสอบของนักเรียน 5 คนล่ะ? ด้วยสิ่งที่เรารู้ตอนนี้ เราต้องใช้ตัวแปร 5 ตัว:

```python
score1 = 85
score2 = 92
score3 = 78
score4 = 90
score5 = 66
```

แค่นี้ก็เยอะแล้ว แล้วถ้ามีนักเรียน 30 คนล่ะ? หรือ 300 คน?

**ลิสต์ (list)** ช่วยแก้ปัญหานี้ ลิสต์คือตัวแปรตัวเดียวที่เก็บ **ค่าได้หลายค่าตามลำดับ**

ลองนึกภาพลิสต์เป็น **ตู้ล็อกเกอร์** ที่เรียงกันเป็นแถว:
- ตู้ทั้งแถวมีชื่อเดียว (ชื่อของลิสต์)
- ตู้แต่ละช่องเก็บค่าได้หนึ่งค่า
- ตู้แต่ละช่องมีหมายเลขกำกับ เพื่อให้หาเจอได้

วิธีสร้างลิสต์:
- ใส่ค่าต่าง ๆ ไว้ในวงเล็บเหลี่ยม `[ ]`
- คั่นค่าแต่ละค่าด้วยเครื่องหมายจุลภาค `,`

```python
scores = [85, 92, 78, 90, 66]
```

````python
# ลิสต์ของจำนวนเต็ม
# ตอนนี้ตัวแปรตัวเดียว (scores) เก็บค่าได้ 5 ค่า
scores = [85, 92, 78, 90, 66]
print(scores)           # print() แสดงลิสต์ทั้งหมด รวมถึงวงเล็บเหลี่ยมด้วย
````

````python
# ลิสต์เก็บข้อมูลชนิดไหนก็ได้ที่เราเรียนในบทที่ 1
fruits = ["apple", "banana", "cherry"]      # ลิสต์ของสตริง (str)
heights = [175.5, 162.3, 180.1]             # ลิสต์ของทศนิยม (float)
answers = [True, False, True, True]         # ลิสต์ของค่าบูลีน (bool)

print(fruits)
print(heights)
print(answers)
````

````python
# ข้อมูลต่างชนิดกันอยู่ในลิสต์เดียวกันได้
# ทำได้ แต่ส่วนใหญ่ลิสต์ที่เก็บค่าชนิดเดียวจะเข้าใจง่ายกว่า
student = ["Bill", 20, 175.5, True]         # ชื่อ อายุ ส่วนสูง เป็นนักเรียนหรือไม่
print(student)
````

````python
# ลิสต์ว่างเปล่าก็ได้ (ไม่มีค่าอะไรเลย)
# มีประโยชน์เมื่ออยากเริ่มจากว่างเปล่าแล้วค่อยเพิ่มค่าทีหลัง (เราจะเรียนวิธีเพิ่มในหัวข้อที่ 5)
empty_list = []
print(empty_list)
````

````python
# ลิสต์เป็นชนิดข้อมูลใหม่: list
# type() บอกชนิดข้อมูลของค่า
print(type(scores))     # <class 'list'> หมายถึง "นี่คือลิสต์"
print(type(85))         # <class 'int'> หมายถึง "นี่คือจำนวนเต็ม" (หนึ่งในค่าที่อยู่ในลิสต์)
````

## 2. อินเด็กซ์ (index): หาค่าจากตำแหน่ง

ค่าแต่ละค่าในลิสต์มีหมายเลขตำแหน่ง หมายเลขนี้เรียกว่า **อินเด็กซ์ (index)** ให้นึกภาพว่าเป็นหมายเลขที่เขียนไว้บนตู้แต่ละช่อง

สิ่งที่สำคัญที่สุดที่ต้องจำ: **Python เริ่มนับจาก 0 ไม่ใช่ 1**

ดังนั้นค่าแรกอยู่ที่อินเด็กซ์ 0 ค่าที่สองอยู่ที่อินเด็กซ์ 1 ค่าที่สามอยู่ที่อินเด็กซ์ 2 ไปเรื่อย ๆ (จำ `range(5)` ในบทที่ 3 ได้ไหม? มันก็เริ่มจาก 0 เหมือนกัน: 0, 1, 2, 3, 4)

```python
fruits = ["apple", "banana", "cherry", "grape", "melon"]
```

| ค่า | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| อินเด็กซ์ | 0 | 1 | 2 | 3 | 4 |
| ลำดับ (แบบที่คนนับ) | ที่ 1 | ที่ 2 | ที่ 3 | ที่ 4 | ที่ 5 |

วิธีจำง่าย ๆ: **อินเด็กซ์ = ลำดับ - 1** ค่าที่ 1 มีอินเด็กซ์ 0 ค่าที่ 5 มีอินเด็กซ์ 4

การดึงค่าออกมา ให้เขียนชื่อลิสต์ ตามด้วยอินเด็กซ์ในวงเล็บเหลี่ยม:

```python
fruits[0]     # "ค่าที่อินเด็กซ์ 0 ใน fruits"
```

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# ดึงค่าด้วยอินเด็กซ์
print(fruits[0])        # ค่าที่ 1 อยู่ที่อินเด็กซ์ 0 -> apple
print(fruits[1])        # ค่าที่ 2 อยู่ที่อินเด็กซ์ 1 -> banana
print(fruits[2])        # ค่าที่ 3 อยู่ที่อินเด็กซ์ 2 -> cherry
print(fruits[3])        # ค่าที่ 4 อยู่ที่อินเด็กซ์ 3 -> grape
print(fruits[4])        # ค่าที่ 5 (ตัวสุดท้าย) อยู่ที่อินเด็กซ์ 4 -> melon
````

````python
# แบบเดียวกันกับลิสต์ของตัวเลข
# ระวังอย่าสับสนระหว่างอินเด็กซ์ (ตำแหน่ง) กับค่า (สิ่งที่อยู่ข้างใน)
numbers = [10, 20, 30, 40]
#  อินเด็กซ์: 0   1   2   3

print(numbers[0])       # อินเด็กซ์ 0 -> ค่าคือ 10
print(numbers[2])       # อินเด็กซ์ 2 -> ค่าคือ 30 (ไม่ใช่ 2!)
print(numbers[3])       # อินเด็กซ์ 3 -> ค่าคือ 40
````

````python
# ค่าต่าง ๆ ไม่จำเป็นต้องเรียงลำดับแบบพิเศษใด ๆ
# อินเด็กซ์บอกแค่ว่าค่าอยู่ "ตรงไหน" ไม่ได้บอกว่าค่ามากหรือน้อย
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์: 0   1   2   3   4

print(scores[1])        # คะแนนของนักเรียนคนที่ 2 -> 92
print(scores[4])        # คะแนนของนักเรียนคนที่ 5 -> 66
````

````python
# ดึงค่าจากอินเด็กซ์เดิมกี่ครั้งก็ได้
# การดึงค่าเป็นแค่การ "อ่าน" ลิสต์ไม่เปลี่ยนแปลง
colors = ["red", "green", "blue"]

print(colors[1])        # green
print(colors[1])        # ยังเป็น green
print(colors)           # ลิสต์ยังเหมือนเดิม
````

````python
# ค่าที่ดึงออกมาจากลิสต์ใช้งานได้เหมือนตัวแปรธรรมดา
scores = [85, 92, 78, 90, 66]

first = scores[0]                  # เก็บคะแนนแรก (85) ไว้ในตัวแปรใหม่
print(first)

print(scores[0] + scores[1])       # 85 + 92 = 177 (นำมาคำนวณได้)
print(scores[1] > scores[2])       # 92 > 78 เป็น True (นำมาเปรียบเทียบได้ เหมือนในบทที่ 2)
````

````python
# สตริงที่ดึงออกมาจากลิสต์นำมาต่อกันด้วย + ได้ เหมือนในบทที่ 1
words = ["Hello", "nice", "to", "meet", "you"]

print(words[0] + ", " + words[1] + " " + words[2] + " " + words[3] + " " + words[4])
````

````python
# อินเด็กซ์เป็นตัวแปรก็ได้
# ตรงนี้ i เก็บค่า 2 ดังนั้น scores[i] ก็คือ scores[2]
scores = [85, 92, 78, 90, 66]

i = 2
print(scores[i])        # scores[2] -> 78

i = 0
print(scores[i])        # ตอนนี้ i เก็บค่า 0 ดังนั้น scores[0] -> 85
````

````python
# อินเด็กซ์เป็นการคำนวณก็ได้ด้วย
# Python จะคำนวณตัวเลขใน [ ] ก่อน แล้วจึงไปหาอินเด็กซ์นั้น
scores = [85, 92, 78, 90, 66]

print(scores[1 + 2])    # 1 + 2 = 3 -> scores[3] -> 90

i = 1
print(scores[i + 1])    # i + 1 = 2 -> scores[2] -> 78
````

### อินเด็กซ์ติดลบ: นับจากท้าย

ตอนนี้เรารู้จักอินเด็กซ์แบบปกติ (อินเด็กซ์บวก) แล้ว มาดูอีกวิธีหนึ่งในการหาค่ากัน

Python ยังให้เรา **นับย้อนจากท้าย** ลิสต์ได้ด้วยตัวเลขติดลบ

- `-1` คือค่า **ตัวสุดท้าย**
- `-2` คือค่ารองสุดท้าย
- `-3` คือค่าที่สามนับจากท้าย ไปเรื่อย ๆ

| ค่า | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| อินเด็กซ์ | 0 | 1 | 2 | 3 | 4 |
| อินเด็กซ์ติดลบ | -5 | -4 | -3 | -2 | -1 |

สังเกตว่าอินเด็กซ์ติดลบเริ่มจาก **-1** ไม่ใช่ -0 (ไม่มี -0 เพราะ -0 ก็คือ 0 ซึ่งเป็นค่าแรกนั่นเอง)

**ทำไมต้องใช้อินเด็กซ์ติดลบ?**

เหตุผลหลักคือเพื่อดึง **ค่าตัวสุดท้ายได้อย่างรวดเร็ว**

ถ้าใช้อินเด็กซ์บวก เราต้องรู้ก่อนว่าลิสต์มีค่ากี่ค่า จึงจะดึงค่าตัวสุดท้ายได้ ถ้าลิสต์มี 5 ค่า อินเด็กซ์สุดท้ายคือ 4 ถ้ามี 100 ค่า อินเด็กซ์สุดท้ายคือ 99 และถ้าเพิ่มหรือลบค่า อินเด็กซ์สุดท้ายก็เปลี่ยนตามไปด้วย

ถ้าใช้อินเด็กซ์ติดลบ ไม่ต้องนับอะไรเลย **`-1` คือค่าตัวสุดท้ายเสมอ** ไม่ว่าลิสต์จะยาวแค่ไหน

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# อินเด็กซ์ติดลบ: นับจากท้าย
print(fruits[-1])       # -1 คือค่าตัวสุดท้ายเสมอ -> melon
print(fruits[-2])       # -2 คือค่ารองสุดท้าย -> grape
print(fruits[-3])       # -3 คือค่าที่สามนับจากท้าย -> cherry
````

````python
# อินเด็กซ์บวกกับอินเด็กซ์ติดลบชี้ไปที่ค่าเดียวกันได้
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[4])        # ค่าตัวสุดท้าย ใช้อินเด็กซ์บวก -> melon
print(fruits[-1])       # ค่าตัวสุดท้าย ใช้อินเด็กซ์ติดลบ -> melon (เหมือนกัน!)

print(fruits[0])        # ค่าแรก ใช้อินเด็กซ์บวก -> apple
print(fruits[-5])       # ค่าแรก ใช้อินเด็กซ์ติดลบ -> apple (เหมือนกัน!)
````

````python
# ทำไม -1 จึงมีประโยชน์: ใช้ได้กับลิสต์ "ทุกลิสต์" ไม่ว่าจะยาวเท่าไร
# ถ้าใช้อินเด็กซ์บวก อินเด็กซ์สุดท้ายของแต่ละลิสต์จะไม่เท่ากัน (2, 6, 0 ...)
# ถ้าใช้ -1 จะเหมือนเดิมเสมอ
short_list = ["a", "b", "c"]
long_list = [10, 20, 30, 40, 50, 60, 70]
one_item = ["only"]

print(short_list[-1])   # c       (ถ้าใช้อินเด็กซ์บวกจะเป็น 2)
print(long_list[-1])    # 70      (ถ้าใช้อินเด็กซ์บวกจะเป็น 6)
print(one_item[-1])     # only    (ถ้าใช้อินเด็กซ์บวกจะเป็น 0)
````

## 3. การเปลี่ยนค่าและ len()

### การเปลี่ยนค่า

ลิสต์ไม่ได้ตายตัว เราสามารถ **เปลี่ยน** ค่าข้างในได้ทุกเมื่อ เหมือนเอาของใหม่ใส่ลงไปในตู้ช่องหนึ่ง:

```python
ชื่อลิสต์[อินเด็กซ์] = ค่าใหม่
```

ทำงานเหมือนกับการเก็บค่าในตัวแปรทุกประการ (`=` หมายถึง "เก็บค่า") แต่จะแทนที่เฉพาะค่าในตำแหน่งนั้นตำแหน่งเดียว ค่าอื่น ๆ ยังเหมือนเดิม

````python
# เปลี่ยนค่าที่อินเด็กซ์ที่กำหนด
fruits = ["apple", "banana", "cherry"]
print(fruits)

fruits[1] = "orange"        # แทนที่ค่าที่อินเด็กซ์ 1 ("banana") ด้วย "orange"
print(fruits)               # เปลี่ยนแค่อินเด็กซ์ 1 ส่วน "apple" กับ "cherry" ยังเหมือนเดิม
````

````python
# อินเด็กซ์ติดลบก็ใช้เปลี่ยนค่าได้เช่นกัน
fruits = ["apple", "banana", "cherry"]

fruits[-1] = "grape"        # แทนที่ค่าตัวสุดท้าย ("cherry") ด้วย "grape"
print(fruits)
````

````python
# เปลี่ยนค่าโดยใช้ค่าเดิมของมัน
# แนวคิดเดียวกับ age = age + 1 ในบทที่ 1
scores = [85, 92, 78]

scores[2] = scores[2] + 5   # นักเรียนคนที่สามได้คะแนนพิเศษ 5 คะแนน (78 + 5 = 83)
print(scores)
````

### len(): ลิสต์มีค่ากี่ค่า?

`len()` บอกว่าลิสต์มีค่าอยู่กี่ค่า คำว่า `len` ย่อมาจาก **length** (ความยาว)

```python
len(ชื่อลิสต์)
```

ใส่ลิสต์ไว้ในวงเล็บ แล้ว `len()` จะให้ตัวเลขกลับมา คือจำนวนค่าในลิสต์นั้น

````python
# len(): นับว่าลิสต์มีค่ากี่ค่า
fruits = ["apple", "banana", "cherry", "grape", "melon"]
print(len(fruits))          # มี 5 ค่า จึงได้ 5
````

````python
# ลิสต์ที่ยาวไม่เท่ากัน
three = ["a", "b", "c"]
five = [1, 2, 3, 4, 5]
one = ["only"]
empty_list = []

print(len(three))           # 3 ค่า -> 3
print(len(five))            # 5 ค่า -> 5
print(len(one))             # 1 ค่า -> 1
print(len(empty_list))      # ลิสต์ว่างมี 0 ค่า -> 0
````

````python
# len() ทำงานเหมือนกันกับลิสต์ทุกชนิด
names = ["Bill", "Anna", "Tom", "Kate"]          # ลิสต์ของสตริง
numbers = [3, 7, 1, 9, 4, 6]                     # ลิสต์ของจำนวนเต็ม
answers = [True, False, True]                    # ลิสต์ของค่าบูลีน

print(len(names))           # 4
print(len(numbers))         # 6
print(len(answers))         # 3
````

````python
# len() นับว่ามีค่า "กี่ค่า" ไม่ได้นำค่ามาบวกกัน
# ค่าแต่ละค่ามีขนาดใหญ่ แต่มีแค่ 3 ค่า
prices = [1000, 2500, 3000]
print(len(prices))          # 3 (ไม่ใช่ 6500!)
````

````python
# ผลลัพธ์ของ len() เป็นแค่ตัวเลข (int) จึงเก็บไว้ในตัวแปรได้
students = ["Bill", "Anna", "Tom", "Kate", "Mike"]

count = len(students)       # เก็บ 5 ไว้ในตัวแปร count
print(count)
print("There are", count, "students.")
````

````python
# len() ใช้กับสตริงได้ด้วย: มันจะนับจำนวนตัวอักษร
# (ช่องว่างก็นับเป็นหนึ่งตัวอักษร)
print(len("apple"))         # a, p, p, l, e -> 5
print(len("hi there"))      # h, i, (ช่องว่าง), t, h, e, r, e -> 8
````

### อินเด็กซ์สุดท้ายคือ len() - 1

ตอนนี้เราเชื่อมโยง `len()` กับอินเด็กซ์ได้แล้ว

ถ้าลิสต์มี 5 ค่า `len()` จะได้ 5 แต่อินเด็กซ์สุดท้ายคือ **4** เพราะอินเด็กซ์เริ่มจาก 0 **อินเด็กซ์สุดท้ายจึงเป็น `len() - 1` เสมอ**

ดังนั้นการดึงค่าตัวสุดท้ายมี 2 วิธี:
- `fruits[len(fruits) - 1]`: นับจำนวนค่า ลบ 1 แล้วใช้อินเด็กซ์นั้น
- `fruits[-1]`: ใช้ -1 ไปเลย

ทั้งสองวิธีได้ค่าเดียวกัน แต่ `-1` สั้นและง่ายกว่ามาก นี่คือเหตุผลที่อินเด็กซ์ติดลบมีประโยชน์

````python
# อินเด็กซ์สุดท้ายคือ len() - 1
fruits = ["apple", "banana", "cherry", "grape", "melon"]

last_index = len(fruits) - 1       # 5 - 1 = 4
print(last_index)
print(fruits[last_index])          # fruits[4] -> melon
````

````python
# เปรียบเทียบสองวิธีในการดึงค่าตัวสุดท้าย
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[len(fruits) - 1])     # วิธียาว: len(fruits) คือ 5, 5 - 1 = 4, fruits[4] -> melon
print(fruits[-1])                  # วิธีสั้น: -1 คือค่าตัวสุดท้าย -> melon (ผลลัพธ์เหมือนกัน!)
````

## 4. ตรวจสอบว่ามีค่านั้นอยู่ในลิสต์หรือไม่ (in)

บางครั้งเราอยากรู้ว่า **"ค่านี้อยู่ในลิสต์ไหม?"**

เราใช้ `in` สำหรับเรื่องนี้

```python
ค่า in ชื่อลิสต์
```

- ถ้าค่านั้นอยู่ในลิสต์ ผลลัพธ์คือ `True`
- ถ้าไม่อยู่ ผลลัพธ์คือ `False`

ผลลัพธ์เป็น `True` หรือ `False` เหมือนตัวดำเนินการเปรียบเทียบในบทที่ 2 แปลว่านำไปใช้เป็นเงื่อนไขของ `if` ได้ทันที!

ถ้าต้องการตรวจว่าค่า **ไม่** อยู่ในลิสต์ ให้ใช้ `not in`

**หมายเหตุ:** เราเคยเห็นคำว่า `in` ในบทที่ 3 มาแล้ว (`for i in range(5)`) `in` ตัวนั้นเป็นส่วนหนึ่งของลูป `for` ส่วน `in` ที่เรียนตรงนี้ต่างกัน: มัน **ตรวจสอบ** ว่าค่าอยู่ในลิสต์หรือไม่ แล้วให้ผลเป็น True หรือ False

````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)       # มี "banana" อยู่ในลิสต์ -> True
print("grape" in fruits)        # ไม่มี "grape" อยู่ในลิสต์ -> False
````

````python
# not in: ตรงข้ามกับ in
print("grape" not in fruits)    # "grape" ไม่อยู่ในลิสต์ -> True
print("apple" not in fruits)    # "apple" อยู่ในลิสต์ "not in" จึงเป็น False
````

````python
# ใช้ in เป็นเงื่อนไขของ if (บทที่ 2)
fruits = ["apple", "banana", "cherry"]
wanted = "banana"

if wanted in fruits:                    # "banana" in fruits เป็น True
    print("Yes, we have it!")
else:
    print("Sorry, it's sold out.")
````

````python
# ตัวอย่างจากชีวิตจริง: ตรวจรายชื่อแขก
guests = ["Bill", "Anna", "Tom"]
name = "Kate"

if name in guests:                      # "Kate" in guests เป็น False
    print("Welcome!")
else:
    print("Sorry, you are not on the list.")    # บรรทัดนี้จึงทำงาน
````

````python
# in ตรวจค่าแบบตรงทุกตัวอักษร ตัวพิมพ์ใหญ่กับตัวพิมพ์เล็กถือว่าต่างกัน
fruits = ["apple", "banana", "cherry"]
print("Apple" in fruits)        # "Apple" (A ตัวใหญ่) กับ "apple" เป็นสตริงคนละตัว -> False
````

## 5. การเพิ่มและลบค่า

ลิสต์ยาวขึ้นและสั้นลงได้ เราเพิ่มค่าใหม่และลบค่าที่ไม่ต้องการออกได้

| วิธีเขียน | ทำอะไร |
|---|---|
| `ชื่อลิสต์.append(ค่า)` | เพิ่มค่าไว้ **ท้าย** ลิสต์ |
| `ชื่อลิสต์.insert(อินเด็กซ์, ค่า)` | เพิ่มค่าไว้ใน **ตำแหน่งที่เลือก** |
| `ชื่อลิสต์.remove(ค่า)` | ลบค่าที่ระบุ (ตัวแรกที่เจอ) |
| `ชื่อลิสต์.pop()` | ดึงค่า **ตัวสุดท้าย** ออก |
| `ชื่อลิสต์.pop(อินเด็กซ์)` | ดึงค่าใน **ตำแหน่งที่เลือก** ออก |

จุด `.` หมายถึง "ทำสิ่งนี้กับลิสต์นี้" ตัวอย่างเช่น `fruits.append("grape")` หมายถึง "เพิ่ม "grape" เข้าไปในลิสต์ fruits" คำสั่งที่เขียนตามหลังจุดแบบนี้เรียกว่า **เมธอด (method)**

เมธอดเหล่านี้เปลี่ยนลิสต์ตัวมันเองโดยตรง ไม่ต้องเขียน `fruits = ...`

````python
# append(): เพิ่มค่าไว้ท้ายลิสต์
fruits = ["apple", "banana"]
print(fruits)

fruits.append("cherry")         # เพิ่ม "cherry" ไว้ท้ายลิสต์
print(fruits)

fruits.append("grape")          # เพิ่ม "grape" ไว้ท้ายลิสต์อีกครั้ง
print(fruits)
print(len(fruits))              # ลิสต์ยาวขึ้นจาก 2 เป็น 4 ค่า
````

````python
# เริ่มจากลิสต์ว่าง แล้วเพิ่มค่าทีละค่า
# เป็นรูปแบบที่ใช้กันบ่อยมาก
shopping = []                   # เริ่มด้วยลิสต์ว่าง
shopping.append("milk")
shopping.append("eggs")
shopping.append("bread")
print(shopping)
````

````python
# insert(): เพิ่มค่าในตำแหน่งที่กำหนด
# insert(อินเด็กซ์, ค่า) -> ค่าใหม่จะไปอยู่ที่อินเด็กซ์นั้น และค่าที่อยู่ถัดไปจะถูกเลื่อนไปข้างหลังหนึ่งตำแหน่ง
fruits = ["apple", "banana", "cherry"]

fruits.insert(1, "orange")      # ใส่ "orange" ไว้ที่อินเด็กซ์ 1
print(fruits)                   # "banana" กับ "cherry" ถูกเลื่อนไปข้างหลังคนละหนึ่งตำแหน่ง

fruits.insert(0, "melon")       # ใส่ "melon" ไว้ที่อินเด็กซ์ 0 (หน้าสุด)
print(fruits)
````

````python
# remove(): ลบค่าโดยระบุตัวค่าเอง
fruits = ["apple", "banana", "cherry"]

fruits.remove("banana")         # หา "banana" แล้วลบออก
print(fruits)                   # ค่าที่อยู่ถัดไปจะเลื่อนขึ้นมาเติมช่องว่าง
````

````python
# ถ้ามีค่าเดียวกันหลายตัว remove() จะลบแค่ "ตัวแรก" เท่านั้น
numbers = [1, 2, 3, 2, 1]

numbers.remove(2)               # ลบแค่เลข 2 ตัวแรก (ที่อินเด็กซ์ 1)
print(numbers)                  # เลข 2 ตัวที่สองยังอยู่
````

````python
# pop(): ดึงค่าออกตามตำแหน่ง
# ถ้าไม่ใส่อินเด็กซ์ pop() จะดึงค่า "ตัวสุดท้าย" ออก
fruits = ["apple", "banana", "cherry"]

fruits.pop()                    # ดึงค่าตัวสุดท้าย ("cherry") ออก
print(fruits)
````

````python
# pop(อินเด็กซ์): ดึงค่าที่อินเด็กซ์นั้นออก
fruits = ["apple", "banana", "cherry"]

fruits.pop(0)                   # ดึงค่าที่อินเด็กซ์ 0 ("apple") ออก
print(fruits)
````

````python
# pop() ยัง "คืน" ค่าที่ดึงออกมาให้ด้วย จึงเก็บไว้ในตัวแปรได้
# remove() แค่ลบอย่างเดียว ส่วน pop() ลบ "และ" ส่งค่านั้นให้เรา
tasks = ["wash dishes", "do homework", "clean room"]

done = tasks.pop(0)             # ดึงงานแรกออกมา แล้วเก็บไว้ใน done
print("Finished:", done)
print("Still to do:", tasks)
````

````python
# ดู remove() กับ pop() แบบสรุป
# remove(ค่า):      "ลบค่านี้"                 -> เรารู้ค่า
# pop(อินเด็กซ์):   "ลบสิ่งที่อยู่ในตำแหน่งนี้"  -> เรารู้ตำแหน่ง
colors = ["red", "green", "blue"]

colors.remove("green")          # ลบตามค่า
print(colors)

colors.pop(0)                   # ลบตามตำแหน่ง (ตอนนี้อินเด็กซ์ 0 คือ "red")
print(colors)
````

## 6. ข้อผิดพลาดที่พบบ่อย

นี่คือข้อผิดพลาดที่เกือบทุกคนเคยทำตอนเรียนเรื่องลิสต์ ถ้าเจอ error ให้ตรวจสอบข้อเหล่านี้ก่อน

1. **ลืมว่าอินเด็กซ์เริ่มจาก 0**: `fruits[1]` คือค่า **ที่สอง** ไม่ใช่ค่าแรก
2. **ใช้อินเด็กซ์ที่ไม่มีอยู่**: ถ้าลิสต์มี 3 ค่า อินเด็กซ์คือ 0, 1, 2 `fruits[3]` จะเกิด `IndexError: list index out of range` ("อินเด็กซ์อยู่นอกขอบเขตของลิสต์")
3. **พยายามเพิ่มค่าด้วยอินเด็กซ์ใหม่**: `fruits[3] = "grape"` กับลิสต์ที่มี 3 ค่า **ไม่ได้** เพิ่มค่า แต่จะเกิด `IndexError: list assignment index out of range` ("เก็บค่าไว้ที่อินเด็กซ์นอกลิสต์ไม่ได้") ถ้าต้องการเพิ่ม ให้ใช้ `append()`
4. **ลบค่าที่ไม่มีอยู่ในลิสต์**: `fruits.remove("grape")` จะเกิด `ValueError: list.remove(x): x not in list` ("ไม่มีค่านี้ในลิสต์") ให้ตรวจด้วย `in` ก่อน
5. **สับสนระหว่าง `( )` กับ `[ ]`**: ใช้ `[ ]` สำหรับสร้างลิสต์และดึงค่าด้วยอินเด็กซ์ (`fruits[0]`) ใช้ `( )` สำหรับคำสั่ง เช่น `len(fruits)` และ `fruits.append("grape")`

เซลล์ข้างล่างแสดงให้เห็นว่าข้อผิดพลาดแต่ละแบบเป็นอย่างไร โค้ดที่ผิดถูกเขียนไว้เป็นคอมเมนต์ เพื่อไม่ให้ notebook พัง ลองลบ `#` ออกจากบรรทัดโค้ดเพื่อดูข้อความ error จริง แล้วค่อยใส่ `#` กลับเหมือนเดิม

````python
# ข้อผิดพลาด 1: ลืมว่าอินเด็กซ์เริ่มจาก 0
fruits = ["apple", "banana", "cherry"]

print(fruits[1])        # หลายคนคิดว่าจะได้ "apple" แต่นี่คือค่า "ที่สอง" -> banana
print(fruits[0])        # ถูกต้อง: ค่าแรกอยู่ที่อินเด็กซ์ 0 -> apple
````

````python
# ข้อผิดพลาด 2: ใช้อินเด็กซ์ที่ไม่มีอยู่
fruits = ["apple", "banana", "cherry"]      # 3 ค่า -> อินเด็กซ์คือ 0, 1, 2

# print(fruits[3])      # IndexError: list index out of range -> ไม่มีอินเด็กซ์ 3!

print(fruits[2])                    # ถูกต้อง: อินเด็กซ์สุดท้ายคือ 2
print(fruits[len(fruits) - 1])      # ถูกต้องเช่นกัน: len(fruits) - 1 = 3 - 1 = 2
print(fruits[-1])                   # ถูกต้องเช่นกัน: -1 คือค่าตัวสุดท้ายเสมอ
````

````python
# ข้อผิดพลาด 3: พยายามเพิ่มค่าด้วยอินเด็กซ์ใหม่
fruits = ["apple", "banana", "cherry"]

# fruits[3] = "grape"   # IndexError: list assignment index out of range -> "เปลี่ยน" ค่าได้เฉพาะอินเด็กซ์ที่มีอยู่แล้วเท่านั้น

fruits.append("grape")  # ถูกต้อง: ใช้ append() เพื่อเพิ่มค่าใหม่ไว้ท้ายลิสต์
print(fruits)
````

````python
# ข้อผิดพลาด 4: ลบค่าที่ไม่มีอยู่ในลิสต์
fruits = ["apple", "banana", "cherry"]

# fruits.remove("grape")    # ValueError: list.remove(x): x not in list -> ไม่มี "grape" ในลิสต์!

# ถูกต้อง: ตรวจด้วย in ก่อน (หัวข้อที่ 4) แล้วลบเฉพาะเมื่อมีค่านั้นอยู่
if "grape" in fruits:
    fruits.remove("grape")
else:
    print("grape is not in the list, so nothing was removed.")
````

````python
# ข้อผิดพลาด 5: สับสนระหว่าง ( ) กับ [ ]
fruits = ["apple", "banana", "cherry"]

# print(fruits(0))          # TypeError: 'list' object is not callable -> ( ) ใช้กับคำสั่ง ไม่ใช่กับอินเด็กซ์
# fruits.append["grape"]    # TypeError -> append เป็นคำสั่ง จึงต้องใช้ ( ) ไม่ใช่ [ ]

print(fruits[0])            # ถูกต้อง: ใช้ [ ] ดึงค่าด้วยอินเด็กซ์
fruits.append("grape")      # ถูกต้อง: ใช้ ( ) กับคำสั่ง
print(fruits)
````
