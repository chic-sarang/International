## 1. Зачем нужен словарь?

Начнём с того, что мы уже знаем: со **списка**.

Список хранит много значений по порядку, внутри квадратных скобок `[ ]`. Значение мы получаем по его **индексу** (номеру позиции), и индекс начинается с **0**.

Теперь представьте, что нужно сохранить оценки за экзамен 3 учеников: Anna, Tom и Bill. Со списками нам понадобятся **два** списка: один для имён, другой для оценок.

```python
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]
```

Два списка связаны между собой позицией:

| Индекс | 0 | 1 | 2 |
|---|---|---|---|
| `names` | "Anna" | "Tom" | "Bill" |
| `scores` | 85 | 92 | 78 |

Значит, `names[1]` - это Tom, а `scores[1]` - оценка Tom.

**Но у этого способа есть проблемы:**

- Чтобы найти оценку Tom, нужно сначала **помнить**, что Tom стоит под индексом 1. Python не может найти её по имени "Tom".
- Если добавить или удалить имя в одном списке и забыть сделать то же самое в другом, имена и оценки **перепутаются**.

Разве не было бы проще просто спросить: **"Какая оценка у Tom?"**

Именно это и делает словарь.

````python
# С двумя списками имя и оценка хранятся в разных местах
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

# Чтобы получить оценку Tom, нужно знать, что Tom стоит под индексом 1
print(names[1])         # Tom
print(scores[1])        # 92 -> оценка Tom
````

````python
# Проблема: если один список изменился, а другой нет, они перепутываются
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

names.remove("Anna")    # Мы удалили Anna из names, но забыли удалить 85 из scores

print(names[0])         # Tom
print(scores[0])        # 85 -> Это оценка Anna, а не Tom! Теперь они не совпадают
````

## 2. Что такое словарь?

Вспомните настоящий словарь - книгу, по которой учат слова.

- Вы ищете **слово** (например, "apple").
- Находите его **значение** ("круглый фрукт").

Не нужно знать, на какой странице это слово. Вы просто ищете слово, а значение стоит прямо рядом с ним.

**Словарь** в Python работает так же. Он хранит значения **парами**:

| Слово в Python | Что значит | В настоящем словаре | В нашем примере |
|---|---|---|---|
| **ключ (key)** | имя, которое вы ищете | слово | имя ученика ("Tom") |
| **значение (value)** | то, что вы получаете | толкование | оценка (92) |
| **пара (pair)** | ключ и его значение вместе | слово и его толкование | "Tom" и 92 |

Можно представить это и как **шкафчики с табличками с именами**:
- В **списке** у каждого шкафчика есть **номер** (0, 1, 2 ...).
- В **словаре** у каждого шкафчика есть **табличка с именем** ("Anna", "Tom", "Bill" ...).

Чтобы открыть шкафчик, вы используете табличку с именем, а не номер.

## 3. Создание словаря

Начнём с самого маленького словаря: всего **одна пара**.

```python
scores = {"Tom": 92}
```

Разберём каждую часть:

| Часть | Что значит |
|---|---|
| `{` | словарь начинается (это **фигурная скобка**, а не квадратная) |
| `"Tom"` | **ключ** (здесь это строка, поэтому в кавычках) |
| `:` | **двоеточие**. Оно связывает ключ с его значением |
| `92` | **значение** |
| `}` | словарь заканчивается |

Помните: `=` означает **"сохрани значение справа в коробку слева"**. Поэтому эта строка сохраняет весь словарь в переменную `scores`.

````python
# Словарь из одной пары
scores = {"Tom": 92}

print(scores)           # {'Tom': 92}
````

### Больше одной пары

Чтобы положить в словарь несколько пар, разделите их **запятыми** `,`. Так же, как в списке значения разделяются запятыми.

```python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
```

- `"Anna": 85` - одна пара
- `"Tom": 92` - одна пара
- `"Bill": 78` - одна пара

Значит, в этом словаре **3 пары**.

````python
# Словарь из двух пар
scores = {"Anna": 85, "Tom": 92}
print(scores)
````

````python
# Словарь из трёх пар
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)
````

### По одной паре на строку

Когда пар много, одну длинную строку трудно читать. Можно писать **по одной паре на строку**. Python читает это абсолютно так же. Не забывайте запятую в конце каждой пары.

````python
# Тот же словарь, по одной паре на строку (так легче читать)
scores = {
    "Anna": 85,
    "Tom": 92,
    "Bill": 78,
}
print(scores)
````

### Сравните: список и словарь

| | Список | Словарь |
|---|---|---|
| Скобки | квадратные `[ ]` | фигурные `{ }` |
| Что внутри | только значения | **пары** `ключ: значение` |
| Пример | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92, "Bill": 78}` |

````python
# Те же оценки в виде списка и в виде словаря
score_list = [85, 92, 78]
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list)       # Только оценки. Мы не знаем, чьи они
print(score_dict)       # Рядом с каждой оценкой есть имя
````

````python
# Словарь - это новый тип данных: dict (сокращение от "dictionary")
# type() показывает, какого вида значение
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(type(scores))     # <class 'dict'>
````

## 4. Получение значения по ключу

Чтобы получить значение из словаря, напишите имя словаря и **ключ** в квадратных скобках `[ ]`:

```python
scores["Tom"]      # "значение ключа 'Tom' в scores"
```

Это очень похоже на список. Единственная разница - то, что внутри квадратных скобок:

| | Список | Словарь |
|---|---|---|
| Что пишем в `[ ]` | **индекс** (номер позиции) | **ключ** (имя) |
| Пример | `score_list[1]` | `score_dict["Tom"]` |
| Значит | "значение на позиции 1" | "значение, у которого ключ 'Tom'" |

**Важно:** хотя для **создания** словаря мы используем фигурные скобки `{ }`, чтобы **получить** из него значение, мы используем квадратные скобки `[ ]`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # Значение ключа "Tom" -> 92
print(scores["Anna"])       # Значение ключа "Anna" -> 85
print(scores["Bill"])       # Значение ключа "Bill" -> 78
````

````python
# Сравните: получаем оценку Tom из списка и из словаря
score_list = [85, 92, 78]                           # Anna, Tom, Bill
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list[1])        # Список: нужно помнить, что Tom под индексом 1
print(score_dict["Tom"])    # Словарь: просто используем имя "Tom"
````

````python
# Ещё примеры
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

print(capitals["Korea"])    # Seoul
print(capitals["France"])   # Paris
````

````python
english = {"apple": "a round fruit", "dog": "an animal that barks"}

print(english["dog"])       # Ищем слово "dog" -> его значение
````

### Значение можно использовать как обычную переменную

Значение, полученное из словаря, - это просто обычное значение. Его можно сохранить в переменную, считать с ним или сравнивать.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

tom_score = scores["Tom"]                   # Сохраняем оценку Tom в новую переменную
print(tom_score)

print(scores["Anna"] + scores["Tom"])       # 85 + 92 = 177 (можно считать)
print(scores["Tom"] > scores["Bill"])       # 92 > 78 - это True (можно сравнивать)
````

````python
# Используем значение в if (if выполняет код с отступом, только когда условие равно True)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

if scores["Bill"] >= 80:
    print("Bill passed with a high score!")
else:
    print("Bill needs to study more.")
````

### Ключ может быть переменной

Вместо того чтобы писать ключ напрямую, можно сначала сохранить ключ в переменную. Тогда в `[ ]` пишется переменная, **без кавычек**.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Anna"               # В переменной name хранится строка "Anna"
print(scores[name])         # scores[name] - то же самое, что scores["Anna"] -> 85

name = "Bill"               # Теперь в переменной "Bill"
print(scores[name])         # scores["Bill"] -> 78
````

**О выводе нескольких значений:** в следующей ячейке мы пишем `print(name, scores[name])`. Когда в `print()` несколько значений через **запятую** `,`, они выводятся в одну строку, и Python автоматически ставит между ними **один пробел**.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Tom"
print(name, scores[name])   # Tom 92 -> имя и оценка в одной строке
````

## 5. Изменение значения и добавление новой пары

Чтобы изменить значение или добавить новую пару, используется одна и та же форма:

```python
имя_словаря[ключ] = значение
```

Python делает одно из двух, в зависимости от того, есть ли уже такой ключ в словаре:

| Ключ уже есть в словаре? | Что происходит |
|---|---|
| **Да** | Старое значение **заменяется** новым |
| **Нет** | Добавляется **новая пара** |

Это отличается от списка. В списке можно менять значения только под уже существующими индексами. Для добавления нужен `append()`. В словаре одно и то же `[ключ] = значение` и меняет, и добавляет.

### Изменение значения (ключ уже есть)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Bill"] = 88         # "Bill" уже есть среди ключей -> значение 78 заменяется на 88
print(scores)               # Изменилась только оценка Bill
````

````python
# Изменяем значение на основе старого (та же идея, что и age = age + 1)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Anna"] = scores["Anna"] + 5     # Сначала правая часть: 85 + 5 = 90, потом сохраняем обратно
print(scores["Anna"])                   # 90
````

````python
# Короткая запись += тоже работает (то же самое, что: scores["Tom"] = scores["Tom"] + 3)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Tom"] += 3
print(scores["Tom"])                    # 95
````

### Добавление новой пары (ключа ещё нет)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Kate"] = 95         # "Kate" ещё НЕТ среди ключей -> добавляется новая пара
print(scores)               # Теперь 4 пары
````

````python
# Начинаем с пустого словаря и добавляем пары по одной
capitals = {}                   # Начинаем с пустого
print(capitals)

capitals["Korea"] = "Seoul"     # Добавляем пару
print(capitals)

capitals["Japan"] = "Tokyo"     # Добавляем ещё одну
print(capitals)

capitals["France"] = "Paris"    # И ещё одну
print(capitals)
````

````python
# Сравните со списком: для добавления списку нужен append(), а словарю - просто [ключ] = значение
fruit_list = ["apple", "banana"]
fruit_list.append("cherry")         # Список: append() добавляет в конец
print(fruit_list)

prices = {"apple": 1000, "banana": 500}
prices["cherry"] = 3000             # Словарь: [новый ключ] = значение
print(prices)
````

## 6. len() и in

### len(): сколько пар?

Вспомните `len()`: она показывает, сколько значений в списке. Для словаря `len()` показывает, сколько в нём **пар**.

Одна пара (`"Tom": 92`) считается за **1**, а не за 2.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(len(scores))          # 3 пары -> 3

one_pair = {"Tom": 92}
print(len(one_pair))        # 1 пара -> 1 (а не 2!)

empty_dict = {}
print(len(empty_dict))      # 0 пар -> 0
````

````python
# len() меняется, когда мы добавляем пару
scores = {"Anna": 85, "Tom": 92}
print(len(scores))          # 2

scores["Bill"] = 78         # Добавляем новую пару
print(len(scores))          # 3
````

### in: есть ли такой ключ в словаре?

Вспомните `in`: `значение in список` проверяет, есть ли значение в списке, и даёт `True` или `False`.

Со словарём `in` проверяет **ключи**.

```python
"Tom" in scores      # Есть ли ключ "Tom"?
```

**Будьте внимательны:** `in` проверяет **только ключи**, а не значения. Даже если 92 есть в словаре как значение, `92 in scores` даёт `False`.

Чтобы проверить, что ключа **нет** в словаре, используйте `not in`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" - это ключ? -> True
print("Kate" in scores)     # "Kate" - это ключ? -> False
print("Kate" not in scores) # "Kate" НЕ ключ? -> True
````

````python
# in проверяет только КЛЮЧИ, а не значения
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" - ключ -> True
print(92 in scores)         # 92 - это значение, а не ключ -> False
````

````python
# in вместе с if: сначала проверяем, потом получаем значение
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    print(name, scores[name])
else:
    print(name, "is not in the list.")
````

````python
# Простая программа-словарь
english = {"apple": "a round fruit", "dog": "an animal that barks"}
word = "cat"

if word in english:
    print(english[word])
else:
    print("Sorry, I don't know that word.")
````

## 7. Удаление пары

Чтобы удалить пару, используйте `del` и ключ:

```python
del имя_словаря[ключ]
```

`del` - сокращение от **delete** (удалить). Оно удаляет ключ **и** его значение вместе.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

del scores["Tom"]           # Удаляем ключ "Tom" и его значение 92
print(scores)
print(len(scores))          # Теперь 2 пары
````

````python
# Удаление несуществующего ключа вызывает ошибку (увидим в разделе 10)
# Поэтому сначала проверяем с помощью in
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
    print(name, "was deleted.")
else:
    print(name, "is not in the dictionary.")
````

## 8. Цикл for со словарём

Вспомним, как цикл for работает со списком:

```python
for score in score_list:
    print(score)
```

- При каждом повторении одно значение из списка сохраняется в переменную `score`.
- Строка `for` заканчивается двоеточием `:`.
- Повторяемый код пишется **с отступом** (4 пробела). По отступу Python понимает, какие строки относятся к циклу.
- Первая строка без отступа находится вне цикла и выполняется только один раз после цикла.

Со словарём цикл for даёт **ключи**, по одному.

```python
for name in scores:
    print(name)          # Anna, Tom, Bill
```

А когда ключ есть, значение можно получить через `scores[name]`, как в разделе 4.

````python
# Цикл for даёт КЛЮЧИ словаря
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:         # name становится "Anna", потом "Tom", потом "Bill"
    print(name)
````

````python
# По ключу получаем значение
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(scores[name])     # scores["Anna"], потом scores["Tom"], потом scores["Bill"]
````

````python
# Ключ и значение в одной строке (print с запятыми)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])
````

### Имя переменной цикла

Переменная после `for` (здесь `name`) - новая переменная. Ей можно дать любое имя. Выбирайте имя, которое говорит, что за **ключи** в словаре: имена → `name`, страны → `country`, слова → `word`.

````python
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

for country in capitals:            # Ключи - это страны, поэтому называем переменную country
    print(country, capitals[country])
````

### Отступ решает, что повторяется

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])       # С отступом -> повторяется для каждого ключа
print("That's everyone!")           # Без отступа -> вне цикла -> выполняется один раз
````

### Складываем значения

Вспомните **накопитель**:

1. **Перед циклом:** `total = 0`
2. **Внутри цикла:** прибавляем текущее значение → `total = total + scores[name]`
3. **После цикла:** в `total` лежит сумма

`total = total + scores[name]` значит: сначала вычислить правую часть (старая сумма + оценка этого ученика), а потом сохранить результат обратно в `total`. Короткая запись - `total += scores[name]`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0                           # Перед циклом: начинаем с 0

for name in scores:
    total = total + scores[name]    # старая сумма + оценка этого ученика

print("Total:", total)              # После цикла (без отступа): 85 + 92 + 78 = 255
````

````python
# Среднее = сумма / количество значений
# len(scores) - количество пар (= количество учеников)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:
    total += scores[name]           # то же самое, что: total = total + scores[name]

average = total / len(scores)       # После цикла: 255 / 3
print("Average:", average)
````

### if внутри цикла

Когда `if` находится внутри `for`, получается два уровня отступа: 4 пробела для `if` (внутри `for`) и 8 пробелов для кода внутри `if`.

````python
# Выводим только сдавших учеников (60 и выше)
scores = {"Anna": 85, "Tom": 42, "Bill": 78, "Kate": 55}

for name in scores:
    if scores[name] >= 60:          # 4 пробела: проверяем оценку каждого ученика
        print(name, "passed")       # 8 пробелов: только когда условие равно True
````

````python
# Кто сегодня пришёл на урок?
attendance = {"Anna": True, "Tom": False, "Bill": True}

for name in attendance:
    if attendance[name]:            # Значение уже True или False, поэтому его можно сразу использовать как условие
        print(name, "is here.")
    else:
        print(name, "is absent.")
````

## 9. Список и словарь

| | Список | Словарь |
|---|---|---|
| Создание | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92}` |
| Скобки для создания | `[ ]` | `{ }` |
| Значение ищем по | **индексу** (номеру позиции) | **ключу** (имени) |
| Получить значение | `score_list[1]` | `score_dict["Tom"]` |
| Изменить значение | `score_list[1] = 95` | `score_dict["Tom"] = 95` |
| Добавить | `score_list.append(88)` | `score_dict["Kate"] = 88` |
| Удалить | `score_list.remove(92)` | `del score_dict["Tom"]` |
| `len()` считает | значения | пары |
| `in` проверяет | значения | **ключи** |
| `for x in ...` даёт | значения | **ключи** |

**Что когда использовать?**
- Если важен **порядок** или значениям не нужны имена → **список** (например, оценки за сегодняшний тест, по порядку)
- Если нужно находить значение **по имени** → **словарь** (например, оценка каждого ученика, слово и его значение)

## 10. Частые ошибки

Эти ошибки делает почти каждый, кто изучает словари. Для каждой мы посмотрим:

- ❌ **Неправильный код**
- 🔍 **Что пошло не так**
- ✅ **Правильный способ**

Код, который вызывает ошибку, записан в виде комментария (с `#` в начале), чтобы он не останавливал блокнот. Если хотите увидеть настоящее сообщение об ошибке, уберите `#`, запустите ячейку, а потом верните `#` обратно.

### Ошибка 1: Несуществующий ключ

````python
# ❌ Неправильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores["Kate"])     # KeyError: 'Kate'
# print(scores["tom"])      # KeyError: 'tom'
````

🔍 **Что пошло не так?**

- `"Kate"` - не ключ этого словаря. Python не может его найти и показывает `KeyError` ("такого ключа нет").
- `"tom"` (маленькая t) и `"Tom"` (большая T) - **разные** строки. Ключ нужно писать **точно так же**, включая заглавные и строчные буквы.

✅ **Правильный способ:** Пишите ключ точно так же, как в словаре. Если не уверены, что ключ существует, сначала проверьте с помощью `in`.

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # Точно так же, как ключ

name = "Kate"
if name in scores:          # Сначала проверяем
    print(scores[name])
else:
    print(name, "is not in the dictionary.")
````

### Ошибка 2: Квадратные скобки при создании словаря

````python
# ❌ Неправильно
# scores = ["Anna": 85, "Tom": 92]      # SyntaxError
````

🔍 **Что пошло не так?**

- Квадратные скобки `[ ]` нужны для создания **списка**. В списке только значения, поэтому Python не понимает `"Anna": 85` внутри `[ ]`.
- Ошибка - `SyntaxError` ("в коде нарушена грамматика").

✅ **Правильный способ:** Для создания словаря используйте **фигурные скобки** `{ }`. (Чтобы **получить** значение, по-прежнему используются квадратные скобки: `scores["Anna"]`.)

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom": 92}    # Создание: фигурные скобки
print(scores["Anna"])               # Получение: квадратные скобки
````

### Ошибка 3: Забытые кавычки вокруг ключа

````python
# ❌ Неправильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[Anna])       # NameError: name 'Anna' is not defined
````

🔍 **Что пошло не так?**

- Ключ - это **строка** `"Anna"`, в кавычках.
- Без кавычек Python думает, что `Anna` - имя **переменной**. Но переменной с именем `Anna` нет.
- Ошибка - `NameError` ("переменной с таким именем нет").

✅ **Правильный способ:** Ставьте кавычки вокруг строкового ключа. (Кавычки не нужны, только когда ключ хранится в переменной, как `scores[name]` в разделе 4.)

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # Строковый ключ в кавычках

name = "Anna"
print(scores[name])         # Переменная без кавычек (в ней хранится "Anna")
````

### Ошибка 4: Номер индекса, как в списке

````python
# ❌ Неправильно: мы хотим получить оценку первого ученика
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[0])          # KeyError: 0
````

🔍 **Что пошло не так?**

- В списке `[0]` значит "первое значение". Но в словаре нет номеров индексов. Там используются **ключи**.
- Поэтому Python ищет **ключ** с именем `0`. Такого ключа нет, и он показывает `KeyError: 0`.

✅ **Правильный способ:** Используйте ключ (имя).

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # Используем ключ, а не число
````

### Ошибка 5: Один и тот же ключ два раза

````python
# ❌ Неправильно: два разных ученика по имени Tom?
scores = {"Anna": 85, "Tom": 92, "Tom": 60}

print(scores)               # {'Anna': 85, 'Tom': 60} -> первый Tom (92) пропал!
print(len(scores))          # 2, а не 3
````

🔍 **Что пошло не так?**

- В словаре каждый ключ может встречаться **только один раз**.
- Если написать тот же ключ ещё раз, это работает как в разделе 5: ключ уже есть, поэтому старое значение **заменяется**. 92 заменяется на 60.
- Ошибки нет, поэтому это легко пропустить.

✅ **Правильный способ:** Сделайте все ключи разными. Если у двух людей одинаковые имена, дайте им разные ключи.

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom K.": 92, "Tom L.": 60}

print(scores)
print(len(scores))          # 3
````

### Ошибка 6: Ждать, что in проверяет значения

````python
# ❌ Неправильно: мы хотим узнать, получил ли кто-то 92 балла
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(92 in scores)         # False -> но Tom получил 92!
````

🔍 **Что пошло не так?**

- Со словарём `in` проверяет **только ключи** (имена), а не значения (оценки).
- 92 - это значение, а не ключ, поэтому результат `False`.
- Ошибки нет, но ответ неправильный.

✅ **Правильный способ:** Пройдите по словарю циклом for и сравните каждое значение.

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    if scores[name] == 92:      # Сравниваем каждое ЗНАЧЕНИЕ с 92
        print(name, "got 92 points.")
````

### Ошибка 7: Думать, что цикл for даёт значения

````python
# ❌ Неправильно: мы хотим сложить все оценки
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

# for score in scores:
#     total += score        # TypeError: unsupported operand type(s) for +=: 'int' and 'str'
````

🔍 **Что пошло не так?**

- Со словарём цикл for даёт **ключи**, а не значения. Поэтому `score` - это `"Anna"`, `"Tom"`, `"Bill"`, а не 85, 92, 78.
- `total += score` пытается прибавить строку (`"Anna"`) к числу (`0`). Python не умеет складывать число и строку, поэтому показывает `TypeError`.
- Имя переменной `score` только запутало ещё больше. На самом деле это имя ученика.

✅ **Правильный способ:** Называйте переменную цикла по ключам (`name`) и получайте значение через `scores[name]`.

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:             # name - это ключ
    total += scores[name]       # scores[name] - это значение

print("Total:", total)
````

### Ошибка 8: Удаление несуществующего ключа

````python
# ❌ Неправильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# del scores["Kate"]        # KeyError: 'Kate'
````

🔍 **Что пошло не так?**

- Ключа `"Kate"` в словаре нет, поэтому удалять нечего.
- Как и в Ошибке 1, Python показывает `KeyError`.

✅ **Правильный способ:** Сначала проверьте с помощью `in` и удаляйте, только если ключ есть.

````python
# ✅ Правильно
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
else:
    print(name, "is not in the dictionary, so nothing was deleted.")
````
