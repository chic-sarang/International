## 1. Variables and Data Types

A variable is like a box that stores a value. If you give the box a name, later you can just call that name to get the value inside. If you put a new value in, the contents of the box change too.

However, there are different kinds of values you can put into the box. This kind of value is called a data type. The main data types in Python are as follows.

- `str` : string (text)
- `int` : integer
- `float` : a number with a decimal point
- `bool` : boolean (True/False)

By the way, `print()` is a command that shows the value inside the parentheses on the screen.

```python
# = doesn't mean 'equal to' - it means 'store the value on the right into the box named on the left'
# The variable names (name, age, height, is_student) are the names given to the boxes

# A string (str) just needs to be wrapped in quotes. You can use double quotes (" ") or single quotes (' ')
name = "Bill"                    # A word wrapped in double quotes
city = 'Seoul'                    # A word wrapped in single quotes
greeting = "Hello, nice to meet you"  # A sentence is still a string
zip_code = "06236"               # Even if it looks like a number, if it's inside quotes it's a string

# An integer (int) is a number without a decimal point. Positive numbers, negative numbers, and 0 are all integers
age = 20              # A positive integer
temperature = -5      # A negative integer is fine too
score = 0             # 0 is an integer too

# A float is a number with a decimal point
height = 175.5        # A positive float
account_balance = -3.75  # A negative float is fine too
pi_value = 3.14159    # It doesn't matter how many digits are after the decimal point

# A boolean (bool) can only be True or False (the first letter is always capitalized)
is_student = True     # is a student
is_married = False    # is not married
has_passed = True     # passed the exam

# print() is a command that shows a value on the screen
print(name)
print(city)
print(greeting)
print(zip_code)
print(age)
print(temperature)
print(score)
print(height)
print(account_balance)
print(pi_value)
print(is_student)
print(is_married)
print(has_passed)
```

```python
# You can change the value of a variable (integer example)
age = 20
print(age)

age = 21   # If you store a new value in the same variable, the old value is gone and replaced by the new one
print(age)

age = age + 1   # You can also create a new value based on the old one (21 + 1 = 22)
print(age)
```

```python
# You can change the value of a variable (boolean example)
is_raining = False
print(is_raining)

is_raining = True   # If you store a new value in the same variable, the old value is gone and replaced by the new one
print(is_raining)

is_raining = not is_raining   # Using not flips the value: True becomes False, and False becomes True
print(is_raining)
```

```python
# You can change the value of a variable (string example)
city = "Seoul"
print(city)

city = "Busan"   # If you store a new value in the same variable, the old value is gone and replaced by the new one
print(city)

city = city + " Travel"   # Strings can be joined together with + ("Busan" + " Travel" = "Busan Travel")
print(city)
```

```python
# You can change the value of a variable (float example)
height = 175.5
print(height)

height = 178.2   # If you store a new value in the same variable, the old value is gone and replaced by the new one
print(height)

height = height + 0.5   # You can also create a new value based on the old one (178.2 + 0.5 = 178.7)
print(height)
```

## 2. Basic Operators

Symbols used to calculate numbers or compare values.

```python
# The four basic arithmetic operations
a = 10
b = 3

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division (result is a float)
print(a // b)  # Division (only the quotient remains)
print(a % b)   # Remainder
```

```python
# Comparison operators -> the result is always True or False
print(a > b)   # a is greater than b
print(a < b)   # a is less than b
print(a == b)  # a is equal to b
print(a != b)  # a is not equal to b
```
