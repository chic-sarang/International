## 1. Что такое цикл for?

Представьте, что нужно вывести "Hello" 5 раз. Можно написать `print("Hello")` 5 раз. А если 100 раз? Или 1000? Писать одну и ту же строку снова и снова утомительно, и легко ошибиться.

Эту проблему решает **цикл**. Цикл выполняет один и тот же код снова и снова, столько раз, сколько вам нужно.

Самый распространённый цикл в Python - это цикл `for`. Базовая форма выглядит так:

```python
for i in range(5):
    код, который нужно повторять
```

Это можно прочитать так: **"для каждого числа i из range(5) выполни код ниже"**.

- `range(5)` даёт числа 0, 1, 2, 3, 4 (всего 5 чисел). Подробнее `range()` мы изучим в следующем разделе.
- `i` - это переменная. При каждом повторе цикла в `i` сохраняется следующее число.
- Как и у `if`, строка с `for` должна заканчиваться двоеточием `:`.
- Как и у `if`, повторяемый код должен быть **с отступом** (4 пробела). Повторяются только строки с отступом.

````python
# Повторяем print("Hello") 5 раз
# range(5) даёт 5 чисел (0, 1, 2, 3, 4), поэтому код с отступом выполняется 5 раз
for i in range(5):
    print("Hello")      # Эта строка находится внутри цикла for (с отступом), поэтому она повторяется
````

````python
# Посмотрим, что хранится в i каждый раз
# 1-й раз: i = 0, 2-й раз: i = 1, 3-й раз: i = 2, 4-й раз: i = 3, 5-й раз: i = 4
for i in range(5):
    print(i)            # Выводим текущее значение i
````

````python
# Один полный проход по коду с отступом называется "повторением" (итерацией)
# Каждая строка с отступом выполняется один раз за повторение, по порядку сверху вниз
for i in range(3):
    print("Repetition number", i)     # print() может показать несколько значений сразу, если разделить их запятыми
    print("----------")               # Эта строка тоже внутри цикла (такой же отступ)

print("The loop is finished.")        # БЕЗ отступа, значит она вне цикла. Выполняется только один раз, после окончания цикла
````

````python
# Переменная не обязательно должна называться i
# i - это просто имя, которое чаще всего используют. Можно выбрать любое имя, если оно соответствует правилам именования переменных
for number in range(3):
    print(number)

for count in range(3):
    print("count is", count)
````

## 2. Подробнее о range()

`range()` создаёт последовательность чисел. Есть три способа его использовать.

| Как писать | Значение | Какие числа даёт |
|---|---|---|
| `range(end)` | от 0 до end (end не включается) | `range(5)` → 0, 1, 2, 3, 4 |
| `range(start, end)` | от start до end (end не включается) | `range(1, 6)` → 1, 2, 3, 4, 5 |
| `range(start, end, step)` | от start до end, с шагом step | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |

Самое важное, что нужно запомнить:
- **Конечное число никогда не включается.** `range(5)` останавливается на 4, а не на 5.
- Если не указать начальное число, отсчёт начинается с **0**.
- Если не указать шаг, числа увеличиваются на **1**.

````python
# range(end): начинает с 0 и останавливается прямо перед end
# range(4) даёт 0, 1, 2, 3 -> 4 НЕ включается
for i in range(4):
    print(i)
````

````python
# range(start, end): начинает со start и останавливается прямо перед end
# Если нужны числа от 1 до 5, end должен быть 6 (потому что конечное число не включается)
for i in range(1, 6):
    print(i)
````

````python
# range(start, end, step): каждый раз прыгает на step
# Начинаем с 0, шаг 2, останавливаемся перед 10 -> 0, 2, 4, 6, 8
for i in range(0, 10, 2):
    print(i)
````

````python
# Ещё один пример с шагом: считаем пятёрками от 5 до 25
# end равен 26, чтобы 25 включилось (можно было написать и 30, ведь само 30 не включилось бы)
for i in range(5, 26, 5):
    print(i)
````

````python
# Если шаг отрицательный, числа идут ВНИЗ (обратный отсчёт)
# Начинаем с 5, уменьшаем на 1, останавливаемся перед 0 -> 5, 4, 3, 2, 1
for i in range(5, 0, -1):
    print(i)

print("Lift off!")      # Вне цикла, выполняется один раз в конце
````

````python
# Используем i в вычислениях
# Выводим таблицу умножения на 3 (от 3 x 1 до 3 x 9)
for i in range(1, 10):           # i идёт от 1 до 9 (10 не включается)
    print(3, "x", i, "=", 3 * i)  # 3 * i каждый раз вычисляется заново с новым i
````

## 3. Накопление суммы с помощью цикла

Одна из самых частых задач для цикла - **сложить** числа. Например, сколько будет 1 + 2 + 3 + ... + 10?

Хитрость в том, чтобы использовать переменную как "накопитель".
1. Перед циклом создайте переменную и присвойте ей 0. (`total = 0`)
2. Внутри цикла каждый раз прибавляйте к ней текущее число. (`total = total + i`)
3. После цикла в переменной будет итоговый результат.

Помните `age = age + 1` из Урока 1? Это та же самая идея: берём старое значение, что-то прибавляем и сохраняем обратно в ту же коробку.

````python
# Складываем 1 + 2 + 3 + ... + 10
total = 0                     # Шаг 1: накопитель начинается с 0 (ПЕРЕД циклом)

for i in range(1, 11):        # i идёт от 1 до 10
    total = total + i         # Шаг 2: прибавляем текущее i к накопителю

print("Sum:", total)          # Шаг 3: ПОСЛЕ цикла выводим итог (вне цикла, поэтому выводится только один раз)
````

````python
# Посмотрим, как меняется total при каждом повторении
# Это тот же код, что и выше, но с print() внутри цикла
total = 0

for i in range(1, 6):         # i идёт от 1 до 5
    total = total + i
    print("i =", i, "-> total =", total)   # Внутри цикла, поэтому выводится каждый раз

# 1-й раз: 0 + 1 = 1
# 2-й раз: 1 + 2 = 3
# 3-й раз: 3 + 3 = 6
# 4-й раз: 6 + 4 = 10
# 5-й раз: 10 + 5 = 15
````

````python
# total = total + i можно записать короче: total += i
# Обе строки означают одно и то же. Многие программисты на Python используют короткую форму
total = 0

for i in range(1, 101):       # i идёт от 1 до 100
    total += i                # То же самое, что: total = total + i

print("Sum of 1 to 100:", total)
````

````python
# Накопитель не обязательно должен быть суммой. Можно и умножать
# 1 x 2 x 3 x 4 x 5 (в математике это называется "факториал")
result = 1                    # Для умножения начинаем с 1, а не с 0 (любое число x 0 даёт 0!)

for i in range(1, 6):         # i идёт от 1 до 5
    result = result * i       # Умножаем результат на текущее i

print("1 x 2 x 3 x 4 x 5 =", result)
````

````python
# Подсчёт того, сколько раз выполнился цикл, - это тоже накопление
# Здесь мы каждый раз прибавляем 1, а не i
count = 0

for i in range(10, 20):       # i идёт от 10 до 19
    count += 1                # Прибавляем 1 при каждом повторении

print("The loop ran", count, "times.")   # В range(10, 20) 10 чисел, поэтому цикл выполнился 10 раз
````

## 4. for и if вместе

`if` можно поместить **внутрь** цикла `for`. Так можно проверить каждое число по очереди и что-то сделать только с нужными числами.

```python
for i in range(...):
    if условие:
        код, который выполняется только когда условие равно True
```

- `if` находится внутри `for`, поэтому у него отступ 4 пробела.
- Код внутри `if` сдвинут ещё на один уровень, поэтому у него отступ 8 пробелов.

````python
# Выводим только чётные числа от 1 до 10
# Вспомните Урок 2: число чётное, если number % 2 == 0
for i in range(1, 11):        # Проверяем каждое число от 1 до 10 по очереди
    if i % 2 == 0:            # Текущее i чётное? (4 пробела: внутри for)
        print(i)              # Выводятся только чётные числа (8 пробелов: внутри if)
````

````python
# Тот же результат можно получить и только с помощью шага range(), без if
# Начинаем с 2, шаг 2 -> 2, 4, 6, 8, 10
for i in range(2, 11, 2):
    print(i)

# Оба способа правильные. Но с if можно проверить любое условие, а не только "каждое второе число"
````

````python
# Используем if - else внутри цикла
for i in range(1, 6):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
````

````python
# for + if + накопитель вместе
# Складываем только числа, кратные 3, от 1 до 20 (3 + 6 + 9 + 12 + 15 + 18)
total = 0

for i in range(1, 21):        # i идёт от 1 до 20
    if i % 3 == 0:            # i кратно 3? (делится на 3 без остатка)
        total += i            # Только тогда прибавляем его к сумме

print("Sum of multiples of 3:", total)
````

````python
# Классическое упражнение: FizzBuzz
# Для чисел от 1 до 15:
#   - если число кратно и 3, и 5, выведите "FizzBuzz"
#   - если число кратно только 3, выведите "Fizz"
#   - если число кратно только 5, выведите "Buzz"
#   - иначе выведите само число
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:     # Сначала проверяем самое строгое условие (вспомните Урок 2: порядок важен!)
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

## 5. Цикл while

Цикл `for` хорош, когда вы знаете, **сколько раз** нужно повторить. Но иногда это число заранее неизвестно. Известно только, **когда нужно остановиться**. Например:

- Продолжай копить деньги, **пока** сумма меньше 10000.
- Продолжай резать бумагу пополам, **пока** она длиннее 1 см.

Для этого используется цикл `while`. `while` означает "пока". Он продолжает повторяться, **пока условие равно True**, и останавливается, когда условие становится False.

```python
while условие:
    код, который нужно повторять
```

Как это работает:
1. Проверить условие.
2. Если оно True, выполнить код с отступом и вернуться к шагу 1.
3. Если оно False, остановить цикл и перейти к коду после него.

**Очень важно:** Внутри цикла что-то должно меняться, чтобы в какой-то момент условие стало False. Если условие остаётся True навсегда, цикл никогда не закончится. Это называется **бесконечным циклом**.

````python
# Выводим числа от 1 до 5 с помощью while
# В while мы сами создаём, проверяем и изменяем переменную
i = 1                     # 1. Создаём переменную ПЕРЕД циклом

while i <= 5:             # 2. Проверяем условие: продолжаем, пока i меньше или равно 5
    print(i)
    i = i + 1             # 3. Изменяем переменную. Без этой строки i навсегда останется 1 (бесконечный цикл!)

print("Done. Now i is", i)    # Когда i стало 6, 6 <= 5 оказалось False, и цикл остановился
````

````python
# Сравним: то же самое с помощью for
# for делает работу "создать, проверить, изменить" автоматически с помощью range()
for i in range(1, 6):
    print(i)

# Поэтому, когда известно, сколько раз повторять, for обычно проще
````

````python
# Случай, где while подходит лучше: мы заранее не знаем, сколько раз повторять
# Вы каждый день откладываете 3000 вон. Сколько дней нужно, чтобы накопить 20000 вон или больше?
money = 0
days = 0

while money < 20000:          # Продолжаем копить, пока денег меньше 20000
    money += 3000             # Откладываем 3000 вон
    days += 1                 # Прошёл ещё один день

print("Days needed:", days)
print("Money saved:", money)
````

````python
# Ещё один пример: делим число пополам, пока оно не станет 1 или меньше
# Используем // (только целая часть, из Урока 1), чтобы число оставалось целым
number = 100
steps = 0

while number > 1:             # Продолжаем, пока number больше 1
    number = number // 2      # Делим пополам (100 -> 50 -> 25 -> 12 -> 6 -> 3 -> 1)
    steps += 1
    print("Step", steps, "->", number)

print("It took", steps, "steps.")
````

````python
# Если условие равно False с самого начала, цикл while не выполнится ни разу
i = 10

while i < 5:                  # 10 < 5 - это False сразу
    print("This never prints.")
    i += 1

print("The loop was skipped.")
````

## 6. break и continue

Иногда нужно изменить ход цикла посередине.

- `break`: **немедленно остановить весь цикл** и выйти из него.
- `continue`: **пропустить остаток текущего повторения** и сразу перейти к следующему.

Оба обычно используются вместе с `if`, потому что прерывать или пропускать нужно только в определённой ситуации.

````python
# break: останавливаем цикл, как только нашли то, что искали
# Найдём первое число от 1 до 100, которое кратно 7
for i in range(1, 101):
    if i % 7 == 0:                # i кратно 7?
        print("Found it:", i)
        break                     # Останавливаем цикл прямо здесь. Числа после 7 вообще не проверяются

print("The loop is over.")
````

````python
# Посмотрим, где именно break останавливает цикл
for i in range(1, 11):
    if i == 4:                    # Когда i становится 4...
        break                     # ...сразу останавливаем цикл
    print(i)                      # Поэтому выводятся только 1, 2, 3 (4 не выводится, потому что break срабатывает раньше)
````

````python
# continue: пропускаем только это повторение и продолжаем со следующим
for i in range(1, 6):
    if i == 3:                    # Когда i равно 3...
        continue                  # ...пропускаем print() ниже и сразу переходим к i = 4
    print(i)                      # Поэтому выводятся 1, 2, 4, 5 (не хватает только 3)
````

````python
# break вместе с while: частый приём "while True" + break
# while True означает "повторять вечно", поэтому break - ЕДИНСТВЕННЫЙ способ выйти
count = 0

while True:                       # Условие всегда True -> без break это был бы бесконечный цикл
    count += 1
    print("count:", count)
    if count == 3:                # Когда count достигает 3...
        break                     # ...выходим из цикла

print("Out of the loop!")
````

````python
# Сравниваем break и continue рядом
print("With break:")
for i in range(1, 6):
    if i == 3:
        break                     # Останавливает всё на 3
    print(i)                      # 1, 2

print("With continue:")
for i in range(1, 6):
    if i == 3:
        continue                  # Пропускает только 3
    print(i)                      # 1, 2, 4, 5
````

## 7. Частые ошибки

Эти ошибки делает почти каждый, кто изучает циклы. Если что-то пошло не так, сначала проверьте эти пункты.

1. **Забыть, что конечное число в range() не включается**: `range(1, 10)` останавливается на 9. Чтобы включить 10, пишите `range(1, 11)`.
2. **Забытое двоеточие `:`**: Строки `for` и `while` должны заканчиваться на `:`.
3. **Неправильный отступ**: Повторяются только строки с отступом. Строка, которая должна быть внутри цикла, но не имеет отступа, выполнится только один раз.
4. **Бесконечный цикл с while**: Если забыть изменить переменную внутри цикла `while`, условие навсегда останется True.
5. **Обнуление накопителя внутри цикла**: `total = 0` нужно писать **перед** циклом, а не внутри него.

Ячейки ниже показывают, как выглядит каждая ошибка. Ошибочный код записан в виде комментариев, чтобы не ломать блокнот.

**Если вы случайно запустили бесконечный цикл**, нажмите кнопку ■ (стоп / interrupt) в верхней части Jupyter или дважды нажмите `I` (`I`, `I`), когда ячейка выбрана.

````python
# Ошибка 1: забыть, что конечное число не включается
# Нам нужны числа от 1 до 5, но...
for i in range(1, 5):     # Неправильно: цикл останавливается на 4!
    print(i)

print("---")

for i in range(1, 6):     # Правильно: end должен быть на 1 больше последнего нужного числа
    print(i)
````

````python
# Ошибка 2: забытое двоеточие

# for i in range(3)       # SyntaxError! В конце не хватает двоеточия
#     print(i)

for i in range(3):        # Правильно: строка заканчивается двоеточием
    print(i)
````

````python
# Ошибка 3: неправильный отступ
# Второй print() БЕЗ отступа, поэтому он вне цикла и выполняется только один раз
for i in range(3):
    print("Inside the loop:", i)
print("Outside the loop:", i)     # Выполняется один раз, после цикла. (i всё ещё хранит последнее значение - 2)

# for i in range(3):
# print(i)                # IndentationError! Внутри for вообще нет кода с отступом
````

````python
# Ошибка 4: бесконечный цикл с while
# НЕ убирайте # в этом примере! Он никогда не остановится.

# i = 1
# while i <= 5:
#     print(i)            # i никогда не меняется, поэтому i <= 5 всегда True -> бесконечный цикл!

i = 1
while i <= 5:
    print(i)
    i += 1                # Правильно: каждый раз меняем i, чтобы условие в итоге стало False
````

````python
# Ошибка 5: обнуление накопителя внутри цикла
# Неправильно: total каждый раз сбрасывается в 0, поэтому остаётся только последнее число
for i in range(1, 6):
    total = 0             # Неправильное место! Эта строка выполняется 5 раз
    total += i
print("Wrong total:", total)      # Только 5

# Правильно: присваиваем total = 0 только один раз, ПЕРЕД циклом
total = 0
for i in range(1, 6):
    total += i
print("Correct total:", total)    # 1 + 2 + 3 + 4 + 5 = 15
````
