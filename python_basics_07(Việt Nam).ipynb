## 1. Tại sao chúng ta cần dictionary?

Hãy bắt đầu từ thứ chúng ta đã biết: **list**.

List chứa nhiều giá trị theo thứ tự, bên trong dấu ngoặc vuông `[ ]`. Chúng ta lấy giá trị bằng **chỉ số (index)** của nó (số chỉ vị trí), và chỉ số bắt đầu từ **0**.

Bây giờ hãy tưởng tượng chúng ta muốn lưu điểm thi của 3 học sinh: Anna, Tom và Bill. Với list, chúng ta sẽ cần **hai** list: một cho tên, một cho điểm.

```python
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]
```

Hai list này được nối với nhau bằng vị trí:

| Chỉ số | 0 | 1 | 2 |
|---|---|---|---|
| `names` | "Anna" | "Tom" | "Bill" |
| `scores` | 85 | 92 | 78 |

Vậy `names[1]` là Tom, và `scores[1]` là điểm của Tom.

**Nhưng cách này có vấn đề:**

- Để tìm điểm của Tom, trước tiên bạn phải **nhớ** rằng Tom ở chỉ số 1. Python không thể tìm từ cái tên "Tom".
- Nếu bạn thêm hoặc xóa một tên ở list này mà quên làm điều tương tự ở list kia, tên và điểm sẽ bị **lẫn lộn**.

Sẽ dễ hơn nhiều nếu chúng ta chỉ cần hỏi: **"Điểm của Tom là bao nhiêu?"**

Đó chính là điều mà dictionary làm được.

````python
# Với hai list, tên và điểm nằm ở hai nơi riêng biệt
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

# Để lấy điểm của Tom, chúng ta phải biết Tom ở chỉ số 1
print(names[1])         # Tom
print(scores[1])        # 92 -> điểm của Tom
````

````python
# Vấn đề: nếu một list thay đổi còn list kia thì không, chúng sẽ bị lẫn lộn
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

names.remove("Anna")    # Chúng ta đã xóa Anna khỏi names, nhưng quên xóa 85 khỏi scores

print(names[0])         # Tom
print(scores[0])        # 85 -> Đây là điểm của Anna, không phải của Tom! Bây giờ chúng không khớp nhau
````

## 2. Dictionary là gì?

Hãy nghĩ đến một cuốn từ điển thật, cuốn sách bạn dùng để học từ.

- Bạn tra một **từ** (ví dụ, "apple").
- Bạn tìm thấy **nghĩa** của nó ("một loại quả tròn").

Bạn không cần biết từ đó nằm ở trang nào. Bạn chỉ cần tìm từ, và nghĩa nằm ngay bên cạnh.

**Dictionary** trong Python hoạt động giống như vậy. Nó lưu giá trị theo **cặp**:

| Từ trong Python | Ý nghĩa | Trong từ điển thật | Trong ví dụ của chúng ta |
|---|---|---|---|
| **khóa (key)** | cái tên bạn tìm | từ | tên học sinh ("Tom") |
| **giá trị (value)** | thứ bạn nhận được | nghĩa | điểm số (92) |
| **cặp (pair)** | một khóa và giá trị của nó đi cùng nhau | một từ và nghĩa của nó | "Tom" và 92 |

Bạn cũng có thể hình dung nó như **những ngăn tủ có gắn bảng tên**:
- Trong **list**, mỗi ngăn tủ có một **con số** (0, 1, 2 ...).
- Trong **dictionary**, mỗi ngăn tủ có một **bảng tên** ("Anna", "Tom", "Bill" ...).

Để mở một ngăn tủ, bạn dùng bảng tên thay vì con số.

## 3. Tạo một dictionary

Hãy bắt đầu với dictionary nhỏ nhất: chỉ có **một cặp**.

```python
scores = {"Tom": 92}
```

Hãy xem từng phần:

| Phần | Ý nghĩa |
|---|---|
| `{` | dictionary bắt đầu (là **dấu ngoặc nhọn**, không phải dấu ngoặc vuông) |
| `"Tom"` | **khóa** (ở đây là một chuỗi, nên nằm trong dấu ngoặc kép) |
| `:` | **dấu hai chấm**. Nó nối khóa với giá trị của nó |
| `92` | **giá trị** |
| `}` | dictionary kết thúc |

Hãy nhớ: `=` có nghĩa là **"lưu giá trị bên phải vào chiếc hộp bên trái"**. Vì vậy dòng này lưu cả dictionary vào một biến tên `scores`.

````python
# Một dictionary có một cặp
scores = {"Tom": 92}

print(scores)           # {'Tom': 92}
````

### Nhiều hơn một cặp

Để đặt nhiều cặp vào dictionary, hãy ngăn cách chúng bằng **dấu phẩy** `,`. Giống như trong list, các giá trị được ngăn cách bằng dấu phẩy.

```python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
```

- `"Anna": 85` là một cặp
- `"Tom": 92` là một cặp
- `"Bill": 78` là một cặp

Vậy dictionary này có **3 cặp**.

````python
# Một dictionary có hai cặp
scores = {"Anna": 85, "Tom": 92}
print(scores)
````

````python
# Một dictionary có ba cặp
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)
````

### Viết mỗi cặp trên một dòng

Khi có nhiều cặp, một dòng dài sẽ khó đọc. Bạn có thể viết **mỗi cặp trên một dòng**. Python đọc nó hoàn toàn giống nhau. Đừng quên dấu phẩy ở cuối mỗi cặp.

````python
# Cùng dictionary đó, viết mỗi cặp trên một dòng (dễ đọc hơn)
scores = {
    "Anna": 85,
    "Tom": 92,
    "Bill": 78,
}
print(scores)
````

### So sánh: list và dictionary

| | List | Dictionary |
|---|---|---|
| Dấu ngoặc | ngoặc vuông `[ ]` | ngoặc nhọn `{ }` |
| Bên trong có gì | chỉ có giá trị | các **cặp** `khóa: giá trị` |
| Ví dụ | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92, "Bill": 78}` |

````python
# Cùng các điểm đó, dưới dạng list và dưới dạng dictionary
score_list = [85, 92, 78]
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list)       # Chỉ có điểm. Chúng ta không biết đó là điểm của ai
print(score_dict)       # Mỗi điểm có một cái tên bên cạnh
````

````python
# Dictionary là một kiểu dữ liệu mới: dict (viết tắt của "dictionary")
# type() cho biết một giá trị thuộc loại gì
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(type(scores))     # <class 'dict'>
````

### Các loại dictionary khác

Giá trị có thể là bất kỳ kiểu dữ liệu nào: chuỗi, số nguyên, số thực, bool. Khóa thường là **chuỗi**.

````python
# Giá trị là chuỗi: một từ -> nghĩa của nó, giống như từ điển thật
english = {"apple": "a round fruit", "dog": "an animal that barks"}
print(english)
````

````python
# Giá trị là chuỗi: một quốc gia -> thủ đô của nó
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}
print(capitals)
````

````python
# Giá trị là số thực: một người -> chiều cao
heights = {"Anna": 162.5, "Tom": 175.0}
print(heights)
````

````python
# Giá trị là bool: một học sinh -> hôm nay có đến lớp không?
attendance = {"Anna": True, "Tom": False, "Bill": True}
print(attendance)
````

````python
# Một dictionary rỗng: không có cặp nào
# Chúng ta có thể bắt đầu với dictionary rỗng rồi thêm cặp sau (phần 5)
empty_dict = {}
print(empty_dict)
````

## 4. Lấy giá trị bằng khóa (key)

Để lấy một giá trị từ dictionary, hãy viết tên dictionary và **khóa** trong dấu ngoặc vuông `[ ]`:

```python
scores["Tom"]      # "giá trị của khóa 'Tom' trong scores"
```

Cách này trông rất giống list. Điểm khác duy nhất là thứ nằm trong dấu ngoặc vuông:

| | List | Dictionary |
|---|---|---|
| Viết gì trong `[ ]` | một **chỉ số** (số chỉ vị trí) | một **khóa** (một cái tên) |
| Ví dụ | `score_list[1]` | `score_dict["Tom"]` |
| Ý nghĩa | "giá trị ở vị trí 1" | "giá trị có khóa là 'Tom'" |

**Quan trọng:** dù chúng ta dùng dấu ngoặc nhọn `{ }` để **tạo** dictionary, chúng ta dùng dấu ngoặc vuông `[ ]` để **lấy** giá trị từ nó.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # Giá trị của khóa "Tom" -> 92
print(scores["Anna"])       # Giá trị của khóa "Anna" -> 85
print(scores["Bill"])       # Giá trị của khóa "Bill" -> 78
````

````python
# So sánh: lấy điểm của Tom từ list và từ dictionary
score_list = [85, 92, 78]                           # Anna, Tom, Bill
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list[1])        # List: chúng ta phải nhớ rằng Tom ở chỉ số 1
print(score_dict["Tom"])    # Dictionary: chúng ta chỉ cần dùng tên "Tom"
````

````python
# Thêm ví dụ
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

print(capitals["Korea"])    # Seoul
print(capitals["France"])   # Paris
````

````python
english = {"apple": "a round fruit", "dog": "an animal that barks"}

print(english["dog"])       # Tra từ "dog" -> nghĩa của nó
````

### Giá trị có thể được dùng như một biến bình thường

Giá trị bạn lấy từ dictionary chỉ là một giá trị bình thường. Bạn có thể lưu nó vào biến, tính toán hoặc so sánh với nó.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

tom_score = scores["Tom"]                   # Lưu điểm của Tom vào một biến mới
print(tom_score)

print(scores["Anna"] + scores["Tom"])       # 85 + 92 = 177 (có thể tính toán)
print(scores["Tom"] > scores["Bill"])       # 92 > 78 là True (có thể so sánh)
````

````python
# Dùng giá trị trong if (if chỉ chạy đoạn mã thụt lề khi điều kiện là True)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

if scores["Bill"] >= 80:
    print("Bill passed with a high score!")
else:
    print("Bill needs to study more.")
````

### Khóa có thể là một biến

Thay vì viết khóa trực tiếp, bạn có thể lưu khóa vào một biến trước. Sau đó đặt biến vào trong `[ ]`, **không có dấu ngoặc kép**.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Anna"               # Biến name chứa chuỗi "Anna"
print(scores[name])         # scores[name] giống với scores["Anna"] -> 85

name = "Bill"               # Bây giờ biến chứa "Bill"
print(scores[name])         # scores["Bill"] -> 78
````

**Về việc in nhiều giá trị:** ở ô tiếp theo, chúng ta viết `print(name, scores[name])`. Khi đặt nhiều giá trị vào `print()` và ngăn cách bằng **dấu phẩy** `,`, chúng được in trên một dòng, và Python tự động đặt **một dấu cách** giữa chúng.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Tom"
print(name, scores[name])   # Tom 92 -> tên và điểm trên một dòng
````

## 5. Thay đổi giá trị và thêm một cặp mới

Để thay đổi giá trị hoặc thêm cặp mới, chúng ta dùng cùng một dạng:

```python
tên_dictionary[khóa] = giá_trị
```

Python sẽ làm một trong hai việc, tùy vào việc khóa đó **đã có** trong dictionary hay chưa:

| Khóa đã có trong dictionary chưa? | Điều gì xảy ra |
|---|---|
| **Có rồi** | Giá trị cũ bị **thay thế** bằng giá trị mới |
| **Chưa có** | Một **cặp mới** được thêm vào |

Điều này khác với list. Trong list, bạn chỉ có thể thay đổi giá trị ở những chỉ số đã tồn tại. Để thêm, bạn cần `append()`. Trong dictionary, cùng một `[khóa] = giá_trị` vừa thay đổi vừa thêm được.

### Thay đổi giá trị (khóa đã tồn tại)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Bill"] = 88         # "Bill" đã là một khóa -> giá trị 78 bị thay bằng 88
print(scores)               # Chỉ có điểm của Bill thay đổi
````

````python
# Thay đổi giá trị dựa trên giá trị cũ (cùng ý tưởng với age = age + 1)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Anna"] = scores["Anna"] + 5     # Vế phải trước: 85 + 5 = 90, rồi lưu lại
print(scores["Anna"])                   # 90
````

````python
# Cách viết ngắn += cũng dùng được (giống với: scores["Tom"] = scores["Tom"] + 3)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Tom"] += 3
print(scores["Tom"])                    # 95
````

### Thêm một cặp mới (khóa chưa tồn tại)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Kate"] = 95         # "Kate" CHƯA là khóa -> một cặp mới được thêm vào
print(scores)               # Bây giờ có 4 cặp
````

````python
# Bắt đầu với một dictionary rỗng, rồi thêm từng cặp một
capitals = {}                   # Bắt đầu rỗng
print(capitals)

capitals["Korea"] = "Seoul"     # Thêm một cặp
print(capitals)

capitals["Japan"] = "Tokyo"     # Thêm một cặp nữa
print(capitals)

capitals["France"] = "Paris"    # Và một cặp nữa
print(capitals)
````

````python
# So sánh với list: để thêm, list cần append(), còn dictionary chỉ cần [khóa] = giá_trị
fruit_list = ["apple", "banana"]
fruit_list.append("cherry")         # List: append() thêm vào cuối
print(fruit_list)

prices = {"apple": 1000, "banana": 500}
prices["cherry"] = 3000             # Dictionary: [khóa mới] = giá_trị
print(prices)
````

## 6. len() và in

### len(): có bao nhiêu cặp?

Hãy nhớ `len()`: nó cho biết list có bao nhiêu giá trị. Với dictionary, `len()` cho biết có bao nhiêu **cặp**.

Một cặp (`"Tom": 92`) được tính là **1**, không phải 2.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(len(scores))          # 3 cặp -> 3

one_pair = {"Tom": 92}
print(len(one_pair))        # 1 cặp -> 1 (không phải 2!)

empty_dict = {}
print(len(empty_dict))      # 0 cặp -> 0
````

````python
# len() thay đổi khi thêm một cặp
scores = {"Anna": 85, "Tom": 92}
print(len(scores))          # 2

scores["Bill"] = 78         # Thêm một cặp mới
print(len(scores))          # 3
````

### in: khóa này có trong dictionary không?

Hãy nhớ `in`: `giá_trị in list` kiểm tra xem giá trị có nằm trong list không, và cho ra `True` hoặc `False`.

Với dictionary, `in` kiểm tra các **khóa**.

```python
"Tom" in scores      # Có khóa "Tom" không?
```

**Hãy cẩn thận:** `in` **chỉ kiểm tra khóa**, không kiểm tra giá trị. Dù 92 có trong dictionary dưới dạng giá trị, `92 in scores` vẫn là `False`.

Để kiểm tra một khóa **không** có trong dictionary, hãy dùng `not in`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" có phải là khóa không? -> True
print("Kate" in scores)     # "Kate" có phải là khóa không? -> False
print("Kate" not in scores) # "Kate" KHÔNG phải là khóa? -> True
````

````python
# in chỉ kiểm tra KHÓA, không kiểm tra giá trị
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" là khóa -> True
print(92 in scores)         # 92 là giá trị, không phải khóa -> False
````

````python
# Dùng in với if: kiểm tra trước, rồi mới lấy giá trị
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    print(name, scores[name])
else:
    print(name, "is not in the list.")
````

````python
# Một chương trình tra từ điển đơn giản
english = {"apple": "a round fruit", "dog": "an animal that barks"}
word = "cat"

if word in english:
    print(english[word])
else:
    print("Sorry, I don't know that word.")
````

## 7. Xóa một cặp

Để xóa một cặp, hãy dùng `del` và khóa:

```python
del tên_dictionary[khóa]
```

`del` là viết tắt của **delete** (xóa). Nó xóa khóa **và** giá trị của nó cùng nhau.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

del scores["Tom"]           # Xóa khóa "Tom" và giá trị 92 của nó
print(scores)
print(len(scores))          # Bây giờ còn 2 cặp
````

````python
# Xóa một khóa không tồn tại sẽ gây ra lỗi (chúng ta sẽ thấy ở phần 10)
# Vì vậy hãy kiểm tra bằng in trước
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
    print(name, "was deleted.")
else:
    print(name, "is not in the dictionary.")
````

## 8. Dùng vòng lặp for với dictionary

Hãy nhớ lại cách vòng lặp for hoạt động với list:

```python
for score in score_list:
    print(score)
```

- Ở mỗi lần lặp, một giá trị từ list được lưu vào biến `score`.
- Dòng `for` kết thúc bằng dấu hai chấm `:`.
- Đoạn mã cần lặp lại được **thụt lề** (4 dấu cách). Python dùng thụt lề để biết những dòng nào thuộc về vòng lặp.
- Dòng đầu tiên không thụt lề nằm ngoài vòng lặp, và chỉ chạy một lần sau vòng lặp.

Với dictionary, vòng lặp for cho bạn các **khóa**, lần lượt từng cái một.

```python
for name in scores:
    print(name)          # Anna, Tom, Bill
```

Và khi đã có khóa, bạn có thể lấy giá trị bằng `scores[name]`, giống như ở phần 4.

````python
# Vòng lặp for cho ra các KHÓA của dictionary
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:         # name lần lượt là "Anna", rồi "Tom", rồi "Bill"
    print(name)
````

````python
# Dùng khóa để lấy giá trị
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(scores[name])     # scores["Anna"], rồi scores["Tom"], rồi scores["Bill"]
````

````python
# Khóa và giá trị cùng trên một dòng (print với dấu phẩy)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])
````

### Tên của biến lặp

Biến đứng sau `for` (ở đây là `name`) là một biến mới. Bạn có thể đặt cho nó bất kỳ tên nào. Hãy chọn tên nói lên **khóa** là gì: tên người → `name`, quốc gia → `country`, từ → `word`.

````python
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

for country in capitals:            # Các khóa là quốc gia, nên chúng ta gọi nó là country
    print(country, capitals[country])
````

### Thụt lề quyết định cái gì được lặp lại

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])       # Được thụt lề -> lặp lại cho mỗi khóa
print("That's everyone!")           # Không thụt lề -> bên ngoài vòng lặp -> chạy một lần
````

### Cộng dồn các giá trị

Hãy nhớ **tổng tích lũy**:

1. **Trước vòng lặp:** `total = 0`
2. **Bên trong vòng lặp:** cộng giá trị hiện tại vào → `total = total + scores[name]`
3. **Sau vòng lặp:** `total` chứa tổng

`total = total + scores[name]` có nghĩa là: tính vế phải trước (tổng cũ + điểm của học sinh này), rồi lưu kết quả trở lại vào `total`. Cách viết ngắn là `total += scores[name]`.

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0                           # Trước vòng lặp: bắt đầu từ 0

for name in scores:
    total = total + scores[name]    # tổng cũ + điểm của học sinh này

print("Total:", total)              # Sau vòng lặp (không thụt lề): 85 + 92 + 78 = 255
````

````python
# Trung bình = tổng / số lượng giá trị
# len(scores) là số cặp (= số học sinh)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:
    total += scores[name]           # giống với: total = total + scores[name]

average = total / len(scores)       # Sau vòng lặp: 255 / 3
print("Average:", average)
````

### if bên trong vòng lặp

Khi đặt `if` bên trong `for`, sẽ có hai mức thụt lề: 4 dấu cách cho `if` (bên trong `for`), và 8 dấu cách cho đoạn mã bên trong `if`.

````python
# Chỉ in những học sinh đậu (từ 60 trở lên)
scores = {"Anna": 85, "Tom": 42, "Bill": 78, "Kate": 55}

for name in scores:
    if scores[name] >= 60:          # 4 dấu cách: kiểm tra điểm của từng học sinh
        print(name, "passed")       # 8 dấu cách: chỉ khi điều kiện là True
````

````python
# Hôm nay ai đến lớp?
attendance = {"Anna": True, "Tom": False, "Bill": True}

for name in attendance:
    if attendance[name]:            # Giá trị đã là True hoặc False, nên có thể dùng trực tiếp làm điều kiện
        print(name, "is here.")
    else:
        print(name, "is absent.")
````

## 9. List và dictionary

| | List | Dictionary |
|---|---|---|
| Tạo | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92}` |
| Dấu ngoặc để tạo | `[ ]` | `{ }` |
| Tìm giá trị bằng | **chỉ số** (số chỉ vị trí) | **khóa** (tên) |
| Lấy giá trị | `score_list[1]` | `score_dict["Tom"]` |
| Thay đổi giá trị | `score_list[1] = 95` | `score_dict["Tom"] = 95` |
| Thêm | `score_list.append(88)` | `score_dict["Kate"] = 88` |
| Xóa | `score_list.remove(92)` | `del score_dict["Tom"]` |
| `len()` đếm | giá trị | cặp |
| `in` kiểm tra | giá trị | **khóa** |
| `for x in ...` cho ra | giá trị | **khóa** |

**Khi nào dùng cái nào?**
- Nếu **thứ tự** quan trọng, hoặc các giá trị không cần tên → **list** (ví dụ: điểm bài kiểm tra hôm nay, theo thứ tự)
- Nếu bạn muốn tìm giá trị **theo tên** của nó → **dictionary** (ví dụ: điểm của từng học sinh, một từ và nghĩa của nó)

## 10. Những lỗi thường gặp

Đây là những lỗi mà hầu như ai cũng mắc phải khi học dictionary. Với mỗi lỗi, chúng ta sẽ xem:

- ❌ **Mã sai**
- 🔍 **Sai ở đâu**
- ✅ **Cách làm đúng**

Mã gây ra lỗi được viết dưới dạng chú thích (có `#` ở đầu) để không làm notebook dừng lại. Nếu muốn xem thông báo lỗi thật, hãy xóa `#`, chạy ô đó, rồi đặt `#` lại.

### Lỗi 1: Dùng một khóa không tồn tại

````python
# ❌ Sai
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores["Kate"])     # KeyError: 'Kate'
# print(scores["tom"])      # KeyError: 'tom'
````

🔍 **Sai ở đâu?**

- `"Kate"` không phải là khóa trong dictionary. Python không tìm thấy nó, nên báo lỗi `KeyError` ("không có khóa như vậy").
- `"tom"` (chữ t thường) và `"Tom"` (chữ T hoa) là hai chuỗi **khác nhau**. Khóa phải được viết **chính xác** như vậy, kể cả chữ hoa và chữ thường.

✅ **Cách làm đúng:** Viết khóa chính xác như trong dictionary. Nếu không chắc khóa có tồn tại không, hãy kiểm tra bằng `in` trước.

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # Giống hệt khóa

name = "Kate"
if name in scores:          # Kiểm tra trước
    print(scores[name])
else:
    print(name, "is not in the dictionary.")
````

### Lỗi 2: Dùng dấu ngoặc vuông để tạo dictionary

````python
# ❌ Sai
# scores = ["Anna": 85, "Tom": 92]      # SyntaxError
````

🔍 **Sai ở đâu?**

- Dấu ngoặc vuông `[ ]` dùng để tạo **list**. List chỉ chứa giá trị, nên Python không hiểu `"Anna": 85` bên trong `[ ]`.
- Lỗi là `SyntaxError` ("ngữ pháp của mã bị sai").

✅ **Cách làm đúng:** Dùng **dấu ngoặc nhọn** `{ }` để tạo dictionary. (Để **lấy** giá trị, bạn vẫn dùng dấu ngoặc vuông: `scores["Anna"]`.)

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom": 92}    # Tạo: dấu ngoặc nhọn
print(scores["Anna"])               # Lấy: dấu ngoặc vuông
````

### Lỗi 3: Quên dấu ngoặc kép quanh khóa

````python
# ❌ Sai
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[Anna])       # NameError: name 'Anna' is not defined
````

🔍 **Sai ở đâu?**

- Khóa là **chuỗi** `"Anna"`, có dấu ngoặc kép.
- Không có dấu ngoặc kép, Python nghĩ `Anna` là tên của một **biến**. Nhưng không có biến nào tên `Anna`.
- Lỗi là `NameError` ("không có biến nào mang tên này").

✅ **Cách làm đúng:** Đặt dấu ngoặc kép quanh khóa là chuỗi. (Chỉ không cần dấu ngoặc kép khi khóa được lưu trong một biến, như `scores[name]` ở phần 4.)

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # Khóa là chuỗi, có dấu ngoặc kép

name = "Anna"
print(scores[name])         # Biến, không có dấu ngoặc kép (biến chứa "Anna")
````

### Lỗi 4: Dùng số chỉ số giống như list

````python
# ❌ Sai: chúng ta muốn điểm của học sinh đầu tiên
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[0])          # KeyError: 0
````

🔍 **Sai ở đâu?**

- Trong list, `[0]` có nghĩa là "giá trị đầu tiên". Nhưng dictionary không dùng số chỉ số. Nó dùng **khóa**.
- Vì vậy Python tìm một **khóa** tên là `0`. Không có khóa như vậy, nên nó báo lỗi `KeyError: 0`.

✅ **Cách làm đúng:** Dùng khóa (tên).

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # Dùng khóa, không dùng số
````

### Lỗi 5: Viết cùng một khóa hai lần

````python
# ❌ Sai: hai học sinh khác nhau cùng tên Tom?
scores = {"Anna": 85, "Tom": 92, "Tom": 60}

print(scores)               # {'Anna': 85, 'Tom': 60} -> Tom đầu tiên (92) đã biến mất!
print(len(scores))          # 2, không phải 3
````

🔍 **Sai ở đâu?**

- Trong dictionary, mỗi khóa **chỉ được xuất hiện một lần**.
- Nếu bạn viết lại cùng một khóa, nó hoạt động giống như phần 5: khóa đã tồn tại, nên giá trị cũ bị **thay thế**. 92 bị thay bằng 60.
- Không có lỗi nào, nên rất dễ bỏ sót.

✅ **Cách làm đúng:** Làm cho mọi khóa đều khác nhau. Nếu hai người trùng tên, hãy đặt cho họ những khóa khác nhau.

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom K.": 92, "Tom L.": 60}

print(scores)
print(len(scores))          # 3
````

### Lỗi 6: Tưởng rằng in kiểm tra giá trị

````python
# ❌ Sai: chúng ta muốn biết có ai được 92 điểm không
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(92 in scores)         # False -> nhưng Tom được 92 điểm!
````

🔍 **Sai ở đâu?**

- Với dictionary, `in` **chỉ kiểm tra khóa** (tên), không kiểm tra giá trị (điểm).
- 92 là giá trị, không phải khóa, nên kết quả là `False`.
- Không có lỗi nào, nhưng câu trả lời bị sai.

✅ **Cách làm đúng:** Duyệt dictionary bằng vòng lặp for, và so sánh từng giá trị.

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    if scores[name] == 92:      # So sánh từng GIÁ TRỊ với 92
        print(name, "got 92 points.")
````

### Lỗi 7: Tưởng rằng vòng lặp for cho ra giá trị

````python
# ❌ Sai: chúng ta muốn cộng tất cả các điểm
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

# for score in scores:
#     total += score        # TypeError: unsupported operand type(s) for +=: 'int' and 'str'
````

🔍 **Sai ở đâu?**

- Với dictionary, vòng lặp for cho ra các **khóa**, không phải giá trị. Vì vậy `score` là `"Anna"`, `"Tom"`, `"Bill"`, không phải 85, 92, 78.
- `total += score` cố cộng một chuỗi (`"Anna"`) vào một số (`0`). Python không thể cộng số với chuỗi, nên báo lỗi `TypeError`.
- Đặt tên biến là `score` càng làm rối hơn. Thực ra nó là một cái tên.

✅ **Cách làm đúng:** Đặt tên biến lặp theo khóa (`name`), và dùng `scores[name]` để lấy giá trị.

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:             # name là khóa
    total += scores[name]       # scores[name] là giá trị

print("Total:", total)
````

### Lỗi 8: Xóa một khóa không tồn tại

````python
# ❌ Sai
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# del scores["Kate"]        # KeyError: 'Kate'
````

🔍 **Sai ở đâu?**

- Không có khóa `"Kate"` trong dictionary, nên không có gì để xóa.
- Giống như Lỗi 1, Python báo lỗi `KeyError`.

✅ **Cách làm đúng:** Kiểm tra bằng `in` trước, và chỉ xóa khi khóa tồn tại.

````python
# ✅ Đúng
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
else:
    print(name, "is not in the dictionary, so nothing was deleted.")
````
