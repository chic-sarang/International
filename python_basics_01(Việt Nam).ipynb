## 1. Biến và kiểu dữ liệu

Biến là một chiếc hộp để lưu trữ giá trị. Khi bạn đặt tên cho hộp, sau này bạn chỉ cần gọi tên đó để lấy giá trị bên trong ra sử dụng. Nếu bạn đặt một giá trị mới vào, nội dung bên trong hộp cũng sẽ thay đổi.

Tuy nhiên, giá trị có thể đặt vào trong hộp có nhiều loại khác nhau. Loại giá trị này được gọi là kiểu dữ liệu. Các kiểu dữ liệu tiêu biểu trong Python như sau.

- `str` : chuỗi ký tự (chữ)
- `int` : số nguyên
- `float` : số thực (số có phần thập phân)
- `bool` : boolean (đúng/sai)

Lưu ý, `print()` là lệnh hiển thị giá trị trong ngoặc lên màn hình.

```python
# = không có nghĩa là 'bằng nhau', mà có nghĩa là 'lưu giá trị bên phải vào hộp mang tên bên trái'
# Tên biến (name, age, height, is_student) là tên được đặt cho hộp

# Chuỗi ký tự (str) chỉ cần bọc trong dấu ngoặc kép. Bạn có thể dùng cả dấu ngoặc kép (" ") và dấu ngoặc đơn (' ')
name = "Bill"                    # Từ được bọc trong dấu ngoặc kép
city = 'Seoul'                    # Từ được bọc trong dấu ngoặc đơn
greeting = "Hello, nice to meet you"  # Một câu cũng vẫn là một chuỗi ký tự
zip_code = "06236"               # Dù trông giống số, nhưng nếu nằm trong dấu ngoặc thì đó là chuỗi ký tự

# Số nguyên (int) là số không có phần thập phân. Số dương, số âm, và 0 đều là số nguyên
age = 20              # Số nguyên dương
temperature = -5      # Số nguyên âm cũng được
score = 0             # 0 cũng là số nguyên

# Số thực (float) là số có phần thập phân
height = 175.5        # Số thực dương
account_balance = -3.75  # Số thực âm cũng được
pi_value = 3.14159    # Dù có nhiều chữ số sau dấu thập phân cũng không sao

# Boolean (bool) chỉ có hai giá trị: True (đúng) hoặc False (sai) (chữ cái đầu luôn viết hoa)
is_student = True     # là học sinh/sinh viên
is_married = False    # chưa kết hôn
has_passed = True     # đã thi đỗ

# print() là lệnh hiển thị giá trị lên màn hình
print(name)
print(city)
print(greeting)
print(zip_code)
print(age)
print(temperature)
print(score)
print(height)
print(account_balance)
print(pi_value)
print(is_student)
print(is_married)
print(has_passed)
```

```python
# Giá trị của biến có thể thay đổi được (ví dụ với số nguyên)
age = 20
print(age)

age = 21   # Nếu lưu giá trị mới vào cùng một biến, giá trị cũ sẽ biến mất và được thay bằng giá trị mới
print(age)

age = age + 1   # Cũng có thể tạo giá trị mới dựa trên giá trị cũ (21 + 1 = 22)
print(age)
```

```python
# Giá trị của biến có thể thay đổi được (ví dụ với boolean)
is_raining = False
print(is_raining)

is_raining = True   # Nếu lưu giá trị mới vào cùng một biến, giá trị cũ sẽ biến mất và được thay bằng giá trị mới
print(is_raining)

is_raining = not is_raining   # Dùng not sẽ đảo ngược: True thành False, False thành True
print(is_raining)
```

```python
# Giá trị của biến có thể thay đổi được (ví dụ với chuỗi ký tự)
city = "Seoul"
print(city)

city = "Busan"   # Nếu lưu giá trị mới vào cùng một biến, giá trị cũ sẽ biến mất và được thay bằng giá trị mới
print(city)

city = city + " Travel"   # Chuỗi ký tự có thể nối với nhau bằng dấu + ("Busan" + " Travel" = "Busan Travel")
print(city)
```

```python
# Giá trị của biến có thể thay đổi được (ví dụ với số thực)
height = 175.5
print(height)

height = 178.2   # Nếu lưu giá trị mới vào cùng một biến, giá trị cũ sẽ biến mất và được thay bằng giá trị mới
print(height)

height = height + 0.5   # Cũng có thể tạo giá trị mới dựa trên giá trị cũ (178.2 + 0.5 = 178.7)
print(height)
```

## 2. Toán tử cơ bản

Các ký hiệu dùng để tính toán số hoặc so sánh giá trị.

```python
# Bốn phép tính cơ bản
a = 10
b = 3

print(a + b)   # Cộng
print(a - b)   # Trừ
print(a * b)   # Nhân
print(a / b)   # Chia (kết quả là số thực)
print(a // b)  # Chia lấy phần nguyên (chỉ lấy thương)
print(a % b)   # Chia lấy phần dư
```

```python
# Toán tử so sánh -> kết quả luôn là True hoặc False
print(a > b)   # a lớn hơn b
print(a < b)   # a nhỏ hơn b
print(a == b)  # a bằng b
print(a != b)  # a khác b
```
