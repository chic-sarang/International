## 1. Что такое срез?

Начнём с того, что мы уже знаем.

**Список** хранит много значений по порядку, внутри квадратных скобок `[ ]`:

```python
scores = [85, 92, 78, 90, 66]
```

У каждого значения есть номер позиции - **индекс**. Самое важное правило: **Python начинает счёт с 0, а не с 1.**

| Значение | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |

Чтобы получить **одно** значение, мы пишем имя списка и индекс в квадратных скобках: `scores[0]` - это `85`, `scores[2]` - это `78`.

А что, если нужно **несколько значений сразу**? Например, только оценки первых 3 учеников?

С тем, что мы знаем, пришлось бы брать их по одному: `scores[0]`, `scores[1]`, `scores[2]`.

**Срез** позволяет взять **часть** списка за один шаг. Слово "срез" (slice) означает "кусок, отрезанный от чего-то", как ломтик хлеба или кусок пиццы. Мы отрезаем кусок от списка.

Результат среза - это **новый список**, в котором есть только эта часть.

````python
scores = [85, 92, 78, 90, 66]

# Получаем ОДНО значение по индексу (то, что мы уже знаем)
print(scores[0])        # индекс 0 -> 1-е значение -> 85
print(scores[1])        # индекс 1 -> 2-е значение -> 92
print(scores[2])        # индекс 2 -> 3-е значение -> 78
````

````python
scores = [85, 92, 78, 90, 66]

# Получаем НЕСКОЛЬКО значений сразу с помощью среза (то, что мы изучаем сегодня)
# scores[0:3] означает "от индекса 0 до индекса 3 (3 не включается)"
print(scores[0:3])      # [85, 92, 78] -> новый список из первых 3 значений
````

Внимательно посмотрите на два результата выше:

- `scores[0]` даёт **одно значение**: `85` (без квадратных скобок)
- `scores[0:3]` даёт **список**: `[85, 92, 78]` (с квадратными скобками)

Это различие очень важно. Индекс даёт значение. Срез даёт список.

## 2. Базовая форма: `список[start:end]`

```python
имя_списка[start:end]
```

- `start`: индекс, с которого срез **начинается** (это значение **включается**)
- `end`: индекс, на котором срез **останавливается** (это значение **НЕ** включается)
- `:` двоеточие между двумя числами

`scores[1:4]` можно прочитать так: **"от индекса 1 до индекса 4, не включая 4"**.

### end не включается

Это то, что больше всего путает новичков. Разберём медленно, с помощью таблицы.

```python
scores = [85, 92, 78, 90, 66]
```

| Индекс | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Значение | 85 | 92 | 78 | 90 | 66 |
| Входит в `scores[1:4]`? | ❌ | ✅ | ✅ | ✅ | ❌ |

`scores[1:4]` → индексы 1, 2, 3 → `[92, 78, 90]`. На индексе 4 срез **останавливается**, поэтому 66 не входит.

**Вы уже встречали это правило?** Да! В Уроке 3 `range(1, 4)` давал числа 1, 2, 3, а 4 **не** включалось. Срез использует точно такое же правило: **start включается, end - нет**.

### Сколько значений вы получите?

Поскольку end не включается, количество значений равно просто **end - start**.

- `scores[1:4]` → 4 - 1 = **3** значения
- `scores[0:2]` → 2 - 0 = **2** значения
- `scores[2:5]` → 5 - 2 = **3** значения

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[1:4])      # индексы 1, 2, 3 (4 не включается) -> [92, 78, 90]
````

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[0:2])      # индексы 0, 1 -> [85, 92]
print(scores[2:5])      # индексы 2, 3, 4 -> [78, 90, 66]
print(scores[3:4])      # только индекс 3 -> [90] (список из одного значения)
````

````python
# То же правило, что и у range() в Уроке 3: end не включается
# range(1, 4) даёт 1, 2, 3
for i in range(1, 4):
    print(i)
````

````python
# Со строками всё работает точно так же
fruits = ["apple", "banana", "cherry", "grape", "melon"]
#  индекс:     0        1         2        3        4

print(fruits[0:2])      # индексы 0, 1 -> ['apple', 'banana']
print(fruits[1:3])      # индексы 1, 2 -> ['banana', 'cherry']
print(fruits[2:5])      # индексы 2, 3, 4 -> ['cherry', 'grape', 'melon']
````

### Считаем количество значений с помощью len()

Помните `len()`? Она показывает, **сколько значений** в списке. `len([85, 92, 78])` равно `3`.

Срез - это список, поэтому к нему тоже можно применить `len()`. Проверим правило "end - start".

````python
scores = [85, 92, 78, 90, 66]

print(len(scores))          # Во всём списке 5 значений
print(len(scores[1:4]))     # 4 - 1 = 3 значения
print(len(scores[0:2]))     # 2 - 0 = 2 значения
````

### Срез - это новый список. Исходный список не меняется.

Срез **копирует** часть списка в **новый список**. Исходный список остаётся точно таким же.

Если хотите использовать срез позже, сохраните его в переменную, как любое другое значение. Помните: `=` означает **"сохрани значение справа в коробку слева"**.

````python
scores = [85, 92, 78, 90, 66]

first_three = scores[0:3]   # Сохраняем новый список в переменную first_three

print(first_three)          # [85, 92, 78] -> новый список
print(scores)               # [85, 92, 78, 90, 66] -> исходный список остался полным
````

````python
# Новый список - это настоящий список, поэтому его тип - list
# type() показывает, какого вида значение
scores = [85, 92, 78, 90, 66]
first_three = scores[0:3]

print(type(first_three))    # <class 'list'>
print(type(scores[0]))      # <class 'int'> -> индекс даёт одно значение (целое число), а не список
````

````python
# К новому списку тоже можно применить индекс
# Помните: новый список снова начинается с индекса 0!
scores = [85, 92, 78, 90, 66]
last_two = scores[3:5]      # [90, 66]

print(last_two[0])          # 1-е значение НОВОГО списка -> 90 (в исходном списке это был индекс 3)
print(last_two[1])          # 2-е значение НОВОГО списка -> 66
````

## 3. Пропуск start или end

Не обязательно всегда писать оба числа. Python сам подставит недостающее число.

| Как писать | Значение | Python понимает как |
|---|---|---|
| `scores[:3]` | от **начала** до индекса 3 (не включая) | `scores[0:3]` |
| `scores[2:]` | от индекса 2 до **конца** | `scores[2:5]` (когда значений 5) |
| `scores[:]` | от начала до конца (весь список) | `scores[0:5]` |

- Нет start → начинаем с **начала** (индекс 0)
- Нет end → идём до самого **конца** (последнее значение включается)

**Чем это удобно?**
- `scores[:3]` короче, чем `scores[0:3]`, и легко читается: "первые 3".
- `scores[2:]` работает, даже если не знать длину списка. Не нужно считать значения.

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

# Нет start -> с начала
print(scores[:3])       # То же самое, что scores[0:3] -> [85, 92, 78]
print(scores[0:3])      # Тот же результат
````

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

# Нет end -> до конца (последнее значение ВКЛЮЧАЕТСЯ)
print(scores[2:])       # индексы 2, 3, 4 -> [78, 90, 66]
print(scores[1:])       # Всё, кроме первого -> [92, 78, 90, 66]
````

````python
# [:n] = "первые n значений"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[:1])       # Первое 1 -> ['apple']
print(fruits[:2])       # Первые 2 -> ['apple', 'banana']
print(fruits[:4])       # Первые 4 -> ['apple', 'banana', 'cherry', 'grape']
````

````python
# [n:] = "пропусти первые n значений и возьми остальные"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[1:])       # Пропускаем 1 -> ['banana', 'cherry', 'grape', 'melon']
print(fruits[3:])       # Пропускаем 3 -> ['grape', 'melon']
````

````python
# [n:] работает, даже когда список становится длиннее
# append() добавляет значение в конец списка
scores = [85, 92, 78]
scores.append(90)           # Теперь: [85, 92, 78, 90]
scores.append(66)           # Теперь: [85, 92, 78, 90, 66]

print(scores[1:])           # Нам не нужно было знать, что теперь значений 5
````

````python
# [:] = весь список (его копия)
scores = [85, 92, 78, 90, 66]
print(scores[:])
````

````python
# Делим список на две части
# В [:2] и [2:] одно и то же число, поэтому вместе они дают весь список
# (потому что 2 НЕ входит в [:2], но ВХОДИТ в [2:])
scores = [85, 92, 78, 90, 66]

front = scores[:2]
back = scores[2:]

print(front)            # [85, 92]
print(back)             # [78, 90, 66]
````

## 4. Срезы с отрицательными индексами

Вспомним **отрицательный индекс**. Python умеет также считать **с конца списка в обратную сторону**:

- `-1` - это **последнее** значение
- `-2` - предпоследнее значение
- `-3` - третье с конца значение, и так далее

| Значение | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |
| Отрицательный индекс | -5 | -4 | -3 | -2 | -1 |

Главная причина использовать отрицательный индекс: **можно добраться до конца списка, не считая, сколько в нём значений**. `-1` - всегда последнее значение, будь в списке 5 значений или 500.

Отрицательные индексы работают и в срезах. Правило то же: **start включается, end - нет**.

| Как писать | Значение |
|---|---|
| `scores[-3:]` | **последние 3** значения |
| `scores[-2:]` | **последние 2** значения |
| `scores[:-1]` | всё, **кроме последнего** значения |
| `scores[:-2]` | всё, **кроме последних 2** значений |

````python
scores = [85, 92, 78, 90, 66]
#  отриц. индекс: -5  -4  -3  -2  -1

# [-n:] = "последние n значений"
print(scores[-3:])      # От -3 до конца -> [78, 90, 66]
print(scores[-2:])      # От -2 до конца -> [90, 66]
print(scores[-1:])      # От -1 до конца -> [66] (список из одного значения)
````

````python
# Сравните: индекс даёт одно значение, срез даёт список
scores = [85, 92, 78, 90, 66]

print(scores[-1])       # 66   -> одно значение (последнее)
print(scores[-1:])      # [66] -> список с последним значением
````

````python
scores = [85, 92, 78, 90, 66]
#  отриц. индекс: -5  -4  -3  -2  -1

# [:-n] = "всё, кроме последних n значений"
print(scores[:-1])      # До -1, но -1 НЕ включается -> [85, 92, 78, 90]
print(scores[:-2])      # До -2, но -2 НЕ включается -> [85, 92, 78]
````

````python
# Последние 3 значения, при любой длине списка
# Нам не нужно знать, сколько значений в списке
short_list = [1, 2, 3, 4]
long_list = [10, 20, 30, 40, 50, 60, 70, 80]

print(short_list[-3:])  # [2, 3, 4]
print(long_list[-3:])   # [60, 70, 80]
````

````python
# Пример из жизни: 3 последних сообщения
# Новые сообщения добавляются в конец с помощью append(), поэтому самые новые - в конце
messages = ["Hi", "How are you?", "Fine, thanks", "See you", "Bye"]

print(messages[-3:])    # 3 самых новых сообщения
````

## 5. Срез в цикле for

Срез - это список. Значит, всё, что можно делать со списком, можно делать и со срезом. В том числе использовать **цикл for**.

Вспомним, как работает цикл for:

```python
for score in scores:
    print(score)
```

- `for score in scores:` означает **"для каждого score в scores выполни код ниже"**.
- При каждом повторении одно значение из списка сохраняется в переменную `score`, от первого до последнего.
- Строка должна заканчиваться двоеточием `:`.
- Повторяемый код должен быть **с отступом** (4 пробела в начале). По отступу Python понимает, какие строки относятся к циклу. Эта группа строк с отступом называется **блоком**.
- Первая строка **без** отступа находится вне цикла и выполняется только один раз, после окончания цикла.

Теперь, если поставить после `in` **срез**, цикл пройдёт **только по этой части** списка.

```python
for score in scores[:3]:      # только первые 3 значения
    print(score)
```

````python
# Весь список
scores = [85, 92, 78, 90, 66]

for score in scores:            # 5 значений -> цикл выполняется 5 раз
    print(score)
````

````python
# Только первые 3 значения
scores = [85, 92, 78, 90, 66]

for score in scores[:3]:        # scores[:3] - это [85, 92, 78] -> цикл выполняется 3 раза
    print(score)                # С отступом (4 пробела) -> внутри цикла -> повторяется
````

````python
# Только последние 2 значения
scores = [85, 92, 78, 90, 66]

for score in scores[-2:]:       # scores[-2:] - это [90, 66] -> цикл выполняется 2 раза
    print(score)
````

````python
# Всё, кроме первого
# Удобно, когда первое значение - это заголовок, а не настоящие данные
row = ["Scores", 85, 92, 78]

for value in row[1:]:           # row[1:] - это [85, 92, 78] -> пропускаем "Scores"
    print(value)
````

### У переменной цикла может быть любое имя

Помните: переменная после `for` (здесь `score`) - это **новая переменная**, которую создаёт цикл. Ей можно дать любое имя. Но хорошая привычка - использовать **единственное число** от имени списка: `scores` → `score`, `fruits` → `fruit`. Тогда легко отличить "весь список" от "одного значения".

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[1:4]:       # fruits[1:4] - это ['banana', 'cherry', 'grape']
    print(fruit)                # fruit (одно значение), fruits (весь список)
````

### Отступ решает, что повторяется

Как и в Уроке 5, строки с отступом повторяются, а первая строка без отступа выполняется только один раз после цикла.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[:2]:        # Только 'apple' и 'banana'
    print("I like", fruit)      # С отступом -> выполняется 2 раза
print("That's all!")            # Без отступа -> вне цикла -> выполняется один раз
````

**О `print("I like", fruit)`:** когда вы ставите в `print()` несколько значений через **запятую** `,`, они выводятся в одну строку, и Python автоматически ставит между ними **один пробел**. Текст в кавычках выводится как есть, а вместо переменной выводится её значение.

## 6. Вычисления со срезом

### Складываем часть списка

В Уроке 5 мы складывали все значения списка с помощью **накопителя**:

1. **Перед циклом:** создайте переменную и присвойте ей 0 → `total = 0`
2. **Внутри цикла:** прибавляйте текущее значение → `total = total + score`
3. **После цикла:** в переменной лежит сумма

Вспомните, что значит `total = total + score`: `=` значит "сохрани". Python сначала вычисляет правую часть (старая сумма + текущая оценка), а потом сохраняет результат обратно в `total`.

Со срезом можно сложить **только часть** списка. Просто поставьте срез после `in`.

````python
# Складываем только первые 3 оценки
scores = [85, 92, 78, 90, 66]
total = 0                           # Перед циклом: начинаем с 0

for score in scores[:3]:            # Только 85, 92, 78
    total = total + score           # старая сумма + текущая оценка -> сохраняем обратно в total

print("Total of the first 3:", total)   # После цикла (без отступа): 85 + 92 + 78 = 255
````

````python
# Посмотрим, как меняется total
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total = total + score
    print("score:", score, "total:", total)     # С отступом -> выводится при каждом повторении
````

**Вспомните короткую запись `+=`:** `total += score` означает точно то же самое, что `total = total + score`. "Прибавь правую часть к переменной и сохрани результат обратно в неё."

````python
# То же самое, но с +=
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total += score                  # то же самое, что: total = total + score

print("Total of the first 3:", total)
````

````python
# Складываем только последние 2 оценки
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[-2:]:           # Только 90, 66
    total += score                  # то же самое, что: total = total + score

print("Total of the last 2:", total)    # 90 + 66 = 156
````

### Среднее значение части списка

**Среднее** - это сумма, делённая на количество значений: **среднее = сумма / количество значений**

Количество значений мы получаем с помощью `len()`. Но будьте внимательны: нужно количество значений **в срезе**, а не во всём списке. Поэтому применяем `len()` к срезу.

И помните: вычисляйте среднее **после** цикла (без отступа), когда сумма готова.

````python
# Среднее первых 3 оценок
scores = [85, 92, 78, 90, 66]
first_three = scores[:3]            # Сохраняем срез в переменную: [85, 92, 78]
total = 0

for score in first_three:
    total += score                  # то же самое, что: total = total + score

average = total / len(first_three)  # 255 / 3 (длина СРЕЗА, а не всего списка)
print("Total:", total)
print("Count:", len(first_three))
print("Average:", average)
````

````python
# Сравниваем среднее первой и второй половины
scores = [60, 65, 70, 85, 90, 95]
first_half = scores[:3]             # [60, 65, 70]
second_half = scores[3:]            # [85, 90, 95]

total = 0
for score in first_half:
    total += score
print("First half average:", total / len(first_half))

total = 0                           # Для второй половины снова начинаем с 0!
for score in second_half:
    total += score
print("Second half average:", total / len(second_half))
````

### Наибольшее значение в части списка

В Уроке 5 мы находили наибольшее значение так, как это сделал бы человек на бумаге:

1. Запомнить **первое значение** как "самое большое на данный момент".
2. Смотреть на каждое значение. Если оно больше запомненного, запомнить его вместо старого.
3. То, что запомнено в конце, и есть самое большое.

Со срезом можно найти наибольшее значение **только в части** списка.

````python
# Самая высокая оценка среди последних 3 учеников
scores = [95, 82, 74, 88, 79]
last_three = scores[-3:]            # [74, 88, 79]

highest = last_three[0]             # Шаг 1: начинаем с первого значения СРЕЗА (74)

for score in last_three:
    if score > highest:             # Шаг 2: больше запомненного?
        highest = score             #        если да, запоминаем его вместо старого

print("Highest of the last 3:", highest)    # Шаг 3: 88
````

## 7. Срез вместе с if, in и append()

### if внутри цикла

Когда `if` находится внутри цикла `for`, получается **два уровня** отступа:

```python
for score in scores[:3]:      # без пробелов
    if score >= 80:           # 4 пробела -> внутри for (выполняется для каждой оценки)
        print(score)          # 8 пробелов -> внутри if (выполняется, только когда условие равно True)
```

Вспомните, как работает `if`: код с отступом под ним выполняется **только когда условие равно True**.

````python
# Из первых 4 оценок выводим только те, что 80 или выше
scores = [85, 72, 91, 78, 95, 60]

for score in scores[:4]:            # Только 85, 72, 91, 78
    if score >= 80:                 # 4 пробела: проверяем каждую оценку
        print(score)                # 8 пробелов: выводим, только если 80 или выше
````

````python
# Подсчёт: сколько из последних 3 оценок сданы (60 и выше)?
# Подсчёт работает как накопитель, только прибавляем 1, а не значение
scores = [85, 42, 78, 55, 90]
pass_count = 0                      # Перед циклом: начинаем с 0

for score in scores[-3:]:           # Только 78, 55, 90
    if score >= 60:
        pass_count += 1             # то же самое, что: pass_count = pass_count + 1

print("Passed among the last 3:", pass_count)
````

### in

`значение in список` проверяет, есть ли значение в списке, и даёт `True` или `False`. Срез - это список, поэтому `in` можно использовать и со срезом. Тогда мы проверяем **только часть** списка.

````python
# Тройка лучших учеников (список уже по порядку: 1-е место, 2-е место, ...)
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]

print("Tom" in ranking[:3])         # Tom в тройке лучших? -> True
print("Kate" in ranking[:3])        # Kate в тройке лучших? -> False
print("Kate" in ranking)            # Kate есть во всём списке? -> True
````

````python
# Используем in как условие if
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]
name = "Bill"

if name in ranking[:3]:             # Bill в тройке лучших?
    print(name, "gets a medal!")
else:
    print(name, "doesn't get a medal.")
````

### append(): создаём новый список

Вспомните `append()`: `список.append(значение)` добавляет значение в **конец** списка.

Можно начать с пустого списка `[]` и внутри цикла добавлять с помощью `append()` только нужные значения из среза.

````python
# Из первых 4 оценок создаём новый список только со сданными
scores = [85, 42, 78, 55, 90, 30]
passed = []                         # Перед циклом: пустой список

for score in scores[:4]:            # Только 85, 42, 78, 55
    if score >= 60:
        passed.append(score)        # Добавляем в конец passed

print(passed)                       # [85, 78]
````

````python
# Создаём новый список, где последние 3 цены умножены на 2
prices = [1000, 2500, 3000, 1500, 4000]
doubled = []

for price in prices[-3:]:           # Только 3000, 1500, 4000
    doubled.append(price * 2)       # Сначала вычисляется price * 2, потом добавляется

print(doubled)
````

## 8. Частые ошибки

Эти ошибки делает почти каждый, кто изучает срезы. Для каждой мы посмотрим:

- ❌ **Неправильный код**
- 🔍 **Что пошло не так**
- ✅ **Правильный способ**

Некоторый неправильный код вызывает ошибку. Такой код записан в виде комментария (с `#` в начале), чтобы он не останавливал блокнот. Если хотите увидеть настоящее сообщение об ошибке, уберите `#`, запустите ячейку, а потом верните `#` обратно.

Другой неправильный код **не вызывает ошибку, но даёт неправильный результат**. Такие ошибки опаснее, потому что Python вас не предупреждает.

### Ошибка 1: Думать, что end включается

````python
# ❌ Неправильно: нам нужны значения с индексами 1, 2, 3
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[1:3])      # [92, 78] -> только 2 значения! 90 (индекс 3) пропало
````

🔍 **Что пошло не так?**

- `scores[1:3]` означает "от индекса 1 до индекса 3, **не включая** 3".
- Поэтому мы получаем только индексы 1 и 2. На индексе 3 срез **останавливается**.
- Ошибки нет, поэтому это легко пропустить.

✅ **Правильный способ:** end должен быть **на 1 больше** последнего нужного индекса. Чтобы включить индекс 3, пишите `4`. (Как с `range()` в Уроке 3: чтобы включить 3, пишут `range(1, 4)`.)

````python
# ✅ Правильно
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # индексы 1, 2, 3 -> [92, 78, 90]
````

### Ошибка 2: Запятая вместо двоеточия

````python
# ❌ Неправильно
scores = [85, 92, 78, 90, 66]

# print(scores[1, 4])     # TypeError: list indices must be integers or slices, not tuple
````

🔍 **Что пошло не так?**

- Для среза между двумя числами нужно **двоеточие** `:`, а не запятая `,`.
- С запятой Python не понимает, что это срез, и показывает ошибку.
- Сообщение об ошибке значит: "внутри `[ ]` нужно писать одно число (индекс) или срез с `:`".

✅ **Правильный способ:** Используйте двоеточие `:`.

````python
# ✅ Правильно
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # Двоеточие между числами
````

### Ошибка 3: Круглые скобки вместо квадратных

````python
# ❌ Неправильно
scores = [85, 92, 78, 90, 66]

# print(scores(1:4))      # SyntaxError
````

🔍 **Что пошло не так?**

- Чтобы получить значения из списка (по индексу или срезом), всегда используют **квадратные скобки** `[ ]`.
- Круглые скобки `( )` нужны для команд, например `print()`, `len()` и `append()`.
- Python не понимает `scores(1:4)` и показывает `SyntaxError` ("в коде нарушена грамматика").

✅ **Правильный способ:** Используйте квадратные скобки `[ ]`.

````python
# ✅ Правильно
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # Квадратные скобки
````

### Ошибка 4: Путать срез и индекс

````python
# ❌ Неправильно: мы хотим прибавить 5 к первой оценке
scores = [85, 92, 78]

first = scores[0:1]     # Это СПИСОК: [85], а не число 85
print(first)

# print(first + 5)      # TypeError: can only concatenate list (not "int") to list
````

🔍 **Что пошло не так?**

- `scores[0:1]` - это **срез**, поэтому результат - **список** с одним значением: `[85]`.
- Список и число нельзя сложить через `+`, поэтому Python показывает ошибку.
- Сообщение об ошибке значит: "к списку можно присоединить только список, а не число (int)".
- Посмотрите на квадратные скобки в выводе: `[85]`. Если видите скобки, это список.

✅ **Правильный способ:** Чтобы получить **одно значение**, используйте **индекс** (одно число, без двоеточия): `scores[0]`.

````python
# ✅ Правильно
scores = [85, 92, 78]

first = scores[0]       # Это число 85 (без скобок)
print(first)
print(first + 5)        # 90
````

### Ошибка 5: Думать, что срез меняет исходный список

````python
# ❌ Неправильно: мы хотим оставить только первые 3 оценки
scores = [85, 92, 78, 90, 66]

scores[:3]              # Это создаёт новый список... но он нигде не сохраняется!
print(scores)           # [85, 92, 78, 90, 66] -> ничего не изменилось
````

🔍 **Что пошло не так?**

- Срез создаёт **новый список**. Он **не** меняет исходный список.
- В строке `scores[:3]` Python создал новый список `[85, 92, 78]`, но мы не сохранили его ни в какую переменную. Поэтому он просто исчез.
- Ошибки нет, поэтому это трудно заметить.

✅ **Правильный способ:** Сохраните срез в переменную с помощью `=`. Если хотите заменить исходный список, сохраните срез обратно в ту же переменную (как в `age = age + 1`: новое значение заменяет старое).

````python
# ✅ Правильно (1): сохраняем в новую переменную
scores = [85, 92, 78, 90, 66]

first_three = scores[:3]
print(first_three)      # [85, 92, 78]
print(scores)           # Исходный список по-прежнему полный
````

````python
# ✅ Правильно (2): сохраняем обратно в ту же переменную
scores = [85, 92, 78, 90, 66]

scores = scores[:3]     # Сначала вычисляется правая часть ([85, 92, 78]), потом она сохраняется в scores
print(scores)           # [85, 92, 78] -> старый список заменён
````

### Ошибка 6: start стоит после end

````python
# ❌ Неправильно: нам нужны значения с индекса 1 по индекс 3
scores = [85, 92, 78, 90, 66]

print(scores[4:1])      # [] -> пустой список!
````

🔍 **Что пошло не так?**

- Срез всегда идёт **слева направо**: от start к end.
- `scores[4:1]` начинается с индекса 4 и пытается дойти до индекса 1. Но 1 находится **левее** 4, поэтому между ними ничего нет.
- Результат - пустой список `[]`. Ошибки нет.

✅ **Правильный способ:** Пишите **меньший** индекс первым (start), а **больший** - вторым (end).

````python
# ✅ Правильно
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # С индекса 1 по индекс 3 -> [92, 78, 90]
````

### Ошибка 7: [-3:-1], чтобы получить последние 3 значения

````python
# ❌ Неправильно: нам нужны последние 3 значения
scores = [85, 92, 78, 90, 66]
#  отриц. индекс: -5  -4  -3  -2  -1

print(scores[-3:-1])    # [78, 90] -> только 2 значения! Последнего (66) нет
````

🔍 **Что пошло не так?**

- end **не включается**. С отрицательными индексами это правило не меняется.
- `scores[-3:-1]` означает "от -3 до -1, **не включая** -1". Поэтому последнее значение (-1) не попадает.

✅ **Правильный способ:** Чтобы включить последнее значение, **не пишите end**: `scores[-3:]`. Отсутствие end значит "до самого конца".

````python
# ✅ Правильно
scores = [85, 92, 78, 90, 66]

print(scores[-3:])      # От -3 до конца -> [78, 90, 66]
````

### Ошибка 8: Длина всего списка для среднего значения среза

````python
# ❌ Неправильно: нам нужно среднее первых 3 оценок
scores = [80, 90, 100, 50, 60]
total = 0

for score in scores[:3]:
    total += score              # 80 + 90 + 100 = 270

average = total / len(scores)   # 270 / 5 -> неправильно! Мы сложили только 3 оценки
print("Average:", average)      # 54.0
````

🔍 **Что пошло не так?**

- Мы сложили только первые **3** оценки (270).
- Но `len(scores)` считает **все 5** значений всего списка.
- Поэтому мы разделили на 5, а не на 3. Среднее получилось сильно заниженным.
- Ошибки нет.

✅ **Правильный способ:** Делите на количество значений **в срезе**. Проще всего сначала сохранить срез в переменную и применить `len()` к этой переменной.

````python
# ✅ Правильно
scores = [80, 90, 100, 50, 60]
first_three = scores[:3]        # [80, 90, 100]
total = 0

for score in first_three:
    total += score

average = total / len(first_three)  # 270 / 3
print("Average:", average)          # 90.0
````

### Ошибка 9: Ждать ошибку, когда срез выходит за конец списка

````python
# ❓ Сюрприз: это НЕ вызывает ошибку
scores = [85, 92, 78]           # Всего 3 значения (индексы 0, 1, 2)

print(scores[0:10])             # [85, 92, 78] -> ошибки нет!

# А вот слишком большой индекс ВЫЗЫВАЕТ ошибку
# print(scores[10])             # IndexError: list index out of range
````

🔍 **Что происходит?**

- С **индексом** Python должен найти **одно** значение ровно в этой позиции. Если такой позиции нет, он показывает `IndexError` ("индекс выходит за пределы списка").
- Со **срезом** Python просто отдаёт **те значения, которые смог найти** в этом диапазоне. Если диапазон выходит за конец, он останавливается на конце. Ошибки нет.
- Это не совсем ошибка, но может сбить с толку: срез, который выглядит неправильно, всё равно может "сработать", и вы не заметите, что числа были написаны неверно.

✅ **Совет:** Если нужно всё до конца, не угадывайте большое число. Просто не пишите end: `scores[0:]` или `scores[:]`.

````python
# ✅ Лучше
scores = [85, 92, 78]

print(scores[0:])               # С индекса 0 до конца -> [85, 92, 78]
````
