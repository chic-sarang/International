## 1. slicing คืออะไร?

เริ่มจากสิ่งที่เรารู้แล้วกันก่อน

**ลิสต์** เก็บค่าได้หลายค่าตามลำดับ ไว้ในวงเล็บเหลี่ยม `[ ]`:

```python
scores = [85, 92, 78, 90, 66]
```

ค่าแต่ละค่ามีหมายเลขตำแหน่งเรียกว่า **อินเด็กซ์ (index)** กฎที่สำคัญที่สุด: **Python เริ่มนับจาก 0 ไม่ใช่ 1**

| ค่า | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| อินเด็กซ์ | 0 | 1 | 2 | 3 | 4 |

ถ้าต้องการดึง **หนึ่ง** ค่า เราเขียนชื่อลิสต์ตามด้วยอินเด็กซ์ในวงเล็บเหลี่ยม: `scores[0]` คือ `85` และ `scores[2]` คือ `78`

แต่ถ้าเราต้องการ **หลายค่าพร้อมกัน** ล่ะ? เช่น แค่คะแนนของนักเรียน 3 คนแรก?

ด้วยสิ่งที่เรารู้ตอนนี้ เราต้องดึงทีละค่า: `scores[0]`, `scores[1]`, `scores[2]`

**slicing** ช่วยให้ดึง **บางส่วน** ของลิสต์ได้ในขั้นตอนเดียว คำว่า "slice" แปลว่า "ชิ้นที่ตัดออกมาจากบางอย่าง" เหมือนขนมปังหนึ่งแผ่น หรือพิซซ่าหนึ่งชิ้น เราตัดชิ้นหนึ่งออกมาจากลิสต์

ผลลัพธ์ของ slicing คือ **ลิสต์ใหม่** ที่มีแค่ส่วนนั้น

````python
scores = [85, 92, 78, 90, 66]

# ดึง "หนึ่ง" ค่าด้วยอินเด็กซ์ (สิ่งที่เรารู้แล้ว)
print(scores[0])        # อินเด็กซ์ 0 -> ค่าที่ 1 -> 85
print(scores[1])        # อินเด็กซ์ 1 -> ค่าที่ 2 -> 92
print(scores[2])        # อินเด็กซ์ 2 -> ค่าที่ 3 -> 78
````

````python
scores = [85, 92, 78, 90, 66]

# ดึง "หลาย" ค่าพร้อมกันด้วย slicing (สิ่งที่เราเรียนวันนี้)
# scores[0:3] หมายถึง "ตั้งแต่อินเด็กซ์ 0 ถึงก่อนอินเด็กซ์ 3 (ไม่รวม 3)"
print(scores[0:3])      # [85, 92, 78] -> ลิสต์ใหม่ที่มี 3 ค่าแรก
````

ดูผลลัพธ์สองแบบข้างบนให้ดี:

- `scores[0]` ได้ **ค่าเดียว**: `85` (ไม่มีวงเล็บเหลี่ยม)
- `scores[0:3]` ได้ **ลิสต์**: `[85, 92, 78]` (มีวงเล็บเหลี่ยม)

ความแตกต่างนี้สำคัญมาก อินเด็กซ์ได้ค่า slice ได้ลิสต์

## 2. รูปแบบพื้นฐาน: `ลิสต์[start:end]`

```python
ชื่อลิสต์[start:end]
```

- `start`: อินเด็กซ์ที่ slice **เริ่ม** (ค่านี้ **ถูกรวม**)
- `end`: อินเด็กซ์ที่ slice **หยุด** (ค่านี้ **ไม่ถูกรวม**)
- `:` โคลอนระหว่างตัวเลขสองตัว

อ่าน `scores[1:4]` ได้ว่า: **"ตั้งแต่อินเด็กซ์ 1 ถึงก่อนอินเด็กซ์ 4 (ไม่รวม 4)"**

### end ไม่ถูกรวม

นี่คือส่วนที่ทำให้มือใหม่สับสนมากที่สุด มาดูช้า ๆ ด้วยตารางกัน

```python
scores = [85, 92, 78, 90, 66]
```

| อินเด็กซ์ | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| ค่า | 85 | 92 | 78 | 90 | 66 |
| อยู่ใน `scores[1:4]` ไหม? | ❌ | ✅ | ✅ | ✅ | ❌ |

`scores[1:4]` → อินเด็กซ์ 1, 2, 3 → `[92, 78, 90]` อินเด็กซ์ 4 คือจุดที่ slice **หยุด** จึงไม่รวม 66

**เคยเห็นกฎนี้มาก่อนไหม?** เคยแน่นอน! ในบทที่ 3 `range(1, 4)` ให้ตัวเลข 1, 2, 3 และ **ไม่** รวม 4 slicing ใช้กฎเดียวกันทุกประการ: **start ถูกรวม แต่ end ไม่ถูกรวม**

### ได้กี่ค่า?

เพราะ end ไม่ถูกรวม จำนวนค่าจึงเท่ากับ **end - start** ง่าย ๆ

- `scores[1:4]` → 4 - 1 = **3** ค่า
- `scores[0:2]` → 2 - 0 = **2** ค่า
- `scores[2:5]` → 5 - 2 = **3** ค่า

````python
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์: 0   1   2   3   4

print(scores[1:4])      # อินเด็กซ์ 1, 2, 3 (ไม่รวม 4) -> [92, 78, 90]
````

````python
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์: 0   1   2   3   4

print(scores[0:2])      # อินเด็กซ์ 0, 1 -> [85, 92]
print(scores[2:5])      # อินเด็กซ์ 2, 3, 4 -> [78, 90, 66]
print(scores[3:4])      # อินเด็กซ์ 3 อย่างเดียว -> [90] (ลิสต์ที่มีค่าเดียว)
````

````python
# กฎเดียวกับ range() ในบทที่ 3: end ไม่ถูกรวม
# range(1, 4) ให้ 1, 2, 3
for i in range(1, 4):
    print(i)
````

````python
# กับสตริงก็ทำงานแบบเดียวกันทุกประการ
fruits = ["apple", "banana", "cherry", "grape", "melon"]
#  อินเด็กซ์:   0        1         2        3        4

print(fruits[0:2])      # อินเด็กซ์ 0, 1 -> ['apple', 'banana']
print(fruits[1:3])      # อินเด็กซ์ 1, 2 -> ['banana', 'cherry']
print(fruits[2:5])      # อินเด็กซ์ 2, 3, 4 -> ['cherry', 'grape', 'melon']
````

### นับจำนวนค่าด้วย len()

จำ `len()` ได้ไหม? มันบอกว่าลิสต์มี **ค่ากี่ค่า** `len([85, 92, 78])` คือ `3`

slice ก็เป็นลิสต์ เราจึงใช้ `len()` กับมันได้ มาตรวจกฎ "end - start" กัน

````python
scores = [85, 92, 78, 90, 66]

print(len(scores))          # ทั้งลิสต์มี 5 ค่า
print(len(scores[1:4]))     # 4 - 1 = 3 ค่า
print(len(scores[0:2]))     # 2 - 0 = 2 ค่า
````

### slice คือลิสต์ใหม่ ลิสต์เดิมไม่เปลี่ยน

slicing **คัดลอก** บางส่วนของลิสต์ไปไว้ใน **ลิสต์ใหม่** ลิสต์เดิมยังเหมือนเดิมทุกประการ

ถ้าต้องการใช้ slice ภายหลัง ให้เก็บไว้ในตัวแปร เหมือนค่าอื่น ๆ จำไว้ว่า: `=` หมายถึง **"เก็บค่าทางขวาไว้ในกล่องทางซ้าย"**

````python
scores = [85, 92, 78, 90, 66]

first_three = scores[0:3]   # เก็บลิสต์ใหม่ไว้ในตัวแปรชื่อ first_three

print(first_three)          # [85, 92, 78] -> ลิสต์ใหม่
print(scores)               # [85, 92, 78, 90, 66] -> ลิสต์เดิมยังครบ
````

````python
# ลิสต์ใหม่เป็นลิสต์จริง ๆ ชนิดของมันจึงเป็น list
# type() บอกว่าค่านั้นเป็นชนิดอะไร
scores = [85, 92, 78, 90, 66]
first_three = scores[0:3]

print(type(first_three))    # <class 'list'>
print(type(scores[0]))      # <class 'int'> -> อินเด็กซ์ได้ค่าเดียว (จำนวนเต็ม) ไม่ใช่ลิสต์
````

````python
# ใช้อินเด็กซ์กับลิสต์ใหม่ได้ด้วย
# จำไว้ว่า: ลิสต์ใหม่เริ่มจากอินเด็กซ์ 0 อีกครั้ง!
scores = [85, 92, 78, 90, 66]
last_two = scores[3:5]      # [90, 66]

print(last_two[0])          # ค่าที่ 1 ของลิสต์ "ใหม่" -> 90 (ในลิสต์เดิมคืออินเด็กซ์ 3)
print(last_two[1])          # ค่าที่ 2 ของลิสต์ "ใหม่" -> 66
````

## 3. การไม่ใส่ start หรือ end

ไม่จำเป็นต้องเขียนตัวเลขทั้งสองตัวเสมอไป Python จะเติมตัวเลขที่ขาดให้เอง

| วิธีเขียน | ความหมาย | Python อ่านเป็น |
|---|---|---|
| `scores[:3]` | ตั้งแต่ **ต้น** ถึงก่อนอินเด็กซ์ 3 | `scores[0:3]` |
| `scores[2:]` | ตั้งแต่อินเด็กซ์ 2 ถึง **ท้าย** | `scores[2:5]` (เมื่อมี 5 ค่า) |
| `scores[:]` | ตั้งแต่ต้นจนถึงท้าย (ทั้งลิสต์) | `scores[0:5]` |

- ไม่มี start → เริ่มจาก **ต้น** (อินเด็กซ์ 0)
- ไม่มี end → ไปจนถึง **ท้าย** (รวมค่าสุดท้ายด้วย)

**มีประโยชน์อย่างไร?**
- `scores[:3]` สั้นกว่า `scores[0:3]` และอ่านง่าย: "3 ตัวแรก"
- `scores[2:]` ใช้ได้โดยไม่ต้องรู้ว่าลิสต์ยาวเท่าไร ไม่ต้องนับจำนวนค่า

````python
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์: 0   1   2   3   4

# ไม่มี start -> เริ่มจากต้น
print(scores[:3])       # เหมือนกับ scores[0:3] -> [85, 92, 78]
print(scores[0:3])      # ผลลัพธ์เหมือนกัน
````

````python
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์: 0   1   2   3   4

# ไม่มี end -> ไปจนถึงท้าย (ค่าสุดท้าย "ถูกรวม")
print(scores[2:])       # อินเด็กซ์ 2, 3, 4 -> [78, 90, 66]
print(scores[1:])       # ทุกค่ายกเว้นตัวแรก -> [92, 78, 90, 66]
````

````python
# [:n] = "n ค่าแรก"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[:1])       # 1 ตัวแรก -> ['apple']
print(fruits[:2])       # 2 ตัวแรก -> ['apple', 'banana']
print(fruits[:4])       # 4 ตัวแรก -> ['apple', 'banana', 'cherry', 'grape']
````

````python
# [n:] = "ข้าม n ค่าแรก แล้วเอาที่เหลือ"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[1:])       # ข้าม 1 -> ['banana', 'cherry', 'grape', 'melon']
print(fruits[3:])       # ข้าม 3 -> ['grape', 'melon']
````

````python
# [n:] ยังใช้ได้แม้ลิสต์ยาวขึ้น
# append() เพิ่มค่าไว้ท้ายลิสต์
scores = [85, 92, 78]
scores.append(90)           # ตอนนี้: [85, 92, 78, 90]
scores.append(66)           # ตอนนี้: [85, 92, 78, 90, 66]

print(scores[1:])           # เราไม่ต้องรู้เลยว่าตอนนี้มี 5 ค่า
````

````python
# [:] = ทั้งลิสต์ (สำเนาของมัน)
scores = [85, 92, 78, 90, 66]
print(scores[:])
````

````python
# ตัดลิสต์ออกเป็นสองส่วน
# [:2] กับ [2:] ใช้ตัวเลขเดียวกัน รวมกันจึงได้ทั้งลิสต์
# (เพราะ 2 "ไม่" อยู่ใน [:2] แต่ "อยู่" ใน [2:])
scores = [85, 92, 78, 90, 66]

front = scores[:2]
back = scores[2:]

print(front)            # [85, 92]
print(back)             # [78, 90, 66]
````

## 4. slicing ด้วยอินเด็กซ์ติดลบ

มาทบทวน **อินเด็กซ์ติดลบ** กัน Python นับ **ย้อนจากท้าย** ลิสต์ได้ด้วย:

- `-1` คือค่า **ตัวสุดท้าย**
- `-2` คือค่ารองสุดท้าย
- `-3` คือค่าที่สามนับจากท้าย ไปเรื่อย ๆ

| ค่า | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| อินเด็กซ์ | 0 | 1 | 2 | 3 | 4 |
| อินเด็กซ์ติดลบ | -5 | -4 | -3 | -2 | -1 |

เหตุผลหลักที่ใช้อินเด็กซ์ติดลบ: **ไปถึงท้ายลิสต์ได้โดยไม่ต้องนับว่ามีกี่ค่า** `-1` คือค่าตัวสุดท้ายเสมอ ไม่ว่าลิสต์จะมี 5 ค่าหรือ 500 ค่า

อินเด็กซ์ติดลบใช้ใน slicing ได้ด้วย กฎยังเหมือนเดิม: **start ถูกรวม แต่ end ไม่ถูกรวม**

| วิธีเขียน | ความหมาย |
|---|---|
| `scores[-3:]` | **3 ค่าสุดท้าย** |
| `scores[-2:]` | **2 ค่าสุดท้าย** |
| `scores[:-1]` | ทุกค่า **ยกเว้นค่าสุดท้าย** |
| `scores[:-2]` | ทุกค่า **ยกเว้น 2 ค่าสุดท้าย** |

````python
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์ติดลบ: -5  -4  -3  -2  -1

# [-n:] = "n ค่าสุดท้าย"
print(scores[-3:])      # ตั้งแต่ -3 ถึงท้าย -> [78, 90, 66]
print(scores[-2:])      # ตั้งแต่ -2 ถึงท้าย -> [90, 66]
print(scores[-1:])      # ตั้งแต่ -1 ถึงท้าย -> [66] (ลิสต์ที่มีค่าเดียว)
````

````python
# เปรียบเทียบ: อินเด็กซ์ได้ค่าเดียว slice ได้ลิสต์
scores = [85, 92, 78, 90, 66]

print(scores[-1])       # 66   -> ค่าเดียว (ค่าสุดท้าย)
print(scores[-1:])      # [66] -> ลิสต์ที่มีค่าสุดท้ายอยู่ข้างใน
````

````python
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์ติดลบ: -5  -4  -3  -2  -1

# [:-n] = "ทุกค่ายกเว้น n ค่าสุดท้าย"
print(scores[:-1])      # ถึง -1 แต่ "ไม่รวม" -1 -> [85, 92, 78, 90]
print(scores[:-2])      # ถึง -2 แต่ "ไม่รวม" -2 -> [85, 92, 78]
````

````python
# 3 ค่าสุดท้าย กับลิสต์ยาวเท่าไรก็ได้
# ไม่ต้องรู้ว่ามีกี่ค่า
short_list = [1, 2, 3, 4]
long_list = [10, 20, 30, 40, 50, 60, 70, 80]

print(short_list[-3:])  # [2, 3, 4]
print(long_list[-3:])   # [60, 70, 80]
````

````python
# ตัวอย่างจากชีวิตจริง: 3 ข้อความล่าสุด
# ข้อความใหม่ถูกเพิ่มไว้ท้ายด้วย append() ข้อความใหม่สุดจึงอยู่ท้ายสุด
messages = ["Hi", "How are you?", "Fine, thanks", "See you", "Bye"]

print(messages[-3:])    # 3 ข้อความล่าสุด
````

## 5. การใช้ slice กับลูป for

slice คือลิสต์ ดังนั้นทุกอย่างที่ทำกับลิสต์ได้ ก็ทำกับ slice ได้ รวมถึง **ลูป for** ด้วย

มาทบทวนวิธีทำงานของลูป for:

```python
for score in scores:
    print(score)
```

- `for score in scores:` หมายถึง **"สำหรับ score แต่ละตัวใน scores ให้รันโค้ดข้างล่าง"**
- แต่ละรอบ ค่าหนึ่งค่าจากลิสต์จะถูกเก็บไว้ในตัวแปร `score` ตั้งแต่ค่าแรกจนถึงค่าสุดท้าย
- บรรทัดนี้ต้องลงท้ายด้วยโคลอน `:`
- โค้ดที่ต้องการทำซ้ำต้อง **ย่อหน้า** (มี 4 ช่องว่างข้างหน้า) Python ใช้การย่อหน้าเพื่อรู้ว่าบรรทัดไหนเป็นของลูป กลุ่มบรรทัดที่ย่อหน้านี้เรียกว่า **บล็อก (block)**
- บรรทัดแรกที่ **ไม่** ย่อหน้าอยู่นอกลูป และทำงานแค่ครั้งเดียว หลังจากลูปจบ

ทีนี้ ถ้าเราใส่ **slice** ไว้หลัง `in` ลูปจะไล่ดู **แค่ส่วนนั้น** ของลิสต์

```python
for score in scores[:3]:      # แค่ 3 ค่าแรก
    print(score)
```

````python
# ทั้งลิสต์
scores = [85, 92, 78, 90, 66]

for score in scores:            # 5 ค่า -> ลูปทำงาน 5 รอบ
    print(score)
````

````python
# แค่ 3 ค่าแรก
scores = [85, 92, 78, 90, 66]

for score in scores[:3]:        # scores[:3] คือ [85, 92, 78] -> ลูปทำงาน 3 รอบ
    print(score)                # ย่อหน้า (4 ช่องว่าง) -> อยู่ในลูป -> ถูกทำซ้ำ
````

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[1:4]:       # fruits[1:4] คือ ['banana', 'cherry', 'grape']
    print(fruit)                # fruit (ค่าเดียว), fruits (ทั้งลิสต์)
````

### การย่อหน้าเป็นตัวกำหนดว่าอะไรถูกทำซ้ำ

## 6. การคำนวณด้วย slice

### บวกบางส่วนของลิสต์

ในบทที่ 5 เราบวกทุกค่าในลิสต์ด้วย **ผลรวมสะสม**:

1. **ก่อนลูป:** สร้างตัวแปรและกำหนดเป็น 0 → `total = 0`
2. **ในลูป:** บวกค่าปัจจุบันเข้าไป → `total = total + score`
3. **หลังลูป:** ตัวแปรนั้นเก็บผลรวมไว้

จำความหมายของ `total = total + score`: `=` หมายถึง "เก็บ" Python จะคำนวณทางขวาก่อน (ผลรวมเดิม + คะแนนปัจจุบัน) แล้วเก็บผลลัพธ์กลับไปใน `total`

ด้วย slice เราบวก **แค่บางส่วน** ของลิสต์ได้ แค่ใส่ slice ไว้หลัง `in`

````python
# บวกแค่ 3 คะแนนแรก
scores = [85, 92, 78, 90, 66]
total = 0                           # ก่อนลูป: เริ่มจาก 0

for score in scores[:3]:            # แค่ 85, 92, 78
    total = total + score           # ผลรวมเดิม + คะแนนปัจจุบัน -> เก็บกลับเข้า total

print("Total of the first 3:", total)   # หลังลูป (ไม่ย่อหน้า): 85 + 92 + 78 = 255
````

````python
# มาดูกันว่า total เปลี่ยนไปอย่างไร
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total = total + score
    print("score:", score, "total:", total)     # ย่อหน้า -> แสดงทุกรอบ
````

**จำวิธีเขียนสั้น `+=`:** `total += score` มีความหมายเหมือนกับ `total = total + score` ทุกประการ "บวกค่าทางขวาเข้าไปในตัวแปร แล้วเก็บผลลัพธ์กลับไว้ในตัวแปรนั้น"

````python
# แบบเดียวกัน แต่เขียนด้วย +=
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total += score                  # เหมือนกับ: total = total + score

print("Total of the first 3:", total)
````

````python
# บวกแค่ 2 คะแนนสุดท้าย
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[-2:]:           # แค่ 90, 66
    total += score                  # เหมือนกับ: total = total + score

print("Total of the last 2:", total)    # 90 + 66 = 156
````

### ค่าเฉลี่ยของบางส่วน

**ค่าเฉลี่ย** คือผลรวมหารด้วยจำนวนค่า: **ค่าเฉลี่ย = ผลรวม / จำนวนค่า**

เราหาจำนวนค่าด้วย `len()` แต่ระวัง: เราต้องการจำนวนค่า **ใน slice** ไม่ใช่ในทั้งลิสต์ จึงต้องใช้ `len()` กับ slice

และจำไว้ว่า: คำนวณค่าเฉลี่ย **หลัง** ลูป (ไม่ย่อหน้า) เมื่อผลรวมครบแล้ว

````python
# ค่าเฉลี่ยของ 3 คะแนนแรก
scores = [85, 92, 78, 90, 66]
first_three = scores[:3]            # เก็บ slice ไว้ในตัวแปร: [85, 92, 78]
total = 0

for score in first_three:
    total += score                  # เหมือนกับ: total = total + score

average = total / len(first_three)  # 255 / 3 (ความยาวของ "slice" ไม่ใช่ทั้งลิสต์)
print("Total:", total)
print("Count:", len(first_three))
print("Average:", average)
````

````python
# เปรียบเทียบค่าเฉลี่ยครึ่งแรกกับครึ่งหลัง
scores = [60, 65, 70, 85, 90, 95]
first_half = scores[:3]             # [60, 65, 70]
second_half = scores[3:]            # [85, 90, 95]

total = 0
for score in first_half:
    total += score
print("First half average:", total / len(first_half))

total = 0                           # เริ่มจาก 0 ใหม่สำหรับครึ่งหลัง!
for score in second_half:
    total += score
print("Second half average:", total / len(second_half))
````

### ค่ามากที่สุดในบางส่วน

ในบทที่ 5 เราหาค่ามากที่สุดแบบที่คนทำบนกระดาษ:

1. จำ **ค่าแรก** ไว้ว่าเป็น "ตัวที่มากที่สุดจนถึงตอนนี้"
2. ดูแต่ละค่า ถ้ามากกว่าตัวที่จำไว้ ให้จำตัวนี้แทน
3. ตัวที่จำไว้ตอนท้ายคือตัวที่มากที่สุด

ด้วย slice เราหาค่ามากที่สุด **ในแค่บางส่วน** ของลิสต์ได้

````python
# คะแนนสูงสุดในนักเรียน 3 คนสุดท้าย
scores = [95, 82, 74, 88, 79]
last_three = scores[-3:]            # [74, 88, 79]

highest = last_three[0]             # ขั้นที่ 1: เริ่มจากค่าแรกของ "slice" (74)

for score in last_three:
    if score > highest:             # ขั้นที่ 2: มากกว่าตัวที่จำไว้ไหม?
        highest = score             #           ถ้าใช่ ให้จำตัวนี้แทน

print("Highest of the last 3:", highest)    # ขั้นที่ 3: 88
````

## 7. การใช้ slice กับ if, in และ append()

### if ในลูป

เมื่อใส่ `if` ไว้ในลูป `for` จะมีการย่อหน้า **สองระดับ**:

```python
for score in scores[:3]:      # ไม่มีช่องว่าง
    if score >= 80:           # 4 ช่องว่าง -> อยู่ใน for (ทำงานกับทุก score)
        print(score)          # 8 ช่องว่าง -> อยู่ใน if (ทำงานเฉพาะเมื่อเงื่อนไขเป็น True)
```

จำวิธีทำงานของ `if`: โค้ดที่ย่อหน้าอยู่ข้างใต้จะทำงาน **เฉพาะเมื่อเงื่อนไขเป็น True**

````python
# ใน 4 คะแนนแรก แสดงเฉพาะคะแนนที่ได้ 80 ขึ้นไป
scores = [85, 72, 91, 78, 95, 60]

for score in scores[:4]:            # แค่ 85, 72, 91, 78
    if score >= 80:                 # 4 ช่องว่าง: ตรวจทีละคะแนน
        print(score)                # 8 ช่องว่าง: แสดงเฉพาะเมื่อได้ 80 ขึ้นไป
````

````python
# การนับ: ใน 3 คะแนนสุดท้ายมีกี่คะแนนที่ผ่าน (60 ขึ้นไป)?
# การนับทำงานเหมือนผลรวมสะสม แต่บวก 1 แทนการบวกค่า
scores = [85, 42, 78, 55, 90]
pass_count = 0                      # ก่อนลูป: เริ่มจาก 0

for score in scores[-3:]:           # แค่ 78, 55, 90
    if score >= 60:
        pass_count += 1             # เหมือนกับ: pass_count = pass_count + 1

print("Passed among the last 3:", pass_count)
````

### in

`ค่า in ลิสต์` ตรวจว่าค่านั้นอยู่ในลิสต์หรือไม่ แล้วให้ผลเป็น `True` หรือ `False` slice เป็นลิสต์ เราจึงใช้ `in` กับ slice ได้ด้วย แบบนี้จะตรวจ **แค่บางส่วน** ของลิสต์

````python
# นักเรียน 3 อันดับแรก (ลิสต์เรียงตามอันดับอยู่แล้ว: ที่ 1, ที่ 2, ...)
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]

print("Tom" in ranking[:3])         # Tom อยู่ใน 3 อันดับแรกไหม? -> True
print("Kate" in ranking[:3])        # Kate อยู่ใน 3 อันดับแรกไหม? -> False
print("Kate" in ranking)            # Kate อยู่ในทั้งลิสต์ไหม? -> True
````

````python
# ใช้ in เป็นเงื่อนไขของ if
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]
name = "Bill"

if name in ranking[:3]:             # Bill อยู่ใน 3 อันดับแรกไหม?
    print(name, "gets a medal!")
else:
    print(name, "doesn't get a medal.")
````

### append(): สร้างลิสต์ใหม่

จำ `append()` ได้ไหม: `ลิสต์.append(ค่า)` จะเพิ่มค่าไว้ **ท้าย** ลิสต์

เราเริ่มจากลิสต์ว่าง `[]` แล้วในลูปใช้ `append()` เพิ่มเฉพาะค่าที่ต้องการจาก slice ได้

````python
# จาก 4 คะแนนแรก สร้างลิสต์ใหม่ที่มีแค่คะแนนที่ผ่าน
scores = [85, 42, 78, 55, 90, 30]
passed = []                         # ก่อนลูป: ลิสต์ว่าง

for score in scores[:4]:            # แค่ 85, 42, 78, 55
    if score >= 60:
        passed.append(score)        # เพิ่มไว้ท้าย passed

print(passed)                       # [85, 78]
````

````python
# สร้างลิสต์ใหม่ที่ 3 ราคาสุดท้ายถูกคูณ 2
prices = [1000, 2500, 3000, 1500, 4000]
doubled = []

for price in prices[-3:]:           # แค่ 3000, 1500, 4000
    doubled.append(price * 2)       # price * 2 ถูกคำนวณก่อน แล้วจึงถูกเพิ่มเข้าไป

print(doubled)
````

## 8. ข้อผิดพลาดที่พบบ่อย

นี่คือข้อผิดพลาดที่เกือบทุกคนเคยทำตอนเรียน slicing ในแต่ละข้อเราจะดู:

- ❌ **โค้ดที่ผิด**
- 🔍 **ผิดตรงไหน**
- ✅ **วิธีที่ถูกต้อง**

โค้ดที่ผิดบางตัวทำให้เกิด error โค้ดเหล่านั้นถูกเขียนไว้เป็นคอมเมนต์ (มี `#` ข้างหน้า) เพื่อไม่ให้ notebook หยุด ถ้าอยากเห็นข้อความ error จริง ให้ลบ `#` รันเซลล์ แล้วค่อยใส่ `#` กลับ

โค้ดที่ผิดบางตัว **ไม่มี error แต่ได้ผลลัพธ์ผิด** แบบนี้อันตรายกว่า เพราะ Python ไม่เตือนเรา

### ข้อผิดพลาด 1: คิดว่า end ถูกรวม

````python
# ❌ ผิด: เราต้องการค่าที่อินเด็กซ์ 1, 2, 3
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์: 0   1   2   3   4

print(scores[1:3])      # [92, 78] -> ได้แค่ 2 ค่า! ขาด 90 (อินเด็กซ์ 3)
````

🔍 **ผิดตรงไหน?**

- `scores[1:3]` หมายถึง "ตั้งแต่อินเด็กซ์ 1 ถึงก่อนอินเด็กซ์ 3 (**ไม่รวม** 3)"
- จึงได้แค่อินเด็กซ์ 1 กับ 2 อินเด็กซ์ 3 คือจุดที่ slice **หยุด**
- ไม่มี error จึงมองข้ามได้ง่าย

✅ **วิธีที่ถูกต้อง:** end ต้อง **มากกว่า** อินเด็กซ์สุดท้ายที่ต้องการอยู่ **1** ถ้าจะรวมอินเด็กซ์ 3 ให้เขียน `4` (เหมือน `range()` ในบทที่ 3: ถ้าจะรวม 3 ต้องเขียน `range(1, 4)`)

````python
# ✅ ถูกต้อง
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # อินเด็กซ์ 1, 2, 3 -> [92, 78, 90]
````

### ข้อผิดพลาด 2: ใช้จุลภาคแทนโคลอน

````python
# ❌ ผิด
scores = [85, 92, 78, 90, 66]

# print(scores[1, 4])     # TypeError: list indices must be integers or slices, not tuple
````

🔍 **ผิดตรงไหน?**

- slicing ต้องใช้ **โคลอน** `:` คั่นระหว่างตัวเลขสองตัว ไม่ใช่จุลภาค `,`
- ถ้าใช้จุลภาค Python จะไม่เข้าใจว่าเป็น slice แล้วแสดง error
- ข้อความ error หมายถึง: "ใน `[ ]` ต้องเขียนตัวเลขตัวเดียว (อินเด็กซ์) หรือ slice ที่มี `:`"

✅ **วิธีที่ถูกต้อง:** ใช้โคลอน `:`

````python
# ✅ ถูกต้อง
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # มีโคลอนระหว่างตัวเลข
````

### ข้อผิดพลาด 3: ใช้วงเล็บกลมแทนวงเล็บเหลี่ยม

````python
# ❌ ผิด
scores = [85, 92, 78, 90, 66]

# print(scores(1:4))      # SyntaxError
````

🔍 **ผิดตรงไหน?**

- การดึงค่าจากลิสต์ (ด้วยอินเด็กซ์หรือ slice) ใช้ **วงเล็บเหลี่ยม** `[ ]` เสมอ
- วงเล็บกลม `( )` ใช้กับคำสั่ง เช่น `print()`, `len()` และ `append()`
- Python ไม่เข้าใจ `scores(1:4)` จึงแสดง `SyntaxError` ("ไวยากรณ์ของโค้ดผิด")

✅ **วิธีที่ถูกต้อง:** ใช้วงเล็บเหลี่ยม `[ ]`

````python
# ✅ ถูกต้อง
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # วงเล็บเหลี่ยม
````

### ข้อผิดพลาด 4: สับสนระหว่าง slice กับอินเด็กซ์

````python
# ❌ ผิด: เราอยากบวก 5 ให้คะแนนแรก
scores = [85, 92, 78]

first = scores[0:1]     # นี่คือ "ลิสต์": [85] ไม่ใช่ตัวเลข 85
print(first)

# print(first + 5)      # TypeError: can only concatenate list (not "int") to list
````

🔍 **ผิดตรงไหน?**

- `scores[0:1]` เป็น **slice** ผลลัพธ์จึงเป็น **ลิสต์** ที่มีค่าเดียว: `[85]`
- ลิสต์กับตัวเลขบวกกันด้วย `+` ไม่ได้ Python จึงแสดง error
- ข้อความ error หมายถึง: "ต่อลิสต์ได้กับลิสต์เท่านั้น ต่อกับตัวเลข (int) ไม่ได้"
- ดูวงเล็บเหลี่ยมในผลลัพธ์: `[85]` ถ้าเห็นวงเล็บเหลี่ยม แปลว่าเป็นลิสต์

✅ **วิธีที่ถูกต้อง:** ถ้าต้องการ **ค่าเดียว** ให้ใช้ **อินเด็กซ์** (ตัวเลขตัวเดียว ไม่มีโคลอน): `scores[0]`

````python
# ✅ ถูกต้อง
scores = [85, 92, 78]

first = scores[0]       # นี่คือตัวเลข 85 (ไม่มีวงเล็บเหลี่ยม)
print(first)
print(first + 5)        # 90
````

### ข้อผิดพลาด 5: คิดว่า slice เปลี่ยนลิสต์เดิม

````python
# ❌ ผิด: เราอยากเก็บไว้แค่ 3 คะแนนแรก
scores = [85, 92, 78, 90, 66]

scores[:3]              # บรรทัดนี้สร้างลิสต์ใหม่... แต่ไม่ได้เก็บไว้ที่ไหนเลย!
print(scores)           # [85, 92, 78, 90, 66] -> ไม่มีอะไรเปลี่ยน
````

🔍 **ผิดตรงไหน?**

- slicing สร้าง **ลิสต์ใหม่** มัน **ไม่** เปลี่ยนลิสต์เดิม
- ในบรรทัด `scores[:3]` Python สร้างลิสต์ใหม่ `[85, 92, 78]` ขึ้นมา แต่เราไม่ได้เก็บไว้ในตัวแปรไหนเลย มันจึงหายไป
- ไม่มี error จึงสังเกตได้ยาก

✅ **วิธีที่ถูกต้อง:** เก็บ slice ไว้ในตัวแปรด้วย `=` ถ้าต้องการแทนที่ลิสต์เดิม ให้เก็บกลับเข้าไปในตัวแปรเดิม (เหมือน `age = age + 1`: ค่าใหม่แทนที่ค่าเก่า)

````python
# ✅ ถูกต้อง (1): เก็บไว้ในตัวแปรใหม่
scores = [85, 92, 78, 90, 66]

first_three = scores[:3]
print(first_three)      # [85, 92, 78]
print(scores)           # ลิสต์เดิมยังครบ
````

````python
# ✅ ถูกต้อง (2): เก็บกลับเข้าตัวแปรเดิม
scores = [85, 92, 78, 90, 66]

scores = scores[:3]     # ทางขวาถูกคำนวณก่อน ([85, 92, 78]) แล้วจึงถูกเก็บใน scores
print(scores)           # [85, 92, 78] -> ลิสต์เก่าถูกแทนที่แล้ว
````

### ข้อผิดพลาด 6: ใส่ start ไว้หลัง end

````python
# ❌ ผิด: เราต้องการค่าตั้งแต่อินเด็กซ์ 1 ถึงอินเด็กซ์ 3
scores = [85, 92, 78, 90, 66]

print(scores[4:1])      # [] -> ลิสต์ว่าง!
````

🔍 **ผิดตรงไหน?**

- slice ไปจาก **ซ้ายไปขวา** เสมอ: จาก start ไป end
- `scores[4:1]` เริ่มที่อินเด็กซ์ 4 แล้วพยายามไปที่อินเด็กซ์ 1 แต่ 1 อยู่ทาง **ซ้าย** ของ 4 จึงไม่มีอะไรอยู่ระหว่างกลาง
- ผลลัพธ์คือลิสต์ว่าง `[]` ไม่มี error

✅ **วิธีที่ถูกต้อง:** เขียนอินเด็กซ์ที่ **น้อยกว่า** ก่อน (start) แล้วจึงเขียนอินเด็กซ์ที่ **มากกว่า** (end)

````python
# ✅ ถูกต้อง
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # ตั้งแต่อินเด็กซ์ 1 ถึงอินเด็กซ์ 3 -> [92, 78, 90]
````

### ข้อผิดพลาด 7: ใช้ [-3:-1] เพื่อดึง 3 ค่าสุดท้าย

````python
# ❌ ผิด: เราต้องการ 3 ค่าสุดท้าย
scores = [85, 92, 78, 90, 66]
#  อินเด็กซ์ติดลบ: -5  -4  -3  -2  -1

print(scores[-3:-1])    # [78, 90] -> ได้แค่ 2 ค่า! ขาดค่าสุดท้าย (66)
````

🔍 **ผิดตรงไหน?**

- end **ไม่ถูกรวม** กฎนี้ไม่เปลี่ยนแม้ใช้อินเด็กซ์ติดลบ
- `scores[-3:-1]` หมายถึง "ตั้งแต่ -3 ถึงก่อน -1 (**ไม่รวม** -1)" ค่าสุดท้าย (-1) จึงไม่ถูกรวม

✅ **วิธีที่ถูกต้อง:** ถ้าจะรวมค่าสุดท้าย ให้ **ไม่ใส่ end**: `scores[-3:]` ไม่มี end แปลว่า "ไปจนถึงท้าย"

````python
# ✅ ถูกต้อง
scores = [85, 92, 78, 90, 66]

print(scores[-3:])      # ตั้งแต่ -3 ถึงท้าย -> [78, 90, 66]
````

### ข้อผิดพลาด 8: ใช้ความยาวของทั้งลิสต์ในการหาค่าเฉลี่ยของ slice

````python
# ❌ ผิด: เราต้องการค่าเฉลี่ยของ 3 คะแนนแรก
scores = [80, 90, 100, 50, 60]
total = 0

for score in scores[:3]:
    total += score              # 80 + 90 + 100 = 270

average = total / len(scores)   # 270 / 5 -> ผิด! เราบวกแค่ 3 คะแนน
print("Average:", average)      # 54.0
````

🔍 **ผิดตรงไหน?**

- เราบวกแค่ **3** คะแนนแรก (270)
- แต่ `len(scores)` นับ **ทั้ง 5** ค่าในลิสต์
- เราจึงหารด้วย 5 แทนที่จะเป็น 3 ค่าเฉลี่ยจึงต่ำเกินไปมาก
- ไม่มี error

✅ **วิธีที่ถูกต้อง:** หารด้วยจำนวนค่า **ใน slice** วิธีที่ง่ายที่สุด: เก็บ slice ไว้ในตัวแปรก่อน แล้วใช้ `len()` กับตัวแปรนั้น

````python
# ✅ ถูกต้อง
scores = [80, 90, 100, 50, 60]
first_three = scores[:3]        # [80, 90, 100]
total = 0

for score in first_three:
    total += score

average = total / len(first_three)  # 270 / 3
print("Average:", average)          # 90.0
````

### ข้อผิดพลาด 9: คิดว่าจะเกิด error เมื่อ slice เลยท้ายลิสต์

````python
# ❓ น่าแปลกใจ: บรรทัดนี้ "ไม่" ทำให้เกิด error
scores = [85, 92, 78]           # มีแค่ 3 ค่า (อินเด็กซ์ 0, 1, 2)

print(scores[0:10])             # [85, 92, 78] -> ไม่มี error!

# แต่อินเด็กซ์ที่ใหญ่เกินไป "ทำให้" เกิด error
# print(scores[10])             # IndexError: list index out of range
````

🔍 **เกิดอะไรขึ้น?**

- กับ **อินเด็กซ์** Python ต้องหา **ค่าเดียว** ที่ตำแหน่งนั้นพอดี ถ้าไม่มีตำแหน่งนั้น จะแสดง `IndexError` ("อินเด็กซ์อยู่นอกลิสต์")
- กับ **slice** Python แค่ให้ **ค่าเท่าที่หาเจอ** ในช่วงนั้น ถ้าช่วงเลยท้ายลิสต์ ก็จะหยุดที่ท้ายลิสต์ ไม่มี error
- นี่ไม่ใช่ข้อผิดพลาดเสียทีเดียว แต่อาจทำให้สับสนได้: slice ที่ดูเหมือนผิดอาจยัง "ใช้ได้" และเราจะไม่รู้ตัวว่าเขียนตัวเลขผิด

✅ **เคล็ดลับ:** ถ้าต้องการทุกค่าจนถึงท้าย อย่าเดาตัวเลขใหญ่ ๆ แค่ไม่ใส่ end: `scores[0:]` หรือ `scores[:]`

````python
# ✅ ดีกว่า
scores = [85, 92, 78]

print(scores[0:])               # ตั้งแต่อินเด็กซ์ 0 ถึงท้าย -> [85, 92, 78]
````
