## 1. What Is a Function?

### A problem

Let's say we want to print a short greeting:

```python
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
```

Now imagine we need to print the same greeting **3 times** in different places of our program. We would have to write the same 3 lines again and again.

````python
# The same 3 lines, written 3 times
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
````

This works, but there are problems:

- It's long and tiring to write.
- If we want to change "Hello!" to "Hi!", we have to change it in **every** place. If we forget one, the greetings will be different.

### The idea of a function

What if we could write those 3 lines **only once**, give them a **name**, and then just say the name whenever we need them?

That's exactly what a **function** is: **a group of code with a name**.

### Think of a recipe

A function is like a **recipe**.

1. First, you **write the recipe** on paper: "How to make pancakes: step 1, step 2, step 3." Writing the recipe doesn't make any pancakes. It's just instructions saved for later.
2. Later, whenever you want pancakes, you **follow the recipe**. You can follow the same recipe as many times as you want.

A function works the same way:

| Recipe | Function |
|---|---|
| Writing the recipe | **Making** (defining) the function |
| Following the recipe to cook | **Using** (calling) the function |

### You have already used functions!

You've been using functions since Lesson 1. `print()`, `len()` and `type()` are all functions. Someone already wrote them for you, and you just used them by writing their name and parentheses `( )`.

In this lesson, you will learn how to **make your own** functions.

````python
# Functions you have already used
print("Hello")              # print() is a function: it shows a value on the screen
print(len("apple"))         # len() is a function: it counts
print(type(10))             # type() is a function: it tells you the data type
````

## 2. Making a Function and Using It

A function is used in **two separate steps**:

1. **Make** the function (write the recipe)
2. **Use** the function (follow the recipe)

### Step 1: Making a function

```python
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")
```

Let's look at each piece:

| Piece | Meaning |
|---|---|
| `def` | short for **define**. It means "I'm going to make a function" |
| `say_hello` | the **name** of the function. You choose it (the rules are the same as for variable names) |
| `()` | parentheses. They are always needed after the name (we'll see what goes inside in section 3) |
| `:` | a colon at the end of the line, just like `if` and `for` |
| indented lines | the **body** of the function: the code that belongs to the function |

**About indentation:** just like with `if` and `for`, Python uses indentation (4 spaces at the beginning of a line) to know which lines belong to the function. The indented lines under `def` are the function's body. The first line that is not indented is **outside** the function.

### Making a function does NOT run it

This is the most important thing in this section.

When Python reads `def`, it only **remembers** the function. It doesn't run the code inside. Just like writing a recipe doesn't make pancakes.

Look at the cell below. We make a function, but **nothing is printed**.

````python
# Step 1: make the function
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")

# Nothing is printed! We only made (wrote) the function. We didn't use it yet.
````

### Step 2: Using a function

To use a function, write its **name** followed by **parentheses** `()`:

```python
say_hello()
```

Using a function is called **calling** the function. When you call a function, Python goes to the function, runs all the code inside it from top to bottom, and then comes back.

````python
# Step 2: call (use) the function
say_hello()
````

### One recipe, many times

Once a function is made, you can call it **as many times as you want**. Each call runs the whole body again.

````python
# The same greeting 3 times, but now with only 3 short lines
say_hello()
say_hello()
say_hello()
````

Compare this with the beginning of the lesson. Before, we needed 9 lines. Now we need just 3 calls.

And if we want to change "Hello!" to "Hi!", we only change it **in one place**: inside the function. Every call will use the new greeting.

````python
# Change the function in one place...
def say_hello():
    print("Hi!")                        # Changed only here
    print("Welcome to Python class.")
    print("Have a nice day!")

# ...and every call uses the new version
say_hello()
say_hello()
````

### Making and using in the same cell

Usually, we make a function first and call it below. Let's follow what Python does, step by step.

````python
def say_goodbye():          # 1. Python remembers the function "say_goodbye". Nothing runs yet
    print("Goodbye!")
    print("See you next time.")

print("Class is over.")     # 2. This line is not indented -> outside the function -> it runs now
say_goodbye()               # 3. Call the function -> the 2 lines inside run now
print("The end.")           # 4. After the function is finished, Python comes back here
````

Notice the order of the output:

1. `Class is over.` - the first line **outside** the function
2. `Goodbye!` and `See you next time.` - the function's body runs only when we **call** it
3. `The end.` - after the function finishes, Python continues from the line after the call

### Make first, then use

A recipe must be written **before** you can follow it. In the same way, a function must be **made before it is called**. Python reads the code from top to bottom, so if you call a function that hasn't been made yet, Python doesn't know it, and shows an error. (We'll see this error in section 5.)

### More examples

````python
# A function that draws a line
def draw_line():
    print("--------------------")

draw_line()
print("Python Basics")
draw_line()
````

````python
# A function that prints the rules of a game
def show_rules():
    print("Rule 1: Be kind.")
    print("Rule 2: Take turns.")
    print("Rule 3: Have fun!")

print("Before we start, here are the rules:")
show_rules()
````

## 3. Giving Something to a Function

### A problem

Our `say_hello()` function always says exactly the same thing. But what if we want to greet **different people** by name?

- "Hello, Anna!"
- "Hello, Tom!"
- "Hello, Bill!"

We don't want to make a different function for every person. We want **one** function, and we want to **tell** it the name each time.

### Think of ordering at a café

When you order at a café, you **tell** the staff what you want: "One coffee, please." or "One tea, please." The staff does the same job (making a drink), but the result depends on **what you told them**.

A function can work the same way. When we call it, we can **give** it a value, and the function does its job **with that value**.

### Making a function that receives a value

```python
def greet(name):
    print("Hello,", name)
```

Look at the parentheses: there is a word inside, `name`. This is a **variable** that will receive the value we give when we call the function. This variable is called a **parameter**.

### Calling it with a value

```python
greet("Anna")
```

Now we put a value inside the parentheses when we call the function. The value we give is called an **argument**.

| Word | Where it is | Example |
|---|---|---|
| **parameter** | inside `( )` when we **make** the function | `name` in `def greet(name):` |
| **argument** | inside `( )` when we **call** the function | `"Anna"` in `greet("Anna")` |

Don't worry too much about these two words. The important idea is: **the value we give is stored in the variable inside the function.**

````python
# Make a function that receives one value
def greet(name):
    print("Hello,", name)

# Call it with different values
greet("Anna")
greet("Tom")
greet("Bill")
````

**About `print("Hello,", name)`:** when you put several values in `print()` separated by **commas** `,`, they are printed on one line, and Python puts **one space** between them automatically. Text inside quotes is printed as it is, and a variable is replaced by the value inside it.

### What happens, step by step

When we call `greet("Anna")`:

| Step | What Python does |
|---|---|
| 1 | Finds the function `greet` |
| 2 | Stores the given value in the parameter: `name = "Anna"` |
| 3 | Runs the body: `print("Hello,", name)` → `name` is `"Anna"` → prints `Hello, Anna` |
| 4 | The function is finished, so Python goes back to the line after the call |

Then when we call `greet("Tom")`, the same thing happens again, but this time `name = "Tom"`.

So **the code inside the function is the same every time**, but **the result changes** depending on the value we give.

````python
# The parameter is just a variable. We can see what's inside it
def show_value(x):
    print("The function received:", x)

show_value(10)
show_value("apple")
show_value(3.14)
````

````python
# A function that uses the value in a calculation
def show_double(number):
    print(number * 2)       # number is replaced by the value we gave

show_double(5)              # number = 5  -> 5 * 2
show_double(100)            # number = 100 -> 100 * 2
````

````python
# The argument can also be a variable
def greet(name):
    print("Hello,", name)

student = "Kate"
greet(student)              # The value inside student ("Kate") is given to the function
````

### Giving two values

A function can receive **more than one** value. Separate them with commas `,`, both when making and when calling the function.

```python
def add(a, b):
    print(a + b)

add(2, 3)
```

The values are matched **in order**:
- the 1st value (`2`) goes into the 1st parameter (`a`)
- the 2nd value (`3`) goes into the 2nd parameter (`b`)

````python
def add(a, b):
    print(a + b)

add(2, 3)                   # a = 2, b = 3  -> 2 + 3
add(10, 20)                 # a = 10, b = 20 -> 10 + 20
````

````python
# A function with two parameters: a name and an age
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce("Anna", 20)       # name = "Anna", age = 20
introduce("Tom", 25)        # name = "Tom", age = 25
````

### The order matters

Because the values are matched in order, changing the order can change the result.

````python
def subtract(a, b):
    print(a - b)

subtract(10, 3)             # a = 10, b = 3  -> 10 - 3 = 7
subtract(3, 10)             # a = 3, b = 10  -> 3 - 10 = -7 (a different result!)
````

````python
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce(20, "Anna")       # The order is wrong: name = 20, age = "Anna"
````

## 4. Getting a Result Back: return

### A problem

Our `add(a, b)` function **prints** the result on the screen. But what if we want to **use** the result later? For example, to store it in a variable, or to calculate with it?

Printing only **shows** the result. It doesn't **give** the result back to us. Once it's printed, we can't use it anymore.

### Think of an errand

Imagine you ask a friend: "Please go to the store and buy me some bread."

- Your friend goes, buys the bread, and comes back.
- Then your friend **hands you the bread**.
- Now **you have** the bread, and you can do anything with it: eat it, share it, make a sandwich.

`return` is like **handing the result back**. The function does its job, and then gives the result to the place where it was called.

### Making a function with return

```python
def add(a, b):
    return a + b
```

`return` means: **"give this value back to the place where the function was called."**

Notice: this function doesn't print anything. It only gives the result back.

````python
def add(a, b):
    return a + b            # Give the result back

result = add(2, 3)          # The result (5) comes back and is stored in result
print(result)
````

### The call is replaced by the returned value

This is the most important idea of `return`.

When a function returns a value, **the place where you called the function is replaced by that value.**

Let's look at this line slowly:

```python
result = add(2, 3)
```

| Step | What Python sees |
|---|---|
| 1 | `result = add(2, 3)` |
| 2 | Python calls `add(2, 3)`: `a = 2`, `b = 3`, and the function returns `2 + 3`, which is `5` |
| 3 | `add(2, 3)` is **replaced** by `5` → `result = 5` |
| 4 | `5` is stored in `result` |

So after the function returns, it's as if you had written `result = 5`.

You can think of `add(2, 3)` as a box that **turns into 5** when the function is finished.

````python
def add(a, b):
    return a + b

x = add(2, 3)               # add(2, 3) -> 5,  so this becomes: x = 5
y = add(10, 20)             # add(10, 20) -> 30, so this becomes: y = 30

print(x)
print(y)
````

### Using the returned value in a calculation

Because `add(2, 3)` turns into `5`, we can use it anywhere we could use the number `5`.

```python
add(2, 3) + add(3, 4)
```

| Step | What Python sees |
|---|---|
| 1 | `add(2, 3) + add(3, 4)` |
| 2 | `add(2, 3)` runs and returns 5 → **`5`** `+ add(3, 4)` |
| 3 | `add(3, 4)` runs and returns 7 → `5 +` **`7`** |
| 4 | `12` |

````python
def add(a, b):
    return a + b

total = add(2, 3) + add(3, 4)   # becomes 5 + 7
print(total)                    # 12
````

````python
def add(a, b):
    return a + b

print(add(2, 3) * 10)           # add(2, 3) becomes 5 -> print(5 * 10) -> 50
print(add(2, 3) > 4)            # add(2, 3) becomes 5 -> print(5 > 4) -> True
````

### Putting the call directly inside print()

`print(add(2, 3))` also works the same way:

| Step | What Python sees |
|---|---|
| 1 | `print(add(2, 3))` |
| 2 | `add(2, 3)` returns 5 → `print(5)` |
| 3 | `5` is printed |

````python
def add(a, b):
    return a + b

print(add(2, 3))                # becomes print(5)
````

### Giving a returned value to another function

A returned value can even be given to a function again. Python always works **from the inside out**: the inner call is replaced first.

```python
add(add(1, 2), 3)
```

| Step | What Python sees |
|---|---|
| 1 | `add(add(1, 2), 3)` |
| 2 | The inner `add(1, 2)` returns 3 → `add(` **`3`** `, 3)` |
| 3 | `add(3, 3)` returns 6 → **`6`** |

````python
def add(a, b):
    return a + b

result = add(add(1, 2), 3)      # add(1, 2) -> 3, then add(3, 3) -> 6
print(result)
````

### More examples with return

````python
def double(number):
    return number * 2

a = double(4)                   # becomes a = 8
b = double(a)                   # a is 8, so double(8) -> 16
print(a)
print(b)
````

````python
# A function that returns a string
def make_greeting(name):
    return "Hello, " + name     # + joins two strings

message = make_greeting("Anna")  # becomes message = "Hello, Anna"
print(message)
````

````python
# The price after a discount
def discount_price(price):
    return price - 1000

coffee = discount_price(4500)   # becomes coffee = 3500
cake = discount_price(6000)     # becomes cake = 5000

print("Coffee:", coffee)
print("Cake:", cake)
print("Total:", coffee + cake)  # We can calculate with the returned values
````

### print vs return: what's the difference?

Now let's make two functions that look almost the same. One uses `print`, the other uses `return`.

```python
def show_add(a, b):        # uses print
    print(a + b)

def add(a, b):             # uses return
    return a + b
```

| | `print` inside the function | `return` inside the function |
|---|---|---|
| What it does | **shows** the result on the screen | **gives** the result back to the caller |
| Can we store the result in a variable? | ❌ | ✅ |
| Can we calculate with the result? | ❌ | ✅ |
| Like... | a friend who shows you the bread, and keeps it | a friend who hands you the bread |

### What comes back when there is no return?

If a function has no `return`, it still gives something back: a special value called **`None`**. `None` means **"nothing"**.

So `show_add(2, 3)` is replaced by `None`, not by `5`.

````python
def show_add(a, b):
    print(a + b)                # Only shows the result. No return

result = show_add(2, 3)         # 5 is printed on the screen... but what is stored in result?
print("result is:", result)     # None! show_add(2, 3) was replaced by None, not 5
````

Look carefully at the output:

1. `5` - printed **by the function** (by the `print` inside it)
2. `result is: None` - the value stored in `result` is `None`, because the function didn't return anything

The number 5 appeared on the screen, but **we didn't get it back**. It's gone.

### So add(2, 3) + add(3, 4) doesn't work with print

What happens if we try `show_add(2, 3) + show_add(3, 4)`?

| Step | What Python sees |
|---|---|
| 1 | `show_add(2, 3) + show_add(3, 4)` |
| 2 | `show_add(2, 3)` prints 5, and is replaced by **`None`** |
| 3 | `show_add(3, 4)` prints 7, and is replaced by **`None`** |
| 4 | `None + None` → **error!** Python can't add "nothing" and "nothing" |

````python
def show_add(a, b):
    print(a + b)

# total = show_add(2, 3) + show_add(3, 4)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'NoneType'
````

(The line above is a comment so it won't stop the notebook. If you remove the two `#` from the line `total = ...`, you will see `5` and `7` printed first, and then the error.)

The error message means: "`+` can't be used with `None` and `None`". (`NoneType` is the data type of `None`.)

### When to use which?

- If you only want to **show** something on the screen → `print` is fine
- If you want to **use** the result later (store it, calculate with it, give it to another function) → use `return`

Most of the time, functions that calculate something should use `return`. Then you can still print the result whenever you want, with `print(add(2, 3))`.

````python
def add(a, b):
    return a + b

# With return, we can do both: use the result AND print it
total = add(2, 3) + add(3, 4)
print(total)
````

## 5. Common Mistakes

These are the mistakes almost everyone makes when learning functions. For each one, we will look at:

- ❌ **The wrong code**
- 🔍 **What went wrong**
- ✅ **The correct way**

Code that causes an error is written as a comment (with `#` at the beginning) so it won't stop the notebook. If you want to see the real error message, remove the `#`, run the cell, and then put the `#` back.

### Mistake 1: Making a function and expecting it to run

````python
# ❌ Wrong: we want to print "Good morning!"
def morning():
    print("Good morning!")

# Nothing is printed...
````

🔍 **What went wrong?**

- `def` only **makes** the function. It doesn't run it. (Writing a recipe doesn't make food.)
- No error appears, so it can be confusing: "Why didn't anything happen?"

✅ **The correct way:** **Call** the function after making it.

````python
# ✅ Correct
def morning():
    print("Good morning!")

morning()               # Call it
````

### Mistake 2: Forgetting the parentheses when calling

````python
# ❌ Wrong
def morning():
    print("Good morning!")

morning                 # No parentheses -> the function does NOT run
print("Did you see 'Good morning!' above? No!")
````

🔍 **What went wrong?**

- `morning` without `()` is just the **name** of the function. It doesn't call it.
- To call a function, you **must** write the parentheses: `morning()`. The parentheses mean "run it now".
- No error appears, but the function doesn't run.

✅ **The correct way:** Always write `()` after the name when calling.

````python
# ✅ Correct
def morning():
    print("Good morning!")

morning()               # With parentheses
````

### Mistake 3: Calling a function before making it

````python
# ❌ Wrong
# evening()             # NameError: name 'evening' is not defined
#
# def evening():
#     print("Good evening!")
````

🔍 **What went wrong?**

- Python reads the code **from top to bottom**.
- When it reaches `evening()`, the function hasn't been made yet. Python doesn't know any function named `evening`.
- The error is a `NameError` ("there is nothing with this name").

✅ **The correct way:** Make the function first, then call it below.

````python
# ✅ Correct
def evening():
    print("Good evening!")

evening()               # Called after it was made
````

### Mistake 4: Forgetting the colon or the indentation

````python
# ❌ Wrong
# def night()                 # SyntaxError: the colon is missing
#     print("Good night!")

# def night():
# print("Good night!")        # IndentationError: the body is not indented
````

🔍 **What went wrong?**

- These are the same rules as `if` and `for`.
- The `def` line must end with a colon `:`.
- The body of the function must be indented (4 spaces). Without indentation, Python doesn't know which lines belong to the function.

✅ **The correct way:**

````python
# ✅ Correct
def night():                # Colon at the end
    print("Good night!")    # Indented by 4 spaces

night()
````

### Mistake 5: Giving the wrong number of values

````python
# ❌ Wrong
def add(a, b):
    return a + b

# add(2)          # TypeError: add() missing 1 required positional argument: 'b'
# add(2, 3, 4)    # TypeError: add() takes 2 positional arguments but 3 were given
````

🔍 **What went wrong?**

- `add` was made with **2** parameters (`a` and `b`), so it needs exactly **2** values.
- `add(2)`: only 1 value was given. `a` gets 2, but `b` gets nothing. The error says "1 value is missing: `b`".
- `add(2, 3, 4)`: 3 values were given, but there are only 2 places for them. The error says "it takes 2, but 3 were given".

✅ **The correct way:** Give exactly as many values as there are parameters.

````python
# ✅ Correct
def add(a, b):
    return a + b

print(add(2, 3))        # 2 parameters, 2 values
````

### Mistake 6: Using print when you need return

````python
# ❌ Wrong: we want to store the result and use it later
def multiply(a, b):
    print(a * b)                # Only shows the result

answer = multiply(3, 4)         # 12 is printed... but answer is None
print("answer:", answer)
````

🔍 **What went wrong?**

- `multiply` uses `print`, not `return`. It **shows** 12 on the screen, but doesn't **give it back**.
- A function without `return` gives back `None`. So `multiply(3, 4)` is replaced by `None`, and `answer` is `None`.
- If we try to calculate with it (like `answer + 1`), we get an error.

✅ **The correct way:** Use `return` to give the result back. Then print it outside the function if you want to see it.

````python
# ✅ Correct
def multiply(a, b):
    return a * b                # Give the result back

answer = multiply(3, 4)         # becomes answer = 12
print("answer:", answer)
print("answer + 1:", answer + 1)
````

### Mistake 7: Returning but not using the result

````python
# ❌ Wrong: we want to see the result
def multiply(a, b):
    return a * b

multiply(3, 4)                  # The result (12) comes back... but nobody catches it!
print("Where is the result?")
````

🔍 **What went wrong?**

- `multiply(3, 4)` is replaced by `12`. But this line is just `12` by itself. We didn't store it or print it.
- It's like your friend hands you the bread, and you drop it on the floor.
- No error appears, but the result is lost.

✅ **The correct way:** Catch the returned value: store it in a variable, or put the call inside `print()`.

````python
# ✅ Correct
def multiply(a, b):
    return a * b

result = multiply(3, 4)         # Store it
print(result)

print(multiply(3, 4))           # Or print it directly
````
