## 1. What Is Slicing?

Let's start with something we already know.

A **list** holds many values in order, inside square brackets `[ ]`:

```python
scores = [85, 92, 78, 90, 66]
```

Each value has a position number called an **index**. The most important rule: **Python starts counting from 0, not 1.**

| Value | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 |

To get **one** value, we write the list name and the index in square brackets: `scores[0]` is `85`, `scores[2]` is `78`.

But what if we want **several values at once**? For example, the scores of the first 3 students only?

With what we know, we would have to get them one by one: `scores[0]`, `scores[1]`, `scores[2]`.

**Slicing** lets us take a **part** of a list in one step. The word "slice" means "a piece cut from something", like a slice of bread or a slice of pizza. We cut a piece out of the list.

The result of slicing is a **new list** that holds only that part.

````python
scores = [85, 92, 78, 90, 66]

# Getting ONE value with an index (what we already know)
print(scores[0])        # index 0 -> the 1st value -> 85
print(scores[1])        # index 1 -> the 2nd value -> 92
print(scores[2])        # index 2 -> the 3rd value -> 78
````

````python
scores = [85, 92, 78, 90, 66]

# Getting SEVERAL values at once with slicing (what we learn today)
# scores[0:3] means "from index 0 up to (but not including) index 3"
print(scores[0:3])      # [85, 92, 78] -> a new list with the first 3 values
````

Look at the two results above carefully:

- `scores[0]` gives **one value**: `85` (no square brackets)
- `scores[0:3]` gives **a list**: `[85, 92, 78]` (with square brackets)

This difference is very important. An index gives a value. A slice gives a list.

## 2. The Basic Form: `list[start:end]`

```python
list_name[start:end]
```

- `start`: the index where the slice **begins** (this value **is** included)
- `end`: the index where the slice **stops** (this value is **NOT** included)
- `:` a colon between the two numbers

You can read `scores[1:4]` as: **"from index 1, up to but not including index 4"**.

### The end is not included

This is the part that confuses beginners the most. Let's look at it slowly with a table.

```python
scores = [85, 92, 78, 90, 66]
```

| Index | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Value | 85 | 92 | 78 | 90 | 66 |
| Included in `scores[1:4]`? | ❌ | ✅ | ✅ | ✅ | ❌ |

`scores[1:4]` → index 1, 2, 3 → `[92, 78, 90]`. Index 4 is where it **stops**, so 66 is not included.

**Have you seen this rule before?** Yes! In Lesson 3, `range(1, 4)` gave the numbers 1, 2, 3, and 4 was **not** included. Slicing uses exactly the same rule: **start is included, end is not**.

### How many values do you get?

Because the end is not included, the number of values is simply **end - start**.

- `scores[1:4]` → 4 - 1 = **3** values
- `scores[0:2]` → 2 - 0 = **2** values
- `scores[2:5]` → 5 - 2 = **3** values

````python
scores = [85, 92, 78, 90, 66]
#  index:  0   1   2   3   4

print(scores[1:4])      # index 1, 2, 3 (4 is not included) -> [92, 78, 90]
````

````python
scores = [85, 92, 78, 90, 66]
#  index:  0   1   2   3   4

print(scores[0:2])      # index 0, 1 -> [85, 92]
print(scores[2:5])      # index 2, 3, 4 -> [78, 90, 66]
print(scores[3:4])      # index 3 only -> [90] (a list with just one value)
````

````python
# The same rule as range() in Lesson 3: the end is not included
# range(1, 4) gives 1, 2, 3
for i in range(1, 4):
    print(i)
````

````python
# With strings, it works exactly the same way
fruits = ["apple", "banana", "cherry", "grape", "melon"]
#  index:     0        1         2        3        4

print(fruits[0:2])      # index 0, 1 -> ['apple', 'banana']
print(fruits[1:3])      # index 1, 2 -> ['banana', 'cherry']
print(fruits[2:5])      # index 2, 3, 4 -> ['cherry', 'grape', 'melon']
````

### Counting the number of values with len()

Remember `len()`? It tells you **how many values** are in a list. `len([85, 92, 78])` is `3`.

A slice is a list, so we can use `len()` on it too. Let's check the rule "end - start".

````python
scores = [85, 92, 78, 90, 66]

print(len(scores))          # The whole list has 5 values
print(len(scores[1:4]))     # 4 - 1 = 3 values
print(len(scores[0:2]))     # 2 - 0 = 2 values
````

### The slice is a new list. The original list doesn't change.

Slicing **copies** a part of the list into a **new list**. The original list stays exactly the same.

If you want to use the slice later, store it in a variable, just like any other value. Remember: `=` means **"store the value on the right into the box on the left"**.

````python
scores = [85, 92, 78, 90, 66]

first_three = scores[0:3]   # Store the new list in a variable named first_three

print(first_three)          # [85, 92, 78] -> the new list
print(scores)               # [85, 92, 78, 90, 66] -> the original list is still complete
````

````python
# The new list is a real list, so its type is list
# type() tells you what kind of value something is
scores = [85, 92, 78, 90, 66]
first_three = scores[0:3]

print(type(first_three))    # <class 'list'>
print(type(scores[0]))      # <class 'int'> -> an index gives one value (an integer), not a list
````

````python
# You can use an index on the new list too
# Remember: the new list starts from index 0 again!
scores = [85, 92, 78, 90, 66]
last_two = scores[3:5]      # [90, 66]

print(last_two[0])          # The 1st value of the NEW list -> 90 (it was index 3 in the original list)
print(last_two[1])          # The 2nd value of the NEW list -> 66
````

## 3. Leaving Out the Start or the End

You don't always have to write both numbers. Python fills in the missing number for you.

| How to write it | Meaning | Python reads it as |
|---|---|---|
| `scores[:3]` | from the **beginning** up to (not including) index 3 | `scores[0:3]` |
| `scores[2:]` | from index 2 to the **end** | `scores[2:5]` (when there are 5 values) |
| `scores[:]` | from the beginning to the end (the whole list) | `scores[0:5]` |

- No start → start from the **beginning** (index 0)
- No end → go all the way to the **end** (including the last value)

**Why is this useful?**
- `scores[:3]` is shorter than `scores[0:3]`, and it's easy to read: "the first 3".
- `scores[2:]` works without knowing how long the list is. You don't need to count the values.

````python
scores = [85, 92, 78, 90, 66]
#  index:  0   1   2   3   4

# No start -> from the beginning
print(scores[:3])       # Same as scores[0:3] -> [85, 92, 78]
print(scores[0:3])      # The same result
````

````python
scores = [85, 92, 78, 90, 66]
#  index:  0   1   2   3   4

# No end -> to the end (the last value IS included)
print(scores[2:])       # index 2, 3, 4 -> [78, 90, 66]
print(scores[1:])       # Everything except the first one -> [92, 78, 90, 66]
````

````python
# [:n] = "the first n values"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[:1])       # The first 1 -> ['apple']
print(fruits[:2])       # The first 2 -> ['apple', 'banana']
print(fruits[:4])       # The first 4 -> ['apple', 'banana', 'cherry', 'grape']
````

````python
# [n:] = "skip the first n values, and take the rest"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[1:])       # Skip 1 -> ['banana', 'cherry', 'grape', 'melon']
print(fruits[3:])       # Skip 3 -> ['grape', 'melon']
````

````python
# [n:] works even after the list gets longer
# append() adds a value at the end of the list
scores = [85, 92, 78]
scores.append(90)           # Now: [85, 92, 78, 90]
scores.append(66)           # Now: [85, 92, 78, 90, 66]

print(scores[1:])           # We didn't need to know that there are 5 values now
````

````python
# [:] = the whole list (a copy of it)
scores = [85, 92, 78, 90, 66]
print(scores[:])
````

````python
# Cutting a list into two parts
# [:2] and [2:] use the same number, so together they make the whole list
# (because 2 is NOT included in [:2], but IS included in [2:])
scores = [85, 92, 78, 90, 66]

front = scores[:2]
back = scores[2:]

print(front)            # [85, 92]
print(back)             # [78, 90, 66]
````

## 4. Slicing with Negative Indexes

Let's remember the **negative index**. Python can also count **backwards from the end** of the list:

- `-1` is the **last** value
- `-2` is the second to last value
- `-3` is the third to last value, and so on

| Value | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 |
| Negative index | -5 | -4 | -3 | -2 | -1 |

The main reason we use a negative index: **we can reach the end of the list without counting how many values there are**. `-1` is always the last value, whether the list has 5 values or 500.

Negative indexes work in slicing too. The rule is the same: **the start is included, the end is not**.

| How to write it | Meaning |
|---|---|
| `scores[-3:]` | the **last 3** values |
| `scores[-2:]` | the **last 2** values |
| `scores[:-1]` | everything **except the last** value |
| `scores[:-2]` | everything **except the last 2** values |

````python
scores = [85, 92, 78, 90, 66]
#  negative index: -5  -4  -3  -2  -1

# [-n:] = "the last n values"
print(scores[-3:])      # From -3 to the end -> [78, 90, 66]
print(scores[-2:])      # From -2 to the end -> [90, 66]
print(scores[-1:])      # From -1 to the end -> [66] (a list with one value)
````

````python
# Compare: an index gives one value, a slice gives a list
scores = [85, 92, 78, 90, 66]

print(scores[-1])       # 66   -> one value (the last one)
print(scores[-1:])      # [66] -> a list holding the last value
````

````python
scores = [85, 92, 78, 90, 66]
#  negative index: -5  -4  -3  -2  -1

# [:-n] = "everything except the last n values"
print(scores[:-1])      # Up to -1, but -1 is NOT included -> [85, 92, 78, 90]
print(scores[:-2])      # Up to -2, but -2 is NOT included -> [85, 92, 78]
````

````python
# The last 3 values, with any length of list
# We don't need to know how many values there are
short_list = [1, 2, 3, 4]
long_list = [10, 20, 30, 40, 50, 60, 70, 80]

print(short_list[-3:])  # [2, 3, 4]
print(long_list[-3:])   # [60, 70, 80]
````

````python
# A real-life example: the most recent 3 messages
# New messages are added at the end with append(), so the newest ones are at the end
messages = ["Hi", "How are you?", "Fine, thanks", "See you", "Bye"]

print(messages[-3:])    # The 3 most recent messages
````

## 5. Using a Slice with a for Loop

A slice is a list. So everything we can do with a list, we can also do with a slice. That includes the **for loop**.

Let's remember how a for loop works:

```python
for score in scores:
    print(score)
```

- `for score in scores:` means **"for each score in scores, run the code below"**.
- Each repetition, one value from the list is stored in the variable `score`, from the first value to the last.
- The line must end with a colon `:`.
- The code to repeat must be **indented** (4 spaces at the beginning). Python uses the indentation to know which lines belong to the loop. This group of indented lines is called a **block**.
- The first line that is **not** indented is outside the loop, and runs only once, after the loop is finished.

Now, if we put a **slice** after `in`, the loop goes through **only that part** of the list.

```python
for score in scores[:3]:      # only the first 3 values
    print(score)
```

````python
# The whole list
scores = [85, 92, 78, 90, 66]

for score in scores:            # 5 values -> the loop runs 5 times
    print(score)
````

````python
# Only the first 3 values
scores = [85, 92, 78, 90, 66]

for score in scores[:3]:        # scores[:3] is [85, 92, 78] -> the loop runs 3 times
    print(score)                # Indented (4 spaces) -> inside the loop -> repeated
````

````python
# Only the last 2 values
scores = [85, 92, 78, 90, 66]

for score in scores[-2:]:       # scores[-2:] is [90, 66] -> the loop runs 2 times
    print(score)
````

````python
# Everything except the first one
# Useful when the first value is a title or a header, not real data
row = ["Scores", 85, 92, 78]

for value in row[1:]:           # row[1:] is [85, 92, 78] -> skip "Scores"
    print(value)
````

### The loop variable can have any name

Remember: the variable after `for` (here, `score`) is a **new variable** that the loop creates. You can give it any name. But a good habit is to use the **singular** form of the list name: `scores` → `score`, `fruits` → `fruit`. Then it's easy to tell "the whole list" from "one value".

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[1:4]:       # fruits[1:4] is ['banana', 'cherry', 'grape']
    print(fruit)                # fruit (one value), fruits (the whole list)
````

### Indentation decides what is repeated

Just like in Lesson 5, the indented lines are repeated, and the first line without indentation runs only once after the loop.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[:2]:        # Only 'apple' and 'banana'
    print("I like", fruit)      # Indented -> runs 2 times
print("That's all!")            # Not indented -> outside the loop -> runs once
````

**About `print("I like", fruit)`:** when you put several values in `print()` separated by **commas** `,`, they are printed on one line, and Python puts **one space** between them automatically. Text inside quotes is printed as it is, and a variable is replaced by the value inside it.

## 6. Calculating with a Slice

### Adding up a part of the list

In Lesson 5, we added up all the values in a list using a **running total**:

1. **Before the loop:** make a variable and set it to 0 → `total = 0`
2. **Inside the loop:** add the current value → `total = total + score`
3. **After the loop:** the variable holds the sum

Remember what `total = total + score` means: `=` means "store". Python first calculates the right side (the old total + the current score), and then stores the result back into `total`.

With a slice, we can add up **only a part** of the list. Just put the slice after `in`.

````python
# Add up only the first 3 scores
scores = [85, 92, 78, 90, 66]
total = 0                           # Before the loop: start from 0

for score in scores[:3]:            # Only 85, 92, 78
    total = total + score           # old total + current score -> stored back in total

print("Total of the first 3:", total)   # After the loop (not indented): 85 + 92 + 78 = 255
````

````python
# Let's watch how total changes
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total = total + score
    print("score:", score, "total:", total)     # Indented -> printed every repetition
````

**Remember the short form `+=`:** `total += score` means exactly the same as `total = total + score`. "Add the right side to the variable, and store the result back in it."

````python
# The same thing, written with +=
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total += score                  # same as: total = total + score

print("Total of the first 3:", total)
````

````python
# Add up only the last 2 scores
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[-2:]:           # Only 90, 66
    total += score                  # same as: total = total + score

print("Total of the last 2:", total)    # 90 + 66 = 156
````

### The average of a part

The **average** is the total divided by the number of values: **average = total / number of values**

We get the number of values with `len()`. But be careful: we need the number of values **in the slice**, not in the whole list. So we use `len()` on the slice.

Also remember: calculate the average **after** the loop (not indented), when the total is complete.

````python
# The average of the first 3 scores
scores = [85, 92, 78, 90, 66]
first_three = scores[:3]            # Store the slice in a variable: [85, 92, 78]
total = 0

for score in first_three:
    total += score                  # same as: total = total + score

average = total / len(first_three)  # 255 / 3 (the length of the SLICE, not the whole list)
print("Total:", total)
print("Count:", len(first_three))
print("Average:", average)
````

````python
# Comparing the average of the first half and the second half
scores = [60, 65, 70, 85, 90, 95]
first_half = scores[:3]             # [60, 65, 70]
second_half = scores[3:]            # [85, 90, 95]

total = 0
for score in first_half:
    total += score
print("First half average:", total / len(first_half))

total = 0                           # Start from 0 again for the second half!
for score in second_half:
    total += score
print("Second half average:", total / len(second_half))
````

### The highest value in a part

In Lesson 5, we found the biggest value like a person would do it on paper:

1. Remember the **first value** as "the biggest so far".
2. Look at each value. If it is bigger than the one you remember, remember this one instead.
3. The one you remember at the end is the biggest.

With a slice, we can find the biggest value in **only a part** of the list.

````python
# The highest score among the last 3 students
scores = [95, 82, 74, 88, 79]
last_three = scores[-3:]            # [74, 88, 79]

highest = last_three[0]             # Step 1: start with the first value of the SLICE (74)

for score in last_three:
    if score > highest:             # Step 2: bigger than the one we remember?
        highest = score             #         if yes, remember this one instead

print("Highest of the last 3:", highest)    # Step 3: 88
````

## 7. Using a Slice with if, in and append()

### if inside the loop

When you put an `if` inside a `for` loop, there are **two levels** of indentation:

```python
for score in scores[:3]:      # no spaces
    if score >= 80:           # 4 spaces -> inside the for (runs for every score)
        print(score)          # 8 spaces -> inside the if (runs only when the condition is True)
```

Remember how `if` works: the indented code under it runs **only when the condition is True**.

````python
# Among the first 4 scores, print only the ones that are 80 or higher
scores = [85, 72, 91, 78, 95, 60]

for score in scores[:4]:            # Only 85, 72, 91, 78
    if score >= 80:                 # 4 spaces: check each score
        print(score)                # 8 spaces: print only if it is 80 or higher
````

````python
# Counting: how many of the last 3 scores passed (60 or higher)?
# Counting works like the running total, but we add 1 instead of the value
scores = [85, 42, 78, 55, 90]
pass_count = 0                      # Before the loop: start from 0

for score in scores[-3:]:           # Only 78, 55, 90
    if score >= 60:
        pass_count += 1             # same as: pass_count = pass_count + 1

print("Passed among the last 3:", pass_count)
````

### in

`value in list` checks whether the value is inside the list, and gives `True` or `False`. A slice is a list, so we can use `in` with a slice too. Then we only check **a part** of the list.

````python
# The top 3 students (the list is already in order: 1st place, 2nd place, ...)
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]

print("Tom" in ranking[:3])         # Is Tom in the top 3? -> True
print("Kate" in ranking[:3])        # Is Kate in the top 3? -> False
print("Kate" in ranking)            # Is Kate in the whole list? -> True
````

````python
# Using in as the condition of an if
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]
name = "Bill"

if name in ranking[:3]:             # Is Bill in the top 3?
    print(name, "gets a medal!")
else:
    print(name, "doesn't get a medal.")
````

### append(): making a new list

Remember `append()`: `list.append(value)` adds the value to the **end** of the list.

We can start with an empty list `[]`, and inside the loop, `append()` only the values we want from a slice.

````python
# From the first 4 scores, make a new list with only the passing scores
scores = [85, 42, 78, 55, 90, 30]
passed = []                         # Before the loop: an empty list

for score in scores[:4]:            # Only 85, 42, 78, 55
    if score >= 60:
        passed.append(score)        # Add it to the end of passed

print(passed)                       # [85, 78]
````

````python
# Make a new list with the last 3 prices doubled
prices = [1000, 2500, 3000, 1500, 4000]
doubled = []

for price in prices[-3:]:           # Only 3000, 1500, 4000
    doubled.append(price * 2)       # price * 2 is calculated first, then added

print(doubled)
````

## 8. Common Mistakes

These are the mistakes almost everyone makes when learning slicing. For each one, we will look at:

- ❌ **The wrong code**
- 🔍 **What went wrong**
- ✅ **The correct way**

Some wrong code causes an error. That code is written as a comment (with `#` at the beginning) so it won't stop the notebook. If you want to see the real error message, remove the `#`, run the cell, and then put the `#` back.

Other wrong code gives **no error, but the wrong result**. These are more dangerous, because Python doesn't warn you.

### Mistake 1: Thinking the end is included

````python
# ❌ Wrong: we want the values at index 1, 2, 3
scores = [85, 92, 78, 90, 66]
#  index:  0   1   2   3   4

print(scores[1:3])      # [92, 78] -> only 2 values! 90 (index 3) is missing
````

🔍 **What went wrong?**

- `scores[1:3]` means "from index 1, up to but **not including** index 3".
- So we only get index 1 and 2. Index 3 is where the slice **stops**.
- No error appears, so it's easy to miss.

✅ **The correct way:** The end must be **one more** than the last index you want. To include index 3, write `4`. (Same as `range()` in Lesson 3: to include 3, you write `range(1, 4)`.)

````python
# ✅ Correct
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # index 1, 2, 3 -> [92, 78, 90]
````

### Mistake 2: Using a comma instead of a colon

````python
# ❌ Wrong
scores = [85, 92, 78, 90, 66]

# print(scores[1, 4])     # TypeError: list indices must be integers or slices, not tuple
````

🔍 **What went wrong?**

- Slicing needs a **colon** `:` between the two numbers, not a comma `,`.
- With a comma, Python doesn't understand it as a slice, and shows an error.
- The error message means: "inside `[ ]`, you must write one number (an index) or a slice with `:`".

✅ **The correct way:** Use a colon `:`.

````python
# ✅ Correct
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # Colon between the numbers
````

### Mistake 3: Using parentheses instead of square brackets

````python
# ❌ Wrong
scores = [85, 92, 78, 90, 66]

# print(scores(1:4))      # SyntaxError
````

🔍 **What went wrong?**

- Getting values from a list (with an index or a slice) always uses **square brackets** `[ ]`.
- Parentheses `( )` are for commands like `print()`, `len()` and `append()`.
- Python doesn't understand `scores(1:4)`, so it shows a `SyntaxError` ("the grammar of the code is wrong").

✅ **The correct way:** Use square brackets `[ ]`.

````python
# ✅ Correct
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # Square brackets
````

### Mistake 4: Mixing up a slice and an index

````python
# ❌ Wrong: we want to add 5 to the first score
scores = [85, 92, 78]

first = scores[0:1]     # This is a LIST: [85], not the number 85
print(first)

# print(first + 5)      # TypeError: can only concatenate list (not "int") to list
````

🔍 **What went wrong?**

- `scores[0:1]` is a **slice**, so the result is a **list** with one value: `[85]`.
- A list and a number can't be added with `+`, so Python shows an error.
- The error message means: "you can only join a list to a list, not a number (int)".
- Look at the square brackets in the output: `[85]`. When you see brackets, it's a list.

✅ **The correct way:** To get **one value**, use an **index** (one number, no colon): `scores[0]`.

````python
# ✅ Correct
scores = [85, 92, 78]

first = scores[0]       # This is the number 85 (no brackets)
print(first)
print(first + 5)        # 90
````

### Mistake 5: Thinking the slice changes the original list

````python
# ❌ Wrong: we want to keep only the first 3 scores
scores = [85, 92, 78, 90, 66]

scores[:3]              # This makes a new list... but it is not stored anywhere!
print(scores)           # [85, 92, 78, 90, 66] -> nothing changed
````

🔍 **What went wrong?**

- Slicing makes a **new list**. It does **not** change the original list.
- On the line `scores[:3]`, Python made the new list `[85, 92, 78]`, but we didn't store it in any variable. So it just disappeared.
- No error appears, so it's hard to notice.

✅ **The correct way:** Store the slice in a variable with `=`. If you want to replace the original list, store it back into the same variable (just like `age = age + 1`: the new value replaces the old one).

````python
# ✅ Correct (1): store it in a new variable
scores = [85, 92, 78, 90, 66]

first_three = scores[:3]
print(first_three)      # [85, 92, 78]
print(scores)           # The original is still complete
````

````python
# ✅ Correct (2): store it back into the same variable
scores = [85, 92, 78, 90, 66]

scores = scores[:3]     # The right side is calculated first ([85, 92, 78]), then stored in scores
print(scores)           # [85, 92, 78] -> the old list is replaced
````

### Mistake 6: Putting the start after the end

````python
# ❌ Wrong: we want the values from index 1 to index 3
scores = [85, 92, 78, 90, 66]

print(scores[4:1])      # [] -> an empty list!
````

🔍 **What went wrong?**

- A slice always goes from **left to right**: from the start to the end.
- `scores[4:1]` starts at index 4 and tries to go to index 1. But 1 is to the **left** of 4, so there is nothing in between.
- The result is an empty list `[]`. No error appears.

✅ **The correct way:** Write the **smaller** index first (start), and the **bigger** index second (end).

````python
# ✅ Correct
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # From index 1 to index 3 -> [92, 78, 90]
````

### Mistake 7: Using [-3:-1] to get the last 3 values

````python
# ❌ Wrong: we want the last 3 values
scores = [85, 92, 78, 90, 66]
#  negative index: -5  -4  -3  -2  -1

print(scores[-3:-1])    # [78, 90] -> only 2 values! The last one (66) is missing
````

🔍 **What went wrong?**

- The end is **not included**. That rule doesn't change with negative indexes.
- `scores[-3:-1]` means "from -3, up to but **not including** -1". So the last value (-1) is left out.

✅ **The correct way:** To include the last value, **leave out the end**: `scores[-3:]`. No end means "all the way to the end".

````python
# ✅ Correct
scores = [85, 92, 78, 90, 66]

print(scores[-3:])      # From -3 to the end -> [78, 90, 66]
````

### Mistake 8: Using the length of the whole list for the average of a slice

````python
# ❌ Wrong: we want the average of the first 3 scores
scores = [80, 90, 100, 50, 60]
total = 0

for score in scores[:3]:
    total += score              # 80 + 90 + 100 = 270

average = total / len(scores)   # 270 / 5 -> wrong! We only added 3 scores
print("Average:", average)      # 54.0
````

🔍 **What went wrong?**

- We added only the first **3** scores (270).
- But `len(scores)` counts **all 5** values in the whole list.
- So we divided by 5 instead of 3. The average is much too low.
- No error appears.

✅ **The correct way:** Divide by the number of values **in the slice**. The easiest way: store the slice in a variable first, and use `len()` on that variable.

````python
# ✅ Correct
scores = [80, 90, 100, 50, 60]
first_three = scores[:3]        # [80, 90, 100]
total = 0

for score in first_three:
    total += score

average = total / len(first_three)  # 270 / 3
print("Average:", average)          # 90.0
````

### Mistake 9: Expecting an error when the slice goes past the end

````python
# ❓ Surprise: this does NOT cause an error
scores = [85, 92, 78]           # Only 3 values (index 0, 1, 2)

print(scores[0:10])             # [85, 92, 78] -> no error!

# But an index that is too big DOES cause an error
# print(scores[10])             # IndexError: list index out of range
````

🔍 **What is going on?**

- With an **index**, Python must find **one** value at that exact position. If there is no such position, it shows an `IndexError` ("the index is outside the list").
- With a **slice**, Python just gives **whatever values it can find** in that range. If the range goes past the end, it stops at the end. No error.
- This is not exactly a mistake, but it can be confusing: a slice that looks wrong may still "work", and you won't notice that the numbers were wrong.

✅ **Tip:** If you want everything until the end, don't guess a big number. Just leave out the end: `scores[0:]` or `scores[:]`.

````python
# ✅ Better
scores = [85, 92, 78]

print(scores[0:])               # From index 0 to the end -> [85, 92, 78]
````
