## 1. dictionary ဘာလို့ လိုတာလဲ?

သိပြီးသား အရာကနေ စကြရအောင်: **list**။

list က square bracket `[ ]` ထဲမှာ တန်ဖိုး အများကြီးကို အစီအစဉ်လိုက် သိမ်းထားပါတယ်။ တန်ဖိုးကို သူ့ရဲ့ **index** (နေရာ နံပါတ်) နဲ့ ယူပြီး index က **0** ကနေ စပါတယ်။

အခု Anna, Tom, Bill ဆိုတဲ့ ကျောင်းသား 3 ယောက်ရဲ့ စာမေးပွဲ အမှတ်တွေကို သိမ်းချင်တယ်လို့ စဉ်းစားကြည့်ပါ။ list နဲ့ဆိုရင် list **နှစ်ခု** လိုပါမယ်: နာမည်အတွက် တစ်ခု၊ အမှတ်အတွက် တစ်ခု။

```python
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]
```

list နှစ်ခုက နေရာအလိုက် ချိတ်ဆက်နေပါတယ်:

| index | 0 | 1 | 2 |
|---|---|---|---|
| `names` | "Anna" | "Tom" | "Bill" |
| `scores` | 85 | 92 | 78 |

ဒါကြောင့် `names[1]` က Tom ဖြစ်ပြီး `scores[1]` က Tom ရဲ့ အမှတ် ဖြစ်ပါတယ်။

**ဒါပေမဲ့ ဒီနည်းမှာ ပြဿနာတွေ ရှိပါတယ်:**

- Tom ရဲ့ အမှတ်ကို ရှာဖို့ Tom က index 1 မှာ ရှိတယ်ဆိုတာ အရင် **မှတ်ထား** ရပါမယ်။ Python က "Tom" ဆိုတဲ့ နာမည်ကနေ ရှာမပေးနိုင်ပါဘူး။
- list တစ်ခုမှာ နာမည် ထည့်တာ ဒါမှမဟုတ် ဖျက်တာ လုပ်ပြီး နောက် list မှာ အတူတူ လုပ်ဖို့ မေ့သွားရင် နာမည်နဲ့ အမှတ်တွေ **ရောထွေး** သွားပါတယ်။

**"Tom ရဲ့ အမှတ် ဘယ်လောက်လဲ?"** လို့ တန်းမေးလို့ရရင် ပိုလွယ်မှာ မဟုတ်လား?

dictionary က အဲဒါကို အတိအကျ လုပ်ပေးပါတယ်။

````python
# list နှစ်ခုနဲ့ဆိုရင် နာမည်နဲ့ အမှတ်က သီးခြား နေရာတွေမှာ ရှိပါတယ်
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

# Tom ရဲ့ အမှတ်ကို ယူဖို့ Tom က index 1 မှာ ရှိတယ်ဆိုတာ သိရပါမယ်
print(names[1])         # Tom
print(scores[1])        # 92 -> Tom ရဲ့ အမှတ်
````

````python
# ပြဿနာ: list တစ်ခု ပြောင်းပြီး နောက်တစ်ခု မပြောင်းရင် ရောထွေးသွားပါတယ်
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

names.remove("Anna")    # names ထဲက Anna ကို ဖျက်ခဲ့ပေမဲ့ scores ထဲက 85 ကို ဖျက်ဖို့ မေ့သွားပါတယ်

print(names[0])         # Tom
print(scores[0])        # 85 -> ဒါက Tom ရဲ့ မဟုတ်ဘဲ Anna ရဲ့ အမှတ်ပါ! အခု မကိုက်ညီတော့ပါဘူး
````

## 2. dictionary ဆိုတာဘာလဲ?

တကယ့် အဘိဓာန်၊ စကားလုံး လေ့လာဖို့ သုံးတဲ့ စာအုပ်ကို စဉ်းစားကြည့်ပါ။

- **စကားလုံး** တစ်လုံးကို ရှာပါတယ် (ဥပမာ "apple")။
- သူ့ရဲ့ **အဓိပ္ပာယ်** ကို တွေ့ပါတယ် ("အလုံးလိုက် သစ်သီး တစ်မျိုး")။

အဲဒီ စကားလုံးက ဘယ်စာမျက်နှာမှာ ရှိလဲ သိစရာ မလိုပါဘူး။ စကားလုံးကို ရှာလိုက်ရုံနဲ့ အဓိပ္ပာယ်က ဘေးမှာ တန်းရှိနေပါတယ်။

Python ရဲ့ **dictionary** ကလည်း ဒီလိုပဲ အလုပ်လုပ်ပါတယ်။ တန်ဖိုးတွေကို **အတွဲ** လိုက် သိမ်းပါတယ်:

| Python စကားလုံး | အဓိပ္ပာယ် | တကယ့် အဘိဓာန်မှာ | ကျွန်တော်တို့ ဥပမာမှာ |
|---|---|---|---|
| **key** | ရှာတဲ့ နာမည် | စကားလုံး | ကျောင်းသား နာမည် ("Tom") |
| **value (တန်ဖိုး)** | ရလာတဲ့ အရာ | အဓိပ္ပာယ် | အမှတ် (92) |
| **pair (အတွဲ)** | key တစ်ခုနဲ့ သူ့တန်ဖိုး အတူတူ | စကားလုံး တစ်လုံးနဲ့ သူ့အဓိပ္ပာယ် | "Tom" နဲ့ 92 |

**နာမည်တံဆိပ် ကပ်ထားတဲ့ locker** တွေလိုလည်း စဉ်းစားလို့ရပါတယ်:
- **list** မှာ locker တစ်ခန်းစီမှာ **နံပါတ်** (0, 1, 2 ...) ရှိပါတယ်။
- **dictionary** မှာ locker တစ်ခန်းစီမှာ **နာမည်တံဆိပ်** ("Anna", "Tom", "Bill" ...) ရှိပါတယ်။

locker ကို ဖွင့်ဖို့ နံပါတ်အစား နာမည်တံဆိပ်ကို သုံးပါတယ်။

## 3. dictionary ဖန်တီးခြင်း

အသေးဆုံး dictionary နဲ့ စကြရအောင်: **အတွဲ တစ်ခုတည်း**။

```python
scores = {"Tom": 92}
```

အစိတ်အပိုင်း တစ်ခုချင်းစီကို ကြည့်ရအောင်:

| အစိတ်အပိုင်း | အဓိပ္ပာယ် |
|---|---|
| `{` | dictionary စပါတယ် (square bracket မဟုတ်ဘဲ **curly bracket (တွန့်ကွင်း)**) |
| `"Tom"` | **key** (ဒီမှာ string ဖြစ်လို့ quote ထဲမှာ ရှိပါတယ်) |
| `:` | **colon**။ key ကို သူ့တန်ဖိုးနဲ့ ချိတ်ပေးပါတယ် |
| `92` | **တန်ဖိုး** |
| `}` | dictionary ဆုံးပါတယ် |

မှတ်ထားပါ: `=` က **"ညာဘက်က တန်ဖိုးကို ဘယ်ဘက်က သေတ္တာထဲ သိမ်းပါ"** လို့ အဓိပ္ပာယ်ရပါတယ်။ ဒါကြောင့် ဒီစာကြောင်းက dictionary တစ်ခုလုံးကို `scores` ဆိုတဲ့ variable ထဲ သိမ်းပါတယ်။

````python
# အတွဲ တစ်ခုတည်း ပါတဲ့ dictionary
scores = {"Tom": 92}

print(scores)           # {'Tom': 92}
````

### အတွဲ တစ်ခုထက် ပိုခြင်း

dictionary ထဲမှာ အတွဲ အများကြီး ထည့်ဖို့ **comma** `,` နဲ့ ခွဲပါ။ list ထဲမှာ တန်ဖိုးတွေကို comma နဲ့ ခွဲသလိုပါပဲ။

```python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
```

- `"Anna": 85` က အတွဲ တစ်ခု
- `"Tom": 92` က အတွဲ တစ်ခု
- `"Bill": 78` က အတွဲ တစ်ခု

ဒါကြောင့် ဒီ dictionary မှာ **အတွဲ 3 ခု** ရှိပါတယ်။

````python
# အတွဲ နှစ်ခု ပါတဲ့ dictionary
scores = {"Anna": 85, "Tom": 92}
print(scores)
````

````python
# အတွဲ သုံးခု ပါတဲ့ dictionary
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)
````

### စာကြောင်း တစ်ကြောင်းမှာ အတွဲ တစ်ခုစီ ရေးခြင်း

အတွဲ များလာရင် ရှည်လျားတဲ့ စာကြောင်း တစ်ကြောင်းတည်းက ဖတ်ရ ခက်ပါတယ်။ **စာကြောင်း တစ်ကြောင်းမှာ အတွဲ တစ်ခုစီ** ရေးလို့ရပါတယ်။ Python က အတူတူပဲ ဖတ်ပါတယ်။ အတွဲတိုင်းရဲ့ အဆုံးမှာ comma ကို မမေ့ပါနဲ့။

````python
# dictionary အတူတူကို စာကြောင်း တစ်ကြောင်းမှာ အတွဲ တစ်ခုစီ ရေးထားပါတယ် (ဖတ်ရ ပိုလွယ်ပါတယ်)
scores = {
    "Anna": 85,
    "Tom": 92,
    "Bill": 78,
}
print(scores)
````

### နှိုင်းယှဉ်ကြည့်ပါ: list နဲ့ dictionary

| | list | dictionary |
|---|---|---|
| bracket | square bracket `[ ]` | curly bracket `{ }` |
| အထဲမှာ ဘာရှိလဲ | တန်ဖိုးတွေပဲ | `key: တန်ဖိုး` **အတွဲ** တွေ |
| ဥပမာ | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92, "Bill": 78}` |

````python
# အမှတ် အတူတူကို list အဖြစ် နဲ့ dictionary အဖြစ်
score_list = [85, 92, 78]
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list)       # အမှတ်တွေပဲ ရှိပါတယ်။ ဘယ်သူ့ အမှတ်လဲ မသိပါဘူး
print(score_dict)       # အမှတ်တိုင်းရဲ့ ဘေးမှာ နာမည် ရှိပါတယ်
````

````python
# dictionary က data type အသစ်ပါ: dict ("dictionary" ရဲ့ အတိုကောက်)
# type() က တန်ဖိုးတစ်ခုဟာ ဘယ်လို အမျိုးအစားလဲ ပြောပြပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(type(scores))     # <class 'dict'>
````

### တခြား dictionary အမျိုးအစားများ

တန်ဖိုးတွေက data type ဘာမဆို ဖြစ်လို့ရပါတယ်: string, integer, float, bool။ key တွေက များသောအားဖြင့် **string** ပါ။

````python
# တန်ဖိုးတွေက string: စကားလုံး -> သူ့အဓိပ္ပာယ်၊ တကယ့် အဘိဓာန်လိုပဲ
english = {"apple": "a round fruit", "dog": "an animal that barks"}
print(english)
````

````python
# တန်ဖိုးတွေက string: နိုင်ငံ -> မြို့တော်
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}
print(capitals)
````

````python
# တန်ဖိုးတွေက bool: ကျောင်းသား -> ဒီနေ့ အတန်းတက်လား?
attendance = {"Anna": True, "Tom": False, "Bill": True}
print(attendance)
````

## 4. key နဲ့ တန်ဖိုး ယူခြင်း

dictionary ထဲက တန်ဖိုး ယူဖို့ dictionary နာမည်နဲ့ square bracket `[ ]` ထဲမှာ **key** ကို ရေးပါ:

```python
scores["Tom"]      # "scores ထဲက 'Tom' key ရဲ့ တန်ဖိုး"
```

ဒါက list နဲ့ အရမ်းတူပါတယ်။ တစ်ခုတည်းသော ကွာခြားချက်က square bracket ထဲက အရာပါ:

| | list | dictionary |
|---|---|---|
| `[ ]` ထဲမှာ ဘာရေးလဲ | **index** (နေရာ နံပါတ်) | **key** (နာမည်) |
| ဥပမာ | `score_list[1]` | `score_dict["Tom"]` |
| အဓိပ္ပာယ် | "နေရာ 1 က တန်ဖိုး" | "key က 'Tom' ဖြစ်တဲ့ တန်ဖိုး" |

**အရေးကြီးပါတယ်:** dictionary ကို **ဖန်တီး** တဲ့အခါ curly bracket `{ }` ကို သုံးပေမဲ့ အထဲက တန်ဖိုးကို **ယူ** တဲ့အခါ square bracket `[ ]` ကို သုံးပါတယ်။

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # "Tom" key ရဲ့ တန်ဖိုး -> 92
print(scores["Anna"])       # "Anna" key ရဲ့ တန်ဖိုး -> 85
print(scores["Bill"])       # "Bill" key ရဲ့ တန်ဖိုး -> 78
````

````python
# နှိုင်းယှဉ်ကြည့်ပါ: Tom ရဲ့ အမှတ်ကို list ကနေ နဲ့ dictionary ကနေ ယူခြင်း
score_list = [85, 92, 78]                           # Anna, Tom, Bill
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list[1])        # list: Tom က index 1 မှာ ရှိတယ်ဆိုတာ မှတ်ထားရပါမယ်
print(score_dict["Tom"])    # dictionary: "Tom" ဆိုတဲ့ နာမည်ကို သုံးရုံပါပဲ
````

````python
# နောက်ထပ် ဥပမာများ
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

print(capitals["Korea"])    # Seoul
print(capitals["France"])   # Paris
````

````python
english = {"apple": "a round fruit", "dog": "an animal that barks"}

print(english["dog"])       # "dog" ဆိုတဲ့ စကားလုံးကို ရှာပါတယ် -> သူ့အဓိပ္ပာယ်
````

### တန်ဖိုးကို သာမန် variable လို သုံးလို့ရပါတယ်

dictionary ကနေ ရလာတဲ့ တန်ဖိုးက သာမန် တန်ဖိုး တစ်ခုပါပဲ။ variable ထဲ သိမ်းတာ၊ တွက်တာ၊ နှိုင်းယှဉ်တာ လုပ်လို့ရပါတယ်။

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

tom_score = scores["Tom"]                   # Tom ရဲ့ အမှတ်ကို variable အသစ်ထဲ သိမ်းပါတယ်
print(tom_score)

print(scores["Anna"] + scores["Tom"])       # 85 + 92 = 177 (တွက်လို့ရပါတယ်)
print(scores["Tom"] > scores["Bill"])       # 92 > 78 က True (နှိုင်းယှဉ်လို့ရပါတယ်)
````

````python
# တန်ဖိုးကို if ထဲမှာ သုံးခြင်း (if က condition True ဖြစ်မှသာ indent လုပ်ထားတဲ့ code ကို run ပါတယ်)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

if scores["Bill"] >= 80:
    print("Bill passed with a high score!")
else:
    print("Bill needs to study more.")
````

### key က variable ဖြစ်လို့ရပါတယ်

key ကို တိုက်ရိုက် ရေးမယ့်အစား variable ထဲမှာ အရင် သိမ်းထားလို့ရပါတယ်။ ဒါဆိုရင် `[ ]` ထဲမှာ variable ကို **quote မပါဘဲ** ရေးပါ။

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Anna"               # name variable ထဲမှာ "Anna" ဆိုတဲ့ string ရှိပါတယ်
print(scores[name])         # scores[name] က scores["Anna"] နဲ့ အတူတူ -> 85

name = "Bill"               # အခု variable ထဲမှာ "Bill" ရှိပါတယ်
print(scores[name])         # scores["Bill"] -> 78
````

**တန်ဖိုး အများကြီး ပြခြင်း အကြောင်း:** နောက် cell မှာ `print(name, scores[name])` လို့ ရေးပါတယ်။ `print()` ထဲမှာ တန်ဖိုး အများကြီးကို **comma** `,` နဲ့ ခွဲထည့်ရင် စာကြောင်း တစ်ကြောင်းတည်းမှာ ပြပြီး Python က သူတို့ကြားမှာ **space တစ်ခု** ကို အလိုအလျောက် ထည့်ပေးပါတယ်။

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Tom"
print(name, scores[name])   # Tom 92 -> နာမည်နဲ့ အမှတ် စာကြောင်း တစ်ကြောင်းတည်းမှာ
````

## 5. တန်ဖိုး ပြောင်းခြင်းနဲ့ အတွဲအသစ် ထည့်ခြင်း

တန်ဖိုး ပြောင်းဖို့ ဒါမှမဟုတ် အတွဲအသစ် ထည့်ဖို့ ပုံစံ အတူတူကို သုံးပါတယ်:

```python
dictionary_နာမည်[key] = တန်ဖိုး
```

key က dictionary ထဲမှာ **ရှိပြီးသား** လား ဆိုတာပေါ် မူတည်ပြီး Python က နှစ်ခုထဲက တစ်ခုကို လုပ်ပါတယ်:

| key က dictionary ထဲမှာ ရှိပြီးသားလား? | ဘာဖြစ်လဲ |
|---|---|
| **ရှိပါတယ်** | တန်ဖိုးဟောင်းကို တန်ဖိုးအသစ်နဲ့ **အစားထိုး** ပါတယ် |
| **မရှိပါဘူး** | **အတွဲအသစ်** ကို ထည့်ပါတယ် |

ဒါက list နဲ့ မတူပါဘူး။ list မှာ ရှိပြီးသား index က တန်ဖိုးကိုပဲ ပြောင်းလို့ရပါတယ်။ ထည့်ဖို့ `append()` လိုပါတယ်။ dictionary မှာတော့ `[key] = တန်ဖိုး` တစ်ခုတည်းနဲ့ ပြောင်းလည်း ရ၊ ထည့်လည်း ရပါတယ်။

### တန်ဖိုး ပြောင်းခြင်း (key ရှိပြီးသား)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Bill"] = 88         # "Bill" က key ဖြစ်ပြီးသား -> တန်ဖိုး 78 ကို 88 နဲ့ အစားထိုးပါတယ်
print(scores)               # Bill ရဲ့ အမှတ်ပဲ ပြောင်းသွားပါတယ်
````

````python
# တန်ဖိုးဟောင်းကို သုံးပြီး တန်ဖိုး ပြောင်းခြင်း (age = age + 1 နဲ့ စိတ်ကူး အတူတူ)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Anna"] = scores["Anna"] + 5     # ညာဘက် အရင်: 85 + 5 = 90၊ ပြီးမှ ပြန်သိမ်းပါတယ်
print(scores["Anna"])                   # 90
````

````python
# အတိုပုံစံ += လည်း ရပါတယ် (အဓိပ္ပာယ်တူ: scores["Tom"] = scores["Tom"] + 3)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Tom"] += 3
print(scores["Tom"])                    # 95
````

### အတွဲအသစ် ထည့်ခြင်း (key မရှိသေး)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Kate"] = 95         # "Kate" က key "မဟုတ်သေးပါ" -> အတွဲအသစ် ထည့်ပါတယ်
print(scores)               # အခု အတွဲ 4 ခု ရှိပါတယ်
````

````python
# ဗလာ dictionary နဲ့ စပြီး အတွဲတွေကို တစ်ခုချင်း ထည့်ခြင်း
capitals = {}                   # ဗလာနဲ့ စပါတယ်
print(capitals)

capitals["Korea"] = "Seoul"     # အတွဲ တစ်ခု ထည့်ပါတယ်
print(capitals)

capitals["Japan"] = "Tokyo"     # နောက်ထပ် အတွဲ တစ်ခု
print(capitals)

capitals["France"] = "Paris"    # နောက်ထပ် တစ်ခု
print(capitals)
````

````python
# list နဲ့ နှိုင်းယှဉ်ပါ: ထည့်ဖို့ list က append() လိုပေမဲ့ dictionary က [key] = တန်ဖိုး ပဲ လိုပါတယ်
fruit_list = ["apple", "banana"]
fruit_list.append("cherry")         # list: append() က နောက်ဆုံးမှာ ထည့်ပါတယ်
print(fruit_list)

prices = {"apple": 1000, "banana": 500}
prices["cherry"] = 3000             # dictionary: [key အသစ်] = တန်ဖိုး
print(prices)
````

## 6. len() နဲ့ in

### len(): အတွဲ ဘယ်နှစ်ခု ရှိလဲ?

`len()` ကို မှတ်မိပါ: list ထဲမှာ တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ ပြောပြပါတယ်။ dictionary နဲ့ဆိုရင် `len()` က **အတွဲ** ဘယ်နှစ်ခု ရှိလဲ ပြောပြပါတယ်။

အတွဲ တစ်ခု (`"Tom": 92`) ကို 2 မဟုတ်ဘဲ **1** လို့ ရေတွက်ပါတယ်။

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(len(scores))          # အတွဲ 3 ခု -> 3

one_pair = {"Tom": 92}
print(len(one_pair))        # အတွဲ 1 ခု -> 1 (2 မဟုတ်ပါဘူး!)

empty_dict = {}
print(len(empty_dict))      # အတွဲ 0 ခု -> 0
````

````python
# အတွဲ ထည့်ရင် len() ပြောင်းပါတယ်
scores = {"Anna": 85, "Tom": 92}
print(len(scores))          # 2

scores["Bill"] = 78         # အတွဲအသစ် ထည့်ပါတယ်
print(len(scores))          # 3
````

### in: ဒီ key က dictionary ထဲမှာ ရှိလား?

`in` ကို မှတ်မိပါ: `တန်ဖိုး in list` က တန်ဖိုးဟာ list ထဲမှာ ရှိမရှိ စစ်ပြီး `True` ဒါမှမဟုတ် `False` ကို ပေးပါတယ်။

dictionary နဲ့ဆိုရင် `in` က **key** တွေကို စစ်ပါတယ်။

```python
"Tom" in scores      # "Tom" ဆိုတဲ့ key ရှိလား?
```

**သတိထားပါ:** `in` က တန်ဖိုးတွေကို မဟုတ်ဘဲ **key တွေကိုပဲ** စစ်ပါတယ်။ 92 က dictionary ထဲမှာ တန်ဖိုး အဖြစ် ရှိနေပေမဲ့ `92 in scores` က `False` ပါ။

key က dictionary ထဲမှာ **မရှိဘူး** ဆိုတာ စစ်ချင်ရင် `not in` ကို သုံးပါ။

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" က key လား? -> True
print("Kate" in scores)     # "Kate" က key လား? -> False
print("Kate" not in scores) # "Kate" က key "မဟုတ်" ဘူးလား? -> True
````

````python
# in က တန်ဖိုးတွေကို မဟုတ်ဘဲ KEY တွေကိုပဲ စစ်ပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" က key -> True
print(92 in scores)         # 92 က key မဟုတ်ဘဲ တန်ဖိုး -> False
````

````python
# in ကို if နဲ့ သုံးခြင်း: အရင်စစ်ပြီးမှ တန်ဖိုး ယူပါ
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    print(name, scores[name])
else:
    print(name, "is not in the list.")
````

````python
# ရိုးရှင်းတဲ့ အဘိဓာန် program
english = {"apple": "a round fruit", "dog": "an animal that barks"}
word = "cat"

if word in english:
    print(english[word])
else:
    print("Sorry, I don't know that word.")
````

## 8. dictionary နဲ့ for loop ကို သုံးခြင်း

for loop က list နဲ့ ဘယ်လို အလုပ်လုပ်လဲ ပြန်သတိရကြရအောင်:

```python
for score in score_list:
    print(score)
```

- အကြိမ်တိုင်း list ထဲက တန်ဖိုး တစ်ခုကို `score` variable ထဲ သိမ်းပါတယ်။
- `for` စာကြောင်းက colon `:` နဲ့ ဆုံးပါတယ်။
- ထပ်ခါထပ်ခါ run မယ့် code ကို **indent** (space 4 ခု) လုပ်ပါတယ်။ ဘယ်စာကြောင်းတွေက loop ထဲမှာ ပါလဲဆိုတာ Python က indent နဲ့ သိပါတယ်။
- indent မလုပ်ထားတဲ့ ပထမ စာကြောင်းက loop အပြင်မှာ ရှိပြီး loop ပြီးမှ တစ်ကြိမ်ပဲ run ပါတယ်။

dictionary နဲ့ဆိုရင် for loop က **key** တွေကို တစ်ခုချင်း ပေးပါတယ်။

```python
for name in scores:
    print(name)          # Anna, Tom, Bill
```

key ရပြီဆိုရင် အပိုင်း 4 လိုပဲ `scores[name]` နဲ့ တန်ဖိုးကို ယူလို့ရပါတယ်။

````python
# for loop က dictionary ရဲ့ KEY တွေကို ပေးပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:         # name က "Anna"၊ ပြီးတော့ "Tom"၊ ပြီးတော့ "Bill" ဖြစ်ပါတယ်
    print(name)
````

````python
# key ကို သုံးပြီး တန်ဖိုး ယူခြင်း
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(scores[name])     # scores["Anna"]၊ ပြီးတော့ scores["Tom"]၊ ပြီးတော့ scores["Bill"]
````

````python
# key နဲ့ တန်ဖိုးကို စာကြောင်း တစ်ကြောင်းတည်းမှာ (comma ပါတဲ့ print)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])
````

### loop variable ရဲ့ နာမည်

`for` နောက်က variable (ဒီမှာ `name`) က variable အသစ်ပါ။ ကြိုက်တဲ့ နာမည် ပေးလို့ရပါတယ်။ **key** တွေက ဘာလဲဆိုတာ ပြောတဲ့ နာမည်ကို ရွေးပါ: နာမည်တွေ → `name`, နိုင်ငံတွေ → `country`, စကားလုံးတွေ → `word`။

````python
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

for country in capitals:            # key တွေက နိုင်ငံတွေ ဖြစ်လို့ country လို့ ခေါ်ပါတယ်
    print(country, capitals[country])
````

### ဘာကို ထပ်ခါထပ်ခါ run မလဲ ဆိုတာ indent က ဆုံးဖြတ်ပါတယ်

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])       # indent လုပ်ထား -> key တစ်ခုစီအတွက် ထပ်ခါထပ်ခါ run ပါတယ်
print("That's everyone!")           # indent မလုပ်ထား -> loop အပြင်မှာ -> တစ်ကြိမ် run ပါတယ်
````

### တန်ဖိုးတွေကို ပေါင်းခြင်း

**ပေါင်းစုတန်ဖိုး** ကို သတိရပါ:

1. **loop မစခင်:** `total = 0`
2. **loop ထဲမှာ:** လက်ရှိ တန်ဖိုးကို ပေါင်းထည့်ပါ → `total = total + scores[name]`
3. **loop ပြီးရင်:** `total` ထဲမှာ ပေါင်းလဒ် ရှိပါတယ်

`total = total + scores[name]` ဆိုတာ: ညာဘက် (စုစုပေါင်းဟောင်း + ဒီကျောင်းသားရဲ့ အမှတ်) ကို အရင်တွက်ပြီး ရလဒ်ကို `total` ထဲ ပြန်သိမ်းပါ လို့ အဓိပ္ပာယ်ရပါတယ်။ အတိုပုံစံက `total += scores[name]` ပါ။

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0                           # loop မစခင်: 0 ကနေ စပါတယ်

for name in scores:
    total = total + scores[name]    # စုစုပေါင်းဟောင်း + ဒီကျောင်းသားရဲ့ အမှတ်

print("Total:", total)              # loop ပြီးမှ (indent မလုပ်ထား): 85 + 92 + 78 = 255
````

````python
# ပျမ်းမျှ = စုစုပေါင်း / တန်ဖိုး အရေအတွက်
# len(scores) က အတွဲ အရေအတွက် (= ကျောင်းသား အရေအတွက်)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:
    total += scores[name]           # အဓိပ္ပာယ်တူ: total = total + scores[name]

average = total / len(scores)       # loop ပြီးမှ: 255 / 3
print("Average:", average)
````

### loop ထဲက if

`for` ထဲမှာ `if` ကို ထည့်ရင် indent နှစ်ဆင့် ဖြစ်ပါတယ်: `if` အတွက် space 4 ခု (`for` ထဲမှာ)၊ `if` ထဲက code အတွက် space 8 ခု။

````python
# အောင်တဲ့ ကျောင်းသားတွေကိုပဲ ပြခြင်း (60 နဲ့အထက်)
scores = {"Anna": 85, "Tom": 42, "Bill": 78, "Kate": 55}

for name in scores:
    if scores[name] >= 60:          # space 4 ခု: ကျောင်းသား တစ်ယောက်စီရဲ့ အမှတ်ကို စစ်ပါတယ်
        print(name, "passed")       # space 8 ခု: condition True ဖြစ်မှသာ
````

````python
# ဒီနေ့ ဘယ်သူတွေ အတန်းတက်လဲ?
attendance = {"Anna": True, "Tom": False, "Bill": True}

for name in attendance:
    if attendance[name]:            # တန်ဖိုးက True ဒါမှမဟုတ် False ဖြစ်ပြီးသားမို့ condition အဖြစ် တိုက်ရိုက် သုံးလို့ရပါတယ်
        print(name, "is here.")
    else:
        print(name, "is absent.")
````

## 9. list နဲ့ dictionary

| | list | dictionary |
|---|---|---|
| ဖန်တီးခြင်း | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92}` |
| ဖန်တီးတဲ့ bracket | `[ ]` | `{ }` |
| တန်ဖိုးကို ဘာနဲ့ ရှာလဲ | **index** (နေရာ နံပါတ်) | **key** (နာမည်) |
| တန်ဖိုး ယူခြင်း | `score_list[1]` | `score_dict["Tom"]` |
| တန်ဖိုး ပြောင်းခြင်း | `score_list[1] = 95` | `score_dict["Tom"] = 95` |
| ထည့်ခြင်း | `score_list.append(88)` | `score_dict["Kate"] = 88` |
| ဖျက်ခြင်း | `score_list.remove(92)` | `del score_dict["Tom"]` |
| `len()` က ဘာကို ရေတွက်လဲ | တန်ဖိုးတွေ | အတွဲတွေ |
| `in` က ဘာကို စစ်လဲ | တန်ဖိုးတွေ | **key** တွေ |
| `for x in ...` က ဘာကို ပေးလဲ | တန်ဖိုးတွေ | **key** တွေ |

**ဘယ်အချိန် ဘာကို သုံးမလဲ?**
- **အစီအစဉ်** က အရေးကြီးရင် ဒါမှမဟုတ် တန်ဖိုးတွေမှာ နာမည် မလိုရင် → **list** (ဥပမာ ဒီနေ့ စာမေးပွဲ အမှတ်တွေ အစီအစဉ်လိုက်)
- တန်ဖိုးကို **နာမည်နဲ့** ရှာချင်ရင် → **dictionary** (ဥပမာ ကျောင်းသား တစ်ယောက်စီရဲ့ အမှတ်၊ စကားလုံးနဲ့ သူ့အဓိပ္ပာယ်)

### အမှား 1: မရှိတဲ့ key ကို သုံးခြင်း

````python
# ❌ မှားပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores["Kate"])     # KeyError: 'Kate'
# print(scores["tom"])      # KeyError: 'tom'
````

🔍 **ဘာမှားသွားလဲ?**

- `"Kate"` က dictionary ထဲမှာ key မဟုတ်ပါဘူး။ Python က ရှာမတွေ့လို့ `KeyError` ("ဒီလို key မရှိပါ") ကို ပြပါတယ်။
- `"tom"` (t အသေး) နဲ့ `"Tom"` (T အကြီး) က string **မတူပါဘူး**။ key ကို စာလုံးအကြီး၊ အသေး အပါအဝင် **အတိအကျ** တူအောင် ရေးရပါမယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** key ကို dictionary ထဲမှာ ရှိတဲ့အတိုင်း အတိအကျ ရေးပါ။ key ရှိမရှိ မသေချာရင် `in` နဲ့ အရင်စစ်ပါ။

````python
# ✅ မှန်ပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # key နဲ့ အတိအကျ တူပါတယ်

name = "Kate"
if name in scores:          # အရင်စစ်ပါတယ်
    print(scores[name])
else:
    print(name, "is not in the dictionary.")
````

### အမှား 2: dictionary ကို square bracket နဲ့ ဖန်တီးခြင်း

````python
# ❌ မှားပါတယ်
# scores = ["Anna": 85, "Tom": 92]      # SyntaxError
````

🔍 **ဘာမှားသွားလဲ?**

- square bracket `[ ]` က **list** ဖန်တီးဖို့ပါ။ list က တန်ဖိုးတွေပဲ သိမ်းလို့ Python က `[ ]` ထဲက `"Anna": 85` ကို နားမလည်ပါဘူး။
- error က `SyntaxError` ("code ရဲ့ သဒ္ဒါ မှားနေပါတယ်") ပါ။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** dictionary ဖန်တီးဖို့ **curly bracket** `{ }` ကို သုံးပါ။ (တန်ဖိုး **ယူ** တဲ့အခါ square bracket ကို ဆက်သုံးပါတယ်: `scores["Anna"]`။)

### အမှား 3: key ပတ်လည်က quote ကို မေ့ခြင်း

````python
# ❌ မှားပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[Anna])       # NameError: name 'Anna' is not defined
````

🔍 **ဘာမှားသွားလဲ?**

- key က quote ပါတဲ့ `"Anna"` ဆိုတဲ့ **string** ပါ။
- quote မပါရင် Python က `Anna` ကို **variable** ရဲ့ နာမည်လို့ ထင်ပါတယ်။ ဒါပေမဲ့ `Anna` ဆိုတဲ့ variable မရှိပါဘူး။
- error က `NameError` ("ဒီနာမည်နဲ့ variable မရှိပါ") ပါ။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** string key ပတ်လည်မှာ quote ထည့်ပါ။ (အပိုင်း 4 က `scores[name]` လို key ကို variable ထဲ သိမ်းထားမှသာ quote မလိုပါဘူး။)

````python
# ✅ မှန်ပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # quote ပါတဲ့ string key

name = "Anna"
print(scores[name])         # quote မပါတဲ့ variable (variable ထဲမှာ "Anna" ရှိပါတယ်)
````

🔍 **ဘာမှားသွားလဲ?**

- list မှာ `[0]` က "ပထမ တန်ဖိုး" လို့ အဓိပ္ပာယ်ရပါတယ်။ ဒါပေမဲ့ dictionary က index နံပါတ် မသုံးပါဘူး။ **key** ကို သုံးပါတယ်။
- ဒါကြောင့် Python က `0` ဆိုတဲ့ နာမည်ရှိတဲ့ **key** ကို ရှာပါတယ်။ ဒီလို key မရှိလို့ `KeyError: 0` ကို ပြပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** key (နာမည်) ကို သုံးပါ။

### အမှား 5: key အတူတူကို နှစ်ခါ ရေးခြင်း

````python
# ❌ မှားပါတယ်: Tom လို့ နာမည်တူတဲ့ ကျောင်းသား နှစ်ယောက်လား?
scores = {"Anna": 85, "Tom": 92, "Tom": 60}

print(scores)               # {'Anna': 85, 'Tom': 60} -> ပထမ Tom (92) ပျောက်သွားပါပြီ!
print(len(scores))          # 3 မဟုတ်ဘဲ 2
````

🔍 **ဘာမှားသွားလဲ?**

- dictionary မှာ key တစ်ခုစီက **တစ်ကြိမ်ပဲ** ပါလို့ရပါတယ်။
- key အတူတူကို ထပ်ရေးရင် အပိုင်း 5 လိုပဲ အလုပ်လုပ်ပါတယ်: key ရှိပြီးသား ဖြစ်လို့ တန်ဖိုးဟောင်းကို **အစားထိုး** ပါတယ်။ 92 ကို 60 နဲ့ အစားထိုးသွားပါတယ်။
- error မရှိလို့ လွတ်သွားဖို့ လွယ်ပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** key အားလုံးကို မတူအောင် လုပ်ပါ။ လူနှစ်ယောက် နာမည်တူရင် key မတူတာ ပေးပါ။

````python
# ✅ မှန်ပါတယ်
scores = {"Anna": 85, "Tom K.": 92, "Tom L.": 60}

print(scores)
print(len(scores))          # 3
````

### အမှား 6: in က တန်ဖိုးတွေကို စစ်တယ်လို့ ထင်ခြင်း

````python
# ❌ မှားပါတယ်: တစ်ယောက်ယောက် 92 မှတ် ရလား သိချင်တာပါ
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(92 in scores)         # False -> ဒါပေမဲ့ Tom က 92 ရခဲ့ပါတယ်!
````

🔍 **ဘာမှားသွားလဲ?**

- dictionary နဲ့ဆိုရင် `in` က တန်ဖိုးတွေ (အမှတ်တွေ) ကို မဟုတ်ဘဲ **key တွေ (နာမည်တွေ) ကိုပဲ** စစ်ပါတယ်။
- 92 က key မဟုတ်ဘဲ တန်ဖိုး ဖြစ်လို့ ရလဒ်က `False` ပါ။
- error မရှိပေမဲ့ အဖြေ မှားပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** for loop နဲ့ dictionary ကို ဖြတ်သွားပြီး တန်ဖိုး တစ်ခုစီကို နှိုင်းယှဉ်ပါ။

````python
# ✅ မှန်ပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    if scores[name] == 92:      # တန်ဖိုး တစ်ခုစီကို 92 နဲ့ နှိုင်းယှဉ်ပါတယ်
        print(name, "got 92 points.")
````

### အမှား 7: for loop က တန်ဖိုးတွေကို ပေးတယ်လို့ ထင်ခြင်း

````python
# ❌ မှားပါတယ်: အမှတ်အားလုံးကို ပေါင်းချင်တာပါ
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

# for score in scores:
#     total += score        # TypeError: unsupported operand type(s) for +=: 'int' and 'str'
````

🔍 **ဘာမှားသွားလဲ?**

- dictionary နဲ့ဆိုရင် for loop က တန်ဖိုးတွေကို မဟုတ်ဘဲ **key** တွေကို ပေးပါတယ်။ ဒါကြောင့် `score` က 85, 92, 78 မဟုတ်ဘဲ `"Anna"`, `"Tom"`, `"Bill"` ဖြစ်ပါတယ်။
- `total += score` က ဂဏန်း (`0`) ထဲကို string (`"Anna"`) ပေါင်းဖို့ ကြိုးစားပါတယ်။ Python က ဂဏန်းနဲ့ string ကို မပေါင်းနိုင်လို့ `TypeError` ကို ပြပါတယ်။
- variable ကို `score` လို့ နာမည်ပေးထားတာက ပိုရှုပ်ထွေးစေပါတယ်။ တကယ်တော့ နာမည်ပါ။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** loop variable ကို key အလိုက် (`name`) နာမည်ပေးပြီး `scores[name]` နဲ့ တန်ဖိုးကို ယူပါ။

````python
# ✅ မှန်ပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:             # name က key
    total += scores[name]       # scores[name] က တန်ဖိုး

print("Total:", total)
````

### အမှား 8: မရှိတဲ့ key ကို ဖျက်ခြင်း

````python
# ❌ မှားပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# del scores["Kate"]        # KeyError: 'Kate'
````

🔍 **ဘာမှားသွားလဲ?**

- dictionary ထဲမှာ `"Kate"` ဆိုတဲ့ key မရှိလို့ ဖျက်စရာ ဘာမှ မရှိပါဘူး။
- အမှား 1 လိုပဲ Python က `KeyError` ကို ပြပါတယ်။

✅ **မှန်ကန်တဲ့ နည်းလမ်း:** `in` နဲ့ အရင်စစ်ပြီး key ရှိမှသာ ဖျက်ပါ။

````python
# ✅ မှန်ပါတယ်
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
else:
    print(name, "is not in the dictionary, so nothing was deleted.")
````
