## 1. ทำไมเราต้องมี dictionary?

เริ่มจากสิ่งที่เรารู้แล้ว: **ลิสต์ (list)**

ลิสต์เก็บค่าได้หลายค่าตามลำดับ ไว้ในวงเล็บเหลี่ยม `[ ]` เราดึงค่าออกมาด้วย **อินเด็กซ์ (index)** (หมายเลขตำแหน่ง) และอินเด็กซ์เริ่มจาก **0**

ทีนี้ลองนึกภาพว่าเราอยากเก็บคะแนนสอบของนักเรียน 3 คน: Anna, Tom และ Bill ถ้าใช้ลิสต์ เราต้องมีลิสต์ **สอง** ลิสต์: ลิสต์หนึ่งเก็บชื่อ อีกลิสต์เก็บคะแนน

```python
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]
```

สองลิสต์นี้เชื่อมกันด้วยตำแหน่ง:

| อินเด็กซ์ | 0 | 1 | 2 |
|---|---|---|---|
| `names` | "Anna" | "Tom" | "Bill" |
| `scores` | 85 | 92 | 78 |

ดังนั้น `names[1]` คือ Tom และ `scores[1]` คือคะแนนของ Tom

**แต่แบบนี้มีปัญหา:**

- จะหาคะแนนของ Tom ได้ เราต้อง **จำ** ให้ได้ก่อนว่า Tom อยู่ที่อินเด็กซ์ 1 Python หาจากชื่อ "Tom" ไม่ได้
- ถ้าเพิ่มหรือลบชื่อในลิสต์หนึ่ง แต่ลืมทำแบบเดียวกันในอีกลิสต์ ชื่อกับคะแนนจะ **สลับกันมั่ว**

คงจะง่ายกว่านี้ถ้าเราถามได้เลยว่า **"Tom ได้คะแนนเท่าไร?"**

นั่นคือสิ่งที่ dictionary ทำได้

````python
# ถ้าใช้สองลิสต์ ชื่อกับคะแนนจะอยู่คนละที่
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

# จะดึงคะแนนของ Tom เราต้องรู้ว่า Tom อยู่ที่อินเด็กซ์ 1
print(names[1])         # Tom
print(scores[1])        # 92 -> คะแนนของ Tom
````

````python
# ปัญหา: ถ้าลิสต์หนึ่งเปลี่ยนแต่อีกลิสต์ไม่เปลี่ยน ทั้งสองจะไม่ตรงกัน
names = ["Anna", "Tom", "Bill"]
scores = [85, 92, 78]

names.remove("Anna")    # เราลบ Anna ออกจาก names แต่ลืมลบ 85 ออกจาก scores

print(names[0])         # Tom
print(scores[0])        # 85 -> นี่คือคะแนนของ Anna ไม่ใช่ของ Tom! ตอนนี้ไม่ตรงกันแล้ว
````

## 2. dictionary คืออะไร?

ลองนึกถึงพจนานุกรมจริง ๆ หนังสือที่ใช้เปิดหาคำศัพท์

- เราเปิดหา **คำ** (เช่น "apple")
- แล้วเจอ **ความหมาย** ของมัน ("ผลไม้ลูกกลม")

เราไม่ต้องรู้ว่าคำนั้นอยู่หน้าไหน แค่หาคำ ความหมายก็อยู่ข้าง ๆ เลย

**dictionary** ใน Python ทำงานแบบเดียวกัน มันเก็บค่าเป็น **คู่**:

| คำใน Python | ความหมาย | ในพจนานุกรมจริง | ในตัวอย่างของเรา |
|---|---|---|---|
| **คีย์ (key)** | ชื่อที่เราใช้หา | คำศัพท์ | ชื่อนักเรียน ("Tom") |
| **ค่า (value)** | สิ่งที่เราได้ | ความหมาย | คะแนน (92) |
| **คู่ (pair)** | คีย์หนึ่งตัวกับค่าของมัน | คำหนึ่งคำกับความหมายของมัน | "Tom" กับ 92 |

หรือจะนึกภาพเป็น **ตู้ล็อกเกอร์ที่มีป้ายชื่อ** ก็ได้:
- ใน **ลิสต์** ตู้แต่ละช่องมี **หมายเลข** (0, 1, 2 ...)
- ใน **dictionary** ตู้แต่ละช่องมี **ป้ายชื่อ** ("Anna", "Tom", "Bill" ...)

เวลาจะเปิดตู้ เราใช้ป้ายชื่อแทนหมายเลข

## 3. การสร้าง dictionary

เริ่มจาก dictionary ที่เล็กที่สุด: มีแค่ **คู่เดียว**

```python
scores = {"Tom": 92}
```

มาดูทีละส่วน:

| ส่วน | ความหมาย |
|---|---|
| `{` | dictionary เริ่มต้น (เป็น **วงเล็บปีกกา** ไม่ใช่วงเล็บเหลี่ยม) |
| `"Tom"` | **คีย์** (ตรงนี้เป็นสตริง จึงอยู่ในเครื่องหมายคำพูด) |
| `:` | **โคลอน** เชื่อมคีย์กับค่าของมัน |
| `92` | **ค่า** |
| `}` | dictionary สิ้นสุด |

จำไว้ว่า: `=` หมายถึง **"เก็บค่าทางขวาไว้ในกล่องทางซ้าย"** บรรทัดนี้จึงเก็บ dictionary ทั้งหมดไว้ในตัวแปรชื่อ `scores`

````python
# dictionary ที่มีคู่เดียว
scores = {"Tom": 92}

print(scores)           # {'Tom': 92}
````

### มากกว่าหนึ่งคู่

ถ้าจะใส่หลายคู่ใน dictionary ให้คั่นแต่ละคู่ด้วย **จุลภาค** `,` เหมือนในลิสต์ที่ค่าต่าง ๆ คั่นด้วยจุลภาค

```python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
```

- `"Anna": 85` คือหนึ่งคู่
- `"Tom": 92` คือหนึ่งคู่
- `"Bill": 78` คือหนึ่งคู่

dictionary นี้จึงมี **3 คู่**

````python
# dictionary ที่มีสองคู่
scores = {"Anna": 85, "Tom": 92}
print(scores)
````

````python
# dictionary ที่มีสามคู่
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)
````

### เขียนบรรทัดละหนึ่งคู่

ถ้ามีหลายคู่ การเขียนบรรทัดยาวบรรทัดเดียวจะอ่านยาก เราเขียน **บรรทัดละหนึ่งคู่** ได้ Python อ่านเหมือนเดิมทุกประการ อย่าลืมจุลภาคท้ายแต่ละคู่

````python
# dictionary เดิม เขียนบรรทัดละหนึ่งคู่ (อ่านง่ายกว่า)
scores = {
    "Anna": 85,
    "Tom": 92,
    "Bill": 78,
}
print(scores)
````

### เปรียบเทียบ: ลิสต์กับ dictionary

| | ลิสต์ | dictionary |
|---|---|---|
| วงเล็บ | วงเล็บเหลี่ยม `[ ]` | วงเล็บปีกกา `{ }` |
| ข้างในมีอะไร | ค่าอย่างเดียว | **คู่** ของ `คีย์: ค่า` |
| ตัวอย่าง | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92, "Bill": 78}` |

````python
# คะแนนชุดเดียวกัน ในรูปลิสต์และในรูป dictionary
score_list = [85, 92, 78]
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list)       # มีแค่คะแนน เราไม่รู้ว่าเป็นของใคร
print(score_dict)       # คะแนนแต่ละค่ามีชื่ออยู่ข้าง ๆ
````

````python
# dictionary เป็นชนิดข้อมูลใหม่: dict (ย่อมาจาก "dictionary")
# type() บอกว่าค่านั้นเป็นชนิดอะไร
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(type(scores))     # <class 'dict'>
````

### dictionary แบบอื่น ๆ

ค่าเป็นข้อมูลชนิดไหนก็ได้: สตริง จำนวนเต็ม ทศนิยม บูลีน ส่วนคีย์มักเป็น **สตริง**

````python
# ค่าเป็นสตริง: ประเทศ -> เมืองหลวง
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}
print(capitals)
````

````python
# ค่าเป็นทศนิยม: คน -> ส่วนสูง
heights = {"Anna": 162.5, "Tom": 175.0}
print(heights)
````

## 4. การดึงค่าด้วยคีย์ (key)

จะดึงค่าออกจาก dictionary ให้เขียนชื่อ dictionary แล้วตามด้วย **คีย์** ในวงเล็บเหลี่ยม `[ ]`:

```python
scores["Tom"]      # "ค่าของคีย์ 'Tom' ใน scores"
```

หน้าตาเหมือนลิสต์มาก ต่างกันแค่สิ่งที่อยู่ในวงเล็บเหลี่ยม:

| | ลิสต์ | dictionary |
|---|---|---|
| เขียนอะไรใน `[ ]` | **อินเด็กซ์** (หมายเลขตำแหน่ง) | **คีย์** (ชื่อ) |
| ตัวอย่าง | `score_list[1]` | `score_dict["Tom"]` |
| ความหมาย | "ค่าที่ตำแหน่ง 1" | "ค่าที่มีคีย์เป็น 'Tom'" |

**สำคัญ:** แม้เราจะใช้วงเล็บปีกกา `{ }` ตอน **สร้าง** dictionary แต่ตอน **ดึงค่า** เราใช้วงเล็บเหลี่ยม `[ ]`

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # ค่าของคีย์ "Tom" -> 92
print(scores["Anna"])       # ค่าของคีย์ "Anna" -> 85
print(scores["Bill"])       # ค่าของคีย์ "Bill" -> 78
````

````python
# เปรียบเทียบ: ดึงคะแนนของ Tom จากลิสต์และจาก dictionary
score_list = [85, 92, 78]                           # Anna, Tom, Bill
score_dict = {"Anna": 85, "Tom": 92, "Bill": 78}

print(score_list[1])        # ลิสต์: ต้องจำว่า Tom อยู่ที่อินเด็กซ์ 1
print(score_dict["Tom"])    # dictionary: แค่ใช้ชื่อ "Tom"
````

````python
# ตัวอย่างเพิ่มเติม
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

print(capitals["Korea"])    # Seoul
print(capitals["France"])   # Paris
````

````python
english = {"apple": "a round fruit", "dog": "an animal that barks"}

print(english["dog"])       # เปิดหาคำว่า "dog" -> ความหมายของมัน
````

### ค่าที่ได้ใช้งานได้เหมือนตัวแปรธรรมดา

ค่าที่ดึงออกมาจาก dictionary ก็เป็นแค่ค่าธรรมดา เก็บไว้ในตัวแปร นำไปคำนวณ หรือเปรียบเทียบได้

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

tom_score = scores["Tom"]                   # เก็บคะแนนของ Tom ไว้ในตัวแปรใหม่
print(tom_score)

print(scores["Anna"] + scores["Tom"])       # 85 + 92 = 177 (คำนวณได้)
print(scores["Tom"] > scores["Bill"])       # 92 > 78 เป็น True (เปรียบเทียบได้)
````

````python
# ใช้ค่าใน if (if จะรันโค้ดที่ย่อหน้าเฉพาะเมื่อเงื่อนไขเป็น True)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

if scores["Bill"] >= 80:
    print("Bill passed with a high score!")
else:
    print("Bill needs to study more.")
````

### คีย์เป็นตัวแปรก็ได้

แทนที่จะเขียนคีย์ลงไปตรง ๆ เราเก็บคีย์ไว้ในตัวแปรก่อนก็ได้ แล้วใส่ตัวแปรใน `[ ]` **โดยไม่มีเครื่องหมายคำพูด**

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Anna"               # ตัวแปร name เก็บสตริง "Anna"
print(scores[name])         # scores[name] เหมือนกับ scores["Anna"] -> 85

name = "Bill"               # ตอนนี้ตัวแปรเก็บ "Bill"
print(scores[name])         # scores["Bill"] -> 78
````

**เรื่องการแสดงหลายค่า:** ในเซลล์ถัดไป เราเขียน `print(name, scores[name])` เมื่อใส่หลายค่าใน `print()` โดยคั่นด้วย **จุลภาค** `,` ค่าเหล่านั้นจะแสดงในบรรทัดเดียว และ Python จะเติม **ช่องว่างหนึ่งช่อง** ระหว่างค่าให้อัตโนมัติ

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

name = "Tom"
print(name, scores[name])   # Tom 92 -> ชื่อกับคะแนนในบรรทัดเดียว
````

## 5. การเปลี่ยนค่าและการเพิ่มคู่ใหม่

จะเปลี่ยนค่าหรือเพิ่มคู่ใหม่ เราใช้รูปแบบเดียวกัน:

```python
ชื่อ_dictionary[คีย์] = ค่า
```

Python จะทำอย่างใดอย่างหนึ่ง ขึ้นอยู่กับว่าคีย์นั้น **มีอยู่แล้ว** ใน dictionary หรือไม่:

| คีย์มีอยู่แล้วใน dictionary ไหม? | เกิดอะไรขึ้น |
|---|---|
| **มี** | ค่าเดิมจะถูก **แทนที่** ด้วยค่าใหม่ |
| **ไม่มี** | จะมีการ **เพิ่มคู่ใหม่** |

ตรงนี้ต่างจากลิสต์ ในลิสต์เราเปลี่ยนค่าได้เฉพาะอินเด็กซ์ที่มีอยู่แล้ว ถ้าจะเพิ่มต้องใช้ `append()` แต่ใน dictionary ใช้ `[คีย์] = ค่า` แบบเดียวกันได้ทั้งเปลี่ยนและเพิ่ม

### การเปลี่ยนค่า (คีย์มีอยู่แล้ว)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Bill"] = 88         # "Bill" เป็นคีย์อยู่แล้ว -> ค่า 78 ถูกแทนที่ด้วย 88
print(scores)               # เปลี่ยนแค่คะแนนของ Bill
````

````python
# เปลี่ยนค่าโดยใช้ค่าเดิม (แนวคิดเดียวกับ age = age + 1)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Anna"] = scores["Anna"] + 5     # ทางขวาก่อน: 85 + 5 = 90 แล้วเก็บกลับเข้าไป
print(scores["Anna"])                   # 90
````

````python
# แบบสั้น += ก็ใช้ได้ (เหมือนกับ: scores["Tom"] = scores["Tom"] + 3)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

scores["Tom"] += 3
print(scores["Tom"])                    # 95
````

### การเพิ่มคู่ใหม่ (คีย์ยังไม่มี)

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(scores)

scores["Kate"] = 95         # "Kate" ยัง "ไม่" เป็นคีย์ -> มีการเพิ่มคู่ใหม่
print(scores)               # ตอนนี้มี 4 คู่
````

````python
# เริ่มจาก dictionary ว่าง แล้วเพิ่มทีละคู่
capitals = {}                   # เริ่มจากว่าง
print(capitals)

capitals["Korea"] = "Seoul"     # เพิ่มหนึ่งคู่
print(capitals)

capitals["Japan"] = "Tokyo"     # เพิ่มอีกหนึ่งคู่
print(capitals)

capitals["France"] = "Paris"    # และอีกหนึ่งคู่
print(capitals)
````

````python
# เปรียบเทียบกับลิสต์: ลิสต์ต้องใช้ append() เพื่อเพิ่ม แต่ dictionary ใช้แค่ [คีย์] = ค่า
fruit_list = ["apple", "banana"]
fruit_list.append("cherry")         # ลิสต์: append() เพิ่มไว้ท้าย
print(fruit_list)

prices = {"apple": 1000, "banana": 500}
prices["cherry"] = 3000             # dictionary: [คีย์ใหม่] = ค่า
print(prices)
````

## 6. len() และ in

### len(): มีกี่คู่?

จำ `len()` ได้ไหม: มันบอกว่าลิสต์มีกี่ค่า กับ dictionary `len()` จะบอกว่ามีกี่ **คู่**

หนึ่งคู่ (`"Tom": 92`) นับเป็น **1** ไม่ใช่ 2

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
print(len(scores))          # 3 คู่ -> 3

one_pair = {"Tom": 92}
print(len(one_pair))        # 1 คู่ -> 1 (ไม่ใช่ 2!)

empty_dict = {}
print(len(empty_dict))      # 0 คู่ -> 0
````

````python
# len() เปลี่ยนเมื่อเราเพิ่มคู่
scores = {"Anna": 85, "Tom": 92}
print(len(scores))          # 2

scores["Bill"] = 78         # เพิ่มคู่ใหม่
print(len(scores))          # 3
````

### in: คีย์นี้อยู่ใน dictionary ไหม?

จำ `in` ได้ไหม: `ค่า in ลิสต์` ตรวจว่าค่านั้นอยู่ในลิสต์หรือไม่ แล้วให้ผลเป็น `True` หรือ `False`

กับ dictionary `in` จะตรวจ **คีย์**

```python
"Tom" in scores      # มีคีย์ "Tom" ไหม?
```

**ระวัง:** `in` ตรวจ **เฉพาะคีย์** ไม่ได้ตรวจค่า แม้ 92 จะอยู่ใน dictionary ในฐานะค่า `92 in scores` ก็เป็น `False`

ถ้าจะตรวจว่าคีย์ **ไม่** อยู่ใน dictionary ให้ใช้ `not in`

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" เป็นคีย์ไหม? -> True
print("Kate" in scores)     # "Kate" เป็นคีย์ไหม? -> False
print("Kate" not in scores) # "Kate" "ไม่" เป็นคีย์ใช่ไหม? -> True
````

````python
# in ตรวจเฉพาะ "คีย์" ไม่ได้ตรวจค่า
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print("Tom" in scores)      # "Tom" เป็นคีย์ -> True
print(92 in scores)         # 92 เป็นค่า ไม่ใช่คีย์ -> False
````

````python
# ใช้ in กับ if: ตรวจก่อน แล้วจึงดึงค่า
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    print(name, scores[name])
else:
    print(name, "is not in the list.")
````

````python
# โปรแกรมพจนานุกรมง่าย ๆ
english = {"apple": "a round fruit", "dog": "an animal that barks"}
word = "cat"

if word in english:
    print(english[word])
else:
    print("Sorry, I don't know that word.")
````

## 7. การลบคู่

del ชื่อ_dictionary[คีย์]

## 8. การใช้ลูป for กับ dictionary

มาทบทวนว่าลูป for ทำงานกับลิสต์อย่างไร:

```python
for score in score_list:
    print(score)
```

- แต่ละรอบ ค่าหนึ่งค่าจากลิสต์จะถูกเก็บไว้ในตัวแปร `score`
- บรรทัด `for` ลงท้ายด้วยโคลอน `:`
- โค้ดที่ต้องทำซ้ำต้อง **ย่อหน้า** (4 ช่องว่าง) Python ใช้การย่อหน้าเพื่อรู้ว่าบรรทัดไหนเป็นของลูป
- บรรทัดแรกที่ไม่ย่อหน้าอยู่นอกลูป และทำงานแค่ครั้งเดียวหลังลูป

กับ dictionary ลูป for จะให้ **คีย์** มาทีละตัว

```python
for name in scores:
    print(name)          # Anna, Tom, Bill
```

และเมื่อมีคีย์แล้ว ก็ดึงค่าได้ด้วย `scores[name]` เหมือนในหัวข้อที่ 4

````python
# ลูป for ให้ "คีย์" ของ dictionary
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:         # name เป็น "Anna" แล้วเป็น "Tom" แล้วเป็น "Bill"
    print(name)
````

````python
# ใช้คีย์ดึงค่า
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(scores[name])     # scores["Anna"] แล้ว scores["Tom"] แล้ว scores["Bill"]
````

````python
# คีย์กับค่าในบรรทัดเดียวกัน (print ด้วยจุลภาค)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])
````

### ชื่อตัวแปรในลูป

ตัวแปรหลัง `for` (ตรงนี้คือ `name`) เป็นตัวแปรใหม่ จะตั้งชื่ออะไรก็ได้ เลือกชื่อที่บอกว่า **คีย์** คืออะไร: ชื่อคน → `name`, ประเทศ → `country`, คำ → `word`

````python
capitals = {"Korea": "Seoul", "Japan": "Tokyo", "France": "Paris"}

for country in capitals:            # คีย์เป็นประเทศ เราจึงตั้งชื่อว่า country
    print(country, capitals[country])
````

### การย่อหน้าเป็นตัวกำหนดว่าอะไรถูกทำซ้ำ

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    print(name, scores[name])       # ย่อหน้า -> ทำซ้ำกับทุกคีย์
print("That's everyone!")           # ไม่ย่อหน้า -> อยู่นอกลูป -> ทำงานครั้งเดียว
````

### การบวกค่าต่าง ๆ รวมกัน

จำ **ผลรวมสะสม** ได้ไหม:

1. **ก่อนลูป:** `total = 0`
2. **ในลูป:** บวกค่าปัจจุบันเข้าไป → `total = total + scores[name]`
3. **หลังลูป:** `total` เก็บผลรวมไว้

`total = total + scores[name]` หมายถึง: คำนวณทางขวาก่อน (ผลรวมเดิม + คะแนนของนักเรียนคนนี้) แล้วเก็บผลลัพธ์กลับไปใน `total` แบบสั้นคือ `total += scores[name]`

````python
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0                           # ก่อนลูป: เริ่มจาก 0

for name in scores:
    total = total + scores[name]    # ผลรวมเดิม + คะแนนของนักเรียนคนนี้

print("Total:", total)              # หลังลูป (ไม่ย่อหน้า): 85 + 92 + 78 = 255
````

````python
# ค่าเฉลี่ย = ผลรวม / จำนวนค่า
# len(scores) คือจำนวนคู่ (= จำนวนนักเรียน)
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:
    total += scores[name]           # เหมือนกับ: total = total + scores[name]

average = total / len(scores)       # หลังลูป: 255 / 3
print("Average:", average)
````

### if ในลูป

เมื่อใส่ `if` ไว้ใน `for` จะมีการย่อหน้าสองระดับ: 4 ช่องว่างสำหรับ `if` (อยู่ใน `for`) และ 8 ช่องว่างสำหรับโค้ดที่อยู่ใน `if`

````python
# แสดงเฉพาะนักเรียนที่สอบผ่าน (60 ขึ้นไป)
scores = {"Anna": 85, "Tom": 42, "Bill": 78, "Kate": 55}

for name in scores:
    if scores[name] >= 60:          # 4 ช่องว่าง: ตรวจคะแนนของนักเรียนแต่ละคน
        print(name, "passed")       # 8 ช่องว่าง: เฉพาะเมื่อเงื่อนไขเป็น True
````

````python
# วันนี้ใครมาเรียนบ้าง?
attendance = {"Anna": True, "Tom": False, "Bill": True}

for name in attendance:
    if attendance[name]:            # ค่าเป็น True หรือ False อยู่แล้ว จึงใช้เป็นเงื่อนไขได้ทันที
        print(name, "is here.")
    else:
        print(name, "is absent.")
````

## 9. ลิสต์กับ dictionary

| | ลิสต์ | dictionary |
|---|---|---|
| สร้าง | `[85, 92, 78]` | `{"Anna": 85, "Tom": 92}` |
| วงเล็บที่ใช้สร้าง | `[ ]` | `{ }` |
| หาค่าด้วย | **อินเด็กซ์** (หมายเลขตำแหน่ง) | **คีย์** (ชื่อ) |
| ดึงค่า | `score_list[1]` | `score_dict["Tom"]` |
| เปลี่ยนค่า | `score_list[1] = 95` | `score_dict["Tom"] = 95` |
| เพิ่ม | `score_list.append(88)` | `score_dict["Kate"] = 88` |
| ลบ | `score_list.remove(92)` | `del score_dict["Tom"]` |
| `len()` นับ | ค่า | คู่ |
| `in` ตรวจ | ค่า | **คีย์** |
| `for x in ...` ให้ | ค่า | **คีย์** |

**ใช้อันไหนเมื่อไร?**
- ถ้า **ลำดับ** สำคัญ หรือค่าไม่จำเป็นต้องมีชื่อ → **ลิสต์** (เช่น คะแนนสอบวันนี้ตามลำดับ)
- ถ้าอยากหาค่า **จากชื่อ** ของมัน → **dictionary** (เช่น คะแนนของนักเรียนแต่ละคน คำกับความหมาย)

## 10. ข้อผิดพลาดที่พบบ่อย

นี่คือข้อผิดพลาดที่เกือบทุกคนเคยทำตอนเรียน dictionary ในแต่ละข้อเราจะดู:

- ❌ **โค้ดที่ผิด**
- 🔍 **ผิดตรงไหน**
- ✅ **วิธีที่ถูกต้อง**

โค้ดที่ทำให้เกิด error ถูกเขียนไว้เป็นคอมเมนต์ (มี `#` ข้างหน้า) เพื่อไม่ให้ notebook หยุด ถ้าอยากเห็นข้อความ error จริง ให้ลบ `#` รันเซลล์ แล้วค่อยใส่ `#` กลับ

### ข้อผิดพลาด 1: ใช้คีย์ที่ไม่มีอยู่

````python
# ❌ ผิด
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores["Kate"])     # KeyError: 'Kate'
# print(scores["tom"])      # KeyError: 'tom'
````

🔍 **ผิดตรงไหน?**

- `"Kate"` ไม่ใช่คีย์ใน dictionary Python หาไม่เจอ จึงแสดง `KeyError` ("ไม่มีคีย์นี้")
- `"tom"` (t ตัวเล็ก) กับ `"Tom"` (T ตัวใหญ่) เป็นสตริง **คนละตัว** ต้องเขียนคีย์ให้ **ตรงทุกตัวอักษร** รวมถึงตัวพิมพ์ใหญ่และตัวพิมพ์เล็ก

✅ **วิธีที่ถูกต้อง:** เขียนคีย์ให้ตรงกับใน dictionary ถ้าไม่แน่ใจว่ามีคีย์นั้นไหม ให้ตรวจด้วย `in` ก่อน

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Tom"])        # ตรงกับคีย์ทุกตัวอักษร

name = "Kate"
if name in scores:          # ตรวจก่อน
    print(scores[name])
else:
    print(name, "is not in the dictionary.")
````

### ข้อผิดพลาด 2: ใช้วงเล็บเหลี่ยมสร้าง dictionary

````python
# ❌ ผิด
# scores = ["Anna": 85, "Tom": 92]      # SyntaxError
````

🔍 **ผิดตรงไหน?**

- วงเล็บเหลี่ยม `[ ]` ใช้สร้าง **ลิสต์** ลิสต์เก็บแค่ค่า Python จึงไม่เข้าใจ `"Anna": 85` ที่อยู่ใน `[ ]`
- error คือ `SyntaxError` ("ไวยากรณ์ของโค้ดผิด")

✅ **วิธีที่ถูกต้อง:** ใช้ **วงเล็บปีกกา** `{ }` สร้าง dictionary (ตอน **ดึง** ค่ายังใช้วงเล็บเหลี่ยมเหมือนเดิม: `scores["Anna"]`)

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom": 92}    # สร้าง: วงเล็บปีกกา
print(scores["Anna"])               # ดึง: วงเล็บเหลี่ยม
````

### ข้อผิดพลาด 3: ลืมเครื่องหมายคำพูดรอบคีย์

````python
# ❌ ผิด
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[Anna])       # NameError: name 'Anna' is not defined
````

🔍 **ผิดตรงไหน?**

- คีย์คือ **สตริง** `"Anna"` ที่มีเครื่องหมายคำพูด
- ถ้าไม่มีเครื่องหมายคำพูด Python จะคิดว่า `Anna` เป็นชื่อ **ตัวแปร** แต่ไม่มีตัวแปรชื่อ `Anna`
- error คือ `NameError` ("ไม่มีตัวแปรชื่อนี้")

✅ **วิธีที่ถูกต้อง:** ใส่เครื่องหมายคำพูดรอบคีย์ที่เป็นสตริง (ไม่ต้องใส่เครื่องหมายคำพูดเฉพาะเมื่อคีย์ถูกเก็บไว้ในตัวแปร เช่น `scores[name]` ในหัวข้อที่ 4)

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # คีย์สตริงที่มีเครื่องหมายคำพูด

name = "Anna"
print(scores[name])         # ตัวแปรที่ไม่มีเครื่องหมายคำพูด (ตัวแปรเก็บ "Anna" ไว้)
````

### ข้อผิดพลาด 4: ใช้หมายเลขอินเด็กซ์เหมือนลิสต์

````python
# ❌ ผิด: เราอยากได้คะแนนของนักเรียนคนแรก
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# print(scores[0])          # KeyError: 0
````

🔍 **ผิดตรงไหน?**

- ในลิสต์ `[0]` หมายถึง "ค่าแรก" แต่ dictionary ไม่ได้ใช้หมายเลขอินเด็กซ์ มันใช้ **คีย์**
- Python จึงไปหา **คีย์** ที่ชื่อ `0` ซึ่งไม่มี จึงแสดง `KeyError: 0`

✅ **วิธีที่ถูกต้อง:** ใช้คีย์ (ชื่อ)

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(scores["Anna"])       # ใช้คีย์ ไม่ใช่ตัวเลข
````

### ข้อผิดพลาด 5: เขียนคีย์เดียวกันสองครั้ง

````python
# ❌ ผิด: นักเรียนสองคนที่ชื่อ Tom เหมือนกัน?
scores = {"Anna": 85, "Tom": 92, "Tom": 60}

print(scores)               # {'Anna': 85, 'Tom': 60} -> Tom คนแรก (92) หายไป!
print(len(scores))          # 2 ไม่ใช่ 3
````

🔍 **ผิดตรงไหน?**

- ใน dictionary คีย์แต่ละตัวมีได้ **แค่ครั้งเดียว**
- ถ้าเขียนคีย์เดิมซ้ำ จะทำงานเหมือนหัวข้อที่ 5: คีย์มีอยู่แล้ว ค่าเดิมจึงถูก **แทนที่** 92 ถูกแทนด้วย 60
- ไม่มี error จึงมองข้ามได้ง่าย

✅ **วิธีที่ถูกต้อง:** ทำให้ทุกคีย์ไม่ซ้ำกัน ถ้ามีสองคนชื่อเดียวกัน ให้ใช้คีย์ที่ต่างกัน

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom K.": 92, "Tom L.": 60}

print(scores)
print(len(scores))          # 3
````

### ข้อผิดพลาด 6: คิดว่า in ตรวจค่า

````python
# ❌ ผิด: เราอยากรู้ว่ามีใครได้ 92 คะแนนไหม
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

print(92 in scores)         # False -> แต่ Tom ได้ 92!
````

🔍 **ผิดตรงไหน?**

- กับ dictionary `in` ตรวจ **เฉพาะคีย์** (ชื่อ) ไม่ได้ตรวจค่า (คะแนน)
- 92 เป็นค่า ไม่ใช่คีย์ ผลลัพธ์จึงเป็น `False`
- ไม่มี error แต่คำตอบผิด

✅ **วิธีที่ถูกต้อง:** ไล่ดู dictionary ด้วยลูป for แล้วเปรียบเทียบทีละค่า

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

for name in scores:
    if scores[name] == 92:      # เปรียบเทียบ "ค่า" แต่ละตัวกับ 92
        print(name, "got 92 points.")
````

### ข้อผิดพลาด 7: คิดว่าลูป for ให้ค่า

````python
# ❌ ผิด: เราอยากบวกคะแนนทั้งหมด
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

# for score in scores:
#     total += score        # TypeError: unsupported operand type(s) for +=: 'int' and 'str'
````

🔍 **ผิดตรงไหน?**

- กับ dictionary ลูป for ให้ **คีย์** ไม่ใช่ค่า `score` จึงเป็น `"Anna"`, `"Tom"`, `"Bill"` ไม่ใช่ 85, 92, 78
- `total += score` พยายามบวกสตริง (`"Anna"`) เข้ากับตัวเลข (`0`) Python บวกตัวเลขกับสตริงไม่ได้ จึงแสดง `TypeError`
- การตั้งชื่อตัวแปรว่า `score` ยิ่งทำให้สับสน ที่จริงมันคือชื่อคน

✅ **วิธีที่ถูกต้อง:** ตั้งชื่อตัวแปรในลูปตามคีย์ (`name`) แล้วใช้ `scores[name]` ดึงค่า

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
total = 0

for name in scores:             # name คือคีย์
    total += scores[name]       # scores[name] คือค่า

print("Total:", total)
````

### ข้อผิดพลาด 8: ลบคีย์ที่ไม่มีอยู่

````python
# ❌ ผิด
scores = {"Anna": 85, "Tom": 92, "Bill": 78}

# del scores["Kate"]        # KeyError: 'Kate'
````

🔍 **ผิดตรงไหน?**

- ใน dictionary ไม่มีคีย์ `"Kate"` จึงไม่มีอะไรให้ลบ
- เหมือนข้อผิดพลาด 1 Python จะแสดง `KeyError`

✅ **วิธีที่ถูกต้อง:** ตรวจด้วย `in` ก่อน แล้วลบเฉพาะเมื่อมีคีย์นั้นอยู่

````python
# ✅ ถูกต้อง
scores = {"Anna": 85, "Tom": 92, "Bill": 78}
name = "Kate"

if name in scores:
    del scores[name]
else:
    print(name, "is not in the dictionary, so nothing was deleted.")
````
