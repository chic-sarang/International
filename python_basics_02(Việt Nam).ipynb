## 1. Câu lệnh if là gì?

Cho đến bây giờ, mã của chúng ta luôn chạy từ trên xuống dưới, lần lượt từng dòng, lần nào cũng vậy. Nhưng trong cuộc sống, chúng ta luôn phải đưa ra lựa chọn.

- **Nếu** trời mưa, hãy mang ô.
- **Nếu** bạn từ 20 tuổi trở lên, bạn có thể mua vé người lớn.

Câu lệnh `if` giúp chương trình đưa ra những lựa chọn như vậy. Nó kiểm tra một **điều kiện**, và chạy đoạn mã bên trong **chỉ khi điều kiện là True**.

Bạn còn nhớ các toán tử so sánh ở Bài 1 (`>`, `<`, `==`, `!=`) không? Kết quả của chúng luôn là `True` hoặc `False`. Đó chính là thứ mà `if` cần để làm điều kiện.

Dạng cơ bản như sau:

```python
if điều_kiện:
    mã chỉ chạy khi điều kiện là True
```

Hai quy tắc rất quan trọng:
- Cuối dòng `if` phải có dấu hai chấm `:`.
- Mã thuộc về `if` phải được **thụt lề** (đẩy sang phải 4 dấu cách). Thụt lề là cách Python biết những dòng nào nằm "bên trong" `if`.

````python
# Điều kiện là age >= 20 ("tuổi lớn hơn hoặc bằng 20")
# Kết quả của phép so sánh này là True hoặc False
age = 25

if age >= 20:                         # 25 >= 20 là True, nên đoạn mã thụt lề bên dưới sẽ chạy
    print("You are an adult.")        # Dòng này nằm trong if (được thụt lề 4 dấu cách)

print("This line always runs.")      # Dòng này KHÔNG thụt lề, nên nó nằm ngoài if. Nó luôn chạy dù thế nào đi nữa
````

````python
# Bây giờ hãy thử trường hợp điều kiện là False
age = 15

if age >= 20:                         # 15 >= 20 là False, nên đoạn mã thụt lề bên dưới bị bỏ qua
    print("You are an adult.")        # Dòng này hoàn toàn không chạy

print("This line always runs.")      # Dòng này nằm ngoài if, nên vẫn chạy
````

````python
# Bên trong if có thể có nhiều hơn một dòng
# Mọi dòng có cùng mức thụt lề đều thuộc về cùng một if
temperature = 32

if temperature > 30:                          # 32 > 30 là True
    print("It's very hot today.")             # Bên trong if
    print("Drink plenty of water.")           # Cũng bên trong if (cùng mức thụt lề)
    print("Try to stay in the shade.")        # Cũng bên trong if (cùng mức thụt lề)

print("Have a nice day!")                    # Bên ngoài if, luôn chạy
````

````python
# Bạn cũng có thể kiểm tra hai giá trị có giống nhau không bằng ==
# Hãy nhớ: = nghĩa là "lưu một giá trị", còn == nghĩa là "hai giá trị này có bằng nhau không?"
password = "python123"

if password == "python123":           # Hai chuỗi giống hệt nhau, nên đây là True
    print("Login successful!")
````

## 2. if - else

Chỉ với `if`, chương trình sẽ làm gì đó khi điều kiện là True, và **không làm gì cả** khi điều kiện là False.

Nhưng thường thì chúng ta muốn làm **việc này hoặc việc kia**. Ví dụ:
- Nếu bạn đậu, in ra "Chúc mừng". **Nếu không**, in ra "Hãy thử lại".

Đó là lúc dùng `else`. `else` có nghĩa là "nếu không thì". Mã của nó chạy **chỉ khi điều kiện của if là False**.

```python
if điều_kiện:
    mã chạy khi điều kiện là True
else:
    mã chạy khi điều kiện là False
```

- `else` không có điều kiện riêng. Nó đơn giản là xử lý mọi trường hợp mà `if` không xử lý.
- `else` cũng cần dấu hai chấm `:` ở cuối, và mã của nó cũng phải được thụt lề.
- Luôn luôn có đúng **một** trong hai khối được chạy. Không bao giờ cả hai, cũng không bao giờ không có khối nào.

````python
# Kiểm tra điểm thi là đậu hay rớt
# Giả sử điểm đậu là 60
score = 75

if score >= 60:                      # 75 >= 60 là True
    print("Pass! Congratulations.")  # Nên dòng này chạy
else:                                # "Nếu không thì" - phần này bị bỏ qua vì if là True
    print("Fail. Try again.")
````

````python
# Cùng đoạn mã, nhưng với điểm thấp hơn
score = 40

if score >= 60:                      # 40 >= 60 là False, nên phần này bị bỏ qua
    print("Pass! Congratulations.")
else:                                # if là False, nên phần else chạy thay thế
    print("Fail. Try again.")
````

````python
# Kiểm tra một số là chẵn hay lẻ
# Bạn còn nhớ % ở Bài 1 không? Nó cho ra số dư của phép chia
# Số chẵn chia cho 2 luôn có số dư là 0
number = 7

if number % 2 == 0:                  # 7 % 2 là 1, và 1 == 0 là False
    print(number, "is even.")        # print() có thể hiển thị nhiều giá trị cùng lúc nếu bạn ngăn cách chúng bằng dấu phẩy
else:
    print(number, "is odd.")         # if là False, nên dòng này chạy
````

## 3. if - elif - else

Đôi khi có **nhiều hơn hai** lựa chọn. Ví dụ, đổi điểm số thành xếp loại:

- Từ 90 trở lên → A
- Từ 80 trở lên → B
- Từ 70 trở lên → C
- Các trường hợp còn lại → F

Trong trường hợp này, chúng ta dùng `elif`. `elif` là viết tắt của "else if", có nghĩa là "nếu các điều kiện ở trên đều là False, thì kiểm tra điều kiện này".

```python
if điều_kiện_1:
    chạy khi điều_kiện_1 là True
elif điều_kiện_2:
    chạy khi điều_kiện_1 là False và điều_kiện_2 là True
elif điều_kiện_3:
    chạy khi điều_kiện_1 và điều_kiện_2 là False và điều_kiện_3 là True
else:
    chạy khi mọi điều kiện ở trên đều là False
```

Cách hoạt động:
- Python kiểm tra các điều kiện **từ trên xuống dưới, theo thứ tự**.
- Ngay khi tìm thấy điều kiện True **đầu tiên**, nó chạy khối đó và **bỏ qua tất cả phần còn lại**.
- Bạn có thể thêm bao nhiêu `elif` tùy thích. Phần `else` ở cuối có thể có hoặc không.

````python
# Đổi điểm số thành xếp loại
score = 85

if score >= 90:                # 85 >= 90 là False -> chuyển sang điều kiện tiếp theo
    print("Grade: A")
elif score >= 80:              # 85 >= 80 là True -> khối này chạy, và mọi thứ bên dưới bị bỏ qua
    print("Grade: B")
elif score >= 70:              # Hoàn toàn không được kiểm tra, vì đã tìm thấy điều kiện True ở trên
    print("Grade: C")
else:                          # Cũng không được kiểm tra
    print("Grade: F")
````

````python
# Thử vài điểm số khác nhau với cùng đoạn mã
# Hãy thay đổi giá trị của score rồi chạy lại để xem kết quả thay đổi thế nào
score = 72

if score >= 90:                # 72 >= 90 là False
    print("Grade: A")
elif score >= 80:              # 72 >= 80 là False
    print("Grade: B")
elif score >= 70:              # 72 >= 70 là True -> khối này chạy
    print("Grade: C")
else:
    print("Grade: F")
````

````python
# Tại sao thứ tự lại quan trọng?
# Python dừng lại ở điều kiện True ĐẦU TIÊN, nên nếu đặt điều kiện nhỏ nhất lên đầu, kết quả sẽ sai
score = 95

if score >= 70:                # 95 >= 70 là True -> khối này chạy trước và mọi thứ khác bị bỏ qua!
    print("Grade: C")
elif score >= 80:              # Không bao giờ được kiểm tra
    print("Grade: B")
elif score >= 90:              # Không bao giờ được kiểm tra, dù 95 xứng đáng được A
    print("Grade: A")

# Vì vậy khi kiểm tra các khoảng giá trị như thế này, hãy bắt đầu từ điều kiện lớn nhất (khắt khe nhất)
````

````python
# Chọn trang phục tùy theo nhiệt độ
temperature = 12

if temperature >= 25:                  # 12 >= 25 là False
    print("Wear a T-shirt.")
elif temperature >= 15:                # 12 >= 15 là False
    print("Wear a light jacket.")
elif temperature >= 5:                 # 12 >= 5 là True -> dòng này chạy
    print("Wear a coat.")
else:                                  # Chỉ chạy khi nhiệt độ dưới 5
    print("Wear a thick padded jacket.")
````

## 4. Dùng biến bool làm điều kiện

Điều kiện của `if` chỉ cần là `True` hoặc `False`. Điều đó có nghĩa là một **biến bool** (ở Bài 1) có thể được dùng trực tiếp làm điều kiện mà không cần so sánh.

```python
if is_student:        # cùng ý nghĩa với: if is_student == True:
```

Viết `== True` không sai, nhưng không cần thiết, nên người lập trình Python thường bỏ đi.

Để kiểm tra một biến bool là `False`, hãy dùng `not`:

```python
if not is_student:    # cùng ý nghĩa với: if is_student == False:
```

````python
# Biến bool đã chứa sẵn True hoặc False, nên có thể dùng trực tiếp làm điều kiện
is_student = True

if is_student:                        # is_student là True, nên dòng này chạy
    print("You get a student discount.")
else:
    print("Regular price.")
````

````python
# Dùng not để kiểm tra False
# not đảo ngược giá trị: not True là False, và not False là True
is_raining = False

if not is_raining:                    # is_raining là False, nên not is_raining là True -> dòng này chạy
    print("No umbrella needed.")
else:
    print("Take an umbrella.")
````

````python
# Bạn có thể lưu kết quả của phép so sánh vào một biến bool trước, rồi mới dùng nó
# Cách này hữu ích khi điều kiện dài, hoặc khi bạn muốn đặt cho nó một cái tên rõ ràng
age = 17
is_adult = age >= 20                  # 17 >= 20 là False, nên is_adult lưu giá trị False
print(is_adult)

if is_adult:                          # is_adult là False, nên phần này bị bỏ qua
    print("You can watch this movie.")
else:
    print("You can't watch this movie.")
````

## 5. Kết hợp điều kiện (and, or, not)

Đôi khi một điều kiện là không đủ. Ví dụ:
- Bạn được đi tàu lượn siêu tốc nếu bạn **từ 10 tuổi trở lên và cao từ 140 cm trở lên**.
- Bạn được vào cửa miễn phí nếu bạn **dưới 7 tuổi hoặc từ 65 tuổi trở lên**.

Python có ba từ để kết hợp các điều kiện.

| Toán tử | Ý nghĩa | Kết quả |
|---|---|---|
| `and` | và | True chỉ khi **cả hai** vế đều True |
| `or` | hoặc | True khi **ít nhất một** vế là True |
| `not` | ngược lại | Đổi True thành False và False thành True |

````python
# Trước tiên, hãy xem and / or tự cho ra kết quả gì
print(True and True)     # cả hai đều True -> True
print(True and False)    # một vế là False -> False
print(True or False)     # ít nhất một vế là True -> True
print(False or False)    # cả hai đều False -> False
print(not True)          # ngược lại của True -> False
````

````python
# and: mọi điều kiện đều phải là True
# Quy tắc: từ 10 tuổi trở lên VÀ cao từ 140 cm trở lên
age = 12
height = 135

if age >= 10 and height >= 140:       # 12 >= 10 là True, nhưng 135 >= 140 là False -> True and False -> False
    print("You can ride the roller coaster.")
else:
    print("Sorry, you can't ride it yet.")  # Điều kiện là False, nên dòng này chạy
````

````python
# or: ít nhất một điều kiện phải là True
# Quy tắc: dưới 7 tuổi HOẶC từ 65 tuổi trở lên được vào cửa miễn phí
age = 70

if age < 7 or age >= 65:              # 70 < 7 là False, nhưng 70 >= 65 là True -> False or True -> True
    print("Free entry!")              # Điều kiện là True, nên dòng này chạy
else:
    print("Please buy a ticket.")
````

````python
# Kiểm tra một số có nằm trong một khoảng hay không
# "score nằm giữa 0 và 100" có nghĩa là "score từ 0 trở lên VÀ score từ 100 trở xuống"
score = 105

if score >= 0 and score <= 100:       # 105 >= 0 là True, nhưng 105 <= 100 là False -> False
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")

# Python cũng cho phép viết điều tương tự ngắn gọn hơn, giống như trong toán học
if 0 <= score <= 100:                 # Ý nghĩa hoàn toàn giống dòng ở trên
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")
````

## 6. Câu lệnh if lồng nhau

Bạn có thể đặt một `if` **bên trong một** `if` khác. Điều này được gọi là if **lồng nhau**.

Cách này hữu ích khi bạn muốn kiểm tra điều kiện thứ hai **chỉ sau khi** điều kiện thứ nhất là True. Ví dụ, trước tiên kiểm tra người dùng đã đăng nhập chưa, và chỉ khi đó mới kiểm tra họ có phải là quản trị viên hay không.

`if` bên trong được thụt lề thêm một bậc (thêm 4 dấu cách) so với `if` bên ngoài.

````python
# Trước tiên kiểm tra đăng nhập, và chỉ khi đăng nhập thành công mới kiểm tra quyền quản trị
is_logged_in = True
is_admin = False

if is_logged_in:                               # if bên ngoài: True, nên đi vào bên trong
    print("Welcome!")                          # Thuộc về if bên ngoài (4 dấu cách)
    if is_admin:                               # if bên trong: is_admin là False
        print("You can open the admin page.")  # Thuộc về if bên trong (8 dấu cách) - bị bỏ qua
    else:
        print("You are a regular user.")       # Thuộc về else bên trong (8 dấu cách) - dòng này chạy
else:                                          # Thuộc về if bên ngoài
    print("Please log in first.")              # Bị bỏ qua, vì if bên ngoài là True
````

````python
# Mua hàng trong cửa hàng
# Bước 1: món hàng còn không?  Bước 2: bạn có đủ tiền không?
in_stock = True
money = 8000
price = 12000

if in_stock:                                   # Món hàng còn -> đi vào bên trong
    if money >= price:                         # 8000 >= 12000 là False
        print("Purchase complete!")
    else:
        print("You don't have enough money.")  # Dòng này chạy
        print("You need", price - money, "more won.")  # 12000 - 8000 = 4000
else:
    print("Sorry, this item is sold out.")
````

**Mẹo:** if lồng nhau rất hữu ích, nhưng lồng quá nhiều tầng sẽ khiến mã khó đọc. Khi có thể, hãy thử kết hợp các điều kiện bằng `and`.

```python
# lồng nhau
if is_logged_in:
    if is_admin:
        print("You can open the admin page.")

# cùng ý nghĩa, viết bằng and
if is_logged_in and is_admin:
    print("You can open the admin page.")
```

## 7. Những lỗi thường gặp

Đây là những lỗi mà hầu như ai cũng mắc phải khi học `if`. Nếu gặp lỗi, hãy kiểm tra những điểm này trước.

1. **Dùng `=` thay vì `==`**: `=` dùng để lưu giá trị, `==` dùng để so sánh giá trị. `if age = 20:` là lỗi.
2. **Quên dấu hai chấm `:`**: Các dòng `if`, `elif` và `else` đều phải kết thúc bằng `:`.
3. **Thụt lề sai**: Mã bên trong phải được thụt lề, và các dòng trong cùng một khối phải thụt lề bằng nhau.
4. **Cho `else` một điều kiện**: `else` không bao giờ có điều kiện. `else age < 20:` là lỗi. Nếu cần điều kiện, hãy dùng `elif`.
5. **So sánh số với chuỗi**: `"20"` (chuỗi) và `20` (số nguyên) là khác nhau. `"20" == 20` là `False`.

Các ô bên dưới cho thấy mỗi lỗi trông như thế nào. Chúng được viết dưới dạng chú thích để không làm hỏng notebook. Hãy thử xóa dấu `#` ở dòng mã để xem thông báo lỗi thật, sau đó đặt lại dấu `#`.

````python
# Lỗi 1: dùng = thay vì ==
age = 20

# if age = 20:          # SyntaxError! = nghĩa là "lưu", nên không thể dùng làm điều kiện
#     print("20 years old")

if age == 20:           # Đúng: == nghĩa là "có bằng nhau không?"
    print("20 years old")
````

````python
# Lỗi 2: quên dấu hai chấm
age = 20

# if age >= 20          # SyntaxError! Thiếu dấu hai chấm ở cuối
#     print("adult")

if age >= 20:           # Đúng: dòng kết thúc bằng dấu hai chấm
    print("adult")
````

````python
# Lỗi 3: thụt lề sai
age = 20

# if age >= 20:
# print("adult")        # IndentationError! Dòng bên trong if không được thụt lề

if age >= 20:
    print("adult")      # Đúng: thụt lề 4 dấu cách
````

````python
# Lỗi 4: cho else một điều kiện
age = 15

# if age >= 20:
#     print("adult")
# else age < 20:        # SyntaxError! else không thể có điều kiện
#     print("minor")

if age >= 20:
    print("adult")
else:                   # Đúng: else không có điều kiện. Nó xử lý mọi trường hợp mà if không xử lý
    print("minor")
````

````python
# Lỗi 5: so sánh số với chuỗi
# Không có lỗi nào hiện ra, nhưng kết quả không như bạn mong đợi, nên lỗi này rất khó phát hiện
age_text = "20"          # Đây là chuỗi, vì nó nằm trong dấu ngoặc kép (hãy nhớ zip_code ở Bài 1)
age_number = 20          # Đây là số nguyên

print(age_text == age_number)     # False! Chuỗi và số nguyên không bao giờ bằng nhau

if age_text == age_number:
    print("Same")
else:
    print("Different")            # Dòng này chạy, dù trông chúng giống nhau
````
