## 1. What Is a for Loop?

Imagine you have to print "Hello" 5 times. You could write `print("Hello")` 5 times. But what about 100 times? Or 1000 times? Writing the same line over and over is tiring, and it is easy to make mistakes.

A **loop** solves this problem. A loop runs the same code again and again, as many times as you want.

The most common loop in Python is the `for` loop. The basic shape looks like this:

```python
for i in range(5):
    code to repeat
```

You can read it like this: **"for each number i in range(5), run the code below"**.

- `range(5)` gives the numbers 0, 1, 2, 3, 4 (5 numbers in total). We will learn `range()` in detail in the next section.
- `i` is a variable. Each time the loop repeats, the next number is stored in `i`.
- Just like `if`, the `for` line must end with a colon `:`.
- Just like `if`, the code to repeat must be **indented** (4 spaces). Only the indented lines are repeated.

````python
# Repeat print("Hello") 5 times
# range(5) gives 5 numbers (0, 1, 2, 3, 4), so the indented code runs 5 times
for i in range(5):
    print("Hello")      # This line is inside the for loop (indented), so it is repeated
````

````python
# Let's see what is stored in i each time
# 1st time: i = 0, 2nd time: i = 1, 3rd time: i = 2, 4th time: i = 3, 5th time: i = 4
for i in range(5):
    print(i)            # Print the current value of i
````

````python
# Each time the loop repeats, one full round of the indented code is called a "repetition"
# Every indented line runs once per repetition, in order from top to bottom
for i in range(3):
    print("Repetition number", i)     # print() can show several values at once if you separate them with commas
    print("----------")               # This line is also inside the loop (same indentation)

print("The loop is finished.")        # NOT indented, so it is outside the loop. It runs only once, after the loop ends
````

````python
# The variable name doesn't have to be i
# i is just the name most people use. You can choose any name, as long as it follows the variable naming rules
for number in range(3):
    print(number)

for count in range(3):
    print("count is", count)
````

## 2. range() in Detail

`range()` creates a series of numbers. There are three ways to use it.

| How to write it | Meaning | Numbers it gives |
|---|---|---|
| `range(end)` | from 0 up to (but not including) end | `range(5)` → 0, 1, 2, 3, 4 |
| `range(start, end)` | from start up to (but not including) end | `range(1, 6)` → 1, 2, 3, 4, 5 |
| `range(start, end, step)` | from start up to (but not including) end, jumping by step | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |

The most important thing to remember:
- **The end number is never included.** `range(5)` stops at 4, not 5.
- If you don't give a start number, it starts from **0**.
- If you don't give a step, it goes up by **1**.

````python
# range(end): starts from 0, and stops just before end
# range(4) gives 0, 1, 2, 3 -> 4 is NOT included
for i in range(4):
    print(i)
````

````python
# range(start, end): starts from start, and stops just before end
# If we want 1 to 5, the end must be 6 (because the end number is not included)
for i in range(1, 6):
    print(i)
````

````python
# range(start, end, step): jumps by step each time
# Start from 0, jump by 2, stop before 10 -> 0, 2, 4, 6, 8
for i in range(0, 10, 2):
    print(i)
````

````python
# Another step example: count by 5 from 5 to 25
# The end is 26 so that 25 is included (we could also write 30, since 30 itself would not be included)
for i in range(5, 26, 5):
    print(i)
````

````python
# If the step is a negative number, the numbers go DOWN (counting backwards)
# Start from 5, go down by 1, stop before 0 -> 5, 4, 3, 2, 1
for i in range(5, 0, -1):
    print(i)

print("Lift off!")      # Outside the loop, runs once at the end
````

````python
# Using i in a calculation
# Print the multiplication table of 3 (3 x 1 to 3 x 9)
for i in range(1, 10):           # i goes from 1 to 9 (10 is not included)
    print(3, "x", i, "=", 3 * i)  # 3 * i is calculated again with the new i each time
````

## 3. Adding Things Up with a Loop

One of the most common jobs for a loop is **adding up** numbers. For example, what is 1 + 2 + 3 + ... + 10?

The trick is to use a variable as a "running total".
1. Before the loop, make a variable and set it to 0. (`total = 0`)
2. Inside the loop, add the current number to it each time. (`total = total + i`)
3. After the loop, the variable holds the final result.

Remember `age = age + 1` from Lesson 1? It's exactly the same idea: take the old value, add something, and store it back into the same box.

````python
# Add up 1 + 2 + 3 + ... + 10
total = 0                     # Step 1: the running total starts at 0 (BEFORE the loop)

for i in range(1, 11):        # i goes from 1 to 10
    total = total + i         # Step 2: add the current i to the running total

print("Sum:", total)          # Step 3: AFTER the loop, print the final result (outside the loop, so it prints only once)
````

````python
# Let's see how total changes in every repetition
# This is the same code as above, but with a print() inside the loop
total = 0

for i in range(1, 6):         # i goes from 1 to 5
    total = total + i
    print("i =", i, "-> total =", total)   # Inside the loop, so it prints every time

# 1st time: 0 + 1 = 1
# 2nd time: 1 + 2 = 3
# 3rd time: 3 + 3 = 6
# 4th time: 6 + 4 = 10
# 5th time: 10 + 5 = 15
````

````python
# You can write total = total + i in a shorter way: total += i
# Both lines mean exactly the same thing. Many Python programmers use the short form
total = 0

for i in range(1, 101):       # i goes from 1 to 100
    total += i                # Same as: total = total + i

print("Sum of 1 to 100:", total)
````

````python
# A running total doesn't have to be a sum. We can multiply too
# 1 x 2 x 3 x 4 x 5 (this is called a "factorial" in math)
result = 1                    # For multiplication, start with 1, not 0 (anything x 0 is 0!)

for i in range(1, 6):         # i goes from 1 to 5
    result = result * i       # Multiply the current i into the result

print("1 x 2 x 3 x 4 x 5 =", result)
````

````python
# Counting how many times the loop ran is also a running total
# Here we add 1 each time, instead of adding i
count = 0

for i in range(10, 20):       # i goes from 10 to 19
    count += 1                # Add 1 every repetition

print("The loop ran", count, "times.")   # range(10, 20) has 10 numbers, so it ran 10 times
````

## 4. Using for and if Together

You can put an `if` **inside** a `for` loop. This lets you check every number one by one, and do something only for the numbers you want.

```python
for i in range(...):
    if condition:
        code that runs only when the condition is True
```

- The `if` is inside the `for`, so it is indented by 4 spaces.
- The code inside the `if` is indented one more step, so it is indented by 8 spaces.

````python
# Print only the even numbers from 1 to 10
# Remember from Lesson 2: a number is even if number % 2 == 0
for i in range(1, 11):        # Check each number from 1 to 10, one at a time
    if i % 2 == 0:            # Is the current i even? (4 spaces: inside the for)
        print(i)              # Only even numbers get printed (8 spaces: inside the if)
````

````python
# The same result can also be made with range() steps only, without if
# Start at 2, jump by 2 -> 2, 4, 6, 8, 10
for i in range(2, 11, 2):
    print(i)

# Both ways are correct. But with if, you can check any condition you want, not just "every 2nd number"
````

````python
# Using if - else inside a loop
for i in range(1, 6):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
````

````python
# for + if + running total together
# Add up only the multiples of 3 between 1 and 20 (3 + 6 + 9 + 12 + 15 + 18)
total = 0

for i in range(1, 21):        # i goes from 1 to 20
    if i % 3 == 0:            # Is i a multiple of 3? (no remainder when divided by 3)
        total += i            # Only then, add it to the total

print("Sum of multiples of 3:", total)
````

````python
# A classic exercise: FizzBuzz
# For numbers 1 to 15:
#   - if it is a multiple of both 3 and 5, print "FizzBuzz"
#   - if it is a multiple of 3 only, print "Fizz"
#   - if it is a multiple of 5 only, print "Buzz"
#   - otherwise, print the number itself
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:     # Check the strictest condition FIRST (remember Lesson 2: order matters!)
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

## 5. while Loop

A `for` loop is good when you know **how many times** to repeat. But sometimes you don't know the number in advance. You only know **when to stop**. For example:

- Keep saving money **while** the total is less than 10000.
- Keep cutting the paper in half **while** it is longer than 1 cm.

For this, we use the `while` loop. `while` means "as long as". It keeps repeating **as long as the condition is True**, and stops when the condition becomes False.

```python
while condition:
    code to repeat
```

How it works:
1. Check the condition.
2. If it is True, run the indented code, then go back to step 1.
3. If it is False, stop the loop and move on to the code after it.

**Very important:** Something inside the loop must change, so that the condition becomes False at some point. If the condition stays True forever, the loop never ends. This is called an **infinite loop**.

````python
# Print 1 to 5 using while
# With while, we must create, check, and change the variable ourselves
i = 1                     # 1. Create the variable BEFORE the loop

while i <= 5:             # 2. Check the condition: keep going as long as i is 5 or less
    print(i)
    i = i + 1             # 3. Change the variable. Without this line, i stays 1 forever (infinite loop!)

print("Done. Now i is", i)    # When i became 6, 6 <= 5 was False, so the loop stopped
````

````python
# Compare: the same thing with for
# for does the "create, check, change" work automatically with range()
for i in range(1, 6):
    print(i)

# So when you know how many times to repeat, for is usually simpler
````

````python
# A case where while fits better: we don't know how many times in advance
# You save 3000 won every day. How many days does it take to have 20000 won or more?
money = 0
days = 0

while money < 20000:          # Keep saving as long as the money is less than 20000
    money += 3000             # Save 3000 won
    days += 1                 # One more day has passed

print("Days needed:", days)
print("Money saved:", money)
````

````python
# Another example: keep cutting a number in half until it becomes 1 or less
# We use // (quotient only, from Lesson 1) so the number stays an integer
number = 100
steps = 0

while number > 1:             # Keep going as long as number is greater than 1
    number = number // 2      # Cut it in half (100 -> 50 -> 25 -> 12 -> 6 -> 3 -> 1)
    steps += 1
    print("Step", steps, "->", number)

print("It took", steps, "steps.")
````

````python
# If the condition is False from the very beginning, the while loop doesn't run even once
i = 10

while i < 5:                  # 10 < 5 is False right away
    print("This never prints.")
    i += 1

print("The loop was skipped.")
````

## 6. break and continue

Sometimes you want to change the flow of a loop in the middle.

- `break`: **stop the whole loop right now**, and jump out of it.
- `continue`: **skip the rest of this repetition**, and go straight to the next one.

Both are usually used together with `if`, because we want to break or continue only in a certain situation.

````python
# break: stop the loop as soon as we find what we are looking for
# Find the first number between 1 and 100 that is a multiple of 7
for i in range(1, 101):
    if i % 7 == 0:                # Is i a multiple of 7?
        print("Found it:", i)
        break                     # Stop the loop right here. The numbers after 7 are not checked at all

print("The loop is over.")
````

````python
# Let's see exactly where break stops the loop
for i in range(1, 11):
    if i == 4:                    # When i becomes 4...
        break                     # ...stop the loop immediately
    print(i)                      # So only 1, 2, 3 are printed (4 is not printed, because break comes first)
````

````python
# continue: skip only this repetition, and keep going with the next one
for i in range(1, 6):
    if i == 3:                    # When i is 3...
        continue                  # ...skip the print() below and go straight to i = 4
    print(i)                      # So 1, 2, 4, 5 are printed (only 3 is missing)
````

````python
# break with while: a common pattern is "while True" + break
# while True means "repeat forever", so break is the ONLY way out
count = 0

while True:                       # The condition is always True -> this would be an infinite loop without break
    count += 1
    print("count:", count)
    if count == 3:                # When count reaches 3...
        break                     # ...get out of the loop

print("Out of the loop!")
````

````python
# Comparing break and continue side by side
print("With break:")
for i in range(1, 6):
    if i == 3:
        break                     # Stops everything at 3
    print(i)                      # 1, 2

print("With continue:")
for i in range(1, 6):
    if i == 3:
        continue                  # Skips only 3
    print(i)                      # 1, 2, 4, 5
````

## 7. Common Mistakes

These are the mistakes almost everyone makes when learning loops. If something goes wrong, check these first.

1. **Forgetting that the end number of range() is not included**: `range(1, 10)` stops at 9. To include 10, write `range(1, 11)`.
2. **Forgetting the colon `:`**: The `for` and `while` lines must end with `:`.
3. **Wrong indentation**: Only indented lines are repeated. A line that should be inside the loop but isn't indented runs only once.
4. **Infinite loop with while**: If you forget to change the variable inside a `while` loop, the condition stays True forever.
5. **Resetting the running total inside the loop**: `total = 0` must be written **before** the loop, not inside it.

The cells below show what each mistake looks like. The broken code is written as comments so it won't break the notebook.

**If you ever start an infinite loop by accident**, click the ■ (stop / interrupt) button at the top of Jupyter, or press `I` twice (`I`, `I`) while the cell is selected.

````python
# Mistake 1: forgetting that the end number is not included
# We want 1 to 5, but...
for i in range(1, 5):     # Wrong: this stops at 4!
    print(i)

print("---")

for i in range(1, 6):     # Correct: the end must be one more than the last number we want
    print(i)
````

````python
# Mistake 2: forgetting the colon

# for i in range(3)       # SyntaxError! The colon at the end is missing
#     print(i)

for i in range(3):        # Correct: the line ends with a colon
    print(i)
````

````python
# Mistake 3: wrong indentation
# The second print() is NOT indented, so it is outside the loop and runs only once
for i in range(3):
    print("Inside the loop:", i)
print("Outside the loop:", i)     # Runs once, after the loop. (i still holds the last value, 2)

# for i in range(3):
# print(i)                # IndentationError! There is no indented code inside the for at all
````

````python
# Mistake 4: infinite loop with while
# DON'T remove the # from this one! It will never stop.

# i = 1
# while i <= 5:
#     print(i)            # i is never changed, so i <= 5 is True forever -> infinite loop!

i = 1
while i <= 5:
    print(i)
    i += 1                # Correct: change i every time, so that the condition eventually becomes False
````

````python
# Mistake 5: resetting the running total inside the loop
# Wrong: total goes back to 0 every repetition, so only the last number remains
for i in range(1, 6):
    total = 0             # Wrong place! This runs 5 times
    total += i
print("Wrong total:", total)      # Only 5

# Correct: set total to 0 only once, BEFORE the loop
total = 0
for i in range(1, 6):
    total += i
print("Correct total:", total)    # 1 + 2 + 3 + 4 + 5 = 15
````
