## 1. Что такое оператор if?

До сих пор наш код выполнялся сверху вниз, строка за строкой, каждый раз одинаково. Но в жизни мы постоянно делаем выбор.

- **Если** идёт дождь, возьми зонт.
- **Если** тебе 20 лет или больше, ты можешь купить взрослый билет.

Оператор `if` позволяет программе делать такой выбор. Он проверяет **условие** и выполняет код внутри себя **только тогда, когда условие равно True**.

Помните операторы сравнения из Урока 1 (`>`, `<`, `==`, `!=`)? Их результат всегда `True` или `False`. Именно это и нужно `if` в качестве условия.

Базовая форма выглядит так:

```python
if условие:
    код, который выполняется только когда условие равно True
```

Два очень важных правила:
- В конце строки с `if` обязательно должно стоять двоеточие `:`.
- Код, который относится к `if`, должен быть **с отступом** (сдвинут вправо на 4 пробела). По отступу Python понимает, какие строки находятся "внутри" `if`.

````python
# Условие здесь - age >= 20 ("возраст больше или равен 20")
# Результат этого сравнения - True или False
age = 25

if age >= 20:                         # 25 >= 20 - это True, поэтому код с отступом ниже выполняется
    print("You are an adult.")        # Эта строка находится внутри if (у неё отступ 4 пробела)

print("This line always runs.")      # У этой строки НЕТ отступа, значит она вне if. Она выполняется в любом случае
````

````python
# Теперь попробуем случай, когда условие равно False
age = 15

if age >= 20:                         # 15 >= 20 - это False, поэтому код с отступом ниже пропускается
    print("You are an adult.")        # Эта строка вообще не выполняется

print("This line always runs.")      # Эта строка вне if, поэтому она всё равно выполняется
````

````python
# Внутри if может быть больше одной строки
# Все строки с одинаковым отступом относятся к одному и тому же if
temperature = 32

if temperature > 30:                          # 32 > 30 - это True
    print("It's very hot today.")             # Внутри if
    print("Drink plenty of water.")           # Тоже внутри if (такой же отступ)
    print("Try to stay in the shade.")        # Тоже внутри if (такой же отступ)

print("Have a nice day!")                    # Вне if, выполняется всегда
````

````python
# С помощью == можно также проверить, одинаковы ли два значения
# Запомните: = значит "сохранить значение", а == значит "равны ли эти значения?"
password = "python123"

if password == "python123":           # Две строки полностью совпадают, поэтому это True
    print("Login successful!")
````

## 2. if - else

Если есть только `if`, программа что-то делает, когда условие равно True, и **ничего не делает**, когда оно равно False.

Но часто нам нужно сделать **либо одно, либо другое**. Например:
- Если ты сдал экзамен, выведи "Поздравляем". **Иначе** выведи "Попробуй ещё раз".

Для этого и нужен `else`. `else` означает "иначе". Его код выполняется **только когда условие if равно False**.

```python
if условие:
    код, который выполняется, когда условие равно True
else:
    код, который выполняется, когда условие равно False
```

- У `else` нет собственного условия. Он просто обрабатывает все случаи, которые не обработал `if`.
- В конце `else` тоже нужно двоеточие `:`, и его код тоже должен быть с отступом.
- Всегда выполняется ровно **один** из двух блоков. Никогда оба, и никогда ни одного.

````python
# Проверяем, сдан экзамен или нет
# Допустим, проходной балл - 60
score = 75

if score >= 60:                      # 75 >= 60 - это True
    print("Pass! Congratulations.")  # Поэтому выполняется эта строка
else:                                # "Иначе" - эта часть пропускается, потому что if был True
    print("Fail. Try again.")
````

````python
# Тот же код, но с более низким баллом
score = 40

if score >= 60:                      # 40 >= 60 - это False, поэтому эта часть пропускается
    print("Pass! Congratulations.")
else:                                # if был False, поэтому вместо него выполняется else
    print("Fail. Try again.")
````

````python
# Проверяем, чётное число или нечётное
# Помните % из Урока 1? Он даёт остаток от деления
# Чётное число при делении на 2 всегда даёт остаток 0
number = 7

if number % 2 == 0:                  # 7 % 2 - это 1, а 1 == 0 - это False
    print(number, "is even.")        # print() может показать несколько значений сразу, если разделить их запятыми
else:
    print(number, "is odd.")         # if был False, поэтому выполняется эта строка
````

## 3. if - elif - else

Иногда вариантов **больше двух**. Например, перевод баллов в оценку:

- 90 и выше → A
- 80 и выше → B
- 70 и выше → C
- Всё остальное → F

Для этого используется `elif`. `elif` - это сокращение от "else if", оно означает "если все условия выше были False, то проверь это условие".

```python
if условие1:
    выполняется, когда условие1 равно True
elif условие2:
    выполняется, когда условие1 равно False, а условие2 равно True
elif условие3:
    выполняется, когда условие1 и условие2 равны False, а условие3 равно True
else:
    выполняется, когда все условия выше равны False
```

Как это работает:
- Python проверяет условия **сверху вниз, по порядку**.
- Как только он находит **первое** условие, равное True, он выполняет этот блок и **пропускает все остальные**.
- Можно добавить сколько угодно `elif`. `else` в конце писать необязательно.

````python
# Переводим баллы в оценку
score = 85

if score >= 90:                # 85 >= 90 - это False -> переходим к следующему условию
    print("Grade: A")
elif score >= 80:              # 85 >= 80 - это True -> выполняется этот блок, а всё ниже пропускается
    print("Grade: B")
elif score >= 70:              # Вообще не проверяется, потому что условие True уже найдено выше
    print("Grade: C")
else:                          # Тоже не проверяется
    print("Grade: F")
````

````python
# Попробуйте разные баллы с тем же кодом
# Измените значение score и запустите ячейку снова, чтобы увидеть, как меняется результат
score = 72

if score >= 90:                # 72 >= 90 - это False
    print("Grade: A")
elif score >= 80:              # 72 >= 80 - это False
    print("Grade: B")
elif score >= 70:              # 72 >= 70 - это True -> выполняется этот блок
    print("Grade: C")
else:
    print("Grade: F")
````

````python
# Почему важен порядок?
# Python останавливается на ПЕРВОМ условии True, поэтому если сначала поставить самую слабую проверку, всё пойдёт не так
score = 95

if score >= 70:                # 95 >= 70 - это True -> этот блок выполняется первым, а всё остальное пропускается!
    print("Grade: C")
elif score >= 80:              # Никогда не проверяется
    print("Grade: B")
elif score >= 90:              # Никогда не проверяется, хотя 95 заслуживает оценку A
    print("Grade: A")

# Поэтому при проверке диапазонов начинайте с самого большого (самого строгого) условия
````

````python
# Выбираем, что надеть, в зависимости от температуры
temperature = 12

if temperature >= 25:                  # 12 >= 25 - это False
    print("Wear a T-shirt.")
elif temperature >= 15:                # 12 >= 15 - это False
    print("Wear a light jacket.")
elif temperature >= 5:                 # 12 >= 5 - это True -> выполняется эта строка
    print("Wear a coat.")
else:                                  # Выполняется только когда температура ниже 5
    print("Wear a thick padded jacket.")
````

## 4. Переменные bool в качестве условий

Условию `if` достаточно быть `True` или `False`. Это значит, что **переменную bool** (из Урока 1) можно использовать как условие напрямую, без всякого сравнения.

```python
if is_student:        # то же самое, что: if is_student == True:
```

Писать `== True` не ошибка, но это лишнее, поэтому программисты на Python обычно это не пишут.

Чтобы проверить, что переменная bool равна `False`, используйте `not`:

```python
if not is_student:    # то же самое, что: if is_student == False:
```

````python
# Переменная bool уже хранит True или False, поэтому её можно использовать как условие напрямую
is_student = True

if is_student:                        # is_student равно True, поэтому выполняется эта строка
    print("You get a student discount.")
else:
    print("Regular price.")
````

````python
# Используем not, чтобы проверить False
# not меняет значение на противоположное: not True - это False, а not False - это True
is_raining = False

if not is_raining:                    # is_raining равно False, значит not is_raining равно True -> выполняется эта строка
    print("No umbrella needed.")
else:
    print("Take an umbrella.")
````

````python
# Можно сначала сохранить результат сравнения в переменную bool, а потом использовать её
# Это удобно, когда условие длинное или когда хочется дать ему понятное имя
age = 17
is_adult = age >= 20                  # 17 >= 20 - это False, поэтому в is_adult сохраняется False
print(is_adult)

if is_adult:                          # is_adult равно False, поэтому эта часть пропускается
    print("You can watch this movie.")
else:
    print("You can't watch this movie.")
````

## 5. Объединение условий (and, or, not)

Иногда одного условия недостаточно. Например:
- На американских горках можно кататься, если тебе **не меньше 10 лет и твой рост не меньше 140 см**.
- Вход бесплатный, если тебе **меньше 7 лет или 65 лет и больше**.

В Python есть три слова для объединения условий.

| Оператор | Значение | Результат |
|---|---|---|
| `and` | и | True только когда **обе** стороны равны True |
| `or` | или | True, когда **хотя бы одна** сторона равна True |
| `not` | наоборот | Меняет True на False, а False на True |

````python
# Сначала посмотрим, что дают and / or сами по себе
print(True and True)     # обе стороны True -> True
print(True and False)    # одна сторона False -> False
print(True or False)     # хотя бы одна сторона True -> True
print(False or False)    # обе стороны False -> False
print(not True)          # противоположность True -> False
````

````python
# and: все условия должны быть True
# Правило: не меньше 10 лет И рост не меньше 140 см
age = 12
height = 135

if age >= 10 and height >= 140:       # 12 >= 10 - это True, но 135 >= 140 - это False -> True and False -> False
    print("You can ride the roller coaster.")
else:
    print("Sorry, you can't ride it yet.")  # Условие было False, поэтому выполняется эта строка
````

````python
# or: хотя бы одно условие должно быть True
# Правило: младше 7 лет ИЛИ 65 лет и старше - вход бесплатный
age = 70

if age < 7 or age >= 65:              # 70 < 7 - это False, но 70 >= 65 - это True -> False or True -> True
    print("Free entry!")              # Условие было True, поэтому выполняется эта строка
else:
    print("Please buy a ticket.")
````

````python
# Проверяем, попадает ли число в диапазон
# "score находится между 0 и 100" значит "score не меньше 0 И score не больше 100"
score = 105

if score >= 0 and score <= 100:       # 105 >= 0 - это True, но 105 <= 100 - это False -> False
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")

# В Python то же самое можно записать короче, как в математике
if 0 <= score <= 100:                 # Точно такое же значение, как в строке выше
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")
````

## 6. Вложенные операторы if

Можно поместить один `if` **внутрь другого** `if`. Это называется **вложенным** if.

Это полезно, когда второе условие нужно проверить **только после того**, как первое условие оказалось True. Например, сначала проверить, вошёл ли пользователь в систему, и только потом проверить, является ли он администратором.

Внутренний `if` имеет отступ на один уровень больше (на 4 пробела больше), чем внешний `if`.

````python
# Сначала проверяем вход в систему, и только если вход выполнен, проверяем права администратора
is_logged_in = True
is_admin = False

if is_logged_in:                               # Внешний if: True, поэтому заходим внутрь
    print("Welcome!")                          # Относится к внешнему if (4 пробела)
    if is_admin:                               # Внутренний if: is_admin равно False
        print("You can open the admin page.")  # Относится к внутреннему if (8 пробелов) - пропускается
    else:
        print("You are a regular user.")       # Относится к внутреннему else (8 пробелов) - выполняется
else:                                          # Относится к внешнему if
    print("Please log in first.")              # Пропускается, потому что внешний if был True
````

````python
# Покупка в магазине
# Шаг 1: есть ли товар в наличии?  Шаг 2: хватает ли денег?
in_stock = True
money = 8000
price = 12000

if in_stock:                                   # Товар в наличии -> заходим внутрь
    if money >= price:                         # 8000 >= 12000 - это False
        print("Purchase complete!")
    else:
        print("You don't have enough money.")  # Выполняется эта строка
        print("You need", price - money, "more won.")  # 12000 - 8000 = 4000
else:
    print("Sorry, this item is sold out.")
````

**Совет:** Вложенные if полезны, но слишком много уровней делает код трудным для чтения. Когда это возможно, попробуйте объединить условия с помощью `and`.

```python
# вложенный вариант
if is_logged_in:
    if is_admin:
        print("You can open the admin page.")

# то же самое с and
if is_logged_in and is_admin:
    print("You can open the admin page.")
```

## 7. Частые ошибки

Эти ошибки делает почти каждый, кто изучает `if`. Если у вас появилась ошибка, сначала проверьте эти пункты.

1. **`=` вместо `==`**: `=` сохраняет значение, `==` сравнивает значения. `if age = 20:` - это ошибка.
2. **Забытое двоеточие `:`**: Строки `if`, `elif` и `else` должны заканчиваться на `:`.
3. **Неправильный отступ**: Код внутри должен быть с отступом, а строки одного блока должны иметь одинаковый отступ.
4. **Условие у `else`**: У `else` никогда не бывает условия. `else age < 20:` - это ошибка. Если нужно условие, используйте `elif`.
5. **Сравнение числа со строкой**: `"20"` (строка) и `20` (целое число) - это разные вещи. `"20" == 20` равно `False`.

Ячейки ниже показывают, как выглядит каждая ошибка. Ошибочный код записан в виде комментариев, чтобы не ломать блокнот. Попробуйте убрать `#` в строке с кодом, чтобы увидеть настоящее сообщение об ошибке, а потом верните `#` обратно.

````python
# Ошибка 1: = вместо ==
age = 20

# if age = 20:          # SyntaxError! = значит "сохранить", поэтому его нельзя использовать как условие
#     print("20 years old")

if age == 20:           # Правильно: == значит "равны ли они?"
    print("20 years old")
````

````python
# Ошибка 2: забытое двоеточие
age = 20

# if age >= 20          # SyntaxError! В конце не хватает двоеточия
#     print("adult")

if age >= 20:           # Правильно: строка заканчивается двоеточием
    print("adult")
````

````python
# Ошибка 3: неправильный отступ
age = 20

# if age >= 20:
# print("adult")        # IndentationError! У строки внутри if нет отступа

if age >= 20:
    print("adult")      # Правильно: отступ 4 пробела
````

````python
# Ошибка 4: условие у else
age = 15

# if age >= 20:
#     print("adult")
# else age < 20:        # SyntaxError! У else не может быть условия
#     print("minor")

if age >= 20:
    print("adult")
else:                   # Правильно: у else нет условия. Он обрабатывает всё, что не обработал if
    print("minor")
````

````python
# Ошибка 5: сравнение числа со строкой
# Сообщения об ошибке нет, но результат не тот, что вы ожидали, поэтому эта ошибка коварная
age_text = "20"          # Это строка, потому что она в кавычках (вспомните zip_code из Урока 1)
age_number = 20          # Это целое число

print(age_text == age_number)     # False! Строка и целое число никогда не равны

if age_text == age_number:
    print("Same")
else:
    print("Different")            # Выполняется эта строка, хотя значения выглядят одинаково
````
