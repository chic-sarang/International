### Function ရဲ့ အယူအဆ

အဲဒီ 3 ကြောင်းကို **တစ်ကြိမ်တည်း** ရေးပြီး **နာမည်** ပေးထားမယ်၊ လိုအပ်တဲ့အခါ အဲဒီနာမည်ကိုပဲ ခေါ်လိုက်ရင် ဘယ်လိုလဲ?

အဲဒါက **function** ပါပဲ: **နာမည်ရှိတဲ့ code အစုအဝေး** တစ်ခု။

### ဟင်းချက်နည်း (recipe) ကို စဉ်းစားကြည့်ပါ

Function ဟာ **ဟင်းချက်နည်း** နဲ့ တူပါတယ်။

1. ပထမ၊ ဟင်းချက်နည်းကို စက္ကူပေါ်မှာ **ရေးထား** ပါတယ်: "ကြက်ဥကြော်နည်း: အဆင့် 1၊ အဆင့် 2၊ အဆင့် 3"။ ဟင်းချက်နည်း ရေးတာက ကြက်ဥကြော် ဖြစ်မလာပါဘူး။ နောက်မှ သုံးဖို့ သိမ်းထားတဲ့ ညွှန်ကြားချက်ပဲ ဖြစ်ပါတယ်။
2. ပြီးရင် ကြက်ဥကြော် စားချင်တိုင်း ဟင်းချက်နည်း **အတိုင်း ချက်** ပါတယ်။ ဟင်းချက်နည်း တစ်ခုတည်းကို အကြိမ်ကြိမ် လိုက်လုပ်လို့ရပါတယ်။

Function လည်း ဒီလိုပဲ အလုပ်လုပ်ပါတယ်:

| ဟင်းချက်နည်း | Function |
|---|---|
| ဟင်းချက်နည်း ရေးခြင်း | Function **ဖန်တီးခြင်း** (define) |
| ဟင်းချက်နည်းအတိုင်း ချက်ခြင်း | Function **အသုံးပြုခြင်း** (call / ခေါ်ခြင်း) |

### သင် function ကို သုံးဖူးပြီးသားပါ!

သင်ခန်းစာ 1 ကတည်းက function တွေကို သုံးခဲ့ပြီးပါပြီ။ `print()`၊ `len()`၊ `type()` အားလုံး function တွေပါ။ တစ်ယောက်ယောက်က ရေးထားပေးပြီးသားဖြစ်လို့ နာမည်နဲ့ ကွင်း `( )` ကို ရေးပြီး သုံးခဲ့တာပါ။

ဒီသင်ခန်းစာမှာ **ကိုယ်ပိုင်** function ဖန်တီးနည်းကို လေ့လာပါမယ်။

````python
# သုံးဖူးပြီးသား function များ
print("Hello")              # print() က function: တန်ဖိုးကို screen ပေါ်မှာ ပြတယ်
print(len("apple"))         # len() က function: အရေအတွက် ရေတွက်တယ်
print(type(10))             # type() က function: data type ကို ပြောပြတယ်
````

## 2. Function ဖန်တီးခြင်းနဲ့ အသုံးပြုခြင်း

Function ကို **သီးခြား အဆင့် 2 ဆင့်** နဲ့ သုံးပါတယ်:

1. Function ကို **ဖန်တီး** (ဟင်းချက်နည်း ရေး)
2. Function ကို **အသုံးပြု** (ဟင်းချက်နည်းအတိုင်း ချက်)

### အဆင့် 1: Function ဖန်တီးခြင်း

```python
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")
```

အပိုင်းတစ်ခုချင်း ကြည့်ရအောင်:

| အပိုင်း | အဓိပ္ပာယ် |
|---|---|
| `def` | **define** (သတ်မှတ်) ရဲ့ အတိုကောက်။ "function တစ်ခု ဖန်တီးတော့မယ်" လို့ ဆိုလိုတယ် |
| `say_hello` | function ရဲ့ **နာမည်**။ ကိုယ်တိုင် ရွေးလို့ရတယ် (variable နာမည်ပေးတဲ့ စည်းမျဉ်းနဲ့ တူတယ်) |
| `()` | ကွင်း။ နာမည်နောက်မှာ အမြဲ လိုတယ် (အထဲမှာ ဘာထည့်လို့ရလဲ ဆိုတာ အပိုင်း 3 မှာ လေ့လာမယ်) |
| `:` | `if`၊ `for` လိုပဲ စာကြောင်းအဆုံးမှာ colon |
| indent လုပ်ထားတဲ့ စာကြောင်းများ | function ရဲ့ **body**: function ထဲမှာ ပါဝင်တဲ့ code |

**Indent အကြောင်း:** `if`၊ `for` လိုပဲ Python က indent (စာကြောင်းအစမှာ space 4 ခု) ကို ကြည့်ပြီး ဘယ်စာကြောင်းတွေက function ထဲမှာ ပါလဲ ဆိုတာ သိပါတယ်။ `def` အောက်က indent လုပ်ထားတဲ့ စာကြောင်းတွေက function body ဖြစ်ပါတယ်။ indent မလုပ်ထားတဲ့ ပထမ စာကြောင်းက function **အပြင်ဘက်** မှာ ရှိပါတယ်။

### Function ဖန်တီးတာက အဲဒါကို RUN မလုပ်ပါဘူး

ဒါက ဒီအပိုင်းမှာ အရေးကြီးဆုံး အချက်ပါ။

Python က `def` ကို ဖတ်တဲ့အခါ function ကို **မှတ်ထား** ရုံပါပဲ။ အထဲက code ကို မ run ပါဘူး။ ဟင်းချက်နည်း ရေးတာက ကြက်ဥကြော် ဖြစ်မလာတာနဲ့ တူပါတယ်။

အောက်က cell ကို ကြည့်ပါ။ function ဖန်တီးထားပေမဲ့ **ဘာမှ print မထွက်ပါဘူး**။

````python
# အဆင့် 1: function ကို ဖန်တီးတယ်
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")

# ဘာမှ print မထွက်ဘူး! function ကို ဖန်တီး (ရေး) ရုံပဲ ရှိသေးတယ်။ မသုံးရသေးဘူး။
````

### အဆင့် 2: Function အသုံးပြုခြင်း

Function ကို သုံးဖို့ **နာမည်** ကို ရေးပြီး နောက်မှာ **ကွင်း** `()` ထည့်ပါ:

```python
say_hello()
```

Function ကို သုံးတာကို function ကို **ခေါ်တယ် (call)** လို့ ခေါ်ပါတယ်။ Function ကို ခေါ်တဲ့အခါ Python က အဲဒီ function ဆီ သွားပြီး အထဲက code အားလုံးကို အပေါ်ကနေ အောက်ကို run ပြီး ပြန်လာပါတယ်။

````python
# အဆင့် 2: function ကို ခေါ် (သုံး) တယ်
say_hello()
````

### ဟင်းချက်နည်း တစ်ခု၊ အကြိမ်ကြိမ်

Function ကို ဖန်တီးပြီးရင် **ကြိုက်သလောက် အကြိမ်** ခေါ်လို့ရပါတယ်။ ခေါ်တိုင်း body တစ်ခုလုံး ပြန် run ပါတယ်။

````python
# တူညီတဲ့ နှုတ်ဆက်စာ 3 ကြိမ်၊ ဒါပေမဲ့ အခု စာကြောင်းတို 3 ကြောင်းနဲ့ပဲ
say_hello()
say_hello()
say_hello()
````

သင်ခန်းစာ အစနဲ့ နှိုင်းယှဉ်ကြည့်ပါ။ အရင်က 9 ကြောင်း လိုခဲ့တယ်။ အခု 3 ကြိမ် ခေါ်ရုံပါပဲ။

"Hello!" ကို "Hi!" ပြောင်းချင်ရင် **တစ်နေရာတည်း**: function ထဲမှာ ပြောင်းပါ။ ခေါ်တိုင်း နှုတ်ဆက်စာ အသစ်ကို သုံးပါလိမ့်မယ်။

````python
# function ကို တစ်နေရာတည်းမှာ ပြောင်းရင်...
def say_hello():
    print("Hi!")                        # ဒီမှာပဲ ပြောင်းထားတယ်
    print("Welcome to Python class.")
    print("Have a nice day!")

# ...ခေါ်တိုင်း version အသစ်ကို သုံးတယ်
say_hello()
say_hello()
````

### Cell တစ်ခုတည်းမှာ ဖန်တီးပြီး သုံးခြင်း

များသောအားဖြင့် function ကို အရင် ဖန်တီးပြီး အောက်မှာ ခေါ်ပါတယ်။ Python ဘာလုပ်လဲ ဆိုတာ တစ်ဆင့်ချင်း လိုက်ကြည့်ရအောင်။

````python
def say_goodbye():          # 1. Python က "say_goodbye" function ကို မှတ်ထားတယ်။ ဘာမှ မ run သေးဘူး
    print("Goodbye!")
    print("See you next time.")

print("Class is over.")     # 2. ဒီစာကြောင်းက indent မရှိဘူး -> function အပြင် -> အခု run တယ်
say_goodbye()               # 3. function ကို ခေါ်တယ် -> အထဲက 2 ကြောင်း အခု run တယ်
print("The end.")           # 4. function ပြီးတဲ့အခါ Python က ဒီကို ပြန်လာတယ်
````

Output ရဲ့ အစီအစဉ်ကို သတိထားကြည့်ပါ:

1. `Class is over.` - function **အပြင်** က ပထမ စာကြောင်း
2. `Goodbye!` နဲ့ `See you next time.` - function body က **ခေါ်** တဲ့အခါမှပဲ run တယ်
3. `The end.` - function ပြီးတဲ့နောက် Python က ခေါ်ခဲ့တဲ့ နေရာရဲ့ နောက်စာကြောင်းကနေ ဆက်လုပ်တယ်

### အရင် ဖန်တီး၊ ပြီးမှ သုံး

ဟင်းချက်နည်းကို လိုက်မချက်ခင် **အရင်** ရေးထားရပါမယ်။ ဒီလိုပဲ function ကို **မခေါ်ခင် ဖန်တီးထား** ရပါမယ်။ Python က code ကို အပေါ်ကနေ အောက်ကို ဖတ်တဲ့အတွက် မဖန်တီးရသေးတဲ့ function ကို ခေါ်ရင် Python က အဲဒါကို မသိဘဲ error ပြပါလိမ့်မယ်။ (ဒီ error ကို အပိုင်း 5 မှာ ကြည့်ပါမယ်။)

### နောက်ထပ် ဥပမာများ

````python
# မျဉ်းဆွဲတဲ့ function
def draw_line():
    print("--------------------")

draw_line()
print("Python Basics")
draw_line()
````

## 3. Function ကို တန်ဖိုး ပေးပို့ခြင်း

### ပြဿနာ

ကျွန်တော်တို့ရဲ့ `say_hello()` function က အမြဲ တူညီတာကိုပဲ ပြောပါတယ်။ ဒါပေမဲ့ **လူအမျိုးမျိုး** ကို သူတို့ နာမည်နဲ့ နှုတ်ဆက်ချင်ရင် ဘယ်လိုလုပ်မလဲ?

- "Hello, Anna!"
- "Hello, Tom!"
- "Hello, Bill!"

လူတိုင်းအတွက် function အသစ် မဖန်တီးချင်ပါဘူး။ function **တစ်ခုတည်း** ရှိပြီး ခေါ်တိုင်း နာမည်ကို **ပြော** ပေးချင်ပါတယ်။

### ကော်ဖီဆိုင်မှာ မှာတာကို စဉ်းစားကြည့်ပါ

ကော်ဖီဆိုင်မှာ မှာတဲ့အခါ ဝန်ထမ်းကို ဘာလိုချင်လဲ **ပြော** ပါတယ်: "ကော်ဖီ တစ်ခွက်ပေးပါ" ဒါမှမဟုတ် "လက်ဖက်ရည် တစ်ခွက်ပေးပါ"။ ဝန်ထမ်းက အလုပ်တူတူ (သောက်စရာ ဖျော်တာ) လုပ်ပေမဲ့ ရလဒ်က **သင်ပြောတာ** ပေါ် မူတည်ပါတယ်။

Function လည်း ဒီလို အလုပ်လုပ်နိုင်ပါတယ်။ ခေါ်တဲ့အခါ function ကို တန်ဖိုး **ပေးပို့** လို့ရပြီး function က **အဲဒီတန်ဖိုးနဲ့** အလုပ်လုပ်ပါတယ်။

### တန်ဖိုး လက်ခံတဲ့ function ဖန်တီးခြင်း

```python
def greet(name):
    print("Hello,", name)
```

ကွင်းထဲကို ကြည့်ပါ: `name` ဆိုတဲ့ စကားလုံး ရှိပါတယ်။ ဒါက function ကို ခေါ်တဲ့အခါ ပေးပို့လိုက်တဲ့ တန်ဖိုးကို လက်ခံမယ့် **variable** ပါ။ ဒီ variable ကို **parameter** လို့ ခေါ်ပါတယ်။

### တန်ဖိုးနဲ့ ခေါ်ခြင်း

```python
greet("Anna")
```

အခု function ကို ခေါ်တဲ့အခါ ကွင်းထဲမှာ တန်ဖိုး ထည့်ပါတယ်။ ပေးပို့လိုက်တဲ့ တန်ဖိုးကို **argument** လို့ ခေါ်ပါတယ်။

| စကားလုံး | ဘယ်မှာ ရှိလဲ | ဥပမာ |
|---|---|---|
| **parameter** | function ကို **ဖန်တီး** တဲ့အခါ `( )` ထဲမှာ | `def greet(name):` ထဲက `name` |
| **argument** | function ကို **ခေါ်** တဲ့အခါ `( )` ထဲမှာ | `greet("Anna")` ထဲက `"Anna"` |

ဒီစကားလုံး နှစ်လုံးအတွက် သိပ်စိတ်ပူစရာ မလိုပါဘူး။ အရေးကြီးတဲ့ အယူအဆက: **ပေးပို့လိုက်တဲ့ တန်ဖိုးက function ထဲက variable မှာ သိမ်းဆည်းခံရတယ်။**

````python
# တန်ဖိုး တစ်ခု လက်ခံတဲ့ function ဖန်တီးတယ်
def greet(name):
    print("Hello,", name)

# မတူတဲ့ တန်ဖိုးတွေနဲ့ ခေါ်တယ်
greet("Anna")
greet("Tom")
greet("Bill")
````

**`print("Hello,", name)` အကြောင်း:** `print()` ထဲမှာ တန်ဖိုး အများကြီးကို **comma** `,` နဲ့ ခြားပြီး ရေးရင် စာကြောင်း တစ်ကြောင်းတည်းမှာ print ထွက်ပြီး Python က ကြားမှာ **space တစ်ခု** အလိုအလျောက် ထည့်ပေးပါတယ်။ quote ထဲက စာက ရေးထားတဲ့အတိုင်း ထွက်ပြီး variable ကတော့ အထဲက တန်ဖိုး ထွက်ပါတယ်။

### ဘာဖြစ်သွားလဲ၊ တစ်ဆင့်ချင်း

`greet("Anna")` ကို ခေါ်တဲ့အခါ:

| အဆင့် | Python ဘာလုပ်လဲ |
|---|---|
| 1 | `greet` function ကို ရှာတယ် |
| 2 | ပေးပို့လိုက်တဲ့ တန်ဖိုးကို parameter မှာ သိမ်းတယ်: `name = "Anna"` |
| 3 | body ကို run တယ်: `print("Hello,", name)` → `name` က `"Anna"` → `Hello, Anna` print ထွက်တယ် |
| 4 | function ပြီးသွားလို့ Python က ခေါ်ခဲ့တဲ့ နေရာရဲ့ နောက်စာကြောင်းကို ပြန်သွားတယ် |

ပြီးရင် `greet("Tom")` ကို ခေါ်တဲ့အခါ အတူတူ ထပ်ဖြစ်ပါတယ်၊ ဒါပေမဲ့ ဒီတစ်ခါ `name = "Tom"`။

ဒါကြောင့် **function ထဲက code က အမြဲ တူပေမဲ့** ပေးပို့တဲ့ တန်ဖိုးပေါ် မူတည်ပြီး **ရလဒ် ပြောင်းပါတယ်**။

````python
# parameter က ရိုးရိုး variable ပဲ။ အထဲမှာ ဘာရှိလဲ ကြည့်လို့ရတယ်
def show_value(x):
    print("The function received:", x)

show_value(10)
show_value("apple")
show_value(3.14)
````

````python
# တန်ဖိုးကို တွက်ချက်မှုမှာ သုံးတဲ့ function
def show_double(number):
    print(number * 2)       # number နေရာမှာ ပေးပို့လိုက်တဲ့ တန်ဖိုး ဝင်လာတယ်

show_double(5)              # number = 5  -> 5 * 2
show_double(100)            # number = 100 -> 100 * 2
````

````python
# argument က variable လည်း ဖြစ်နိုင်တယ်
def greet(name):
    print("Hello,", name)

student = "Kate"
greet(student)              # student ထဲက တန်ဖိုး ("Kate") ကို function ကို ပေးပို့တယ်
````

### တန်ဖိုး နှစ်ခု ပေးပို့ခြင်း

Function က တန်ဖိုး **တစ်ခုထက်ပို** လက်ခံနိုင်ပါတယ်။ function ဖန်တီးတဲ့အခါရော ခေါ်တဲ့အခါရော comma `,` နဲ့ ခြားပါ။

```python
def add(a, b):
    print(a + b)

add(2, 3)
```

တန်ဖိုးတွေက **အစီအစဉ်အတိုင်း** ကိုက်ညီပါတယ်:
- ပထမ တန်ဖိုး (`2`) က ပထမ parameter (`a`) ထဲ ဝင်တယ်
- ဒုတိယ တန်ဖိုး (`3`) က ဒုတိယ parameter (`b`) ထဲ ဝင်တယ်

````python
def add(a, b):
    print(a + b)

add(2, 3)                   # a = 2, b = 3  -> 2 + 3
add(10, 20)                 # a = 10, b = 20 -> 10 + 20
````

````python
# parameter နှစ်ခုပါတဲ့ function: နာမည်နဲ့ အသက်
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce("Anna", 20)       # name = "Anna", age = 20
introduce("Tom", 25)        # name = "Tom", age = 25
````

တန်ဖိုးတွေက အစီအစဉ်အတိုင်း ကိုက်ညီတဲ့အတွက် အစီအစဉ် ပြောင်းရင် ရလဒ် ပြောင်းနိုင်ပါတယ်။

````python
def subtract(a, b):
    print(a - b)

subtract(10, 3)             # a = 10, b = 3  -> 10 - 3 = 7
subtract(3, 10)             # a = 3, b = 10  -> 3 - 10 = -7 (ရလဒ် မတူဘူး!)
````

## 4. ရလဒ်ကို ပြန်ယူခြင်း: return

### ပြဿနာ

ကျွန်တော်တို့ရဲ့ `add(a, b)` function က ရလဒ်ကို screen ပေါ်မှာ **print** လုပ်ပါတယ်။ ဒါပေမဲ့ ရလဒ်ကို နောက်မှ **သုံး** ချင်ရင် ဘယ်လိုလုပ်မလဲ? ဥပမာ variable ထဲ သိမ်းတာ၊ ဒါမှမဟုတ် ဆက်တွက်တာ?

Print လုပ်တာက ရလဒ်ကို **ပြ** ရုံပါပဲ။ ရလဒ်ကို ကျွန်တော်တို့ဆီ **ပြန်မပေး** ပါဘူး။ print ထွက်ပြီးရင် ပြန်သုံးလို့ မရတော့ပါဘူး။

### တစ်ခုခု ဝယ်ခိုင်းတာကို စဉ်းစားကြည့်ပါ

သူငယ်ချင်းကို "ဆိုင်သွားပြီး ပေါင်မုန့် ဝယ်ပေးပါ" လို့ ခိုင်းတယ် ဆိုပါစို့။

- သူငယ်ချင်းက သွားပြီး ပေါင်မုန့် ဝယ်၊ ပြန်လာတယ်။
- ပြီးရင် သူငယ်ချင်းက **ပေါင်မုန့်ကို သင့်လက်ထဲ ပေး** တယ်။
- အခု **သင့်မှာ ပေါင်မုန့် ရှိပြီ**၊ ကြိုက်တာ လုပ်လို့ရတယ်: စား၊ ခွဲဝေ၊ sandwich လုပ်။

`return` ဟာ **ရလဒ်ကို ပြန်ပေးတာ** နဲ့ တူပါတယ်။ function က အလုပ် လုပ်ပြီး ရလဒ်ကို ခေါ်ခဲ့တဲ့ နေရာဆီ ပြန်ပေးပါတယ်။

### return ပါတဲ့ function ဖန်တီးခြင်း

```python
def add(a, b):
    return a + b
```

`return` ရဲ့ အဓိပ္ပာယ်က: **"ဒီတန်ဖိုးကို function ခေါ်ခဲ့တဲ့ နေရာဆီ ပြန်ပေး"**

သတိထားပါ: ဒီ function က ဘာမှ print မလုပ်ပါဘူး။ ရလဒ်ကို ပြန်ပေးရုံပါပဲ။

````python
def add(a, b):
    return a + b            # ရလဒ်ကို ပြန်ပေးတယ်

result = add(2, 3)          # ရလဒ် (5) ပြန်လာပြီး result ထဲ သိမ်းတယ်
print(result)
````

### ခေါ်တဲ့နေရာက return တန်ဖိုးနဲ့ အစားထိုးခံရတယ်

ဒါက `return` ရဲ့ အရေးကြီးဆုံး အယူအဆပါ။

Function က တန်ဖိုး return လုပ်တဲ့အခါ **function ကို ခေါ်ခဲ့တဲ့ နေရာက အဲဒီတန်ဖိုးနဲ့ အစားထိုးခံရပါတယ်။**

ဒီစာကြောင်းကို ဖြည်းဖြည်း ကြည့်ရအောင်:

```python
result = add(2, 3)
```

| အဆင့် | Python ဘာမြင်လဲ |
|---|---|
| 1 | `result = add(2, 3)` |
| 2 | Python က `add(2, 3)` ကို ခေါ်တယ်: `a = 2`, `b = 3`၊ function က `2 + 3` ဆိုတော့ `5` ကို return လုပ်တယ် |
| 3 | `add(2, 3)` က `5` နဲ့ **အစားထိုး** ခံရတယ် → `result = 5` |
| 4 | `5` ကို `result` ထဲ သိမ်းတယ် |

ဒါကြောင့် function က return လုပ်ပြီးနောက် `result = 5` လို့ ရေးထားသလိုပဲ ဖြစ်သွားပါတယ်။

`add(2, 3)` ကို function ပြီးတဲ့အခါ **5 ဖြစ်သွားတဲ့** သေတ္တာ လို့ စိတ်ကူးကြည့်နိုင်ပါတယ်။

````python
def add(a, b):
    return a + b

x = add(2, 3)               # add(2, 3) -> 5 ဆိုတော့ ဒီလိုဖြစ်သွားတယ်: x = 5
y = add(10, 20)             # add(10, 20) -> 30 ဆိုတော့ ဒီလိုဖြစ်သွားတယ်: y = 30

print(x)
print(y)
````

### return တန်ဖိုးကို တွက်ချက်မှုမှာ သုံးခြင်း

`add(2, 3)` က `5` ဖြစ်သွားတဲ့အတွက် ဂဏန်း `5` ကို သုံးလို့ရတဲ့ ဘယ်နေရာမှာမဆို သုံးလို့ရပါတယ်။

```python
add(2, 3) + add(3, 4)
```

| အဆင့် | Python ဘာမြင်လဲ |
|---|---|
| 1 | `add(2, 3) + add(3, 4)` |
| 2 | `add(2, 3)` run ပြီး 5 ကို return လုပ်တယ် → **`5`** `+ add(3, 4)` |
| 3 | `add(3, 4)` run ပြီး 7 ကို return လုပ်တယ် → `5 +` **`7`** |
| 4 | `12` |

````python
def add(a, b):
    return a + b

total = add(2, 3) + add(3, 4)   # 5 + 7 ဖြစ်သွားတယ်
print(total)                    # 12
````

````python
def add(a, b):
    return a + b

print(add(2, 3) * 10)           # add(2, 3) က 5 ဖြစ်သွားတယ် -> print(5 * 10) -> 50
print(add(2, 3) > 4)            # add(2, 3) က 5 ဖြစ်သွားတယ် -> print(5 > 4) -> True
````

### ခေါ်တာကို print() ထဲမှာ တိုက်ရိုက် ရေးခြင်း

`print(add(2, 3))` လည်း ဒီနည်းအတိုင်း အလုပ်လုပ်ပါတယ်:

| အဆင့် | Python ဘာမြင်လဲ |
|---|---|
| 1 | `print(add(2, 3))` |
| 2 | `add(2, 3)` က 5 ကို return လုပ်တယ် → `print(5)` |
| 3 | `5` print ထွက်တယ် |

````python
def add(a, b):
    return a + b

print(add(2, 3))                # print(5) ဖြစ်သွားတယ်
````

### return တန်ဖိုးကို နောက် function တစ်ခုဆီ ပေးပို့ခြင်း

return တန်ဖိုးကို function ဆီ ထပ်ပေးပို့လို့ ရပါတယ်။ Python က အမြဲ **အတွင်းကနေ အပြင်ကို** အလုပ်လုပ်ပါတယ်: အတွင်းက ခေါ်တာ အရင် အစားထိုးခံရပါတယ်။

```python
add(add(1, 2), 3)
```

| အဆင့် | Python ဘာမြင်လဲ |
|---|---|
| 1 | `add(add(1, 2), 3)` |
| 2 | အတွင်းက `add(1, 2)` က 3 ကို return လုပ်တယ် → `add(` **`3`** `, 3)` |
| 3 | `add(3, 3)` က 6 ကို return လုပ်တယ် → **`6`** |

````python
def add(a, b):
    return a + b

result = add(add(1, 2), 3)      # add(1, 2) -> 3၊ ပြီးရင် add(3, 3) -> 6
print(result)
````

### return ပါတဲ့ နောက်ထပ် ဥပမာများ

````python
def double(number):
    return number * 2

a = double(4)                   # a = 8 ဖြစ်သွားတယ်
b = double(a)                   # a က 8 ဆိုတော့ double(8) -> 16
print(a)
print(b)
````

````python
# string ကို return လုပ်တဲ့ function
def make_greeting(name):
    return "Hello, " + name     # + က string နှစ်ခုကို ဆက်တယ်

message = make_greeting("Anna")  # message = "Hello, Anna" ဖြစ်သွားတယ်
print(message)
````

````python
# လျှော့ဈေးပြီးနောက် ဈေးနှုန်း
def discount_price(price):
    return price - 1000

coffee = discount_price(4500)   # coffee = 3500 ဖြစ်သွားတယ်
cake = discount_price(6000)     # cake = 5000 ဖြစ်သွားတယ်

print("Coffee:", coffee)
print("Cake:", cake)
print("Total:", coffee + cake)  # return တန်ဖိုးတွေနဲ့ တွက်လို့ရတယ်
````

### print နဲ့ return: ဘာကွာလဲ?

အခု နီးပါး တူတဲ့ function နှစ်ခု ဖန်တီးကြည့်ရအောင်။ တစ်ခုက `print`၊ နောက်တစ်ခုက `return` သုံးပါတယ်။

```python
def show_add(a, b):        # print ကို သုံးတယ်
    print(a + b)

def add(a, b):             # return ကို သုံးတယ်
    return a + b
```

| | function ထဲက `print` | function ထဲက `return` |
|---|---|---|
| ဘာလုပ်လဲ | ရလဒ်ကို screen ပေါ်မှာ **ပြ** တယ် | ရလဒ်ကို ခေါ်ခဲ့တဲ့ နေရာဆီ **ပြန်ပေး** တယ် |
| ရလဒ်ကို variable ထဲ သိမ်းလို့ရလား? | ❌ | ✅ |
| ရလဒ်နဲ့ တွက်လို့ရလား? | ❌ | ✅ |
| ဘာနဲ့ တူလဲ... | ပေါင်မုန့်ကို ပြပြီး ကိုယ့်ဘာသာ သိမ်းထားတဲ့ သူငယ်ချင်း | ပေါင်မုန့်ကို သင့်လက်ထဲ ပေးတဲ့ သူငယ်ချင်း |

### return မပါရင် ဘာပြန်လာလဲ?

Function မှာ `return` မပါရင်တောင် တစ်ခုခု ပြန်ပေးပါတယ်: **`None`** လို့ ခေါ်တဲ့ အထူး တန်ဖိုး။ `None` ရဲ့ အဓိပ္ပာယ်က **"ဘာမှ မရှိ"** ပါ။

ဒါကြောင့် `show_add(2, 3)` က `5` မဟုတ်ဘဲ `None` နဲ့ အစားထိုးခံရပါတယ်။

````python
def show_add(a, b):
    print(a + b)                # ရလဒ်ကို ပြရုံပဲ။ return မရှိဘူး

result = show_add(2, 3)         # 5 က screen ပေါ် print ထွက်တယ်... ဒါပေမဲ့ result ထဲ ဘာသိမ်းလဲ?
print("result is:", result)     # None! show_add(2, 3) က 5 မဟုတ်ဘဲ None နဲ့ အစားထိုးခံရတယ်
````

Output ကို သေချာ ကြည့်ပါ:

1. `5` - **function ကိုယ်တိုင်** (အထဲက `print`) print လုပ်တာ
2. `result is: None` - function က ဘာမှ return မလုပ်လို့ `result` က `None` ရတယ်

ဂဏန်း 5 က screen ပေါ် ပေါ်လာပေမဲ့ **ကျွန်တော်တို့ ပြန်မရခဲ့ပါဘူး**။ ပျောက်သွားပါပြီ။

### ဒါကြောင့် add(2, 3) + add(3, 4) က print နဲ့ အလုပ်မလုပ်ဘူး

`show_add(2, 3) + show_add(3, 4)` ကို စမ်းရင် ဘာဖြစ်မလဲ?

| အဆင့် | Python ဘာမြင်လဲ |
|---|---|
| 1 | `show_add(2, 3) + show_add(3, 4)` |
| 2 | `show_add(2, 3)` က 5 ကို print လုပ်ပြီး **`None`** နဲ့ အစားထိုးခံရတယ် |
| 3 | `show_add(3, 4)` က 7 ကို print လုပ်ပြီး **`None`** နဲ့ အစားထိုးခံရတယ် |
| 4 | `None + None` → **error!** Python က "ဘာမှ မရှိ" နဲ့ "ဘာမှ မရှိ" ကို ပေါင်းလို့ မရဘူး |

````python
def show_add(a, b):
    print(a + b)

# total = show_add(2, 3) + show_add(3, 4)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'NoneType'
````

(အပေါ်က စာကြောင်းကို notebook မရပ်သွားအောင် comment လုပ်ထားပါတယ်။ `total = ...` စာကြောင်းရဲ့ `#` နှစ်ခုကို ဖျက်ရင် `5` နဲ့ `7` အရင် print ထွက်ပြီးမှ error ဖြစ်တာကို မြင်ရပါလိမ့်မယ်။)

Error message ရဲ့ အဓိပ္ပာယ်က: "`None` နဲ့ `None` ကို `+` သုံးလို့ မရဘူး"။ (`NoneType` က `None` ရဲ့ data type ပါ။)

### ဘယ်အချိန် ဘာကို သုံးမလဲ?

- Screen ပေါ်မှာ တစ်ခုခု **ပြ** ရုံပဲ လိုချင်ရင် → `print` ရပါတယ်
- ရလဒ်ကို နောက်မှ **သုံး** (သိမ်း၊ တွက်၊ နောက် function ဆီ ပေးပို့) ချင်ရင် → `return` ကို သုံးပါ

များသောအားဖြင့် တစ်ခုခု တွက်ချက်တဲ့ function က `return` ကို သုံးသင့်ပါတယ်။ ပြီးရင် ရလဒ်ကို ကြည့်ချင်တဲ့အခါ `print(add(2, 3))` နဲ့ print လုပ်လို့ အမြဲ ရပါတယ်။

````python
def add(a, b):
    return a + b

# return ပါရင် နှစ်ခုလုံး လုပ်လို့ရတယ်: ရလဒ်ကို သုံး ပြီး print လည်း လုပ်
total = add(2, 3) + add(3, 4)
print(total)
````

## 5. အဖြစ်များတဲ့ အမှားများ

- ❌ **မှားတဲ့ code**
- 🔍 **ဘာမှားသွားလဲ**
- ✅ **မှန်ကန်တဲ့ နည်း**

Error ဖြစ်စေတဲ့ code ကို notebook မရပ်သွားအောင် comment (ရှေ့မှာ `#`) အဖြစ် ရေးထားပါတယ်။ တကယ့် error message ကို မြင်ချင်ရင် `#` ကို ဖျက်ပြီး cell ကို run ပါ၊ ပြီးရင် `#` ကို ပြန်ထည့်ပါ။

### အမှား 1: Function ဖန်တီးပြီး သူ့ဘာသာ run မယ်လို့ ထင်ခြင်း

````python
# ❌ မှားပါတယ်: "Good morning!" ကို print လုပ်ချင်တယ်
def morning():
    print("Good morning!")

# ဘာမှ print မထွက်ဘူး...
````

🔍 **ဘာမှားသွားလဲ?**

- `def` က function ကို **ဖန်တီး** ရုံပါပဲ။ run မလုပ်ပါဘူး။ (ဟင်းချက်နည်း ရေးတာက ဟင်း ဖြစ်မလာဘူး။)
- Error မပြတဲ့အတွက် ရှုပ်ထွေးနိုင်ပါတယ်: "ဘာလို့ ဘာမှ မဖြစ်တာလဲ?"

- Error မပြပေမဲ့ function က မ run ပါဘူး။

✅ **မှန်ကန်တဲ့ နည်း:** ခေါ်တိုင်း နာမည်နောက်မှာ `()` ထည့်ပါ။

### အမှား 5: တန်ဖိုး အရေအတွက် မှားပေးခြင်း

````python
# ❌ မှားပါတယ်
def add(a, b):
    return a + b

# add(2)          # TypeError: add() missing 1 required positional argument: 'b'
# add(2, 3, 4)    # TypeError: add() takes 2 positional arguments but 3 were given
````

🔍 **ဘာမှားသွားလဲ?**

- `add` ကို parameter **2** ခု (`a` နဲ့ `b`) နဲ့ ဖန်တီးထားလို့ တန်ဖိုး **2** ခု အတိအကျ လိုပါတယ်။
- `add(2)`: တန်ဖိုး 1 ခုပဲ ပေးထားတယ်။ `a` က 2 ရပေမဲ့ `b` က ဘာမှ မရဘူး။ Error က "တန်ဖိုး 1 ခု လိုနေတယ်: `b`" လို့ ပြောတယ်။
- `add(2, 3, 4)`: တန်ဖိုး 3 ခု ပေးထားပေမဲ့ နေရာ 2 ခုပဲ ရှိတယ်။ Error က "2 ခု လက်ခံတယ်၊ ဒါပေမဲ့ 3 ခု ပေးထားတယ်" လို့ ပြောတယ်။

✅ **မှန်ကန်တဲ့ နည်း:** Parameter အရေအတွက်နဲ့ အတိအကျ တူတဲ့ တန်ဖိုးတွေ ပေးပါ။

````python
# ✅ မှန်ပါတယ်
def add(a, b):
    return a + b

print(add(2, 3))        # parameter 2 ခု၊ တန်ဖိုး 2 ခု
````

### အမှား 6: return လိုတဲ့နေရာမှာ print သုံးခြင်း

````python
# ❌ မှားပါတယ်: ရလဒ်ကို သိမ်းပြီး နောက်မှ သုံးချင်တယ်
def multiply(a, b):
    print(a * b)                # ရလဒ်ကို ပြရုံပဲ

answer = multiply(3, 4)         # 12 print ထွက်တယ်... ဒါပေမဲ့ answer က None
print("answer:", answer)
````

🔍 **ဘာမှားသွားလဲ?**

- `multiply` က `return` မဟုတ်ဘဲ `print` ကို သုံးတယ်။ 12 ကို screen ပေါ်မှာ **ပြ** ပေမဲ့ **ပြန်မပေး** ဘူး။
- `return` မပါတဲ့ function က `None` ကို ပြန်ပေးတယ်။ ဒါကြောင့် `multiply(3, 4)` က `None` နဲ့ အစားထိုးခံရပြီး `answer` က `None` ဖြစ်တယ်။
- အဲဒါနဲ့ တွက်ဖို့ ကြိုးစားရင် (ဥပမာ `answer + 1`) error ဖြစ်ပါလိမ့်မယ်။

✅ **မှန်ကန်တဲ့ နည်း:** ရလဒ်ကို ပြန်ပေးဖို့ `return` ကို သုံးပါ။ ကြည့်ချင်ရင် function အပြင်မှာ print လုပ်ပါ။

````python
# ✅ မှန်ပါတယ်
def multiply(a, b):
    return a * b                # ရလဒ်ကို ပြန်ပေးတယ်

answer = multiply(3, 4)         # answer = 12 ဖြစ်သွားတယ်
print("answer:", answer)
print("answer + 1:", answer + 1)
````

### အမှား 7: return လုပ်ထားပေမဲ့ ရလဒ်ကို မသုံးခြင်း

````python
# ❌ မှားပါတယ်: ရလဒ်ကို မြင်ချင်တယ်
def multiply(a, b):
    return a * b

multiply(3, 4)                  # ရလဒ် (12) ပြန်လာတယ်... ဒါပေမဲ့ ဘယ်သူမှ မဖမ်းထားဘူး!
print("Where is the result?")
````

🔍 **ဘာမှားသွားလဲ?**

- သူငယ်ချင်းက ပေါင်မုန့် ပေးတဲ့အခါ ကြမ်းပြင်ပေါ် ပြုတ်ကျသွားသလိုပါပဲ။
- Error မပြပေမဲ့ ရလဒ် ပျောက်သွားပါပြီ။

✅ **မှန်ကန်တဲ့ နည်း:** Return တန်ဖိုးကို ဖမ်းယူပါ: variable ထဲ သိမ်း၊ ဒါမှမဟုတ် ခေါ်တာကို `print()` ထဲမှာ ရေးပါ။

````python
# ✅ မှန်ပါတယ်
def multiply(a, b):
    return a * b

result = multiply(3, 4)         # သိမ်းထားတယ်
print(result)

print(multiply(3, 4))           # ဒါမှမဟုတ် တိုက်ရိုက် print လုပ်တယ်
````
