## 1. List là gì?

Ở Bài 1, chúng ta đã học rằng biến giống như một chiếc hộp chứa **một** giá trị.

Nhưng nếu bạn muốn lưu điểm thi của 5 học sinh thì sao? Với những gì đã học, bạn sẽ cần 5 biến:

```python
score1 = 85
score2 = 92
score3 = 78
score4 = 90
score5 = 66
```

Như vậy đã nhiều rồi. Còn nếu có 30 học sinh? Hay 300 học sinh?

**List (danh sách)** giải quyết vấn đề này. List là một biến duy nhất có thể chứa **nhiều giá trị theo thứ tự**.

Bạn có thể hình dung list giống như một dãy **tủ đồ (locker)**:
- Cả dãy tủ có một cái tên (tên của list).
- Mỗi ngăn tủ chứa một giá trị.
- Mỗi ngăn tủ có một con số, để bạn có thể tìm thấy nó.

Cách tạo list:
- Đặt các giá trị bên trong dấu ngoặc vuông `[ ]`.
- Ngăn cách các giá trị bằng dấu phẩy `,`.

```python
scores = [85, 92, 78, 90, 66]
```

````python
# Một list các số nguyên
# Một biến (scores) giờ chứa 5 giá trị
scores = [85, 92, 78, 90, 66]
print(scores)           # print() hiển thị toàn bộ list, bao gồm cả dấu ngoặc vuông
````

````python
# List có thể chứa bất kỳ kiểu dữ liệu nào bạn đã học ở Bài 1
fruits = ["apple", "banana", "cherry"]      # Một list các chuỗi (str)
heights = [175.5, 162.3, 180.1]             # Một list các số thực (float)
answers = [True, False, True, True]         # Một list các giá trị bool

print(fruits)
print(heights)
print(answers)
````

````python
# Có thể trộn nhiều kiểu dữ liệu khác nhau trong một list
# Điều này được phép, nhưng thường thì list chỉ chứa một loại giá trị sẽ dễ hiểu hơn
student = ["Bill", 20, 175.5, True]         # tên, tuổi, chiều cao, có phải học sinh không
print(student)
````

````python
# List cũng có thể rỗng (không có giá trị nào)
# Cách này hữu ích khi bạn muốn bắt đầu từ con số không rồi thêm giá trị sau (chúng ta sẽ học cách làm ở phần 5)
empty_list = []
print(empty_list)
````

````python
# List là một kiểu dữ liệu mới: list
# type() cho biết kiểu dữ liệu của một giá trị
print(type(scores))     # <class 'list'> nghĩa là "đây là một list"
print(type(85))         # <class 'int'> nghĩa là "đây là một số nguyên" (một trong các giá trị bên trong list)
````

## 2. Chỉ số (index): tìm giá trị theo vị trí

Mỗi giá trị trong list có một số chỉ vị trí. Số này được gọi là **chỉ số (index)**. Hãy hình dung nó như con số được ghi trên mỗi ngăn tủ.

Điều quan trọng nhất cần nhớ: **Python bắt đầu đếm từ 0, không phải từ 1.**

Vì vậy giá trị thứ nhất ở chỉ số 0, giá trị thứ hai ở chỉ số 1, giá trị thứ ba ở chỉ số 2, và cứ thế tiếp tục. (Bạn còn nhớ `range(5)` ở Bài 3 không? Nó cũng bắt đầu từ 0: 0, 1, 2, 3, 4.)

```python
fruits = ["apple", "banana", "cherry", "grape", "melon"]
```

| Giá trị | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Chỉ số (index) | 0 | 1 | 2 | 3 | 4 |
| Vị trí (cách con người đếm) | thứ 1 | thứ 2 | thứ 3 | thứ 4 | thứ 5 |

Một cách dễ nhớ: **chỉ số = vị trí - 1**. Giá trị thứ 1 có chỉ số 0, giá trị thứ 5 có chỉ số 4.

Để lấy một giá trị, hãy viết tên list rồi đến chỉ số trong dấu ngoặc vuông:

```python
fruits[0]     # "giá trị ở chỉ số 0 trong fruits"
```

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Lấy giá trị theo chỉ số
print(fruits[0])        # Giá trị thứ 1 ở chỉ số 0 -> apple
print(fruits[1])        # Giá trị thứ 2 ở chỉ số 1 -> banana
print(fruits[2])        # Giá trị thứ 3 ở chỉ số 2 -> cherry
print(fruits[3])        # Giá trị thứ 4 ở chỉ số 3 -> grape
print(fruits[4])        # Giá trị thứ 5 (cuối cùng) ở chỉ số 4 -> melon
````

````python
# Điều tương tự với một list các số
# Cẩn thận đừng nhầm lẫn giữa chỉ số (vị trí) và giá trị (thứ chứa bên trong)
numbers = [10, 20, 30, 40]
#  chỉ số:  0   1   2   3

print(numbers[0])       # chỉ số 0 -> giá trị là 10
print(numbers[2])       # chỉ số 2 -> giá trị là 30 (không phải 2!)
print(numbers[3])       # chỉ số 3 -> giá trị là 40
````

````python
# Các giá trị không cần phải theo thứ tự đặc biệt nào
# Chỉ số chỉ cho bạn biết giá trị nằm Ở ĐÂU, không cho biết nó lớn hay nhỏ
scores = [85, 92, 78, 90, 66]
#  chỉ số:  0   1   2   3   4

print(scores[1])        # Điểm của học sinh thứ 2 -> 92
print(scores[4])        # Điểm của học sinh thứ 5 -> 66
````

````python
# Bạn có thể lấy cùng một chỉ số bao nhiêu lần tùy thích
# Lấy giá trị chỉ là ĐỌC nó. List không thay đổi
colors = ["red", "green", "blue"]

print(colors[1])        # green
print(colors[1])        # vẫn là green
print(colors)           # List vẫn giống như trước
````

````python
# Giá trị lấy ra từ list có thể được dùng giống như một biến bình thường
scores = [85, 92, 78, 90, 66]

first = scores[0]                  # Lưu điểm đầu tiên (85) vào một biến mới
print(first)

print(scores[0] + scores[1])       # 85 + 92 = 177 (bạn có thể tính toán với chúng)
print(scores[1] > scores[2])       # 92 > 78 là True (bạn có thể so sánh chúng, giống như ở Bài 2)
````

````python
# Các chuỗi lấy ra từ list có thể nối với nhau bằng +, giống như ở Bài 1
words = ["Hello", "nice", "to", "meet", "you"]

print(words[0] + ", " + words[1] + " " + words[2] + " " + words[3] + " " + words[4])
````

````python
# Chỉ số cũng có thể là một biến
# Ở đây, i chứa 2, nên scores[i] giống với scores[2]
scores = [85, 92, 78, 90, 66]

i = 2
print(scores[i])        # scores[2] -> 78

i = 0
print(scores[i])        # Bây giờ i chứa 0, nên scores[0] -> 85
````

````python
# Chỉ số thậm chí có thể là một phép tính
# Python tính con số bên trong [ ] trước, rồi mới tìm chỉ số đó
scores = [85, 92, 78, 90, 66]

print(scores[1 + 2])    # 1 + 2 = 3 -> scores[3] -> 90

i = 1
print(scores[i + 1])    # i + 1 = 2 -> scores[2] -> 78
````

### Chỉ số âm: đếm từ cuối

Bây giờ bạn đã biết chỉ số thông thường (chỉ số dương), hãy xem thêm một cách nữa để tìm giá trị.

Python còn cho phép bạn **đếm ngược từ cuối** list, bằng các số âm.

- `-1` là giá trị **cuối cùng**
- `-2` là giá trị kế cuối
- `-3` là giá trị thứ ba tính từ cuối, và cứ thế tiếp tục

| Giá trị | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Chỉ số (index) | 0 | 1 | 2 | 3 | 4 |
| Chỉ số âm | -5 | -4 | -3 | -2 | -1 |

Hãy chú ý rằng chỉ số âm bắt đầu từ **-1**, không phải -0. (Không có -0, vì -0 chính là 0, tức là giá trị đầu tiên.)

**Tại sao lại dùng chỉ số âm?**

Lý do chính là để lấy **giá trị cuối cùng một cách nhanh chóng**.

Với chỉ số dương, để lấy giá trị cuối cùng, trước tiên bạn cần biết list có bao nhiêu giá trị. Nếu list có 5 giá trị, chỉ số cuối là 4. Nếu có 100 giá trị, chỉ số cuối là 99. Và nếu bạn thêm hoặc xóa giá trị, chỉ số cuối cũng thay đổi theo.

Với chỉ số âm, bạn không cần đếm gì cả. **`-1` luôn là giá trị cuối cùng**, dù list dài bao nhiêu đi nữa.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Chỉ số âm: đếm từ cuối
print(fruits[-1])       # -1 luôn là giá trị cuối cùng -> melon
print(fruits[-2])       # -2 là giá trị kế cuối -> grape
print(fruits[-3])       # -3 là giá trị thứ ba tính từ cuối -> cherry
````

````python
# Một chỉ số dương và một chỉ số âm có thể chỉ đến cùng một giá trị
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[4])        # Giá trị cuối cùng, dùng chỉ số dương -> melon
print(fruits[-1])       # Giá trị cuối cùng, dùng chỉ số âm -> melon (giống nhau!)

print(fruits[0])        # Giá trị đầu tiên, dùng chỉ số dương -> apple
print(fruits[-5])       # Giá trị đầu tiên, dùng chỉ số âm -> apple (giống nhau!)
````

````python
# Tại sao -1 hữu ích: nó dùng được cho MỌI list, dù dài bao nhiêu
# Với chỉ số dương, chỉ số cuối của mỗi list lại khác nhau (2, 6, 0 ...)
# Với -1, nó luôn giống nhau
short_list = ["a", "b", "c"]
long_list = [10, 20, 30, 40, 50, 60, 70]
one_item = ["only"]

print(short_list[-1])   # c       (chỉ số dương sẽ là 2)
print(long_list[-1])    # 70      (chỉ số dương sẽ là 6)
print(one_item[-1])     # only    (chỉ số dương sẽ là 0)
````

## 3. Thay đổi giá trị và len()

### Thay đổi một giá trị

List không cố định. Bạn có thể **thay đổi** một giá trị bên trong nó bất cứ lúc nào. Giống như đặt một món đồ mới vào một ngăn tủ:

```python
tên_list[chỉ_số] = giá_trị_mới
```

Cách này hoạt động giống hệt như lưu một giá trị vào biến (`=` nghĩa là "lưu"), nhưng chỉ có giá trị ở đúng vị trí đó bị thay thế. Các giá trị khác vẫn giữ nguyên.

````python
# Thay đổi giá trị ở một chỉ số cụ thể
fruits = ["apple", "banana", "cherry"]
print(fruits)

fruits[1] = "orange"        # Thay giá trị ở chỉ số 1 ("banana") bằng "orange"
print(fruits)               # Chỉ có chỉ số 1 thay đổi. "apple" và "cherry" vẫn như trước
````

````python
# Chỉ số âm cũng dùng được để thay đổi giá trị
fruits = ["apple", "banana", "cherry"]

fruits[-1] = "grape"        # Thay giá trị cuối cùng ("cherry") bằng "grape"
print(fruits)
````

````python
# Thay đổi giá trị dựa trên giá trị cũ của nó
# Cùng ý tưởng với age = age + 1 ở Bài 1
scores = [85, 92, 78]

scores[2] = scores[2] + 5   # Học sinh thứ ba được cộng thêm 5 điểm (78 + 5 = 83)
print(scores)
````

### len(): list có bao nhiêu giá trị?

`len()` cho bạn biết list có bao nhiêu giá trị. `len` là viết tắt của **length** (độ dài).

```python
len(tên_list)
```

Đặt list vào bên trong dấu ngoặc tròn, và `len()` trả về một con số: số lượng giá trị trong list đó.

````python
# len(): đếm xem list có bao nhiêu giá trị
fruits = ["apple", "banana", "cherry", "grape", "melon"]
print(len(fruits))          # Có 5 giá trị, nên là 5
````

````python
# Các list có độ dài khác nhau
three = ["a", "b", "c"]
five = [1, 2, 3, 4, 5]
one = ["only"]
empty_list = []

print(len(three))           # 3 giá trị -> 3
print(len(five))            # 5 giá trị -> 5
print(len(one))             # 1 giá trị -> 1
print(len(empty_list))      # List rỗng có 0 giá trị -> 0
````

````python
# len() hoạt động giống nhau với mọi loại list
names = ["Bill", "Anna", "Tom", "Kate"]          # Một list các chuỗi
numbers = [3, 7, 1, 9, 4, 6]                     # Một list các số nguyên
answers = [True, False, True]                    # Một list các giá trị bool

print(len(names))           # 4
print(len(numbers))         # 6
print(len(answers))         # 3
````

````python
# len() đếm xem có BAO NHIÊU giá trị. Nó không cộng các giá trị lại
# Các giá trị rất lớn, nhưng chỉ có 3 giá trị
prices = [1000, 2500, 3000]
print(len(prices))          # 3 (không phải 6500!)
````

````python
# Kết quả của len() chỉ là một con số (int), nên bạn có thể lưu nó vào biến
students = ["Bill", "Anna", "Tom", "Kate", "Mike"]

count = len(students)       # Lưu 5 vào biến count
print(count)
print("There are", count, "students.")
````

````python
# len() cũng dùng được với chuỗi: nó đếm số ký tự
# (Dấu cách cũng được tính là một ký tự)
print(len("apple"))         # a, p, p, l, e -> 5
print(len("hi there"))      # h, i, (dấu cách), t, h, e, r, e -> 8
````

### Chỉ số cuối cùng là len() - 1

Bây giờ chúng ta có thể kết nối `len()` với chỉ số.

Nếu list có 5 giá trị, `len()` cho ra 5, nhưng chỉ số cuối cùng là **4**. Vì chỉ số bắt đầu từ 0, **chỉ số cuối cùng luôn là `len() - 1`**.

Vì vậy có hai cách để lấy giá trị cuối cùng:
- `fruits[len(fruits) - 1]`: đếm số giá trị, trừ đi 1, rồi dùng chỉ số đó
- `fruits[-1]`: chỉ cần dùng -1

Cả hai cho ra cùng một giá trị, nhưng `-1` ngắn và dễ hơn nhiều. Đây chính là lý do chỉ số âm rất hữu ích.

````python
# Chỉ số cuối cùng là len() - 1
fruits = ["apple", "banana", "cherry", "grape", "melon"]

last_index = len(fruits) - 1       # 5 - 1 = 4
print(last_index)
print(fruits[last_index])          # fruits[4] -> melon
````

````python
# So sánh hai cách lấy giá trị cuối cùng
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[len(fruits) - 1])     # Cách dài: len(fruits) là 5, 5 - 1 = 4, fruits[4] -> melon
print(fruits[-1])                  # Cách ngắn: -1 là giá trị cuối cùng -> melon (kết quả giống nhau!)
````

## 4. Kiểm tra một giá trị có nằm trong list không (in)

Đôi khi bạn muốn biết: **"Giá trị này có nằm trong list không?"**

Để làm điều này, chúng ta dùng `in`.

```python
giá_trị in tên_list
```

- Nếu giá trị nằm trong list, kết quả là `True`.
- Nếu không, kết quả là `False`.

Kết quả là `True` hoặc `False`, giống như các toán tử so sánh ở Bài 2. Điều đó có nghĩa là bạn có thể dùng nó trực tiếp làm điều kiện của `if`!

Để kiểm tra một giá trị **không** nằm trong list, hãy dùng `not in`.

**Lưu ý:** Bạn cũng đã thấy từ `in` ở Bài 3 (`for i in range(5)`). `in` đó là một phần của vòng lặp `for`. `in` mà chúng ta học ở đây thì khác: nó **kiểm tra** xem một giá trị có nằm trong list không, và cho ra True hoặc False.

````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)       # "banana" có trong list -> True
print("grape" in fruits)        # "grape" không có trong list -> False
````

````python
# not in: ngược lại với in
print("grape" not in fruits)    # "grape" KHÔNG có trong list -> True
print("apple" not in fruits)    # "apple" CÓ trong list, nên "not in" là False
````

````python
# Dùng in làm điều kiện của if (Bài 2)
fruits = ["apple", "banana", "cherry"]
wanted = "banana"

if wanted in fruits:                    # "banana" in fruits là True
    print("Yes, we have it!")
else:
    print("Sorry, it's sold out.")
````

````python
# Một ví dụ thực tế: kiểm tra danh sách khách mời
guests = ["Bill", "Anna", "Tom"]
name = "Kate"

if name in guests:                      # "Kate" in guests là False
    print("Welcome!")
else:
    print("Sorry, you are not on the list.")    # Nên dòng này chạy
````

````python
# in kiểm tra giá trị một cách chính xác. Chữ hoa và chữ thường là khác nhau
fruits = ["apple", "banana", "cherry"]
print("Apple" in fruits)        # "Apple" (chữ A hoa) và "apple" là hai chuỗi khác nhau -> False
````

## 5. Thêm và xóa giá trị

List có thể dài ra và ngắn lại. Bạn có thể thêm giá trị mới và xóa những giá trị không cần nữa.

| Cách viết | Tác dụng |
|---|---|
| `tên_list.append(giá_trị)` | Thêm giá trị vào **cuối** list |
| `tên_list.insert(chỉ_số, giá_trị)` | Thêm giá trị vào **vị trí bạn chọn** |
| `tên_list.remove(giá_trị)` | Xóa giá trị bạn chỉ định (giá trị đầu tiên tìm thấy) |
| `tên_list.pop()` | Lấy ra giá trị **cuối cùng** |
| `tên_list.pop(chỉ_số)` | Lấy ra giá trị ở **vị trí bạn chọn** |

Dấu chấm `.` có nghĩa là "hãy làm việc này với list này". Ví dụ, `fruits.append("grape")` có nghĩa là "thêm "grape" vào list fruits". Những lệnh được viết sau dấu chấm như thế này được gọi là **phương thức (method)**.

Các phương thức này thay đổi chính list đó. Bạn không cần viết `fruits = ...`.

````python
# append(): thêm một giá trị vào cuối
fruits = ["apple", "banana"]
print(fruits)

fruits.append("cherry")         # "cherry" được thêm vào cuối
print(fruits)

fruits.append("grape")          # "grape" lại được thêm vào cuối
print(fruits)
print(len(fruits))              # List đã tăng từ 2 lên 4 giá trị
````

````python
# Bắt đầu với một list rỗng và thêm từng giá trị một
# Đây là một mẫu rất thường gặp
shopping = []                   # Bắt đầu với một list rỗng
shopping.append("milk")
shopping.append("eggs")
shopping.append("bread")
print(shopping)
````

````python
# insert(): thêm một giá trị vào một vị trí cụ thể
# insert(chỉ_số, giá_trị) -> giá trị mới vào đúng chỉ số đó, và các giá trị phía sau lùi về sau một vị trí
fruits = ["apple", "banana", "cherry"]

fruits.insert(1, "orange")      # Đặt "orange" vào chỉ số 1
print(fruits)                   # "banana" và "cherry" mỗi cái đều lùi về sau một vị trí

fruits.insert(0, "melon")       # Đặt "melon" vào chỉ số 0 (đầu tiên)
print(fruits)
````

````python
# remove(): xóa một giá trị bằng cách gọi tên chính giá trị đó
fruits = ["apple", "banana", "cherry"]

fruits.remove("banana")         # Tìm "banana" và xóa nó
print(fruits)                   # Các giá trị phía sau tiến lên để lấp chỗ trống
````

````python
# Nếu cùng một giá trị xuất hiện nhiều lần, remove() chỉ xóa giá trị ĐẦU TIÊN
numbers = [1, 2, 3, 2, 1]

numbers.remove(2)               # Chỉ số 2 đầu tiên (ở chỉ số 1) bị xóa
print(numbers)                  # Số 2 thứ hai vẫn còn đó
````

````python
# pop(): lấy ra một giá trị theo vị trí của nó
# Không có chỉ số, pop() lấy ra giá trị CUỐI CÙNG
fruits = ["apple", "banana", "cherry"]

fruits.pop()                    # Lấy ra giá trị cuối cùng ("cherry")
print(fruits)
````

````python
# pop(chỉ_số): lấy ra giá trị ở chỉ số đó
fruits = ["apple", "banana", "cherry"]

fruits.pop(0)                   # Lấy ra giá trị ở chỉ số 0 ("apple")
print(fruits)
````

````python
# pop() còn TRẢ LẠI giá trị mà nó lấy ra, nên bạn có thể lưu nó vào một biến
# remove() chỉ xóa. pop() vừa xóa VỪA đưa giá trị đó cho bạn
tasks = ["wash dishes", "do homework", "clean room"]

done = tasks.pop(0)             # Lấy ra công việc đầu tiên và lưu vào done
print("Finished:", done)
print("Still to do:", tasks)
````

````python
# So sánh nhanh remove() và pop()
# remove(giá_trị): "xóa GIÁ TRỊ NÀY"              -> bạn biết giá trị
# pop(chỉ_số):     "xóa thứ đang ở VỊ TRÍ NÀY"   -> bạn biết vị trí
colors = ["red", "green", "blue"]

colors.remove("green")          # Xóa theo giá trị
print(colors)

colors.pop(0)                   # Xóa theo vị trí (chỉ số 0 lúc này là "red")
print(colors)
````

## 6. Những lỗi thường gặp

Đây là những lỗi mà hầu như ai cũng mắc phải khi học về list. Nếu gặp lỗi, hãy kiểm tra những điểm này trước.

1. **Quên rằng chỉ số bắt đầu từ 0**: `fruits[1]` là giá trị **thứ hai**, không phải thứ nhất.
2. **Dùng chỉ số không tồn tại**: Nếu list có 3 giá trị, các chỉ số là 0, 1, 2. `fruits[3]` gây ra lỗi `IndexError: list index out of range` ("chỉ số nằm ngoài phạm vi của list").
3. **Cố thêm giá trị bằng một chỉ số mới**: `fruits[3] = "grape"` với list có 3 giá trị **không** thêm được giá trị. Nó gây ra lỗi `IndexError: list assignment index out of range` ("không thể lưu giá trị vào chỉ số nằm ngoài list"). Để thêm, hãy dùng `append()`.
4. **Xóa một giá trị không có trong list**: `fruits.remove("grape")` gây ra lỗi `ValueError: list.remove(x): x not in list` ("giá trị không có trong list"). Hãy kiểm tra bằng `in` trước.
5. **Nhầm lẫn `( )` và `[ ]`**: Dùng `[ ]` để tạo list và để lấy giá trị theo chỉ số (`fruits[0]`). Dùng `( )` cho các lệnh như `len(fruits)` và `fruits.append("grape")`.

Các ô bên dưới cho thấy mỗi lỗi trông như thế nào. Mã bị lỗi được viết dưới dạng chú thích để không làm hỏng notebook. Hãy thử xóa dấu `#` ở dòng mã để xem thông báo lỗi thật, sau đó đặt lại dấu `#`.

````python
# Lỗi 1: quên rằng chỉ số bắt đầu từ 0
fruits = ["apple", "banana", "cherry"]

print(fruits[1])        # Nhiều người tưởng là "apple", nhưng đây là giá trị THỨ HAI -> banana
print(fruits[0])        # Đúng: giá trị đầu tiên ở chỉ số 0 -> apple
````

````python
# Lỗi 2: dùng chỉ số không tồn tại
fruits = ["apple", "banana", "cherry"]      # 3 giá trị -> các chỉ số là 0, 1, 2

# print(fruits[3])      # IndexError: list index out of range -> không có chỉ số 3!

print(fruits[2])                    # Đúng: chỉ số cuối cùng là 2
print(fruits[len(fruits) - 1])      # Cũng đúng: len(fruits) - 1 = 3 - 1 = 2
print(fruits[-1])                   # Cũng đúng: -1 luôn là giá trị cuối cùng
````

````python
# Lỗi 3: cố thêm giá trị bằng một chỉ số mới
fruits = ["apple", "banana", "cherry"]

# fruits[3] = "grape"   # IndexError: list assignment index out of range -> bạn chỉ có thể THAY ĐỔI giá trị ở những chỉ số đã tồn tại

fruits.append("grape")  # Đúng: dùng append() để thêm giá trị mới vào cuối
print(fruits)
````

````python
# Lỗi 4: xóa một giá trị không có trong list
fruits = ["apple", "banana", "cherry"]

# fruits.remove("grape")    # ValueError: list.remove(x): x not in list -> "grape" không có trong list!

# Đúng: kiểm tra bằng in trước (phần 4), và chỉ xóa nếu nó có trong list
if "grape" in fruits:
    fruits.remove("grape")
else:
    print("grape is not in the list, so nothing was removed.")
````

````python
# Lỗi 5: nhầm lẫn ( ) và [ ]
fruits = ["apple", "banana", "cherry"]

# print(fruits(0))          # TypeError: 'list' object is not callable -> ( ) dùng cho lệnh, không dùng cho chỉ số
# fruits.append["grape"]    # TypeError -> append là một lệnh, nên cần ( ), không phải [ ]

print(fruits[0])            # Đúng: [ ] để lấy giá trị theo chỉ số
fruits.append("grape")      # Đúng: ( ) cho lệnh
print(fruits)
````
