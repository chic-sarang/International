## 1. for давталт гэж юу вэ?

"Hello" гэж 5 удаа хэвлэх хэрэгтэй гэж бодоод үзээрэй. `print("Hello")`-г 5 удаа бичиж болно. Харин 100 удаа бол? Эсвэл 1000 удаа бол? Нэг мөрийг дахин дахин бичих нь ядаргаатай бөгөөд алдаа гаргахад амархан.

**Давталт (loop)** энэ асуудлыг шийддэг. Давталт нь нэг кодыг хүссэн удаагаараа дахин дахин ажиллуулдаг.

Python-д хамгийн түгээмэл давталт бол `for` давталт юм. Үндсэн хэлбэр нь ийм байна:

```python
for i in range(5):
    давтах код
```

Үүнийг ингэж уншиж болно: **"range(5) доторх i тоо бүрийн хувьд доорх кодыг ажиллуул"**.

- `range(5)` нь 0, 1, 2, 3, 4 гэсэн тоонуудыг өгнө (нийт 5 тоо). `range()`-ийг дараагийн хэсэгт дэлгэрэнгүй үзнэ.
- `i` бол хувьсагч. Давталт давтагдах бүрт дараагийн тоо `i`-д хадгалагдана.
- `if`-ийн нэгэн адил `for` мөр хоёр цэг `:`-ээр төгсөх ёстой.
- `if`-ийн нэгэн адил давтах код **догол мөртэй** (4 зай) байх ёстой. Зөвхөн догол мөртэй мөрүүд давтагдана.

````python
# print("Hello")-г 5 удаа давтах
# range(5) нь 5 тоо (0, 1, 2, 3, 4) өгөх тул догол мөртэй код 5 удаа ажиллана
for i in range(5):
    print("Hello")      # Энэ мөр for давталтын дотор (догол мөртэй) байгаа тул давтагдана
````

````python
# Удаа бүрт i-д юу хадгалагдаж байгааг харъя
# 1 дэх удаа: i = 0, 2 дахь удаа: i = 1, 3 дахь удаа: i = 2, 4 дэх удаа: i = 3, 5 дахь удаа: i = 4
for i in range(5):
    print(i)            # i-ийн одоогийн утгыг хэвлэнэ
````

````python
# Догол мөртэй кодыг нэг бүтэн гүйцэтгэхийг нэг "давталт" (давтамж) гэнэ
# Догол мөртэй мөр бүр давталт бүрт нэг удаа, дээрээс доош дарааллаар ажиллана
for i in range(3):
    print("Repetition number", i)     # print() нь таслалаар тусгаарласан хэд хэдэн утгыг нэг дор харуулж чадна
    print("----------")               # Энэ мөр ч давталтын дотор байна (ижил догол мөр)

print("The loop is finished.")        # Догол мөргүй тул давталтын гадна байна. Давталт дууссаны дараа зөвхөн нэг удаа ажиллана
````

````python
# Хувьсагчийн нэр заавал i байх албагүй
# i бол олон хүний хэрэглэдэг нэр л юм. Хувьсагчийн нэрийн дүрмийг баримталж байвал ямар ч нэр сонгож болно
for number in range(3):
    print(number)

for count in range(3):
    print("count is", count)
````

## 2. range()-ийн тухай дэлгэрэнгүй

`range()` нь дараалсан тоонуудыг үүсгэдэг. Үүнийг ашиглах гурван арга бий.

| Бичих арга | Утга | Өгөх тоонууд |
|---|---|---|
| `range(end)` | 0-ээс end хүртэл (end орохгүй) | `range(5)` → 0, 1, 2, 3, 4 |
| `range(start, end)` | start-аас end хүртэл (end орохгүй) | `range(1, 6)` → 1, 2, 3, 4, 5 |
| `range(start, end, step)` | start-аас end хүртэл, step-ээр алгасаж | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |

Хамгийн чухал санах зүйл:
- **Төгсгөлийн тоо хэзээ ч ордоггүй.** `range(5)` нь 5 биш 4 дээр зогсоно.
- Эхлэх тоо өгөхгүй бол **0**-ээс эхэлнэ.
- Алхам (step) өгөхгүй бол **1**-ээр нэмэгдэнэ.

````python
# range(end): 0-ээс эхэлж, end-ийн яг өмнө зогсоно
# range(4) нь 0, 1, 2, 3-ыг өгнө -> 4 ОРОХГҮЙ
for i in range(4):
    print(i)
````

````python
# range(start, end): start-аас эхэлж, end-ийн яг өмнө зогсоно
# 1-ээс 5 хүртэл авахыг хүсвэл end нь 6 байх ёстой (учир нь төгсгөлийн тоо ордоггүй)
for i in range(1, 6):
    print(i)
````

````python
# range(start, end, step): удаа бүр step-ээр алгасна
# 0-ээс эхэлж, 2-оор алгасаж, 10-ын өмнө зогсоно -> 0, 2, 4, 6, 8
for i in range(0, 10, 2):
    print(i)
````

````python
# step-ийн өөр нэг жишээ: 5-аас 25 хүртэл 5-аар тоолох
# 25 орохын тулд end нь 26 (30 гэж бичсэн ч болно, учир нь 30 өөрөө орохгүй)
for i in range(5, 26, 5):
    print(i)
````

````python
# step сөрөг тоо байвал тоонууд БУУРНА (буцаж тоолох)
# 5-аас эхэлж, 1-ээр буурч, 0-ийн өмнө зогсоно -> 5, 4, 3, 2, 1
for i in range(5, 0, -1):
    print(i)

print("Lift off!")      # Давталтын гадна, төгсгөлд нэг удаа ажиллана
````

````python
# i-г тооцоололд ашиглах
# 3-ын үржүүлэх хүснэгтийг хэвлэх (3 x 1-ээс 3 x 9 хүртэл)
for i in range(1, 10):           # i нь 1-ээс 9 хүртэл (10 орохгүй)
    print(3, "x", i, "=", 3 * i)  # Удаа бүр шинэ i-ээр 3 * i дахин бодогдоно
````

## 3. Давталтаар нийлбэр хуримтлуулах

Давталтын хамгийн түгээмэл ажлын нэг бол тоонуудыг **нэмж хуримтлуулах**. Жишээ нь, 1 + 2 + 3 + ... + 10 хэд вэ?

Арга нь нэг хувьсагчийг "хуримтлагдах нийлбэр" болгон ашиглах юм.
1. Давталтын өмнө хувьсагч үүсгээд 0 гэж оноо. (`total = 0`)
2. Давталтын дотор удаа бүр одоогийн тоог түүн дээр нэм. (`total = total + i`)
3. Давталтын дараа тэр хувьсагчид эцсийн үр дүн хадгалагдсан байна.

Хичээл 1-ийн `age = age + 1`-ийг санаж байна уу? Яг ижил санаа: хуучин утгыг аваад, ямар нэг зүйл нэмээд, нөгөө хайрцагтаа буцааж хадгална.

````python
# 1 + 2 + 3 + ... + 10-ыг нэмэх
total = 0                     # 1-р алхам: хуримтлагдах нийлбэр 0-ээс эхэлнэ (давталтын ӨМНӨ)

for i in range(1, 11):        # i нь 1-ээс 10 хүртэл
    total = total + i         # 2-р алхам: одоогийн i-г хуримтлагдах нийлбэр дээр нэмнэ

print("Sum:", total)          # 3-р алхам: давталтын ДАРАА эцсийн үр дүнг хэвлэнэ (давталтын гадна тул нэг л удаа хэвлэнэ)
````

````python
# Давталт бүрт total хэрхэн өөрчлөгдөж байгааг харъя
# Дээрх кодтой ижил, гэхдээ давталтын дотор print() нэмсэн
total = 0

for i in range(1, 6):         # i нь 1-ээс 5 хүртэл
    total = total + i
    print("i =", i, "-> total =", total)   # Давталтын дотор байгаа тул удаа бүр хэвлэнэ

# 1 дэх удаа: 0 + 1 = 1
# 2 дахь удаа: 1 + 2 = 3
# 3 дахь удаа: 3 + 3 = 6
# 4 дэх удаа: 6 + 4 = 10
# 5 дахь удаа: 10 + 5 = 15
````

````python
# total = total + i-г илүү товч бичиж болно: total += i
# Хоёр мөр яг ижил утгатай. Python програмистуудын олонх нь товч хэлбэрийг ашигладаг
total = 0

for i in range(1, 101):       # i нь 1-ээс 100 хүртэл
    total += i                # Дараахтай ижил: total = total + i

print("Sum of 1 to 100:", total)
````

````python
# Хуримтлуулах утга заавал нийлбэр байх албагүй. Үржүүлж ч болно
# 1 x 2 x 3 x 4 x 5 (математикт үүнийг "факториал" гэдэг)
result = 1                    # Үржүүлэхэд 0 биш 1-ээс эхэлнэ (ямар ч тоог 0-ээр үржүүлбэл 0 болно!)

for i in range(1, 6):         # i нь 1-ээс 5 хүртэл
    result = result * i       # Одоогийн i-ээр үр дүнг үржүүлнэ

print("1 x 2 x 3 x 4 x 5 =", result)
````

````python
# Давталт хэдэн удаа ажилласныг тоолох нь бас хуримтлуулалт юм
# Энд удаа бүр i биш 1-ийг нэмнэ
count = 0

for i in range(10, 20):       # i нь 10-аас 19 хүртэл
    count += 1                # Давталт бүрт 1 нэмнэ

print("The loop ran", count, "times.")   # range(10, 20)-д 10 тоо байгаа тул 10 удаа ажилласан
````

## 4. for болон if-ийг хамт ашиглах

`for` давталтын **дотор** `if`-г хийж болно. Ингэснээр тоо бүрийг нэг нэгээр нь шалгаж, зөвхөн хүссэн тоонууд дээрээ ямар нэг зүйл хийх боломжтой.

```python
for i in range(...):
    if нөхцөл:
        зөвхөн нөхцөл True үед ажиллах код
```

- `if` нь `for`-ийн дотор байгаа тул 4 зайн догол мөртэй.
- `if`-ийн доторх код нэг шат илүү догол мөртэй тул 8 зайн догол мөртэй.

````python
# 1-ээс 10 хүртэлх тэгш тоонуудыг л хэвлэх
# Хичээл 2-оос санаарай: number % 2 == 0 бол тэгш тоо
for i in range(1, 11):        # 1-ээс 10 хүртэлх тоо бүрийг нэг нэгээр шалгана
    if i % 2 == 0:            # Одоогийн i тэгш үү? (4 зай: for-ийн дотор)
        print(i)              # Зөвхөн тэгш тоо хэвлэгдэнэ (8 зай: if-ийн дотор)
````

````python
# Ижил үр дүнг if-гүйгээр, зөвхөн range()-ийн алхмаар ч гаргаж болно
# 2-оос эхэлж, 2-оор алгасна -> 2, 4, 6, 8, 10
for i in range(2, 11, 2):
    print(i)

# Хоёр арга хоёулаа зөв. Гэхдээ if ашиглавал "2 тутамд" гэхээс гадна ямар ч нөхцөл шалгаж болно
````

````python
# Давталтын дотор if - else ашиглах
for i in range(1, 6):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
````

````python
# for + if + хуримтлагдах нийлбэрийг хамт ашиглах
# 1-ээс 20 хүртэлх 3-ын үржвэрүүдийг л нэмэх (3 + 6 + 9 + 12 + 15 + 18)
total = 0

for i in range(1, 21):        # i нь 1-ээс 20 хүртэл
    if i % 3 == 0:            # i нь 3-ын үржвэр үү? (3-т хуваахад үлдэгдэлгүй)
        total += i            # Тэгвэл л нийлбэр дээр нэмнэ

print("Sum of multiples of 3:", total)
````

````python
# Сонгодог дасгал: FizzBuzz
# 1-ээс 15 хүртэлх тоонуудын хувьд:
#   - 3 болон 5-ын аль алиных нь үржвэр бол "FizzBuzz" гэж хэвлэ
#   - зөвхөн 3-ын үржвэр бол "Fizz" гэж хэвлэ
#   - зөвхөн 5-ын үржвэр бол "Buzz" гэж хэвлэ
#   - бусад тохиолдолд тоог өөрийг нь хэвлэ
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:     # Хамгийн хатуу нөхцөлийг ЭХЭЛЖ шалгана (Хичээл 2-ыг санаарай: дараалал чухал!)
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

## 5. while давталт

`for` давталт нь **хэдэн удаа** давтахаа мэдэж байгаа үед тохиромжтой. Гэхдээ заримдаа тэр тоог урьдчилан мэддэггүй. Зөвхөн **хэзээ зогсохоо** л мэддэг. Жишээ нь:

- Нийт мөнгө 10000-аас бага **байх хугацаанд** мөнгө хуримтлуулсаар бай.
- Цаас 1 см-ээс урт **байх хугацаанд** хагаслан хайчилсаар бай.

Үүнд `while` давталтыг ашигладаг. `while` нь "... байх хугацаанд" гэсэн утгатай. **Нөхцөл True байх хугацаанд** давтагдсаар, нөхцөл False болоход зогсоно.

```python
while нөхцөл:
    давтах код
```

Хэрхэн ажилладаг вэ:
1. Нөхцөлийг шалгана.
2. True бол догол мөртэй кодыг ажиллуулаад 1-р алхам руу буцна.
3. False бол давталтыг зогсоож, түүний дараах код руу шилжинэ.

**Маш чухал:** Давталтын дотор ямар нэг зүйл өөрчлөгдөж, хэзээ нэгэн цагт нөхцөл False болох ёстой. Нөхцөл үүрд True хэвээр байвал давталт хэзээ ч дуусахгүй. Үүнийг **төгсгөлгүй давталт** гэнэ.

````python
# while ашиглан 1-ээс 5 хүртэл хэвлэх
# while-д хувьсагчийг бид өөрсдөө үүсгэж, шалгаж, өөрчлөх ёстой
i = 1                     # 1. Давталтын ӨМНӨ хувьсагч үүсгэнэ

while i <= 5:             # 2. Нөхцөлийг шалгана: i нь 5-аас бага буюу тэнцүү байх хугацаанд үргэлжлүүлнэ
    print(i)
    i = i + 1             # 3. Хувьсагчийг өөрчилнө. Энэ мөр байхгүй бол i үүрд 1 хэвээр (төгсгөлгүй давталт!)

print("Done. Now i is", i)    # i нь 6 болоход 6 <= 5 нь False болсон тул давталт зогссон
````

````python
# Харьцуулъя: ижил зүйлийг for-оор
# for нь range()-ийн тусламжтайгаар "үүсгэх, шалгах, өөрчлөх" ажлыг автоматаар хийдэг
for i in range(1, 6):
    print(i)

# Тиймээс хэдэн удаа давтахаа мэдэж байвал ихэвчлэн for илүү энгийн
````

````python
# while илүү тохиромжтой тохиолдол: хэдэн удаа давтахаа урьдчилан мэдэхгүй
# Та өдөр бүр 3000 вон хуримтлуулдаг. 20000 вон буюу түүнээс их болоход хэдэн өдөр хэрэгтэй вэ?
money = 0
days = 0

while money < 20000:          # Мөнгө 20000-аас бага байх хугацаанд хуримтлуулсаар байна
    money += 3000             # 3000 вон хуримтлуулна
    days += 1                 # Нэг өдөр өнгөрлөө

print("Days needed:", days)
print("Money saved:", money)
````

````python
# Өөр нэг жишээ: тоо 1 буюу түүнээс бага болтол хоёр хуваасаар байх
# Тоо бүхэл хэвээр байхын тулд // (зөвхөн ноогдвор, Хичээл 1-ээс)-ийг ашиглана
number = 100
steps = 0

while number > 1:             # number нь 1-ээс их байх хугацаанд үргэлжлүүлнэ
    number = number // 2      # Хоёр хуваана (100 -> 50 -> 25 -> 12 -> 6 -> 3 -> 1)
    steps += 1
    print("Step", steps, "->", number)

print("It took", steps, "steps.")
````

````python
# Нөхцөл анхнаасаа False бол while давталт нэг ч удаа ажиллахгүй
i = 10

while i < 5:                  # 10 < 5 нь шууд False
    print("This never prints.")
    i += 1

print("The loop was skipped.")
````

## 6. break ба continue

Заримдаа давталтын явцыг дундаас нь өөрчлөхийг хүсдэг.

- `break`: **бүх давталтыг яг одоо зогсоож**, гадагш гарна.
- `continue`: **энэ удаагийн давталтын үлдсэн хэсгийг алгасаж**, шууд дараагийнх руу шилжинэ.

Тодорхой нөхцөлд л зогсоох эсвэл алгасахыг хүсдэг тул хоёулаа ихэвчлэн `if`-тэй хамт хэрэглэгддэг.

````python
# break: хайж байгаа зүйлээ олмогц давталтыг зогсооно
# 1-ээс 100 хүртэлх тоонуудаас 7-гийн үржвэр болох эхний тоог олъё
for i in range(1, 101):
    if i % 7 == 0:                # i нь 7-гийн үржвэр үү?
        print("Found it:", i)
        break                     # Давталтыг яг энд зогсооно. 7-гийн дараах тоонууд огт шалгагдахгүй

print("The loop is over.")
````

````python
# break давталтыг яг хаана зогсоож байгааг харъя
for i in range(1, 11):
    if i == 4:                    # i нь 4 болоход...
        break                     # ...давталтыг шууд зогсооно
    print(i)                      # Тиймээс зөвхөн 1, 2, 3 хэвлэгдэнэ (break түрүүлж ажилласан тул 4 хэвлэгдэхгүй)
````

````python
# continue: зөвхөн энэ удаагийн давталтыг алгасаад дараагийнхаар үргэлжлүүлнэ
for i in range(1, 6):
    if i == 3:                    # i нь 3 байхад...
        continue                  # ...доорх print()-г алгасаад шууд i = 4 руу шилжинэ
    print(i)                      # Тиймээс 1, 2, 4, 5 хэвлэгдэнэ (зөвхөн 3 байхгүй)
````

````python
# while-тай break: түгээмэл хэрэглэгддэг хэлбэр бол "while True" + break
# while True нь "үүрд давтах" гэсэн утгатай тул break л гарах ЦОРЫН ГАНЦ арга
count = 0

while True:                       # Нөхцөл үргэлж True -> break байхгүй бол төгсгөлгүй давталт болно
    count += 1
    print("count:", count)
    if count == 3:                # count нь 3 хүрэхэд...
        break                     # ...давталтаас гарна

print("Out of the loop!")
````

````python
# break болон continue-г зэрэгцүүлэн харьцуулах
print("With break:")
for i in range(1, 6):
    if i == 3:
        break                     # 3 дээр бүгдийг зогсооно
    print(i)                      # 1, 2

print("With continue:")
for i in range(1, 6):
    if i == 3:
        continue                  # Зөвхөн 3-ыг алгасна
    print(i)                      # 1, 2, 4, 5
````

## 7. Түгээмэл алдаанууд

Давталтыг сурч байхдаа бараг бүх хүн эдгээр алдааг гаргадаг. Ямар нэг зүйл буруу болвол эхлээд эдгээрийг шалгаарай.

1. **range()-ийн төгсгөлийн тоо ордоггүйг мартах**: `range(1, 10)` нь 9 дээр зогсоно. 10-ыг оруулахын тулд `range(1, 11)` гэж бичнэ.
2. **Хоёр цэг `:`-ийг мартах**: `for` болон `while` мөрүүд `:`-ээр төгсөх ёстой.
3. **Буруу догол мөр**: Зөвхөн догол мөртэй мөрүүд давтагдана. Давталтын дотор байх ёстой ч догол мөргүй мөр нэг л удаа ажиллана.
4. **while-тай төгсгөлгүй давталт**: `while` давталтын дотор хувьсагчийг өөрчлөхөө мартвал нөхцөл үүрд True хэвээр байна.
5. **Хуримтлагдах нийлбэрийг давталтын дотор дахин 0 болгох**: `total = 0`-г давталтын дотор биш **өмнө** нь бичих ёстой.

Доорх нүднүүд алдаа бүр ямар харагддагийг үзүүлнэ. Notebook эвдрэхгүйн тулд алдаатай кодыг тайлбар хэлбэрээр бичсэн.

**Хэрэв санамсаргүйгээр төгсгөлгүй давталт эхлүүлсэн бол** Jupyter-ийн дээд талын ■ (зогсоох / interrupt) товчийг дарах, эсвэл нүдийг сонгосон үед `I`-г хоёр удаа (`I`, `I`) дараарай.

````python
# Алдаа 1: төгсгөлийн тоо ордоггүйг мартах
# Бид 1-ээс 5 хүртэл авахыг хүссэн ч...
for i in range(1, 5):     # Буруу: энэ нь 4 дээр зогсоно!
    print(i)

print("---")

for i in range(1, 6):     # Зөв: end нь хүссэн сүүлийн тооноос 1-ээр их байх ёстой
    print(i)
````

````python
# Алдаа 2: хоёр цэгийг мартах

# for i in range(3)       # SyntaxError! Төгсгөлд хоёр цэг дутуу байна
#     print(i)

for i in range(3):        # Зөв: мөр хоёр цэгээр төгсөж байна
    print(i)
````

````python
# Алдаа 3: буруу догол мөр
# Хоёр дахь print() догол мөргүй тул давталтын гадна байж, нэг л удаа ажиллана
for i in range(3):
    print("Inside the loop:", i)
print("Outside the loop:", i)     # Давталтын дараа нэг удаа ажиллана. (i сүүлийн утга болох 2-ыг хадгалсан хэвээр)

# for i in range(3):
# print(i)                # IndentationError! for-ийн дотор догол мөртэй код огт алга
````

````python
# Алдаа 4: while-тай төгсгөлгүй давталт
# Энэ жишээнээс #-г БҮҮ арилгаарай! Хэзээ ч зогсохгүй.

# i = 1
# while i <= 5:
#     print(i)            # i хэзээ ч өөрчлөгдөхгүй тул i <= 5 үүрд True -> төгсгөлгүй давталт!

i = 1
while i <= 5:
    print(i)
    i += 1                # Зөв: удаа бүр i-г өөрчилж, эцэст нь нөхцөл False болно
````

````python
# Алдаа 5: хуримтлагдах нийлбэрийг давталтын дотор дахин 0 болгох
# Буруу: давталт бүрт total 0 болж байгаа тул зөвхөн сүүлийн тоо үлдэнэ
for i in range(1, 6):
    total = 0             # Буруу байрлал! Энэ мөр 5 удаа ажиллана
    total += i
print("Wrong total:", total)      # Зөвхөн 5

# Зөв: total-г давталтын ӨМНӨ ганцхан удаа 0 болгоно
total = 0
for i in range(1, 6):
    total += i
print("Correct total:", total)    # 1 + 2 + 3 + 4 + 5 = 15
````
