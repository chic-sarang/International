## 1. Зүсэлт гэж юу вэ?

Мэддэг зүйлээсээ эхэлье.

**Жагсаалт** нь дөрвөлжин хаалт `[ ]` дотор олон утгыг дарааллаар нь хадгалдаг:

```python
scores = [85, 92, 78, 90, 66]
```

Утга бүр **индекс** гэж нэрлэгдэх байрлалын дугаартай. Хамгийн чухал дүрэм: **Python 1-ээс биш 0-ээс тоолж эхэлдэг.**

| Утга | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |

**Нэг** утга авахын тулд жагсаалтын нэр болон дөрвөлжин хаалтанд индексийг бичнэ: `scores[0]` нь `85`, `scores[2]` нь `78`.

Гэтэл **хэд хэдэн утгыг зэрэг** авахыг хүсвэл яах вэ? Жишээ нь, зөвхөн эхний 3 сурагчийн оноог?

Одоогийн мэдлэгээрээ бол нэг нэгээр нь авах хэрэгтэй: `scores[0]`, `scores[1]`, `scores[2]`.

**Зүсэлт** нь жагсаалтын **нэг хэсгийг** нэг алхмаар авах боломж олгодог. "Slice" гэдэг үг нь "ямар нэг зүйлээс зүсэж авсан хэсэг" гэсэн утгатай, яг л нэг зүсэм талх эсвэл нэг хэсэг пицца шиг. Бид жагсаалтаас нэг хэсгийг зүсэж авна.

Зүсэлтийн үр дүн нь зөвхөн тэр хэсгийг агуулсан **шинэ жагсаалт** юм.

````python
scores = [85, 92, 78, 90, 66]

# Индексээр НЭГ утга авах (бидний мэддэг зүйл)
print(scores[0])        # индекс 0 -> 1-р утга -> 85
print(scores[1])        # индекс 1 -> 2-р утга -> 92
print(scores[2])        # индекс 2 -> 3-р утга -> 78
````

````python
scores = [85, 92, 78, 90, 66]

# Зүсэлтээр ХЭД ХЭДЭН утгыг зэрэг авах (өнөөдөр сурах зүйл)
# scores[0:3] нь "0 индексээс 3 индекс хүртэл (3 орохгүй)" гэсэн утгатай
print(scores[0:3])      # [85, 92, 78] -> эхний 3 утгатай шинэ жагсаалт
````

Дээрх хоёр үр дүнг анхааралтай хараарай:

- `scores[0]` нь **нэг утга** өгнө: `85` (дөрвөлжин хаалтгүй)
- `scores[0:3]` нь **жагсаалт** өгнө: `[85, 92, 78]` (дөрвөлжин хаалттай)

Энэ ялгаа маш чухал. Индекс нь утга өгдөг. Зүсэлт нь жагсаалт өгдөг.

## 2. Үндсэн хэлбэр: `жагсаалт[start:end]`

```python
жагсаалтын_нэр[start:end]
```

- `start`: зүсэлт **эхлэх** индекс (энэ утга **орно**)
- `end`: зүсэлт **зогсох** индекс (энэ утга **ОРОХГҮЙ**)
- `:` хоёр тооны хоорондох хоёр цэг

`scores[1:4]`-ийг ингэж уншиж болно: **"1 индексээс эхлээд, 4 индекс хүртэл (4 орохгүй)"**.

### end ордоггүй

Энэ бол эхлэгчдийг хамгийн их будлиулдаг хэсэг. Хүснэгтээр удаанаар харъя.

```python
scores = [85, 92, 78, 90, 66]
```

| Индекс | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Утга | 85 | 92 | 78 | 90 | 66 |
| `scores[1:4]`-д орох уу? | ❌ | ✅ | ✅ | ✅ | ❌ |

`scores[1:4]` → 1, 2, 3 индекс → `[92, 78, 90]`. 4 индекс дээр зүсэлт **зогсдог** тул 66 орохгүй.

**Энэ дүрмийг өмнө нь харж байсан уу?** Тийм! Хичээл 3-т `range(1, 4)` нь 1, 2, 3 тоонуудыг өгсөн бөгөөд 4 **ороогүй**. Зүсэлт яг ижил дүрмийг ашигладаг: **start орно, end орохгүй**.

### Хэдэн утга авах вэ?

end ордоггүй тул утгын тоо нь зүгээр л **end - start** байна.

- `scores[1:4]` → 4 - 1 = **3** утга
- `scores[0:2]` → 2 - 0 = **2** утга
- `scores[2:5]` → 5 - 2 = **3** утга

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[1:4])      # 1, 2, 3 индекс (4 орохгүй) -> [92, 78, 90]
````

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[0:2])      # 0, 1 индекс -> [85, 92]
print(scores[2:5])      # 2, 3, 4 индекс -> [78, 90, 66]
print(scores[3:4])      # зөвхөн 3 индекс -> [90] (нэг утгатай жагсаалт)
````

````python
# Хичээл 3-ын range()-тэй ижил дүрэм: end орохгүй
# range(1, 4) нь 1, 2, 3-ыг өгнө
for i in range(1, 4):
    print(i)
````

````python
# Тэмдэгт мөрүүдтэй ч яг адилхан ажиллана
fruits = ["apple", "banana", "cherry", "grape", "melon"]
#  индекс:     0        1         2        3        4

print(fruits[0:2])      # 0, 1 индекс -> ['apple', 'banana']
print(fruits[1:3])      # 1, 2 индекс -> ['banana', 'cherry']
print(fruits[2:5])      # 2, 3, 4 индекс -> ['cherry', 'grape', 'melon']
````

### len()-ээр утгын тоог тоолох

`len()`-ийг санаж байна уу? Энэ нь жагсаалтад **хэдэн утга** байгааг хэлж өгдөг. `len([85, 92, 78])` нь `3`.

Зүсэлт бол жагсаалт тул түүн дээр ч `len()` ашиглаж болно. "end - start" дүрмийг шалгаж үзье.

````python
scores = [85, 92, 78, 90, 66]

print(len(scores))          # Бүх жагсаалт 5 утгатай
print(len(scores[1:4]))     # 4 - 1 = 3 утга
print(len(scores[0:2]))     # 2 - 0 = 2 утга
````

### Зүсэлт бол шинэ жагсаалт. Анхны жагсаалт өөрчлөгдөхгүй.

Зүсэлт нь жагсаалтын нэг хэсгийг **шинэ жагсаалт** руу **хуулдаг**. Анхны жагсаалт яг хэвээрээ үлдэнэ.

Зүсэлтийг дараа нь ашиглахыг хүсвэл бусад утгын нэгэн адил хувьсагчид хадгалаарай. Санаарай: `=` нь **"баруун талын утгыг зүүн талын хайрцагт хадгал"** гэсэн утгатай.

````python
scores = [85, 92, 78, 90, 66]

first_three = scores[0:3]   # Шинэ жагсаалтыг first_three нэртэй хувьсагчид хадгална

print(first_three)          # [85, 92, 78] -> шинэ жагсаалт
print(scores)               # [85, 92, 78, 90, 66] -> анхны жагсаалт бүтэн хэвээр
````

````python
# Шинэ жагсаалт бол жинхэнэ жагсаалт тул төрөл нь list
# type() нь утга ямар төрлийнх болохыг хэлж өгдөг
scores = [85, 92, 78, 90, 66]
first_three = scores[0:3]

print(type(first_three))    # <class 'list'>
print(type(scores[0]))      # <class 'int'> -> индекс нь жагсаалт биш нэг утга (бүхэл тоо) өгдөг
````

````python
# Шинэ жагсаалт дээр ч индекс ашиглаж болно
# Санаарай: шинэ жагсаалт дахиад 0 индексээс эхэлнэ!
scores = [85, 92, 78, 90, 66]
last_two = scores[3:5]      # [90, 66]

print(last_two[0])          # ШИНЭ жагсаалтын 1-р утга -> 90 (анхны жагсаалтад 3 индекс байсан)
print(last_two[1])          # ШИНЭ жагсаалтын 2-р утга -> 66
````

## 3. start эсвэл end-ийг орхих

Хоёр тоог үргэлж бичих шаардлагагүй. Дутуу тоог Python өөрөө нөхдөг.

| Бичих арга | Утга | Python үүнийг юу гэж ойлгох вэ |
|---|---|---|
| `scores[:3]` | **эхнээс** 3 индекс хүртэл (3 орохгүй) | `scores[0:3]` |
| `scores[2:]` | 2 индексээс **төгсгөл** хүртэл | `scores[2:5]` (5 утга байх үед) |
| `scores[:]` | эхнээс төгсгөл хүртэл (бүх жагсаалт) | `scores[0:5]` |

- start байхгүй → **эхнээс** (0 индексээс) эхэлнэ
- end байхгүй → **төгсгөл** хүртэл явна (сүүлийн утга орно)

**Энэ яагаад хэрэгтэй вэ?**
- `scores[:3]` нь `scores[0:3]`-аас богино бөгөөд уншихад амархан: "эхний 3".
- `scores[2:]` нь жагсаалт хэр урт болохыг мэдэхгүй байсан ч ажилладаг. Утгуудыг тоолох шаардлагагүй.

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

# start байхгүй -> эхнээс
print(scores[:3])       # scores[0:3]-тай ижил -> [85, 92, 78]
print(scores[0:3])      # Ижил үр дүн
````

````python
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

# end байхгүй -> төгсгөл хүртэл (сүүлийн утга ОРНО)
print(scores[2:])       # 2, 3, 4 индекс -> [78, 90, 66]
print(scores[1:])       # Эхнийхээс бусад бүгд -> [92, 78, 90, 66]
````

````python
# [:n] = "эхний n утга"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[:1])       # Эхний 1 -> ['apple']
print(fruits[:2])       # Эхний 2 -> ['apple', 'banana']
print(fruits[:4])       # Эхний 4 -> ['apple', 'banana', 'cherry', 'grape']
````

````python
# [n:] = "эхний n утгыг алгасаад үлдсэнийг ав"
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[1:])       # 1-ийг алгасна -> ['banana', 'cherry', 'grape', 'melon']
print(fruits[3:])       # 3-ыг алгасна -> ['grape', 'melon']
````

````python
# [n:] нь жагсаалт уртассан ч ажилладаг
# append() нь жагсаалтын төгсгөлд утга нэмдэг
scores = [85, 92, 78]
scores.append(90)           # Одоо: [85, 92, 78, 90]
scores.append(66)           # Одоо: [85, 92, 78, 90, 66]

print(scores[1:])           # Одоо 5 утга байгааг мэдэх шаардлагагүй байсан
````

````python
# [:] = бүх жагсаалт (түүний хуулбар)
scores = [85, 92, 78, 90, 66]
print(scores[:])
````

````python
# Жагсаалтыг хоёр хэсэгт хуваах
# [:2] ба [2:] ижил тоо ашигладаг тул хамтдаа бүх жагсаалт болно
# (учир нь 2 нь [:2]-т ОРОХГҮЙ, харин [2:]-т ОРНО)
scores = [85, 92, 78, 90, 66]

front = scores[:2]
back = scores[2:]

print(front)            # [85, 92]
print(back)             # [78, 90, 66]
````

## 4. Сөрөг индексээр зүсэх

**Сөрөг индекс**-ийг санацгаая. Python нь жагсаалтын **төгсгөлөөс буцаж** ч тоолж чаддаг:

- `-1` нь **сүүлийн** утга
- `-2` нь сүүлээсээ хоёр дахь утга
- `-3` нь сүүлээсээ гурав дахь утга гэх мэт

| Утга | 85 | 92 | 78 | 90 | 66 |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |
| Сөрөг индекс | -5 | -4 | -3 | -2 | -1 |

Сөрөг индекс ашигладаг гол шалтгаан: **хэдэн утга байгааг тоолохгүйгээр жагсаалтын төгсгөлд хүрч болно**. Жагсаалт 5 утгатай ч, 500 утгатай ч `-1` нь үргэлж сүүлийн утга.

Сөрөг индекс зүсэлтэд ч ажилладаг. Дүрэм нь адилхан: **start орно, end орохгүй**.

| Бичих арга | Утга |
|---|---|
| `scores[-3:]` | **сүүлийн 3** утга |
| `scores[-2:]` | **сүүлийн 2** утга |
| `scores[:-1]` | **сүүлийнхээс бусад** бүх утга |
| `scores[:-2]` | **сүүлийн 2-оос бусад** бүх утга |

````python
scores = [85, 92, 78, 90, 66]
#  сөрөг индекс: -5  -4  -3  -2  -1

# [-n:] = "сүүлийн n утга"
print(scores[-3:])      # -3-аас төгсгөл хүртэл -> [78, 90, 66]
print(scores[-2:])      # -2-оос төгсгөл хүртэл -> [90, 66]
print(scores[-1:])      # -1-ээс төгсгөл хүртэл -> [66] (нэг утгатай жагсаалт)
````

````python
# Харьцуулаарай: индекс нэг утга, зүсэлт жагсаалт өгдөг
scores = [85, 92, 78, 90, 66]

print(scores[-1])       # 66   -> нэг утга (сүүлийнх)
print(scores[-1:])      # [66] -> сүүлийн утгыг агуулсан жагсаалт
````

````python
scores = [85, 92, 78, 90, 66]
#  сөрөг индекс: -5  -4  -3  -2  -1

# [:-n] = "сүүлийн n утгаас бусад бүгд"
print(scores[:-1])      # -1 хүртэл, гэхдээ -1 ОРОХГҮЙ -> [85, 92, 78, 90]
print(scores[:-2])      # -2 хүртэл, гэхдээ -2 ОРОХГҮЙ -> [85, 92, 78]
````

````python
# Ямар ч урттай жагсаалтын сүүлийн 3 утга
# Хэдэн утга байгааг мэдэх шаардлагагүй
short_list = [1, 2, 3, 4]
long_list = [10, 20, 30, 40, 50, 60, 70, 80]

print(short_list[-3:])  # [2, 3, 4]
print(long_list[-3:])   # [60, 70, 80]
````

````python
# Бодит жишээ: хамгийн сүүлийн 3 мессеж
# Шинэ мессеж append()-ээр төгсгөлд нэмэгддэг тул хамгийн шинэ нь төгсгөлд байна
messages = ["Hi", "How are you?", "Fine, thanks", "See you", "Bye"]

print(messages[-3:])    # Хамгийн сүүлийн 3 мессеж
````

## 5. Зүсэлтийг for давталттай ашиглах

Зүсэлт бол жагсаалт. Тиймээс жагсаалтаар хийж болох бүх зүйлийг зүсэлтээр ч хийж болно. Үүнд **for давталт** ч орно.

for давталт хэрхэн ажилладгийг санацгаая:

```python
for score in scores:
    print(score)
```

- `for score in scores:` нь **"scores доторх score бүрийн хувьд доорх кодыг ажиллуул"** гэсэн утгатай.
- Давталт бүрд жагсаалтын нэг утга эхнийхээс сүүлийнх хүртэл `score` хувьсагчид хадгалагдана.
- Мөр хоёр цэг `:`-ээр төгсөх ёстой.
- Давтах код **догол мөртэй** (эхэндээ 4 зай) байх ёстой. Python догол мөрөөр аль мөрүүд давталтад хамаарахыг мэддэг. Догол мөртэй мөрүүдийн энэ бүлгийг **блок** гэдэг.
- Догол мөр**гүй** эхний мөр давталтын гадна байх бөгөөд давталт дууссаны дараа зөвхөн нэг удаа ажиллана.

Одоо `in`-ий дараа **зүсэлт** бичвэл давталт жагсаалтын **зөвхөн тэр хэсгээр** гүйнэ.

```python
for score in scores[:3]:      # зөвхөн эхний 3 утга
    print(score)
```

````python
# Бүх жагсаалт
scores = [85, 92, 78, 90, 66]

for score in scores:            # 5 утга -> давталт 5 удаа ажиллана
    print(score)
````

````python
# Зөвхөн эхний 3 утга
scores = [85, 92, 78, 90, 66]

for score in scores[:3]:        # scores[:3] нь [85, 92, 78] -> давталт 3 удаа ажиллана
    print(score)                # Догол мөртэй (4 зай) -> давталтын дотор -> давтагдана
````

````python
# Зөвхөн сүүлийн 2 утга
scores = [85, 92, 78, 90, 66]

for score in scores[-2:]:       # scores[-2:] нь [90, 66] -> давталт 2 удаа ажиллана
    print(score)
````

````python
# Эхнийхээс бусад бүгд
# Эхний утга нь жинхэнэ өгөгдөл биш, гарчиг байх үед хэрэгтэй
row = ["Scores", 85, 92, 78]

for value in row[1:]:           # row[1:] нь [85, 92, 78] -> "Scores"-ыг алгасна
    print(value)
````

### Давталтын хувьсагч ямар ч нэртэй байж болно

Санаарай: `for`-ийн дараах хувьсагч (энд `score`) нь давталтын үүсгэдэг **шинэ хувьсагч**. Түүнд ямар ч нэр өгч болно. Гэхдээ жагсаалтын нэрийн **ганц тооны** хэлбэрийг ашиглах нь сайн дадал: `scores` → `score`, `fruits` → `fruit`. Тэгвэл "бүх жагсаалт" ба "нэг утга"-ыг ялгахад амархан.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[1:4]:       # fruits[1:4] нь ['banana', 'cherry', 'grape']
    print(fruit)                # fruit (нэг утга), fruits (бүх жагсаалт)
````

### Юу давтагдахыг догол мөр шийддэг

Хичээл 5-ын нэгэн адил догол мөртэй мөрүүд давтагдаж, догол мөргүй эхний мөр давталтын дараа зөвхөн нэг удаа ажиллана.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

for fruit in fruits[:2]:        # Зөвхөн 'apple' ба 'banana'
    print("I like", fruit)      # Догол мөртэй -> 2 удаа ажиллана
print("That's all!")            # Догол мөргүй -> давталтын гадна -> нэг удаа ажиллана
````

**`print("I like", fruit)`-ийн тухай:** `print()` дотор хэд хэдэн утгыг **таслал** `,`-аар тусгаарлаж бичвэл тэдгээр нь нэг мөрөнд хэвлэгдэж, Python тэдгээрийн хооронд **нэг зай** автоматаар тавина. Хашилт доторх бичвэр байгаагаараа хэвлэгдэж, хувьсагчийн оронд түүн доторх утга хэвлэгдэнэ.

## 6. Зүсэлтээр тооцоолох

### Жагсаалтын нэг хэсгийг нэмэх

Хичээл 5-д бид **хуримтлагдах нийлбэр** ашиглан жагсаалтын бүх утгыг нэмсэн:

1. **Давталтын өмнө:** хувьсагч үүсгээд 0 гэж оноо → `total = 0`
2. **Давталтын дотор:** одоогийн утгыг нэм → `total = total + score`
3. **Давталтын дараа:** хувьсагчид нийлбэр байна

`total = total + score` юу гэсэн үг болохыг санаарай: `=` нь "хадгал" гэсэн утгатай. Python эхлээд баруун талыг (хуучин нийлбэр + одоогийн оноо) бодоод, үр дүнг `total` руу буцааж хадгална.

Зүсэлтээр бол жагсаалтын **зөвхөн нэг хэсгийг** нэмж болно. `in`-ий дараа зүсэлтийг бичихэд л болно.

````python
# Зөвхөн эхний 3 оноог нэмэх
scores = [85, 92, 78, 90, 66]
total = 0                           # Давталтын өмнө: 0-ээс эхэлнэ

for score in scores[:3]:            # Зөвхөн 85, 92, 78
    total = total + score           # хуучин нийлбэр + одоогийн оноо -> total руу буцааж хадгална

print("Total of the first 3:", total)   # Давталтын дараа (догол мөргүй): 85 + 92 + 78 = 255
````

````python
# total хэрхэн өөрчлөгдөхийг харъя
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total = total + score
    print("score:", score, "total:", total)     # Догол мөртэй -> давталт бүрт хэвлэгдэнэ
````

**Богино хэлбэр `+=`-ийг санаарай:** `total += score` нь `total = total + score`-тэй яг ижил утгатай. "Баруун талыг хувьсагч дээр нэмээд, үр дүнг түүнд буцааж хадгал."

````python
# Ижил зүйлийг += ашиглан бичих
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[:3]:
    total += score                  # дараахтай ижил: total = total + score

print("Total of the first 3:", total)
````

````python
# Зөвхөн сүүлийн 2 оноог нэмэх
scores = [85, 92, 78, 90, 66]
total = 0

for score in scores[-2:]:           # Зөвхөн 90, 66
    total += score                  # дараахтай ижил: total = total + score

print("Total of the last 2:", total)    # 90 + 66 = 156
````

### Нэг хэсгийн дундаж

**Дундаж** гэдэг нь нийлбэрийг утгын тоонд хуваасан утга: **дундаж = нийлбэр / утгын тоо**

Утгын тоог `len()`-ээр олно. Гэхдээ анхаараарай: бүх жагсаалтын биш, **зүсэлт доторх** утгын тоо хэрэгтэй. Тиймээс `len()`-ийг зүсэлт дээр ашиглана.

Мөн санаарай: нийлбэр бүрэн болсон үед, давталтын **дараа** (догол мөргүй) дунджийг бодно.

````python
# Эхний 3 онооны дундаж
scores = [85, 92, 78, 90, 66]
first_three = scores[:3]            # Зүсэлтийг хувьсагчид хадгална: [85, 92, 78]
total = 0

for score in first_three:
    total += score                  # дараахтай ижил: total = total + score

average = total / len(first_three)  # 255 / 3 (бүх жагсаалтын биш ЗҮСЭЛТИЙН урт)
print("Total:", total)
print("Count:", len(first_three))
print("Average:", average)
````

````python
# Эхний хагас ба хоёр дахь хагасын дунджийг харьцуулах
scores = [60, 65, 70, 85, 90, 95]
first_half = scores[:3]             # [60, 65, 70]
second_half = scores[3:]            # [85, 90, 95]

total = 0
for score in first_half:
    total += score
print("First half average:", total / len(first_half))

total = 0                           # Хоёр дахь хагаст дахин 0-ээс эхэлнэ!
for score in second_half:
    total += score
print("Second half average:", total / len(second_half))
````

### Нэг хэсэг доторх хамгийн их утга

Хичээл 5-д бид хамгийн их утгыг хүн цаасан дээр олдог шиг олсон:

1. **Эхний утгыг** "одоогоор хамгийн том" гэж санана.
2. Утга бүрийг харна. Санасан утгаас том бол оронд нь энийг санана.
3. Эцэст нь санаж буй утга бол хамгийн том нь.

Зүсэлтээр бол жагсаалтын **зөвхөн нэг хэсгээс** хамгийн их утгыг олж болно.

````python
# Сүүлийн 3 сурагчийн хамгийн өндөр оноо
scores = [95, 82, 74, 88, 79]
last_three = scores[-3:]            # [74, 88, 79]

highest = last_three[0]             # 1-р алхам: ЗҮСЭЛТИЙН эхний утгаас (74) эхэлнэ

for score in last_three:
    if score > highest:             # 2-р алхам: санасан утгаас том уу?
        highest = score             #            том бол оронд нь энийг санана

print("Highest of the last 3:", highest)    # 3-р алхам: 88
````

## 7. Зүсэлтийг if, in, append()-тэй ашиглах

### Давталтын доторх if

`for` давталтын дотор `if` хийхэд догол мөр **хоёр шат** болно:

```python
for score in scores[:3]:      # зайгүй
    if score >= 80:           # 4 зай -> for-ийн дотор (оноо бүрд ажиллана)
        print(score)          # 8 зай -> if-ийн дотор (зөвхөн нөхцөл True үед ажиллана)
```

`if` хэрхэн ажилладгийг санаарай: түүний доорх догол мөртэй код **зөвхөн нөхцөл True үед** ажиллана.

````python
# Эхний 4 онооноос зөвхөн 80 ба түүнээс дээшийг хэвлэх
scores = [85, 72, 91, 78, 95, 60]

for score in scores[:4]:            # Зөвхөн 85, 72, 91, 78
    if score >= 80:                 # 4 зай: оноо бүрийг шалгана
        print(score)                # 8 зай: 80 ба түүнээс дээш бол л хэвлэнэ
````

````python
# Тоолох: сүүлийн 3 онооноос хэд нь тэнцсэн бэ (60 ба түүнээс дээш)?
# Тоолох нь хуримтлагдах нийлбэр шиг ажилладаг, гэхдээ утгын оронд 1-ийг нэмнэ
scores = [85, 42, 78, 55, 90]
pass_count = 0                      # Давталтын өмнө: 0-ээс эхэлнэ

for score in scores[-3:]:           # Зөвхөн 78, 55, 90
    if score >= 60:
        pass_count += 1             # дараахтай ижил: pass_count = pass_count + 1

print("Passed among the last 3:", pass_count)
````

### in

`утга in жагсаалт` нь утга жагсаалтад байгаа эсэхийг шалгаж, `True` эсвэл `False` өгдөг. Зүсэлт бол жагсаалт тул `in`-ийг зүсэлттэй ч ашиглаж болно. Тэгвэл бид жагсаалтын **зөвхөн нэг хэсгийг** шалгана.

````python
# Шилдэг 3 сурагч (жагсаалт аль хэдийн дараалсан: 1-р байр, 2-р байр, ...)
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]

print("Tom" in ranking[:3])         # Tom шилдэг 3-т байна уу? -> True
print("Kate" in ranking[:3])        # Kate шилдэг 3-т байна уу? -> False
print("Kate" in ranking)            # Kate бүх жагсаалтад байна уу? -> True
````

````python
# in-ийг if-ийн нөхцөл болгон ашиглах
ranking = ["Anna", "Tom", "Bill", "Kate", "Mike"]
name = "Bill"

if name in ranking[:3]:             # Bill шилдэг 3-т байна уу?
    print(name, "gets a medal!")
else:
    print(name, "doesn't get a medal.")
````

### append(): шинэ жагсаалт үүсгэх

`append()`-ийг санаарай: `жагсаалт.append(утга)` нь утгыг жагсаалтын **төгсгөлд** нэмдэг.

Хоосон жагсаалт `[]`-ээс эхлээд, давталтын дотор зүсэлтээс зөвхөн хүссэн утгуудаа `append()` хийж болно.

````python
# Эхний 4 онооноос зөвхөн тэнцсэн оноотой шинэ жагсаалт үүсгэх
scores = [85, 42, 78, 55, 90, 30]
passed = []                         # Давталтын өмнө: хоосон жагсаалт

for score in scores[:4]:            # Зөвхөн 85, 42, 78, 55
    if score >= 60:
        passed.append(score)        # passed-ийн төгсгөлд нэмнэ

print(passed)                       # [85, 78]
````

````python
# Сүүлийн 3 үнийг 2 дахин өсгөсөн шинэ жагсаалт үүсгэх
prices = [1000, 2500, 3000, 1500, 4000]
doubled = []

for price in prices[-3:]:           # Зөвхөн 3000, 1500, 4000
    doubled.append(price * 2)       # Эхлээд price * 2 бодогдоод, дараа нь нэмэгдэнэ

print(doubled)
````

## 8. Түгээмэл алдаанууд

Эдгээр нь зүсэлтийг сурч байхдаа бараг бүх хүний гаргадаг алдаанууд. Алдаа бүрийн хувьд бид дараахыг харна:

- ❌ **Буруу код**
- 🔍 **Юу буруу болсон бэ**
- ✅ **Зөв арга**

Зарим буруу код алдаа гаргадаг. Notebook зогсохгүйн тулд ийм кодыг тайлбар хэлбэрээр (эхэнд нь `#`-тэй) бичсэн. Жинхэнэ алдааны мэдэгдлийг харахыг хүсвэл `#`-г арилгаж, нүдийг ажиллуулаад, дараа нь `#`-г буцааж тавиарай.

Бусад буруу код **алдаа гаргахгүй ч буруу үр дүн өгдөг**. Python анхааруулдаггүй тул эдгээр нь илүү аюултай.

### Алдаа 1: end орно гэж бодох

````python
# ❌ Буруу: 1, 2, 3 индекс дээрх утгуудыг авахыг хүсэж байна
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[1:3])      # [92, 78] -> ердөө 2 утга! 90 (3 индекс) алга
````

🔍 **Юу буруу болсон бэ?**

- `scores[1:3]` нь "1 индексээс 3 индекс хүртэл, 3 **орохгүй**" гэсэн утгатай.
- Тиймээс бид зөвхөн 1 ба 2 индексийг авна. 3 индекс дээр зүсэлт **зогсдог**.
- Алдаа гарахгүй тул анзааралгүй өнгөрөхөд амархан.

✅ **Зөв арга:** end нь хүссэн сүүлийн индексээс **1-ээр их** байх ёстой. 3 индексийг оруулахын тулд `4` гэж бичнэ. (Хичээл 3-ын `range()`-тэй адил: 3-ыг оруулахын тулд `range(1, 4)` гэж бичдэг.)

````python
# ✅ Зөв
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # 1, 2, 3 индекс -> [92, 78, 90]
````

### Алдаа 4: Зүсэлт ба индексийг андуурах

````python
# ❌ Буруу: эхний оноон дээр 5 нэмэхийг хүсэж байна
scores = [85, 92, 78]

first = scores[0:1]     # Энэ бол ЖАГСААЛТ: [85], 85 гэсэн тоо биш
print(first)

# print(first + 5)      # TypeError: can only concatenate list (not "int") to list
````

🔍 **Юу буруу болсон бэ?**

- `scores[0:1]` бол **зүсэлт** тул үр дүн нь нэг утгатай **жагсаалт**: `[85]`.
- Жагсаалт ба тоог `+`-ээр нэмж болохгүй тул Python алдаа гаргана.
- Алдааны мэдэгдэл нь: "жагсаалтад зөвхөн жагсаалт залгаж болно, тоо (int) биш" гэсэн утгатай.
- Гаралт дахь дөрвөлжин хаалтыг хараарай: `[85]`. Хаалт харагдвал энэ бол жагсаалт.

✅ **Зөв арга:** **Нэг утга** авахын тулд **индекс** (хоёр цэггүй, нэг тоо) ашиглана: `scores[0]`.

````python
# ✅ Зөв
scores = [85, 92, 78]

first = scores[0]       # Энэ бол 85 гэсэн тоо (хаалтгүй)
print(first)
print(first + 5)        # 90
````

### Алдаа 5: Зүсэлт анхны жагсаалтыг өөрчилнө гэж бодох

````python
# ❌ Буруу: зөвхөн эхний 3 оноог үлдээхийг хүсэж байна
scores = [85, 92, 78, 90, 66]

scores[:3]              # Энэ нь шинэ жагсаалт үүсгэнэ... гэхдээ хаана ч хадгалагдаагүй!
print(scores)           # [85, 92, 78, 90, 66] -> юу ч өөрчлөгдөөгүй
````

🔍 **Юу буруу болсон бэ?**

- Зүсэлт нь **шинэ жагсаалт** үүсгэдэг. Анхны жагсаалтыг **өөрчилдөггүй**.
- `scores[:3]` мөрөнд Python `[85, 92, 78]` гэсэн шинэ жагсаалт үүсгэсэн ч бид түүнийг ямар ч хувьсагчид хадгалаагүй. Тиймээс зүгээр л алга болсон.
- Алдаа гарахгүй тул анзаарахад хэцүү.

✅ **Зөв арга:** Зүсэлтийг `=`-ээр хувьсагчид хадгална. Анхны жагсаалтыг солихыг хүсвэл ижил хувьсагч руу буцааж хадгална (`age = age + 1` шиг: шинэ утга хуучныг солино).

````python
# ✅ Зөв (1): шинэ хувьсагчид хадгалах
scores = [85, 92, 78, 90, 66]

first_three = scores[:3]
print(first_three)      # [85, 92, 78]
print(scores)           # Анхны жагсаалт бүтэн хэвээр
````

````python
# ✅ Зөв (2): ижил хувьсагч руу буцааж хадгалах
scores = [85, 92, 78, 90, 66]

scores = scores[:3]     # Эхлээд баруун тал бодогдоод ([85, 92, 78]), дараа нь scores-д хадгалагдана
print(scores)           # [85, 92, 78] -> хуучин жагсаалт солигдсон
````

### Алдаа 6: start-ыг end-ийн ард бичих

````python
# ❌ Буруу: 1 индексээс 3 индекс хүртэлх утгуудыг авахыг хүсэж байна
scores = [85, 92, 78, 90, 66]

print(scores[4:1])      # [] -> хоосон жагсаалт!
````

🔍 **Юу буруу болсон бэ?**

- Зүсэлт үргэлж **зүүнээс баруун тийш**: start-аас end рүү явдаг.
- `scores[4:1]` нь 4 индексээс эхлээд 1 индекс рүү явах гэж оролддог. Гэхдээ 1 нь 4-ийн **зүүн** талд байгаа тул хооронд нь юу ч байхгүй.
- Үр дүн нь хоосон жагсаалт `[]`. Алдаа гарахгүй.

✅ **Зөв арга:** **Бага** индексийг эхэнд (start), **их** индексийг дараа нь (end) бичнэ.

````python
# ✅ Зөв
scores = [85, 92, 78, 90, 66]

print(scores[1:4])      # 1 индексээс 3 индекс хүртэл -> [92, 78, 90]
````

### Алдаа 7: Сүүлийн 3 утгыг авахдаа [-3:-1] ашиглах

````python
# ❌ Буруу: сүүлийн 3 утгыг авахыг хүсэж байна
scores = [85, 92, 78, 90, 66]
#  сөрөг индекс: -5  -4  -3  -2  -1

print(scores[-3:-1])    # [78, 90] -> ердөө 2 утга! Сүүлийн утга (66) алга
````

🔍 **Юу буруу болсон бэ?**

- end **ордоггүй**. Сөрөг индекстэй үед ч энэ дүрэм өөрчлөгдөхгүй.
- `scores[-3:-1]` нь "-3-аас -1 хүртэл, -1 **орохгүй**" гэсэн утгатай. Тиймээс сүүлийн утга (-1) үлдэнэ.

✅ **Зөв арга:** Сүүлийн утгыг оруулахын тулд **end-ийг орхино**: `scores[-3:]`. end байхгүй бол "төгсгөл хүртэл" гэсэн утгатай.

````python
# ✅ Зөв
scores = [85, 92, 78, 90, 66]

print(scores[-3:])      # -3-аас төгсгөл хүртэл -> [78, 90, 66]
````

### Алдаа 8: Зүсэлтийн дунджийг бодохдоо бүх жагсаалтын уртыг ашиглах

````python
# ❌ Буруу: эхний 3 онооны дунджийг авахыг хүсэж байна
scores = [80, 90, 100, 50, 60]
total = 0

for score in scores[:3]:
    total += score              # 80 + 90 + 100 = 270

average = total / len(scores)   # 270 / 5 -> буруу! Бид ердөө 3 оноо нэмсэн
print("Average:", average)      # 54.0
````

🔍 **Юу буруу болсон бэ?**

- Бид зөвхөн эхний **3** оноог (270) нэмсэн.
- Гэтэл `len(scores)` нь бүх жагсаалтын **5** утгыг бүгдийг тоолно.
- Тиймээс 3-ын оронд 5-д хуваасан. Дундаж хэт бага гарсан.
- Алдаа гарахгүй.

✅ **Зөв арга:** **Зүсэлт доторх** утгын тоонд хуваана. Хамгийн хялбар арга: эхлээд зүсэлтийг хувьсагчид хадгалаад, тэр хувьсагч дээр `len()` ашиглана.

````python
# ✅ Зөв
scores = [80, 90, 100, 50, 60]
first_three = scores[:3]        # [80, 90, 100]
total = 0

for score in first_three:
    total += score

average = total / len(first_three)  # 270 / 3
print("Average:", average)          # 90.0
````

### Алдаа 9: Зүсэлт жагсаалтын төгсгөлөөс хэтрэхэд алдаа гарна гэж бодох

````python
# ❓ Гайхалтай нь: энэ АЛДАА гаргахгүй
scores = [85, 92, 78]           # Ердөө 3 утга (0, 1, 2 индекс)

print(scores[0:10])             # [85, 92, 78] -> алдаа алга!

# Харин хэт том индекс АЛДАА гаргадаг
# print(scores[10])             # IndexError: list index out of range
````

🔍 **Юу болж байна вэ?**

- **Индекс**-ийн хувьд Python яг тэр байрлалаас **нэг** утга олох ёстой. Тийм байрлал байхгүй бол `IndexError` ("индекс жагсаалтын гадна байна") гаргана.
- **Зүсэлт**-ийн хувьд Python тэр хүрээнээс **олж чадсан утгуудаа** л өгдөг. Хүрээ төгсгөлөөс хэтэрвэл төгсгөл дээр зогсоно. Алдаа гарахгүй.
- Энэ нь яг алдаа биш ч будлиулж болно: буруу харагдаж буй зүсэлт "ажиллаж" магадгүй бөгөөд тоонууд буруу бичигдсэнийг та анзаарахгүй.

✅ **Зөвлөгөө:** Төгсгөл хүртэл бүгдийг авахыг хүсвэл том тоо таамаглах хэрэггүй. end-ийг орхиход л болно: `scores[0:]` эсвэл `scores[:]`.

````python
# ✅ Илүү сайн
scores = [85, 92, 78]

print(scores[0:])               # 0 индексээс төгсгөл хүртэл -> [85, 92, 78]
````
