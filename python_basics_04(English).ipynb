## 1. What Is a List?

In Lesson 1, we learned that a variable is like a box that stores **one** value.

But what if you want to store the test scores of 5 students? With what we know so far, you would need 5 variables:

```python
score1 = 85
score2 = 92
score3 = 78
score4 = 90
score5 = 66
```

That's already a lot. And what if there are 30 students? Or 300?

A **list** solves this problem. A list is a single variable that can hold **many values in order**.

You can think of a list like a row of **lockers**:
- The whole row of lockers has one name (the list's name).
- Each locker holds one value.
- Each locker has a number on it, so you can find it.

How to make a list:
- Put the values inside square brackets `[ ]`.
- Separate the values with commas `,`.

```python
scores = [85, 92, 78, 90, 66]
```

````python
# A list of integers
# One variable (scores) now holds 5 values
scores = [85, 92, 78, 90, 66]
print(scores)           # print() shows the whole list, including the square brackets
````

````python
# A list can hold any data type you learned in Lesson 1
fruits = ["apple", "banana", "cherry"]      # A list of strings (str)
heights = [175.5, 162.3, 180.1]             # A list of floats
answers = [True, False, True, True]         # A list of booleans (bool)

print(fruits)
print(heights)
print(answers)
````

````python
# Different data types can be mixed in one list
# This is allowed, but usually it is easier to understand if a list holds only one kind of value
student = ["Bill", 20, 175.5, True]         # name, age, height, is_student
print(student)
````

````python
# A list can also be empty (no values at all)
# This is useful when you want to start with nothing and add values later (we will learn how in section 5)
empty_list = []
print(empty_list)
````

````python
# A list is a new data type: list
# type() tells you the data type of a value
print(type(scores))     # <class 'list'> means "this is a list"
print(type(85))         # <class 'int'> means "this is an integer" (one of the values inside the list)
````

## 2. Index: Finding a Value by Its Position

Each value in a list has a position number. This number is called an **index**. Think of it as the number written on each locker.

The most important thing to remember: **Python starts counting from 0, not 1.**

So the first value is at index 0, the second value is at index 1, the third value is at index 2, and so on. (Remember `range(5)` in Lesson 3? It also started from 0: 0, 1, 2, 3, 4.)

```python
fruits = ["apple", "banana", "cherry", "grape", "melon"]
```

| Value | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 |
| Position (how people count) | 1st | 2nd | 3rd | 4th | 5th |

An easy way to remember: **index = position - 1**. The 1st value has index 0, the 5th value has index 4.

To get a value, write the list name followed by the index in square brackets:

```python
fruits[0]     # "the value at index 0 in fruits"
```

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Getting a value by its index
print(fruits[0])        # The 1st value is at index 0 -> apple
print(fruits[1])        # The 2nd value is at index 1 -> banana
print(fruits[2])        # The 3rd value is at index 2 -> cherry
print(fruits[3])        # The 4th value is at index 3 -> grape
print(fruits[4])        # The 5th (last) value is at index 4 -> melon
````

````python
# The same thing with a list of numbers
# Be careful not to mix up the index (position) and the value (what's inside)
numbers = [10, 20, 30, 40]
#  index:   0   1   2   3

print(numbers[0])       # index 0 -> the value is 10
print(numbers[2])       # index 2 -> the value is 30 (not 2!)
print(numbers[3])       # index 3 -> the value is 40
````

````python
# The values don't have to be in any special order
# The index only tells you WHERE the value is, not how big it is
scores = [85, 92, 78, 90, 66]
#  index:  0   1   2   3   4

print(scores[1])        # The 2nd student's score -> 92
print(scores[4])        # The 5th student's score -> 66
````

````python
# You can get the same index as many times as you want
# Getting a value only READS it. The list does not change
colors = ["red", "green", "blue"]

print(colors[1])        # green
print(colors[1])        # still green
print(colors)           # The list is the same as before
````

````python
# A value taken from a list can be used just like a normal variable
scores = [85, 92, 78, 90, 66]

first = scores[0]                  # Store the first score (85) in a new variable
print(first)

print(scores[0] + scores[1])       # 85 + 92 = 177 (you can calculate with them)
print(scores[1] > scores[2])       # 92 > 78 is True (you can compare them, just like in Lesson 2)
````

````python
# Strings taken from a list can be joined with +, just like in Lesson 1
words = ["Hello", "nice", "to", "meet", "you"]

print(words[0] + ", " + words[1] + " " + words[2] + " " + words[3] + " " + words[4])
````

````python
# The index can also be a variable
# Here, i holds 2, so scores[i] is the same as scores[2]
scores = [85, 92, 78, 90, 66]

i = 2
print(scores[i])        # scores[2] -> 78

i = 0
print(scores[i])        # Now i holds 0, so scores[0] -> 85
````

````python
# The index can even be a calculation
# Python first calculates the number inside [ ], and then finds that index
scores = [85, 92, 78, 90, 66]

print(scores[1 + 2])    # 1 + 2 = 3 -> scores[3] -> 90

i = 1
print(scores[i + 1])    # i + 1 = 2 -> scores[2] -> 78
````

### Negative index: counting from the end

Now that you know the normal (positive) index, let's look at one more way to find a value.

Python also lets you count **backwards from the end** of the list, using negative numbers.

- `-1` is the **last** value
- `-2` is the second to last value
- `-3` is the third to last value, and so on

| Value | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 |
| Negative index | -5 | -4 | -3 | -2 | -1 |

Notice that a negative index starts from **-1**, not -0. (There is no such thing as -0, because -0 is just 0, which is the first value.)

**Why use a negative index?**

The main reason is to get the **last value quickly**.

With a positive index, to get the last value you first need to know how many values are in the list. If a list has 5 values, the last index is 4. If it has 100 values, the last index is 99. And if you add or remove values, the last index changes too.

With a negative index, you don't need to count anything. **`-1` is always the last value**, no matter how long the list is.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Negative index: counting from the end
print(fruits[-1])       # -1 is always the last value -> melon
print(fruits[-2])       # -2 is the second to last value -> grape
print(fruits[-3])       # -3 is the third to last value -> cherry
````

````python
# A positive index and a negative index can point to the same value
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[4])        # The last value, using a positive index -> melon
print(fruits[-1])       # The last value, using a negative index -> melon (same!)

print(fruits[0])        # The first value, using a positive index -> apple
print(fruits[-5])       # The first value, using a negative index -> apple (same!)
````

````python
# Why -1 is useful: it works for ANY list, whatever its length
# With a positive index, the last index is different for each list (2, 4, 0 ...)
# With -1, it is always the same
short_list = ["a", "b", "c"]
long_list = [10, 20, 30, 40, 50, 60, 70]
one_item = ["only"]

print(short_list[-1])   # c       (positive index would be 2)
print(long_list[-1])    # 70      (positive index would be 6)
print(one_item[-1])     # only    (positive index would be 0)
````

## 3. Changing Values and len()

### Changing a value

A list is not fixed. You can **change** a value inside it at any time. It's just like putting something new into one locker:

```python
list_name[index] = new_value
```

This works exactly like storing a value in a variable (`=` means "store"), but only the value at that one position is replaced. The other values stay the same.

````python
# Changing a value at a specific index
fruits = ["apple", "banana", "cherry"]
print(fruits)

fruits[1] = "orange"        # Replace the value at index 1 ("banana") with "orange"
print(fruits)               # Only index 1 changed. "apple" and "cherry" are the same as before
````

````python
# A negative index works for changing values too
fruits = ["apple", "banana", "cherry"]

fruits[-1] = "grape"        # Replace the last value ("cherry") with "grape"
print(fruits)
````

````python
# Changing a value using its old value
# Same idea as age = age + 1 in Lesson 1
scores = [85, 92, 78]

scores[2] = scores[2] + 5   # The third student gets 5 bonus points (78 + 5 = 83)
print(scores)
````

### len(): how many values are in the list?

`len()` tells you how many values a list has. `len` is short for **length**.

```python
len(list_name)
```

Put the list inside the parentheses, and `len()` gives back a number: the count of values in that list.

````python
# len(): count how many values are in the list
fruits = ["apple", "banana", "cherry", "grape", "melon"]
print(len(fruits))          # There are 5 values, so 5
````

````python
# Lists of different lengths
three = ["a", "b", "c"]
five = [1, 2, 3, 4, 5]
one = ["only"]
empty_list = []

print(len(three))           # 3 values -> 3
print(len(five))            # 5 values -> 5
print(len(one))             # 1 value -> 1
print(len(empty_list))      # An empty list has 0 values -> 0
````

````python
# len() works the same way for any kind of list
names = ["Bill", "Anna", "Tom", "Kate"]          # A list of strings
numbers = [3, 7, 1, 9, 4, 6]                     # A list of integers
answers = [True, False, True]                    # A list of booleans

print(len(names))           # 4
print(len(numbers))         # 6
print(len(answers))         # 3
````

````python
# len() counts how MANY values there are. It does not add the values up
# The values are big, but there are only 3 of them
prices = [1000, 2500, 3000]
print(len(prices))          # 3 (not 6500!)
````

````python
# The result of len() is just a number (int), so you can store it in a variable
students = ["Bill", "Anna", "Tom", "Kate", "Mike"]

count = len(students)       # Store 5 in the variable count
print(count)
print("There are", count, "students.")
````

````python
# len() also works on strings: it counts the number of characters
# (A space also counts as one character)
print(len("apple"))         # a, p, p, l, e -> 5
print(len("hi there"))      # h, i, (space), t, h, e, r, e -> 8
````

### The last index is len() - 1

Now we can connect `len()` with the index.

If a list has 5 values, `len()` gives 5, but the last index is **4**. Because the index starts at 0, **the last index is always `len() - 1`**.

So there are two ways to get the last value:
- `fruits[len(fruits) - 1]`: count the values, subtract 1, then use that index
- `fruits[-1]`: just use -1

Both give the same value, but `-1` is much shorter and easier. This is exactly why the negative index is useful.

````python
# The last index is len() - 1
fruits = ["apple", "banana", "cherry", "grape", "melon"]

last_index = len(fruits) - 1       # 5 - 1 = 4
print(last_index)
print(fruits[last_index])          # fruits[4] -> melon
````

````python
# Comparing the two ways to get the last value
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[len(fruits) - 1])     # The long way: len(fruits) is 5, 5 - 1 = 4, fruits[4] -> melon
print(fruits[-1])                  # The short way: -1 is the last value -> melon (same result!)
````

## 4. Checking Whether a Value Is in a List (in)

Sometimes you want to know: **"Is this value inside the list?"**

For this, we use `in`.

```python
value in list_name
```

- If the value is in the list, the result is `True`.
- If it is not, the result is `False`.

The result is `True` or `False`, just like the comparison operators in Lesson 2. That means you can use it directly as the condition of an `if`!

To check that a value is **not** in a list, use `not in`.

**Note:** You saw the word `in` in Lesson 3 too (`for i in range(5)`). That `in` is part of the `for` loop. The `in` we learn here is different: it **checks** whether a value is inside a list, and gives True or False.

````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)       # "banana" is in the list -> True
print("grape" in fruits)        # "grape" is not in the list -> False
````

````python
# not in: the opposite of in
print("grape" not in fruits)    # "grape" is NOT in the list -> True
print("apple" not in fruits)    # "apple" IS in the list, so "not in" is False
````

````python
# Using in as the condition of an if (Lesson 2)
fruits = ["apple", "banana", "cherry"]
wanted = "banana"

if wanted in fruits:                    # "banana" in fruits is True
    print("Yes, we have it!")
else:
    print("Sorry, it's sold out.")
````

````python
# A real-life example: checking a guest list
guests = ["Bill", "Anna", "Tom"]
name = "Kate"

if name in guests:                      # "Kate" in guests is False
    print("Welcome!")
else:
    print("Sorry, you are not on the list.")    # So this runs
````

````python
# in checks the value exactly. Upper and lower case letters are different
fruits = ["apple", "banana", "cherry"]
print("Apple" in fruits)        # "Apple" (capital A) and "apple" are different strings -> False
````

## 5. Adding and Removing Values

A list can grow and shrink. You can add new values and remove values you don't need.

| How to write it | What it does |
|---|---|
| `list_name.append(value)` | Adds the value at the **end** of the list |
| `list_name.insert(index, value)` | Adds the value at the **position you choose** |
| `list_name.remove(value)` | Removes the value you name (the first one found) |
| `list_name.pop()` | Takes out the **last** value |
| `list_name.pop(index)` | Takes out the value at the **position you choose** |

The dot `.` means "do this to this list". For example, `fruits.append("grape")` means "add "grape" to the fruits list". Commands written after a dot like this are called **methods**.

These methods change the list itself. You don't need to write `fruits = ...`.

````python
# append(): add a value at the end
fruits = ["apple", "banana"]
print(fruits)

fruits.append("cherry")         # "cherry" is added at the end
print(fruits)

fruits.append("grape")          # "grape" is added at the end again
print(fruits)
print(len(fruits))              # The list grew from 2 to 4 values
````

````python
# Starting with an empty list and adding values one by one
# This is a very common pattern
shopping = []                   # Start with an empty list
shopping.append("milk")
shopping.append("eggs")
shopping.append("bread")
print(shopping)
````

````python
# insert(): add a value at a specific position
# insert(index, value) -> the new value goes to that index, and the values after it move back by one
fruits = ["apple", "banana", "cherry"]

fruits.insert(1, "orange")      # Put "orange" at index 1
print(fruits)                   # "banana" and "cherry" each moved back by one position

fruits.insert(0, "melon")       # Put "melon" at index 0 (the very front)
print(fruits)
````

````python
# remove(): remove a value by naming the value itself
fruits = ["apple", "banana", "cherry"]

fruits.remove("banana")         # Find "banana" and remove it
print(fruits)                   # The values after it move forward to fill the gap
````

````python
# If the same value appears more than once, remove() removes only the FIRST one
numbers = [1, 2, 3, 2, 1]

numbers.remove(2)               # Only the first 2 (at index 1) is removed
print(numbers)                  # The second 2 is still there
````

````python
# pop(): take out a value by its position
# Without an index, pop() takes out the LAST value
fruits = ["apple", "banana", "cherry"]

fruits.pop()                    # Take out the last value ("cherry")
print(fruits)
````

````python
# pop(index): take out the value at that index
fruits = ["apple", "banana", "cherry"]

fruits.pop(0)                   # Take out the value at index 0 ("apple")
print(fruits)
````

````python
# pop() also GIVES BACK the value it took out, so you can store it in a variable
# remove() only removes. pop() removes AND gives the value to you
tasks = ["wash dishes", "do homework", "clean room"]

done = tasks.pop(0)             # Take out the first task and store it in done
print("Finished:", done)
print("Still to do:", tasks)
````

````python
# remove() vs pop() at a glance
# remove(value): "remove THIS VALUE"       -> you know the value
# pop(index):    "remove what's at THIS POSITION" -> you know the position
colors = ["red", "green", "blue"]

colors.remove("green")          # Removing by value
print(colors)

colors.pop(0)                   # Removing by position (index 0 is now "red")
print(colors)
````

## 6. Common Mistakes

These are the mistakes almost everyone makes when learning lists. If you get an error, check these first.

1. **Forgetting that the index starts at 0**: `fruits[1]` is the **second** value, not the first.
2. **Using an index that doesn't exist**: If a list has 3 values, the indexes are 0, 1, 2. `fruits[3]` causes `IndexError: list index out of range` ("the index is outside the range of the list").
3. **Trying to add a value with a new index**: `fruits[3] = "grape"` on a list with 3 values does **not** add a value. It causes `IndexError: list assignment index out of range` ("you can't store a value at an index outside the list"). To add, use `append()`.
4. **Removing a value that isn't in the list**: `fruits.remove("grape")` causes `ValueError: list.remove(x): x not in list` ("the value is not in the list"). Check with `in` first.
5. **Mixing up `( )` and `[ ]`**: Use `[ ]` to make a list and to get a value by index (`fruits[0]`). Use `( )` for commands like `len(fruits)` and `fruits.append("grape")`.

The cells below show what each mistake looks like. The broken code is written as comments so it won't break the notebook. Try removing the `#` from the code line to see the real error message, and then put it back.

````python
# Mistake 1: forgetting that the index starts at 0
fruits = ["apple", "banana", "cherry"]

print(fruits[1])        # Many people expect "apple", but this is the SECOND value -> banana
print(fruits[0])        # Correct: the first value is at index 0 -> apple
````

````python
# Mistake 2: using an index that doesn't exist
fruits = ["apple", "banana", "cherry"]      # 3 values -> indexes are 0, 1, 2

# print(fruits[3])      # IndexError: list index out of range -> there is no index 3!

print(fruits[2])                    # Correct: the last index is 2
print(fruits[len(fruits) - 1])      # Also correct: len(fruits) - 1 = 3 - 1 = 2
print(fruits[-1])                   # Also correct: -1 is always the last value
````

````python
# Mistake 3: trying to add a value with a new index
fruits = ["apple", "banana", "cherry"]

# fruits[3] = "grape"   # IndexError: list assignment index out of range -> you can only CHANGE values at indexes that already exist

fruits.append("grape")  # Correct: use append() to add a new value at the end
print(fruits)
````

````python
# Mistake 4: removing a value that isn't in the list
fruits = ["apple", "banana", "cherry"]

# fruits.remove("grape")    # ValueError: list.remove(x): x not in list -> "grape" is not in the list!

# Correct: check with in first (section 4), and remove only if it is there
if "grape" in fruits:
    fruits.remove("grape")
else:
    print("grape is not in the list, so nothing was removed.")
````

````python
# Mistake 5: mixing up ( ) and [ ]
fruits = ["apple", "banana", "cherry"]

# print(fruits(0))          # TypeError: 'list' object is not callable -> ( ) is for commands, not for indexes
# fruits.append["grape"]    # TypeError -> append is a command, so it needs ( ), not [ ]

print(fruits[0])            # Correct: [ ] to get a value by index
fruits.append("grape")      # Correct: ( ) for a command
print(fruits)
````
