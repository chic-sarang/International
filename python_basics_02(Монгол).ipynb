## 1. if хэллэг гэж юу вэ?

Одоог хүртэл бидний код дээрээс доош, мөр мөрөөр, үргэлж ижилхэн ажиллаж ирсэн. Гэтэл бодит амьдралд бид байнга сонголт хийдэг.

- **Хэрэв** бороо орж байвал шүхрээ ав.
- **Хэрэв** чи 20 ба түүнээс дээш настай бол том хүний тасалбар авч болно.

`if` хэллэг нь програмд ийм сонголт хийх боломж олгодог. Энэ нь **нөхцөл**-ийг шалгаж, **зөвхөн нөхцөл True байх үед** дотор нь байгаа кодыг ажиллуулна.

Хичээл 1-ийн харьцуулах операторуудыг (`>`, `<`, `==`, `!=`) санаж байна уу? Тэдгээрийн үр дүн үргэлж `True` эсвэл `False` байдаг. Яг үүнийг л `if` нөхцөл болгон шаарддаг.

Үндсэн хэлбэр нь ийм байна:

```python
if нөхцөл:
    зөвхөн нөхцөл True байх үед ажиллах код
```

Маш чухал хоёр дүрэм:
- `if` мөрийн төгсгөлд заавал хоёр цэг `:` байх ёстой.
- `if`-д хамаарах код **догол мөртэй** (баруун тийш 4 зайгаар шилжүүлсэн) байх ёстой. Догол мөрөөр Python аль мөрүүд `if`-ийн "дотор" байгааг мэддэг.

````python
# Нөхцөл нь age >= 20 ("нас 20-оос их буюу тэнцүү")
# Энэ харьцуулалтын үр дүн True эсвэл False байна
age = 25

if age >= 20:                         # 25 >= 20 нь True, тиймээс доорх догол мөртэй код ажиллана
    print("You are an adult.")        # Энэ мөр if-ийн дотор байна (4 зайн догол мөртэй)

print("This line always runs.")      # Энэ мөр догол мөргүй тул if-ийн гадна байна. Юу ч болсон ажиллана
````

````python
# Одоо нөхцөл False байх тохиолдлыг туршиж үзье
age = 15

if age >= 20:                         # 15 >= 20 нь False, тиймээс доорх догол мөртэй код алгасагдана
    print("You are an adult.")        # Энэ мөр огт ажиллахгүй

print("This line always runs.")      # Энэ мөр if-ийн гадна байгаа тул ажилласаар байна
````

````python
# if-ийн дотор нэгээс олон мөр байж болно
# Ижил догол мөртэй бүх мөр нэг if-д хамаарна
temperature = 32

if temperature > 30:                          # 32 > 30 нь True
    print("It's very hot today.")             # if-ийн дотор
    print("Drink plenty of water.")           # Мөн if-ийн дотор (ижил догол мөр)
    print("Try to stay in the shade.")        # Мөн if-ийн дотор (ижил догол мөр)

print("Have a nice day!")                    # if-ийн гадна, үргэлж ажиллана
````

````python
# == ашиглан хоёр утга ижил эсэхийг ч шалгаж болно
# Санаарай: = нь "утга хадгалах", харин == нь "эдгээр утга тэнцүү юу?" гэсэн утгатай
password = "python123"

if password == "python123":           # Хоёр тэмдэгт мөр яг ижил тул энэ нь True
    print("Login successful!")
````

## 2. if - else

Зөвхөн `if` байвал програм нөхцөл True үед ямар нэг юм хийж, False үед **юу ч хийхгүй**.

Гэхдээ бид ихэвчлэн **энийг эсвэл тэрийг** хийхийг хүсдэг. Жишээ нь:
- Шалгалтад тэнцсэн бол "Баяр хүргэе" гэж хэвлэ. **Үгүй бол** "Дахин оролдоорой" гэж хэвлэ.

Үүний тулд `else`-г ашигладаг. `else` нь "үгүй бол" гэсэн утгатай. Түүний код **зөвхөн if-ийн нөхцөл False байх үед** ажиллана.

```python
if нөхцөл:
    нөхцөл True үед ажиллах код
else:
    нөхцөл False үед ажиллах код
```

- `else` өөрийн гэсэн нөхцөлгүй. Энэ нь `if`-ийн барьж аваагүй бүх тохиолдлыг л хүлээн авдаг.
- `else`-ийн төгсгөлд мөн хоёр цэг `:` хэрэгтэй бөгөөд түүний код ч догол мөртэй байх ёстой.
- Хоёр блокийн яг **нэг** нь үргэлж ажиллана. Хэзээ ч хоёулаа биш, хэзээ ч аль нь ч биш гэж байхгүй.

````python
# Шалгалтын оноо тэнцсэн эсвэл унасан эсэхийг шалгах
# Тэнцэх оноо 60 гэж үзье
score = 75

if score >= 60:                      # 75 >= 60 нь True
    print("Pass! Congratulations.")  # Тиймээс энэ мөр ажиллана
else:                                # "Үгүй бол" - if True байсан тул энэ хэсэг алгасагдана
    print("Fail. Try again.")
````

````python
# Ижил код, гэхдээ бага оноотой
score = 40

if score >= 60:                      # 40 >= 60 нь False, тиймээс энэ хэсэг алгасагдана
    print("Pass! Congratulations.")
else:                                # if False байсан тул оронд нь else хэсэг ажиллана
    print("Fail. Try again.")
````

````python
# Тоо тэгш эсвэл сондгой эсэхийг шалгах
# Хичээл 1-ийн %-ийг санаж байна уу? Энэ нь хуваалтын үлдэгдлийг өгдөг
# Тэгш тоог 2-т хуваахад үлдэгдэл үргэлж 0 байна
number = 7

if number % 2 == 0:                  # 7 % 2 нь 1, харин 1 == 0 нь False
    print(number, "is even.")        # print() нь таслалаар тусгаарласан хэд хэдэн утгыг нэг дор харуулж чадна
else:
    print(number, "is odd.")         # if False байсан тул энэ мөр ажиллана
````

## 3. if - elif - else

Заримдаа сонголт **хоёроос олон** байдаг. Жишээ нь, оноог үнэлгээ болгох:

- 90 ба түүнээс дээш → A
- 80 ба түүнээс дээш → B
- 70 ба түүнээс дээш → C
- Бусад бүх тохиолдол → F

Үүнд `elif`-г ашигладаг. `elif` нь "else if"-ийн товчлол бөгөөд "дээрх нөхцөлүүд False байсан бол энэ нөхцөлийг шалга" гэсэн утгатай.

```python
if нөхцөл1:
    нөхцөл1 True үед ажиллана
elif нөхцөл2:
    нөхцөл1 False, нөхцөл2 True үед ажиллана
elif нөхцөл3:
    нөхцөл1 ба нөхцөл2 False, нөхцөл3 True үед ажиллана
else:
    дээрх бүх нөхцөл False үед ажиллана
```

Хэрхэн ажилладаг вэ:
- Python нөхцөлүүдийг **дээрээс доош, дарааллаар** шалгана.
- **Анхны** True нөхцөлийг олмогц тэр блокийг ажиллуулаад **үлдсэн бүгдийг алгасна**.
- `elif`-г хүссэн хэмжээгээрээ нэмж болно. Төгсгөлийн `else` заавал байх албагүй.

````python
# Оноог үнэлгээ болгох
score = 85

if score >= 90:                # 85 >= 90 нь False -> дараагийн нөхцөл рүү шилжинэ
    print("Grade: A")
elif score >= 80:              # 85 >= 80 нь True -> энэ блок ажиллаж, доорх бүгд алгасагдана
    print("Grade: B")
elif score >= 70:              # Дээр True нөхцөл аль хэдийн олдсон тул огт шалгагдахгүй
    print("Grade: C")
else:                          # Мөн шалгагдахгүй
    print("Grade: F")
````

````python
# Ижил кодоор өөр өөр оноо туршиж үзээрэй
# score-ийн утгыг өөрчлөөд дахин ажиллуулж үр дүн хэрхэн өөрчлөгдөхийг хараарай
score = 72

if score >= 90:                # 72 >= 90 нь False
    print("Grade: A")
elif score >= 80:              # 72 >= 80 нь False
    print("Grade: B")
elif score >= 70:              # 72 >= 70 нь True -> энэ ажиллана
    print("Grade: C")
else:
    print("Grade: F")
````

````python
# Дараалал яагаад чухал вэ?
# Python АНХНЫ True нөхцөл дээр зогсдог тул хамгийн бага шалгалтыг эхэнд тавибал буруу болно
score = 95

if score >= 70:                # 95 >= 70 нь True -> энэ хамгийн түрүүнд ажиллаж, бусад бүгд алгасагдана!
    print("Grade: C")
elif score >= 80:              # Хэзээ ч шалгагдахгүй
    print("Grade: B")
elif score >= 90:              # 95 нь A авах ёстой ч хэзээ ч шалгагдахгүй
    print("Grade: A")

# Тиймээс ийм хязгаар шалгахдаа хамгийн том (хамгийн хатуу) нөхцөлөөс эхлээрэй
````

````python
# Агаарын температураас хамаарч юу өмсөхөө шийдэх
temperature = 12

if temperature >= 25:                  # 12 >= 25 нь False
    print("Wear a T-shirt.")
elif temperature >= 15:                # 12 >= 15 нь False
    print("Wear a light jacket.")
elif temperature >= 5:                 # 12 >= 5 нь True -> энэ ажиллана
    print("Wear a coat.")
else:                                  # Зөвхөн температур 5-аас доош үед ажиллана
    print("Wear a thick padded jacket.")
````

## 4. bool хувьсагчийг нөхцөл болгон ашиглах

`if`-ийн нөхцөл зөвхөн `True` эсвэл `False` байхад л хангалттай. Энэ нь **bool хувьсагч**-ийг (Хичээл 1-ээс) ямар ч харьцуулалтгүйгээр шууд нөхцөл болгон ашиглаж болно гэсэн үг.

```python
if is_student:        # дараахтай ижил утгатай: if is_student == True:
```

`== True` гэж бичих нь буруу биш, гэхдээ шаардлагагүй тул Python програмистууд ихэвчлэн бичдэггүй.

bool хувьсагч `False` эсэхийг шалгахын тулд `not`-ийг ашиглана:

```python
if not is_student:    # дараахтай ижил утгатай: if is_student == False:
```

````python
# bool хувьсагч аль хэдийн True эсвэл False-г хадгалж байгаа тул шууд нөхцөл болгон ашиглаж болно
is_student = True

if is_student:                        # is_student нь True тул энэ ажиллана
    print("You get a student discount.")
else:
    print("Regular price.")
````

````python
# False-г шалгахын тулд not ашиглах
# not нь утгыг эсрэгээр нь эргүүлнэ: not True нь False, not False нь True
is_raining = False

if not is_raining:                    # is_raining нь False тул not is_raining нь True -> энэ ажиллана
    print("No umbrella needed.")
else:
    print("Take an umbrella.")
````

````python
# Харьцуулалтын үр дүнг эхлээд bool хувьсагчид хадгалаад дараа нь ашиглаж болно
# Нөхцөл урт үед эсвэл түүнд ойлгомжтой нэр өгөхийг хүсвэл энэ нь хэрэгтэй
age = 17
is_adult = age >= 20                  # 17 >= 20 нь False тул is_adult-д False хадгалагдана
print(is_adult)

if is_adult:                          # is_adult нь False тул энэ хэсэг алгасагдана
    print("You can watch this movie.")
else:
    print("You can't watch this movie.")
````

## 5. Нөхцөлүүдийг нэгтгэх (and, or, not)

Заримдаа нэг нөхцөл хангалтгүй байдаг. Жишээ нь:
- Чи **10 ба түүнээс дээш настай, мөн 140 см ба түүнээс өндөр** бол галзуу хулгана (roller coaster) унаж болно.
- Чи **7-оос доош настай эсвэл 65 ба түүнээс дээш настай** бол үнэгүй нэвтэрнэ.

Python-д нөхцөлүүдийг нэгтгэх гурван үг бий.

| Оператор | Утга | Үр дүн |
|---|---|---|
| `and` | ба | Зөвхөн **хоёр** тал True үед True |
| `or` | эсвэл | **Ядаж нэг** тал True үед True |
| `not` | эсрэгээр | True-г False, False-г True болгоно |

````python
# Эхлээд and / or дангаараа ямар үр дүн өгөхийг харъя
print(True and True)     # хоёулаа True -> True
print(True and False)    # нэг нь False -> False
print(True or False)     # ядаж нэг нь True -> True
print(False or False)    # хоёулаа False -> False
print(not True)          # True-ийн эсрэг -> False
````

````python
# and: бүх нөхцөл True байх ёстой
# Дүрэм: 10 ба түүнээс дээш настай БА 140 см ба түүнээс өндөр
age = 12
height = 135

if age >= 10 and height >= 140:       # 12 >= 10 нь True, гэхдээ 135 >= 140 нь False -> True and False -> False
    print("You can ride the roller coaster.")
else:
    print("Sorry, you can't ride it yet.")  # Нөхцөл False байсан тул энэ ажиллана
````

````python
# or: ядаж нэг нөхцөл True байх ёстой
# Дүрэм: 7-оос доош ЭСВЭЛ 65 ба түүнээс дээш настай бол үнэгүй нэвтэрнэ
age = 70

if age < 7 or age >= 65:              # 70 < 7 нь False, гэхдээ 70 >= 65 нь True -> False or True -> True
    print("Free entry!")              # Нөхцөл True байсан тул энэ ажиллана
else:
    print("Please buy a ticket.")
````

````python
# Тоо тодорхой хязгаарт багтаж байгаа эсэхийг шалгах
# "score нь 0-ээс 100-ийн хооронд" гэдэг нь "score 0 ба түүнээс их БА score 100 ба түүнээс бага" гэсэн үг
score = 105

if score >= 0 and score <= 100:       # 105 >= 0 нь True, гэхдээ 105 <= 100 нь False -> False
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")

# Python-д үүнийг математик шиг илүү товч бичиж болно
if 0 <= score <= 100:                 # Дээрх мөртэй яг ижил утгатай
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")
````

## 6. Давхар (үүрлэсэн) if хэллэг

Нэг `if`-г **өөр нэг** `if`-ийн дотор хийж болно. Үүнийг **давхар (үүрлэсэн)** if гэдэг.

Эхний нөхцөл True болсны **дараа л** хоёр дахь нөхцөлийг шалгахыг хүсвэл энэ нь хэрэгтэй. Жишээ нь, эхлээд хэрэглэгч нэвтэрсэн эсэхийг шалгаж, зөвхөн дараа нь админ эсэхийг шалгах.

Дотоод `if` нь гадаад `if`-ээс нэг шат илүү (4 зай илүү) догол мөртэй байна.

````python
# Эхлээд нэвтрэлтийг шалгаж, нэвтэрсэн тохиолдолд л админ эрхийг шалгана
is_logged_in = True
is_admin = False

if is_logged_in:                               # Гадаад if: True тул дотогш орно
    print("Welcome!")                          # Гадаад if-д хамаарна (4 зай)
    if is_admin:                               # Дотоод if: is_admin нь False
        print("You can open the admin page.")  # Дотоод if-д хамаарна (8 зай) - алгасагдана
    else:
        print("You are a regular user.")       # Дотоод else-д хамаарна (8 зай) - энэ ажиллана
else:                                          # Гадаад if-д хамаарна
    print("Please log in first.")              # Гадаад if True байсан тул алгасагдана
````

````python
# Дэлгүүрээс юм худалдаж авах
# 1-р алхам: бараа байгаа юу?  2-р алхам: мөнгө хүрэлцэх үү?
in_stock = True
money = 8000
price = 12000

if in_stock:                                   # Бараа байгаа -> дотогш орно
    if money >= price:                         # 8000 >= 12000 нь False
        print("Purchase complete!")
    else:
        print("You don't have enough money.")  # Энэ ажиллана
        print("You need", price - money, "more won.")  # 12000 - 8000 = 4000
else:
    print("Sorry, this item is sold out.")
````

**Зөвлөгөө:** Давхар if хэрэгтэй ч хэт олон шат байвал кодыг уншихад хэцүү болгодог. Боломжтой үед нөхцөлүүдийг `and`-аар нэгтгэхийг оролдоорой.

```python
# давхар хэлбэр
if is_logged_in:
    if is_admin:
        print("You can open the admin page.")

# and ашигласан ижил утгатай хэлбэр
if is_logged_in and is_admin:
    print("You can open the admin page.")
```

## 7. Түгээмэл алдаанууд

`if`-г сурч байхдаа бараг бүх хүн эдгээр алдааг гаргадаг. Алдаа гарвал эхлээд эдгээрийг шалгаарай.

1. **`==`-ийн оронд `=` ашиглах**: `=` нь утга хадгалдаг, `==` нь утгуудыг харьцуулдаг. `if age = 20:` нь алдаа.
2. **Хоёр цэг `:`-ийг мартах**: `if`, `elif`, `else` мөрүүд бүгд `:`-ээр төгсөх ёстой.
3. **Буруу догол мөр**: Доторх код догол мөртэй байх ёстой бөгөөд нэг блокийн мөрүүд ижил хэмжээний догол мөртэй байх ёстой.
4. **`else`-д нөхцөл өгөх**: `else` хэзээ ч нөхцөлгүй. `else age < 20:` нь алдаа. Нөхцөл хэрэгтэй бол `elif` ашиглаарай.
5. **Тоог тэмдэгт мөртэй харьцуулах**: `"20"` (тэмдэгт мөр) ба `20` (бүхэл тоо) нь өөр. `"20" == 20` нь `False`.

Доорх нүднүүд алдаа бүр ямар харагддагийг үзүүлнэ. Notebook эвдрэхгүйн тулд алдаатай кодыг тайлбар хэлбэрээр бичсэн. Жинхэнэ алдааны мэдэгдлийг харахын тулд кодын мөрөөс `#`-г арилгаж үзээд, дараа нь буцааж тавиарай.

````python
# Алдаа 1: == -ийн оронд = ашиглах
age = 20

# if age = 20:          # SyntaxError! = нь "хадгалах" гэсэн утгатай тул нөхцөл болгон ашиглаж болохгүй
#     print("20 years old")

if age == 20:           # Зөв: == нь "тэнцүү юу?" гэсэн утгатай
    print("20 years old")
````

````python
# Алдаа 2: хоёр цэгийг мартах
age = 20

# if age >= 20          # SyntaxError! Төгсгөлд хоёр цэг дутуу байна
#     print("adult")

if age >= 20:           # Зөв: мөр хоёр цэгээр төгсөж байна
    print("adult")
````

````python
# Алдаа 3: буруу догол мөр
age = 20

# if age >= 20:
# print("adult")        # IndentationError! if-ийн доторх мөр догол мөргүй байна

if age >= 20:
    print("adult")      # Зөв: 4 зайн догол мөртэй
````

````python
# Алдаа 4: else-д нөхцөл өгөх
age = 15

# if age >= 20:
#     print("adult")
# else age < 20:        # SyntaxError! else нөхцөлтэй байж болохгүй
#     print("minor")

if age >= 20:
    print("adult")
else:                   # Зөв: else нөхцөлгүй. if-ийн барьж аваагүй бүх тохиолдлыг хүлээн авна
    print("minor")
````

````python
# Алдаа 5: тоог тэмдэгт мөртэй харьцуулах
# Алдааны мэдэгдэл гарахгүй ч үр дүн нь таны бодсоноос өөр байх тул энэ алдаа их зальтай
age_text = "20"          # Энэ нь хашилтанд байгаа тул тэмдэгт мөр (Хичээл 1-ийн zip_code-г санаарай)
age_number = 20          # Энэ нь бүхэл тоо

print(age_text == age_number)     # False! Тэмдэгт мөр ба бүхэл тоо хэзээ ч тэнцүү биш

if age_text == age_number:
    print("Same")
else:
    print("Different")            # Харахад ижил мэт боловч энэ мөр ажиллана
````
