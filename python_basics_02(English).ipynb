## 1. What Is an if Statement?

Until now, our code ran from top to bottom, one line after another, every single time. But in real life, we make choices all the time.

- **If** it is raining, take an umbrella.
- **If** you are 20 or older, you can buy a ticket for adults.

The `if` statement lets the program make choices like this. It checks a **condition**, and runs the code inside it **only when the condition is True**.

Remember the comparison operators from Lesson 1 (`>`, `<`, `==`, `!=`)? Their result is always `True` or `False`. That is exactly what `if` needs as a condition.

The basic shape looks like this:

```python
if condition:
    code that runs only when the condition is True
```

Two very important rules:
- There must be a colon `:` at the end of the `if` line.
- The code that belongs to the `if` must be **indented** (pushed to the right by 4 spaces). Indentation is how Python knows which lines are "inside" the `if`.

````python
# The condition is age >= 20 ("age is greater than or equal to 20")
# The result of this comparison is True or False
age = 25

if age >= 20:                         # 25 >= 20 is True, so the indented code below runs
    print("You are an adult.")        # This line is inside the if (it is indented by 4 spaces)

print("This line always runs.")      # This line is NOT indented, so it is outside the if. It runs no matter what
````

````python
# Now let's try a case where the condition is False
age = 15

if age >= 20:                         # 15 >= 20 is False, so the indented code below is skipped
    print("You are an adult.")        # This line does not run at all

print("This line always runs.")      # This line is outside the if, so it still runs
````

````python
# More than one line can be inside an if
# Every line with the same indentation belongs to the same if
temperature = 32

if temperature > 30:                          # 32 > 30 is True
    print("It's very hot today.")             # Inside the if
    print("Drink plenty of water.")           # Also inside the if (same indentation)
    print("Try to stay in the shade.")        # Also inside the if (same indentation)

print("Have a nice day!")                    # Outside the if, always runs
````

````python
# You can also check whether two values are the same with ==
# Remember: = means "store a value", but == means "are these equal?"
password = "python123"

if password == "python123":           # The two strings are exactly the same, so this is True
    print("Login successful!")
````

## 2. if - else

With only `if`, the program does something when the condition is True, and does **nothing** when it is False.

But often we want to do **one thing or the other**. For example:
- If you passed, print "Congratulations". **Otherwise**, print "Try again".

This is what `else` is for. `else` means "otherwise". Its code runs **only when the if condition is False**.

```python
if condition:
    code that runs when the condition is True
else:
    code that runs when the condition is False
```

- `else` does not have its own condition. It simply catches every case the `if` did not.
- `else` also needs a colon `:` at the end, and its code must be indented too.
- Exactly **one** of the two blocks always runs. Never both, never neither.

````python
# Checking whether an exam score is a pass or a fail
# Let's say the passing score is 60
score = 75

if score >= 60:                      # 75 >= 60 is True
    print("Pass! Congratulations.")  # So this runs
else:                                # "Otherwise" - this part is skipped because the if was True
    print("Fail. Try again.")
````

````python
# Same code, but with a lower score
score = 40

if score >= 60:                      # 40 >= 60 is False, so this part is skipped
    print("Pass! Congratulations.")
else:                                # The if was False, so the else part runs instead
    print("Fail. Try again.")
````

````python
# Checking whether a number is even or odd
# Remember % from Lesson 1? It gives the remainder of a division
# An even number divided by 2 always has a remainder of 0
number = 7

if number % 2 == 0:                  # 7 % 2 is 1, and 1 == 0 is False
    print(number, "is even.")        # print() can show several values at once if you separate them with commas
else:
    print(number, "is odd.")         # The if was False, so this runs
````

## 3. if - elif - else

Sometimes there are **more than two** choices. For example, turning a score into a grade:

- 90 or above → A
- 80 or above → B
- 70 or above → C
- Anything else → F

For this we use `elif`. `elif` is short for "else if", and it means "if the conditions above were False, then check this condition".

```python
if condition1:
    runs when condition1 is True
elif condition2:
    runs when condition1 is False and condition2 is True
elif condition3:
    runs when condition1 and condition2 are False and condition3 is True
else:
    runs when every condition above is False
```

How it works:
- Python checks the conditions **from top to bottom, in order**.
- As soon as it finds the **first** True condition, it runs that block and **skips all the rest**.
- You can add as many `elif` as you want. The `else` at the end is optional.

````python
# Turning a score into a grade
score = 85

if score >= 90:                # 85 >= 90 is False -> move on to the next condition
    print("Grade: A")
elif score >= 80:              # 85 >= 80 is True -> this block runs, and everything below is skipped
    print("Grade: B")
elif score >= 70:              # Not checked at all, because a True condition was already found above
    print("Grade: C")
else:                          # Not checked either
    print("Grade: F")
````

````python
# Try a few different scores with the same code
# Change the value of score and run it again to see how the result changes
score = 72

if score >= 90:                # 72 >= 90 is False
    print("Grade: A")
elif score >= 80:              # 72 >= 80 is False
    print("Grade: B")
elif score >= 70:              # 72 >= 70 is True -> this one runs
    print("Grade: C")
else:
    print("Grade: F")
````

````python
# Why does the order matter?
# Python stops at the FIRST True condition, so if we put the smallest check first, things go wrong
score = 95

if score >= 70:                # 95 >= 70 is True -> this runs first and everything else is skipped!
    print("Grade: C")
elif score >= 80:              # Never checked
    print("Grade: B")
elif score >= 90:              # Never checked, even though 95 deserves an A
    print("Grade: A")

# So when you check ranges like this, start with the biggest (strictest) condition first
````

````python
# Deciding what to wear depending on the temperature
temperature = 12

if temperature >= 25:                  # 12 >= 25 is False
    print("Wear a T-shirt.")
elif temperature >= 15:                # 12 >= 15 is False
    print("Wear a light jacket.")
elif temperature >= 5:                 # 12 >= 5 is True -> this runs
    print("Wear a coat.")
else:                                  # Only runs when the temperature is below 5
    print("Wear a thick padded jacket.")
````

## 4. Using bool Variables as Conditions

The condition of an `if` just needs to be `True` or `False`. That means a **bool variable** (from Lesson 1) can be used as a condition directly, without any comparison.

```python
if is_student:        # same meaning as: if is_student == True:
```

Writing `== True` is not wrong, but it is unnecessary, so Python programmers usually leave it out.

To check that a bool variable is `False`, use `not`:

```python
if not is_student:    # same meaning as: if is_student == False:
```

````python
# A bool variable already holds True or False, so it can be used as the condition as it is
is_student = True

if is_student:                        # is_student is True, so this runs
    print("You get a student discount.")
else:
    print("Regular price.")
````

````python
# Using not to check for False
# not flips the value: not True is False, and not False is True
is_raining = False

if not is_raining:                    # is_raining is False, so not is_raining is True -> this runs
    print("No umbrella needed.")
else:
    print("Take an umbrella.")
````

````python
# You can store the result of a comparison in a bool variable first, and then use it
# This is useful when the condition is long, or when you want to give it a clear name
age = 17
is_adult = age >= 20                  # 17 >= 20 is False, so is_adult stores False
print(is_adult)

if is_adult:                          # is_adult is False, so this part is skipped
    print("You can watch this movie.")
else:
    print("You can't watch this movie.")
````

## 5. Combining Conditions (and, or, not)

Sometimes one condition is not enough. For example:
- You can ride the roller coaster if you are **at least 10 years old and at least 140 cm tall**.
- You get free entry if you are **under 7 or 65 and over**.

Python has three words for combining conditions.

| Operator | Meaning | Result |
|---|---|---|
| `and` | both | True only when **both** sides are True |
| `or` | either | True when **at least one** side is True |
| `not` | the opposite | Flips True to False and False to True |

````python
# First, let's see what and / or give us on their own
print(True and True)     # both are True -> True
print(True and False)    # one is False -> False
print(True or False)     # at least one is True -> True
print(False or False)    # both are False -> False
print(not True)          # the opposite of True -> False
````

````python
# and: every condition must be True
# Rule: at least 10 years old AND at least 140 cm tall
age = 12
height = 135

if age >= 10 and height >= 140:       # 12 >= 10 is True, but 135 >= 140 is False -> True and False -> False
    print("You can ride the roller coaster.")
else:
    print("Sorry, you can't ride it yet.")  # The condition was False, so this runs
````

````python
# or: at least one condition must be True
# Rule: under 7 OR 65 and over gets free entry
age = 70

if age < 7 or age >= 65:              # 70 < 7 is False, but 70 >= 65 is True -> False or True -> True
    print("Free entry!")              # The condition was True, so this runs
else:
    print("Please buy a ticket.")
````

````python
# Checking whether a number is inside a range
# "score is between 0 and 100" means "score is 0 or more AND score is 100 or less"
score = 105

if score >= 0 and score <= 100:       # 105 >= 0 is True, but 105 <= 100 is False -> False
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")

# Python also lets you write the same thing in a shorter way, just like in math
if 0 <= score <= 100:                 # Exactly the same meaning as the line above
    print("Valid score.")
else:
    print("Invalid score. It must be between 0 and 100.")
````

## 6. Nested if Statements

You can put an `if` **inside another** `if`. This is called a **nested** if.

It is useful when you want to check the second condition **only after** the first condition is True. For example, first check whether the user is logged in, and only then check whether they are an administrator.

The inner `if` is indented one more step (4 more spaces) than the outer `if`.

````python
# First check the login, and only if the login is OK, check the admin status
is_logged_in = True
is_admin = False

if is_logged_in:                               # Outer if: True, so go inside
    print("Welcome!")                          # Belongs to the outer if (4 spaces)
    if is_admin:                               # Inner if: is_admin is False
        print("You can open the admin page.")  # Belongs to the inner if (8 spaces) - skipped
    else:
        print("You are a regular user.")       # Belongs to the inner else (8 spaces) - this runs
else:                                          # Belongs to the outer if
    print("Please log in first.")              # Skipped, because the outer if was True
````

````python
# Buying something in a shop
# Step 1: is the item in stock?  Step 2: do you have enough money?
in_stock = True
money = 8000
price = 12000

if in_stock:                                   # The item is in stock -> go inside
    if money >= price:                         # 8000 >= 12000 is False
        print("Purchase complete!")
    else:
        print("You don't have enough money.")  # This runs
        print("You need", price - money, "more won.")  # 12000 - 8000 = 4000
else:
    print("Sorry, this item is sold out.")
````

**Tip:** Nested ifs are useful, but too many levels make code hard to read. When it is possible, try combining the conditions with `and` instead.

```python
# nested
if is_logged_in:
    if is_admin:
        print("You can open the admin page.")

# the same thing with and
if is_logged_in and is_admin:
    print("You can open the admin page.")
```

## 7. Common Mistakes

These are the mistakes almost everyone makes when learning `if`. If you get an error, check these first.

1. **Using `=` instead of `==`**: `=` stores a value, `==` compares values. `if age = 20:` is an error.
2. **Forgetting the colon `:`**: `if`, `elif`, and `else` lines must all end with `:`.
3. **Wrong indentation**: The code inside must be indented, and lines in the same block must be indented by the same amount.
4. **Giving `else` a condition**: `else` never has a condition. `else age < 20:` is an error. Use `elif` if you need a condition.
5. **Comparing a number with a string**: `"20"` (a string) and `20` (an integer) are different. `"20" == 20` is `False`.

The cells below show what each mistake looks like. They are written as comments so they won't break the notebook. Try removing the `#` from the code line to see the real error message, and then put it back.

````python
# Mistake 1: using = instead of ==
age = 20

# if age = 20:          # SyntaxError! = means "store", so it cannot be used as a condition
#     print("20 years old")

if age == 20:           # Correct: == means "are they equal?"
    print("20 years old")
````

````python
# Mistake 2: forgetting the colon
age = 20

# if age >= 20          # SyntaxError! The colon at the end is missing
#     print("adult")

if age >= 20:           # Correct: the line ends with a colon
    print("adult")
````

````python
# Mistake 3: wrong indentation
age = 20

# if age >= 20:
# print("adult")        # IndentationError! The line inside the if is not indented

if age >= 20:
    print("adult")      # Correct: indented by 4 spaces
````

````python
# Mistake 4: giving else a condition
age = 15

# if age >= 20:
#     print("adult")
# else age < 20:        # SyntaxError! else cannot have a condition
#     print("minor")

if age >= 20:
    print("adult")
else:                   # Correct: else has no condition. It covers everything the if didn't
    print("minor")
````

````python
# Mistake 5: comparing a number with a string
# No error appears, but the result is not what you expected, so this one is sneaky
age_text = "20"          # This is a string, because it is inside quotes (remember zip_code in Lesson 1)
age_number = 20          # This is an integer

print(age_text == age_number)     # False! A string and an integer are never equal

if age_text == age_number:
    print("Same")
else:
    print("Different")            # This runs, even though they look the same
````
