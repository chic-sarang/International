## 1. list ဆိုတာဘာလဲ?

သင်ခန်းစာ 1 မှာ variable ဆိုတာ တန်ဖိုး **တစ်ခု** ကို သိမ်းထားတဲ့ သေတ္တာနဲ့ တူတယ်လို့ လေ့လာခဲ့ပါတယ်။

ဒါပေမဲ့ ကျောင်းသား 5 ယောက်ရဲ့ စာမေးပွဲ အမှတ်တွေကို သိမ်းချင်ရင် ဘယ်လိုလုပ်မလဲ? အခုထိ သိထားတာနဲ့ဆိုရင် variable 5 ခု လိုပါမယ်:

```python
score1 = 85
score2 = 92
score3 = 78
score4 = 90
score5 = 66
```

ဒါတောင် များနေပါပြီ။ ကျောင်းသား 30 ယောက် ဆိုရင်ရော? 300 ယောက် ဆိုရင်ရော?

**list (စာရင်း)** က ဒီပြဿနာကို ဖြေရှင်းပေးပါတယ်။ list ဆိုတာ **တန်ဖိုး အများကြီးကို အစီအစဉ်လိုက်** သိမ်းထားနိုင်တဲ့ variable တစ်ခုတည်း ဖြစ်ပါတယ်။

list ကို တန်းစီထားတဲ့ **locker (ပစ္စည်းထည့်သော သော့ခတ်ဗီရို)** တစ်တန်းလို့ စဉ်းစားကြည့်ပါ:
- locker တစ်တန်းလုံးမှာ နာမည် တစ်ခုတည်း ရှိပါတယ် (list ရဲ့ နာမည်)။
- locker တစ်ခန်းစီမှာ တန်ဖိုး တစ်ခုစီ ရှိပါတယ်။
- locker တစ်ခန်းစီမှာ နံပါတ် ရေးထားလို့ ရှာလို့ ရပါတယ်။

list ဖန်တီးနည်း:
- တန်ဖိုးတွေကို square bracket `[ ]` ထဲမှာ ထည့်ပါ။
- တန်ဖိုးတွေကို comma `,` နဲ့ ခွဲပါ။

```python
scores = [85, 92, 78, 90, 66]
```

````python
# integer တွေရဲ့ list
# variable တစ်ခု (scores) က အခု တန်ဖိုး 5 ခုကို သိမ်းထားပါတယ်
scores = [85, 92, 78, 90, 66]
print(scores)           # print() က square bracket အပါအဝင် list တစ်ခုလုံးကို ပြပါတယ်
````

````python
# list ထဲမှာ သင်ခန်းစာ 1 မှာ လေ့လာခဲ့တဲ့ data type ဘာမဆို ထည့်လို့ရပါတယ်
fruits = ["apple", "banana", "cherry"]      # string (str) တွေရဲ့ list
heights = [175.5, 162.3, 180.1]             # ဒသမကိန်း (float) တွေရဲ့ list
answers = [True, False, True, True]         # bool တန်ဖိုးတွေရဲ့ list

print(fruits)
print(heights)
print(answers)
````

````python
# list တစ်ခုထဲမှာ data type မတူတာတွေကို ရောထည့်လို့ ရပါတယ်
# ရပေမဲ့ များသောအားဖြင့် list တစ်ခုမှာ တန်ဖိုး အမျိုးအစား တစ်မျိုးတည်း ထည့်ရင် နားလည်ရ ပိုလွယ်ပါတယ်
student = ["Bill", 20, 175.5, True]         # နာမည်၊ အသက်၊ အရပ်၊ ကျောင်းသား ဟုတ်မဟုတ်
print(student)
````

````python
# list က ဗလာ (တန်ဖိုး တစ်ခုမှ မရှိ) ဖြစ်လို့လည်း ရပါတယ်
# ဘာမှမရှိဘဲ စပြီး နောက်မှ တန်ဖိုးတွေ ထည့်ချင်တဲ့အခါ အသုံးဝင်ပါတယ် (ထည့်နည်းကို အပိုင်း 5 မှာ လေ့လာပါမယ်)
empty_list = []
print(empty_list)
````

````python
# list က data type အသစ်တစ်ခု ဖြစ်ပါတယ်: list
# type() က တန်ဖိုးတစ်ခုရဲ့ data type ကို ပြောပြပါတယ်
print(type(scores))     # <class 'list'> ဆိုတာ "ဒါက list ပါ" လို့ အဓိပ္ပာယ်ရပါတယ်
print(type(85))         # <class 'int'> ဆိုတာ "ဒါက integer ပါ" လို့ အဓိပ္ပာယ်ရပါတယ် (list ထဲက တန်ဖိုးတစ်ခု)
````

## 2. index: နေရာကို သုံးပြီး တန်ဖိုးကို ရှာခြင်း

list ထဲက တန်ဖိုးတိုင်းမှာ နေရာ နံပါတ် ရှိပါတယ်။ ဒီနံပါတ်ကို **index** လို့ ခေါ်ပါတယ်။ locker တစ်ခန်းစီပေါ်မှာ ရေးထားတဲ့ နံပါတ်လို့ စဉ်းစားကြည့်ပါ။

မှတ်ထားရမယ့် အရေးအကြီးဆုံး အချက်: **Python က 1 ကနေ မဟုတ်ဘဲ 0 ကနေ စရေတွက်ပါတယ်။**

ဒါကြောင့် ပထမ တန်ဖိုးက index 0 မှာ၊ ဒုတိယ တန်ဖိုးက index 1 မှာ၊ တတိယ တန်ဖိုးက index 2 မှာ စသဖြင့် ရှိပါတယ်။ (သင်ခန်းစာ 3 က `range(5)` ကို မှတ်မိသေးလား? သူလည်း 0 ကနေ စခဲ့ပါတယ်: 0, 1, 2, 3, 4)

```python
fruits = ["apple", "banana", "cherry", "grape", "melon"]
```

| တန်ဖိုး | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| index | 0 | 1 | 2 | 3 | 4 |
| နေရာ (လူတွေ ရေတွက်ပုံ) | 1 ခုမြောက် | 2 ခုမြောက် | 3 ခုမြောက် | 4 ခုမြောက် | 5 ခုမြောက် |

မှတ်ရလွယ်တဲ့ နည်း: **index = နေရာ - 1**။ 1 ခုမြောက် တန်ဖိုးရဲ့ index က 0၊ 5 ခုမြောက် တန်ဖိုးရဲ့ index က 4 ပါ။

တန်ဖိုးကို ယူချင်ရင် list နာမည်ကို ရေးပြီး နောက်မှာ square bracket ထဲမှာ index ကို ရေးပါ:

```python
fruits[0]     # "fruits ထဲက index 0 မှာရှိတဲ့ တန်ဖိုး"
```

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# index နဲ့ တန်ဖိုး ယူခြင်း
print(fruits[0])        # 1 ခုမြောက် တန်ဖိုးက index 0 မှာ -> apple
print(fruits[1])        # 2 ခုမြောက် တန်ဖိုးက index 1 မှာ -> banana
print(fruits[2])        # 3 ခုမြောက် တန်ဖိုးက index 2 မှာ -> cherry
print(fruits[3])        # 4 ခုမြောက် တန်ဖိုးက index 3 မှာ -> grape
print(fruits[4])        # 5 ခုမြောက် (နောက်ဆုံး) တန်ဖိုးက index 4 မှာ -> melon
````

````python
# ဂဏန်း list နဲ့လည်း အတူတူပါပဲ
# index (နေရာ) နဲ့ တန်ဖိုး (အထဲမှာ ရှိတာ) ကို မရောထွေးအောင် သတိထားပါ
numbers = [10, 20, 30, 40]
#  index:   0   1   2   3

print(numbers[0])       # index 0 -> တန်ဖိုးက 10
print(numbers[2])       # index 2 -> တန်ဖိုးက 30 (2 မဟုတ်ပါဘူး!)
print(numbers[3])       # index 3 -> တန်ဖိုးက 40
````

````python
# တန်ဖိုးတွေက အထူး အစီအစဉ်တစ်ခုခုအတိုင်း ရှိစရာ မလိုပါဘူး
# index က တန်ဖိုး "ဘယ်နေရာမှာ" ရှိလဲ ဆိုတာပဲ ပြောပြပါတယ်၊ ဘယ်လောက်ကြီးလဲ မပြောပါဘူး
scores = [85, 92, 78, 90, 66]
#  index:   0   1   2   3   4

print(scores[1])        # 2 ယောက်မြောက် ကျောင်းသားရဲ့ အမှတ် -> 92
print(scores[4])        # 5 ယောက်မြောက် ကျောင်းသားရဲ့ အမှတ် -> 66
````

````python
# index တစ်ခုတည်းကို လိုသလောက် အကြိမ်ကြိမ် ယူလို့ရပါတယ်
# တန်ဖိုး ယူတာက "ဖတ်" တာပဲ ဖြစ်ပါတယ်။ list က မပြောင်းလဲပါဘူး
colors = ["red", "green", "blue"]

print(colors[1])        # green
print(colors[1])        # green အတိုင်းပဲ
print(colors)           # list က အရင်အတိုင်းပဲ ရှိပါတယ်
````

````python
# list ထဲက ယူလာတဲ့ တန်ဖိုးကို သာမန် variable လို သုံးလို့ရပါတယ်
scores = [85, 92, 78, 90, 66]

first = scores[0]                  # ပထမ အမှတ် (85) ကို variable အသစ်ထဲ သိမ်းပါတယ်
print(first)

print(scores[0] + scores[1])       # 85 + 92 = 177 (တွက်ချက်လို့ ရပါတယ်)
print(scores[1] > scores[2])       # 92 > 78 က True (သင်ခန်းစာ 2 လိုပဲ နှိုင်းယှဉ်လို့ ရပါတယ်)
````

````python
# list ထဲက ယူလာတဲ့ string တွေကို သင်ခန်းစာ 1 လိုပဲ + နဲ့ ဆက်လို့ ရပါတယ်
words = ["Hello", "nice", "to", "meet", "you"]

print(words[0] + ", " + words[1] + " " + words[2] + " " + words[3] + " " + words[4])
````

````python
# index က variable လည်း ဖြစ်လို့ရပါတယ်
# ဒီမှာ i ထဲမှာ 2 ရှိလို့ scores[i] က scores[2] နဲ့ အတူတူပါပဲ
scores = [85, 92, 78, 90, 66]

i = 2
print(scores[i])        # scores[2] -> 78

i = 0
print(scores[i])        # အခု i ထဲမှာ 0 ရှိလို့ scores[0] -> 85
````

````python
# index က တွက်ချက်မှု တစ်ခုတောင် ဖြစ်လို့ရပါတယ်
# Python က [ ] ထဲက ဂဏန်းကို အရင်တွက်ပြီးမှ အဲဒီ index ကို ရှာပါတယ်
scores = [85, 92, 78, 90, 66]

print(scores[1 + 2])    # 1 + 2 = 3 -> scores[3] -> 90

i = 1
print(scores[i + 1])    # i + 1 = 2 -> scores[2] -> 78
````

### အနုတ် index: နောက်ဆုံးကနေ ရေတွက်ခြင်း

သာမန် (အပေါင်း) index ကို သိသွားပြီဆိုတော့ တန်ဖိုးရှာနည်း နောက်တစ်မျိုးကို ကြည့်ရအောင်။

Python က အနုတ်ကိန်းတွေကို သုံးပြီး list ရဲ့ **နောက်ဆုံးကနေ နောက်ပြန်** ရေတွက်ခွင့်လည်း ပေးပါတယ်။

- `-1` က **နောက်ဆုံး** တန်ဖိုး
- `-2` က နောက်ဆုံးကနေ ဒုတိယ တန်ဖိုး
- `-3` က နောက်ဆုံးကနေ တတိယ တန်ဖိုး၊ စသဖြင့်

| တန်ဖိုး | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| index | 0 | 1 | 2 | 3 | 4 |
| အနုတ် index | -5 | -4 | -3 | -2 | -1 |

အနုတ် index က -0 ကနေ မဟုတ်ဘဲ **-1** ကနေ စတာကို သတိပြုပါ။ (-0 ဆိုတာ မရှိပါဘူး၊ -0 က 0 ပဲ ဖြစ်ပြီး ပထမ တန်ဖိုး ဖြစ်လို့ပါ။)

**အနုတ် index ကို ဘာလို့ သုံးတာလဲ?**

အဓိက အကြောင်းရင်းက **နောက်ဆုံး တန်ဖိုးကို မြန်မြန် ယူဖို့** ပါ။

အပေါင်း index နဲ့ဆိုရင် နောက်ဆုံး တန်ဖိုးကို ယူဖို့ list ထဲမှာ တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ အရင် သိရပါမယ်။ list မှာ တန်ဖိုး 5 ခု ရှိရင် နောက်ဆုံး index က 4 ပါ။ 100 ခု ရှိရင် နောက်ဆုံး index က 99 ပါ။ တန်ဖိုး ထည့်တာ ဖျက်တာ လုပ်ရင်လည်း နောက်ဆုံး index က ပြောင်းသွားပါတယ်။

အနုတ် index နဲ့ဆိုရင် ဘာမှ ရေတွက်စရာ မလိုပါဘူး။ list က ဘယ်လောက်ပဲ ရှည်ရှည် **`-1` က အမြဲတမ်း နောက်ဆုံး တန်ဖိုး** ပါ။

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# အနုတ် index: နောက်ဆုံးကနေ ရေတွက်ခြင်း
print(fruits[-1])       # -1 က အမြဲ နောက်ဆုံး တန်ဖိုး -> melon
print(fruits[-2])       # -2 က နောက်ဆုံးကနေ ဒုတိယ တန်ဖိုး -> grape
print(fruits[-3])       # -3 က နောက်ဆုံးကနေ တတိယ တန်ဖိုး -> cherry
````

````python
# အပေါင်း index နဲ့ အနုတ် index က တန်ဖိုး တစ်ခုတည်းကို ညွှန်ပြနိုင်ပါတယ်
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[4])        # အပေါင်း index နဲ့ နောက်ဆုံး တန်ဖိုး -> melon
print(fruits[-1])       # အနုတ် index နဲ့ နောက်ဆုံး တန်ဖိုး -> melon (အတူတူပဲ!)

print(fruits[0])        # အပေါင်း index နဲ့ ပထမ တန်ဖိုး -> apple
print(fruits[-5])       # အနုတ် index နဲ့ ပထမ တန်ဖိုး -> apple (အတူတူပဲ!)
````

````python
# -1 ဘာလို့ အသုံးဝင်လဲ: list က ဘယ်လောက်ပဲ ရှည်ရှည် "list တိုင်း" မှာ အလုပ်လုပ်ပါတယ်
# အပေါင်း index နဲ့ဆိုရင် list တစ်ခုစီရဲ့ နောက်ဆုံး index က မတူပါဘူး (2, 6, 0 ...)
# -1 နဲ့ဆိုရင် အမြဲ အတူတူပါပဲ
short_list = ["a", "b", "c"]
long_list = [10, 20, 30, 40, 50, 60, 70]
one_item = ["only"]

print(short_list[-1])   # c       (အပေါင်း index ဆိုရင် 2)
print(long_list[-1])    # 70      (အပေါင်း index ဆိုရင် 6)
print(one_item[-1])     # only    (အပေါင်း index ဆိုရင် 0)
````

## 3. တန်ဖိုးပြောင်းခြင်းနဲ့ len()

### တန်ဖိုး ပြောင်းခြင်း

list က ပုံသေ မဟုတ်ပါဘူး။ အထဲက တန်ဖိုးကို ကြိုက်တဲ့အချိန် **ပြောင်း** လို့ ရပါတယ်။ locker တစ်ခန်းထဲကို ပစ္စည်းအသစ် ထည့်တာနဲ့ တူပါတယ်:

```python
list_နာမည်[index] = တန်ဖိုး_အသစ်
```

ဒါက variable ထဲကို တန်ဖိုး သိမ်းတာနဲ့ အတူတူ အလုပ်လုပ်ပါတယ် (`=` က "သိမ်းပါ" လို့ အဓိပ္ပာယ်ရပါတယ်)၊ ဒါပေမဲ့ အဲဒီ နေရာ တစ်ခုတည်းက တန်ဖိုးပဲ အစားထိုးခံရပါတယ်။ ကျန်တဲ့ တန်ဖိုးတွေက အတူတူပဲ ရှိနေပါတယ်။

````python
# သတ်မှတ်ထားတဲ့ index မှာ တန်ဖိုး ပြောင်းခြင်း
fruits = ["apple", "banana", "cherry"]
print(fruits)

fruits[1] = "orange"        # index 1 က တန်ဖိုး ("banana") ကို "orange" နဲ့ အစားထိုးပါတယ်
print(fruits)               # index 1 ပဲ ပြောင်းသွားပါတယ်။ "apple" နဲ့ "cherry" က အရင်အတိုင်းပါ
````

````python
# အနုတ် index နဲ့လည်း တန်ဖိုး ပြောင်းလို့ ရပါတယ်
fruits = ["apple", "banana", "cherry"]

fruits[-1] = "grape"        # နောက်ဆုံး တန်ဖိုး ("cherry") ကို "grape" နဲ့ အစားထိုးပါတယ်
print(fruits)
````

````python
# တန်ဖိုးဟောင်းကို သုံးပြီး တန်ဖိုး ပြောင်းခြင်း
# သင်ခန်းစာ 1 က age = age + 1 နဲ့ စိတ်ကူး အတူတူပါပဲ
scores = [85, 92, 78]

scores[2] = scores[2] + 5   # တတိယ ကျောင်းသားက ဆုအမှတ် 5 မှတ် ရပါတယ် (78 + 5 = 83)
print(scores)
````

### len(): list ထဲမှာ တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ?

`len()` က list ထဲမှာ တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ ပြောပြပါတယ်။ `len` က **length** (အရှည်) ရဲ့ အတိုကောက် ဖြစ်ပါတယ်။

```python
len(list_နာမည်)
```

list ကို ကွင်း `( )` ထဲမှာ ထည့်လိုက်ရင် `len()` က ဂဏန်း တစ်ခုကို ပြန်ပေးပါတယ်: အဲဒီ list ထဲက တန်ဖိုး အရေအတွက် ဖြစ်ပါတယ်။

````python
# len(): list ထဲမှာ တန်ဖိုး ဘယ်နှစ်ခု ရှိလဲ ရေတွက်ခြင်း
fruits = ["apple", "banana", "cherry", "grape", "melon"]
print(len(fruits))          # တန်ဖိုး 5 ခု ရှိလို့ 5
````

````python
# အရှည် မတူတဲ့ list တွေ
three = ["a", "b", "c"]
five = [1, 2, 3, 4, 5]
one = ["only"]
empty_list = []

print(len(three))           # တန်ဖိုး 3 ခု -> 3
print(len(five))            # တန်ဖိုး 5 ခု -> 5
print(len(one))             # တန်ဖိုး 1 ခု -> 1
print(len(empty_list))      # ဗလာ list မှာ တန်ဖိုး 0 ခု -> 0
````

````python
# len() က list အမျိုးအစား ဘာပဲဖြစ်ဖြစ် အတူတူ အလုပ်လုပ်ပါတယ်
names = ["Bill", "Anna", "Tom", "Kate"]          # string တွေရဲ့ list
numbers = [3, 7, 1, 9, 4, 6]                     # integer တွေရဲ့ list
answers = [True, False, True]                    # bool တန်ဖိုးတွေရဲ့ list

print(len(names))           # 4
print(len(numbers))         # 6
print(len(answers))         # 3
````

````python
# len() က တန်ဖိုး "ဘယ်နှစ်ခု" ရှိလဲ ရေတွက်ပါတယ်။ တန်ဖိုးတွေကို မပေါင်းပါဘူး
# တန်ဖိုးတွေက ကြီးပေမဲ့ 3 ခုပဲ ရှိပါတယ်
prices = [1000, 2500, 3000]
print(len(prices))          # 3 (6500 မဟုတ်ပါဘူး!)
````

````python
# len() ရဲ့ ရလဒ်က ဂဏန်း (int) တစ်ခုပဲ ဖြစ်လို့ variable ထဲ သိမ်းလို့ ရပါတယ်
students = ["Bill", "Anna", "Tom", "Kate", "Mike"]

count = len(students)       # variable count ထဲကို 5 သိမ်းပါတယ်
print(count)
print("There are", count, "students.")
````

````python
# len() က string နဲ့လည်း အလုပ်လုပ်ပါတယ်: စာလုံး အရေအတွက်ကို ရေတွက်ပါတယ်
# (space လည်း စာလုံး တစ်လုံးအဖြစ် ရေတွက်ပါတယ်)
print(len("apple"))         # a, p, p, l, e -> 5
print(len("hi there"))      # h, i, (space), t, h, e, r, e -> 8
````

### နောက်ဆုံး index က len() - 1

အခု `len()` နဲ့ index ကို ချိတ်ဆက်လို့ ရပါပြီ။

list မှာ တန်ဖိုး 5 ခု ရှိရင် `len()` က 5 ကို ပေးပေမဲ့ နောက်ဆုံး index က **4** ပါ။ index က 0 ကနေ စလို့ **နောက်ဆုံး index က အမြဲတမ်း `len() - 1`** ဖြစ်ပါတယ်။

ဒါကြောင့် နောက်ဆုံး တန်ဖိုးကို ယူနည်း နှစ်မျိုး ရှိပါတယ်:
- `fruits[len(fruits) - 1]`: တန်ဖိုးတွေကို ရေတွက်၊ 1 နုတ်ပြီး အဲဒီ index ကို သုံးပါတယ်
- `fruits[-1]`: -1 ကို တန်းသုံးလိုက်ပါတယ်

နှစ်မျိုးလုံး တန်ဖိုး အတူတူ ရပေမဲ့ `-1` က ပိုတိုပြီး ပိုလွယ်ပါတယ်။ အနုတ် index အသုံးဝင်တဲ့ အကြောင်းရင်းက ဒါပါပဲ။

````python
# နောက်ဆုံး index က len() - 1
fruits = ["apple", "banana", "cherry", "grape", "melon"]

last_index = len(fruits) - 1       # 5 - 1 = 4
print(last_index)
print(fruits[last_index])          # fruits[4] -> melon
````

````python
# နောက်ဆုံး တန်ဖိုး ယူနည်း နှစ်မျိုးကို နှိုင်းယှဉ်ခြင်း
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[len(fruits) - 1])     # ရှည်တဲ့နည်း: len(fruits) က 5၊ 5 - 1 = 4၊ fruits[4] -> melon
print(fruits[-1])                  # တိုတဲ့နည်း: -1 က နောက်ဆုံး တန်ဖိုး -> melon (ရလဒ် အတူတူပဲ!)
````

## 4. တန်ဖိုးတစ်ခု list ထဲမှာ ရှိမရှိ စစ်ခြင်း (in)

တစ်ခါတလေ **"ဒီတန်ဖိုး list ထဲမှာ ရှိလား?"** ဆိုတာ သိချင်ပါတယ်။

ဒါအတွက် `in` ကို သုံးပါတယ်။

```python
တန်ဖိုး in list_နာမည်
```

- တန်ဖိုးက list ထဲမှာ ရှိရင် ရလဒ်က `True` ပါ။
- မရှိရင် ရလဒ်က `False` ပါ။

ရလဒ်က သင်ခန်းစာ 2 က နှိုင်းယှဉ် operator တွေလိုပဲ `True` ဒါမှမဟုတ် `False` ဖြစ်ပါတယ်။ ဆိုလိုတာက `if` ရဲ့ condition အဖြစ် တိုက်ရိုက် သုံးလို့ရပါတယ်!

တန်ဖိုးတစ်ခု list ထဲမှာ **မရှိဘူး** ဆိုတာ စစ်ချင်ရင် `not in` ကို သုံးပါ။

**မှတ်ချက်:** `in` ဆိုတဲ့ စကားလုံးကို သင်ခန်းစာ 3 မှာလည်း မြင်ခဲ့ဖူးပါတယ် (`for i in range(5)`)။ အဲဒီ `in` က `for` loop ရဲ့ အစိတ်အပိုင်း ဖြစ်ပါတယ်။ ဒီမှာ လေ့လာတဲ့ `in` က မတူပါဘူး: တန်ဖိုးတစ်ခု list ထဲမှာ ရှိမရှိ **စစ်ဆေးပြီး** True ဒါမှမဟုတ် False ကို ပေးပါတယ်။

````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)       # "banana" က list ထဲမှာ ရှိပါတယ် -> True
print("grape" in fruits)        # "grape" က list ထဲမှာ မရှိပါဘူး -> False
````

````python
# not in: in ရဲ့ ဆန့်ကျင်ဘက်
print("grape" not in fruits)    # "grape" က list ထဲမှာ မရှိ -> True
print("apple" not in fruits)    # "apple" က list ထဲမှာ ရှိလို့ "not in" က False
````

````python
# in ကို if ရဲ့ condition အဖြစ် သုံးခြင်း (သင်ခန်းစာ 2)
fruits = ["apple", "banana", "cherry"]
wanted = "banana"

if wanted in fruits:                    # "banana" in fruits က True
    print("Yes, we have it!")
else:
    print("Sorry, it's sold out.")
````

````python
# လက်တွေ့ ဥပမာ: ဧည့်သည်စာရင်း စစ်ခြင်း
guests = ["Bill", "Anna", "Tom"]
name = "Kate"

if name in guests:                      # "Kate" in guests က False
    print("Welcome!")
else:
    print("Sorry, you are not on the list.")    # ဒါကြောင့် ဒါ run ပါတယ်
````

````python
# in က တန်ဖိုးကို အတိအကျ စစ်ပါတယ်။ စာလုံးအကြီး နဲ့ စာလုံးအသေး မတူပါဘူး
fruits = ["apple", "banana", "cherry"]
print("Apple" in fruits)        # "Apple" (A အကြီး) နဲ့ "apple" က string မတူပါဘူး -> False
````

## 5. တန်ဖိုးထည့်ခြင်းနဲ့ ဖျက်ခြင်း

list က ရှည်လာနိုင်သလို တိုသွားနိုင်ပါတယ်။ တန်ဖိုး အသစ်တွေ ထည့်လို့ရပြီး မလိုတော့တဲ့ တန်ဖိုးတွေကို ဖျက်လို့ ရပါတယ်။

| ရေးပုံ | ဘာလုပ်လဲ |
|---|---|
| `list_နာမည်.append(တန်ဖိုး)` | တန်ဖိုးကို list ရဲ့ **နောက်ဆုံး** မှာ ထည့်ပါတယ် |
| `list_နာမည်.insert(index, တန်ဖိုး)` | တန်ဖိုးကို **ရွေးထားတဲ့ နေရာ** မှာ ထည့်ပါတယ် |
| `list_နာမည်.remove(တန်ဖိုး)` | ပြောထားတဲ့ တန်ဖိုးကို ဖျက်ပါတယ် (ပထမဆုံး တွေ့တာကို) |
| `list_နာမည်.pop()` | **နောက်ဆုံး** တန်ဖိုးကို ထုတ်ယူပါတယ် |
| `list_နာမည်.pop(index)` | **ရွေးထားတဲ့ နေရာ** က တန်ဖိုးကို ထုတ်ယူပါတယ် |

အစက် `.` က "ဒီ list ကို ဒါလုပ်ပါ" လို့ အဓိပ္ပာယ်ရပါတယ်။ ဥပမာ `fruits.append("grape")` က "fruits list ထဲကို "grape" ထည့်ပါ" လို့ အဓိပ္ပာယ်ရပါတယ်။ ဒီလို အစက်နောက်မှာ ရေးတဲ့ command တွေကို **method** လို့ ခေါ်ပါတယ်။

ဒီ method တွေက list ကိုယ်တိုင်ကို ပြောင်းလဲပါတယ်။ `fruits = ...` လို့ ရေးစရာ မလိုပါဘူး။

````python
# append(): တန်ဖိုးကို နောက်ဆုံးမှာ ထည့်ခြင်း
fruits = ["apple", "banana"]
print(fruits)

fruits.append("cherry")         # "cherry" ကို နောက်ဆုံးမှာ ထည့်ပါတယ်
print(fruits)

fruits.append("grape")          # "grape" ကို နောက်ဆုံးမှာ ထပ်ထည့်ပါတယ်
print(fruits)
print(len(fruits))              # list က တန်ဖိုး 2 ခုကနေ 4 ခု ဖြစ်လာပါတယ်
````

````python
# ဗလာ list နဲ့ စပြီး တန်ဖိုးတွေကို တစ်ခုချင်း ထည့်ခြင်း
# ဒါက အရမ်းအသုံးများတဲ့ ပုံစံ ဖြစ်ပါတယ်
shopping = []                   # ဗလာ list နဲ့ စပါတယ်
shopping.append("milk")
shopping.append("eggs")
shopping.append("bread")
print(shopping)
````

````python
# insert(): သတ်မှတ်ထားတဲ့ နေရာမှာ တန်ဖိုး ထည့်ခြင်း
# insert(index, တန်ဖိုး) -> တန်ဖိုးအသစ်က အဲဒီ index ကို ရောက်သွားပြီး နောက်က တန်ဖိုးတွေက တစ်နေရာစီ နောက်ကို ရွှေ့သွားပါတယ်
fruits = ["apple", "banana", "cherry"]

fruits.insert(1, "orange")      # "orange" ကို index 1 မှာ ထည့်ပါတယ်
print(fruits)                   # "banana" နဲ့ "cherry" က တစ်နေရာစီ နောက်ကို ရွှေ့သွားပါတယ်

fruits.insert(0, "melon")       # "melon" ကို index 0 (ရှေ့ဆုံး) မှာ ထည့်ပါတယ်
print(fruits)
````

````python
# remove(): တန်ဖိုးကိုယ်တိုင်ကို ပြောပြီး ဖျက်ခြင်း
fruits = ["apple", "banana", "cherry"]

fruits.remove("banana")         # "banana" ကို ရှာပြီး ဖျက်ပါတယ်
print(fruits)                   # နောက်က တန်ဖိုးတွေက ရှေ့ကို ရွှေ့ပြီး နေရာလွတ်ကို ဖြည့်ပါတယ်
````

````python
# တန်ဖိုး တစ်ခုတည်း အကြိမ်ကြိမ် ပါနေရင် remove() က "ပထမဆုံး" တစ်ခုကိုပဲ ဖျက်ပါတယ်
numbers = [1, 2, 3, 2, 1]

numbers.remove(2)               # ပထမ 2 (index 1 က) ကိုပဲ ဖျက်ပါတယ်
print(numbers)                  # ဒုတိယ 2 က ရှိနေဆဲပါ
````

````python
# pop(): နေရာကို သုံးပြီး တန်ဖိုးကို ထုတ်ယူခြင်း
# index မပေးရင် pop() က "နောက်ဆုံး" တန်ဖိုးကို ထုတ်ယူပါတယ်
fruits = ["apple", "banana", "cherry"]

fruits.pop()                    # နောက်ဆုံး တန်ဖိုး ("cherry") ကို ထုတ်ယူပါတယ်
print(fruits)
````

````python
# pop(index): အဲဒီ index က တန်ဖိုးကို ထုတ်ယူခြင်း
fruits = ["apple", "banana", "cherry"]

fruits.pop(0)                   # index 0 က တန်ဖိုး ("apple") ကို ထုတ်ယူပါတယ်
print(fruits)
````

````python
# pop() က ထုတ်ယူလိုက်တဲ့ တန်ဖိုးကို "ပြန်ပေး" လည်း ပေးလို့ variable ထဲ သိမ်းလို့ ရပါတယ်
# remove() က ဖျက်ရုံပဲ ဖျက်ပါတယ်။ pop() က ဖျက်ပြီး "ပြီးတော့" တန်ဖိုးကို ကိုယ့်ကို ပေးပါတယ်
tasks = ["wash dishes", "do homework", "clean room"]

done = tasks.pop(0)             # ပထမ အလုပ်ကို ထုတ်ယူပြီး done ထဲ သိမ်းပါတယ်
print("Finished:", done)
print("Still to do:", tasks)
````

````python
# remove() နဲ့ pop() ကို အကျဉ်းချုပ် ကြည့်ခြင်း
# remove(တန်ဖိုး): "ဒီတန်ဖိုးကို ဖျက်ပါ"           -> တန်ဖိုးကို သိပါတယ်
# pop(index):      "ဒီနေရာမှာ ရှိတာကို ဖျက်ပါ"      -> နေရာကို သိပါတယ်
colors = ["red", "green", "blue"]

colors.remove("green")          # တန်ဖိုးနဲ့ ဖျက်ခြင်း
print(colors)

colors.pop(0)                   # နေရာနဲ့ ဖျက်ခြင်း (အခု index 0 က "red")
print(colors)
````

## 6. မကြာခဏ ဖြစ်တတ်သော အမှားများ

list ကို လေ့လာတဲ့အခါ လူတိုင်းနီးပါး ဒီအမှားတွေ လုပ်မိကြပါတယ်။ error တက်ရင် ဒါတွေကို အရင်စစ်ကြည့်ပါ။

1. **index က 0 ကနေ စတာကို မေ့ခြင်း**: `fruits[1]` က ပထမ မဟုတ်ဘဲ **ဒုတိယ** တန်ဖိုး ဖြစ်ပါတယ်။
2. **မရှိတဲ့ index ကို သုံးခြင်း**: list မှာ တန်ဖိုး 3 ခု ရှိရင် index တွေက 0, 1, 2 ပါ။ `fruits[3]` က `IndexError: list index out of range` ("index က list ရဲ့ အပိုင်းအခြား အပြင်ကို ရောက်နေပါတယ်") ဖြစ်စေပါတယ်။
3. **index အသစ်နဲ့ တန်ဖိုး ထည့်ဖို့ ကြိုးစားခြင်း**: တန်ဖိုး 3 ခုရှိတဲ့ list မှာ `fruits[3] = "grape"` က တန်ဖိုးကို **မထည့်ပါဘူး**။ `IndexError: list assignment index out of range` ("list အပြင်က index မှာ တန်ဖိုး သိမ်းလို့ မရပါ") ဖြစ်စေပါတယ်။ ထည့်ချင်ရင် `append()` ကို သုံးပါ။
4. **list ထဲမှာ မရှိတဲ့ တန်ဖိုးကို ဖျက်ခြင်း**: `fruits.remove("grape")` က `ValueError: list.remove(x): x not in list` ("ဒီတန်ဖိုး list ထဲမှာ မရှိပါ") ဖြစ်စေပါတယ်။ `in` နဲ့ အရင်စစ်ပါ။
5. **`( )` နဲ့ `[ ]` ကို ရောထွေးခြင်း**: list ဖန်တီးတဲ့အခါ နဲ့ index နဲ့ တန်ဖိုး ယူတဲ့အခါ (`fruits[0]`) `[ ]` ကို သုံးပါ။ `len(fruits)`၊ `fruits.append("grape")` လို command တွေအတွက် `( )` ကို သုံးပါ။

အောက်က cell တွေမှာ အမှားတစ်ခုချင်းစီ ဘယ်လိုပုံစံလဲ ပြထားပါတယ်။ notebook မပျက်အောင် အမှား code ကို comment အဖြစ် ရေးထားပါတယ်။ တကယ့် error message ကို မြင်ချင်ရင် code စာကြောင်းက `#` ကို ဖျက်ကြည့်ပြီး ပြီးရင် `#` ကို ပြန်ထည့်ပါ။

````python
# အမှား 1: index က 0 ကနေ စတာကို မေ့ခြင်း
fruits = ["apple", "banana", "cherry"]

print(fruits[1])        # လူအများက "apple" လို့ ထင်ပေမဲ့ ဒါက "ဒုတိယ" တန်ဖိုး -> banana
print(fruits[0])        # မှန်ပါတယ်: ပထမ တန်ဖိုးက index 0 မှာ -> apple
````

````python
# အမှား 2: မရှိတဲ့ index ကို သုံးခြင်း
fruits = ["apple", "banana", "cherry"]      # တန်ဖိုး 3 ခု -> index တွေက 0, 1, 2

# print(fruits[3])      # IndexError: list index out of range -> index 3 မရှိပါဘူး!

print(fruits[2])                    # မှန်ပါတယ်: နောက်ဆုံး index က 2
print(fruits[len(fruits) - 1])      # ဒါလည်း မှန်ပါတယ်: len(fruits) - 1 = 3 - 1 = 2
print(fruits[-1])                   # ဒါလည်း မှန်ပါတယ်: -1 က အမြဲ နောက်ဆုံး တန်ဖိုး
````

````python
# အမှား 3: index အသစ်နဲ့ တန်ဖိုး ထည့်ဖို့ ကြိုးစားခြင်း
fruits = ["apple", "banana", "cherry"]

# fruits[3] = "grape"   # IndexError: list assignment index out of range -> ရှိပြီးသား index တွေက တန်ဖိုးကိုပဲ "ပြောင်း" လို့ ရပါတယ်

fruits.append("grape")  # မှန်ပါတယ်: တန်ဖိုး အသစ်ကို နောက်ဆုံးမှာ ထည့်ဖို့ append() ကို သုံးပါ
print(fruits)
````

````python
# အမှား 4: list ထဲမှာ မရှိတဲ့ တန်ဖိုးကို ဖျက်ခြင်း
fruits = ["apple", "banana", "cherry"]

# fruits.remove("grape")    # ValueError: list.remove(x): x not in list -> "grape" က list ထဲမှာ မရှိပါဘူး!

# မှန်ပါတယ်: in နဲ့ အရင်စစ်ပြီး (အပိုင်း 4) ရှိမှသာ ဖျက်ပါ
if "grape" in fruits:
    fruits.remove("grape")
else:
    print("grape is not in the list, so nothing was removed.")
````

````python
# အမှား 5: ( ) နဲ့ [ ] ကို ရောထွေးခြင်း
fruits = ["apple", "banana", "cherry"]

# print(fruits(0))          # TypeError: 'list' object is not callable -> ( ) က command အတွက်ပါ၊ index အတွက် မဟုတ်ပါ
# fruits.append["grape"]    # TypeError -> append က command ဖြစ်လို့ [ ] မဟုတ်ဘဲ ( ) လိုပါတယ်

print(fruits[0])            # မှန်ပါတယ်: index နဲ့ တန်ဖိုး ယူဖို့ [ ]
fruits.append("grape")      # မှန်ပါတယ်: command အတွက် ( )
print(fruits)
````
