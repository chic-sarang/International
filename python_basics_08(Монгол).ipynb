## 1. Функц гэж юу вэ?

### Асуудал

Богино мэндчилгээ хэвлэхийг хүсэж байна гэж бодъё:

```python
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
```

Одоо энэ мэндчилгээг програмынхаа өөр өөр газарт **3 удаа** хэвлэх хэрэгтэй гэж төсөөлөөрэй. Нөгөө 3 мөрөө дахин дахин бичих болно.

````python
# Ижил 3 мөрийг 3 удаа бичсэн
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
````

Энэ нь ажиллана, гэхдээ асуудал бий:

- Урт бөгөөд бичихэд ядаргаатай.
- "Hello!"-г "Hi!" болгож өөрчлөхийг хүсвэл **бүх** газарт өөрчлөх хэрэгтэй. Нэгийг нь мартвал мэндчилгээнүүд өөр өөр болно.

### Функцийн санаа

Тэр 3 мөрийг **ганцхан удаа** бичээд, **нэр** өгөөд, хэрэгтэй үедээ тэр нэрийг л хэлдэг бол яах вэ?

Яг үүнийг **функц** гэдэг: **нэртэй кодын бүлэг**.

### Хоолны жорыг бодоорой

Функц бол **хоолны жор** шиг.

1. Эхлээд та **жорыг** цаасан дээр **бичнэ**: "Хуушуур хийх арга: 1-р алхам, 2-р алхам, 3-р алхам." Жор бичих нь хуушуур бий болгодоггүй. Энэ бол зүгээр л дараа ашиглахаар хадгалсан заавар.
2. Дараа нь хуушуур идмээр санагдах бүрт та **жорын дагуу хийнэ**. Нэг жорыг хүссэн удаагаараа дагаж болно.

Функц ч ингэж ажилладаг:

| Жор | Функц |
|---|---|
| Жорыг бичих | Функцийг **үүсгэх** (тодорхойлох) |
| Жорын дагуу хоол хийх | Функцийг **ашиглах** (дуудах) |

### Та функцийг аль хэдийн ашиглаж байсан!

Та Хичээл 1-ээс хойш функц ашиглаж ирсэн. `print()`, `len()`, `type()` бүгд функц. Хэн нэгэн тэдгээрийг танд аль хэдийн бичиж өгсөн бөгөөд та зүгээр л нэр болон дугуй хаалт `( )` бичиж ашиглаж байсан.

Энэ хичээлээр та **өөрийн** функцийг хэрхэн үүсгэхийг сурна.

````python
# Таны аль хэдийн ашиглаж байсан функцүүд
print("Hello")              # print() бол функц: утгыг дэлгэцэнд харуулна
print(len("apple"))         # len() бол функц: тоолно
print(type(10))             # type() бол функц: өгөгдлийн төрлийг хэлнэ
````

## 2. Функц үүсгэх ба ашиглах

Функцийг **хоёр тусдаа алхмаар** ашигладаг:

1. Функцийг **үүсгэх** (жорыг бичих)
2. Функцийг **ашиглах** (жорын дагуу хийх)

### 1-р алхам: Функц үүсгэх

```python
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")
```

Хэсэг бүрийг харъя:

| Хэсэг | Утга |
|---|---|
| `def` | **define** (тодорхойлох)-ийн товчлол. "Би функц үүсгэх гэж байна" гэсэн утгатай |
| `say_hello` | функцийн **нэр**. Та өөрөө сонгоно (дүрэм нь хувьсагчийн нэрийнхтэй ижил) |
| `()` | дугуй хаалт. Нэрийн дараа үргэлж хэрэгтэй (дотор нь юу бичихийг 3-р хэсэгт үзнэ) |
| `:` | `if`, `for`-ийн нэгэн адил мөрийн төгсгөлд хоёр цэг |
| догол мөртэй мөрүүд | функцийн **бие**: функцэд хамаарах код |

**Догол мөрийн тухай:** `if`, `for`-ийн нэгэн адил Python догол мөрөөр (мөрийн эхэнд 4 зай) аль мөрүүд функцэд хамаарахыг мэддэг. `def`-ийн доорх догол мөртэй мөрүүд бол функцийн бие. Догол мөргүй эхний мөр функцийн **гадна** байна.

### Функц үүсгэх нь түүнийг АЖИЛЛУУЛДАГГҮЙ

Энэ бол энэ хэсгийн хамгийн чухал зүйл.

Python `def`-ийг уншихдаа зөвхөн функцийг **санаж** авдаг. Доторх кодыг ажиллуулдаггүй. Жор бичих нь хуушуур бий болгодоггүйтэй адил.

Доорх нүдийг хараарай. Бид функц үүсгэсэн ч **юу ч хэвлэгдэхгүй**.

````python
# 1-р алхам: функцийг үүсгэнэ
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")

# Юу ч хэвлэгдсэнгүй! Бид зөвхөн функцийг үүсгэсэн (бичсэн). Хараахан ашиглаагүй.
````

### 2-р алхам: Функц ашиглах

Функцийг ашиглахын тулд түүний **нэр**-ийг бичээд араас нь **дугуй хаалт** `()` бичнэ:

```python
say_hello()
```

Функцийг ашиглахыг функцийг **дуудах (call)** гэдэг. Функцийг дуудахад Python тэр функц руу очиж, доторх бүх кодыг дээрээс доош ажиллуулаад, буцаж ирнэ.

````python
# 2-р алхам: функцийг дуудна (ашиглана)
say_hello()
````

### Нэг жор, олон удаа

Функц нэгэнт үүссэн бол түүнийг **хүссэн удаагаараа** дуудаж болно. Дуудах бүрд бүх бие дахин ажиллана.

````python
# Ижил мэндчилгээ 3 удаа, гэхдээ одоо ердөө 3 богино мөрөөр
say_hello()
say_hello()
say_hello()
````

Хичээлийн эхлэлтэй харьцуулаарай. Өмнө нь бидэнд 9 мөр хэрэгтэй байсан. Одоо ердөө 3 удаа дуудахад л болно.

"Hello!"-г "Hi!" болгохыг хүсвэл **нэг л газарт**: функцийн дотор өөрчилнө. Бүх дуудлага шинэ мэндчилгээг ашиглана.

````python
# Функцийг нэг газарт өөрчилбөл...
def say_hello():
    print("Hi!")                        # Зөвхөн энд өөрчилсөн
    print("Welcome to Python class.")
    print("Have a nice day!")

# ...бүх дуудлага шинэ хувилбарыг ашиглана
say_hello()
say_hello()
````

### Нэг нүдэнд үүсгэж, ашиглах

Ихэвчлэн бид эхлээд функц үүсгээд, доор нь дуудна. Python юу хийж байгааг алхам алхмаар дагаж үзье.

````python
def say_goodbye():          # 1. Python "say_goodbye" функцийг санана. Одоохондоо юу ч ажиллахгүй
    print("Goodbye!")
    print("See you next time.")

print("Class is over.")     # 2. Энэ мөр догол мөргүй -> функцийн гадна -> одоо ажиллана
say_goodbye()               # 3. Функцийг дуудна -> доторх 2 мөр одоо ажиллана
print("The end.")           # 4. Функц дууссаны дараа Python энд буцаж ирнэ
````

Гаралтын дарааллыг анхаараарай:

1. `Class is over.` - функцийн **гаднах** эхний мөр
2. `Goodbye!` ба `See you next time.` - функцийн бие зөвхөн бид **дуудах** үед ажиллана
3. `The end.` - функц дууссаны дараа Python дуудлагын дараагийн мөрөөс үргэлжлүүлнэ

### Эхлээд үүсгэ, дараа нь ашигла

Жорыг дагахаас **өмнө** бичсэн байх ёстой. Үүний нэгэн адил функцийг **дуудахаас өмнө үүсгэсэн** байх ёстой. Python кодыг дээрээс доош уншдаг тул хараахан үүсгээгүй функцийг дуудвал Python түүнийг мэдэхгүй бөгөөд алдаа гаргана. (Энэ алдааг 5-р хэсэгт үзнэ.)

### Өөр жишээнүүд

````python
# Зураас зурдаг функц
def draw_line():
    print("--------------------")

draw_line()
print("Python Basics")
draw_line()
````

````python
# Тоглоомын дүрмийг хэвлэдэг функц
def show_rules():
    print("Rule 1: Be kind.")
    print("Rule 2: Take turns.")
    print("Rule 3: Have fun!")

print("Before we start, here are the rules:")
show_rules()
````

## 3. Функцэд ямар нэг зүйл дамжуулах

### Асуудал

Бидний `say_hello()` функц үргэлж яг ижил зүйл хэлдэг. Гэтэл **өөр өөр хүмүүстэй** нэрээр нь мэндлэхийг хүсвэл яах вэ?

- "Hello, Anna!"
- "Hello, Tom!"
- "Hello, Bill!"

Хүн бүрд өөр функц үүсгэмээргүй байна. **Нэг** функцтэй байгаад, удаа бүр нэрийг нь **хэлж** өгмөөр байна.

### Кофе шопд захиалах үеийг бодоорой

Кофе шопд захиалахдаа та ажилтанд юу хүсч байгаагаа **хэлдэг**: "Нэг кофе өгөөч." эсвэл "Нэг цай өгөөч." Ажилтан ижил ажил (ундаа бэлтгэх) хийдэг ч үр дүн нь **таны юу хэлснээс** хамаардаг.

Функц ч ингэж ажиллаж болно. Дуудахдаа бид түүнд утга **дамжуулж** болох бөгөөд функц ажлаа **тэр утгаар** хийнэ.

### Утга хүлээн авдаг функц үүсгэх

```python
def greet(name):
    print("Hello,", name)
```

Дугуй хаалтыг хараарай: дотор нь `name` гэсэн үг байна. Энэ бол функцийг дуудахад дамжуулсан утгыг хүлээн авах **хувьсагч**. Энэ хувьсагчийг **параметр (parameter)** гэдэг.

### Утгатайгаар дуудах

```python
greet("Anna")
```

Одоо функцийг дуудахдаа хаалтан дотор утга бичнэ. Бидний дамжуулсан утгыг **аргумент (argument)** гэдэг.

| Үг | Хаана байх вэ | Жишээ |
|---|---|---|
| **параметр (parameter)** | функцийг **үүсгэх** үеийн `( )` дотор | `def greet(name):` дахь `name` |
| **аргумент (argument)** | функцийг **дуудах** үеийн `( )` дотор | `greet("Anna")` дахь `"Anna"` |

Эдгээр хоёр үгэнд нэг их санаа зовох хэрэггүй. Чухал санаа нь: **бидний дамжуулсан утга функцийн доторх хувьсагчид хадгалагдана.**

````python
# Нэг утга хүлээн авдаг функц үүсгэнэ
def greet(name):
    print("Hello,", name)

# Өөр өөр утгатайгаар дуудна
greet("Anna")
greet("Tom")
greet("Bill")
````

**`print("Hello,", name)`-ийн тухай:** `print()` дотор хэд хэдэн утгыг **таслал** `,`-аар тусгаарлаж бичвэл тэдгээр нь нэг мөрөнд хэвлэгдэж, Python хооронд нь **нэг зай** автоматаар тавина. Хашилт доторх бичвэр байгаагаараа хэвлэгдэж, хувьсагчийн оронд түүн доторх утга хэвлэгдэнэ.

### Юу болох вэ, алхам алхмаар

`greet("Anna")`-г дуудахад:

| Алхам | Python юу хийх вэ |
|---|---|
| 1 | `greet` функцийг олно |
| 2 | Дамжуулсан утгыг параметрт хадгална: `name = "Anna"` |
| 3 | Биеийг ажиллуулна: `print("Hello,", name)` → `name` нь `"Anna"` → `Hello, Anna` гэж хэвлэнэ |
| 4 | Функц дууссан тул Python дуудлагын дараагийн мөр рүү буцна |

Дараа нь `greet("Tom")`-г дуудахад дахиад ижил зүйл болно, гэхдээ энэ удаа `name = "Tom"`.

Тэгэхээр **функцийн доторх код удаа бүр ижил**, гэхдээ бидний дамжуулсан утгаас хамаарч **үр дүн өөрчлөгдөнө**.

````python
# Параметр бол зүгээр л хувьсагч. Дотор нь юу байгааг харж болно
def show_value(x):
    print("The function received:", x)

show_value(10)
show_value("apple")
show_value(3.14)
````

````python
# Утгыг тооцоололд ашигладаг функц
def show_double(number):
    print(number * 2)       # number-ийн оронд бидний дамжуулсан утга орно

show_double(5)              # number = 5  -> 5 * 2
show_double(100)            # number = 100 -> 100 * 2
````

````python
# Аргумент нь хувьсагч ч байж болно
def greet(name):
    print("Hello,", name)

student = "Kate"
greet(student)              # student доторх утга ("Kate") функцэд дамжуулагдана
````

### Хоёр утга дамжуулах

Функц **нэгээс олон** утга хүлээн авч болно. Функцийг үүсгэхдээ ч, дуудахдаа ч тэдгээрийг таслал `,`-аар тусгаарлана.

```python
def add(a, b):
    print(a + b)

add(2, 3)
```

Утгууд **дарааллаараа** таарна:
- 1-р утга (`2`) 1-р параметрт (`a`) орно
- 2-р утга (`3`) 2-р параметрт (`b`) орно

````python
def add(a, b):
    print(a + b)

add(2, 3)                   # a = 2, b = 3  -> 2 + 3
add(10, 20)                 # a = 10, b = 20 -> 10 + 20
````

````python
# Хоёр параметртэй функц: нэр ба нас
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce("Anna", 20)       # name = "Anna", age = 20
introduce("Tom", 25)        # name = "Tom", age = 25
````

### Дараалал чухал

Утгууд дарааллаараа таардаг тул дарааллыг өөрчилбөл үр дүн өөрчлөгдөж болно.

````python
def subtract(a, b):
    print(a - b)

subtract(10, 3)             # a = 10, b = 3  -> 10 - 3 = 7
subtract(3, 10)             # a = 3, b = 10  -> 3 - 10 = -7 (өөр үр дүн!)
````

````python
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce(20, "Anna")       # Дараалал буруу: name = 20, age = "Anna"
````

## 4. Үр дүнг буцааж авах: return

### Асуудал

Бидний `add(a, b)` функц үр дүнг дэлгэцэнд **хэвлэдэг**. Гэтэл үр дүнг дараа нь **ашиглахыг** хүсвэл яах вэ? Жишээ нь, хувьсагчид хадгалах эсвэл түүгээр тооцоолох?

Хэвлэх нь үр дүнг зөвхөн **харуулдаг**. Үр дүнг бидэнд **буцааж өгдөггүй**. Нэгэнт хэвлэгдсэн бол бид түүнийг дахин ашиглаж чадахгүй.

### Даалгавар өгөхийг бодоорой

Найздаа "Дэлгүүр орж надад талх аваад ирээч" гэж гуйсан гэж төсөөлөөрэй.

- Найз чинь очиж, талх аваад, буцаж ирнэ.
- Дараа нь найз чинь **талхыг гарт чинь өгнө**.
- Одоо **талх танд байна**, та түүгээр юу ч хийж болно: идэх, хуваах, сэндвич хийх.

`return` бол **үр дүнг буцааж өгөх** шиг. Функц ажлаа хийгээд, үр дүнг өөрийг нь дуудсан газар руу буцааж өгнө.

### return-тэй функц үүсгэх

```python
def add(a, b):
    return a + b
```

`return` нь: **"энэ утгыг функцийг дуудсан газар руу буцааж өг"** гэсэн утгатай.

Анхаараарай: энэ функц юу ч хэвлэдэггүй. Зөвхөн үр дүнг буцааж өгдөг.

````python
def add(a, b):
    return a + b            # Үр дүнг буцааж өгнө

result = add(2, 3)          # Үр дүн (5) буцаж ирээд result-д хадгалагдана
print(result)
````

### Дуудлага буцаасан утгаар солигдоно

Энэ бол `return`-ийн хамгийн чухал санаа.

Функц утга буцаахад **таны функцийг дуудсан газар тэр утгаар солигдоно.**

Энэ мөрийг удаанаар харъя:

```python
result = add(2, 3)
```

| Алхам | Python юу харах вэ |
|---|---|
| 1 | `result = add(2, 3)` |
| 2 | Python `add(2, 3)`-г дуудна: `a = 2`, `b = 3`, функц `2 + 3`, өөрөөр хэлбэл `5`-ыг буцаана |
| 3 | `add(2, 3)` нь `5`-аар **солигдоно** → `result = 5` |
| 4 | `5` нь `result`-д хадгалагдана |

Тэгэхээр функц утга буцаасны дараа та `result = 5` гэж бичсэнтэй адил болно.

`add(2, 3)`-ийг функц дуусахад **5 болж хувирдаг** хайрцаг гэж төсөөлж болно.

````python
def add(a, b):
    return a + b

x = add(2, 3)               # add(2, 3) -> 5,  тиймээс энэ нь: x = 5 болно
y = add(10, 20)             # add(10, 20) -> 30, тиймээс энэ нь: y = 30 болно

print(x)
print(y)
````

### Буцаасан утгыг тооцоололд ашиглах

### print ба return: ялгаа нь юу вэ?

Одоо бараг адилхан харагддаг хоёр функц үүсгэе. Нэг нь `print`, нөгөө нь `return` ашиглана.

```python
def show_add(a, b):        # print ашиглана
    print(a + b)

def add(a, b):             # return ашиглана
    return a + b
```

| | Функц доторх `print` | Функц доторх `return` |
|---|---|---|
| Юу хийх вэ | үр дүнг дэлгэцэнд **харуулна** | үр дүнг дуудсан газар руу **буцааж өгнө** |
| Үр дүнг хувьсагчид хадгалж болох уу? | ❌ | ✅ |
| Үр дүнгээр тооцоолж болох уу? | ❌ | ✅ |
| Адилхан... | талхаа харуулаад өөртөө үлдээдэг найз | талхыг гарт чинь өгдөг найз |

### return байхгүй бол юу буцаж ирэх вэ?

Функцэд `return` байхгүй ч ямар нэг зүйл буцаадаг: **`None`** гэдэг тусгай утга. `None` нь **"юу ч биш"** гэсэн утгатай.

Тиймээс `show_add(2, 3)` нь `5` биш, `None`-оор солигдоно.

````python
def show_add(a, b):
    print(a + b)                # Зөвхөн үр дүнг харуулна. return байхгүй

result = show_add(2, 3)         # 5 дэлгэцэнд хэвлэгдэнэ... гэхдээ result-д юу хадгалагдсан бэ?
print("result is:", result)     # None! show_add(2, 3) нь 5 биш, None-оор солигдсон
````

Гаралтыг анхааралтай хараарай:

1. `5` - **функц өөрөө** (доторх `print`) хэвлэсэн
2. `result is: None` - функц юу ч буцаагаагүй тул `result`-д `None` хадгалагдсан

5 гэсэн тоо дэлгэцэнд гарсан ч **бид түүнийг буцааж аваагүй**. Алга болсон.

### Тиймээс add(2, 3) + add(3, 4) нь print-тэй ажиллахгүй

`show_add(2, 3) + show_add(3, 4)`-г оролдвол юу болох вэ?

| Алхам | Python юу харах вэ |
|---|---|
| 1 | `show_add(2, 3) + show_add(3, 4)` |
| 2 | `show_add(2, 3)` 5-ыг хэвлээд **`None`**-оор солигдоно |
| 3 | `show_add(3, 4)` 7-г хэвлээд **`None`**-оор солигдоно |
| 4 | `None + None` → **алдаа!** Python "юу ч биш" ба "юу ч биш"-ийг нэмж чадахгүй |

````python
def show_add(a, b):
    print(a + b)

# total = show_add(2, 3) + show_add(3, 4)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'NoneType'
````

(Дээрх мөр нь notebook зогсохгүйн тулд тайлбар хэлбэртэй. `total = ...` мөрний хоёр `#`-г арилгавал эхлээд `5` ба `7` хэвлэгдээд, дараа нь алдаа гарахыг харна.)

Алдааны мэдэгдэл нь: "`None` ба `None`-д `+` ашиглаж болохгүй" гэсэн утгатай. (`NoneType` бол `None`-ийн өгөгдлийн төрөл.)

### Алийг нь хэзээ ашиглах вэ?

- Зөвхөн дэлгэцэнд ямар нэг зүйл **харуулахыг** хүсвэл → `print` болно
- Үр дүнг дараа нь **ашиглахыг** (хадгалах, тооцоолох, өөр функцэд дамжуулах) хүсвэл → `return` ашиглана

Ихэнх тохиолдолд ямар нэг зүйл тооцоолдог функц `return` ашиглах ёстой. Тэгвэл үр дүнг хүссэн үедээ `print(add(2, 3))` гэж хэвлэж болно.

````python
def add(a, b):
    return a + b

# return-тэй бол хоёуланг нь хийж болно: үр дүнг ашиглах БА хэвлэх
total = add(2, 3) + add(3, 4)
print(total)
````

## 5. Түгээмэл алдаанууд

Эдгээр нь функцийг сурч байхдаа бараг бүх хүний гаргадаг алдаанууд. Алдаа бүрийн хувьд бид дараахыг харна:

- ❌ **Буруу код**
- 🔍 **Юу буруу болсон бэ**
- ✅ **Зөв арга**

Алдаа гаргадаг кодыг notebook зогсохгүйн тулд тайлбар хэлбэрээр (эхэнд нь `#`-тэй) бичсэн. Жинхэнэ алдааны мэдэгдлийг харахыг хүсвэл `#`-г арилгаж, нүдийг ажиллуулаад, дараа нь `#`-г буцааж тавиарай.

### Алдаа 1: Функц үүсгээд өөрөө ажиллана гэж бодох

````python
# ❌ Буруу: "Good morning!" гэж хэвлэхийг хүсэж байна
def morning():
    print("Good morning!")

# Юу ч хэвлэгдсэнгүй...
````

🔍 **Юу буруу болсон бэ?**

- `def` нь функцийг зөвхөн **үүсгэдэг**. Ажиллуулдаггүй. (Жор бичих нь хоол бий болгодоггүй.)
- Алдаа гарахгүй тул будлиулж болно: "Яагаад юу ч болсонгүй вэ?"

✅ **Зөв арга:** Функцийг үүсгэсний дараа **дуудна**.

````python
# ✅ Зөв
def morning():
    print("Good morning!")

morning()               # Дуудна
````

### Алдаа 2: Дуудахдаа хаалтаа мартах

````python
# ❌ Буруу
def morning():
    print("Good morning!")

morning                 # Хаалтгүй -> функц АЖИЛЛАХГҮЙ
print("Did you see 'Good morning!' above? No!")
````

🔍 **Юу буруу болсон бэ?**

- `()`-гүй `morning` бол функцийн зүгээр л **нэр**. Дуудлага биш.
- Функцийг дуудахын тулд хаалтыг **заавал** бичих ёстой: `morning()`. Хаалт нь "одоо ажиллуул" гэсэн утгатай.
- Алдаа гарахгүй ч функц ажиллахгүй.

✅ **Зөв арга:** Дуудахдаа нэрийн дараа үргэлж `()` бичнэ.

````python
# ✅ Зөв
def morning():
    print("Good morning!")

morning()               # Хаалттай
````

### Алдаа 5: Утгын тоог буруу өгөх

````python
# ❌ Буруу
def add(a, b):
    return a + b

# add(2)          # TypeError: add() missing 1 required positional argument: 'b'
# add(2, 3, 4)    # TypeError: add() takes 2 positional arguments but 3 were given
````

🔍 **Юу буруу болсон бэ?**

- `add` нь **2** параметртэй (`a` ба `b`) үүсгэгдсэн тул яг **2** утга хэрэгтэй.

- `add(2, 3, 4)`: 3 утга өгсөн ч ердөө 2 байр бий. Алдаа нь "2 авдаг, гэтэл 3 өгсөн" гэж хэлж байна.

✅ **Зөв арга:** Параметрийн тоотой яг ижил тооны утга өгнө.

````python
# ✅ Зөв
def add(a, b):
    return a + b

print(add(2, 3))        # 2 параметр, 2 утга
````

### Алдаа 6: return хэрэгтэй үед print ашиглах

````python
# ❌ Буруу: үр дүнг хадгалаад дараа нь ашиглахыг хүсэж байна
def multiply(a, b):
    print(a * b)                # Зөвхөн үр дүнг харуулна

answer = multiply(3, 4)         # 12 хэвлэгдэнэ... гэхдээ answer нь None
print("answer:", answer)
````

🔍 **Юу буруу болсон бэ?**

- `multiply` нь `return` биш `print` ашигладаг. Энэ нь 12-ыг дэлгэцэнд **харуулдаг** ч **буцааж өгдөггүй**.
- `return`-гүй функц `None` буцаадаг. Тиймээс `multiply(3, 4)` нь `None`-оор солигдож, `answer` нь `None` болно.
- Түүгээр тооцоолох гэвэл (жишээ нь `answer + 1`) алдаа гарна.

✅ **Зөв арга:** Үр дүнг буцааж өгөхийн тулд `return` ашиглана. Харахыг хүсвэл функцийн гадна хэвлэнэ.

````python
# ✅ Зөв
def multiply(a, b):
    return a * b                # Үр дүнг буцааж өгнө

answer = multiply(3, 4)         # answer = 12 болно
print("answer:", answer)
print("answer + 1:", answer + 1)
````

- Алдаа гарахгүй ч үр дүн алга болсон.
