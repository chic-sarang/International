## 1. Что такое функция?

### Проблема

Допустим, мы хотим вывести короткое приветствие:

```python
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
```

Теперь представьте, что это приветствие нужно вывести **3 раза** в разных местах программы. Придётся снова и снова писать те же 3 строки.

````python
# Те же 3 строки, написанные 3 раза
print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")

print("Hello!")
print("Welcome to Python class.")
print("Have a nice day!")
````

Это работает, но есть проблемы:

- Это длинно, и писать утомительно.
- Если захочется заменить "Hello!" на "Hi!", придётся менять это **везде**. Если забыть хотя бы одно место, приветствия будут разными.

### Идея функции

А что, если написать эти 3 строки **только один раз**, дать им **имя**, а потом просто называть это имя, когда они нужны?

Именно это и есть **функция**: **группа кода с именем**.

### Представьте рецепт

Функция похожа на **рецепт**.

1. Сначала вы **записываете рецепт** на бумаге: "Как приготовить блины: шаг 1, шаг 2, шаг 3." Если записать рецепт, блины не появятся. Это просто инструкция, сохранённая на потом.
2. Позже, когда захотите блинов, вы **готовите по рецепту**. По одному и тому же рецепту можно готовить сколько угодно раз.

Функция работает так же:

| Рецепт | Функция |
|---|---|
| Записать рецепт | **Создать** (определить) функцию |
| Приготовить по рецепту | **Использовать** (вызвать) функцию |

### Вы уже пользовались функциями!

Вы пользуетесь функциями с Урока 1. `print()`, `len()` и `type()` - это всё функции. Кто-то уже написал их для вас, а вы просто использовали их, написав имя и круглые скобки `( )`.

В этом уроке вы научитесь **создавать собственные** функции.

````python
# Функции, которыми вы уже пользовались
print("Hello")              # print() - функция: она показывает значение на экране
print(len("apple"))         # len() - функция: она считает
print(type(10))             # type() - функция: она показывает тип данных
````

## 2. Создание функции и её использование

Функцией пользуются в **два отдельных шага**:

1. **Создать** функцию (записать рецепт)
2. **Использовать** функцию (приготовить по рецепту)

### Шаг 1: Создаём функцию

```python
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")
```

Разберём каждую часть:

| Часть | Что значит |
|---|---|
| `def` | сокращение от **define** (определить). Означает "сейчас я создам функцию" |
| `say_hello` | **имя** функции. Его выбираете вы (правила те же, что для имён переменных) |
| `()` | круглые скобки. После имени они нужны всегда (что пишется внутри, увидим в разделе 3) |
| `:` | двоеточие в конце строки, как у `if` и `for` |
| строки с отступом | **тело** функции: код, который относится к функции |

**Об отступах:** как и у `if` и `for`, Python по отступу (4 пробела в начале строки) понимает, какие строки относятся к функции. Строки с отступом под `def` - это тело функции. Первая строка без отступа находится **вне** функции.

### Создание функции НЕ запускает её

Это самое важное в этом разделе.

Когда Python читает `def`, он только **запоминает** функцию. Код внутри он не выполняет. Так же, как запись рецепта не даёт блинов.

Посмотрите на ячейку ниже. Мы создаём функцию, но **ничего не выводится**.

````python
# Шаг 1: создаём функцию
def say_hello():
    print("Hello!")
    print("Welcome to Python class.")
    print("Have a nice day!")

# Ничего не выведено! Мы только создали (записали) функцию. Мы её ещё не использовали.
````

### Шаг 2: Используем функцию

Чтобы использовать функцию, напишите её **имя** и после него **круглые скобки** `()`:

```python
say_hello()
```

Использование функции называется **вызовом** функции. Когда вы вызываете функцию, Python идёт в неё, выполняет весь код внутри сверху вниз, а потом возвращается обратно.

````python
# Шаг 2: вызываем (используем) функцию
say_hello()
````

### Один рецепт - много раз

Когда функция создана, её можно вызывать **сколько угодно раз**. Каждый вызов снова выполняет всё тело.

````python
# То же приветствие 3 раза, но теперь всего 3 короткие строки
say_hello()
say_hello()
say_hello()
````

Сравните с началом урока. Раньше нам было нужно 9 строк. Теперь нужно всего 3 вызова.

А если захочется заменить "Hello!" на "Hi!", менять нужно **только в одном месте**: внутри функции. Все вызовы будут использовать новое приветствие.

````python
# Меняем функцию в одном месте...
def say_hello():
    print("Hi!")                        # Изменили только здесь
    print("Welcome to Python class.")
    print("Have a nice day!")

# ...и все вызовы используют новую версию
say_hello()
say_hello()
````

### Создание и использование в одной ячейке

Обычно мы сначала создаём функцию, а ниже вызываем её. Проследим, что делает Python, шаг за шагом.

````python
def say_goodbye():          # 1. Python запоминает функцию "say_goodbye". Пока ничего не выполняется
    print("Goodbye!")
    print("See you next time.")

print("Class is over.")     # 2. Эта строка без отступа -> вне функции -> она выполняется сейчас
say_goodbye()               # 3. Вызываем функцию -> 2 строки внутри выполняются сейчас
print("The end.")           # 4. Когда функция закончилась, Python возвращается сюда
````

Обратите внимание на порядок вывода:

1. `Class is over.` - первая строка **вне** функции
2. `Goodbye!` и `See you next time.` - тело функции выполняется, только когда мы её **вызываем**
3. `The end.` - когда функция закончилась, Python продолжает со строки после вызова

### Сначала создать, потом использовать

Рецепт нужно записать **до того**, как готовить по нему. Точно так же функцию нужно **создать до того, как её вызвать**. Python читает код сверху вниз, поэтому если вызвать функцию, которая ещё не создана, Python её не знает и показывает ошибку. (Эту ошибку мы увидим в разделе 5.)

## 3. Передаём функции значение

### Проблема

Наша функция `say_hello()` всегда говорит одно и то же. А что, если мы хотим поздороваться с **разными людьми** по имени?

- "Hello, Anna!"
- "Hello, Tom!"
- "Hello, Bill!"

Мы не хотим создавать отдельную функцию для каждого человека. Мы хотим **одну** функцию и каждый раз **сообщать** ей имя.

### Представьте заказ в кафе

Когда вы делаете заказ в кафе, вы **говорите** сотруднику, что хотите: "Один кофе, пожалуйста." или "Один чай, пожалуйста." Сотрудник делает одну и ту же работу (готовит напиток), но результат зависит от того, **что вы сказали**.

Функция может работать так же. Вызывая её, мы можем **передать** ей значение, и функция делает свою работу **с этим значением**.

### Создаём функцию, которая принимает значение

```python
def greet(name):
    print("Hello,", name)
```

Посмотрите на круглые скобки: внутри есть слово `name`. Это **переменная**, которая получит значение, переданное при вызове функции. Такая переменная называется **параметром (parameter)**.

### Вызываем её со значением

```python
greet("Anna")
```

Теперь при вызове функции мы пишем значение внутри скобок. Переданное значение называется **аргументом (argument)**.

| Слово | Где находится | Пример |
|---|---|---|
| **параметр (parameter)** | внутри `( )`, когда мы **создаём** функцию | `name` в `def greet(name):` |
| **аргумент (argument)** | внутри `( )`, когда мы **вызываем** функцию | `"Anna"` в `greet("Anna")` |

Не беспокойтесь об этих двух словах. Главная мысль: **переданное значение сохраняется в переменную внутри функции.**

````python
# Создаём функцию, которая принимает одно значение
def greet(name):
    print("Hello,", name)

# Вызываем её с разными значениями
greet("Anna")
greet("Tom")
greet("Bill")
````

**О `print("Hello,", name)`:** когда в `print()` несколько значений через **запятую** `,`, они выводятся в одну строку, и Python автоматически ставит между ними **один пробел**. Текст в кавычках выводится как есть, а вместо переменной выводится её значение.

### Что происходит, шаг за шагом

Когда мы вызываем `greet("Anna")`:

| Шаг | Что делает Python |
|---|---|
| 1 | Находит функцию `greet` |
| 2 | Сохраняет переданное значение в параметр: `name = "Anna"` |
| 3 | Выполняет тело: `print("Hello,", name)` → `name` равно `"Anna"` → выводит `Hello, Anna` |
| 4 | Функция закончилась, и Python возвращается к строке после вызова |

Потом, когда мы вызываем `greet("Tom")`, всё повторяется, но на этот раз `name = "Tom"`.

Значит, **код внутри функции каждый раз один и тот же**, но **результат меняется** в зависимости от переданного значения.

````python
# Параметр - это просто переменная. Можно посмотреть, что в ней лежит
def show_value(x):
    print("The function received:", x)

show_value(10)
show_value("apple")
show_value(3.14)
````

````python
# Функция, которая использует значение в вычислении
def show_double(number):
    print(number * 2)       # вместо number подставляется переданное значение

show_double(5)              # number = 5  -> 5 * 2
show_double(100)            # number = 100 -> 100 * 2
````

````python
# Аргументом может быть и переменная
def greet(name):
    print("Hello,", name)

student = "Kate"
greet(student)              # Значение из student ("Kate") передаётся функции
````

### Передаём два значения

Функция может принимать **больше одного** значения. Разделяйте их запятыми `,` и при создании, и при вызове функции.

```python
def add(a, b):
    print(a + b)

add(2, 3)
```

Значения сопоставляются **по порядку**:
- 1-е значение (`2`) попадает в 1-й параметр (`a`)
- 2-е значение (`3`) попадает во 2-й параметр (`b`)

````python
def add(a, b):
    print(a + b)

add(2, 3)                   # a = 2, b = 3  -> 2 + 3
add(10, 20)                 # a = 10, b = 20 -> 10 + 20
````

````python
# Функция с двумя параметрами: имя и возраст
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce("Anna", 20)       # name = "Anna", age = 20
introduce("Tom", 25)        # name = "Tom", age = 25
````

### Порядок важен

Поскольку значения сопоставляются по порядку, если поменять порядок, может измениться результат.

````python
def subtract(a, b):
    print(a - b)

subtract(10, 3)             # a = 10, b = 3  -> 10 - 3 = 7
subtract(3, 10)             # a = 3, b = 10  -> 3 - 10 = -7 (другой результат!)
````

````python
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce(20, "Anna")       # Порядок неправильный: name = 20, age = "Anna"
````

## 4. Получаем результат обратно: return

### Проблема

Наша функция `add(a, b)` **выводит** результат на экран. А что, если мы хотим **использовать** результат потом? Например, сохранить его в переменную или посчитать с ним что-то?

Вывод на экран только **показывает** результат. Он **не отдаёт** результат нам. Когда он выведен, использовать его мы уже не можем.

### Представьте поручение

Представьте, что вы просите друга: "Сходи, пожалуйста, в магазин и купи мне хлеба."

- Друг идёт, покупает хлеб и возвращается.
- Потом друг **отдаёт вам хлеб в руки**.
- Теперь **хлеб у вас**, и с ним можно делать что угодно: съесть, поделиться, сделать бутерброд.

`return` - это как **отдать результат обратно**. Функция делает свою работу, а потом отдаёт результат в то место, откуда её вызвали.

### Создаём функцию с return

```python
def add(a, b):
    return a + b
```

`return` означает: **"верни это значение туда, откуда вызвали функцию."**

Обратите внимание: эта функция ничего не выводит. Она только отдаёт результат обратно.

````python
def add(a, b):
    return a + b            # Отдаём результат обратно

result = add(2, 3)          # Результат (5) возвращается и сохраняется в result
print(result)
````

### Вызов заменяется возвращённым значением

Это самая важная идея `return`.

Когда функция возвращает значение, **место, где вы вызвали функцию, заменяется этим значением.**

Посмотрим медленно на эту строку:

```python
result = add(2, 3)
```

| Шаг | Что видит Python |
|---|---|
| 1 | `result = add(2, 3)` |
| 2 | Python вызывает `add(2, 3)`: `a = 2`, `b = 3`, и функция возвращает `2 + 3`, то есть `5` |
| 3 | `add(2, 3)` **заменяется** на `5` → `result = 5` |
| 4 | `5` сохраняется в `result` |

Значит, после возврата из функции всё выглядит так, будто вы написали `result = 5`.

Можно представить `add(2, 3)` как коробку, которая **превращается в 5**, когда функция закончила работу.

````python
def add(a, b):
    return a + b

x = add(2, 3)               # add(2, 3) -> 5,  поэтому строка становится: x = 5
y = add(10, 20)             # add(10, 20) -> 30, поэтому строка становится: y = 30

print(x)
print(y)
````

### Используем возвращённое значение в вычислении

Поскольку `add(2, 3)` превращается в `5`, его можно использовать везде, где можно использовать число `5`.

```python
add(2, 3) + add(3, 4)
```

| Шаг | Что видит Python |
|---|---|
| 1 | `add(2, 3) + add(3, 4)` |
| 2 | `add(2, 3)` выполняется и возвращает 5 → **`5`** `+ add(3, 4)` |
| 3 | `add(3, 4)` выполняется и возвращает 7 → `5 +` **`7`** |
| 4 | `12` |

````python
def add(a, b):
    return a + b

total = add(2, 3) + add(3, 4)   # становится 5 + 7
print(total)                    # 12
````

````python
def add(a, b):
    return a + b

print(add(2, 3) * 10)           # add(2, 3) становится 5 -> print(5 * 10) -> 50
print(add(2, 3) > 4)            # add(2, 3) становится 5 -> print(5 > 4) -> True
````

### Вызов прямо внутри print()

`print(add(2, 3))` работает так же:

| Шаг | Что видит Python |
|---|---|
| 1 | `print(add(2, 3))` |
| 2 | `add(2, 3)` возвращает 5 → `print(5)` |
| 3 | выводится `5` |

````python
def add(a, b):
    return a + b

print(add(2, 3))                # становится print(5)
````

### Передаём возвращённое значение другой функции

Возвращённое значение можно даже снова передать функции. Python всегда работает **изнутри наружу**: сначала заменяется внутренний вызов.

```python
add(add(1, 2), 3)
```

| Шаг | Что видит Python |
|---|---|
| 1 | `add(add(1, 2), 3)` |
| 2 | Внутренний `add(1, 2)` возвращает 3 → `add(` **`3`** `, 3)` |
| 3 | `add(3, 3)` возвращает 6 → **`6`** |

````python
def add(a, b):
    return a + b

result = add(add(1, 2), 3)      # add(1, 2) -> 3, потом add(3, 3) -> 6
print(result)
````

````python
# Функция, которая возвращает строку
def make_greeting(name):
    return "Hello, " + name     # + соединяет две строки

message = make_greeting("Anna")  # становится message = "Hello, Anna"
print(message)
````

### print или return: в чём разница?

Создадим две функции, которые выглядят почти одинаково. Одна использует `print`, другая - `return`.

```python
def show_add(a, b):        # использует print
    print(a + b)

def add(a, b):             # использует return
    return a + b
```

| | `print` внутри функции | `return` внутри функции |
|---|---|---|
| Что делает | **показывает** результат на экране | **отдаёт** результат тому, кто вызвал |
| Можно сохранить результат в переменную? | ❌ | ✅ |
| Можно считать с результатом? | ❌ | ✅ |
| Похоже на... | друга, который показал вам хлеб и оставил его себе | друга, который отдал вам хлеб в руки |

### Что возвращается, если return нет?

Если в функции нет `return`, она всё равно что-то возвращает: особое значение **`None`**. `None` означает **"ничего"**.

Поэтому `show_add(2, 3)` заменяется на `None`, а не на `5`.

````python
def show_add(a, b):
    print(a + b)                # Только показывает результат. return нет

result = show_add(2, 3)         # 5 выводится на экран... но что сохранилось в result?
print("result is:", result)     # None! show_add(2, 3) заменилось на None, а не на 5
````

Внимательно посмотрите на вывод:

1. `5` - выведено **самой функцией** (командой `print` внутри неё)
2. `result is: None` - в `result` сохранено `None`, потому что функция ничего не вернула

Число 5 появилось на экране, но **мы его не получили обратно**. Оно пропало.

### Поэтому add(2, 3) + add(3, 4) не работает с print

Что будет, если попробовать `show_add(2, 3) + show_add(3, 4)`?

| Шаг | Что видит Python |
|---|---|
| 1 | `show_add(2, 3) + show_add(3, 4)` |
| 2 | `show_add(2, 3)` выводит 5 и заменяется на **`None`** |
| 3 | `show_add(3, 4)` выводит 7 и заменяется на **`None`** |
| 4 | `None + None` → **ошибка!** Python не может сложить "ничего" и "ничего" |

````python
def show_add(a, b):
    print(a + b)

# total = show_add(2, 3) + show_add(3, 4)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'NoneType'
````

(Строка выше - комментарий, чтобы не останавливать блокнот. Если убрать два `#` в строке `total = ...`, вы сначала увидите выведенные `5` и `7`, а потом ошибку.)

Сообщение об ошибке значит: "`+` нельзя применить к `None` и `None`". (`NoneType` - это тип данных `None`.)

### Что когда использовать?

- Если нужно только **показать** что-то на экране → подойдёт `print`
- Если результат нужно **использовать** потом (сохранить, посчитать с ним, передать другой функции) → используйте `return`

Чаще всего функции, которые что-то вычисляют, должны использовать `return`. Тогда результат всё равно можно вывести в любой момент: `print(add(2, 3))`.

````python
def add(a, b):
    return a + b

# С return можно и то, и другое: использовать результат И вывести его
total = add(2, 3) + add(3, 4)
print(total)
````

## 5. Частые ошибки

Эти ошибки делает почти каждый, кто изучает функции. Для каждой мы посмотрим:

- ❌ **Неправильный код**
- 🔍 **Что пошло не так**
- ✅ **Правильный способ**

Код, который вызывает ошибку, записан в виде комментария (с `#` в начале), чтобы он не останавливал блокнот. Если хотите увидеть настоящее сообщение об ошибке, уберите `#`, запустите ячейку, а потом верните `#` обратно.

### Ошибка 1: Создать функцию и ждать, что она выполнится

````python
# ❌ Неправильно: мы хотим вывести "Good morning!"
def morning():
    print("Good morning!")

# Ничего не выведено...
````

🔍 **Что пошло не так?**

- `def` только **создаёт** функцию. Он её не выполняет. (Если записать рецепт, еда не появится.)
- Ошибки нет, поэтому это может сбить с толку: "Почему ничего не произошло?"

✅ **Правильный способ:** После создания функции **вызовите** её.

````python
# ✅ Правильно
def morning():
    print("Good morning!")

morning()               # Вызываем её
````

### Ошибка 2: Забыть скобки при вызове

````python
# ❌ Неправильно
def morning():
    print("Good morning!")

morning                 # Без скобок -> функция НЕ выполняется
print("Did you see 'Good morning!' above? No!")
````

🔍 **Что пошло не так?**

- `morning` без `()` - это просто **имя** функции. Это не вызов.
- Чтобы вызвать функцию, **обязательно** пишите скобки: `morning()`. Скобки значат "выполни её сейчас".
- Ошибки нет, но функция не выполняется.

✅ **Правильный способ:** При вызове всегда пишите `()` после имени.

````python
# ✅ Правильно
def morning():
    print("Good morning!")

morning()               # Со скобками
````

### Ошибка 3: Вызвать функцию до того, как её создали

````python
# ❌ Неправильно
# evening()             # NameError: name 'evening' is not defined
#
# def evening():
#     print("Good evening!")
````

🔍 **Что пошло не так?**

- Python читает код **сверху вниз**.
- Когда он доходит до `evening()`, функция ещё не создана. Python не знает функции с именем `evening`.
- Ошибка - `NameError` ("ничего с таким именем нет").

✅ **Правильный способ:** Сначала создайте функцию, а ниже вызовите её.

````python
# ✅ Правильно
def evening():
    print("Good evening!")

evening()               # Вызвана после создания
````

### Ошибка 4: Забытое двоеточие или отступ

````python
# ❌ Неправильно
# def night()                 # SyntaxError: нет двоеточия
#     print("Good night!")

# def night():
# print("Good night!")        # IndentationError: у тела нет отступа
````

🔍 **Что пошло не так?**

- Это те же правила, что и для `if` и `for`.
- Строка `def` должна заканчиваться двоеточием `:`.
- Тело функции должно быть с отступом (4 пробела). Без отступа Python не знает, какие строки относятся к функции.

✅ **Правильный способ:**

````python
# ✅ Правильно
def night():                # Двоеточие в конце
    print("Good night!")    # Отступ 4 пробела

night()
````

### Ошибка 5: Неправильное количество значений

````python
# ❌ Неправильно
def add(a, b):
    return a + b

# add(2)          # TypeError: add() missing 1 required positional argument: 'b'
# add(2, 3, 4)    # TypeError: add() takes 2 positional arguments but 3 were given
````

🔍 **Что пошло не так?**

- `add` создана с **2** параметрами (`a` и `b`), поэтому ей нужно ровно **2** значения.
- `add(2)`: передано только 1 значение. `a` получает 2, а `b` не получает ничего. Ошибка говорит "не хватает 1 значения: `b`".
- `add(2, 3, 4)`: передано 3 значения, а места есть только для 2. Ошибка говорит "принимает 2, а передано 3".

✅ **Правильный способ:** Передавайте ровно столько значений, сколько параметров.

````python
# ✅ Правильно
def add(a, b):
    return a + b

print(add(2, 3))        # 2 параметра, 2 значения
````

### Ошибка 6: print там, где нужен return

````python
# ❌ Неправильно: мы хотим сохранить результат и использовать его потом
def multiply(a, b):
    print(a * b)                # Только показывает результат

answer = multiply(3, 4)         # 12 выводится... но в answer лежит None
print("answer:", answer)
````

🔍 **Что пошло не так?**

- `multiply` использует `print`, а не `return`. Она **показывает** 12 на экране, но **не отдаёт** его обратно.
- Функция без `return` возвращает `None`. Поэтому `multiply(3, 4)` заменяется на `None`, и в `answer` лежит `None`.
- Если попробовать посчитать с ним (например, `answer + 1`), будет ошибка.

✅ **Правильный способ:** Используйте `return`, чтобы отдать результат. А если хотите его увидеть, выведите его вне функции.

````python
# ✅ Правильно
def multiply(a, b):
    return a * b                # Отдаём результат обратно

answer = multiply(3, 4)         # становится answer = 12
print("answer:", answer)
print("answer + 1:", answer + 1)
````

### Ошибка 7: Вернуть результат, но не использовать его

````python
# ❌ Неправильно: мы хотим увидеть результат
def multiply(a, b):
    return a * b

multiply(3, 4)                  # Результат (12) возвращается... но его никто не ловит!
print("Where is the result?")
````

🔍 **Что пошло не так?**

- `multiply(3, 4)` заменяется на `12`. Но в этой строке просто `12` само по себе. Мы его не сохранили и не вывели.
- Это как если бы друг отдал вам хлеб, а вы уронили его на пол.
- Ошибки нет, но результат потерян.

✅ **Правильный способ:** Поймайте возвращённое значение: сохраните его в переменную или поставьте вызов внутрь `print()`.

````python
# ✅ Правильно
def multiply(a, b):
    return a * b

result = multiply(3, 4)         # Сохраняем
print(result)

print(multiply(3, 4))           # Или выводим сразу
````
