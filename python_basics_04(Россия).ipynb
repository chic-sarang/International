## 1. Что такое список?

В Уроке 1 мы узнали, что переменная похожа на коробку, в которой хранится **одно** значение.

А что, если нужно сохранить оценки за экзамен 5 учеников? С тем, что мы знаем сейчас, понадобится 5 переменных:

```python
score1 = 85
score2 = 92
score3 = 78
score4 = 90
score5 = 66
```

Это уже много. А если учеников 30? Или 300?

Эту проблему решает **список (list)**. Список - это одна переменная, которая может хранить **много значений по порядку**.

Список можно представить как ряд **шкафчиков**:
- У всего ряда шкафчиков одно имя (имя списка).
- В каждом шкафчике лежит одно значение.
- На каждом шкафчике написан номер, чтобы его можно было найти.

Как создать список:
- Поместите значения внутрь квадратных скобок `[ ]`.
- Разделите значения запятыми `,`.

```python
scores = [85, 92, 78, 90, 66]
```

````python
# Список целых чисел
# Одна переменная (scores) теперь хранит 5 значений
scores = [85, 92, 78, 90, 66]
print(scores)           # print() показывает весь список, включая квадратные скобки
````

````python
# В списке может храниться любой тип данных, который вы изучили в Уроке 1
fruits = ["apple", "banana", "cherry"]      # Список строк (str)
heights = [175.5, 162.3, 180.1]             # Список дробных чисел (float)
answers = [True, False, True, True]         # Список логических значений (bool)

print(fruits)
print(heights)
print(answers)
````

````python
# В одном списке можно смешивать разные типы данных
# Так можно, но обычно список проще понять, если в нём значения только одного вида
student = ["Bill", 20, 175.5, True]         # имя, возраст, рост, студент ли
print(student)
````

````python
# Список может быть и пустым (без единого значения)
# Это удобно, когда нужно начать с нуля и добавлять значения позже (как это делать, узнаем в разделе 5)
empty_list = []
print(empty_list)
````

````python
# Список - это новый тип данных: list
# type() показывает тип данных значения
print(type(scores))     # <class 'list'> означает "это список"
print(type(85))         # <class 'int'> означает "это целое число" (одно из значений внутри списка)
````

## 2. Индекс: поиск значения по его позиции

У каждого значения в списке есть номер позиции. Этот номер называется **индексом (index)**. Представьте его как номер, написанный на каждом шкафчике.

Самое важное, что нужно запомнить: **Python начинает счёт с 0, а не с 1.**

Поэтому первое значение находится под индексом 0, второе - под индексом 1, третье - под индексом 2, и так далее. (Помните `range(5)` из Урока 3? Он тоже начинался с 0: 0, 1, 2, 3, 4.)

```python
fruits = ["apple", "banana", "cherry", "grape", "melon"]
```

| Значение | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |
| Позиция (как считают люди) | 1-е | 2-е | 3-е | 4-е | 5-е |

Простой способ запомнить: **индекс = позиция - 1**. У 1-го значения индекс 0, у 5-го значения индекс 4.

Чтобы получить значение, напишите имя списка, а затем индекс в квадратных скобках:

```python
fruits[0]     # "значение под индексом 0 в fruits"
```

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Получаем значение по индексу
print(fruits[0])        # 1-е значение под индексом 0 -> apple
print(fruits[1])        # 2-е значение под индексом 1 -> banana
print(fruits[2])        # 3-е значение под индексом 2 -> cherry
print(fruits[3])        # 4-е значение под индексом 3 -> grape
print(fruits[4])        # 5-е (последнее) значение под индексом 4 -> melon
````

````python
# То же самое со списком чисел
# Будьте внимательны: не путайте индекс (позицию) и значение (то, что внутри)
numbers = [10, 20, 30, 40]
#  индекс:  0   1   2   3

print(numbers[0])       # индекс 0 -> значение 10
print(numbers[2])       # индекс 2 -> значение 30 (а не 2!)
print(numbers[3])       # индекс 3 -> значение 40
````

````python
# Значения не обязаны идти в каком-то особом порядке
# Индекс говорит только о том, ГДЕ находится значение, а не о том, насколько оно большое
scores = [85, 92, 78, 90, 66]
#  индекс:  0   1   2   3   4

print(scores[1])        # Оценка 2-го ученика -> 92
print(scores[4])        # Оценка 5-го ученика -> 66
````

````python
# Можно получать значение по одному и тому же индексу сколько угодно раз
# Получение значения - это только ЧТЕНИЕ. Список при этом не меняется
colors = ["red", "green", "blue"]

print(colors[1])        # green
print(colors[1])        # всё ещё green
print(colors)           # Список остался таким же, как раньше
````

````python
# Значение, взятое из списка, можно использовать как обычную переменную
scores = [85, 92, 78, 90, 66]

first = scores[0]                  # Сохраняем первую оценку (85) в новую переменную
print(first)

print(scores[0] + scores[1])       # 85 + 92 = 177 (с ними можно считать)
print(scores[1] > scores[2])       # 92 > 78 - это True (их можно сравнивать, как в Уроке 2)
````

````python
# Строки, взятые из списка, можно соединять с помощью +, как в Уроке 1
words = ["Hello", "nice", "to", "meet", "you"]

print(words[0] + ", " + words[1] + " " + words[2] + " " + words[3] + " " + words[4])
````

````python
# Индекс может быть и переменной
# Здесь в i хранится 2, поэтому scores[i] - это то же самое, что scores[2]
scores = [85, 92, 78, 90, 66]

i = 2
print(scores[i])        # scores[2] -> 78

i = 0
print(scores[i])        # Теперь в i хранится 0, поэтому scores[0] -> 85
````

````python
# Индекс может быть даже вычислением
# Python сначала вычисляет число внутри [ ], а потом ищет этот индекс
scores = [85, 92, 78, 90, 66]

print(scores[1 + 2])    # 1 + 2 = 3 -> scores[3] -> 90

i = 1
print(scores[i + 1])    # i + 1 = 2 -> scores[2] -> 78
````

### Отрицательный индекс: счёт с конца

Теперь, когда вы знаете обычный (положительный) индекс, давайте посмотрим ещё один способ найти значение.

Python позволяет также считать **с конца списка в обратную сторону**, с помощью отрицательных чисел.

- `-1` - это **последнее** значение
- `-2` - предпоследнее значение
- `-3` - третье с конца значение, и так далее

| Значение | "apple" | "banana" | "cherry" | "grape" | "melon" |
|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 |
| Отрицательный индекс | -5 | -4 | -3 | -2 | -1 |

Обратите внимание: отрицательный индекс начинается с **-1**, а не с -0. (Индекса -0 не бывает, потому что -0 - это просто 0, то есть первое значение.)

**Зачем нужен отрицательный индекс?**

Главная причина - **быстро получить последнее значение**.

С положительным индексом, чтобы получить последнее значение, сначала нужно знать, сколько значений в списке. Если в списке 5 значений, последний индекс - 4. Если 100 значений - последний индекс 99. А если добавить или удалить значения, последний индекс тоже изменится.

С отрицательным индексом ничего считать не нужно. **`-1` - это всегда последнее значение**, какой бы длины ни был список.

````python
fruits = ["apple", "banana", "cherry", "grape", "melon"]

# Отрицательный индекс: счёт с конца
print(fruits[-1])       # -1 - это всегда последнее значение -> melon
print(fruits[-2])       # -2 - предпоследнее значение -> grape
print(fruits[-3])       # -3 - третье с конца значение -> cherry
````

````python
# Положительный и отрицательный индексы могут указывать на одно и то же значение
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[4])        # Последнее значение через положительный индекс -> melon
print(fruits[-1])       # Последнее значение через отрицательный индекс -> melon (то же самое!)

print(fruits[0])        # Первое значение через положительный индекс -> apple
print(fruits[-5])       # Первое значение через отрицательный индекс -> apple (то же самое!)
````

````python
# Почему -1 удобен: он работает для ЛЮБОГО списка, какой бы длины он ни был
# С положительным индексом последний индекс у каждого списка разный (2, 6, 0 ...)
# С -1 он всегда одинаковый
short_list = ["a", "b", "c"]
long_list = [10, 20, 30, 40, 50, 60, 70]
one_item = ["only"]

print(short_list[-1])   # c       (положительный индекс был бы 2)
print(long_list[-1])    # 70      (положительный индекс был бы 6)
print(one_item[-1])     # only    (положительный индекс был бы 0)
````

## 3. Изменение значений и len()

### Изменение значения

Список не застывший. Значение внутри него можно **изменить** в любой момент. Это как положить что-то новое в один шкафчик:

```python
имя_списка[индекс] = новое_значение
```

Это работает точно так же, как сохранение значения в переменную (`=` значит "сохранить"), но заменяется только значение в одной этой позиции. Остальные значения остаются прежними.

````python
# Изменяем значение под определённым индексом
fruits = ["apple", "banana", "cherry"]
print(fruits)

fruits[1] = "orange"        # Заменяем значение под индексом 1 ("banana") на "orange"
print(fruits)               # Изменился только индекс 1. "apple" и "cherry" остались прежними
````

````python
# Отрицательный индекс тоже подходит для изменения значений
fruits = ["apple", "banana", "cherry"]

fruits[-1] = "grape"        # Заменяем последнее значение ("cherry") на "grape"
print(fruits)
````

````python
# Изменяем значение на основе его старого значения
# Та же идея, что и age = age + 1 в Уроке 1
scores = [85, 92, 78]

scores[2] = scores[2] + 5   # Третий ученик получает 5 бонусных баллов (78 + 5 = 83)
print(scores)
````

### len(): сколько значений в списке?

`len()` показывает, сколько значений в списке. `len` - это сокращение от **length** (длина).

```python
len(имя_списка)
```

Поместите список внутрь круглых скобок, и `len()` вернёт число: количество значений в этом списке.

````python
# len(): считаем, сколько значений в списке
fruits = ["apple", "banana", "cherry", "grape", "melon"]
print(len(fruits))          # Значений 5, поэтому 5
````

````python
# Списки разной длины
three = ["a", "b", "c"]
five = [1, 2, 3, 4, 5]
one = ["only"]
empty_list = []

print(len(three))           # 3 значения -> 3
print(len(five))            # 5 значений -> 5
print(len(one))             # 1 значение -> 1
print(len(empty_list))      # В пустом списке 0 значений -> 0
````

````python
# len() работает одинаково для любого вида списка
names = ["Bill", "Anna", "Tom", "Kate"]          # Список строк
numbers = [3, 7, 1, 9, 4, 6]                     # Список целых чисел
answers = [True, False, True]                    # Список логических значений

print(len(names))           # 4
print(len(numbers))         # 6
print(len(answers))         # 3
````

````python
# len() считает, СКОЛЬКО значений в списке. Он не складывает значения
# Значения большие, но их всего 3
prices = [1000, 2500, 3000]
print(len(prices))          # 3 (а не 6500!)
````

````python
# Результат len() - это просто число (int), поэтому его можно сохранить в переменную
students = ["Bill", "Anna", "Tom", "Kate", "Mike"]

count = len(students)       # Сохраняем 5 в переменную count
print(count)
print("There are", count, "students.")
````

````python
# len() работает и со строками: он считает количество символов
# (Пробел тоже считается одним символом)
print(len("apple"))         # a, p, p, l, e -> 5
print(len("hi there"))      # h, i, (пробел), t, h, e, r, e -> 8
````

### Последний индекс - это len() - 1

Теперь мы можем связать `len()` с индексом.

Если в списке 5 значений, `len()` даёт 5, но последний индекс - **4**. Поскольку индекс начинается с 0, **последний индекс всегда равен `len() - 1`**.

Значит, есть два способа получить последнее значение:
- `fruits[len(fruits) - 1]`: посчитать значения, вычесть 1 и использовать этот индекс
- `fruits[-1]`: просто использовать -1

Оба способа дают одно и то же значение, но `-1` гораздо короче и проще. Именно поэтому отрицательный индекс так удобен.

````python
# Последний индекс - это len() - 1
fruits = ["apple", "banana", "cherry", "grape", "melon"]

last_index = len(fruits) - 1       # 5 - 1 = 4
print(last_index)
print(fruits[last_index])          # fruits[4] -> melon
````

````python
# Сравниваем два способа получить последнее значение
fruits = ["apple", "banana", "cherry", "grape", "melon"]

print(fruits[len(fruits) - 1])     # Длинный способ: len(fruits) равно 5, 5 - 1 = 4, fruits[4] -> melon
print(fruits[-1])                  # Короткий способ: -1 - последнее значение -> melon (результат тот же!)
````

## 4. Проверка, есть ли значение в списке (in)

Иногда нужно узнать: **"Есть ли это значение в списке?"**

Для этого используется `in`.

```python
значение in имя_списка
```

- Если значение есть в списке, результат - `True`.
- Если нет, результат - `False`.

Результат - `True` или `False`, как у операторов сравнения из Урока 2. Это значит, что его можно сразу использовать как условие `if`!

Чтобы проверить, что значения **нет** в списке, используйте `not in`.

**Обратите внимание:** Слово `in` вы уже видели в Уроке 3 (`for i in range(5)`). Тот `in` - часть цикла `for`. А `in`, который мы изучаем здесь, другой: он **проверяет**, есть ли значение в списке, и даёт True или False.

````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)       # "banana" есть в списке -> True
print("grape" in fruits)        # "grape" нет в списке -> False
````

````python
# not in: противоположность in
print("grape" not in fruits)    # "grape" НЕТ в списке -> True
print("apple" not in fruits)    # "apple" ЕСТЬ в списке, поэтому "not in" даёт False
````

````python
# Используем in как условие if (Урок 2)
fruits = ["apple", "banana", "cherry"]
wanted = "banana"

if wanted in fruits:                    # "banana" in fruits - это True
    print("Yes, we have it!")
else:
    print("Sorry, it's sold out.")
````

````python
# Пример из жизни: проверка списка гостей
guests = ["Bill", "Anna", "Tom"]
name = "Kate"

if name in guests:                      # "Kate" in guests - это False
    print("Welcome!")
else:
    print("Sorry, you are not on the list.")    # Поэтому выполняется эта строка
````

````python
# in проверяет значение точно. Заглавные и строчные буквы различаются
fruits = ["apple", "banana", "cherry"]
print("Apple" in fruits)        # "Apple" (с заглавной A) и "apple" - разные строки -> False
````

## 5. Добавление и удаление значений

Список может расти и уменьшаться. Можно добавлять новые значения и удалять ненужные.

| Как писать | Что делает |
|---|---|
| `имя_списка.append(значение)` | Добавляет значение в **конец** списка |
| `имя_списка.insert(индекс, значение)` | Добавляет значение в **выбранную позицию** |
| `имя_списка.remove(значение)` | Удаляет указанное значение (первое найденное) |
| `имя_списка.pop()` | Вынимает **последнее** значение |
| `имя_списка.pop(индекс)` | Вынимает значение из **выбранной позиции** |

Точка `.` означает "сделай это с этим списком". Например, `fruits.append("grape")` означает "добавь "grape" в список fruits". Команды, которые пишутся после точки вот так, называются **методами**.

Эти методы изменяют сам список. Писать `fruits = ...` не нужно.

````python
# append(): добавляем значение в конец
fruits = ["apple", "banana"]
print(fruits)

fruits.append("cherry")         # "cherry" добавляется в конец
print(fruits)

fruits.append("grape")          # "grape" снова добавляется в конец
print(fruits)
print(len(fruits))              # Список вырос с 2 до 4 значений
````

````python
# Начинаем с пустого списка и добавляем значения по одному
# Это очень распространённый приём
shopping = []                   # Начинаем с пустого списка
shopping.append("milk")
shopping.append("eggs")
shopping.append("bread")
print(shopping)
````

````python
# insert(): добавляем значение в определённую позицию
# insert(индекс, значение) -> новое значение встаёт под этим индексом, а значения после него сдвигаются на одну позицию назад
fruits = ["apple", "banana", "cherry"]

fruits.insert(1, "orange")      # Ставим "orange" под индекс 1
print(fruits)                   # "banana" и "cherry" сдвинулись на одну позицию назад

fruits.insert(0, "melon")       # Ставим "melon" под индекс 0 (в самое начало)
print(fruits)
````

````python
# remove(): удаляем значение, называя само значение
fruits = ["apple", "banana", "cherry"]

fruits.remove("banana")         # Находим "banana" и удаляем его
print(fruits)                   # Значения после него сдвигаются вперёд и заполняют пустое место
````

````python
# Если одно и то же значение встречается несколько раз, remove() удаляет только ПЕРВОЕ
numbers = [1, 2, 3, 2, 1]

numbers.remove(2)               # Удаляется только первая 2 (под индексом 1)
print(numbers)                  # Вторая 2 всё ещё на месте
````

````python
# pop(): вынимаем значение по его позиции
# Без индекса pop() вынимает ПОСЛЕДНЕЕ значение
fruits = ["apple", "banana", "cherry"]

fruits.pop()                    # Вынимаем последнее значение ("cherry")
print(fruits)
````

````python
# pop(индекс): вынимаем значение под этим индексом
fruits = ["apple", "banana", "cherry"]

fruits.pop(0)                   # Вынимаем значение под индексом 0 ("apple")
print(fruits)
````

````python
# pop() ещё и ВОЗВРАЩАЕТ вынутое значение, поэтому его можно сохранить в переменную
# remove() только удаляет. pop() удаляет И отдаёт значение вам
tasks = ["wash dishes", "do homework", "clean room"]

done = tasks.pop(0)             # Вынимаем первое дело и сохраняем его в done
print("Finished:", done)
print("Still to do:", tasks)
````

````python
# remove() и pop() с первого взгляда
# remove(значение): "удали ЭТО ЗНАЧЕНИЕ"                -> вы знаете значение
# pop(индекс):      "удали то, что стоит в ЭТОЙ ПОЗИЦИИ" -> вы знаете позицию
colors = ["red", "green", "blue"]

colors.remove("green")          # Удаление по значению
print(colors)

colors.pop(0)                   # Удаление по позиции (под индексом 0 сейчас "red")
print(colors)
````

## 6. Частые ошибки

Эти ошибки делает почти каждый, кто изучает списки. Если у вас появилась ошибка, сначала проверьте эти пункты.

1. **Забыть, что индекс начинается с 0**: `fruits[1]` - это **второе** значение, а не первое.
2. **Использовать несуществующий индекс**: Если в списке 3 значения, индексы - 0, 1, 2. `fruits[3]` вызывает `IndexError: list index out of range` ("индекс выходит за пределы списка").
3. **Пытаться добавить значение через новый индекс**: `fruits[3] = "grape"` в списке из 3 значений **не** добавляет значение. Это вызывает `IndexError: list assignment index out of range` ("нельзя сохранить значение под индексом за пределами списка"). Чтобы добавить, используйте `append()`.
4. **Удалять значение, которого нет в списке**: `fruits.remove("grape")` вызывает `ValueError: list.remove(x): x not in list` ("этого значения нет в списке"). Сначала проверьте с помощью `in`.
5. **Путать `( )` и `[ ]`**: `[ ]` используются, чтобы создать список и получить значение по индексу (`fruits[0]`). `( )` используются для команд, например `len(fruits)` и `fruits.append("grape")`.

Ячейки ниже показывают, как выглядит каждая ошибка. Ошибочный код записан в виде комментариев, чтобы не ломать блокнот. Попробуйте убрать `#` в строке с кодом, чтобы увидеть настоящее сообщение об ошибке, а потом верните `#` обратно.

````python
# Ошибка 1: забыть, что индекс начинается с 0
fruits = ["apple", "banana", "cherry"]

print(fruits[1])        # Многие ожидают "apple", но это ВТОРОЕ значение -> banana
print(fruits[0])        # Правильно: первое значение под индексом 0 -> apple
````

````python
# Ошибка 2: использовать несуществующий индекс
fruits = ["apple", "banana", "cherry"]      # 3 значения -> индексы 0, 1, 2

# print(fruits[3])      # IndexError: list index out of range -> индекса 3 нет!

print(fruits[2])                    # Правильно: последний индекс - 2
print(fruits[len(fruits) - 1])      # Тоже правильно: len(fruits) - 1 = 3 - 1 = 2
print(fruits[-1])                   # Тоже правильно: -1 - всегда последнее значение
````

````python
# Ошибка 3: пытаться добавить значение через новый индекс
fruits = ["apple", "banana", "cherry"]

# fruits[3] = "grape"   # IndexError: list assignment index out of range -> ИЗМЕНЯТЬ можно только значения под уже существующими индексами

fruits.append("grape")  # Правильно: используем append(), чтобы добавить новое значение в конец
print(fruits)
````

````python
# Ошибка 4: удалять значение, которого нет в списке
fruits = ["apple", "banana", "cherry"]

# fruits.remove("grape")    # ValueError: list.remove(x): x not in list -> "grape" нет в списке!

# Правильно: сначала проверяем с помощью in (раздел 4) и удаляем, только если значение есть
if "grape" in fruits:
    fruits.remove("grape")
else:
    print("grape is not in the list, so nothing was removed.")
````

````python
# Ошибка 5: путать ( ) и [ ]
fruits = ["apple", "banana", "cherry"]

# print(fruits(0))          # TypeError: 'list' object is not callable -> ( ) нужны для команд, а не для индексов
# fruits.append["grape"]    # TypeError -> append - это команда, поэтому нужны ( ), а не [ ]

print(fruits[0])            # Правильно: [ ] чтобы получить значение по индексу
fruits.append("grape")      # Правильно: ( ) для команды
print(fruits)
````
